# Introduction

Recurrent Neural Networks (RNNs) were the first neural networks designed for sequential data.

However, they suffered from a major limitation:

❌ **Poor long-term memory**

To overcome this problem, **LSTM (Long Short-Term Memory)** was introduced in **1997**.

Although LSTM successfully learned long-term dependencies, its architecture became relatively complex.

To simplify LSTM while maintaining similar performance, researchers introduced the **Gated Recurrent Unit (GRU)** in **2014**.

---

# Evolution of Sequence Models

```
Simple RNN (1986)

        │

Cannot remember
long-term information

        ▼

LSTM (1997)

        │

Learns long-term dependencies
using three gates

        ▼

GRU (2014)

        │

Simplifies LSTM architecture
using only two gates

        ▼

Transformer (2017)

        │

Parallel processing
using Self-Attention
```

# Why Was GRU Introduced?

Researchers observed the following progression:

### Step 1: Simple RNN

Problem:

- Could not remember long-term information.
- Suffered from the vanishing gradient problem.

---

### Step 2: LSTM

Solution:

- Introduced Cell State.
- Added Forget, Input, and Output Gates.
- Successfully captured long-term dependencies.

New Problems:

- Complex architecture.
- More trainable parameters.
- Slower training.
- Higher computational cost.

---

### Step 3: GRU

Researchers asked:

> **"Can we keep the memory power of LSTM but make the architecture simpler?"**

The answer was **GRU**.

---

# What is a GRU?

## Definition

**GRU (Gated Recurrent Unit)** is a simplified version of LSTM that uses fewer gates while still learning long-term dependencies effectively.

It is often described as a **lightweight version of LSTM**.

---

# Why is GRU Simpler?

Unlike LSTM, GRU:

- Does not maintain a separate Cell State.
- Uses only one memory representation (Hidden State).
- Uses only two gates instead of three.

This makes the architecture:

- Simpler
- Faster
- More computationally efficient

---

# Memory in GRU

## LSTM

LSTM maintains:

- Cell State (Long-Term Memory)
- Hidden State (Short-Term Memory)

```
Cell State

+

Hidden State
```

---

## GRU

GRU merges both memories into a single:

**Hidden State**

```
Hidden State

(Long-Term + Short-Term Information)
```

Instead of managing two separate memories, GRU stores both long-term and short-term information in the Hidden State.

---

# LSTM vs GRU Memory

| LSTM | GRU |
|-------|-----|
| Cell State + Hidden State | Hidden State only |
| Two memories | One merged memory |
| More complex | Simpler |

---

# Gates in GRU

Unlike LSTM, which has **three gates**, GRU has only **two gates**.

1. Update Gate
2. Reset Gate

These gates are **trainable** and are learned automatically during training through backpropagation.

They are **not hard-coded**.

---

# 1. Update Gate

## Purpose

The Update Gate decides:

> **"How much of the previous information should be kept?"**

It determines whether old information is still useful.

---

### Example

Sentence:

```
The movie was fantastic.

It was...
```

The word **fantastic** is still important.

The Update Gate keeps this information in memory.

---

### Human Analogy

Suppose you're reading a book.

You don't forget the main character after every chapter.

You continue remembering important details.

The Update Gate behaves similarly.

---

# 2. Reset Gate

## Purpose

The Reset Gate decides:

> **"How much past information should be ignored when processing the current input?"**

It helps the model forget irrelevant context.

---

### Example

Sentence:

```
Yesterday it rained heavily.

Today the weather is sunny.
```

When processing **today**, the importance of **yesterday's rain** decreases.

The Reset Gate reduces the influence of outdated information.

---

### Human Analogy

Imagine changing topics in a conversation.

When discussing a new subject, you naturally ignore details from the previous topic.

The Reset Gate performs this function.

---

# Working of a GRU

At each time step, the GRU receives:

- Current Input (Xt)
- Previous Hidden State (ht−1)

Then:

```
Current Input

        │

Previous Hidden State

        │

──────────────
Update Gate
──────────────

Keep useful information

        │

──────────────
Reset Gate
──────────────

Forget irrelevant information

        │

Updated Hidden State

        │

Prediction
```

Unlike LSTM, there is:

- No Cell State.
- No Output Gate.

Everything is managed using the Hidden State.

# 2. Interview Questions & Answers

## Beginner Questions

### Q1. What does GRU stand for?

**Answer:** GRU stands for **Gated Recurrent Unit**, a simplified version of LSTM designed for sequence modeling.

### Q2. Why was GRU introduced?

**Answer:** GRU was introduced in 2014 to simplify the complex architecture of LSTM while still capturing long-term dependencies effectively.

### Q3. How many gates does a GRU have?

**Answer:** GRU has **two gates**:

* Update Gate
* Reset Gate

### Q4. Does GRU have a Cell State?

**Answer:** No. GRU combines long-term and short-term information into a single **Hidden State**, eliminating the need for a separate Cell State.

### Q5. Is GRU faster than LSTM?

**Answer:** Yes. GRU has fewer parameters and a simpler architecture, making it generally faster to train than LSTM.

---

## Intermediate Questions

### Q1. What is the role of the Update Gate?

**Answer:** The Update Gate decides how much information from the previous Hidden State should be retained and how much new information should be incorporated.

### Q2. What is the role of the Reset Gate?

**Answer:** The Reset Gate determines how much of the previous Hidden State should be ignored when processing the current input.

### Q3. Why does GRU have fewer parameters than LSTM?

**Answer:** GRU removes the separate Cell State and Output Gate, using only two gates instead of three, which reduces the number of trainable parameters.

### Q4. When would you choose GRU over LSTM?

**Answer:** GRU is often preferred when faster training, lower memory usage, and computational efficiency are priorities, especially on smaller datasets.

### Q5. Does GRU completely solve the limitations of LSTM?

**Answer:** No. While it simplifies the architecture and improves efficiency, both GRU and LSTM still process sequences sequentially and are generally outperformed by Transformers on many modern NLP tasks.

---

## Scenario-Based Questions

### Q1. Your NLP model performs well but training is slow due to LSTM complexity. What could you try?

**Answer:** Replace the LSTM with a GRU to reduce parameters and speed up training while maintaining good performance.

### Q2. You're deploying a sequence model on a mobile device with limited resources. Which architecture is suitable?

**Answer:** GRU, because its simpler architecture and fewer parameters make it more computationally efficient.

### Q3. Your task involves extremely long documents and you have powerful GPUs. Would you choose GRU?

**Answer:** For very long documents, a Transformer is usually a better choice because it captures long-range dependencies more effectively and supports parallel processing.

### Q4. If the Reset Gate always outputs 0, what happens?

**Answer:** The model ignores most of the previous Hidden State when processing the current input, focusing mainly on the current input.

### Q5. If the Update Gate always outputs 1, what happens?

**Answer:** The model retains most of the previous Hidden State, preserving past information while incorporating very little new information.

# GRU (Gated Recurrent Unit)

A **GRU (Gated Recurrent Unit)** is a type of **Recurrent Neural Network (RNN)** designed to solve the **vanishing gradient problem** that affects traditional RNNs.

GRU was introduced by Cho et al. (2014) as a **simpler and faster alternative to LSTM**.

---

# Why Do We Need GRU?

## Problem with Traditional RNN

Imagine reading a sentence:

> "I grew up in France. I speak fluent ______."

To predict the last word, the model needs to remember **France**.

A standard RNN often forgets information after many time steps because of:

* Vanishing gradients
* Exploding gradients
* Short-term memory

So researchers introduced gated architectures:

* LSTM
* GRU

Both help the network remember important information for longer.

---

# Basic Idea

Instead of updating memory blindly, GRU asks two questions:

1. **Should I forget the old information?**
2. **Should I store the new information?**

It uses **gates** to make these decisions.

---

# GRU Architecture

```
                Previous Hidden State
                       h(t−1)
                          |
              -----------------------
              |                     |
       Reset Gate              Update Gate
              |                     |
              -----------     -------
                         |   |
                   Candidate Memory
                         |
                    New Hidden State
                         h(t)
```

Unlike LSTM:

* No Cell State
* Only Hidden State

This makes GRU simpler.

---

# Components of GRU

There are only **two gates**.

## 1. Update Gate (z)

Controls:

> How much old information should be kept?

Formula

[
z_t=\sigma(W_z[h_{t-1},x_t])
]

where:

* (x_t) = current input
* (h_{t-1}) = previous hidden state
* (\sigma) = sigmoid

Output ranges between **0 and 1**.

### Example

If

```
z = 1
```

Keep almost everything from previous memory.

If

```
z = 0
```

Ignore old memory.

---

## 2. Reset Gate (r)

Controls:

> How much previous memory should be ignored while creating new memory?

Formula

[
r_t=\sigma(W_r[h_{t-1},x_t])
]

If

```
r = 0
```

Forget previous state.

If

```
r = 1
```

Use previous memory fully.

---

## 3. Candidate Hidden State

GRU creates a temporary memory.

Formula

[
\tilde{h_t}=\tanh(W[r_t*h_{t-1},x_t])
]

Notice

Previous hidden state is multiplied by

```
Reset Gate
```

---

## 4. Final Hidden State

This is the actual memory passed to the next timestep.

Formula

[
h_t=(1-z_t)*h_{t-1}+z_t*\tilde{h_t}
]

Meaning

```
New Memory

=
Old Memory
+
Candidate Memory
```

weighted by the update gate.

---

# Flow Diagram

```
Input xt
   |
   |
Previous Hidden ht-1
   |
   |---------------------------
   |                          |
Update Gate               Reset Gate
   |                          |
   |                    Candidate Memory
   |                          |
   |--------------------------|
              |
        Final Hidden State
              |
           Next Time Step
```

---

# Step-by-Step Example

Suppose

```
Previous Memory = 0.8
```

Current input suggests

```
New Memory = 0.2
```

Update gate

```
z = 0.1
```

Output

```
h = (1−0.1)(0.8)+0.1(0.2)

=0.72+0.02

=0.74
```

The network keeps most of the old memory.

---

Another example

```
Old Memory = 0.8

Candidate = 0.2

Update Gate = 0.9
```

Output

```
0.08+0.18

=0.26
```

Now it mostly trusts the new information.

# Why Is GRU Faster Than LSTM?

LSTM has:

* Input Gate
* Forget Gate
* Output Gate
* Cell State

GRU has only:

* Update Gate
* Reset Gate
* Hidden State

Fewer parameters mean:

* Faster training
* Less memory usage
* Simpler architecture

---

# GRU vs LSTM

| Feature                    | RNN  | GRU              | LSTM      |
| -------------------------- | ---- | ---------------- | --------- |
| Vanishing Gradient Problem | ❌    | ✅                | ✅         |
| Gates                      | 0    | 2                | 3         |
| Cell State                 | ❌    | ❌                | ✅         |

| Training Speed             | Fast | Faster than LSTM | Slower    |
| Long-Term Memory           | Poor | Good             | Excellent |
| Memory Consumption         | Low  | Medium           | High      |

---

# Advantages of GRU

* Simpler architecture than LSTM
* Fewer parameters
* Faster training
* Faster inference
* Works well with smaller datasets
* Handles vanishing gradients much better than a vanilla RNN

---

# Disadvantages

* Slightly less expressive than LSTM for very long sequences
* May underperform LSTM on tasks requiring very long-term dependencies
* Less control because it combines some functions that LSTM keeps separate

---

# PyTorch Example

```python
import torch
import torch.nn as nn

# input_size = number of features
# hidden_size = number of hidden units
# num_layers = stacked GRU layers

gru = nn.GRU(
    input_size=10,
    hidden_size=64,
    num_layers=2,
    batch_first=True
)

# Batch size = 32
# Sequence length = 100
# Features = 10

x = torch.randn(32, 100, 10)

output, hidden = gru(x)

print(output.shape)  # (32, 100, 64)
print(hidden.shape)  # (2, 32, 64)
```

* `output`: hidden state for every time step.
* `hidden`: final hidden state from each GRU layer.

---

# TensorFlow/Keras Example

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

model = Sequential([
    GRU(64, input_shape=(100, 10)),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

# Summary

* **GRU (Gated Recurrent Unit)** is an improved RNN that combats the vanishing gradient problem.
* It uses **two gates**:

  * **Update Gate**: controls how much previous information is retained.
  * **Reset Gate**: controls how much previous information influences the candidate state.
* Unlike LSTM, it **does not maintain a separate cell state**; the hidden state serves as the model's memory.
* GRUs are computationally efficient and often achieve performance comparable to LSTMs while training faster, making them a strong default choice for many sequence modeling tasks.
