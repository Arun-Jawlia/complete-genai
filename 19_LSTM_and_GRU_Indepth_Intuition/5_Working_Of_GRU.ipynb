# Working of GRU (Gated Recurrent Unit)

A **Gated Recurrent Unit (GRU)** is a type of **Recurrent Neural Network (RNN)** that improves the ability of a network to learn long-term dependencies. Unlike a traditional RNN, a GRU uses **gates** to decide what information should be remembered and what should be forgotten at each time step.

The GRU contains only **two gates**:

1. **Update Gate (zₜ)**
2. **Reset Gate (rₜ)**

Unlike an LSTM, a GRU **does not have a separate cell state**. It maintains only a **hidden state (hₜ)**.

---

# GRU Architecture

```text
                Previous Hidden State h(t−1)
                          |
               -----------------------
               |                     |
         Update Gate (z)       Reset Gate (r)
               |                     |
               |                Candidate State
               |                     |
               ------- Combine -------
                        |
                 New Hidden State h(t)
                        |
                   Next Time Step
```

---

# Step-by-Step Working of GRU

Suppose we are processing a sentence one word at a time.

At each time step, the GRU receives:

* Current input: **xₜ**
* Previous hidden state: **hₜ₋₁**

It then performs the following steps.

---

## Step 1: Calculate the Update Gate

The **update gate** determines **how much of the previous hidden state should be carried forward**.

### Formula

[
$z_t = \sigma(W_zx_t + U_zh_{t-1} + b_z)$
]

Where:

* (x_t) = current input
* (h_{t-1}) = previous hidden state
* (W_z, U_z) = weight matrices
* (b_z) = bias
* (\sigma) = Sigmoid activation

The sigmoid function outputs values between **0 and 1**.

### Interpretation

* **z = 0** → Ignore previous memory and use mostly new information.
* **z = 1** → Keep almost all previous memory.

---

## Step 2: Calculate the Reset Gate

The **reset gate** determines **how much of the previous hidden state should be ignored while creating the candidate hidden state**.

### Formula

[
$r_t = \sigma(W_rx_t + U_rh_{t-1} + b_r)$
]

### Interpretation

* **r = 0** → Forget previous memory.
* **r = 1** → Use previous memory completely.

The reset gate is useful when the model needs to ignore irrelevant past information and focus on the current input.

---

## Step 3: Compute the Candidate Hidden State

The GRU creates a **candidate hidden state**, which represents the new information learned from the current input.

Before computing it, the previous hidden state is filtered using the reset gate.

### Formula

[
$\tilde{h_t}=\tanh(W_hx_t+U_h(r_t \odot h_{t-1})+b_h)$
]

Here:

* (\odot) denotes element-wise multiplication.
* (\tanh) produces values between **−1 and 1**.

If the reset gate is close to 0, the previous hidden state contributes very little to the candidate state.

---

## Step 4: Compute the Final Hidden State

The final hidden state is obtained by combining:

* Previous hidden state
* Candidate hidden state

The update gate determines the proportion of each.

### Formula

[
h_t=(1-z_t)\odot h_{t-1}+z_t\odot\tilde{h_t}
]

### Interpretation

* Large **zₜ** → More weight to the candidate hidden state (new information).
* Small **zₜ** → More weight to the previous hidden state (old information).

This hidden state becomes:

* The output for the current time step.
* The input hidden state for the next time step.

---

# Complete Workflow

```text
Current Input (xₜ)
        |
        |
Previous Hidden State (hₜ₋₁)
        |
        |-----------------------------|
        |                             |
        |                      Reset Gate (rₜ)
        |                             |
        |                       rₜ × hₜ₋₁
        |                             |
        |                      Candidate State (h̃ₜ)
        |                             |
        |-----------------------------|
                      |
               Update Gate (zₜ)
                      |
        ---------------------------------
        |                               |
Previous Hidden State            Candidate State
      (1−zₜ) × hₜ₋₁               zₜ × h̃ₜ
        |                               |
        ----------- Add Together --------
                      |
              Final Hidden State (hₜ)
                      |
                Next Time Step
```

[
$h_t=(1-z_t)h_{t-1}+z_t\tilde{h_t}$

# Real-Life Analogy

Imagine you are reading a news article.

### Previous memory

> "The company reported strong profits."

Now you read:

> "However, the CEO resigned unexpectedly."

The GRU makes two decisions:

### Reset Gate

"Should I ignore some of the earlier context while interpreting this new sentence?"

If yes, the reset gate becomes small.

### Update Gate

"Should this new information replace what I currently believe?"

If the CEO's resignation is highly significant, the update gate becomes large, allowing the new information to strongly influence the hidden state.

# Why GRU Works Better Than a Simple RNN

A standard RNN updates its hidden state directly at every time step, which can cause important information to fade over long sequences due to the **vanishing gradient problem**.

A GRU addresses this by:

* Selectively preserving important information using the **update gate**.
* Ignoring irrelevant past information using the **reset gate**.
* Learning long-term dependencies more effectively.
* Training faster than an LSTM because it has fewer parameters.

---

# Summary

| Step | Operation                            | Purpose                                               |
| ---- | ------------------------------------ | ----------------------------------------------------- |
| 1    | Compute Update Gate (zₜ)             | Decide how much previous information to keep          |
| 2    | Compute Reset Gate (rₜ)              | Decide how much previous information to ignore        |
| 3    | Compute Candidate Hidden State (h̃ₜ) | Generate new information from the current input       |
| 4    | Compute Final Hidden State (hₜ)      | Combine old and new information using the update gate |

### Key Points

* **Input:** Current input (**xₜ**) and previous hidden state (**hₜ₋₁**).
* **Memory:** Only one hidden state is maintained (no separate cell state).
* **Gates:** Two gates—Update Gate and Reset Gate.
* **Output:** The new hidden state (**hₜ**) is passed to the next time step and can also be used as the output for the current step.

In essence, a GRU continuously decides **what to remember, what to forget, and what to update**, enabling it to model sequential data much more effectively than a standard RNN while remaining simpler than an LSTM.

# GRU vs Simple RNN

| Feature                    | Simple RNN                         | GRU                                           |
| -------------------------- | ---------------------------------- | --------------------------------------------- |
| Long-term Memory           | Poor (forgets long dependencies)   | Good                                          |
| Vanishing Gradient Problem | Severe                             | Greatly Reduced                               |
| Training Stability         | Low                                | High                                          |
| Number of Gates            | None                               | 2 (Update, Reset)                             |
| Hidden State Control       | Direct update                      | Controlled using gates                        |
| Training Speed             | Faster per step but less effective | Slightly slower than RNN, much more effective |

> **Key Point:** A GRU introduces gates that allow it to selectively retain or discard information, making it much better at learning long-term dependencies than a vanilla RNN.

---

# GRU vs LSTM

| Feature                            | LSTM                               | GRU               |
| ---------------------------------- | ---------------------------------- | ----------------- |
| Number of Gates                    | 3 (Input, Forget, Output)          | 2 (Update, Reset) |
| Cell State                         | Separate Cell State + Hidden State | Only Hidden State |
| Number of Parameters               | More                               | Fewer             |
| Training Speed                     | Slower                             | Faster            |
| Memory Usage                       | Higher                             | Lower             |
| Ability to Learn Long Dependencies | Excellent                          | Very Good         |
| Simplicity                         | More Complex                       | Simpler           |

**Key Point**

> On many NLP and time-series tasks, GRU achieves performance comparable to LSTM while requiring fewer parameters and training faster.

# When to Use GRU

Choose GRU when:

1. The dataset is **small to medium-sized**.
2. Faster training and inference are important.
3. Memory efficiency matters.
4. You need good performance with fewer parameters.
5. You want a simpler model than LSTM.

> **Note:** Instead of saying "except lakhs of rows," it's more accurate to say **small to medium datasets**. Dataset size alone doesn't determine whether GRU or LSTM is better.

---

# When to Use LSTM Instead

Prefer LSTM when:

* Very long-term dependencies must be captured.
* The sequence is extremely long.
* Slightly higher accuracy is more important than speed.
* Additional parameters and computation are acceptable.

# Drawbacks of GRU

1. **Sequential Computation**

   * Like all RNN-based models, GRUs process one time step after another.
   * This limits parallelization during training.

2. **Struggles with Very Long Context**

   * Although much better than a simple RNN, performance can degrade for extremely long sequences.

3. **Less Expressive than LSTM**

   * GRU has fewer parameters, which can make it slightly less powerful on some tasks requiring complex memory management.

4. **Not State-of-the-Art for Modern NLP**

   * Transformers generally outperform GRUs on most large-scale NLP tasks.

---

# Why Were Transformers Introduced?

Both RNNs, LSTMs, and GRUs suffer from two main limitations:

* They process tokens **sequentially**, making training slower.
* Capturing dependencies across very long sequences is difficult.

Transformers solve these problems using the **Attention Mechanism**, which:

* Processes all tokens in parallel.
* Captures relationships between any two words regardless of their distance.
* Scales much better for large datasets and long contexts.
