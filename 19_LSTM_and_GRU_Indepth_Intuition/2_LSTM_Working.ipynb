# Introduction

A Simple RNN updates its memory at every time step.

As sequences become longer:

- Important information gets overwritten.
- Old context is forgotten.
- Long-term dependencies are difficult to learn.

LSTM (Long Short-Term Memory) solves this by introducing **two types of memory** and a **gating mechanism** that controls the flow of information.

---

# How Does an LSTM Work?

Unlike a Simple RNN, an LSTM does not rely only on one hidden state.

It maintains two memories:

1. **Cell State (Long-Term Memory)**
2. **Hidden State (Short-Term Memory)**

These memories work together to process sequential data efficiently.

---

# Two Types of Memory in LSTM

## 1. Cell State (Long-Term Memory)

### Definition

The **Cell State (Ct)** is the long-term memory of an LSTM.

It carries important information across many time steps with minimal modification.

### Characteristics

- Stores long-term information.
- Preserves important context.
- Helps reduce the vanishing gradient problem.
- Updated selectively using gates.

### Example

Sentence:

```
My name is Arun.

...

I am a software engineer.
```

The information **"Arun"** may need to be remembered even after many words.

The Cell State stores this important information.

---

## 2. Hidden State (Short-Term Memory)

### Definition

The **Hidden State (ht)** is the short-term memory of an LSTM.

It represents the current context and is passed:

- To the next time step.
- To the output layer.

### Characteristics

- Stores recent information.
- Used for immediate predictions.
- Changes at every time step.

Think of it as the "current understanding" of the sequence.

---

# Cell State vs Hidden State

| Feature | Cell State (Ct) | Hidden State (ht) |
|----------|-----------------|-------------------|
| Purpose | Long-term memory | Short-term memory |
| Stores | Important information over long sequences | Current context |
| Updated by | Forget & Input Gates | Output Gate |
| Passed to | Next Cell State | Next Hidden State & Output |
| Main Role | Preserve knowledge | Generate predictions |

---

# Inputs to an LSTM Cell

At every time step, the LSTM receives:

- Current input (Xt)
- Previous hidden state (ht−1)
- Previous cell state (Ct−1)

These inputs are processed by the gates.

```
Previous Hidden State (ht-1)
            │
            ▼

Current Input (Xt)

            │

Previous Cell State (Ct-1)
```

---

# Three Gates in LSTM

The gates act like intelligent filters.

They decide:

- What to forget.
- What to remember.
- What to output.

---

## 1. Forget Gate

### Purpose

Decides what information should be removed from the previous Cell State.

Question:

> **"Is this information still useful?"**

### Example

Sentence:

```
I live in Delhi.

Last year I lived in Mumbai.
```

For the current context, the old location (**Mumbai**) becomes less relevant.

The Forget Gate can remove this outdated information.

---

## 2. Input Gate

### Purpose

Determines what new information should be stored in the Cell State.

Question:

> **"Is this new information important?"**

### Example

Sentence:

```
Today I started learning Deep Learning.
```

The Input Gate stores the important concept:

```
Deep Learning
```

into long-term memory.

---

## 3. Output Gate

### Purpose

Determines what information from the Cell State should be exposed as the Hidden State.

Question:

> **"What information should be used right now?"**

This Hidden State is then used:

- For prediction.
- As input to the next LSTM cell.

---

# Complete LSTM Workflow

```
Current Input (Xt)
        │
        ▼

Previous Hidden State (ht−1)

        │

Previous Cell State (Ct−1)

        │

───────────────
Forget Gate
───────────────

Removes unnecessary information

        │

───────────────
Input Gate
───────────────

Adds important new information

        │

───────────────
Updated Cell State (Ct)
───────────────

Long-Term Memory

        │

───────────────
Output Gate
───────────────

Creates Hidden State (ht)

        │

Prediction
```

---

# Information Flow in LSTM

```
Sentence

↓

Tokenizer

↓

Embedding Layer

↓

LSTM Cell

↓

Forget Gate

↓

Input Gate

↓

Cell State Update

↓

Output Gate

↓

Hidden State

↓

Dense Layer

↓

Prediction
```

# Intuition Behind LSTM

Imagine preparing for an exam.

You don't memorize everything.

Instead, you:

- Forget irrelevant topics.
- Learn new important concepts.
- Recall only the information needed to answer a question.

This is exactly how an LSTM works.

It intelligently manages memory instead of storing everything.

---

# Problems Solved by LSTM

## 1. Vanishing Gradient Problem

### Problem

In Simple RNNs, gradients become very small during Backpropagation Through Time (BPTT).

As a result:

- Earlier layers stop learning.
- Long-term information is lost.

### Solution

The Cell State provides a smoother path for gradients, allowing important information to persist across many time steps.

---

## 2. Long-Term Dependency Problem

### Problem

Simple RNN remembers only recent information.

Important details from earlier in the sequence are forgotten.

### Solution

LSTM preserves important information in the Cell State for long durations.

---

## 3. Context Understanding

Sentence 1

```
The bank approved my loan.
```

Sentence 2

```
The children played near the bank.
```

The meaning of **bank** depends on the surrounding context.

The combination of Embedding + LSTM helps the model interpret the correct meaning using long-term context.

---

## 4. Information Filtering

Simple RNN:

Stores almost everything.

LSTM:

Learns to:

- Keep useful information.
- Remove unnecessary information.
- Use only relevant information for prediction.

# 2. Interview Questions & Answers

## Beginner Questions

### Q1. What are the two types of memory in an LSTM?

**Answer:** LSTM maintains a **Cell State (Ct)** for long-term memory and a **Hidden State (ht)** for short-term memory and current context.

### Q2. What is the purpose of the Cell State?

**Answer:** It stores important information across many time steps, enabling the model to capture long-term dependencies.

### Q3. What is the role of the Hidden State?

**Answer:** It represents the current context and is used for predictions and passed to the next LSTM cell.

### Q4. Name the three gates in an LSTM.

**Answer:** Forget Gate, Input Gate, and Output Gate.

### Q5. Why is the Forget Gate important?

**Answer:** It removes outdated or irrelevant information from the Cell State, preventing unnecessary memory accumulation.

---

## Intermediate Questions

### Q1. How do the Cell State and Hidden State work together?

**Answer:** The Cell State preserves long-term information, while the Hidden State provides the current context for predictions. Together, they allow LSTM to balance long-term memory with immediate processing.

### Q2. Why does LSTM perform better than Simple RNN on long sequences?

**Answer:** LSTM selectively updates memory through gates, allowing important information to persist and reducing the vanishing gradient problem.

### Q3. What inputs does an LSTM cell receive?

**Answer:** Each LSTM cell receives the current input (**Xt**), the previous Hidden State (**ht−1**), and the previous Cell State (**Ct−1**).

### Q4. Does the Cell State change at every time step?

**Answer:** Yes, but only selectively. The Forget and Input Gates determine which information is removed or added.

### Q5. Why does LSTM have more parameters than a Simple RNN?

**Answer:** Because it contains additional gates and maintains both Hidden State and Cell State, increasing the number of trainable weights.

---

## Scenario-Based Questions

### Q1. Your chatbot forgets information mentioned earlier in a conversation. Which model would you choose?

**Answer:** LSTM, because it can retain important information across longer sequences using its Cell State.

### Q2. A time-series model fails to capture seasonal trends. What improvement would you suggest?

**Answer:** Replace the Simple RNN with an LSTM to better learn long-term temporal dependencies.

### Q3. Why might an LSTM still struggle with extremely long documents?

**Answer:** Although LSTM handles long-term dependencies better than RNNs, very long sequences can still be challenging. Transformer models are generally more effective for such tasks.

### Q4. Why doesn't the Forget Gate remove all previous information?

**Answer:** It learns, during training, which information is important and which is not. Only irrelevant information is forgotten.

### Q5. What would happen if the Input Gate blocked all new information?

**Answer:** The Cell State would stop learning new patterns, causing the model to rely only on old information and reducing its adaptability.
