# `Steamlit in Python`
---
- Open Source Python Library used to build Beautiful, interactive web applicatoins for data science and machine learning with minimal effort.
- A faster way to build and share data apps
- Turn your data scripts into shareable web apps in minutes. All in pure Python. No front‑end experience required.

## Streamlit is one of the fastest ways to turn Python scripts into interactive web applications, especially for:

- Data science dashboards
- ML model demos
- Internal tools
- Analytics apps
- AI applications
- Quick prototypes

## Why to use Streamlit:
1. Easy to Use
2. Fast Development
3. Interactive UI Elements ( Buttons, Sliders , Text Input, Charts etc )
4. Great for Data Science ( Support Pandas, Matplotlib, Plotly and other libraries )
5. Deployment is Easy

## Installation
`pip install streamlit`
