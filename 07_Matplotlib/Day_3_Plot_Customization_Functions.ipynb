## `Plot Creation Functions`:
---


### 1. `plt.plot()`

- Purpose: Used to create Line Plots.
- Syntax: `plt.plot(x, y)`
- Example: plt.plot([1,2,3], [10,20,30])
- Common Use Cases:
  - Time Series Analysis
  - Trend Analysis
  - Sales Growth


### 2. `plt.bar()`

- Purpose: Used to create Vertical Bar Charts.
- Syntax: `plt.bar(x, height)`
- Example: plt.bar(['A','B','C'], [10,20,15])
- Common Use Cases:
  - Comparing categories
  - Product Sales Comparison
  - Department-wise Performance


### 3. `plt.barh()`

- Purpose: Used to create Horizontal Bar Charts.
- Syntax: `plt.barh(y, width)`
- Example: plt.barh(['A','B','C'], [10,20,15])
- Common Use Cases:
  - Long category names
  - Ranking analysis


### 4. `plt.scatter()`

- Purpose: Used to create Scatter Plots.
- Syntax: `plt.scatter(x, y)`
- Example: plt.scatter(age, salary)
- Common Use Cases:
  - Correlation Analysis
  - Relationship between two numerical columns


### 5. `plt.hist()`

- Purpose: Used to create Histograms.
- Syntax: `plt.hist(data)`
- Example: plt.hist(df['Age'])
- Common Use Cases:
  - Distribution Analysis
  - Frequency Distribution


### 6. `plt.pie()`

- Purpose: Used to create Pie Charts.
- Syntax: `plt.pie(values)`
- Example: plt.pie([40,30,20,10])
- Common Use Cases:
  - Percentage Analysis
  - Market Share Analysis


### 7. `plt.boxplot()`

- Purpose:Used to create Box Plots.
- Syntax: `plt.boxplot(data)`
- Example: plt.boxplot(df['Salary'])
- Common Use Cases:
  - Outlier Detection
  - Distribution Analysis

## `Plot Customization Functions`:
---


### 1. `plt.title()`

- Adds title to the chart.
- Syntax: `plt.title("Chart Title")`
- Example: plt.title("Monthly Sales")
- Parameters: color( What color the title is going to have)


### 2. `plt.xlabel()`

- Adds label to X-axis.
- Syntax: `plt.xlabel("X Label")`
- Example: plt.xlabel("Months")


### 3. `plt.ylabel()`

- Purpose: Adds label to Y-axis.
- Syntax: `plt.ylabel("Sales")`
- Example: plt.ylabel("Revenue")


### 4. `plt.legend()`

- Purpose: Displays labels for multiple plots.
- Syntax: `plt.legend()`
- Example:
  - plt.plot(x, y1, label="Sales")
  - plt.plot(x, y2, label="Profit")
  - plt.legend()


### 5. `plt.grid()`

- Purpose: Displays grid lines on chart.
- Syntax: `plt.grid()`
- Example: plt.grid()
- Benefit: Makes chart easier to read.


### 6. `plt.xticks()`

- Purpose: Customize X-axis tick labels.
- Syntax: `plt.xticks()`
- Example: plt.xticks(rotation=45)


### 7. `plt.yticks()`

- Purpose: Customize Y-axis tick labels.
- Syntax: `plt.yticks()`


### 8. `plt.figure()`

- Purpose: Creates a new figure and controls chart size.
- Syntax: `plt.figure(figsize=(width, height))`
- Example: plt.figure(figsize=(10,5))


### 9. `plt.show()`

- Purpose: Displays the final chart.
- Syntax: `plt.show()`


### 10. `plt.xlim()`: X values --> [10,1000] --> [100,1000] --> [700,800]
- it puts a limitation that how much data mush be represented on the X-Axis

### 11. `plt.ylim()`: X values --> [500,900] --> [600,700]
- it puts a limitation that how much data mush be represented on the Y-Axis

# Line Plot Customization Parameters
---

Matplotlib provides several parameters to customize the appearance of a line plot. These parameters help make the visualization more attractive, readable, and informative.

### 1. `color`

Used to define the color of the line.

**Examples:**

* `'red'`
* `'green'`
* `'blue'`
* `'black'`
* Hex Color Codes: `'#FF5733'`, `'#2ECC71'`

```python
plt.plot(x, y, color='red')
```

---

### 2. `marker`

Used to define how each data point appears on the graph.

**Common Marker Styles:**

| Marker | Shape       |
| ------ | ----------- |
| `'o'`  | Circle      |
| `'^'`  | Triangle Up |
| `'s'`  | Square      |
| `'+'`  | Plus        |
| `'d'`  | Diamond     |
| `'h'`  | Hexagon     |
| `'p'`  | Pentagon    |
| `'*'`  | Star        |
| `'x'`  | Cross       |

```python
plt.plot(x, y, marker='o')
```

---

### 3. `linestyle`

Used to define the pattern or style of the line.

**Common Line Styles:**

| Style       | Description   |
| ----------- | ------------- |
| `'-'`       | Solid Line    |
| `'--'`      | Dashed Line   |
| `'-.'`      | Dash-Dot Line |
| `':'`       | Dotted Line   |
| `'solid'`   | Solid Line    |
| `'dashed'`  | Dashed Line   |
| `'dashdot'` | Dash-Dot Line |
| `'dotted'`  | Dotted Line   |
| `'None'`    | No Line       |

```python
plt.plot(x, y, linestyle='--')
```

---

### 4. `linewidth`

Used to control the thickness of the line.

**Examples:**

* `1`
* `1.5`
* `2`
* `3`
* `5`

```python
plt.plot(x, y, linewidth=3)
```

A larger value creates a thicker line.

---

### 5. `markersize`

Used to control the size of the marker.

**Examples:**

* `4`
* `6`
* `8`
* `10`
* `12`

```python
plt.plot(x, y, marker='o', markersize=10)
```

A larger value creates a bigger marker.

---

### 6. `label`

Used to provide a name for a line. It is especially useful when multiple lines are plotted on the same graph because it helps identify what each line represents.

```python
plt.plot(x, y, label='Sales')
plt.legend()
```

The label becomes visible only when `plt.legend()` is used.

---

### Example

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 120, 150, 180]

plt.plot(
    months,
    sales,
    color='blue',
    marker='o',
    linestyle='--',
    linewidth=2,
    markersize=8,
    label='Monthly Sales'
)

plt.legend()
plt.show()
```

In the above example:

* The line color is blue.
* Circular markers are displayed at each data point.
* The line style is dashed.
* The line width is 2.
* The marker size is 8.
* A legend is displayed with the label "Monthly Sales".


````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 120, 150, 180]

plt.plot(
    months,
    sales,
    color='blue',
    marker='o',
    linestyle='--',
    linewidth=2,
    markersize=8,
    label='Monthly Sales'
)

plt.legend()
plt.show()
````
