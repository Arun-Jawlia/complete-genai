# `2D Line Plot`
---

- A 2D Line Plot is a graph used to show the relationship between two variables/columns (X and Y) by connecting data points with lines.

- It is one of the most commonly used charts for `time series analysis and trend analysis`.

### `Why Use a Line Plot?`

A line plot helps us:

- Identify trends over time
- Compare values across intervals
- Detect increases and decreases
- Observe patterns and seasonality

---
### `When to Use a Line Plot`

- Time Series Analysis

    - Monthly Sales
    - Daily Temperature
    - Stock Prices
    - Website Traffic

- Trend Analysis

    - Revenue Growth
    - Population Growth
    - Profit Analysis

---
### Advantages
- Easy to understand
- Best for time-based data
- Shows trends clearly
- Helps compare multiple datasets

### Limitations
- Not suitable for categorical comparison with many categories
- Can become cluttered with too many lines
- Not ideal for part-to-whole analysis

---

| Feature             | 2D Line Plot         |
| ------------------- | -------------------- |
| Dimensions          | X and Y              |
| Analysis Type       | Bivariate            |
| Best For            | Trend Analysis       |
| Common Use          | Time Series Data     |
| Matplotlib Function | `plt.plot()`         |
| Axes Required       | X-axis and Y-axis    |
| Output              | Connected Line Graph |


---

## `Structure of a 2D Line Plot`

1. X-Axis (Horizontal Axis) → Represent Independent Variable
   - x = ["Jan", "Feb", "Mar"]    

2. Y-Axis (Vertical Axis) → Represent Dependent Variable
   - y = [100, 150, 200]

3. Data Points: Each (x,y) pair forms a point.
("Jan",100)
("Feb",150)
("Mar",200)

1. Connectings Lines → CMatplotlib joins all points to show trends.
2. plt.plot(x,y)

## `Components of a Line Plot`
1. X-Axis: Represents the independent variable.

   - Examples: Date, Month, Year, Time

   - x = ["Jan", "Feb", "Mar"] 

2. Y-Axis: Represents the dependent variable.

   - Examples: Sales, Temperature, Revenue, Profit

   - y = [100, 150, 200]

3. Data Points: Each (x,y) pair forms a point.

    - ("Jan",100)
    - ("Feb",150)
    - ("Mar",200)
  
1. Connecting Lines: Matplotlib joins all points to show trends.
   - `plt.plot(x,y)`

## To convert a Date Column to datetime
- Use `pd.to_datetime()` 
- Extract year value: `df['column_name].dt.year()`
- Extract month value: `df['column_name].dt.month()`
- Extract month name: `df['column_name].dt.month_name()`

```python
sales['Date'] = pd.to_datetime(sales['Date'], format='%d/%m/%Y')
```

```python
sales['Date']= sales['Date'].dt.month_name()
```

```python
sales['Date']= sales['Date'].astype("category")
```

```python
sales['Date']
```

```python
sales['Amount'] = sales['Amount'].str.strip("$")
```

```python
sales['Amount'] = sales['Amount'].str.replace(",", '')
```

```python
sales['Amount'] = sales['Amount'].astype("float").astype("int32")
```

```python
sales['Amount']
```

```python
sales['Boxes Shipped'] = sales['Boxes Shipped'].astype('int32')
```

```python
sales.info()
```

```python
# Group Date by Amount
sales_data = sales.groupby('Date')["Amount"].sum()
```

```python
sales_data.index = pd.CategoricalIndex(sales_data.index, categories=['January', 'February','March',"April", "May", "June", "July", "August"], ordered=True)
```

```python
sales_data = sales_data.sort_index()
```

```python
x = sales_data.index
```

```python
plt.figure(figsize=(10,5))
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.plot(x,y, linewidth=2, linestyle='--', color="red", marker='o')
plt.grid()
plt.show()
```
