# Step 2: Embeddings in Transformers

## Definition

After **Tokenization**, the Transformer receives **Token IDs (integers)**. However, these IDs are just unique identifiers and **do not contain any meaning**.

The next step is **Embeddings**, where each token ID is converted into a **dense vector of real numbers** that captures its **semantic meaning**.

> **Definition:**
> An **Embedding** is a dense numerical vector that represents the semantic meaning of a token, allowing the Transformer to understand relationships between words.

---

# Complete Transformer Pipeline

```text
Input Text
      ↓
Tokenizer
      ↓
Token IDs
      ↓
Embeddings
      ↓
Positional Encoding
      ↓
Transformer Layers
      ↓
Output
```

---

# Why Do We Need Embeddings?

Suppose we have the sentence:

```text
I love AI
```

After tokenization:

| Token | Token ID |
| ----- | -------- |
| I     | 15       |
| love  | 87       |
| AI    | 324      |

These IDs are simply labels.

The model cannot infer that:

* "AI" and "Machine Learning" are related.
* "love" has a positive sentiment.
* "cat" and "dog" are more similar than "cat" and "car".

Therefore, token IDs must be converted into vectors that encode semantic relationships.

---

# Token IDs vs Embeddings

## Token IDs

```text
Hello → 15496
World → 2159
```

These numbers only identify tokens.

They do **not** indicate:

* Similarity
* Meaning
* Context
* Relationships

---

## Embeddings

Each token is converted into a vector.

Example:

```text
Hello

↓

[0.23, -0.84, 1.12, 0.45, ...]
```

```text
World

↓

[-0.91, 0.34, 0.67, 1.08, ...]
```

Instead of one integer, every token becomes a vector with hundreds or thousands of values.

---

# What Does an Embedding Represent?

An embedding captures many hidden features learned during training.

For example, the embedding of the word:

```text
King
```

might implicitly encode:

* Gender
* Royalty
* Human
* Power
* Leadership

Similarly,

```text
Queen
```

will have a similar embedding but differ mainly in gender-related dimensions.

These meanings are **not manually assigned**; they are learned automatically during training.

---

# Real-World Analogy

Imagine every person has an ID card.

```text
Rahul → Employee ID = 501
Priya → Employee ID = 502
```

The ID tells you who the person is, but nothing else.

Now imagine a detailed profile:

```text
Age
Experience
Department
Skills
Location
Education
```

This profile describes the person meaningfully.

Similarly:

* **Token ID** = Identity
* **Embedding** = Rich description of the token

# Embedding Matrix

The Transformer stores embeddings in a large **Embedding Matrix**.

Suppose the vocabulary has **50,000 tokens** and the embedding size is **768**.

The embedding matrix size will be:

```text
Vocabulary Size × Embedding Dimension

50,000 × 768
```

Every row represents one token.

Example:

| Token ID | Embedding Vector      |
| -------- | --------------------- |
| 0        | [0.1, 0.3, -0.4, ...] |
| 1        | [0.9, -0.2, 0.5, ...] |
| 2        | [-0.8, 0.7, 0.1, ...] |

When the tokenizer outputs token ID **1205**, the model simply retrieves row **1205** from the embedding matrix.

---

# How Are Embeddings Learned?

Embeddings are **not hard-coded**.

Initially:

* The embedding matrix is initialized with random values.

During training:

1. The model makes predictions.
2. Loss is calculated.
3. Backpropagation updates the embedding vectors.
4. Similar words gradually move closer together in the vector space.

Over millions or billions of training examples, embeddings become highly meaningful.

---

# Semantic Similarity

Words with similar meanings tend to have similar embeddings.

Example:

```text
King
Queen
Prince
Princess
```

These vectors will be close together.

Likewise:

```text
Car
Truck
Bus
Vehicle
```

will occupy a nearby region in the embedding space.

Whereas:

```text
Apple
Galaxy
Football
Quantum
```

will generally be farther apart because they represent different concepts.

---

# Embedding Dimension

An embedding is not just one number.

It is a vector with many dimensions.

Common sizes include:

| Model       | Embedding Size       |
| ----------- | -------------------- |
| BERT Base   | 768                  |
| BERT Large  | 1024                 |
| GPT-2 Small | 768                  |
| GPT-3       | 12,288 (hidden size) |

Larger models typically use larger embedding dimensions to capture richer information.

---

# Do Embeddings Understand Word Order?

No.

Embeddings only represent the meaning of individual tokens.

For example:

```text
Dog bites man
```

and

```text
Man bites dog
```

contain the same token embeddings but have different meanings because of the order.

To encode order, Transformers add **Positional Encoding** after embeddings.

> **Next Step:** Token Embeddings + Positional Encoding = Input to the Transformer.

# Why Are Embeddings Important?

Without embeddings:

* Tokens would just be random IDs.
* The model could not understand semantic relationships.
* Words with similar meanings would appear unrelated.

Embeddings allow the Transformer to:

* Understand semantic similarity.
* Learn context.
* Capture relationships between words.
* Generalize to unseen combinations.

# Embeddings in Transformers

## What is an Embedding?

An **Embedding** is a dense vector representation of a token (word, subword, or character) that captures its semantic meaning.

After tokenization, the input text becomes **token IDs**, but token IDs themselves carry **no semantic information**.

For example:

```text
I          → 15
love       → 89
AI         → 700
```

The numbers **15**, **89**, and **700** are just identifiers. They do not indicate that "AI" is related to "Machine Learning" or "Deep Learning".

The **Embedding Layer** converts these IDs into meaningful vectors that the Transformer can learn from.

---

# Why Do We Need Embeddings?

Computers understand numbers, but token IDs are arbitrary.

Example:

```text
Dog → 10

Cat → 11

Car → 12
```

From these IDs, the model cannot infer that **Dog** and **Cat** are more similar than **Dog** and **Car**.

The embedding layer transforms them into vectors like:

```text
Dog

↓

[0.21, -0.65, 0.88, 0.14]

Cat

↓

[0.24, -0.61, 0.85, 0.12]

Car

↓

[-0.71, 0.42, -0.58, 0.91]
```

Notice that the vectors for **Dog** and **Cat** are more similar than the vector for **Car**.

# Example

Sentence:

```text
I love AI
```

### Step 1: Tokenization

```text
["I", "love", "AI"]
```

### Step 2: Token IDs

```text
I      → 15

love   → 89

AI     → 700
```

Input:

```text
[15, 89, 700]
```

### Step 3: Embedding Lookup

Suppose the embedding dimension is **4**.

```text
15

↓

[0.12, 0.45, -0.78, 0.34]

89

↓

[-0.21, 0.90, 0.11, -0.45]

700

↓

[0.65, -0.34, 0.88, 0.29]
```

These vectors become the input to the Transformer.

# Embedding Matrix

The embedding layer is essentially a **trainable lookup table**.

Suppose:

* Vocabulary Size = **50,000**
* Embedding Dimension = **768**

The embedding matrix has the shape:

```text
(50000 × 768)
```

Each row corresponds to one token in the vocabulary.

Example:

| Token ID | Embedding (4 dimensions shown) |
| -------: | ------------------------------ |
|        0 | [0.12, 0.31, -0.42, 0.91]      |
|        1 | [0.76, -0.24, 0.13, 0.55]      |
|        2 | [-0.61, 0.18, 0.77, -0.33]     |
|      ... | ...                            |
|      700 | [0.65, -0.34, 0.88, 0.29]      |

When the input ID is **700**, the model simply retrieves the 700th row.

---

# How Embeddings are Learned

Initially, embedding vectors are initialized randomly.

Example:

```text
Dog

↓

[0.31, -0.82, 0.15]
```

During training:

* The model predicts the next token.
* It computes the loss.
* Backpropagation updates **all model parameters**, including the embedding matrix.

After training:

```text
Dog

↓

[0.74, -0.21, 0.64]
```

Words with similar meanings gradually move closer together in the embedding space.

# Why Dense Vectors?

Without embeddings:

```text
Dog

↓

10
```

The model sees only a single integer.

With embeddings:

```text
Dog

↓

[0.42,0.13,-0.84,...]
```

The model now has a rich numerical representation that can capture semantic relationships.

# Types of Embeddings in Transformers

## 1. Token Embeddings

Represent the meaning of each token.

```text
AI

↓

Vector
```

---

## 2. Positional Embeddings

Represent the token's position in the sequence.

Example:

```text
I love AI
```

Position vectors:

```text
Position 0

↓

Vector

Position 1

↓

Vector

Position 2

↓

Vector
```

---

## 3. Segment Embeddings (BERT)

Used when two sentences are provided.

Example:

```text
Sentence A

Sentence B
```

Each token is marked as belonging to segment A or B.

> Decoder-only models like GPT generally do **not** use segment embeddings.

---

# Final Input to Transformer

The actual input is the sum of different embeddings.

For models such as **BERT**:

[
\text{Input Embedding}
======================

\text{Token Embedding}
+
\text{Position Embedding}
+
\text{Segment Embedding}
]

For decoder-only models such as **GPT**, the input is typically:

[
\text{Input Embedding}
======================

\text{Token Embedding}
+
\text{Position Embedding}
]

# Implementation in TensorFlow/Keras

## Basic Embedding Layer

```python
from tensorflow.keras.layers import Embedding
import tensorflow as tf

# Vocabulary size
vocab_size = 10000

# Embedding dimension
embedding_dim = 128

embedding_layer = Embedding(
    input_dim=vocab_size,
    output_dim=embedding_dim
)

# Example token IDs
x = tf.constant([[15, 89, 700]])

output = embedding_layer(x)

print(output.shape)
```

### Output

```text
(1, 3, 128)
```

Explanation:

* Batch Size = **1**
* Sequence Length = **3**
* Embedding Dimension = **128**

---

# Implementation in PyTorch

```python
import torch
import torch.nn as nn

vocab_size = 10000
embedding_dim = 128

embedding = nn.Embedding(
    num_embeddings=vocab_size,
    embedding_dim=embedding_dim
)

x = torch.tensor([[15, 89, 700]])

output = embedding(x)

print(output.shape)
```

### Output

```text
torch.Size([1, 3, 128])
```

# Building a Small Embedding Model (Keras)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense

model = Sequential([
    Embedding(input_dim=10000, output_dim=128, input_length=50),
    GlobalAveragePooling1D(),
    Dense(64, activation='relu'),
    Dense(2, activation='softmax')
])

model.summary()
```

Here:

* `Embedding` converts token IDs into dense vectors.
* `GlobalAveragePooling1D` aggregates the sequence into a fixed-size vector.
* `Dense` layers perform classification.

# Interview Summary

> An **embedding** is a trainable dense vector representation of a token. After tokenization, token IDs are passed to an embedding layer, which acts as a lookup table and converts each token ID into a high-dimensional vector. These vectors capture semantic relationships, so words with similar meanings tend to have similar embeddings. In Transformers, token embeddings are combined with positional embeddings (and, in some architectures like BERT, segment embeddings) before being passed into the Transformer layers. During training, the embedding matrix is updated through backpropagation, allowing the model to learn meaningful representations automatically.
