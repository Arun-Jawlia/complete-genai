# Tokenization and `tiktoken` in Transformers

## Definition

Before a Transformer model can understand any text, it must convert the text into **numbers**. This process begins with **Tokenization**, where the input text is split into smaller units called **tokens**, and each token is mapped to a unique integer ID.

Since deep learning models work with **numerical data**, tokenization acts as the bridge between **human language** and **machine learning models**.

---

# Why Do We Need Tokenization?

Deep Learning models, including Transformers, cannot understand raw text.

They only process **numerical values**.

### Input Flow

```text
Text
   ↓
Tokenizer
   ↓
Tokens
   ↓
Token IDs (Integers)
   ↓
Embeddings
   ↓
Transformer Model
   ↓
Output
```

---

# Transformer Overview

A **Transformer** is a neural network architecture that transforms an input sequence into an output sequence.

### It can perform tasks like:

* Text → Text (Translation, Chatbots, Summarization)
* Text → Image
* Text → Audio
* Text → Video
* Text → Code

---

## Example

### Input

```text
Hey I am A
```

### Transformer Output (Next Token Prediction)

```text
Hey I am A → r → u → n
```

The model predicts one token at a time until the complete output is generated.

> **Interview Point:** LLMs generate text **token by token**, not the entire sentence at once.

---

# Why Convert Text into Numbers?

Machine Learning, Deep Learning, and NLP models perform mathematical computations.

A mathematical model can only process **numeric input**.

### Human Perspective

```text
Hello
```

### Machine Perspective

```text
[15496]
```

The model understands only the numeric representation, not the word itself.

---

# Step 1: Tokenization (Encoding)

## What is a Tokenizer?

A **Tokenizer** is the first component of the Transformer pipeline.

Its responsibilities are:

* Split input text into tokens.
* Convert tokens into unique integer IDs.
* Prepare the data for embedding generation.

---

## Tokenization Pipeline

```text
Input Text
      ↓
Tokenizer
      ↓
Tokens
      ↓
Token IDs
      ↓
Embeddings
      ↓
Transformer
```

---

# Example of Tokenization

### Original Text

```text
Hello I am Arun
```

### Possible Tokens

```text
["Hello", "I", "am", "Arun"]
```

### Integer Mapping

| Token | Token ID |
| ----- | -------- |
| Hello | 101      |
| I     | 45       |
| am    | 89       |
| Arun  | 712      |

> The actual token IDs depend on the tokenizer used.

---

# Different Tokenization Strategies

Different Transformer models use different tokenization strategies.

There is **no single universal tokenizer**.

The strategy depends on factors such as:

* Model architecture
* Training data
* Vocabulary size
* Language support
* Efficiency requirements

---

## 1. Word-Level Tokenization

### Definition

Each complete word is treated as one token.

### Example

```text
I love AI
```

↓

```text
["I", "love", "AI"]
```

### Advantages

* Easy to understand.
* Fewer tokens for common words.

### Disadvantages

* Very large vocabulary required.
* Cannot handle unseen words efficiently.

---

## 2. Character-Level Tokenization

### Definition

Each individual character becomes one token.

### Example

```text
Hello
```

↓

```text
["H", "e", "l", "l", "o"]
```

### Advantages

* Small vocabulary.
* Handles unknown words well.

### Disadvantages

* Produces long token sequences.
* Slower training and inference.

---

## 3. Subword Tokenization (Most Common in Modern LLMs)

Modern LLMs such as GPT generally use **subword tokenization**, where common words may stay intact while rare words are split into meaningful pieces.

### Example

```text
unbelievable
```

↓

```text
["un", "believ", "able"]
```

### Advantages

* Smaller vocabulary than word-level tokenization.
* Better handling of rare and unseen words.
* Efficient for multiple languages.

> **Interview Point:** Most modern Transformer models use **subword tokenization**, not purely word-level or character-level tokenization.

---

# Vocabulary Size and Tokenization

A tokenizer maintains a **vocabulary**, which is the set of all tokens it recognizes.

The chosen tokenization strategy affects vocabulary size:

| Vocabulary Size  | Common Strategy                          |
| ---------------- | ---------------------------------------- |
| Large vocabulary | Word-level or larger subword units       |
| Small vocabulary | Character-level or smaller subword units |

In practice, modern LLMs balance vocabulary size and sequence length using **subword tokenization**.

---

# Token IDs

After tokenization, each token is assigned a unique integer ID.

Example:

```text
Hello → 101
I → 45
am → 89
Arun → 712
```

The Transformer processes these integer IDs instead of raw text.

---

# What Happens After Tokenization?

Token IDs themselves do not capture meaning.

The next step is **Embedding Generation**.

## Flow

```text
Text
   ↓
Tokenization
   ↓
Token IDs
   ↓
Embeddings
   ↓
Transformer
```

Embeddings convert token IDs into dense vectors that capture the **semantic meaning** of words and their relationships.

# Tokenization in Transformers

## What is Tokenization?

**Tokenization** is the process of converting raw text into **tokens** (smaller units of text) that a Transformer model can understand.

Since neural networks cannot process text directly, the first step is to break the text into tokens and map each token to a unique numerical ID.

---

# Why Do We Need Tokenization?

Computers understand **numbers**, not words.

Example:

```text
I love Artificial Intelligence
```

A Transformer cannot process this sentence directly.

It must first convert it into numbers.

```text
Text
    ↓
Tokens
    ↓
Token IDs
    ↓
Embeddings
```

---

# Transformer Pipeline

```text
User Input
      │
      ▼
Tokenization
      │
      ▼
Token IDs
      │
      ▼
Embedding Layer
      │
      ▼
Transformer Model
      │
      ▼
Prediction
```

---

# Example

Sentence

```text
I love machine learning
```

### Step 1: Tokenization

```text
["I", "love", "machine", "learning"]
```

### Step 2: Vocabulary Lookup

Suppose the vocabulary is

```text
I          → 15
love       → 89
machine    → 325
learning   → 451
```

### Step 3: Convert to IDs

```text
[15, 89, 325, 451]
```

These IDs are passed to the embedding layer.

---

# What is a Vocabulary?

A **vocabulary** is a dictionary containing all tokens known by the model.

Example:

| Token       |  ID |
| ----------- | --: |
| I           |  15 |
| love        |  89 |
| machine     | 325 |
| learning    | 451 |
| AI          | 700 |
| Transformer | 920 |

Modern LLMs typically have vocabularies ranging from **tens of thousands to a few hundred thousand** tokens.

---

# Types of Tokenization

There are three common approaches.

## 1. Word Tokenization

Each word becomes one token.

Example

```text
I love AI
```

↓

```text
["I", "love", "AI"]
```

### Advantages

* Easy to understand
* Simple implementation

### Disadvantages

* Very large vocabulary
* Cannot handle unseen words well

---

## 2. Character Tokenization

Each character becomes a token.

Example

```text
Chat
```

↓

```text
["C", "h", "a", "t"]
```

### Advantages

* Can represent any word
* Small vocabulary

### Disadvantages

* Much longer sequences
* Slower training and inference

---

## 3. Subword Tokenization (Used in Modern LLMs)

Words are split into meaningful pieces.

Example

```text
unhappiness
```

↓

```text
["un", "happi", "ness"]
```

Another example

```text
playing
```

↓

```text
["play", "ing"]
```

This balances vocabulary size and the ability to handle unknown words.

---

# Why Do Transformers Use Subword Tokenization?

Consider the word:

```text
electroencephalography
```

A word-based tokenizer might never have seen it.

Instead, it can split it into familiar pieces, for example:

```text
electro
encephalo
graphy
```

Even if the full word is unknown, the model can still process it using known subwords.

---

# Popular Tokenization Algorithms

## 1. Byte Pair Encoding (BPE)

Used by many GPT-family models.

### Idea

Start with individual characters and repeatedly merge the most frequent adjacent pairs.

Example

```text
l o w
```

↓

```text
lo w
```

↓

```text
low
```

Frequent patterns become single tokens over time.

---

## 2. WordPiece

Used in BERT.

Example

```text
playing
```

↓

```text
play ##ing
```

The `##` prefix indicates that `ing` continues a previous subword.

---

## 3. SentencePiece

Used by models such as T5 and LLaMA.

Unlike BPE and WordPiece, it treats the input as a raw stream of text and does not require pre-tokenization based on spaces.

Example

```text
I love AI
```

↓

```text
▁I ▁love ▁AI
```

The special symbol `▁` marks the beginning of a new word.

---

# Special Tokens

Transformers use reserved tokens with specific purposes.

| Token    | Purpose                                                       |
| -------- | ------------------------------------------------------------- |
| `<PAD>`  | Pads sequences to equal length                                |
| `<UNK>`  | Represents unknown tokens (less common in subword tokenizers) |
| `<CLS>`  | Classification token (BERT)                                   |
| `<SEP>`  | Separates sentences (BERT)                                    |
| `<MASK>` | Masked token for masked language modeling (BERT)              |
| `<BOS>`  | Beginning of sequence                                         |
| `<EOS>`  | End of sequence                                               |

Example

```text
<BOS> I love AI <EOS>
```

# What Happens After Tokenization?

```text
Sentence

↓

Tokenizer

↓

Token IDs

↓

Embedding Layer

↓

Position Information

↓

Transformer

↓

Prediction
```

Tokenization itself **does not assign meaning**; it simply converts text into IDs. The embedding layer learns vector representations that capture semantic relationships.

# Python Example (Hugging Face)

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love Transformers."

tokens = tokenizer.tokenize(text)
print(tokens)

ids = tokenizer.convert_tokens_to_ids(tokens)
print(ids)
```

Possible output:

```text
['i', 'love', 'transformers', '.']

[1045, 2293, 19081, 1012]
```

* Learns reusable prefixes, suffixes, and word roots.
* Works well across multiple languages.

# Interview Summary

> **Tokenization** is the process of converting raw text into tokens and then mapping those tokens to numerical IDs that a Transformer can process. Modern language models primarily use **subword tokenization** (such as BPE, WordPiece, or SentencePiece) because it balances vocabulary size with the ability to represent rare and unseen words. After tokenization, the token IDs are passed to an embedding layer, which converts them into dense vectors before they are processed by the Transformer. Tokenization is the first step in every LLM inference and training pipeline.
