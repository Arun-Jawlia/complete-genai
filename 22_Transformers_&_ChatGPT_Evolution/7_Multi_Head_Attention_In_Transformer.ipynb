# Multi-Head Attention in Transformers

## Definition

**Multi-Head Attention (MHA)** is an extension of **Self-Attention** where the Transformer performs **multiple self-attention operations in parallel**. Each attention head learns to focus on a different type of relationship or feature in the input sequence.

> **Definition:**
> **Multi-Head Attention** allows the Transformer to look at the same sentence from multiple perspectives simultaneously, helping it capture richer contextual information.

---

# Quick Recap: Self-Attention

In **Self-Attention**, each token creates:

* Query (Q)
* Key (K)
* Value (V)

The model computes attention **once**.

Example:

```text
Sentence

↓

Self-Attention

↓

Context-Aware Output
```

This works well, but there is one limitation.

---

# Why Isn't One Attention Head Enough?

Consider the sentence:

```text
The boy who won the race is smiling because he is happy.
```

When processing **"he"**, the model may need to learn different types of information:

* Who does **"he"** refer to?
* What is the action?
* What is the grammatical structure?
* What is the overall meaning?

A **single attention head** may struggle to capture all these relationships at the same time.

---

# Real-World Analogy

Imagine watching a cricket match.

One commentator focuses on:

* Batting

Another focuses on:

* Bowling

Another focuses on:

* Field placements

Another focuses on:

* Match strategy

All commentators watch the **same match**, but each provides a different perspective.

Finally, all perspectives are combined to give a complete understanding.

**Multi-Head Attention works the same way.**

---

# Why "Multi-Head"?

Instead of using one Self-Attention mechanism:

```text
Sentence

↓

Self-Attention

↓

Output
```

The Transformer creates multiple independent attention heads.

Example:

```text
Sentence

↓

Head 1
Head 2
Head 3
Head 4

↓

Concatenate

↓

Linear Layer

↓

Final Output
```

Each head learns something different.

---

# Example

Sentence:

```text
The cat sat on the mat because it was tired.
```

Possible focus of different heads:

| Head   | Learns              |
| ------ | ------------------- |
| Head 1 | Grammar             |
| Head 2 | Subject ("cat")     |
| Head 3 | Verb relationships  |
| Head 4 | Context and meaning |

Together, these heads build a richer representation than a single attention mechanism.

---

# Step-by-Step Working

Suppose the input is:

```text
I love Deep Learning
```

After:

* Tokenization
* Embeddings
* Positional Encoding

the input reaches the Multi-Head Attention layer.

---

## Step 1: Input Embeddings

```text
I

↓

Embedding
```

```text
love

↓

Embedding
```

```text
Deep

↓

Embedding
```

```text
Learning

↓

Embedding
```

---

## Step 2: Create Multiple Q, K, V Sets

In Self-Attention:

```text
One Q
One K
One V
```

In Multi-Head Attention:

```text
Head 1

Q₁
K₁
V₁

Head 2

Q₂
K₂
V₂

Head 3

Q₃
K₃
V₃

Head 4

Q₄
K₄
V₄
```

Each head has **its own learnable weight matrices**, so they learn different representations.

---

## Step 3: Perform Self-Attention in Each Head

Every head independently computes:

```text
Attention(Q,K,V)
```

So if there are 8 heads:

```text
Head 1 → Attention

Head 2 → Attention

Head 3 → Attention

...

Head 8 → Attention
```

All computations happen **in parallel**.

---

## Step 4: Concatenate Outputs

Suppose each head produces:

```text
Head 1

[....]

Head 2

[....]

Head 3

[....]

Head 4

[....]
```

These outputs are joined together.

```text
Head1

↓

Head2

↓

Head3

↓

Head4

↓

Concatenation
```

This creates one larger vector.

---

## Step 5: Final Linear Layer

The concatenated vector passes through a linear (fully connected) layer.

```text
Concatenated Output

↓

Linear Layer

↓

Final Output
```

This combines information from all heads into a single context-aware representation.

---

# Visual Flow

```text
Input Embeddings
        ↓

Generate Q, K, V

        ↓

 ┌──────────────┐
 │ Head 1       │
 │ SelfAttention│
 └──────────────┘

 ┌──────────────┐
 │ Head 2       │
 │ SelfAttention│
 └──────────────┘

 ┌──────────────┐
 │ Head 3       │
 │ SelfAttention│
 └──────────────┘

 ┌──────────────┐
 │ Head 4       │
 │ SelfAttention│
 └──────────────┘

        ↓

Concatenate Outputs

        ↓

Linear Layer

        ↓

Final Representation
```

---

# Mathematical Formula

For each attention head:

```text
head_i = Attention(Q_i, K_i, V_i)
```

After computing all heads:

```text
MultiHead(Q,K,V)

=

Concat(head₁, head₂, ..., headₕ)

×

Wᴼ
```

Where:

* **h** = Number of attention heads
* **Wᴼ** = Output projection matrix

Each head learns different patterns within its assigned subspace.

# Multi-Head Attention vs Self-Attention

| Self-Attention          | Multi-Head Attention             |
| ----------------------- | -------------------------------- |
| One attention mechanism | Multiple attention mechanisms    |
| One representation      | Multiple learned representations |
| Limited perspective     | Multiple perspectives            |
| Simpler                 | More expressive and powerful     |

---

# Advantages

* Learns multiple types of relationships simultaneously.
* Captures richer contextual information.
* Improves language understanding.
* Handles complex sentence structures.
* Supports parallel computation.

---

# Disadvantages

* More parameters than single-head attention.
* Higher computational and memory requirements.
* Increased training time.

# Multi-Head Attention in Transformers

## What is Multi-Head Attention?

**Multi-Head Attention (MHA)** is an extension of the **Self-Attention** mechanism where the Transformer computes **multiple self-attention operations in parallel** instead of just one.

Each attention head learns **different relationships** between words in a sentence.

Instead of asking **one question** about the sentence, the model asks **many questions simultaneously**, with each head focusing on a different aspect of the input.

---

# Why Do We Need Multi-Head Attention?

Suppose the sentence is:

```text
The animal didn't cross the road because it was tired.
```

When processing the word **"it"**, one attention mechanism may focus on:

* **animal** (to resolve the pronoun)

But language contains many different relationships.

For example:

* Grammar
* Subject–verb agreement
* Long-distance dependencies
* Semantic similarity
* Context

A single attention mechanism may not capture all of them effectively.

**Multi-Head Attention allows different heads to specialize in different types of relationships.**

---

# Real-Life Analogy

Imagine a cricket match.

Different commentators focus on different things:

* Commentator 1 → Batting
* Commentator 2 → Bowling
* Commentator 3 → Field placement
* Commentator 4 → Match statistics

All of them observe the **same match**, but each provides different insights.

Similarly,

* Head 1 may learn grammar.
* Head 2 may learn semantic meaning.
* Head 3 may learn coreference.
* Head 4 may learn word order.

The Transformer combines all these perspectives.

---

# Transformer Pipeline

```text
Input Sentence
      │
      ▼
Tokenization
      │
      ▼
Embeddings
      │
      ▼
Positional Encoding
      │
      ▼
Multi-Head Attention
      │
      ▼
Feed Forward Network
      │
      ▼
Output
```

---

# Working of Multi-Head Attention

Suppose the sentence is:

```text
I love Deep Learning
```

---

## Step 1: Token Embeddings

After embedding:

```text
I          → Vector

love       → Vector

Deep       → Vector

Learning   → Vector
```

These vectors become the input to Multi-Head Attention.

---

## Step 2: Create Query, Key, and Value

Each attention head has **its own trainable weight matrices**.

For **Head 1**:

```text
Embedding

↓

WQ₁ → Query₁

WK₁ → Key₁

WV₁ → Value₁
```

For **Head 2**:

```text
Embedding

↓

WQ₂ → Query₂

WK₂ → Key₂

WV₂ → Value₂
```

The same input embeddings are transformed differently by each head.

If there are **8 heads**, there are **8 independent sets of Q, K, and V matrices**.

---

# Step 3: Self-Attention in Every Head

Each head performs standard self-attention independently.

For Head 1:

[
\text{Head}_1=
\text{Attention}(Q_1,K_1,V_1)
]

For Head 2:

[
\text{Head}_2=
\text{Attention}(Q_2,K_2,V_2)
]

and so on.

Each head computes

[
\text{Softmax}
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)
V
]

independently.

---

# Visual Representation

```text
               Input Embeddings

                     │

    ---------------------------------------

    │         │         │         │

  Head1     Head2     Head3     Head4

    │         │         │         │

Self Att.  Self Att. Self Att. Self Att.

    │         │         │         │

    ---------------------------------------

                     │

              Concatenate

                     │

            Linear Projection

                     │

                  Output
```

---

# Example

Sentence:

```text
The cat sat on the mat
```

Different heads may learn different patterns.

### Head 1

Focuses on

```text
cat ↔ sat
```

(Subject–verb relationship)

---

### Head 2

Focuses on

```text
sat ↔ mat
```

(Location relationship)

---

### Head 3

Focuses on

```text
the ↔ cat

the ↔ mat
```

(Article–noun relationship)

---

### Head 4

Focuses on

Long-range dependencies.

Every head captures a different aspect of the sentence.

---

# Step 4: Concatenate Outputs

Suppose there are **8 heads**.

Each head outputs a vector.

```text
Head1 Output

↓

64 Dimensions
```

```text
Head2 Output

↓

64 Dimensions
```

...

After concatenation:

```text
64 × 8

=

512 Dimensions
```

The outputs are joined together.

---

# Step 5: Final Linear Layer

The concatenated output is passed through a trainable linear projection matrix (W_O).

[
\text{MultiHead}(Q,K,V)=
\text{Concat}(head_1,\dots,head_h)W_O
]

This mixes information from all attention heads into one output representation.

---

# Complete Formula

For each head:

[
head_i=
\text{Attention}(Q_i,K_i,V_i)
]

Final output:

[
\boxed{
\text{MultiHead}(Q,K,V)=
\text{Concat}(head_1,\dots,head_h)W_O
}
]

---

# Complete Workflow

```text
Input Embeddings

      │

      ▼

Generate Q, K, V

      │

      ▼

-------------------------------------

|    |    |    |    |    |    |    |

H1   H2   H3   H4   H5   H6   H7   H8

|    |    |    |    |    |    |    |

Self Attention in Every Head

-------------------------------------

      │

      ▼

Concatenate

      │

      ▼

Linear Layer

      │

      ▼

Output
```

# Why Multiple Heads?

A single attention mechanism has limited capacity.

Multiple heads allow the model to learn:

* Grammar
* Syntax
* Semantics
* Long-distance relationships
* Pronoun resolution
* Negation
* Subject–verb agreement
* Context

all at the same time.

---

# Example

Sentence:

```text
The bank approved the loan.
```

One head may focus on

```text
bank ↔ loan
```

(financial meaning)

Another head may focus on

```text
approved ↔ loan
```

(verb–object relationship)

Another head may focus on

```text
The ↔ bank
```

(determiner–noun relationship)

---

# Tensor Dimensions

Suppose

* Batch Size = **32**
* Sequence Length = **20**
* Embedding Size = **512**
* Heads = **8**

Then:

Each head receives

```text
512 / 8

=

64 Dimensions
```

Each head works on **64-dimensional** projections.

After concatenation:

```text
64 × 8

=

512
```

The original embedding size is restored.

> **Note:** In practice, the input embedding is projected into separate Q, K, and V matrices for each head. Each head operates on a lower-dimensional representation (e.g., 64 dimensions), not by simply slicing the original embedding.

---

# PyTorch Implementation

```python
import torch
import torch.nn as nn

attention = nn.MultiheadAttention(
    embed_dim=512,
    num_heads=8,
    batch_first=True
)

x = torch.randn(32, 20, 512)

output, weights = attention(x, x, x)

print(output.shape)
print(weights.shape)
```

Output

```text
torch.Size([32, 20, 512])

torch.Size([32, 20, 20])
```

---

# TensorFlow/Keras Implementation

```python
import tensorflow as tf
from tensorflow.keras.layers import MultiHeadAttention

attention = MultiHeadAttention(
    num_heads=8,
    key_dim=64
)

x = tf.random.normal((32, 20, 512))

output = attention(
    query=x,
    key=x,
    value=x
)

print(output.shape)
```

Output

```text
(32, 20, 512)
```

# Self-Attention vs Multi-Head Attention

| Feature                         | Self-Attention     | Multi-Head Attention                   |
| ------------------------------- | ------------------ | -------------------------------------- |
| Number of Attention Mechanisms  | 1                  | Multiple (e.g., 8, 12, 16)             |
| Context Learned                 | One representation | Multiple complementary representations |
| Expressiveness                  | Lower              | Higher                                 |
| Captures Multiple Relationships | Limited            | Yes                                    |
| Performance                     | Good               | Better in practice                     |

---

# Advantages

* Learns multiple types of relationships simultaneously.
* Captures richer contextual information.
* Improves language understanding.
* Helps model long-range dependencies.
* One of the key innovations that made Transformers highly effective.

---

# Limitations

* More computationally expensive than a single attention head.
* Memory usage grows with sequence length because each head computes an attention matrix.
* Standard attention has **quadratic complexity** with respect to sequence length.

---

# Interview Summary

> **Multi-Head Attention (MHA)** extends self-attention by running several independent attention mechanisms in parallel. Each head learns different relationships within the input sequence using its own Query, Key, and Value projection matrices. The outputs from all heads are concatenated and passed through a final linear projection. This enables the Transformer to simultaneously capture grammatical structure, semantic meaning, long-range dependencies, and contextual relationships, making it far more expressive than a single self-attention mechanism.
