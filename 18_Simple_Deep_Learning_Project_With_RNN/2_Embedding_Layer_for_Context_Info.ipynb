# Introduction

Natural language consists of words, but neural networks only understand numbers.

A simple mapping like:

```
I      → 1
Love   → 2
AI     → 3
```

does **not** tell the model anything about the meaning of the words.

The number **2 ("Love")** is not twice as meaningful as **1 ("I")**.

This is why we use an **Embedding Layer**.

---

# Definition

## What is an Embedding Layer?

An **Embedding Layer** is the first trainable layer in most NLP models that converts integer-encoded words into **dense vectors** containing semantic meaning.

Instead of representing a word with a single number, it represents it using multiple learned numerical features.

Example:

```
Love → [0.82, -0.34, 0.56, 0.91]
```

These numbers are **learned during training**, not manually designed.

They represent:

- Meaning
- Context
- Usage of the word

---

# Why is Embedding Critical for RNN?

Without embeddings:

- RNN only receives meaningless integer IDs.
- It cannot understand relationships between words.
- Context learning becomes weak.
- Semantic information is lost.

Example:

```
Good → 15
Excellent → 92
Amazing → 108
```

The numbers have **no relationship** with each other.

---

With embeddings:

- Similar words have similar vectors.
- Different words have distant vectors.
- RNN learns meaningful language patterns.
- Training becomes more efficient.

---

## Why This Matters

The quality of an RNN largely depends on the quality of the word representations it receives.

Better input representations → Better memory → Better predictions.

---

# Word2Vec Intuition

Embedding vectors are inspired by ideas such as **Word2Vec**.

The basic principle is:

- Words with similar meanings are located close together in vector space.
- Words with different meanings are located far apart.

Example:

```
Excellent
      ●

Amazing
      ●

Good
      ●



Car
                     ●

Banana
                              ●
```

Nearby vectors indicate similar meanings.

# Embedding: The Bridge Between Language and RNN Memory

Think of the pipeline like this:

```
Human Language
      ↓
Embedding Layer
      ↓
Meaningful Numerical Vectors
      ↓
Simple RNN Memory
      ↓
Prediction
```

**Embedding defines what a word means.**

**RNN defines how that meaning evolves over time.**

# How Embedding and RNN Work Together

At each time step **t**:

The RNN receives:

- Current word embedding
- Previous hidden state

Then it updates its memory.

```
Current Word
      ↓
Embedding Vector (xₜ)
      ↓
Simple RNN
      ↑
Previous Hidden State (hₜ₋₁)

      ↓

Updated Hidden State (hₜ)
```

Mathematically:

```
Input at time t

Embedding(xₜ)

+

Previous Hidden State (hₜ₋₁)

↓

Simple RNN

↓

New Hidden State (hₜ)
```

The hidden state stores information learned from previous words.

# Capturing Context

Consider the word:

```
Bank
```

Sentence 1

```
I am having an ICICI bank nearby.
```

Sentence 2

```
I have a river bank nearby.
```

Although the word **bank** is the same, its meaning changes depending on the surrounding words.

Without embeddings:

- The model struggles to distinguish the meanings.

With embeddings + RNN:

- The embedding provides the word representation.
- The RNN uses surrounding words to understand the intended meaning.

Here,

**Embedding defines the meaning of the word.**

**RNN tracks how the meaning changes across the sentence.**

# Is the Embedding Layer Trainable?

Yes.

In a Simple RNN, the Embedding Layer is usually **trainable**.

Training process:

Step 1

Random initialization

```
Good → [0.12, -0.45, 0.81]

Bad → [-0.92, 0.11, 0.25]
```

Initially these vectors are random.

---

Step 2

Training

During backpropagation, the vectors are continuously updated.

The model gradually learns semantic relationships.

---

Step 3

After Training

```
Good
Excellent
Amazing
```

become close together.

```
Bad
Poor
Worst
```

also become close together.

The embedding now captures semantic similarity.

---

# Task-Specific Learning

Embedding vectors are learned according to the task.

Example:

Sentiment Analysis

Initially

```
Good → Random Vector

Bad → Random Vector
```

After training

```
Good
Excellent
Amazing
```

represent positive sentiment.

```
Bad
Poor
Worst
```

represent negative sentiment.

Meanwhile,

The RNN captures the emotional flow throughout the sentence.

# Choosing Embedding Size

General guideline:

Small dataset

→ Smaller embedding dimension

Large dataset

→ Larger embedding dimension

Important Note:

**Bigger embeddings are not always better.**

Very large embeddings may:

- Increase training time
- Increase memory usage
- Cause overfitting

Choose an embedding size appropriate for the dataset and task.

# Step-by-Step Working of an Embedding Layer

## Step 1: Tokenization

Sentence

```
I love AI
```

becomes

```
["I", "love", "AI"]
```

---

## Step 2: Integer Encoding

```
I → 1

Love → 7

AI → 12
```

These numbers are only indices.

They contain **no semantic meaning**.

---

## Step 3: Embedding Lookup

The embedding layer maintains an **Embedding Matrix**.

Example:

| Word | Embedding Vector |
|------|------------------|
| I | [0.15, -0.21, 0.48] |
| Love | [0.82, 0.35, -0.18] |
| AI | [0.43, 0.71, 0.24] |

Each word index retrieves its corresponding vector.

---

## Step 4: Sequence of Embeddings

Instead of integers:

```
[1, 7, 12]
```

The RNN receives:

```
[
[0.15, -0.21, 0.48],
[0.82, 0.35, -0.18],
[0.43, 0.71, 0.24]
]
```

Now every word carries semantic information.

The RNN can effectively learn sequential patterns.

Embedding Layer

# Advantages

- Learns semantic meaning automatically.
- Makes RNN training more efficient.
- Captures relationships between words.
- Reduces sparse word representations.
- Improves performance on NLP tasks.

---

# Limitations

- Traditional embeddings assign one vector per word.
- The same word gets the same embedding regardless of context.
- Cannot fully resolve ambiguous words (e.g., "bank") on their own.
- Contextual models like BERT generate different embeddings based on surrounding words.

# 2. Interview Questions & Answers

## Beginner Questions

### Q1. What is an Embedding Layer?

**Answer:** An Embedding Layer converts integer-encoded words into dense vectors that capture semantic meaning. It is typically the first trainable layer in an NLP model.

### Q2. Why can't we feed integer-encoded words directly to an RNN?

**Answer:** Integer IDs are arbitrary labels and do not represent relationships or meanings between words. The RNN would treat them as meaningless numbers.

### Q3. Are embedding vectors fixed?

**Answer:** In most Simple RNN models, they are trainable. They start with random values and are updated using backpropagation.

### Q4. What is semantic similarity?

**Answer:** It means words with similar meanings (e.g., *good*, *excellent*, *amazing*) have similar embedding vectors and are close in vector space.

### Q5. What does the hidden state in an RNN do?

**Answer:** It stores information from previous time steps, allowing the model to remember context while processing a sequence.

---

## Intermediate Questions

### Q1. Why are embeddings considered dense representations?

**Answer:** Unlike one-hot vectors, embeddings use compact continuous-valued vectors where each dimension captures learned features of the word.

### Q2. How do embeddings improve RNN performance?

**Answer:** They provide semantically meaningful inputs, enabling the RNN to learn patterns and dependencies more effectively.

### Q3. Why isn't a larger embedding dimension always better?

**Answer:** Larger embeddings increase computation, memory usage, and the risk of overfitting. The dimension should match the dataset size and task complexity.

### Q4. How are embedding vectors updated?

**Answer:** During training, gradients from the loss function are backpropagated through the network, updating the embedding matrix along with other trainable parameters.

### Q5. How is an Embedding Layer different from Word2Vec?

**Answer:** An Embedding Layer learns task-specific vectors during model training, while Word2Vec is a standalone algorithm that learns general-purpose word embeddings from large text corpora.

---

## Scenario-Based Questions

### Q1. Your sentiment analysis model performs poorly when using integer encoding. What would you do?

**Answer:** Add a trainable Embedding Layer so the model learns meaningful word representations instead of relying on arbitrary integer IDs.

### Q2. Two words (*good* and *excellent*) are far apart in the learned embedding space. What might this indicate?

**Answer:** The model may need more training, better data, or a more suitable embedding dimension to learn semantic relationships.

### Q3. You have a small dataset. Should you choose a 1024-dimensional embedding?

**Answer:** Usually no. A very large embedding dimension can overfit and waste computational resources.

### Q4. Why can the word "bank" still be ambiguous with traditional embeddings?

**Answer:** Traditional embeddings assign a single vector to each word, regardless of context. Contextual embeddings (e.g., BERT) solve this by generating different vectors based on surrounding words.

### Q5. What information does the RNN combine at each time step?

**Answer:** It combines the current word's embedding vector with the previous hidden state to produce an updated hidden state.
