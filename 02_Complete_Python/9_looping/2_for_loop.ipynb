# For Loop
---

A for loop is used to iterate over a sequence (iterable) such as:

- list
- tuple
- string
- set
- dictionary

### Basic Synatx

```
for variable in iterable:
    # code block
```

### Simple Example

````python
numbers = [1, 2, 3, 4]

for num in numbers:
    print(num)
````

## Looping Through Different Data Types

````python
# List
lst = [10, 20, 30]

for item in lst:
    print(item)
````

````python
# String

for ch in "Python":
    print(ch)
````

````python
# Tuple

t = (1, 2, 3)

for i in t:
    print(i)
````

````python
# Dictionary
d = {"a": 1, "b": 2}

for key in d:
    print(key, d[key])

print("--------------")
# Better way
for key, value in d.items():
    print(key, value)
````

## Using Range() with For loop  --> range(start, stop, step)
- start	Optional. An integer number specifying at which position to start. Default is 0
- stop	Required. An integer number specifying at which position to stop (not included).
- step	Optional. An integer number specifying the incrementation. Default is 1

````python
# Only Stop Value : Starts from 0 by default and stops before 5.
for i in range(5):
    print(i)

print('----------------')
# State and Stop Value: Create a sequence of number from 1 to 5 and print each item in the sequence
for i in range(1, 6):
    print(i)   # 1 to 5

print('----------------')
# Start, stop, and step value: Create a sequence of numbers from 1 to 10 but incremented by 2 instead of 1
for i in range(1, 10, 2):
    print(i)   # 1,3,5,7,9

print('----------------')
````

````python
# Reverse order using negative step

for i in range(10, 0, -2):
    print(i)

print('-------------------------')
````

## Loop Control Statements
- Being used for termination a loop in between when a conditoin is met or if you want to have some control over your iterating process or while iterating over the series of numbers
- There are 3 loop control statements: 
  1. break 
  2. continue 
  3. pass

### break--> terminate the loop

````python
#- break is our first loop control statement which will help you out to terminate the loop when a certain condition is met
for i in range(5):
    if i == 3:
        break
    print(i)

print('---------------')
for i in range(1,51):
    if i == 18:
        break
    print(i)
````

### continue --> Skip Iteration

````python
# continue: secound loop control statement which helps us in iterating over a series of numbers but helps us with skipping the values 
# in between once we meet a condition

# print the numbers from 0 to 5 but once 2 is occuring skip it
for i in range(5):
    if i == 2:
        continue
    print(i)

# print the numbers from 0 to 50 but only print even numbers and for odd number skip those values
print('-------------------')
for i in range(51):
    if i % 2 == 0:
        print(i)
    else:
        continue
````

### pass --> does nothing

````python
# pass: third control loop statement is being used for doing nothing.

for i in range(3):
    pass
````

````python
# Division of numbers where the denominator is 0
num1 = int(input("enter the numberator value: "))
num2 = int(input("enter the denominator value: "))

# print(num1/num2) # ZeroDivisionError: division by zero

if num2 == 0 :
    pass  # it will throw nothing ( no error)
else:
    print(num1/num2)
````

## Nested Loop
- Loop inside loop

````python
# Nested Loop can help us out with iterating over a series of numbers as with the proper combination of iterating numbers

for i in range(3):
    for j in range(2):
        print(f"i value is {i}, j value is {j}")
````

````python
# Print the sum of first n natural number

sum = 0
n_values = int(input("Enter the numbers: "))
for i in range(1, n_values + 1):
    sum += i

print(f"sum of first { n_values} natural number is {sum}")
````

## Loop with Index( enumerate )

````python
lst = ["a", "b", "c"]

for index, value in enumerate(lst):
    print(index, value)
````

## Loop Multiple Lists(zip)

````python
# Both list should be of same length
names = ["A", "B"]
marks = [90, 80]

for n, m in zip(names, marks):
    print(n, m)
````

## List Comprehension ( Advanced Loop )

````python
squares = [x*x for x in range(5)]
cubes = [x**3 for x in range(5)]

evens = [x for x in range(10) if x % 2 == 0]
odds = [x for x in range(10) if x % 2 == 1]

print(squares)
print(cubes)
print(evens)
print(odds)
````

## Time Complexity
- Loop runs n times → O(n)
- Nested loops → O(n²)

````python
# Check whether the number is prime or not 

n_value = int(input("Enter number: "))

is_prime = True

if n_value <= 1:
    is_prime = False
else:
    for i in range(2, int(n_value**0.5) + 1):
        if n_value % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime number")
else:
    print("Not a prime number")
````
