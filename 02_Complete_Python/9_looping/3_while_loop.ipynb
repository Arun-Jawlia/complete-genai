# While Loop
---

- Definition: A While loop is being used for iterating over a series of numbers when a condition is True
- It is a condition-controlled loop (not sequence-based like for).

### Syntax
```
while condition:
    # code runs
```

### How it Works
1. Check condition
2. If True → run code
3. Repeat
4. Stop when condition becomes False

````python
# By making the use of while loop we need to print the number from 0 to 10
i=0
while i <= 10:
    print(i)
    i+=1
````

### Real World Examples

````python
# Example: ATM PIN Validation
# Scenario: Keep asking for PIN until correct

correct_pin = "1234"

while True:
    pin = input("Enter PIN: ")

    if pin == correct_pin:
        print("Access Granted")
        break
    else:
        print("Wrong PIN, try again")

'''
Why while loop here?
We don’t know how many attempts the user will take
Loop runs until condition is satisfied

👉 This is exactly where while is preferred over for
'''
````

````python
# Example: Game Loop

lives = 3

while lives > 0:
    print("Playing game...")
    lives -= 1

print("Game Over")
````

## Loop Control Statements
- Being used for termination a loop in between when a conditoin is met or if you want to have some control over your iterating process or while iterating over the series of numbers
- There are 3 loop control statements: 
  1. break 
  2. continue 
  3. pass

### break --> terminate the loop when a certain condition is met

````python
# print the number from 0 to 50 but once number is 18 then terminate the loop or don't print 
i = 1

while i <= 5:
    if i == 3:
        break
    print(i)
    i += 1
````

### continue --> skip iteration

````python
i = 0

while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
````

### pass --> do nothing

````python
while True:
    pass #infinite loop
````

## else with while ( Interview)

````python
i = 1

while i <= 3:
    print(i)
    i += 1
else:
    print("Loop finished")
````

## # Common Mistake: Infinite Loop ❌

````python
i = 1

while i <= 5:
    print(i)
    # missing i += 1 ❌
````
