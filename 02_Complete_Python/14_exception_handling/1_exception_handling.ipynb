# `EXCEPTION HANDLING`
---

Exception handling is a mechanism used to handle runtime errors so that the program does not crash unexpectedly.

It allows a program to:

- detect errors
- respond gracefully
- continue execution when possible


`What is an Exception:` An exception is an event that occurs during program execution that disrupts the normal flow of the program. In other word, error while execution.

### Why Exception Handling is Needed

Without exception handling:

- applications crash
- user experience becomes poor
- resources may not close properly
- debugging becomes difficult
  
Exception handling helps:
  
- prevent crashes
- provide meaningful error messages
- maintain application stability

## `Basic Exception Handling`

###  `Try-Except`

Syntax
```
try:
    # risky code

except:
    # handling code
```

````python

try:
    num = 10 / 0

except:
    print("An error occurred") # there is no except raise or block the code - Program does not crash.
````

### Catching Specific Exceptions : Best practice: catch only expected exceptions

````python
try: 
    a = b
except  NameError:
    print("Name error")
````

````python
try:
    x = int("abc")

except ValueError:
    print("Value Error : Invalid number")
````

### Using `as` Keyword: Used to capture exception object.

````python
deno = int(input("Enter your number: ")) # if user select 0 then it will throw error
try: 
    result = 10 / deno # # ZeroDivisionError: division by zero
    print(result)
except ZeroDivisionError as exception:
    print(exception)
````

### `try-except-else` 

````python
n = int(input("Enter your number: "))
try:
    num = 10 / n

except ZeroDivisionError:
    print("Zero Division Error")
except ValueError:
    print("Value: Error")
else:
    print("Success and Result : ", num)
````

````python
try:
    num = int(input("Enter number: "))
    result = 10 / num

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Invalid input")

else:
    print("Result:", result)

finally:
    print("Program ended")
````

### `Raising Exceptions` : Python allows manually generating exceptions. using `raise` 

````python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
````

### `Custom Exceptions`: You can create your own exception classes.

````python
class InvalidAgeError(Exception):
    pass

age = -1

if age < 0:
    raise InvalidAgeError("Invalid age") # InvalidAgeError: Invalid age
````

### `File Handling with Exceptions`

````python
try:
    with open("data.txt", "r") as f:
        print(f.read())

except FileNotFoundError:
    print("File not found")
````

````python
## Exception ---> FileNotFoundError
try:
    file= open("example.txt", 'r')
    content = file.read()
    print(content)
except FileNotFoundError as ex:
    print(ex)
finally :
    if 'f' in locals () and not file.closed():
        f.close()
        print('File Closed')
````

### `Multiple Exceptions in One Block`

````python
try:
    x = int("abc")

except (ValueError, TypeError):
    print("Handled")
````

### `Nested Exception Handling`

````python
try:
    try:
        print(10 / 0)

    except ZeroDivisionError:
        print("Inner exception")

except:
    print("Outer exception")
````
