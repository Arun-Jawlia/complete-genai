### Types of Operators
1. Arithemtic Operator 
   1.  Addititon ( + )
   2.  Substraction ( - )

   4.  Division ( / )
   5.  Modulus ( % )

   7.  Expoent ( ** )
   
2. Comparison Operatos
   1. Greater than  ( > )
   2. less than ( < )
   3. Greater than equal to ( >= )
   4. Less than equal to  ( <= )
   5. Equality Check ( == )
   6. Not Equal to ( != ) 

3. Logical Operators
   1. and --> whether if both conditions are true or not
   2. or --> whether if any of the operation is true or not
   3. not --> negation or to inverse the result of the given operations

```python
# Simple Calculator

first = int(input("Enter your first value: "))
second = int(input("Enter your second value: "))
operator = input("Enter what do you want to do (+,-,*,/,%,//,**) : ")

if operator == '+':
    print(f"The Addition of {first} and {second} = {first + second}")

elif operator == '-':
    print(f"The Subtraction of {first} and {second} = {first - second}")

elif operator == '*':
    print(f"The Multiplication of {first} and {second} = {first * second}")

elif operator == '/':
    if second != 0:
        print(f"The Division of {first} and {second} = {first / second}")
    else:
        print("Division by zero is not allowed")

elif operator == '%':
    if second != 0:
        print(f"The Modulus of {first} and {second} = {first % second}")
    else:
        print("Modulo by zero is not allowed")

elif operator == '//':
    if second != 0:
        print(f"The Floor Division of {first} and {second} = {first // second}")
    else:
        print("Floor division by zero is not allowed")

elif operator == '**':
    print(f"The Power of {first} and {second} = {first ** second}")

else:
    print("Invalid operator")
```
