# SETS in Data Structure

### Sets are going to be our third data structure which can store values  inside within the curly braces format {} and are known for storing the values which are Non Duplicates and uniques

or 

### A set is a built-in Python data structure that stores unique elements in an unordered collection.

1. No duplicates
2. No order
3. Fast operations
   
---

## Why Do We Use Sets?

Sets exist for efficiency + uniqueness.

✅ Use cases:
- Remove duplicates
- Fast lookup (O(1))
- Mathematical operations (union, intersection)
- lst = [1, 2, 2, 3]
- unique = set(lst)

print(unique)   # {1, 2, 3}

👉 This is much faster than manual duplicate removal.

---
## Creating Sets

```python
# Empty set
s1 = set()

# With values
s2 = {1, 2, 3}

# From list
s3 = set([1, 2, 3])

s = {}   # ❌ This is a dictionary , not Set
```

## Methods in Set

1. Adding Elements
    - add()
    - update()

2. Removing Elements
    - remove()
    - discard()
    - pop()
    - clear()


```python
# Adding Elements 

# add()--> inserting values inside a set from the last

set_values = {1,2,3,4,5}

set_values.add(1000)
print(set_values)
```

```python
# set only consiste of unique values inside it and non duplicate values

set_values.add(1) # try to add 1 , it is already exists
print(set_values)
```

```python
# Update()---> add multiple values inside the set

set_empty = set()

set_empty.update([1,2,3,4])
print(set_empty)
```

## Removing Elements

```python
remove_sets = set()
remove_sets.update([100,200,300,500,800])
print(remove_sets)

# remove_sets.remove(1000) # error if not found  error = KeyError: 1000
print(remove_sets)
remove_sets.discard(100) # safe ( no error )
print(remove_sets)
remove_sets.pop() # removes random element
print(remove_sets)
remove_sets.clear() # empty the set
print(remove_sets)
```

## Core Sets Operations 


```python
a = {1,2,3}
b = {3,4,5}


# Union (All Elements)
print(a | b)  # {1,2,3,4,5}

# Intersection (Common)
print(a & b)  # {3}

# Difference
print(a - b)   # {1, 2}
print(b - a)   # {4, 5}

# Symmetric Difference
print(a ^ b)  # {1, 2, 4, 5}
```

## Relationships Methods

```python
a = {1, 2}
b = {1, 2, 3}

print(a.issubset(b))   # True
print(b.issuperset(a)) # True
```

```python
# Memberships Test using in

s = {1, 2, 3}

print(2 in s)   # True
```

## Set Comprehension

```python
## Set Comprehensions
s = {x*x for x in range(5)}

print(s) # {0, 1, 4, 9, 16}
```
