# `OS LIBRARY `
---

- Able to interact with our Operating System and work with Multiple Directories.
- The os module in Python provides an interface to interact with the operating system—files, directories, environment variables, processes, and paths. It abstracts platform differences (Windows/Linux/macOS) so your code stays portable.

### Why os Module?
- Use it when you need to:   
    1. Navigate directories
    2. Create/delete files or folders
    3. Work with file paths
    4. Access environment variables
    5. Execute system-level operations

### Basic Usage: `import os`

---
## Working with Directories

### 1. Current Directory: `getcwd()` --> Get Current Working Directory

```python
import os
```

```python
current_dir = os.getcwd()
print("Current Directory is :", current_dir)
```

### 2. Make Directory: `mkdir()` --> Creating The Directory / Folder

```python
new_path = "New_Folder"
os.mkdir(new_path)
print(f'New Directory is : {new_path} is created successfully')
```

```python
nested_folder  = 'new_folder/new_folder_1/new_folder_11'
os.makedirs(nested_folder) # Create nested folders
print(nested_folder)
```

### 3. Listing Directory: `listdir()` ---> return all files and folders

```python
all_dir = os.listdir('.')
print(all_dir) # All files and folder
```

### 4. Removing Directories: `rmdir()` and `removedirs()`

```python
# FileNotFoundError: [WinError 2] The system cannot find the file specified: 'test_folder' 
# os.rmdir("test_folder")  # Remove empty folder

nested_folder  = 'new_folder/new_folder_1/new_folder_11'
os.removedirs(nested_folder)  # Remove nested folders
```

### 5. Working with Paths: `os.path`

```python
path = 'C:/Users/arunk/OneDrive/Desktop/Learning/complete-genai/README.md'
```

Path Joins: `os.path.join(directory_name, file_name)`

```python
dir_name = 'FOLDER_DATA'
file_name = 'FILE_DATA'
full_path = os.path.join(dir_name, file_name)
print("Full path: ", full_path)
```

```python
# Join multiple path with the current path

current_complete_path = os.path.join(os.getcwd(),dir_name, file_name )
print("current complete path", current_complete_path)
```

Path exist: `os.path.exists()`

```python
given_path = "Desktop\Learning\complete-genai\README.md" #Not Exists
# given_path = "C:/Users/arunk/OneDrive/Desktop/Learning/complete-genai/README.md" # Exists

if os.path.exists(given_path):
    print("Path exists")
else:
    print("Not Exists")
```

Chekc File type: `os.path.isfile(file_name)`

```python
given_path = "C:/Users/arunk/OneDrive/Desktop/Learning/complete-genai/README.md" 
if os.path.isfile(given_path):
    print("Current path is file")
else:
    print("Current Path is not a file")
```

Check Folder/Directory Type: `os.path.isdir(directory_name)`

```python
given_path = "C:/Users/arunk/OneDrive/Desktop/Learning/complete-genai" 

if os.path.isdir(given_path):
    print("The current path is belongs to directory/folder")
else:
    print("Current Path is not a folder/directory")
```

```python
given_path = "C:/Users/arunk/OneDrive/Desktop/Learning/complete-genai" 

if os.path.isfile(given_path):
    print("Current path is file")
elif os.path.isdir(given_path):
    print("The current path is belongs to directory/folder")
else:
    print("Current Path is neither file nor a folder")
```

Get File Size: `os.path.getsize(path)`

```python
given_path = "C:/Users/arunk/OneDrive/Desktop/Learning/complete-genai" 
print(os.path.getsize(given_path))     # Size in bytes
```

Get File Absolute Path: `os.path.abspath(path)`

```python
# The Complete path <---> Absolute path <---> os.path.abspath()
print(os.path.abspath("data.txt"))
```

Environment Variables: `os.environ.get("Path")`

### 6. Cross Platform Handling: `os.name`

## `REAL WORLD EXAMPLES`

#### Example 1: List only .txt files

```python
for file in os.listdir():
    if file.endswith(".ipynb"):
        print(file)
```

#### Example 2: Create folder if not exists

```python
if not os.path.exists("data"):
    os.mkdir("data")
```

### `Notes:`

4. Avoid os.system() in secure applications
