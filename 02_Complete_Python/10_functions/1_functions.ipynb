```python
def even_or_odd_instance(num):
    if isinstance(num, int):
        if num % 2 == 0:
            print(f"{num} is an even number.")
        else:
            print(f"{num} is an odd number.")
    elif isinstance(num,str):
        print("Invalid data type")
    else:
        print(f"{num} is not a valid value for this function")


even_or_odd_instance(5) #5 is an odd number.
even_or_odd_instance('s') #Invalid data type
even_or_odd_instance(14.4) #14.4 is not a valid value for this function
```

## Recursive Function
- function is going to call itself again and again 
- Base Condition: which will stop the loop or which will let the function to stop calling itself

```python
# Write a function to find factorial of a number

# Brute Force
def factorial_brute(n):
    if n ==0 or n==1:
        return 1
    
    else:
        fact = 1
        for i in range(1, n+1):
            fact = fact*i
        
        print(f"factorial of {n} is : ", fact)

factorial_brute(5)       


# Recursive Function
def factorial_recursive(n):
    if n == 1 or n==0: # base condition to exit
        return n
    
    fact = n * factorial_recursive(n-1) # calling factorial function again and again
    return fact

result = factorial_recursive(5)
print(result)
```

## Parameters

```python
# Function with multiple paramerters
# pass multiple values ar our parameter or more than one parameter we can pass

# additon of number
def addition(num1, num2):
    return f"the addition of {num1} and num2 is : {num1 + num2}"

result = addition(1,3)
# result2 = addition(1,3,5) # TypeError: addition() takes 2 positional arguments but 3 were given
print(result)
# print(result2)
```

```python
# Function with Variable Length Arguments
# Positional arguments --> multiple values inside our function as an argument irrespective of the number of values we have mentioned as the parameter inside our 
# funciton

#function to add multipe numbers irrespective of how many number we are passing

def addition_positiona(*num):
    sum = 0
    for i in num:
        sum += i
    
    print(sum)

addition_positiona(1,2)
addition_positiona(1,2,3)
addition_positiona(1,2,3,4)
addition_positiona(1,2,3,4, 555,7,77,4,3,5)
```

### Type of Function Arguments
- Python supports various types of arguments that can be passed at the time of the function call. Below are types of function argument types:
1. Default Arguments
2. Keyword Arguments
3. Positional Arguments
4. Arbitary Arguements

```python
# Default Parameter

def user_name(name, role="user"):
    print(f"name is {name}, role is {role}")

user_name('Arun')
user_name('Arun', "admin")
```

```python
# Keyword Arguments: values are passed by explicitly specifying the parameter names, so the order doesn’t matter.

def student(fname, lname):
    print(fname, lname)

student(fname='Geeks', lname='Practice')
student(lname='Practice', fname='Geeks')
```

```python
# Positional Arguments: values are assigned to parameters based on their order in the function call.

def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)

print("Case-1:")
nameAge("Olivia", 27)

print("\nCase-2:")
nameAge(27, "Olivia")
```

### Arbitrary Arguments: allow a function to accept a variable number of inputs.
- `*args`: collects extra positional (non-keyword) arguments as a tuple.
- `**kwargs`: collects extra keyword arguments as a dictionary.

```python
# Arbitrary Arguments: 

def addition_positiona(*num):
    sum = 0
    for i in num:
        sum += i
    
    print(sum)

addition_positiona(1,2)
addition_positiona(1,2,3,4)
addition_positiona(1,2,3,4, 555,7,77,4,3,5)

print('--------------------------')
def data(**user_data):
    for key,value in user_data.items():
        print(f"{key}-{value}")

data(name='Amit', age=25, gender="Male")

print("--------------------------")

def variable(*positional, **keyword):
    print(keyword)
    for i in positional:
        print(f"Positional value are :", i)
    
    for key, value in keyword.items():
        print(f"The key is {key}-value is {value}")

variable(1,2,3,4, name="arun", age=26)
```
