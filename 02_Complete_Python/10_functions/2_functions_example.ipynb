### 1. Temperature Convertor

```python
'''
Create a function that converts temperature from Celsius to Fahrenheit or Fahrenheit to Celsius

-->if temperature is in Celsius , then Celsius to Fahrenheit conversion is : 
1 deg Celsius : 9/5 + 32 F
45 deg Celsius = 45*9/5 + 32 F


-->if temperature is in Fahrenheit , then Fahrenheit to Celsius conversion is : 
1 Fahrenheit : 5/9 - 32 C
45 deg Celsius = 45*5/9 - 32 C

'''

def temperature_convertor(value, unit):
    if unit.lower() == 'c':
        temperature =( value * 9/5 ) + 32 
        print(f"The temperatur for {value} is: {(temperature):.2f}°F")
    elif unit.lower() == 'f':
        temperature = (value - 32) * 5/9 
        print(f"The temperatur for {value} is: {(temperature):.2f}°C")
    else:
        print("Invalid unit! Use 'C' or 'F'")

temperature_convertor(80, "c")

# With Error Handling
def temperature_converter(value, unit):
    try:
        value = float(value)
    except ValueError:
        return "Invalid temperature value"

    if unit.lower() == "c":
        return f"{value}°C = {(value * 9/5 + 32):.2f}°F"

    elif unit.lower() == "f":
        return f"{value}°F = {((value - 32) * 5/9):.2f}°C"

    else:
        return "Invalid unit"


print(temperature_converter("30", "c"))
```

### 2. Password Strength Checker
- Strong Password: 
  1. Lenght is greater than 8 character
  2. password must consist of atleast a lowercase character ( a, b, c , ... , z)
  3. password must consist of atleast a uppercase character ( A, B, C , ... , Z)
  4. password must consist of atleast a digit (1,2,3,4,....,9)
  5. password must consist of atleast a special character ( !,@,#,$,%,^,&,*,(,),_)

- Weak Password
  1. doesn't match any above condition then it will consider as a weak password

```python
def password_strength_checker(password):
    special_chars = "!@#$%^&*()_"

    if len(password) <= 8:
        print("Weak Password")

    elif not any(char.islower() for char in password):
        print("Weak Password")

    elif not any(char.isupper() for char in password):
        print("Weak Password")

    elif not any(char.isdigit() for char in password):
        print("Weak Password")

    elif not any(char in special_chars for char in password):
        print("Weak Password")

    else:
        print("Strong Password")

password_strength_checker('arun')
password_strength_checker('arunjawlia')
password_strength_checker('Arunjawlia@')
password_strength_checker('Arunjawlia@1998')

    
```

```python
def check_password_strength(password):
    if len(password) <= 8:
        return "Weak Password (must be > 8 characters)"

    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False

    special_chars = "!@#$%^&*()_"

    for ch in password:
        if ch.islower():
            has_lower = True
        elif ch.isupper():
            has_upper = True
        elif ch.isdigit():
            has_digit = True
        elif ch in special_chars:
            has_special = True

    if has_lower and has_upper and has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

pwd = input("Enter password: ")
print(check_password_strength(pwd))
```

## 3. check that whether if the current string is a palindromw or not
- A palindrome is a sequence—such as a word, phrase, or number—that reads the same forward and backward
- "madam", "racecar", "level"
- 121, 12321, 55

```python
# Using Slicing (Best & Pythonic)
def palindrome(str):
    str_value = str.lower().replace(" ", "")
    return str_value == str_value[::-1]


text = input("Enter your string")
if palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")

```

### 4. Calculate the Total Cost of Items in a shopping cart

```python
cart = [
    {
        "item1":"apple",
        "cost":5, 
        "qty": 20
    },
    {
        "item2":"mango",
        "cost":1000, 
        "qty": 10
    },
    {
        "item3":"pineapple",
        "cost":600, 
        "qty": 40
    },
    {
        "item4":"pine-nipple",
        "cost":40000, 
        "qty": 2
    },
]

def cart_total(cart_items):
    total = 0

    for item in cart_items:
        total += item['cost']  * item['qty']
    
    return total

result = cart_total(cart)
print(result)
```
