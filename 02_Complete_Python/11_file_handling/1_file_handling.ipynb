# File Handling in Python
---

- File handling in Python is how you create, read, write, and manage files on disk. 
- It allows your programs to store data permanently on a storage device rather than losing it once the program finishes execution.
- It’s a fundamental I/O (input/output) operation used in logging, data processing, configuration management, etc.

- RAM Memory --> Information will be their temporary
- File --> Information will be their Permanently

### Use cases of file in AI/Ml

1. Natural Language Processing (NLP): Text Data --> Text Format File
    - Data Category: Text Data (Sequential)
    - Common Formats: .txt, .csv, .json, .mdStorage 
    - Type: Text Format

2. Machine Learning (Image Classification) : Image Classifiers --> png or jpeg --> Binary format file
    - Example: Cat/Dog Image Classifier
    - Common Formats: .png, .jpeg, .jpg, .bmp
    - Storage Type: Binary Format

3. Deep Learning (Object Detection) : Object Classifiers --> png or jpeg --> Binary format file
    - Example: Multi-Object Classifiers
    - Common Formats: .png, .jpeg, .tiff, .webp
    - Storage Type: Binary Format

### Types of Files

2. Binary File --> .png, .mp4, .mp3

### File Handling Mode

| Mode  | Meaning                       |
| ----- | ----------------------------- |
| `"r"` | Read (default)                |
| `"w"` | Write (overwrite)             |
| `"a"` | Append                        |
| `"r+"`| read first and then write     |
| `"w+"`| write first and then read     |
| `"x"` | Create (fails if file exists) |
| `"b"` | Binary mode                   |
| `"t"` | Text mode (default)           |



### File Methods

| Method         | Description            |
| -------------- | ---------------------- |
| `read()`       | Read entire file       |
| `readline()`   | Read one line          |
| `readlines()`  | Read all lines as list |
| `write()`      | Write string           |
| `writelines()` | Write list of strings  |
| `seek()`       | Move cursor            |
| `tell()`       | Get cursor position    |
| `close()`      | Close file             |


### 1. Read the content using `read()` from my `example.txt` file in the system

```python
# read() --> read entire file
f =  open('example.txt',"r")
content = f.read()
print(content)
f.close()
```

```python
# Read line by line
with open("data.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
# Read specific lines
with open("data.txt", "r") as f:
    print(f.readline())   # one line
    print(f.readlines())  # list of lines
```

### 2. Write the content using `write()` into the file `example.txt` 
    -  write() --> it overwrite the whole content into the file 
    -  append() --> it adds the data from the end of the file

```python
# write() --> overwrite the file
f = open('example.txt', "w")
f.write("Hello!  this is example of write() into a file ")
f.close()
```

### 3. Appending into the file using `append()`

```python
f = open("example.txt", "a")
f.write("\nThis is append mode text append into the end of the file")
f.close()
```

## `with()` Statement
- Always prefer with → it automatically closes the file (no resource leak).
- used for first of all opening the file as a object and there is not requirement for closing the file as well

```python
with open('example.txt','r') as f:
    content = f.read()
    print(content)
```

```python
with open('example.txt', "w") as file:
    file.write("Hello!  this is example of write() into a file ")
```

```python
with open('example.txt', "a") as file:
    file.write("\nThis is append mode text append into the end of the file")
```

### Write the content into our file line by line --> writelines()

```python
lines = ["first_lines\n", "second_lines\n", "third_line\n"]

with open("example.txt", "w") as data:
    data.writelines(lines)
```

## Working with Binary Files
    - Used for images, videos, etc.

```python
# write content from the binary file: wb
data = b'\x00\x01\x02\x04\x05'

with open('example.txt', 'wb') as binary_data:
    binary_data.write(data)
```

```python
# Read content into the binary file: rb
with open('example.txt', 'rb') as binary_read:
    content = binary_read.read()
    print(content)
```

```python
### Read the content from source.txt and then writing the content into destination file

with open("source.txt", "r") as source:
    content = source.read()

with open("destination.txt", "w") as destination:
    destination.write(content)
```

```python
## Reading the total lines , Total Words and Total Characters from a filter --> readlines()
def count_lines(file):
    with open(file, "r") as data:
        content = data.readlines()
        total_lines = len(content)
        total_word = sum(len(line.split()) for line in content)
        total_char = sum(len(line) for line in content)
        
        return content, total_lines, total_word, total_char

content, total_lines,total_word, total_char,  = count_lines('data.txt')
print("Content: ",content)
print("Total_lines: ",total_lines)
print("Total words: ",total_word)
print("Total Characters: ",total_char)
```

```python
# Write the content into file and then read the content that file as well ---> seek() and w+ mode
# seek ==> help us in reading the content after writing from the starting only

with open("data.txt", 'w+') as write_read_file:
    write_read_file.write("\n We are looking into W+ operations")
    write_read_file.write("\n This line I am going to add")
    write_read_file.seek(0)

    content = write_read_file.read()
    print(content)
```
