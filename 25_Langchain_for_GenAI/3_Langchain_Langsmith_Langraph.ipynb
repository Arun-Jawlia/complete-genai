# `Detailed Notes 1`

# LangChain vs LangSmith vs LangServe vs LangGraph

## Overview

The **LangChain ecosystem** provides different tools for building, orchestrating, debugging, evaluating, and deploying LLM applications.

A simple way to remember them:

> **LangChain → Build**
> **LangGraph → Orchestrate**
> **LangSmith → Observe & Evaluate**
> **LangServe → Serve as API (legacy/deprecated)**

**Important:** LangServe is now deprecated. The official repository was archived in May 2026, and LangChain recommends **LangGraph Platform** instead for new projects. 

---

## 1. LangChain

### Definition

**LangChain is an open-source framework for building LLM applications and agents.**

It provides abstractions and integrations for:

* LLMs / chat models
* Prompts
* Tools
* Agents
* Retrieval
* RAG
* Structured output
* Model integrations

LangChain provides a higher-level way to build agents quickly. ([Docs by LangChain][2])

### What Problem Does LangChain Solve?

Without a framework, we may have to manually connect:

```text
User
 ↓
LLM
 ↓
Tool
 ↓
Database
 ↓
LLM
 ↓
Response
```

LangChain provides reusable components for building these workflows.

### Example

Suppose we want to create a weather agent.

```text
User: What is the weather in Delhi?

        ↓

     LangChain
        ↓
      LLM
        ↓
  Weather Tool
        ↓
 Weather API
        ↓
      Result
        ↓
      LLM
        ↓
    Final Answer
```

LangChain tools can represent functions that allow an agent to fetch data, query databases, execute code, or perform actions. ([Docs by LangChain][3])

### LangChain and Agents

LangChain provides a high-level agent abstraction.

An agent can:

1. Receive a user request.
2. Understand the task.
3. Decide which tool to use.
4. Execute the tool.
5. Observe the result.
6. Continue reasoning.
7. Return the final answer.

LangChain's current agents are built on top of LangGraph. ([Docs by LangChain][4])

### When to Use LangChain?

Use LangChain when:

* You want to build an LLM application quickly.
* You need model/tool integrations.
* You are building a basic or moderately complex agent.
* You want higher-level abstractions.

---

# 2. LangGraph

## Definition

**LangGraph is a low-level orchestration framework and runtime for building stateful, long-running, and complex AI agents.** ([Docs by LangChain][5])

Think of LangGraph as the **workflow engine behind an agent**.

---

## Why LangGraph?

Simple agents can follow:

```text
User → LLM → Tool → LLM → Response
```

But real-world agents can be much more complicated:

```text
             ┌──→ Search ──→
User → Agent ┤
             ├──→ Database ─→
             │
             └──→ Human Approval
                       ↓
                    Continue
```

LangGraph allows us to explicitly define these workflows as a **graph**.

---

## Core Concepts

### Nodes

A **node** represents a step in the workflow.

Example:

```text
LLM Node
Tool Node
Database Node
Human Approval Node
```

### Edges

An **edge** defines how execution moves from one node to another.

```text
LLM → Tool
Tool → LLM
LLM → Human Approval
```

### State

**State** contains information shared between nodes during execution.

Example:

```text
State:
- user_query
- messages
- tool_result
- user_preferences
- current_step
```

---

## Important LangGraph Features

### 1. Stateful Workflows

LangGraph can maintain state throughout an agent's execution.

### 2. Persistence

LangGraph can save graph state using checkpoints.

This enables:

* Memory
* Human-in-the-loop
* Resume after interruption
* Time-travel debugging
* Fault tolerance ([Docs by LangChain][6])

### 3. Human-in-the-Loop

An agent can stop and ask a human for approval.

Example:

```text
Agent
 ↓
Generate SQL
 ↓
Human Approval
 ↓
Execute SQL
```

The human can approve, modify, or reject the action. ([Docs by LangChain][7])

### 4. Durable Execution

Long-running workflows can resume after failures instead of starting completely from scratch.

### 5. Streaming

LangGraph supports streaming execution and responses.

---

## When to Use LangGraph?

Use LangGraph when you need:

* Complex agent workflows
* Stateful agents
* Long-running tasks
* Human-in-the-loop
* Multiple agents
* Conditional routing
* Persistence
* Fine-grained control over execution

Official documentation describes LangGraph as the lower-level orchestration layer, while LangChain provides higher-level agent abstractions. ([Docs by LangChain][8])

---

# 3. LangSmith

## Definition

**LangSmith is a platform for tracing, debugging, evaluating, and monitoring LLM applications and agents.**

It is **not primarily the framework that builds your agent**.

Think:

> **LangChain/LangGraph build the application.
> LangSmith helps you understand whether it is working correctly.**

LangSmith can capture traces showing model calls, tool calls, execution steps, inputs, outputs, and other runtime information. ([Docs by LangChain][9])

---

## What Problem Does LangSmith Solve?

Imagine your agent gives the wrong answer.

You need to know:

```text
Why did the agent fail?

Was the prompt wrong?
       ↓
Did the LLM make a bad decision?
       ↓
Did it select the wrong tool?
       ↓
Did the tool return incorrect data?
       ↓
Did RAG retrieve the wrong documents?
       ↓
Did the final response generation fail?
```

LangSmith helps you inspect the execution trace.

---

## LangSmith Tracing

Example:

```text
User Request
     ↓
Agent
     ↓
Prompt
     ↓
LLM Call
     ↓
Tool Selection
     ↓
Database Query
     ↓
Tool Result
     ↓
LLM
     ↓
Final Response
```

LangSmith can visualize these steps as traces.

This makes debugging complex agent behavior much easier. ([Docs by LangChain][10])

---

## LangSmith Evaluation

LangSmith can evaluate LLM applications before and after deployment.

### Offline Evaluation

Used before deployment.

```text
Dataset
   ↓
Run Agent
   ↓
Evaluate
   ↓
Compare Versions
```

Useful for:

* Benchmarking
* Regression testing
* Comparing prompts
* Comparing models
* Testing RAG quality

### Online Evaluation

Used with production traffic.

```text
Real User
    ↓
Production Agent
    ↓
LangSmith
    ↓
Evaluation
    ↓
Monitoring / Alerts
```

LangSmith supports both offline and online evaluation workflows. ([Docs by LangChain][11])

---

## When to Use LangSmith?

Use LangSmith when you need:

* Tracing
* Debugging
* Evaluation
* Monitoring
* Prompt experimentation
* Dataset-based testing
* Production observability

---

# 4. LangServe

## Definition

**LangServe was a library for deploying LangChain runnables and chains as REST APIs.**

It integrated with **FastAPI** and **Pydantic** for API serving and validation. ([GitHub][1])

Example:

```text
LangChain Application
        ↓
     LangServe
        ↓
     REST API
        ↓
Frontend / Mobile App / Other Services
```

For example:

```text
POST /chat
POST /invoke
POST /batch
```

---

## Important: LangServe Is Deprecated

This is an important **current interview point**.

LangServe was deprecated in November 2024 and its GitHub repository was archived on **May 5, 2026**.

The LangChain team recommends **LangGraph Platform** rather than LangServe for new projects. ([GitHub][1])

Therefore:

> **Do not learn LangServe as the primary modern deployment solution for new projects.**

You should understand it because you may encounter it in older tutorials, projects, or interview questions.

# LangChain vs LangGraph vs LangSmith vs LangServe

| Tool          | Main Purpose                        | Think Of It As                        |
| ------------- | ----------------------------------- | ------------------------------------- |
| **LangChain** | Build LLM apps and agents           | 🧱 Building framework                 |
| **LangGraph** | Orchestrate complex/stateful agents | 🔀 Workflow engine                    |
| **LangSmith** | Trace, debug, evaluate, monitor     | 🔍 Observability platform             |
| **LangServe** | Expose LangChain apps as APIs       | 🌐 API serving layer — **deprecated** |

---

# How They Work Together

A modern GenAI application can look like this:

```text
                    ┌─────────────────┐
                    │   User / App    │
                    └────────┬────────┘
                             ↓
                    ┌─────────────────┐
                    │   LangChain     │
                    │ Build Agent/App │
                    └────────┬────────┘
                             ↓
                    ┌─────────────────┐
                    │   LangGraph     │
                    │ Orchestration   │
                    └────────┬────────┘
                             ↓
              ┌──────────────┼──────────────┐
              ↓              ↓              ↓
            LLM           Tools           RAG
              └──────────────┼──────────────┘
                             ↓
                       Final Response

                             │
                             ↓
                    ┌─────────────────┐
                    │   LangSmith     │
                    │ Trace/Evaluate  │
                    │   /Monitor      │
                    └─────────────────┘
```

LangChain's agents are currently built on LangGraph, while LangSmith provides tracing and evaluation across these applications. ([Docs by LangChain][2])

# Simple Real-World Example

Suppose we build an **AI Customer Support Agent**.

### Step 1 — LangChain

Use LangChain to connect:

```text
LLM
+
Tools
+
Retriever
+
Agent
```

---

### Step 2 — LangGraph

Our workflow might be:

```text
User Question
      ↓
Classify Question
      ↓
 ┌────┴─────┐
 ↓          ↓
Billing   Technical
 ↓          ↓
Tool      RAG
 └────┬─────┘
      ↓
Human Approval?
      ↓
Final Response
```

LangGraph manages the workflow, state, persistence, and routing.

---

### Step 3 — LangSmith

Monitor the application:

```text
Question
   ↓
Agent Decision
   ↓
Tool Call
   ↓
RAG Retrieval
   ↓
LLM Response
```

LangSmith lets us inspect and evaluate these steps.

---

### Step 4 — Deployment

For a **new application**, use the current LangGraph/LangSmith deployment options rather than starting a new project with LangServe. LangServe is deprecated. ([GitHub][1])

# 2. Interview Questions & Answers

## Beginner Questions

### Q1. What is LangChain?

**Answer:**
LangChain is an open-source framework for building LLM applications and agents. It provides integrations and abstractions for models, tools, retrieval, prompts, and agents. ([Docs by LangChain][2])

---

### Q2. What is LangGraph?

**Answer:**
LangGraph is a low-level orchestration framework and runtime for building complex, stateful, long-running agent workflows.

---

### Q3. What is LangSmith?

**Answer:**
LangSmith is a platform used for tracing, debugging, evaluating, and monitoring LLM applications and agents.

---

### Q4. What was LangServe?

**Answer:**
LangServe was a library used to expose LangChain runnables and chains as REST APIs. It is now deprecated. ([GitHub][1])

---

### Q5. Is LangSmith used to build agents?

**Answer:**
Not primarily. LangChain and LangGraph are used to build/orchestrate agents, while LangSmith is used to observe, debug, evaluate, and monitor them.

---

## Intermediate Questions

### Q6. What is the difference between LangChain and LangGraph?

**Answer:**

| LangChain                      | LangGraph                         |
| ------------------------------ | --------------------------------- |
| Higher-level framework         | Lower-level orchestration runtime |
| Easier to get started          | More control                      |
| Build agents/apps              | Control agent workflows           |
| Provides integrations          | Provides stateful graph execution |
| Good for common agent patterns | Good for complex workflows        |

LangChain agents themselves are built on LangGraph. ([Docs by LangChain][2])

---

### Q7. Why would you use LangGraph instead of LangChain alone?

**Answer:**
Use LangGraph when the agent requires complex routing, persistent state, long-running execution, human approval, fault tolerance, or fine-grained orchestration.

---

### Q8. What is a node in LangGraph?

**Answer:**
A node represents an individual step or operation in a graph, such as calling an LLM, executing a tool, querying a database, or performing validation.

---

### Q9. What is an edge in LangGraph?

**Answer:**
An edge defines the transition between nodes and determines which step executes next.

---

### Q10. What is LangSmith tracing?

**Answer:**
Tracing records the execution steps of an LLM application, including model calls, tool calls, inputs, outputs, and decision points, allowing developers to debug and analyze behavior. ([Docs by LangChain][12])

---

## Scenario-Based Questions

### Q11. You built an agent, but sometimes it selects the wrong tool. What would you use?

**Answer:**
Use **LangSmith** to trace the agent execution and inspect why the model selected that tool.

---

### Q12. You need an agent that pauses for human approval before sending an email. What would you use?

**Answer:**
Use **LangGraph**, because it supports stateful workflows and human-in-the-loop execution. ([Docs by LangChain][7])

---

### Q13. You need to compare two different prompts on 1,000 test questions. What would you use?

**Answer:**
Use **LangSmith Evaluation** to create a dataset, run experiments, and compare the results. ([Docs by LangChain][11])

---

### Q14. You need to build a simple RAG chatbot quickly. Which would you start with?

**Answer:**
Start with **LangChain**, because it provides higher-level abstractions and integrations for retrieval and LLM applications. ([Docs by LangChain][13])

---

### Q15. You are designing a complex multi-step agent with multiple conditional paths. Which would you use?

**Answer:**
Use **LangGraph** for explicit workflow orchestration and state management.

---

### Q16. You find an old project using LangServe. Should you use it for a new project?

**Answer:**
No. LangServe is deprecated. For new projects, use the current LangGraph/LangSmith deployment options recommended by LangChain. ([GitHub][1])
