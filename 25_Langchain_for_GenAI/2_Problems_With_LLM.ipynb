# Problems with Using LLMs Alone

Large Language Models (LLMs) are powerful, but by themselves they have several limitations when building real-world AI applications.

---

## 1. LLMs Can't Remember Context Beyond the Prompt

A standalone LLM only knows the information included in the current prompt (or conversation context that is explicitly provided).

### Example

**User**

> Who won the Cricket World Cup in 2011?

**LLM**

> India.

Next question:

> Which country played against India?

Without previous conversation history, the model cannot reliably determine that you're referring to the **2011 Cricket World Cup Final**.

### Solution: LangChain Memory

LangChain provides **Memory** components that store and retrieve conversation history.

```text
User
   │
Conversation History
   │
Memory
   │
LLM
   │
Answer
```

Now the model understands the context of follow-up questions.

---

# 2. LLMs Cannot Directly Connect to Databases, APIs, or Websites

Suppose a company has an SQL database containing employee records.

```text
Employee Database

Name
Salary
Department
Leave Records
```

If we ask:

> Who has the highest salary?

The LLM cannot access the database on its own.

Similarly, it cannot directly retrieve:

* SQL data
* MongoDB data
* AWS S3 files
* Google Cloud Storage
* Azure Blob Storage
* REST APIs
* Websites

### Solution: Data Connectors

LangChain provides integrations to connect LLMs with external data sources.

Examples include:

* SQL Databases
* PostgreSQL
* MySQL
* MongoDB
* AWS
* Azure
* Google Cloud (GCP)
* PDFs
* CSV files
* APIs
* Websites

```text
User
   │
LangChain
   │
Database/API
   │
Retrieve Data
   │
LLM
   │
Answer
```

---

# 3. LLMs Cannot Perform Actions

LLMs generate text, but they do not inherently perform actions such as:

* Searching the web
* Running Python code
* Executing SQL queries
* Calling APIs
* Sending emails
* Booking appointments
* Reading files

### Example

User:

> What's the weather in Delhi today?

A standalone LLM cannot fetch live weather information.

### Solution: Tools and Agents

LangChain allows LLMs to use external **Tools**.

Examples of tools:

* Weather API
* Python Interpreter
* SQL Database
* Search Engine
* Calculator
* File Reader

```text
User
   │
LangChain Agent
   │
Weather API
   │
Current Temperature
   │
LLM
   │
Final Answer
```

The LLM decides when to use a tool, receives the result, and produces a natural-language response.

---

# 4. Building Multi-Step AI Workflows Is Difficult

Many AI applications require multiple coordinated steps.

Example:

* User asks a question.
* Search relevant documents.
* Split documents into chunks.
* Generate embeddings.
* Retrieve the most relevant chunks.
* Send them to the LLM.
* Produce the final answer.

Implementing this manually is complex.

### Solution: Chains

LangChain introduces **Chains**, which connect multiple components into a structured workflow.

Example (RAG Pipeline):

```text
User Question
       │
       ▼
Load Documents
       │
       ▼
Split into Chunks
       │
       ▼
Generate Embeddings
       │
       ▼
Store in Vector Database
       │
       ▼
Retrieve Relevant Chunks
       │
       ▼
LLM
       │
       ▼
Final Answer
```

# Benefits of LangChain

## 1. Supports RAG (Retrieval-Augmented Generation)

One of LangChain's biggest strengths is simplifying the development of **RAG applications**.

RAG Workflow:

```text
PDF
 │
 ▼
Document Loader
 │
 ▼
Text Splitter
 │
 ▼
Embeddings
 │
 ▼
Vector Database
 │
 ▼
Retriever
 │
 ▼
LLM
 │
 ▼
Answer
```

Applications include:

* PDF Chatbots
* Company Knowledge Assistants
* Research Assistants
* Legal Document Search
* Medical Knowledge Bots

---

## 2. Model-Agnostic Development

LangChain is **model agnostic**, meaning your application isn't tied to a single LLM provider.

You can easily switch between models, for example:

* OpenAI GPT
* Google Gemini
* Anthropic Claude
* Meta Llama
* Mistral
* Gemma
* Hugging Face models

For example, you might choose:

* **GPT-5** for strong reasoning and coding.
* **Gemini Flash** for low latency and speed.
* **Grok** for coding or other specialized tasks.

This flexibility makes it easier to replace one model with another as new models become available, often with minimal code changes.

---

## 3. Complete Ecosystem

LangChain provides a rich ecosystem of reusable components.

Some important modules include:

* **Document Loaders** – Load PDFs, Word files, CSVs, websites, etc.
* **Text Splitters** – Break large documents into manageable chunks.
* **Embedding Models** – Convert text into vector representations.
* **Vector Stores** – Store and search embeddings efficiently.
* **Retrievers** – Fetch the most relevant information.
* **Prompt Templates** – Create reusable prompts.
* **Memory** – Maintain conversation context.
* **Chains** – Build multi-step workflows.
* **Tools** – Integrate external capabilities.
* **Agents** – Enable LLMs to decide which tools to use.
* **Output Parssers** – Structure and validate model outputs.

This modular design allows developers to build production-ready AI systems more efficiently.

# What is LangSmith?

## Definition

**LangSmith** is the official developer platform from the LangChain ecosystem for **debugging, testing, evaluating, and monitoring LLM applications**.

Think of it as:

> **GitHub Actions + Debugger + Analytics Dashboard for LLM Applications**

While LangChain helps you **build** AI applications, **LangSmith helps you improve and operate them**.

---

# Why Do We Need LangSmith?

Suppose your chatbot gives an incorrect answer.

Questions you might ask:

* Which prompt was sent to the LLM?
* Which documents were retrieved?
* Which tool was used?
* How many tokens were consumed?
* How much did the request cost?
* Where did the workflow fail?

Without tracing, these questions are difficult to answer.

LangSmith records the execution of your application step by step.

# Features of LangSmith

### 1. Tracing

Visualizes every step of your AI application's execution, making it easier to understand how a response was generated.

---

### 2. Debugging

Helps identify issues such as:

* Incorrect prompts
* Retrieval failures
* Tool execution errors
* Agent decision mistakes

---

### 3. Evaluation

Measure application quality using metrics like:

* Correctness
* Relevance
* Helpfulness
* Faithfulness (especially for RAG)

You can compare different prompts or LLMs to determine which performs better.

---

### 4. Monitoring

Track production metrics such as:

* Latency
* Token usage
* Request volume
* Cost
* Error rates

---

### 5. Prompt Versioning

Manage and compare different prompt versions without changing application code, making experimentation safer and more systematic.
