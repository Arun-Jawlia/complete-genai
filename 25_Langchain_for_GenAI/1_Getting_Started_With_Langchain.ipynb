# What is LangChain?

**LangChain** is an **open-source framework** that helps developers build applications powered by Large Language Models (LLMs).

Instead of writing everything from scratch, LangChain provides **ready-made components** that make it easier to connect LLMs with memory, documents, databases, APIs, tools, and external services.

### Definition

> **LangChain is an open-source framework that simplifies the development of LLM-powered applications by providing modular components for integrating language models with external data sources, tools, memory, and workflows.**

---

# Why Do We Need LangChain?

Although LLMs are powerful, they have several limitations when used by themselves.

Let's understand these problems.

---

# Problem 1: LLM Has No Long-Term Memory

Suppose we ask:

```text
Who won the Cricket World Cup in 2011?
```

Answer:

```text
India
```

Next question:

```text
Which country was playing against India?
```

Without conversation history or memory, the LLM has no way to know that the second question refers to the **2011 Cricket World Cup**.

It only sees the current prompt.

### Limitation

LLMs do not automatically remember previous conversations beyond the context they are given.

---

# How LangChain Solves It

LangChain provides **Memory** components.

Example:

```text
User
 │
 ▼
Conversation Memory
 │
 ▼
LLM
 │
 ▼
Response
```

Memory stores previous interactions and sends the relevant history along with the current prompt.

Now the model understands:

> "Which country was playing against India?"

refers to

> "2011 Cricket World Cup."

---

# Problem 2: LLM Cannot Access Your Personal Data

Suppose a company has an employee database.

```text
Employee Database
```

Contains:

* Employee Names
* Salaries
* Departments
* Leave Records

If we ask:

```text
Who is the highest paid employee?
```

The LLM cannot answer because it has no direct access to the database.

---

# How LangChain Solves It

LangChain can connect the LLM to external data sources.

```text
User Question
      │
      ▼
LangChain
      │
      ▼
SQL Database
      │
      ▼
Retrieve Data
      │
      ▼
LLM
      │
      ▼
Final Answer
```

The framework retrieves the relevant data first, then asks the LLM to explain or summarize it.

---

# Problem 3: LLM Cannot Browse the Internet by Itself

Ask:

```text
What is today's weather?
```

A standalone LLM cannot fetch live weather information on its own.

---

# How LangChain Solves It

LangChain integrates with APIs.

```text
User
 │
 ▼
Weather API
 │
 ▼
Latest Weather
 │
 ▼
LLM
 │
 ▼
Natural Language Response
```

The LLM uses fresh data instead of relying only on what it learned during training.

---

# Problem 4: LLM Cannot Execute Code or Perform Actions Alone

Suppose we ask:

```text
Calculate the average salary from my employee database.
```

A plain LLM cannot:

* Execute Python code
* Query SQL databases
* Read Excel files
* Search the web
* Send emails

without additional systems around it.

---

# How LangChain Solves It

LangChain allows LLMs to use **Tools**.

Examples of tools:

* Python Interpreter
* SQL Database
* Search Engine
* APIs
* Email Services
* File Readers
* Custom Functions

```text
User
 │
 ▼
LangChain
 │
 ├── SQL Tool
 ├── Python Tool
 ├── Search Tool
 └── API Tool
 │
 ▼
LLM
 │
 ▼
Answer
```

The LLM decides when to use a tool, receives the result, and generates a final response.

---

# Problem 5: LLM Cannot Answer Questions About Your Documents

Suppose you upload:

* PDF
* Word document
* Company policies
* Research papers

A standalone LLM does not automatically know their contents.

---

# How LangChain Solves It with RAG

LangChain supports **Retrieval-Augmented Generation (RAG)**.

### RAG Pipeline

```text
PDF Documents
      │
      ▼
Load Documents
      │
      ▼
Split into Chunks
      │
      ▼
Create Embeddings
      │
      ▼
Store in Vector Database
      │
      ▼
User Question
      │
      ▼
Similarity Search
      │
      ▼
Relevant Chunks
      │
      ▼
LLM
      │
      ▼
Accurate Answer
```

Instead of relying only on the model's training data, the LLM retrieves relevant information from your documents before answering.

# Basic LangChain Example

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4.1-mini"
)

response = llm.invoke("Explain AI.")

print(response.content)
```

---

# Prompt Template Example

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} for a {level}."
)

messages = prompt.invoke({
    "topic": "Python",
    "level": "Beginner"
})

print(messages)
```

---

# Simple Chain Example

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | llm | StrOutputParser()

result = chain.invoke({
    "topic": "Machine Learning",
    "level": "Beginner"
})

print(result)
```

The `|` operator connects components into a chain.
