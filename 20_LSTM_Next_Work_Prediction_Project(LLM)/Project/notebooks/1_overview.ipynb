## `We are going to build Next Word Prediction using LSTM`

eg. 
- Suggest Next work in Keyboards ( google , iphone)
- In Whatsapp, we communicate with people and our keyboard automatically suggest us the next set of words as a prediction of next words. 
- ChatGPT --> GPT --> APP --> ChatGPT --> task Pretraining --> Next word prediction --> Fine Tuning
- Next word Prediction is the task of predicting the most likely next word given a sequence of previous words.

This is how exactly : Autocomplete works, Chatbots thinks, LLM model begain thinking from next word prediction

## Why use LSTM is needed for Next Word Prediction
- If use ANN, No memory
- If use CNN, used with image or videos ( Computer vision) bases project
- If use RNN, short term memory, it is not good with long term context - no access to long term memory
- That's why use LSTM, have Both Short term and Long term memory

Language Neds word order, context, long term dependency
LSTM is specially designed for this Task


## Real Life Analogy
- Sentence : I am going to the ____________
- our brain instantly predicts : [market], [office], [school], [gym]
- LSTM also work in a similar work - grammer, context, past word

## High level Architecture
- Next Work Prediction App
- Text --> Tokenization --> Embedding Layer ( exact semantic meaning) --> LSTM --> Dense ( softmax ) --> Next work probability
- Each component has a clear role 
- Next work prediction can be treated as Text Generator Task --> Supervised Learning Strategy
- More amount of data --> Much accurate prediction
- Right now --> Not Supervised --> we have to conver this task into --> Supervised Task

# Possible Improvements

* Replace LSTM with **Bidirectional LSTM**.

* Extend the model to predict multiple words instead of just one.
