# `Distributional Plot`
---

- A Distribution Plot is used to understand how values of a numerical column are distributed across a dataset.
- It answers questions like:

    - What values occur most frequently?
    - Is the data normally distributed?
    - Is the data skewed?
    - Are there multiple peaks?
    - What is the spread of the data?

- Axes Level Func --> histplot()
- 
- Figure Level Func --> displot() --> Kind = 'hist'

`Distribution Plot in Seaborn`

- The most common function is:

- `sns.histplot()`

and previously:

- `sns.distplot()`

⚠️Note:  distplot() is deprecated and should not be used in new projects.

## 1. `histplot` 
- used to check the distribution of a Numerical Columns wrt bins as the size of the bin reflects into equal interval
- Histogram divides data into intervals called bins.
- Synatx: `histplot()`
- Parameter
  - data --> dataframe
  - x --> column_name
  - bins --> Controls the number of intervals.
  - kde --> Adds a smooth density curve.

```python
student_data.head()
```

```python
# Age Distribution
# axes level function
sns.histplot(data=student_data, x='Age')
plt.show()
```

```python
# Age Distribution
# axes level function
sns.histplot(data=student_data, x='Age',bins =20)
plt.show()
```

```python
# axes level function
sns.histplot(data=student_data, x='Age',hue='Gender')
plt.show()
```

```python
# Figure Level Functions

sns.displot(data= student_data, x='Age', hue='Gender', kind='hist')
```


## `KDE Plot `
- KDE stands for Kernel Density Estimation Plot
- It is a smooth version of a histogram.
- probability density function representing the Distribution of the data as with curve How data is distributed exactly. ( Used when only the density curve is needed. )


Axis Level Fucntion --> sns.kdeplot()
Parameter
1. data -- dataframe
2. x --> column name whose distribution we want to find
3. fill --> to fill under the curve --> True/ False ( bdefault --> False)

```python
sns.kdeplot(data = student_data, x='Age')
```

```python
sns.kdeplot(data = student_data, x='Age', fill= True, hue='Gender')
```

```python
sns.kdeplot(data = tips, x='total_bill', fill= True)
```

```python
sns.kdeplot(data = tips, x='total_bill', fill= True, hue='sex')
plt.show()
```

```python
sns.displot(data = gaps, x='lifeExp', fill= True, kind='kde')
plt.show()
```

## `Rugplot`
-- it adds the small vertical lines like ticks along the axis and that shows the individual data points 
# More deep the Rug plot ticks --> more density we can exepect

- It is usally used in conjection with other pltos like kde plot as --> Kde Plot + Rugplot


```python
sns.rugplot(data = gaps, x='lifeExp')
plt.show()
```

```python
sns.rugplot(data = tips, x='total_bill')
plt.show()
```

```python
sns.kdeplot(data = tips, x = 'total_bill')
sns.rugplot(data = tips, x = 'total_bill')
plt.show()
```

### `Bivariate Histogram`:
- Creates the Histogram in both X and Y axis , as Histogram with one Numerical Qty in X direction and One Numerical Qty in Y direction
- A Bivariate Histogram divides the x-axis and y-axis into bins and counts how many observations fall inside each rectangular region.
- Bivariate Histogram provides us with the Rectangular Grid as the Conjunction in both Direction

| Color  | Meaning               |
| ------ | --------------------- |
| Light  | Few observations      |
| Medium | Moderate observations |
| Dark   | Many observations     |


What is a Bivariate Histogram in Seaborn?

A Bivariate Histogram is a two-dimensional histogram used to analyze the joint distribution of two numerical variables. It divides both axes into bins and uses color intensity to represent the number of observations in each bin. In Seaborn, it is created using sns.histplot() by specifying both x and y parameters and is useful for density analysis of large datasets.

```python
# Bivariate Histogram
sns.histplot(data= tips, x='total_bill', y = 'tip')
```

```python

sns.kdeplot(data= tips, x='total_bill', y = 'tip')
```

Bivariate Histogram with KDE Plot smoothes the x,y observation with a 2D Gaussian Plane 
it will create the circular plots and more closer the circular --> more density we can find

Circular figure closeness ( More ) --> Density ( More )

## `What is a Distribution Plot in Seaborn?`

A Distribution Plot is used to visualize how values of a numerical variable are spread across a dataset. It helps identify the frequency, density, shape, skewness, and spread of data. In Seaborn, distribution plots are commonly created using histplot() and kdeplot(), which are mainly used for univariate analysis.
