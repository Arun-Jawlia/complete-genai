# ` Regression Plot in Seaborn`
---

## Definition

A **Regression Plot** is used to visualize the relationship between two numerical variables along with the **best-fit regression line**.

It helps determine:

- Whether a relationship exists between variables
- Positive or negative correlation
- Strength of the relationship
- Trend prediction - predicts the future values os if trend is strong then we can predict easily

---

## Why Use Regression Plot?

Suppose we have:

| Study Hours | Exam Score |
|------------|-----------|
| 2 | 45 |
| 4 | 60 |
| 6 | 75 |
| 8 | 90 |

Questions:

- Does studying more increase scores?
- What trend exists in the data?
- Can future values be predicted?

A regression plot helps answer these questions.

---

## 1. `regplot()`
---

## Purpose

Draws a scatter plot with a regression line.

## Syntax

```python
sns.regplot(
    data=df,
    x="Study Hours",
    y="Exam Score"
)
```

## Example

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(
    data=df,
    x="Study Hours",
    y="Exam Score"
)

plt.show()
```

---

## Output

```text
•     Regression Line
•   /
•  /
• /
•/
```

# Understanding the Regression Line

The line represents:

```text
y = mx + c
```

Where:

- x = Independent Variable
- y = Dependent Variable
- m = Slope
- c = Intercept

The line shows the overall trend in the data.

---

# Positive Correlation

```text
     •
   •
 •
•
+---------------->
```

As X increases, Y increases.

Example:

```text
Study Hours vs Marks
```

---

# Negative Correlation

```text
•
  •
    •
      •
+---------------->
```

As X increases, Y decreases.

Example:

```text
Stress Level vs Sleep Hours
```

---

# No Correlation

```text
•      •
   •
      •
 •
```

No clear trend exists.

# `Important Parameters`

## scatter=False

Hide scatter points.

```python
sns.regplot(
    data=df,
    x="Study Hours",
    y="Exam Score",
    scatter=False
)
```

Shows only regression line.

---

## ci

Confidence Interval

```python
sns.regplot(
    data=df,
    x="Study Hours",
    y="Exam Score",
    ci=95
)
```

Default:

```python
95%
```

The shaded area around the line represents confidence interval.

---

## marker

Change marker style.

```python
sns.regplot(
    data=df,
    x="Study Hours",
    y="Exam Score",
    marker="o"
)
```

---

## line_kws

Customize regression line.

```python
sns.regplot(
    data=df,
    x="Study Hours",
    y="Exam Score",
    line_kws={"color":"red"}
)
```

## 2. `lmplot()`
---

## Purpose

Figure-level version of regplot.

Supports faceting and grouping.

## Syntax

```python
sns.lmplot(
    data=df,
    x="Study Hours",
    y="Exam Score"
)
```

---

## Example with Categories

```python
sns.lmplot(
    data=df,
    x="Study Hours",
    y="Exam Score",
    hue="Gender"
)
```

Output:

```text
Male Regression Line
Female Regression Line
```

Useful for comparing trends across categories.

---

# regplot() vs lmplot()

| regplot() | lmplot() |
|------------|----------|
| Axes-level function | Figure-level function |
| Single plot | Multiple plots possible |
| Faster | More flexible |
| No faceting | Supports faceting |
