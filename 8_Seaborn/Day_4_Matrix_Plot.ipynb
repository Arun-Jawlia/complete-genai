## `Matrix Plot in Seaborn`

A Matrix Plot is a visualization used to represent data stored in a matrix (rows × columns) using colors.

Instead of bars, lines, or points, matrix plots use color intensity to represent values.

## Simple Definition

A Matrix Plot visualizes values in a two-dimensional matrix using colors, making it easy to identify patterns, correlations, missing values, and clusters.

---

## `Why Do We Use Matrix Plots?`

Suppose we have the following data:

| Student | Math | Science | English |
| ------- | ---- | ------- | ------- |
| A       | 90   | 85      | 80      |
| B       | 70   | 75      | 65      |
| C       | 95   | 92      | 88      |


Looking at numbers is difficult.

A matrix plot converts numbers into colors:

```code
       Math  Sci  Eng
A      ████ ███  ██
B      ██   ██   █
C      ████ ████ ███
```

Darker color = Higher value

Lighter color = Lower value

Note: In Matrix Plot, only Axes Level Functions, not having Figure Level Functions

## `Heatmap`
---

- Definition: A **Heatmap** is a matrix plot that represents values using different colors.
- Higher values are shown with darker/intense colors and lower values with lighter colors.
- Heatmap uses color shades to show the size of Number in a Matrix 2D table
- In every classification Machine Learning Problem Statement, we use the Heatmap
- For coming up an Heatmpa or an Matrix which we can convert into an Heatmap, we create a Pivot Table and then that Pivot Table, we can use to create an Heatmap


- pivot() --> index, columns, values --> gap.pivot(index='', columns = '', values = '' )

### Why Use Heatmap?
- Visualize correlation between numerical features
- Detect patterns and trends
- Identify missing values
- Easy representation of matrix data

### Syntax: `sns.heatmap(data)`

### Correlation Heatmap

```code
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

sns.heatmap(corr)
plt.show()
```

### Important Parameters

```code
sns.heatmap(
    corr,
    annot=True,      # Show values inside cells --> shows number as well for the rectangular grid
    cmap="coolwarm", # Color palette --> special styling for the heatmap itself
    linewidths=1,    # Border width
    cbar=True        # Show color bar
)
```

### Common Color Maps --> values for cmap

```code
"coolwarm"
"viridis"
"Blues"
"YlGnBu"
"rocket"
```

### Missing Value Heatmap

```code
sns.heatmap(df.isnull())
plt.show()
```

### Applications
- Correlation Analysis
- Missing Value Detection
- Feature Selection
- Data Pattern Analysis

# Clustermap
---
- A **Clustermap** is an advanced heatmap that automatically groups similar rows and columns using hierarchical clustering.
- combination of Both HeatMpa and Automatic Grouping using clustering
- Groups the similar rows and similar columns using Machine Learning Algorithm internally ( heirarchial Cluster Technique - to cluser the Rows Together and it uses Algomerative Heirarchial clustering)
- Helps in Finding the hidden patterns in your data.

### Why Use Clustermap?
- Find hidden patterns
- Group similar features
- Identify clusters in data
- Useful in Machine Learning and Data Analysis

### Syntax: `sns.clustermap(data)`

### Example

```code
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

sns.clustermap(corr)
plt.show()
```

### Important Parameters

```code
sns.clustermap(
    corr,
    annot=True,
    cmap="coolwarm",
    figsize=(8,6)
)
```

### Difference Between Heatmap and Clustermap

| Heatmap | Clustermap |
|----------|------------|
| Shows matrix values using colors | Shows matrix values + clustering |
| No grouping | Automatic grouping |
| Simpler | More advanced |
| Correlation visualization | Pattern and cluster detection |

### Applications
- Feature Clustering
- Pattern Recognition
- Gene Expression Analysis
- Similarity Detection
- Machine Learning EDA

### `What is the difference between Heatmap and Clustermap?`

Heatmap visualizes matrix values using colors, while Clustermap extends a heatmap by automatically clustering similar rows and columns, making it easier to identify groups and hidden patterns in the data.
