# `Multi Plots in Seaborn`
---

## Definition

Multi Plots are used to visualize **multiple variables simultaneously** in a single figure.

Combination of Multi Plots Together to Analyse the Relationships between Variables in a similary way.

They help answer questions like:

- How are all numerical columns related?
- Which features are correlated?
- What patterns exist across multiple variables?
- How do distributions compare?

---

## Types of Multi Plots in Seaborn

### 1. Pair Plot (`pairplot`)
### 2. Joint Plot (`jointplot`)
### 3. Pair Grid (`PairGrid`)
### 4. Joint Grid (`JointGrid`)
### 5. Facet Grid (`FacetGrid`)


````python
# Import required libraries
import numpy as np                  # Used for numerical operations
import pandas as pd                 # Used for loading and working with datasets
import matplotlib.pyplot as plt     # Used to display plots
import seaborn as sns               # Used for statistical visualizations
import plotly.express as px         # Used to load Plotly sample datasets
````

````python
# Load custom datasets from the Datasets folder
student_data = pd.read_csv("Datasets/student_depression_dataset.csv")
chocolate_data = pd.read_csv("Datasets/Chocolate Sales.csv")
gaps = px.data.gapminder()
tips = sns.load_dataset('tips')
````


# 1. `Pair Plot`

## Definition

A Pair Plot creates scatter plots for every combination of numerical columns and distributions on the diagonal.

It plots Multiple Scatter Plot and Histograms for all combination of Numerical columns

Quick overview of the entire dataset

## Syntax: `sns.pairplot(data)`

## Output

```text
        Age   CGPA  Score
Age      H     S      S
CGPA     S     H      S
Score    S     S      H
```

Where:

```text
H = Histogram/KDE
S = Scatter Plot
```

## Use Cases

- Feature Relationship Analysis
- Correlation Detection
- Exploratory Data Analysis (EDA)

---

## Important Parameters

### `hue` : Colors plots based on categories.

```python
sns.pairplot(
    df,
    hue="Gender"
)
```

### `vars` : Select specific columns.

```python
sns.pairplot(
    df,
    vars=["Age","CGPA","Score"]
)
```

### `diag_kind` : Diagonal plots become KDE plots.

```python
sns.pairplot(
    df,
    diag_kind="kde"
)
```
---


````python
sns.pairplot(data = student_data)
````

````python
sns.pairplot(data = tips)
````

````python
sns.pairplot(data = chocolate_data)
````

````python
sns.pairplot(data = gaps)
````

# 2. Joint Plot
---

## Definition

Shows the relationship between two numerical variables along with their individual distributions.

Scatter + Histogram Together

Useful when we want a detailed View of 2 Numerical Variables.

## Syntax

```python
sns.jointplot(
    data=df,
    x="Study Hours",
    y="Depression Score"
)
```

## Output

```text
Histogram
    ↑
Scatter Plot
    ↓
Histogram
```

## Use Cases

- Correlation Analysis
- Bivariate Analysis

---

## Types of Joint Plot

### Scatter

```python
sns.jointplot(
    data=df,
    x="Age",
    y="CGPA",
    kind="scatter"
)
```

### KDE

```python
sns.jointplot(
    data=df,
    x="Age",
    y="CGPA",
    kind="kde"
)
```

### Hex

```python
sns.jointplot(
    data=df,
    x="Age",
    y="CGPA",
    kind="hex"
)
```

### Regression

```python
sns.jointplot(
    data=df,
    x="Age",
    y="CGPA",
    kind="reg"
)
```

---


````python
sns.jointplot(data = tips, x = 'total_bill', y = 'tip')
````

````python
sns.jointplot(data = tips, x = 'total_bill', y = 'tip', kind = 'reg')
````

````python
sns.jointplot(data = tips, x = 'total_bill', y = 'tip', kind='hex')
````

````python
sns.jointplot(data = tips, x = 'total_bill', y = 'tip', kind = 'kde')
````

# 3. PairGrid
---

## Definition

Advanced customizable version of Pair Plot.

## Syntax

```python
g = sns.PairGrid(df)
```

## Example

```python
g = sns.PairGrid(df)

g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

## Advantage

Full control over each section of the grid.

---

````python
g = sns.PairGrid(tips)
g.map_diag(sns.histplot)
````

# 4. JointGrid
---

## Definition

Advanced customizable version of Joint Plot.

## Example

```python
g = sns.JointGrid(
    data=df,
    x="Age",
    y="CGPA"
)

g.plot(
    sns.scatterplot,
    sns.histplot
)
```

## Advantage

Custom visualization for joint and marginal distributions.

---

````python
g = sns.JointGrid(data = student_data, x='Age', y = 'CGPA')
g.plot(
    sns.scatterplot,
    sns.histplot
)
````

# 5. FacetGrid
---

## Definition

Creates multiple subplots based on categorical values.

## Syntax

```python
g = sns.FacetGrid(
    data=df,
    col="Gender"
)

g.map(
    sns.histplot,
    "Age"
)
```

## Output

```text
Male Plot     Female Plot
```

## Use Cases

- Category-wise analysis
- Multiple subgroup comparison

---


````python
g = sns.FacetGrid(data = student_data, col = 'Gender')
g.map(
    sns.scatterplot,
    'Age'
)
````

# `Summary`

| Plot | Purpose |
|--------|----------|
| Pair Plot | Relationships among multiple numerical variables |
| Joint Plot | Relationship between two variables + distributions |
| PairGrid | Custom Pair Plot |
| JointGrid | Custom Joint Plot |
| FacetGrid | Multiple plots by category |

# Interview Question

### What are Multi Plots in Seaborn?

Multi Plots are visualization techniques used to analyze multiple variables simultaneously. Seaborn provides `pairplot()`, `jointplot()`, `PairGrid`, `JointGrid`, and `FacetGrid` to explore relationships, distributions, correlations, and category-wise patterns in datasets.
