# **`Seaborn Overview`**
---

## `What is Seaborn?`

- Seaborn is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for creating attractive, informative, and statistical visualizations with less code.

- It is especially useful during Exploratory Data Analysis (EDA) because it can quickly visualize relationships, distributions, patterns, and trends in data.

### `Definition`: Seaborn is a statistical data visualization library that helps create beautiful and informative graphs with minimal code.
---

## `Why was Seaborn Created?`

- Matplotlib is a powerful visualization library, but creating attractive and complex statistical charts often requires a lot of code.

- Seaborn was developed to:

    - Reduce coding effort
    - Improve graph aesthetics (appearance)
    - Simplify statistical visualizations
    - Work directly with Pandas DataFrames
    - Provide better default styles and themes

---

## `Matplotlib vs Seaborn`

| Feature            | Matplotlib                 | Seaborn             |
| ------------------ | -------------------------- | ------------------- |
| Level              | Low-Level Library          | High-Level Library  |
| Ease of Use        | More Code Required         | Less Code Required  |
| Default Design     | Basic                      | Attractive & Modern |
| Statistical Charts | Difficult                  | Easy                |
| DataFrame Support  | Limited                    | Excellent           |
| Built on           | Core Visualization Library | Built on Matplotlib |
| Learning Curve     | Moderate                   | Easy                |


---


## `Why Use Seaborn Over Matplotlib?`

### 1. `Better Looking Graphs by Default`
- Matplotlib ---> Basic Graph appearance
- Seaborn: Automatically adds:

    - Better colors
    - Better grid
    - Better spacing
    - Better styling

- Result → Professional-looking graph with minimal effort.

### 2. `Direct Integration with Pandas`
- Matplotlib : `plt.plot(df['Age'], df['Salary'])`
- Seaborn: `sns.scatterplot(data=df, x='Age', y='Salary')`

- No need to separately extract columns.
- Works directly with DataFrames.

### 3. `Specialized Statistical Plots`

Seaborn provides built-in statistical charts.

Examples:

1. Distribution Analysis
    - `sns.histplot()`
    - `sns.kdeplot()`
2. Relationship Analysis
    - `sns.scatterplot()`
    - `sns.lineplot()`
3. Categorical Analysis
    - `sns.barplot()`
    - `sns.boxplot()`
    - `sns.violinplot()`
4. Correlation Analysis
    - `sns.heatmap()`

Creating these in Matplotlib usually requires much more code.

### 4. `Easy Multivariable Analysis`

- Suppose you want to analyze: Age,  Salary,  Gender together.

- Seaborn: `sns.scatterplot(data=df,x='Age',y='Salary',hue='Gender')`

- One extra parameter:- hue='Gender'

- adds color grouping automatically.

### 5. `Built-in Statistical Features`

- Seaborn can automatically:

    - Calculate averages
    - Draw confidence intervals
    - Fit regression lines
    - Show distributions

- Example: `sns.regplot(data=df,x='Age',y='Salary')`

- This automatically shows:

    - Scatter points
    - Best fit line

- without manually calculating anything.

### 6. `Easy Correlation Visualization`
- Seaborn: `sns.heatmap(df.corr(), annot=True)`
- Produces a beautiful correlation matrix in a single line.
- In Matplotlib, creating the same visualization requires significantly more code.

---

## `When to Use Matplotlib?`

Use Matplotlib when:

- Need complete customization
- Creating highly customized plots
- Building dashboards
- Fine control over every graph element
- Creating plots not directly available in Seaborn

## `When to Use Seaborn?`

Use Seaborn when:

- Performing Exploratory Data Analysis (EDA)
- Working with Pandas DataFrames
- Creating statistical visualizations
- Analyzing distributions
- Finding correlations
- Comparing categories
- Building attractive graphs quickly

---

## `Relationship Between Seaborn and Matplotlib`
```code
            Seaborn
               |
               V
          Matplotlib
               |
               V
          Graph Output
```
- Seaborn is built on top of Matplotlib.
- Seaborn internally uses Matplotlib to draw graphs.
- Any Seaborn graph can be further customized using Matplotlib functions.

### `Key Takeaway`

Matplotlib = Foundation Library

- More control
- More coding
- More customization

Seaborn = High-Level Library built on Matplotlib

- Less code
- Better visuals
- Statistical plots
- Perfect for EDA and Data Science

In real-world Data Science projects, Seaborn and Matplotlib are usually used together.

# `Types Of Plot`
---

### Seaborn Plots divided into two categories

1. `Figure-Level Functions`
2. `Axes-Level Functions`

---

## `What are Figure-Level Functions?`

Figure-level functions create and manage the entire figure automatically.

They can:

- Create multiple plots at once
- Handle subplots automatically
- Apply grouping using rows and columns
- Return a FacetGrid, PairGrid, or JointGrid object

You don't need to manually create:

- plt.figure()
- plt.subplot()

Seaborn handles it internally.

## `Common Figure-Level Functions`
| Function       | Purpose                          |
| -------------- | -------------------------------- |
| `relplot()`    | Relationship plots               |
| `displot()`    | Distribution plots               |
| `catplot()`    | Categorical plots                |
| `lmplot()`     | Regression plots                 |
| `jointplot()`  | Relationship + Distribution      |
| `pairplot()`   | Multiple numerical relationships |
| `clustermap()` | Clustered heatmap                |


---


### 1. `relplot()`

Used to visualize relationships between variables.

Internally uses:

- Scatter Plot
- Line Plot

Example: `sns.relplot(data=df,x="Age",y="Salary")`

With Scatter Plot: `sns.relplot(data=df,x="Age",y="Salary",kind="scatter")`

With Line Plot: `sns.relplot(data=df,x="Year",y="Sales",kind="line")`

### 2. `displot()`

Used for Distribution Analysis.

Internally supports:

- Histogram
- KDE Plot
- ECDF Plot

Example: `sns.displot(data=df,x="Age")`

Histogram: `sns.displot(data=df,x="Age",kind="hist")`

KDE: `sns.displot(data=df,x="Age",kind="kde")`

### 3. `catplot()`

- Used for Categorical Data Analysis.
- Supports multiple categorical charts.

Bar Plot : `sns.catplot(data=df,x="Gender",y="Salary",kind="bar")`

Box Plot: `sns.catplot(data=df,x="Gender",y="Salary",kind="box")`

Violin Plot: `sns.catplot(data=df,x="Gender",y="Salary",kind="violin")`

Count Plot: `sns.catplot(data=df,x="Gender",kind="count")`

### 4. `lmplot()`

Used to visualize:

- Scatter Plot
- Regression Line

Example: `sns.lmplot(data=df,x="Experience",y="Salary")`

Output:

Scatter Plot
```code
      +
    +   +
  +      +
 +          +
------------------
 Regression Line

```

Useful for checking relationships and trends.

### 5. `jointplot()`

Shows:

- Scatter Plot in center
- Distribution of X-axis
- Distribution of Y-axis

Example: `sns.jointplot(data=df,x="Age",y="Salary")`

Visualization:
```code
       Histogram
           ▲
           │
Histogram ◄┼► Scatter Plot
           │
           ▼
```

Useful for bivariate analysis.

### 6. `pairplot()`

Creates plots between every numerical column.

Example: `sns.pairplot(df)`

Suppose: Age, Salary, Experience

Pairplot creates:

- Age vs Salary
- Age vs Experience
- Salary vs Experience

along with distributions.

Very useful during EDA.

### 7. `clustermap()`

Creates:

- Heatmap
- Hierarchical Clustering

Example: `sns.clustermap(df.corr())`

Useful for finding similar groups of features.

---

## `Axes-Level Functions in Seaborn`

Axes-level functions work on a single Matplotlib Axes object (a single graph area).

They:

- Create one plot at a time
- Give more control over customization
- Can be placed inside subplots
- Return a Matplotlib Axes object

Example: `sns.scatterplot(data=df, x="Age", y="Salary")`   

This creates only one graph.

### `Figure-Level vs Axes-Level`
```code
Figure-Level Function
        │
        ▼
 Creates Entire Figure
        │
        ▼
 Uses Axes-Level Function Internally
        │
        ▼
 Creates Actual Plot

```

---

### `Figure-Level → Axes-Level Mapping`

| Feature                | Figure-Level | Axes-Level |
| ---------------------- | ------------ | ---------- |
| Controls Entire Figure | ✅            | ❌          |
| Supports Faceting      | ✅            | ❌          |
| Multiple Subplots      | ✅            | ❌          |
| More Flexible Layout   | ✅            | ❌          |
| Returns Grid Objects   | ✅            | ❌          |
| Works on Single Axes   | ❌            | ✅          |


### `Figure-Level → Axes-Level Mapping`

| Figure-Level | Underlying Axes-Level                                   |
| ------------ | ------------------------------------------------------- |
| `relplot()`  | `scatterplot()`, `lineplot()`                           |
| `displot()`  | `histplot()`, `kdeplot()`                               |
| `catplot()`  | `barplot()`, `boxplot()`, `violinplot()`, `countplot()` |
| `lmplot()`   | `regplot()`                                             |


---
### `Complete Mapping`

| Figure-Level Function     | Axes-Level Function |

| `relplot(kind="line")`    | `lineplot()`        |

| `catplot(kind="count")`   | `countplot()`       |
| `catplot(kind="box")`     | `boxplot()`         |

| `catplot(kind="strip")`   | `stripplot()`       |
| `catplot(kind="swarm")`   | `swarmplot()`       |
| `catplot(kind="point")`   | `pointplot()`       |
| `catplot(kind="boxen")`   | `boxenplot()`       |
| `lmplot()`                | `regplot()`         |
