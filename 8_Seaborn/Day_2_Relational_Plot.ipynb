## `Relational Plot`

---

### Definition

* A **Relational Plot** is used to analyze the relationship between **two or more numerical variables**.
* It helps us understand how changes in one numerical column affect another numerical column.
* Mainly used for **Bivariate Analysis** (**Numerical vs Numerical**).
* Helps identify:

  * Relationship
  * Correlation
  * Trends
  * Patterns

### Example
- Experience ──► Salary
- Study Hours ──► Marks
- Height ──► Weight

---

### Types of Relational Plots

#### 1. Line Plot

* Used for **Time Series Analysis**.
* Shows how a value changes with respect to time.
* X-axis → Time
* Y-axis → Variable that changes over time

Syntax: `sns.lineplot(data=df,x="Month",y="Sales")`


---

#### 2. Scatter Plot

* Used to analyze the relationship between two numerical variables.
* Shows how one numerical variable affects another.
* Helps identify correlation between variables.

 Syntax: `sns.scatterplot(data=df,x="Age",y="Salary")`


---

### Common Parameters

1.  `hue`

* Groups data using colors based on another column.

2. `style`

* Changes marker or line style based on another column.

3.  `size`

* Changes marker size based on another column.

```python
size="Experience"
```

---

## Figure-Level Function

### `sns.relplot()`

* Figure-level function for relational analysis.
* Can create both Line Plot and Scatter Plot.

1.  Line Plot: `sns.relplot(data=df,x="Month",y="Sales",kind="line")`

2. Scatter Plot: `sns.relplot(data=df,x="Age",y="Salary",kind="scatter")`

---

### Syntax
1.  Axes-Level
- Line Plot:  `sns.lineplot(data=df, x="", y="")` 
- Scatter Plot: `sns.scatterplot(data=df, x="", y="")`
  
2. Figure-Level: `sns.relplot(data=df,x="",y="",kind="line")`  kind = 'line' or 'scatter'


---
```code
Relational Plot
│
├── Used for Numerical vs Numerical Analysis
```

# `Facet Plot`
---

Facet Plot can help us divide the plot asa per the mulito;e categories of a categorical column such as let us say, we don't want to use hue and we want to divide the plot or graph as per multiple categories of a categorical column as row wise or column wise.

Facet Plot can help us separate the info Row-wise or Column-wise 

We can use Row or Column Parameter to mention that if we want to separate this info row-wise or column-wise

Facet Plot functionality is only possible in the Figure Level Plot


````python
# sns.scatterplot(data= tips, x='total_bill', y="tip", col='sex') # axes level function
sns.scatterplot(data= tips, x='total_bill', y="tip")
````

````python
sns.relplot(data= tips, x='total_bill', y="tip", row='sex')
plt.show()
````

````python
sns.relplot(data= tips, x='total_bill', y="tip", col='sex')
plt.show()
````

````python
# row --> smoker or not
# column --> Male or Female

# row 1 --> Smoker <--> Male , Smoker <--> Female
# row 2 --> Non Smoker <--> Male   Non <--> Female
````

````python
sns.relplot(data= tips, x='total_bill', y="tip", row='smoker',col='sex', kind='scatter')
plt.show()
````

````python
# col_wrap --> life expentency for a country year wise in multiple columns
# Total countries --> 12 --> 12 columns
# 12 columns --> 1st row --> 3 Columns
# 2nd --> 3 columns -
````

````python
sns.relplot(data= gaps, x='lifeExp', y="pop", col='year', kind='scatter')
plt.show()
````

````python
sns.relplot(data= gaps, x='lifeExp', y="pop", col='year', kind='scatter', col_wrap=3)
plt.show()
````
