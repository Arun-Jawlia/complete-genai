# Introduction

In the previous topic, we learned that **Batch Normalization** normalizes the output (activation) of hidden layers.

Now let's understand **why it is needed**, **how it actually works inside a neural network**, and **why modern GenAI models like GPT do not use Batch Normalization**.

---

# Why Normalization is Important?

Suppose we have two features.

| Feature | Value |
| ------- | ----- |
| CGPA    | 9.2   |
| IQ      | 145   |

These features have different numerical scales.

After **Standardization**,

```text
CGPA

↓

0.65

----------------------

IQ

↓

0.48
```

Now both features have a similar scale.

This makes optimization easier.

---

# Effect on Learning Rate

Recall the Gradient Descent equation.


$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\frac{\partial L}{\partial W}
$$

where

* (W) = Weight
* (\eta) = Learning Rate
* (L) = Loss Function

---

# What Happens Without Normalization?

Suppose

```text
Feature 1 = 10

Feature 2 = 1,000,000
```

The optimization landscape becomes poorly conditioned.

```text
Large Feature Difference

↓

Unstable Gradients

↓

Need Small Learning Rate

↓

Slow Optimization

↓

Slow Training
```

If we increase the learning rate,

```text
High Learning Rate

↓

Optimization Overshoots

↓

Training Becomes Unstable
```

---

# With Normalized Data

```text
All Features Similar Scale

↓

Stable Gradients

↓

Larger Learning Rate Possible

↓

Faster Convergence

↓

Reach Minimum Faster
```

> **Important:** Normalization does **not guarantee** reaching the global minimum. It improves optimization and often helps the optimizer converge faster to a good minimum.

---

# Real-Life Analogy

Imagine driving a car.

### Without Normalization

```text
Road Full of Bumps

↓

Drive Slowly

↓

Long Time to Reach Destination
```

### With Normalization

```text
Smooth Highway

↓

Drive Faster

↓

Reach Destination Earlier
```

Normalization makes the optimization landscape smoother for Gradient Descent.

# Why Batch Normalization Was Introduced?

Data Scaling only normalizes the **input**.

```text
Raw Data

↓

Standardization

↓

Neural Network
```

But what happens **inside** the network?

Every hidden layer produces outputs.

These outputs become inputs to the next layer.

---

# Example Neural Network

```text
Input Layer

     │
     ▼

Hidden Layer 1

     │
     ▼

Hidden Layer 2

     │
     ▼

Output Layer
```

Suppose Hidden Layer 1 contains two neurons.

```text
          x₁      x₂

           │      │
       ┌───┴──────┴───┐
       ▼              ▼

     Perceptron    Perceptron

       │              │

      o₁             o₂

       └──────┬───────┘
              ▼

        Next Hidden Layer
```

---

# Output of a Perceptron

For neuron 1

$$
z_{11}
=
x_1w_1
+
x_2w_2
+
b
$$


Activation

$$
o_{11}
=
g(z_{11})
$$


Similarly,

for neuron 2,

$$
z_{12}
=
x_1w_3
+
x_2w_4
+
b
$$


$$
o_{12}
=
g(z_{12})
$$

where

* (g(\cdot)) is the activation function (ReLU, Sigmoid, etc.)

# What Happens During Training?

Training follows this cycle.

```text
Forward Pass

↓

Prediction

↓

Loss

↓

Backpropagation

↓

Weights Updated
```

After every weight update,

```text
Weights Change

↓

z Values Change

↓

Activation Values Change

↓

Next Layer Receives New Distribution
```

This happens after every mini-batch.

---

# Internal Covariate Shift

```text
Weight Update

↓

New Activation Values

↓

Different Input Distribution

↓

Next Layer Must Readapt

↓

Training Slows Down
```

This changing distribution of hidden activations is called **Internal Covariate Shift**.

---

# How Batch Normalization Solves This

Suppose

```text
Mini Batch Size = 32
```

For every neuron,

BatchNorm computes

```text
Mini Batch

↓

Compute Mean (μ)

↓

Compute Variance (σ²)

↓

Normalize Activations

↓

Mean ≈ 0

Variance ≈ 1
```

---

# Step 1 — Compute Mean

$$
\mu_B
=
\frac{1}{m}
\sum_{i=1}^{m}
x_i
$$

---

# Step 2 — Compute Variance

$$
\sigma_B^2
=
\frac{1}{m}
\sum_{i=1}^{m}
(x_i-\mu_B)^2
$$


---

# Step 3 — Normalize

$$
\hat{x_i}
=
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\varepsilon}}
$$


After normalization,

```text
Mean ≈ 0

Variance ≈ 1
```

Now the activations become much more stable.

---

# Step 4 — Scale and Shift

If we always forced

```text
Mean = 0

Variance = 1
```

the network would lose flexibility.

Therefore,

BatchNorm introduces two learnable parameters.

$$
y_i
=
\gamma
\hat{x_i}
+
\beta
$$


where

* (\gamma) = Scale
* (\beta) = Shift

These parameters are learned during training.

---

# Complete BatchNorm Pipeline

```text
Mini Batch

↓

Dense Layer

↓

Linear Output (z)

↓

Activation

↓

Batch Normalization

↓

Scale (γ)

↓

Shift (β)

↓

Next Layer
```

> **Note:** In many modern implementations, Batch Normalization is placed **between the linear layer and the activation function**:
>
> **Dense → BatchNorm → ReLU**. Some architectures use different ordering depending on the design.

---

Without BatchNorm

```text
Weights Update

↓

Activations Keep Changing

↓

Training Unstable

↓

Small Learning Rate

↓

Slow Convergence
```

With BatchNorm

```text
Normalized Activations

↓

Stable Distribution

↓

Higher Learning Rate Possible

↓

Faster Convergence

↓

Stable Training
```

---

# Does BatchNorm Reach the Global Minimum?

Not necessarily.

BatchNorm

✔ Makes optimization easier.

✔ Allows larger learning rates.

✔ Speeds up convergence.

❌ Does **not** guarantee reaching the global minimum because deep learning optimization landscapes are highly non-convex.

---

# Is BatchNorm Applied to Every Layer?

No.

```text
Input

↓

Dense

↓

BatchNorm ✓

↓

ReLU

↓

Dense

↓

ReLU

↓

Dense

↓

BatchNorm ✓

↓

Output
```

The architecture designer decides where BatchNorm is beneficial.

---

# BatchNorm in CNN

```text
Convolution

↓

BatchNorm

↓

ReLU

↓

Pooling
```

This is one of the most common CNN blocks.

---

# Why Doesn't GPT Use BatchNorm?

Modern LLMs such as

* GPT
* LLaMA
* Gemini
* Claude

generally **do not use Batch Normalization**.

Instead,

they use

```text
Layer Normalization

or

RMSNorm
```

---

# Why?

BatchNorm depends on

```text
Mini Batch Mean

Mini Batch Variance
```

However,

Transformers process

* Variable sequence lengths
* Autoregressive generation (often one token at a time during inference)
* Different batch sizes between training and inference

This makes BatchNorm unsuitable.

---

# LayerNorm

Instead of normalizing across the batch,

LayerNorm normalizes each sample independently.

```text
One Sentence

↓

LayerNorm

↓

Transformer Block

↓

Next Block
```

Therefore,

LayerNorm works perfectly for

* GPT
* LLaMA
* Gemini
* Claude

---

# BatchNorm vs LayerNorm

| Batch Normalization                    | Layer Normalization              |
| -------------------------------------- | -------------------------------- |
| Uses batch statistics                  | Uses statistics from each sample |
| Depends on mini-batch                  | Independent of batch size        |
| Common in ANN and CNN                  | Common in Transformers           |
| Difficult for autoregressive inference | Works well for sequence models   |

# Interview Tips ⭐

* BatchNorm normalizes **hidden layer activations**, not the input data.
* It computes **mean and variance from the current mini-batch**.
* The normalized activations are then scaled and shifted using **γ (gamma)** and **β (beta)**.
* BatchNorm allows **higher learning rates** and generally speeds up training.
* It helps reduce the effects of **internal covariate shift** and improves gradient flow.
* CNNs and ANNs commonly use BatchNorm.
* Transformers and LLMs use **LayerNorm** or **RMSNorm** instead.

# 2-Minute Revision

Batch Normalization normalizes the activations of hidden layers using the mean and variance of the current mini-batch. Since weight updates continuously change these activations, the next layer would otherwise receive inputs with shifting distributions, making training unstable. BatchNorm stabilizes these activations, allowing the use of larger learning rates and enabling faster convergence. After normalization, learnable parameters γ and β restore flexibility. While BatchNorm is widely used in ANNs and CNNs, modern GenAI models such as GPT, LLaMA, Gemini, and Claude use Layer Normalization or RMSNorm because they operate on variable-length sequences and do not rely on batch statistics.
