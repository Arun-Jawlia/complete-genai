# Introduction

As Deep Neural Networks become deeper, training becomes more difficult.

During training, the weights of the network are continuously updated. Because of this, the outputs (activations) of one layer keep changing, making it difficult for the next layer to learn efficiently.

**Batch Normalization (BatchNorm)** solves this problem by normalizing the outputs of a layer, making training **faster, more stable, and more reliable**.

---

# Recap: Data Scaling vs Batch Normalization

Before learning Batch Normalization, let's recall **Data Scaling**.

### Data Scaling

Data Scaling is applied **before** the data enters the neural network.

```text
Raw Data

↓

Standardization

↓

Mean = 0

Std = 1

↓

Neural Network
```

It normalizes the **input features**.

---

### Batch Normalization

Batch Normalization is applied **inside the neural network**.

```text
Input

↓

Hidden Layer

↓

Batch Normalization

↓

Activation Function

↓

Next Layer
```

Instead of normalizing the input,

it normalizes the **activations (outputs) of hidden layers**.

---

# Definition

> **Batch Normalization is a technique that normalizes the outputs (activations) of a layer using the mean and variance of the current mini-batch, making training faster and more stable.**

---

# Why Do We Need Batch Normalization?

During training,

weights keep changing.

```text
Epoch 1

↓

Weights Updated

↓

Layer Output Changes

↓

Next Layer Receives Different Distribution
```

Next epoch

```text
Weights Updated Again

↓

Layer Output Changes Again

↓

Learning Changes Again
```

The distribution keeps shifting throughout training.

This phenomenon is called **Internal Covariate Shift**.

---

# What is Internal Covariate Shift?

## Definition

Internal Covariate Shift is the continuous change in the distribution of hidden layer activations during training because the parameters of previous layers are constantly updated.

```text
Weight Update

↓

Hidden Layer Output Changes

↓

Next Layer Receives Different Distribution

↓

Needs to Readapt

↓

Training Becomes Slower
```

# How Batch Normalization Works

Suppose we are using

```text
Mini Batch Size = 32
```

During one training step,

the network processes 32 samples.

```text
Mini Batch (32 Samples)

↓

Compute Mean (μ)

↓

Compute Variance (σ²)

↓

Normalize Activations

↓

Scale (γ)

↓

Shift (β)

↓

Output
```

---

# Step 1 — Compute Mean

For each feature (or channel),

compute the batch mean.


$$
\mu_B
=
\frac{1}{m}
\sum_{i=1}^{m}
x_i
$$


where

* (m) = Batch Size
* (x_i) = Activation value

---

# Step 2 — Compute Variance

$$
\sigma_B^2
=
\frac{1}{m}
\sum_{i=1}^{m}
(x_i-\mu_B)^2
$$

---

# Step 3 — Normalize

Each activation becomes

$$
\hat{x_i}
=
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\varepsilon}}
$$

where

* (\varepsilon) is a tiny constant added for numerical stability.

Now

```text
Mean ≈ 0

Variance ≈ 1
```

---

# Step 4 — Scale and Shift

Instead of forcing every layer to always have mean 0 and variance 1,

BatchNorm introduces two learnable parameters.

$$
y_i
=
\gamma \hat{x_i}
+
\beta
$$

where

* (\gamma) = Learnable Scale
* (\beta) = Learnable Shift

This allows the network to learn the most useful distribution while still benefiting from normalization.

---

# Complete BatchNorm Pipeline

```text
Mini Batch

↓

Compute Mean

↓

Compute Variance

↓

Normalize

↓

Scale (γ)

↓

Shift (β)

↓

Activation Function

↓

Next Layer
```

---

# Where is Batch Normalization Applied?

Most commonly,

```text
Dense Layer

↓

Batch Normalization

↓

ReLU

↓

Next Layer
```

or

```text
Convolution Layer

↓

Batch Normalization

↓

ReLU
```

Modern implementations usually apply BatchNorm **before** the activation function (e.g., ReLU), although some architectures use different ordering.

# Why Batch Normalization Improves Training

## 1. Speeds Up Training

Stable activations help Gradient Descent converge faster.

```text
Stable Activations

↓

Faster Optimization

↓

Faster Convergence
```

---

## 2. Reduces Vanishing Gradient

Keeping activations in a reasonable range improves gradient flow through deep networks.

```text
Normalized Activations

↓

Better Gradient Flow

↓

Deep Networks Train More Easily
```

> Batch Normalization helps mitigate the vanishing gradient problem, but it is **not a complete solution**. Activation functions (e.g., ReLU) and residual connections also play important roles.

---

## 3. Allows Higher Learning Rates

Without BatchNorm,

a high learning rate may cause unstable training.

With BatchNorm,

the optimization process is generally more stable, so larger learning rates often become practical.

```text
BatchNorm

↓

Stable Training

↓

Higher Learning Rate Possible

↓

Faster Learning
```

---

## 4. Stabilizes Deep Networks

```text
Normalized Activations

↓

Less Internal Covariate Shift

↓

Stable Training

↓

Very Deep Networks Become Practical
```

# Batch Normalization vs Dropout

| Batch Normalization                                                 | Dropout                              |
| ------------------------------------------------------------------- | ------------------------------------ |
| Stabilizes training                                                 | Reduces overfitting                  |
| Normalizes activations                                              | Randomly disables neurons            |
| Speeds up convergence                                               | Does not primarily speed up training |
| Active during training and uses running statistics during inference | Active only during training          |

---

# Batch Normalization in GenAI

Different deep learning architectures use different normalization methods.

| Architecture                               | Normalization Method           |
| ------------------------------------------ | ------------------------------ |
| ANN                                        | Batch Normalization            |
| CNN                                        | Batch Normalization            |
| Transformer                                | Layer Normalization            |
| Large Language Models (GPT, LLaMA, Gemini) | Layer Normalization or RMSNorm |

---

# Why Don't Transformers Use Batch Normalization?

BatchNorm depends on **batch statistics**.

Transformers process

* Variable sequence lengths
* Autoregressive generation
* Different batch sizes during training and inference

Layer Normalization works independently for each sample,

making it more suitable for Transformers and LLMs.

---

# BatchNorm vs LayerNorm

| Batch Normalization          | Layer Normalization                |
| ---------------------------- | ---------------------------------- |
| Uses batch mean and variance | Uses statistics within each sample |
| Depends on batch size        | Independent of batch size          |
| Common in CNNs and ANNs      | Common in Transformers and LLMs    |
| Faster CNN training          | Stable sequence modeling           |

# Interview Tips ⭐

* Batch Normalization normalizes **hidden layer activations**, not the input data.
* It uses the **mean and variance of the current mini-batch** during training.
* It includes two learnable parameters: **γ (scale)** and **β (shift)**.
* BatchNorm speeds up training and stabilizes optimization.
* It helps mitigate vanishing gradients.
* CNNs and ANNs commonly use BatchNorm.
* Transformers and LLMs generally use **Layer Normalization** (or RMSNorm) instead.

---

# Key Takeaways

* Data Scaling normalizes the **input**; BatchNorm normalizes **hidden layer activations**.
* BatchNorm computes mean and variance for each mini-batch.
* It normalizes, then scales and shifts activations using learnable parameters.
* BatchNorm speeds up training, improves stability, and supports deeper networks.
* Modern LLMs typically replace BatchNorm with LayerNorm or RMSNorm because they do not depend on batch statistics.

# Interview Questions

### 1. What is Batch Normalization?

**Answer:**
Batch Normalization is a technique that normalizes the activations of a layer using the mean and variance of the current mini-batch, making training faster and more stable.

---

### 2. What problem does BatchNorm solve?

**Answer:**
It reduces internal covariate shift, stabilizes activations, speeds up training, and improves gradient flow.

---

### 3. Where is BatchNorm applied?

**Answer:**
Typically after a Dense or Convolution layer and before the activation function.

---

### 4. Why are γ (gamma) and β (beta) needed?

**Answer:**
Normalization alone may restrict the network. Gamma and beta allow the model to learn the optimal scale and shift for the normalized activations.

---

### 5. Why do Transformers use LayerNorm instead of BatchNorm?

**Answer:**
Because LayerNorm works independently for each sample and does not rely on batch statistics, making it suitable for variable sequence lengths and autoregressive generation.
