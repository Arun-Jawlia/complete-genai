# Introduction

One of the biggest challenges in Deep Learning is **Overfitting**.

A model that performs extremely well on the training dataset may fail to perform well on new, unseen data.

One of the most effective techniques to reduce overfitting is **Regularization**.

---

# Bias and Variance

Before understanding regularization, we need to understand **Bias** and **Variance**.

```text
                 Machine Learning Error

                         │
          ┌──────────────┴──────────────┐
          ▼                             ▼
       Bias                        Variance
```

### Bias

Bias measures how well the model fits the **training data**.

High bias means

```text
Model

↓

Too Simple

↓

Cannot Learn Patterns

↓

Underfitting
```

---

### Variance

Variance measures how well the model generalizes to **unseen (validation/test) data**.

High variance means

```text
Model

↓

Memorizes Training Data

↓

Fails on New Data

↓

Overfitting
```

---

# Goal

```text
Low Bias

+

Low Variance

↓

Best Generalization
```

---

# Bias-Variance Tradeoff

```text
Model Complexity

Low
│
│   High Bias
│   Low Variance
│
├───────────────────────────────►
│
│                Best Model
│
│
│                     High Variance
│                     Low Bias
│
High
```

A good model balances both bias and variance.

---

# Why Does Overfitting Happen?

Deep Neural Networks contain

* Many hidden layers
* Millions/Billions of parameters

```text
Input

↓

Hidden Layer

↓

Hidden Layer

↓

Hidden Layer

↓

Output
```

These models are powerful enough to memorize the training data.

---

# Ways to Reduce Overfitting

```text
                Overfitting

                     │
     ┌───────────────┼────────────────┐
     ▼               ▼                ▼
 More Data     Reduce Complexity   Regularization
     │               │                │
     ▼               ▼                ▼
Data Augmentation  Dropout      L1 / L2
                   Early Stopping
```

---

# What is Regularization?

## Definition

Regularization is a technique that **adds a penalty to the loss function** to discourage overly complex models.

Instead of allowing the model to use extremely large weights,

regularization encourages **smaller and more stable weights**.

---

# Simple Definition

> **Regularization tells the model: "Don't use extremely large weights unless they are truly necessary."**

---

# Why Large Weights Are a Problem?

Suppose

```text
Feature

↓

Weight = 250
```

A tiny change in the input can produce a very large change in the output.

This makes the model highly sensitive to the training data and more likely to overfit.

---

# Without Regularization

```text
Large Weights

↓

Very Complex Decision Boundary

↓

Memorizes Training Data

↓

Overfitting
```

---

# With Regularization

```text
Small Weights

↓

Simpler Model

↓

Learns General Patterns

↓

Better Generalization
```

---

# Cost Function with Regularization

Normally,

```text
Total Cost = Loss
```

With regularization,

$$
\boxed{
\text{Total Cost}
=
\text{Loss}
+
\lambda \times \text{Penalty}
}
$$

Where

* **Loss** → Prediction error
* **Penalty** → Complexity penalty
* **λ (Lambda)** → Regularization strength

---

# Understanding Lambda (λ)

Lambda controls **how much penalty** is applied.

```text
Small λ

↓

Small Penalty

↓

Model Focuses More on Fitting Data

------------------------

Large λ

↓

Large Penalty

↓

Model Keeps Weights Small
```

> Choosing λ is a hyperparameter tuning problem.

---

# Types of Regularization

```text
Regularization

│

├── L2 Regularization (Weight Decay)

├── L1 Regularization (Lasso)

└── Elastic Net
```

---

# 1. L2 Regularization (Weight Decay)

## Definition

L2 Regularization adds the **sum of squared weights** to the loss function.

Formula

$$
\boxed{
\text{Penalty}
=
\lambda
\sum_{j=1}^{n}
w_j^2
}
$$

$$
\boxed{
J(W)
=
\text{Loss}
+
\lambda
\sum_{j=1}^{n}
w_j^2
}
$$
---

# Why Is It Called Weight Decay?

During optimization,

large weights receive a larger penalty.

Example

```text
Weight = 10

↓

Penalty = 100

----------------------

Weight = 2

↓

Penalty = 4
```

Large weights are pushed toward smaller values.

Hence,

the weights gradually **decay** during training.

---

# Visualization

```text
Large Weight

↓

Penalty

↓

Weight Reduced

↓

Simpler Model

↓

Better Generalization
```

---

# Advantages of L2 Regularization

✅ Keeps weights small and smooth.

✅ Reduces overfitting.

✅ Improves generalization.

✅ Most widely used regularization technique in Deep Learning.

> L2 (Weight Decay) is commonly used in modern deep learning, including many computer vision and language models.

---

# 2. L1 Regularization (Lasso)

## Definition

L1 Regularization adds the **sum of the absolute values of the weights**.

Formula

$$
\boxed{
\text{Penalty}
=
\lambda
\sum_{j=1}^{n}
|w_j|
}
$$


$$
\boxed{
J(W)
=
\text{Loss}
+
\lambda
\sum_{j=1}^{n}
|w_j|
}
$$

---

# Why L1 Is Different

L1 pushes many weights **exactly to zero**.

Example

Before

```text
0.8

0.4

0.2

0.01

0.0001
```

After L1

```text
0.8

0.4

0

0

0
```

---

# Sparsity

When many weights become zero,

the model becomes **sparse**.

```text
Many Weights

↓

Some Become Zero

↓

Simpler Model

↓

Feature Selection
```

This is why L1 is popular in traditional machine learning for **feature selection**.

---

# L2 vs L1

| L2 Regularization            | L1 Regularization                                   |
| ---------------------------- | --------------------------------------------------- |
| Squares the weights          | Uses absolute weights                               |
| Keeps weights small          | Pushes many weights to exactly zero                 |
| Creates smooth models        | Creates sparse models                               |
| Most common in Deep Learning | More common for feature selection and sparse models |

---

# 3. Elastic Net

Elastic Net combines both L1 and L2 regularization.

Formula

$$
\boxed{
J(W)
=
\text{Loss}
+
\lambda_1
\sum_{j=1}^{n}
|w_j|
+
\lambda_2
\sum_{j=1}^{n}
w_j^2
}
$$

It combines

* Feature selection from L1
* Weight smoothing from L2

Elastic Net is often useful when features are highly correlated.

---

# Regularization vs Dropout

| Regularization                        | Dropout                       |
| ------------------------------------- | ----------------------------- |
| Penalizes large weights               | Randomly disables neurons     |
| Keeps model simple                    | Prevents neuron co-adaptation |
| Active through modified loss function | Active only during training   |
| L1 / L2 / Elastic Net                 | Dropout Rate (0.2, 0.5, etc.) |

# 2. Interview Questions & Answers

## Beginner Questions

### 1. What is regularization?

**Answer:**
Regularization is a technique that adds a penalty to the loss function to discourage overly large weights and reduce overfitting.

---

### 2. Why is regularization used?

**Answer:**
It helps the model generalize better by preventing it from memorizing the training data.

---

### 3. What is the regularized cost function?

**Answer:**

[
\boxed{
\text{Total Cost}
=================

\text{Loss}
+
\lambda \times \text{Penalty}
}
]

---

### 4. What does λ (lambda) represent?

**Answer:**
Lambda is the **regularization strength**. A larger λ applies a stronger penalty to large weights.

---

## Intermediate Questions

### 5. Why is L2 called Weight Decay?

**Answer:**
Because it discourages large weights by adding a penalty proportional to the square of the weights, causing them to gradually shrink during training.

---

### 6. What is the main difference between L1 and L2?

**Answer:**

* **L1** pushes many weights exactly to zero, creating sparse models.
* **L2** keeps weights small and smooth without usually making them zero.

---

### 7. Which regularization technique is more common in deep learning?

**Answer:**
**L2 Regularization (Weight Decay)** is the most commonly used regularization technique in modern deep learning.

---

## Scenario-Based Questions

### 8. Your training accuracy is 99%, but validation accuracy is 70%. What would you try?

**Answer:**
The model is likely overfitting. I would consider L2 regularization, Dropout, Early Stopping, data augmentation, or reducing model complexity.

---

### 9. When would L1 Regularization be especially useful?

**Answer:**
When feature selection is important because L1 can drive some weights exactly to zero, effectively removing less useful features.

---

### 10. Does regularization always increase training accuracy?

**Answer:**
No. It often slightly **reduces training accuracy** while **improving validation/test performance**, which indicates better generalization.

# 4. 2-Minute Revision

Regularization is a technique used to reduce overfitting by adding a penalty term to the loss function. Instead of allowing the model to use extremely large weights, it encourages simpler solutions that generalize better. The two most common methods are **L2 Regularization (Weight Decay)**, which keeps weights small and smooth, and **L1 Regularization**, which pushes many weights to zero, creating sparse models useful for feature selection. Elastic Net combines both approaches. The amount of regularization is controlled by **λ (lambda)**, a hyperparameter that balances fitting the training data against keeping the model simple.
