# `Types of Optimizers in Deep Learning`
---

## Definition

An **Optimizer** is an algorithm that updates the weights and biases of a neural network during training to minimize the loss function.

Different optimizers use different strategies to update the weights, resulting in varying convergence speeds, stability, and performance.

---

# Why Are There Different Optimizers?

No single optimizer performs best for every problem.

Different datasets and neural network architectures require different optimization strategies.

An ideal optimizer should:

- Reach the minimum loss quickly.
- Avoid oscillations.
- Escape saddle points and shallow local minima.
- Require minimal manual tuning.
- Converge efficiently.

---

# Classification of Optimizers

Deep Learning optimizers can be broadly classified into two categories.

```
Optimizers
│
├── 1. Non-Adaptive Optimizers
│      ├── Batch Gradient Descent
│      ├── Stochastic Gradient Descent (SGD)
│      ├── Mini-Batch Gradient Descent
│      └── Momentum
│
└── 2. Adaptive Optimizers
       ├── AdaGrad
       ├── AdaDelta
       ├── RMSProp
       ├── Adam
       ├── AdamW
       ├── AdaMax
       └── Nadam
```

---

# 1. Non-Adaptive Optimizers

## Definition

Non-adaptive optimizers use the **same learning rate for all parameters** throughout training.

The learning rate remains fixed unless manually changed.

### Characteristics

- Same learning rate for every weight
- Simpler algorithms
- Require manual learning rate tuning
- Less computational overhead

---

## A. Batch Gradient Descent

### Definition

Uses the **entire training dataset** to calculate the gradient before updating the weights.

### Advantages

- Stable updates
- Accurate gradient estimation

### Disadvantages

- Slow for large datasets
- High memory usage

### Best Used When

- Small datasets
- Research experiments

---

## B. Stochastic Gradient Descent (SGD)

### Definition

Updates the weights after processing **one training example at a time**.

### Advantages

- Faster updates
- Lower memory usage
- Can escape shallow local minima

### Disadvantages

- Noisy updates
- Oscillates frequently
- Less stable

### Best Used When

- Very large datasets
- Online learning

---

## C. Mini-Batch Gradient Descent

### Definition

Updates weights after processing a **small batch of samples** (commonly 32, 64, or 128).

### Advantages

- Faster than Batch GD
- More stable than SGD
- Efficient GPU utilization

### Disadvantages

- Batch size selection affects performance

### Best Used When

- Most Deep Learning applications

---

## D. Momentum Optimizer

### Definition

Momentum improves Gradient Descent by remembering previous gradients.

Instead of moving only in the current gradient direction, it also considers the previous update direction.

### Advantages

- Faster convergence
- Reduces oscillations
- Escapes shallow valleys more easily

### Disadvantages

- Introduces an additional hyperparameter (momentum coefficient)

---

# 2. Adaptive Optimizers

## Definition

Adaptive optimizers automatically adjust the learning rate for each parameter during training.

Different weights can have different effective learning rates.

### Characteristics

- Adaptive learning rates
- Faster convergence
- Less manual tuning
- Better suited for deep neural networks

---

## A. AdaGrad (Adaptive Gradient)

### Definition

AdaGrad adapts the learning rate based on the history of gradients.

Frequently updated parameters receive smaller learning rates, while infrequently updated parameters receive larger learning rates.

### Advantages

- Good for sparse datasets
- Automatic learning rate adjustment

### Disadvantages

- Learning rate keeps decreasing
- Training may stop prematurely

### Applications

- NLP
- Text classification
- Sparse feature problems

---

## B. AdaDelta

### Definition

AdaDelta improves AdaGrad by preventing the learning rate from shrinking indefinitely.

It uses a moving average of previous gradients instead of accumulating all past gradients.

### Advantages

- Eliminates aggressive learning rate decay
- More stable than AdaGrad

### Disadvantages

- Less commonly used today

---

## C. RMSProp (Root Mean Square Propagation)

### Definition

RMSProp modifies AdaGrad by using an exponentially weighted moving average of squared gradients.

This prevents the learning rate from becoming excessively small.

### Advantages

- Faster convergence
- Handles non-stationary objectives
- Stable learning

### Applications

- Recurrent Neural Networks (RNNs)
- Time-series problems

---

## D. Adam (Adaptive Moment Estimation)

### Definition

Adam combines:

- Momentum
- RMSProp

It keeps track of:

- First moment (mean of gradients)
- Second moment (variance of gradients)

### Advantages

- Fast convergence
- Adaptive learning rates
- Stable updates
- Works well for most Deep Learning tasks

### Disadvantages

- May generalize worse than SGD in some cases
- Slightly higher memory usage

### Applications

- CNNs
- RNNs
- Transformers
- Large-scale Deep Learning

---

## E. AdamW

### Definition

AdamW improves Adam by applying **weight decay (L2 regularization)** correctly.

### Advantages

- Better regularization
- Improved generalization
- Preferred for Transformer models

### Applications

- BERT
- GPT
- Vision Transformers

---

## F. AdaMax

### Definition

AdaMax is a variant of Adam that uses the **infinity norm (L∞ norm)** instead of the L2 norm.

### Advantages

- More stable for certain optimization problems
- Handles large gradients effectively

---

## G. Nadam

### Definition

Nadam combines:

- Adam
- Nesterov Accelerated Gradient (NAG)

### Advantages

- Faster convergence
- Better predictive performance in some tasks

# Practical Example

Suppose you're training an image classification model:

- **Batch GD:** Slow but stable.
- **SGD:** Fast but noisy.
- **Mini-Batch GD:** Balanced and GPU-efficient.
- **Momentum:** Reduces oscillations.
- **AdaGrad:** Learns quickly initially but slows down too much.
- **RMSProp:** Maintains a balanced learning rate.
- **Adam:** Fast convergence with adaptive updates.
- **AdamW:** Better regularization and generalization.

---

# Interview Points

- Optimizers update model parameters to minimize loss.
- Optimizers are classified as **Non-Adaptive** and **Adaptive**.
- SGD updates weights after every sample.
- Mini-Batch GD is the most commonly used variant of Gradient Descent.
- Momentum remembers previous gradients.
- AdaGrad adapts learning rates but suffers from aggressive decay.
- RMSProp fixes AdaGrad's learning rate decay.
- Adam combines Momentum and RMSProp.
- AdamW is preferred for Transformer-based models.

---

# Key Takeaways

- Different optimizers use different strategies to update weights.
- Non-adaptive optimizers use a fixed learning rate.
- Adaptive optimizers assign different learning rates to different parameters.
- Adam is the default choice for many Deep Learning tasks.
- AdamW is widely used in modern NLP and Transformer models.
- Choosing the right optimizer can significantly improve training speed and model performance.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. What is an optimizer?

**Answer:**
An optimizer is an algorithm that updates a neural network's weights and biases to minimize the loss function during training.

---

### 2. What are the two main categories of optimizers?

**Answer:**

* Non-Adaptive Optimizers
* Adaptive Optimizers

---

### 3. What is the difference between adaptive and non-adaptive optimizers?

**Answer:**

* **Non-Adaptive:** Uses the same learning rate for all parameters.
* **Adaptive:** Adjusts the learning rate individually for each parameter.

---

### 4. Which optimizer is the most commonly used?

**Answer:**
**Adam** is the most popular optimizer because it combines adaptive learning rates with momentum, leading to fast and stable convergence.

---

### 5. Why is Mini-Batch Gradient Descent preferred over Batch Gradient Descent?

**Answer:**
Mini-Batch GD offers a good balance between computational efficiency and stable gradient updates, and it efficiently utilizes GPUs.

---

## Intermediate Questions

### 6. Why does AdaGrad slow down over time?

**Answer:**
AdaGrad accumulates squared gradients, causing the learning rate to continuously decrease. Eventually, updates become very small, slowing or stopping learning.

---

### 7. How does RMSProp improve AdaGrad?

**Answer:**
RMSProp uses an exponentially weighted moving average of squared gradients instead of accumulating all past gradients, preventing the learning rate from shrinking excessively.

---

### 8. What are the first and second moments used in Adam?

**Answer:**

* **First Moment:** Running average (mean) of gradients.
* **Second Moment:** Running average of squared gradients (variance).

---

### 9. Why is AdamW preferred for Transformers?

**Answer:**
AdamW applies weight decay correctly, improving regularization and helping Transformer models generalize better.

---

### 10. What is the role of Momentum in optimization?

**Answer:**
Momentum remembers previous updates, accelerating learning in consistent directions while reducing oscillations.

---

## Scenario-Based Questions

### Q1. Your model trains very slowly with plain Gradient Descent. What optimizer would you choose?

**Answer:**
Adam or RMSProp, because they provide adaptive learning rates and faster convergence.

---

### Q2. You are fine-tuning a BERT or GPT model. Which optimizer is recommended?

**Answer:**
AdamW, due to its proper implementation of weight decay and improved generalization.

---

### Q3. Your dataset has many sparse features (e.g., text data). Which optimizer is suitable?

**Answer:**
AdaGrad, because it increases learning rates for infrequently updated parameters, making it effective for sparse data.
