# Introduction

An **Activation Function** is one of the most important components of a Neural Network.

It gives a neural network the ability to **learn complex, non-linear relationships** from data.

Without an activation function, even a deep neural network with many layers behaves like a **single linear model**.

---

# Definition

> **An Activation Function is a mathematical function that introduces non-linearity into a neural network, enabling it to learn complex patterns from data.**

---

# Why Do We Need Activation Functions?

Consider a single neuron.

First, the neuron computes a weighted sum.


$$
z = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$


Without an activation function,

the output is simply


$$
y = z
$$

which means

$$
y = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$


This is nothing more than a **linear equation**.

---

# Linear Model

```text
Input Features

↓

Weighted Sum

↓

Output

↓

Straight Line
```

This is exactly what happens in **Linear Regression**.

---

# Example

Suppose

$$
y = 2x + 3
$$


The graph is

```text
y

│
│        /
│      /
│    /
│  /
│/
└──────────────────► x
```

No matter how many linear layers we stack,

the result is still a linear transformation.

---

# Multiple Linear Layers Still Remain Linear

Suppose

```text
Layer 1

↓

Linear

↓

Layer 2

↓

Linear

↓

Layer 3

↓

Linear
```

Mathematically,

$$
y = W_3(W_2(W_1x))
$$


This simplifies to

$$
y = Wx
$$


where

$$
W = W_3W_2W_1
$$


Therefore,

```text
Linear

↓

Linear

↓

Linear

↓

Still Linear
```

Adding more linear layers **does not increase the model's expressive power**.

# Mathematical Representation

A neuron first computes

$$
z = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

Then

the activation function is applied.

$$
a = g(z)
$$

or

$$
\boxed{
a
=
g
\left(
w_1x_1+w_2x_2+\cdots+w_nx_n+b
\right)
}
$$

where

* (x_i) = Inputs
* (w_i) = Weights
* (b) = Bias
* (z) = Linear Combination
* (g(\cdot)) = Activation Function
* (a) = Output of the neuron

# Characteristics of an Ideal Activation Function

A good activation function should satisfy several important properties.

---

# 1. Non-Linearity ⭐

The most important property.

```text
Linear Function

↓

Cannot Learn Complex Data
```

Non-linear activation

```text
Input

↓

Activation

↓

Complex Decision Boundary

↓

Deep Learning
```

Without non-linearity,

Deep Learning would not exist.

---

# 2. Differentiable

Backpropagation requires derivatives.

Gradient Descent updates weights using

$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\frac{\partial L}{\partial W}
$$


To compute gradients,

the activation function should be differentiable almost everywhere (functions like ReLU are acceptable despite having a non-differentiable point at (x=0), where a subgradient is used in practice).

```text
Activation

↓

Derivative

↓

Gradient

↓

Backpropagation

↓

Weight Update
```

---

# 3. Computationally Efficient

Activation functions are evaluated

* Millions of times
* Billions of times

during training.

Therefore,

they should be computationally inexpensive.

Example

```text
Fast Function

↓

Faster Training
```

This is one reason why **ReLU** became more popular than **Sigmoid** in hidden layers.

---

# 4. Zero-Centered (Preferred)

A zero-centered activation produces outputs roughly centered around zero.

Example

```text
Negative Values

↓

0

↓

Positive Values
```

This often leads to more balanced gradient updates.

Examples

✔ Tanh

✔ GELU (approximately zero-centered)

Sigmoid is **not** zero-centered because its outputs lie between 0 and 1.

> Zero-centered outputs are generally beneficial, but **not every successful activation function has this property**. For example, ReLU is not zero-centered yet is widely used because of its other advantages.

---

# 5. Non-Saturating

One of the biggest problems of older activation functions is **saturation**.

Example

Sigmoid

```text
Large Positive Input

↓

Output ≈ 1

↓

Gradient ≈ 0
```

Large Negative Input

```text
Output ≈ 0

↓

Gradient ≈ 0
```

When gradients become very small,

learning slows dramatically.

This contributes to the **Vanishing Gradient Problem**.

---

# Saturating vs Non-Saturating

```text
Saturating Function

↓

Gradient Near Zero

↓

Slow Learning

↓

Vanishing Gradient

────────────────────────

Non-Saturating Function

↓

Healthy Gradient

↓

Fast Learning

↓

Deep Networks
```

ReLU is considered **non-saturating for positive inputs**, which is one reason it became the default choice for hidden layers.

# Interview Tips ⭐

* Activation functions introduce **non-linearity** into neural networks.
* Without activation functions, multiple layers collapse into a single linear transformation.
* Every neuron computes **(z = Wx + b)** before applying an activation function.
* The output of a neuron is **(a = g(z))**.
* Good activation functions are generally non-linear, computationally efficient, and suitable for gradient-based optimization.
* ReLU is widely used in hidden layers, while Softmax is commonly used for multi-class output layers. Modern Transformers often use GELU or gated variants such as SwiGLU.

# 2-Minute Revision

An activation function is a mathematical function applied after the weighted sum of a neuron. Without activation functions, a neural network is equivalent to a linear model, regardless of how many layers it contains. Activation functions introduce non-linearity, allowing networks to learn complex relationships in images, speech, text, and other real-world data. An ideal activation function should be non-linear, computationally efficient, and compatible with gradient-based learning. Modern deep learning models commonly use ReLU, GELU, or gated variants such as SwiGLU depending on the architecture.
