# `Types of Activation Functions`

# Introduction

An **Activation Function** introduces **non-linearity** into a neural network, enabling it to learn complex patterns from data.

Different activation functions are used depending on the **task** and the **network architecture**.

```text
                  Activation Functions
                          │
     ┌──────────┬──────────┬─────────────┬────────────┐
     ▼          ▼          ▼             ▼
  Sigmoid      Tanh       ReLU      Modern Functions
                                        │
                              ┌─────────┴─────────┐
                              ▼                   ▼
                            GELU               Swish/SiLU
```

---

# Comparison Overview

| Activation   | Output Range | Zero Centered | Saturation              | Common Usage                 |
| ------------ | ------------ | ------------- | ----------------------- | ---------------------------- |
| Sigmoid      | (0,1)        | ❌ No          | Yes                     | Binary Classification Output |
| Tanh         | (-1,1)       | ✅ Yes         | Yes                     | Older RNNs                   |
| ReLU         | [0,∞)        | ❌ No          | Only on negative side   | Hidden Layers                |
| GELU         | Continuous   | Approximately | No practical saturation | Transformers, GPT, BERT      |
| Swish (SiLU) | Continuous   | Approximately | No practical saturation | Vision Models, LLM variants  |

---

# 1. Sigmoid Activation Function

## Definition

The **Sigmoid Activation Function** converts any real number into a value between **0 and 1**.

It is widely used whenever the output represents a **probability**.

---

# Formula


$$
\boxed{
\sigma(x)
=
\frac{1}
{1+e^{-x}}
}
$$

where

* (x) = Input
* (\sigma(x)) = Output

---

# Output Range

```text
(-∞ , +∞)

↓

Sigmoid

↓

(0 , 1)
```

---

# Graph

```text
Output
1.0 |                     ●●●●
    |                 ●●●
0.5 |------------●---------
    |        ●●
0.0 |●●●●
    +---------------------------->
           -x        0        +x
```

The output always remains between **0 and 1**.

---

# Working

Suppose

$$
x = 2
$$


Then

$$
\sigma(2)
=
\frac{1}{1+e^{-2}}
\approx
0.88
$$


The neuron outputs

```text
88% Confidence
```

---

# Real-Life Example

Email Spam Detection

```text
Email

↓

Neural Network

↓

Sigmoid

↓

0.92

↓

Spam
```

Decision Rule

```text
Probability > 0.5

↓

Spam

--------------------

Probability ≤ 0.5

↓

Not Spam
```

---

# Where is Sigmoid Used?

✅ Binary Classification Output Layer

✅ Logistic Regression

Examples

* Spam Detection
* Disease Prediction
* Fraud Detection
* Yes/No Decision

---

# Advantages of Sigmoid

✅ Output behaves like a probability.

✅ Non-linear.

✅ Differentiable.

Suitable for Backpropagation.

---

# Disadvantages of Sigmoid

### 1. Saturation

For large positive values

```text
Output ≈ 1

↓

Gradient ≈ 0
```

For large negative values

```text
Output ≈ 0

↓

Gradient ≈ 0
```

This leads to the **Vanishing Gradient Problem**.

---

### 2. Not Zero-Centered

Output

```text
0 → 1
```

instead of

```text
-1 → 1
```

This can slow optimization because gradients are less balanced.

---

# Summary

```text
Input

↓

Sigmoid

↓

0 to 1

↓

Probability

↓

Binary Classification
```

---

# 2. Tanh (Hyperbolic Tangent)

## Definition

Tanh is an improved version of Sigmoid because its output is **centered around zero**.

---

# Formula

$$
\boxed{
\tanh(x)
=
\frac{e^x-e^{-x}}
{e^x+e^{-x}}
}
$$

---

# Output Range

```text
(-∞ , +∞)

↓

Tanh

↓

(-1 , 1)
```

---

# Graph

```text
Output
 1 |                  ●●●
   |              ●●●
 0 |---------●---------
   |      ●●
-1 |●●●
   +---------------------------->
        -x      0       +x
```

---

# Working

Negative input

↓

Negative output

Positive input

↓

Positive output

Zero

↓

Zero

---

# Why is Tanh Better Than Sigmoid?

```text
Sigmoid

↓

0 to 1

↓

Not Zero Centered

----------------------

Tanh

↓

-1 to 1

↓

Zero Centered

↓

Balanced Gradients

↓

Faster Training
```

---

# Advantages

✅ Zero-centered.

✅ Stronger gradients near zero.

✅ Faster convergence than Sigmoid.

---

# Disadvantages

### Saturation

Large positive values

↓

Output ≈ 1

Large negative values

↓

Output ≈ -1

Gradient again becomes close to zero.

Therefore,

deep networks still suffer from vanishing gradients.

---

# Where is Tanh Used?

Older

* RNN
* LSTM (inside gates and cell updates)
* Legacy Neural Networks

> Modern architectures still use **tanh** inside LSTM cells, but it is much less common as the primary hidden-layer activation in feed-forward deep networks.

---

# Summary

```text
Input

↓

Tanh

↓

-1 to 1

↓

Zero Centered

↓

Better than Sigmoid

↓

Still Saturates
```

---

# 3. ReLU (Rectified Linear Unit)

## Definition

ReLU outputs the input if it is positive.

Otherwise,

it outputs zero.

---

# Formula

$$
\boxed{
f(x)
=
\max(0,x)
}
$$


---

# Piecewise Form

$$
f(x)=
\begin{cases}
0, & x<0\\
x, & x\ge0
\end{cases}
$$


---

# Output

```text
Negative Input

↓

0

----------------

Positive Input

↓

Same Value
```

---

# Graph

```text
Output

│
│         /
│       /
│     /
│   /
│ /
└──────────────►
      0
```

---

# Working

Suppose

$$
x=-4
$$

Output


$$
0
$$

Suppose


$$
x=6
$$

Output


$$
6
$$

---

# Why is ReLU So Popular?

```text
Input

↓

ReLU

↓

Healthy Gradient

↓

Fast Learning

↓

Deep Networks
```

---

# Advantages

### 1. Non-linear

Can learn complex decision boundaries.

---

### 2. Does Not Saturate for Positive Inputs

Unlike Sigmoid,

large positive values still produce meaningful gradients.

---

### 3. Computationally Cheap

Only

```text
max(0,x)
```

Very fast.

---

### 4. Faster Convergence

ReLU trains significantly faster than Sigmoid and Tanh in deep feed-forward networks.

---

# Disadvantages

### 1. Not Zero-Centered

Outputs

```text
0 → +∞
```

instead of

```text
-1 → +1
```

This is **not usually a major issue**, and Batch Normalization can help stabilize activations.

---

### 2. Unbounded Output

Large positive values remain large.

---

### 3. Dying ReLU

If a neuron's inputs become consistently negative,

```text
Output = 0

↓

Gradient = 0

↓

Neuron Stops Learning
```

This often happens because of

* Very high learning rate
* Poor weight initialization

---

# Why is it Called Dying ReLU?

```text
Negative Input

↓

Always Zero

↓

Never Updated

↓

Dead Neuron
```

---

# Where is ReLU Used?

Almost every traditional deep network

* ANN
* CNN
* Autoencoders
* Many Computer Vision models

---

# ReLU Summary

```text
Input

↓

ReLU

↓

Positive → Pass

Negative → Ignore

↓

Fast Learning
```

---

# Modern Activation Functions in GenAI

Modern Transformer-based models usually do **not** use ReLU.

Instead they use smoother activation functions.

---

# GELU (Gaussian Error Linear Unit)

Used in

* BERT
* GPT (original GPT family)
* Vision Transformers (many implementations)

Characteristics

✅ Smooth

✅ Better gradient flow

✅ Strong empirical performance in Transformers

---

# Swish / SiLU

Formula

$$
\boxed{
\mathrm{SiLU}(x)
=
x\,\sigma(x)
}
$$


Used in

* EfficientNet
* Diffusion Models (e.g., Stable Diffusion U-Net)
* Some LLM variants

Produces smoother optimization than ReLU.

> **Note:** DALL·E and Midjourney are complex systems whose exact activation functions are not fully public. Many modern diffusion architectures use **SiLU (Swish)**.

---

# SwiGLU (Modern LLMs)

Many recent LLMs

* LLaMA
* PaLM
* Gemma
* Mistral

use **SwiGLU**, a gated activation built on **SiLU/Swish**.

It generally provides better performance than ReLU or standard GELU in large Transformer feed-forward blocks.

# Interview Tips ⭐

* **Sigmoid** is mainly used in the **output layer for binary classification**.
* **Tanh** is zero-centered but still suffers from saturation.
* **ReLU** is the default hidden-layer activation for many CNNs and ANNs.
* **Dying ReLU** occurs when neurons output zero for all inputs and stop learning.
* **GELU** is widely used in Transformer architectures such as GPT and BERT.
* **SwiGLU** is commonly used in many modern large language models like LLaMA and Gemma.

# 2-Minute Revision

Activation functions introduce non-linearity into neural networks. **Sigmoid** outputs values between 0 and 1 and is primarily used for binary classification outputs. **Tanh** outputs values between -1 and 1, making it zero-centered and generally better than Sigmoid, though it still suffers from saturation. **ReLU** outputs zero for negative inputs and the input itself for positive values, making it computationally efficient and the standard choice for hidden layers in CNNs and ANNs. Modern Transformer models prefer **GELU**, while many recent large language models use **SwiGLU**, which provides smoother optimization and improved performance.
