# Models in LangChain

In LangChain, **Models** are the components that interact with AI models such as GPT, Claude, Gemini, Llama, and embedding models.

A useful mental model is:

```text
Your Application
       ↓
    LangChain
       ↓
     Models
       ↓
┌──────┼──────────┐
│      │          │
▼      ▼          ▼
Chat   LLM      Embeddings
Model  Model      Model
```

The model layer is the foundation of a LangChain application.

---

# 1. What is a Model in LangChain?

A **model** is an interface through which your application communicates with an AI model.

Instead of writing provider-specific code everywhere:

```text
Application
    ↓
OpenAI API
```

or:

```text
Application
    ↓
Anthropic API
```

LangChain provides standardized interfaces so your application can work with different model providers.

Conceptually:

```text
                LangChain
                    │
          ┌─────────┼─────────┐
          ▼         ▼         ▼
        OpenAI    Gemini    Anthropic
```

This makes it easier to switch models without rewriting your entire application.

---

# 2. Types of Models in LangChain

The three major categories you should understand are:

```text
Models
│
├── Chat Models
│
├── LLMs
│
└── Embedding Models
```

The distinction is important.

---

# 3. Chat Models

A **Chat Model** is designed to work with conversational messages.

Examples:

* GPT models
* Claude
* Gemini
* Mistral chat models

Instead of passing only a plain string, you can provide structured messages.

```text
System Message
      ↓
Human Message
      ↓
AI Message
      ↓
Human Message
```

---

## Example

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-4.1-mini"
)

response = model.invoke(
    "Explain Transformer in simple terms."
)

print(response.content)
```

The important method here is:

```python
model.invoke(...)
```

It sends the input to the model and returns the model's response.

---

# 4. Messages in Chat Models

Chat models understand different message roles.

### System

Defines the model's behavior.

```python
from langchain_core.messages import SystemMessage

SystemMessage(
    content="You are a Python teacher."
)
```

### Human

Represents the user's request.

```python
HumanMessage(
    content="Explain decorators."
)
```

### AI

Represents a previous model response.

```python
AIMessage(
    content="A decorator is..."
)
```

---

## Complete Example

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    SystemMessage,
    HumanMessage
)

model = ChatOpenAI(
    model="gpt-4.1-mini"
)

messages = [
    SystemMessage(
        content="You are a Python teacher."
    ),
    HumanMessage(
        content="Explain decorators."
    )
]

response = model.invoke(messages)

print(response.content)
```

Conceptually:

```text
System
  ↓
You are a Python teacher.

Human
  ↓
Explain decorators.

  ↓

Chat Model

  ↓

AI Response
```

# 7. Chat Model vs LLM

This distinction is important for interviews.

| Feature             | Chat Model    | LLM                                |
| ------------------- | ------------- | ---------------------------------- |

In current LangChain development, you will generally encounter **chat models more often**.

# 8. Embedding Models

Embedding models are completely different from chat models.

A chat model:

```text
Text
 ↓
Response
```

An embedding model:

```text
Text
 ↓
Vector
```

Example:

```text
"Machine Learning"
       ↓
Embedding Model
       ↓
[0.21, -0.43, 0.76, ...]
```

---

# Why Do We Need Embeddings?

Embeddings are essential for applications such as **RAG** and semantic search.

Example:

```text
Document
   ↓
Text Chunks
   ↓
Embedding Model
   ↓
Vectors
   ↓
Vector Database
```

Then:

```text
User Question
   ↓
Embedding
   ↓
Similarity Search
   ↓
Relevant Documents
```

---

# 9. Example: Embeddings in LangChain

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
)

vector = embeddings.embed_query(
    "What is machine learning?"
)

print(vector)
```

The result is a numerical vector.

# 11. Why Use LangChain Instead of Direct SDKs?

Without LangChain:

```text
Application
     ↓
OpenAI SDK
     ↓
GPT
```

If you switch to another provider:

```text
Application
     ↓
Anthropic SDK
     ↓
Claude
```

```text
Application
     ↓
LangChain Interface
     ↓
Model Provider
```

The application code can remain largely the same while the model integration changes.

# 15. Streaming

Normally:

```text
Request
   ↓
Wait
   ↓
Complete Response
```

With streaming:

```text
Request
   ↓
Token 1
Token 2
Token 3
Token 4
...
```

This is useful for chat applications because users see the response as it is generated.

Conceptually:

```python
for chunk in model.stream("Explain RAG"):
    print(chunk.content, end="")
```

# 17. Structured Output

One of the most useful model capabilities is structured output.

Instead of:

```text
Name: Arun
Age: 25
```

you want:

```json
{
  "name": "Arun",
  "age": 25
}
```

You can define a schema and ask the model to return data matching it.

Example:

```python
from pydantic import BaseModel
from langchain_openai import ChatOpenAI

class User(BaseModel):
    name: str
    age: int

model = ChatOpenAI(
    model="gpt-4.1-mini"
).with_structured_output(User)

result = model.invoke(
    "My name is Arun and I am 25 years old."
)

print(result)
```

This is particularly useful for:

* API responses
* Database insertion
* Information extraction
* Agents
* Backend applications

# 18. Model + Prompt

A very common LangChain pattern is:

```text
Prompt Template
      ↓
Chat Model
      ↓
Output Parser
```

Example:

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} for a {level}."
)

model = ChatOpenAI(
    model="gpt-4.1-mini"
)

parser = StrOutputParser()

chain = prompt | model | parser

result = chain.invoke({
    "topic": "Transformers",
    "level": "beginner"
})

print(result)
```

Pipeline:

```text
Input
  ↓
Prompt Template
  ↓
Chat Model
  ↓
Output Parser
  ↓
String
```

# 19. Model + Retriever + Prompt

This is where things become interesting for GenAI.

A RAG pipeline can look like:

```text
User Question
      ↓
Retriever
      ↓
Relevant Documents
      ↓
Prompt
      ↓
Chat Model
      ↓
Answer
```

Example:

```text
"What is the company's refund policy?"

        ↓

Retriever

        ↓

Relevant PDF chunks

        ↓

Prompt

        ↓

GPT

        ↓

Answer
```

The model itself doesn't need to contain the private company policy in its training weights.

The relevant information is retrieved at runtime.

# 20. Model + Tools

In an agent application:

```text
User
 ↓
Chat Model
 ↓
Tool Call
 ↓
External System
 ↓
Tool Result
 ↓
Chat Model
 ↓
Final Response
```

Example:

```text
User:
Calculate 157 × 48.
```

The model can decide to use:

```text
Calculator Tool
```

Then:

```text
Calculator
   ↓
7536
   ↓
Model
   ↓
"157 × 48 = 7,536."
```

---

# 21. Models Are Not the Same as Agents

This distinction is important.

### Model

```text
Input
 ↓
Model
 ↓
Output
```

### Agent

```text
User
 ↓
Agent
 ↓
Model decides action
 ↓
Tool
 ↓
Result
 ↓
Model
 ↓
Final answer
```

The **model provides intelligence**, while an **agent provides a decision/action loop around the model**.

# 22. Models in a RAG Application

Let's connect everything you've learned.

```text
                 RAG APPLICATION

                     User
                       │
                       ▼
                   Question
                       │
                       ▼
                  Embedding
                    Model
                       │
                       ▼
                 Vector Store
                       │
                       ▼
                   Retriever
                       │
                       ▼
                Relevant Chunks
                       │
                       ▼
                Prompt Template
                       │
                       ▼
                  Chat Model
                       │
                       ▼
                 Final Answer
```

Notice that **two different models** can be involved:

```text
Embedding Model
      ↓
Retrieval

Chat Model
      ↓
Generation
```

This distinction is extremely important in RAG.

# 24. Simple Mental Model

Remember:

```text
                 MODELS
                    │
          ┌─────────┴─────────┐
          │                   │
          ▼                   ▼
      Chat Model        Embedding Model
          │                   │
          ▼                   ▼
    Generate Text       Generate Vectors
          │                   │
          ▼                   ▼
      Chatbot/RAG       Vector Search
```

And the complete GenAI application:

```text
                    LangChain
                        │
        ┌───────────────┼──────────────┐
        │               │              │
     Prompt           Models         Tools
        │               │              │
        │        ┌──────┴──────┐       │
        │        │             │       │
        │      Chat        Embedding   │
        │       Model         Model    │
        │        │             │       │
        └────────┼─────────────┼───────┘
                 │             │
                 ▼             ▼
              Generate      Retrieve
                 │             │
                 └──────┬──────┘
                        ▼
                    Response
```

## Key takeaway

**Models are the core AI components in LangChain.** The two most important categories for GenAI development are **Chat Models**, which generate responses and interact with messages, and **Embedding Models**, which convert text into vectors for semantic search and RAG. LangChain gives these models a standardized interface and lets you compose them with prompts, retrievers, tools, structured output, and other application components.
