# `Detailed Notes `

# Document Loaders in LangChain

> **Document Loader = A component that loads data from different sources and converts it into a standard `Document` format that LangChain can process.**

Document loaders are usually the **first step in a RAG/Indexing pipeline**.

```text
Data Source
    ↓
Document Loader
    ↓
Document
    ↓
Text Splitter
    ↓
Embeddings
    ↓
Vector Store
    ↓
Retriever
    ↓
LLM
```

---

# 1. What Problem Do Document Loaders Solve?

In a GenAI application, your data can come from many different places:

```text
PDF
TXT
CSV
Website
Word Document
Database
YouTube
Notion
Google Drive
Slack
Cloud Storage
```

Each source has a different format.

For example:

```text
PDF → PDF format
CSV → Rows and columns
Website → HTML
TXT → Plain text
Database → Records
```

The LLM doesn't directly need all these different formats.

LangChain's **Document Loaders** help convert them into a common representation.

```text
PDF ───────┐
TXT ───────┤
CSV ───────┤
Website ───┤
Database ──┤
            ↓
     Document Loader
            ↓
        Document
```

---

# 2. What is a `Document`?

LangChain represents loaded content using a `Document`.

Conceptually:

```python
Document(
    page_content="LangChain is a framework for building LLM applications.",
    metadata={
        "source": "langchain.pdf",
        "page": 1
    }
)
```

A `Document` mainly contains:

```text
Document
├── page_content
└── metadata
```

### `page_content`

Contains the actual text.

```text
"LangChain is a framework for building LLM applications."
```

### `metadata`

Contains additional information about the content.

```python
{
    "source": "langchain.pdf",
    "page": 1
}
```

Metadata becomes very useful later during:

* Retrieval
* Filtering
* Debugging
* Citations
* Source tracking

---

# 3. Where Document Loaders Fit in RAG

Document loaders are part of the **indexing/ingestion pipeline**.

```text
             INDEXING / INGESTION

Data Sources
     ↓
Document Loaders
     ↓
Documents
     ↓
Text Splitters
     ↓
Chunks
     ↓
Embedding Model
     ↓
Vector Store
```

Then during querying:

```text
              QUERY PHASE

User Question
      ↓
   Retriever
      ↓
Relevant Chunks
      ↓
     LLM
      ↓
    Answer
```

### Important Interview Point

> **Document loaders are mainly used during the data ingestion/indexing phase, not during every user query.**

# 5. PDF Document Loader

PDFs are extremely common in RAG applications.

Example use cases:

* Company documents
* Research papers
* Books
* Product manuals
* Legal documents
* Course notes

A PDF loader extracts text from PDF files and converts it into LangChain `Document` objects.

Example:

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("company_policy.pdf")

documents = loader.load()
```

Conceptually:

```text
company_policy.pdf
        ↓
   PyPDFLoader
        ↓
Document 1
Document 2
Document 3
...
```

Depending on the loader, metadata may contain information such as the source and page number.

# 16. Document Loader vs Text Splitter

This is a very important distinction.

### Document Loader

**Loads the data.**

```text
PDF
 ↓
Loader
 ↓
Document
```

### Text Splitter

**Splits the loaded document into smaller chunks.**

```text
Document
 ↓
Text Splitter
 ↓
Chunk 1
Chunk 2
Chunk 3
```

### Easy Memory Trick

> **Loader = Bring the data in.**

> **Splitter = Break the data into pieces.**

---

# 17. Document Loader vs Embeddings

Another important distinction.

### Loader

Converts source data into documents.

```text
PDF
 ↓
Document
```

### Embedding

Converts text into numerical vectors.

```text
Document
 ↓
Embedding Model
 ↓
[0.12, 0.43, -0.22, ...]
```

So:

```text
Loader
→ Data → Document

Embedding
→ Text → Vector
```

# 18. Complete Example

Suppose we want to build a chatbot for a company handbook.

We have:

```text
company_handbook.pdf
```

### Step 1 — Load

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("company_handbook.pdf")

documents = loader.load()
```

Now:

```text
PDF
 ↓
PyPDFLoader
 ↓
Documents
```

### Step 2 — Split

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_documents(documents)
```

Now:

```text
Documents
 ↓
Chunks
```

### Step 3 — Embed

```text
Chunks
 ↓
Embedding Model
 ↓
Vectors
```

### Step 4 — Store

```text
Vectors
 ↓
Vector Store
```

### Step 5 — Retrieve

```text
User Question
 ↓
Retriever
 ↓
Relevant Chunks
```

### Step 6 — Generate

```text
Relevant Chunks
      +
User Question
      ↓
     LLM
      ↓
   Answer
```

# 20. Why Metadata Matters

Suppose we load a PDF:

```text
company_policy.pdf
```

The document might contain:

```python
Document(
    page_content="Employees receive 15 days of annual leave.",
    metadata={
        "source": "company_policy.pdf",
        "page": 10
    }
)
```

Later, when the retriever returns the chunk, we know:

```text
Source → company_policy.pdf
Page → 10
```

This allows us to provide citations or trace where an answer came from.

### Interview Point

> **Metadata provides contextual information about a document and can be used for filtering, retrieval, source attribution, and debugging.**

# 21. `load()` vs `lazy_load()`

This is a useful technical concept.

### `load()`

Loads the documents immediately.

Conceptually:

```text
Source
 ↓
Load everything
 ↓
Return documents
```

### `lazy_load()`

Loads documents lazily, generally yielding them as needed rather than materializing the entire collection immediately.

Conceptually:

```text
Large Dataset
     ↓
lazy_load()
     ↓
Document 1
Document 2
Document 3
...
```

### Why is lazy loading useful?

For large datasets, loading everything into memory at once may be inefficient.

# 22. Document Loader vs Indexing

Don't confuse these.

### Document Loader

One component:

```text
Source → Documents
```

### Indexing

The complete preparation process:

```text
Source
 ↓
Loader
 ↓
Documents
 ↓
Splitter
 ↓
Chunks
 ↓
Embeddings
 ↓
Vector Store
```

Therefore:

> **Document loading is one stage of the indexing pipeline.**

# 27. Intermediate Questions

### Q5. What is the difference between a Document Loader and a Text Splitter?

**Answer:**

The loader converts external data into documents, while the text splitter divides those documents into smaller chunks for processing and retrieval.

```text
Loader → Documents
Splitter → Chunks
```

---

### Q6. Why is metadata important?

**Answer:**

Metadata provides information about the document, such as its source, page, document ID, or category. It can be used for filtering, debugging, and source attribution.

---

### Q7. What is the difference between `load()` and `lazy_load()`?

**Answer:**

`load()` materializes the loaded documents immediately, while `lazy_load()` yields documents progressively, which can be useful when working with large datasets.
