# Prompt Engineering, Prompts, Types of Prompts & LangChain Prompts

## 1. What is Prompt Engineering?

### Definition

**Prompt Engineering** is the process of designing and improving instructions given to an AI model so that it produces the desired output.

In simple words:

> **Prompt Engineering = Writing effective instructions for an AI model.**

### Example

A poor prompt:

```text
Tell me about Python.
```

A better prompt:

```text
You are a Python instructor.

Explain Python to a beginner in simple language.
Include:
1. Definition
2. Important features
3. Real-world example
4. One simple code example

Keep the answer under 300 words.
```

The second prompt gives the model:

* Role
* Task
* Context
* Expected output
* Constraints

Therefore, the model has clearer instructions.

---

# 2. Why is Prompt Engineering Important?

LLMs are very powerful, but the quality and usefulness of their responses depend heavily on how we communicate the task.

Prompt engineering helps us:

* Get more relevant answers
* Reduce ambiguity
* Control response format
* Provide context
* Control tone and style
* Improve consistency
* Guide reasoning and task execution
* Generate structured outputs
* Build reliable LLM applications

### Simple Formula

```text
Good Prompt
    ↓
Clear Instructions
    +
Relevant Context
    +
Expected Output
    ↓
Better Response
```

> **Interview Point:** Prompt engineering is not about finding one "magic prompt". It is an iterative process of designing, testing, and improving instructions.

---

# 3. What is a Prompt?

## Definition

A **prompt** is the input or instruction provided to an AI model to tell it what task to perform.

A prompt can contain:

* Instructions
* Questions
* Context
* Examples
* Constraints
* Input data
* Expected output format

### Example

```text
Summarize the following article in 5 bullet points.
```

Here:

```text
Task        → Summarize
Input       → Article
Constraint  → 5 bullet points
Output      → Summary
```

---

# 4. Basic Structure of a Good Prompt

A useful prompt can often be structured as:

```text
Role
  +
Task
  +
Context
  +
Constraints
  +
Output Format
```

### Example

```text
Role:
You are an expert Python teacher.

Task:
Explain decorators.

Context:
The learner is a beginner.

Constraints:
Use simple language and avoid advanced terminology.

Output Format:
1. Definition
2. How it works
3. Example
4. Interview question
```

---

# 5. Types of Prompts

There are several common prompting techniques.

## 5.1 Zero-Shot Prompting

### Definition

In **zero-shot prompting**, we give the model a task without providing examples.

### Example

```text
Classify this review as Positive or Negative:

"The product quality is excellent."
```

Expected output:

```text
Positive
```

There is no example provided.

### Structure

```text
Instruction
    ↓
Input
    ↓
Output
```

### When to Use?

Use zero-shot prompting when:

* The task is simple.
* The model already understands the task.
* You don't need to demonstrate a specific output pattern.

---

# 5.2 One-Shot Prompting

### Definition

In **one-shot prompting**, we provide **one example** before asking the model to perform the task.

### Example

```text
Example:

Input:
"I love this phone."

Output:
Positive

Now classify:

Input:
"The battery life is terrible."

Output:
```

Expected:

```text
Negative
```

### Structure

```text
One Example
    ↓
New Input
    ↓
Model
    ↓
Output
```

---

# 5.3 Few-Shot Prompting

### Definition

In **few-shot prompting**, we provide multiple examples to demonstrate the desired task and output pattern.

### Example

```text
Example 1:
Input: "I love this product."
Output: Positive

Example 2:
Input: "This product is terrible."
Output: Negative

Example 3:
Input: "It is okay."
Output: Neutral

Now classify:

Input: "The product is excellent."
Output:
```

Expected:

```text
Positive
```

### Why Use Few-Shot Prompting?

It helps the model understand:

* Task pattern
* Expected output
* Formatting
* Classification style
* Desired behavior

---

# 5.4 Role Prompting

### Definition

In role prompting, we tell the model to behave like a particular expert or persona.

### Example

```text
You are a senior Python developer.

Review the following code and identify potential problems.
```

Other examples:

```text
You are a SQL expert.
```

```text
You are an experienced system design interviewer.
```

```text
You are a mathematics teacher.
```

### Purpose

Role prompting can provide useful behavioral context, but it does **not** magically give the model real-world credentials or capabilities.

---

# 5.5 Instruction Prompting

### Definition

Instruction prompting directly tells the model what task to perform.

### Example

```text
Convert the following paragraph into 5 bullet points.
```

Another example:

```text
Translate the following English sentence into Hindi.
```

---

# 5.6 Contextual Prompting

### Definition

Contextual prompting provides additional information that the model needs to perform the task.

### Example

```text
You are helping a customer support team.

Company policy:
Customers can return products within 30 days.

Customer:
"I purchased this product 20 days ago. Can I return it?"

Answer based only on the policy provided above.
```

Here, the company policy is the **context**.

---

# 5.7 Chain-of-Thought Prompting

### Definition

Chain-of-thought prompting is a technique intended to encourage a model to reason through a problem step by step.

For example:

```text
Solve the problem carefully and provide the final answer.
```

For production applications, you generally want the model to provide the useful result rather than requiring it to expose private internal reasoning.

### Interview Point

> Modern prompting often focuses on clear task instructions, relevant context, examples, structured outputs, and appropriate reasoning/tool-use strategies rather than simply asking the model to reveal its hidden chain of thought.

---

# 5.8 Structured Prompting

### Definition

Structured prompting organizes instructions into clearly defined sections.

Example:

```text
Role:
You are a technical interviewer.

Task:
Evaluate the candidate's answer.

Candidate Answer:
{answer}

Requirements:
- Identify mistakes.
- Explain the correct concept.
- Give an improved answer.

Output:
Return:
1. Score
2. Mistakes
3. Correct explanation
4. Improved answer
```

This is particularly useful in production GenAI applications.

# 7. What is a Prompt Template?

## Definition

A **Prompt Template** is a reusable prompt containing placeholders for dynamic values.

Instead of writing:

```text
Explain Python in simple language.
```

every time, we can create:

```text
Explain {topic} in simple language.
```

Then:

```text
topic = Python
```

produces:

```text
Explain Python in simple language.
```

If:

```text
topic = LangChain
```

it produces:

```text
Explain LangChain in simple language.
```

### Why Prompt Templates?

They provide:

* Reusability
* Dynamic inputs
* Cleaner code
* Consistency
* Easier maintenance

# 8. LangChain Prompts

LangChain provides prompt abstractions for constructing messages and reusable prompts.

For chat applications, a common approach is to use `ChatPromptTemplate`. This allows us to define a sequence of messages and insert dynamic variables into them.

---

## 8.1 Simple LangChain Prompt Example

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple language."
)

messages = prompt.invoke({
    "topic": "LangChain"
})

print(messages)
```

Conceptually:

```text
Template
    ↓
"Explain {topic} in simple language."
    ↓
topic = LangChain
    ↓
Final Prompt
    ↓
"Explain LangChain in simple language."
```

# 9. LangChain Chat Prompt

Chat models work with messages rather than only a single text string.

A prompt can contain multiple message roles.

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert Python teacher."),
    ("human", "Explain {topic} to a beginner.")
])

messages = prompt.invoke({
    "topic": "Decorators"
})

print(messages)
```

Conceptually:

```text
System:
You are an expert Python teacher.

Human:
Explain Decorators to a beginner.
```

The resulting messages can then be passed to a chat model.

# 10. System, Human and AI Messages

## System Message

Provides high-level instructions about how the model should behave.

```text
You are an expert Python teacher.
```

## Human Message

Represents the user's request.

```text
Explain decorators.
```

## AI Message

Represents a previous model response in a conversation.

```text
Decorators are functions that modify...
```

## Tool Message

Contains the result returned by a tool.

```text
Weather API Result:
Delhi: 32°C
```

### Simplified Flow

```text
System Message
      ↓
Human Message
      ↓
Chat Model
      ↓
AI Message
```

# 11. Dynamic Prompt Example

Suppose we are building a learning application.

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are an expert {role}."
    ),
    (
        "human",
        "Explain {topic} for a {level} learner."
    )
])

messages = prompt.invoke({
    "role": "GenAI mentor",
    "topic": "RAG",
    "level": "beginner"
})

print(messages)
```

The prompt dynamically becomes:

```text
System:
You are an expert GenAI mentor.

Human:
Explain RAG for a beginner learner.
```

### Important Concept

The values:

```text
{role}
{topic}
{level}
```

are **template variables**.

# 12. Prompt + Model

A prompt becomes more useful when connected to a model.

Conceptually:

```text
Prompt Template
      ↓
Dynamic Input
      ↓
Chat Model
      ↓
AI Response
```

Example:

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "openai:gpt-5.5"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert GenAI teacher."),
    ("human", "Explain {topic} in simple language.")
])

chain = prompt | model

response = chain.invoke({
    "topic": "Embeddings"
})

print(response.content)
```

The `prompt | model` syntax represents a composable pipeline:

```text
Input
  ↓
Prompt
  ↓
Model
  ↓
Response
```

LangChain's current model APIs support standardized chat-model interfaces across providers.

# 15. Prompt Engineering in a RAG Application

Prompt engineering is especially important in RAG.

Suppose the retriever returns:

```text
Context:
LangChain is a framework for building LLM applications.
```

We can create a prompt:

```text
You are a helpful AI assistant.

Answer the question using only the provided context.

Context:
{context}

Question:
{question}

If the answer is not present in the context,
say that you don't have enough information.
```

Flow:

```text
User Question
      ↓
Retriever
      ↓
Relevant Context
      ↓
Prompt Template
      ↓
Chat Model
      ↓
Final Answer
```

This is a very common production pattern.

# 16. Common Prompt Engineering Best Practices

## 1. Be Specific

Bad:

```text
Explain Python.
```

Better:

```text
Explain Python to a beginner in 5 bullet points.
Include one real-world example.
```

---

## 2. Provide Context

Instead of:

```text
Summarize this.
```

Provide:

```text
You are preparing notes for a beginner.

Summarize the following lecture into:
- Definition
- Important concepts
- Examples
- Interview points
```

---

## 3. Specify the Output Format

Example:

```text
Return the answer in JSON with:

name
age
skills
experience
```

---

## 4. Give Examples When Necessary

Use:

```text
Input → Output
```

examples when the desired behavior or formatting is difficult to describe.

---

## 5. Define Constraints

Examples:

```text
Use maximum 200 words.
```

```text
Return exactly 5 bullet points.
```

```text
Use simple English.
```

---

## 6. Separate Instructions and Data

A clear structure helps:

```text
INSTRUCTIONS:
Summarize the text.

TEXT:
{input_text}

OUTPUT FORMAT:
5 bullet points.
```

# 17. Prompt Engineering vs RAG

These are different concepts.

### Prompt Engineering

Controls **how we instruct the model**.

```text
Instruction
   ↓
Model
```

### RAG

Provides **external knowledge/context** to the model.

```text
Question
   ↓
Retriever
   ↓
Context
   ↓
Prompt
   ↓
Model
```

They are often used together.
