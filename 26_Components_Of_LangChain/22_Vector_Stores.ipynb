# `Detailed Notes: `

# Vector Stores in GenAI

## 1. What is a Vector Store?

> **A Vector Store is a database/storage system designed to store, manage, and search vector embeddings efficiently.**

In GenAI, we convert text, images, audio, or other data into **embeddings (vectors)** and store those vectors in a vector store.

```text
Document
   ↓
Text Chunk
   ↓
Embedding Model
   ↓
Vector
   ↓
Vector Store
```

When a user asks a question:

```text
User Query
    ↓
Embedding Model
    ↓
Query Vector
    ↓
Vector Store
    ↓
Similarity Search
    ↓
Most Relevant Chunks
```

---

# 2. Why Do We Need Vector Stores?

Traditional databases are good at queries like:

```sql
SELECT * FROM users
WHERE name = 'Arun';
```

This is based mainly on **structured values and exact conditions**.

But GenAI applications often need:

> "Find documents that are semantically similar to this question."

For example:

```text
Query:
"How can I prevent a model from memorizing training data?"
```

A relevant document might contain:

```text
"Overfitting occurs when a machine learning model
learns the training data too closely..."
```

There may be no exact keyword match for "memorizing training data."

A vector store can use the **meaning represented by embeddings** to retrieve the relevant content.

---

# 3. What is a Vector?

A vector is a numerical representation of information.

For example:

```text
"Python is a programming language"
```

might become something conceptually like:

```text
[0.21, -0.14, 0.83, 0.42, ...]
```

Real embedding vectors usually contain many more dimensions.

For example:

```text
Text
 ↓
Embedding Model
 ↓
[0.12, -0.44, 0.87, ..., 0.31]
```

The vector represents semantic information about the text.

---

# 4. Embedding vs Vector Store

These two concepts are often confused.

### Embedding Model

Converts:

```text
Text → Vector
```

Example:

```text
"Machine learning"
       ↓
Embedding Model
       ↓
[0.12, -0.45, 0.78, ...]
```

### Vector Store

Stores and searches:

```text
Vectors + Metadata
```

So:

```text
Embedding Model
      ↓
creates vectors

Vector Store
      ↓
stores/searches vectors
```

### Interview Point

> **An embedding model creates the vector; a vector store stores and retrieves vectors.**

---

# 5. Vector Store in RAG

Vector stores are one of the core components of **Retrieval-Augmented Generation (RAG)**.

The complete architecture is:

```text
                 DOCUMENTS
                     │
                     ▼
              Document Loader
                     │
                     ▼
                Text Splitter
                     │
                     ▼
                   Chunks
                     │
                     ▼
              Embedding Model
                     │
                     ▼
                  Vectors
                     │
                     ▼
                VECTOR STORE
                     │
                     │
                     │
User Query ──────────┘
     │
     ▼
Embedding Model
     │
     ▼
Query Vector
     │
     ▼
Similarity Search
     │
     ▼
Relevant Chunks
     │
     ▼
      LLM
     │
     ▼
   Answer
```

This is one of the most important architectures to understand for GenAI interviews.

---

# 6. What Does a Vector Store Actually Store?

A vector store usually stores more than just the vector.

A simplified record might look like:

```python
{
    "id": "doc_001",
    
    "vector": [0.12, -0.42, 0.88, ...],
    
    "text": "Machine learning is...",
    
    "metadata": {
        "source": "ml_notes.pdf",
        "page": 12,
        "topic": "Machine Learning"
    }
}
```

So conceptually:

```text
Vector Store
│
├── Vector
├── Document/Text
├── ID
└── Metadata
```

# 8. How Similarity Search Works

Suppose we have:

```text
Document A → Vector A
Document B → Vector B
Document C → Vector C
```

User asks:

```text
"What is machine learning?"
```

The query becomes:

```text
Query → Vector Q
```

The vector store compares:

```text
Vector Q
   ↓
 ┌───────┬───────┬───────┐
 ↓       ↓       ↓
Vec A   Vec B   Vec C
```

It calculates some form of vector similarity/distance.

Example ranking:

```text
Document B → 0.92
Document A → 0.84
Document C → 0.51
```

Then:

```text
Top K = 2

→ Document B
→ Document A
```

are returned.

---

# 9. Common Similarity Metrics

Three important concepts are:

### 1. Cosine Similarity

Measures the angle between vectors.

Conceptually:

```text
Similar direction
      ↓
High similarity
```

It is commonly used for semantic similarity.

---

### 2. Euclidean Distance

Measures the geometric distance between vectors.

```text
Smaller distance
      ↓
More similar
```

---

### 3. Dot Product

Calculates the product of corresponding vector values and sums them.

Depending on how vectors are normalized, dot product can behave similarly to cosine similarity.

---

# 10. Important Interview Question

### Q: Which similarity metric does a vector database use?

**Answer:**

> It depends on the vector store/index configuration and embedding setup. Common metrics include cosine similarity, dot product, and Euclidean/L2 distance.

Don't say:

> "Every vector database uses cosine similarity."

That's incorrect.

# 11. Vector Store vs Vector Database

You'll see both terms.

### Vector Database

A dedicated database system designed for storing and searching vector embeddings.

Examples:

* Pinecone
* Qdrant
* Weaviate
* Milvus

### Vector Store

A broader term often used by frameworks such as LangChain for the component/interface used to store and retrieve vectors.

For example:

```text
LangChain Vector Store
       ↓
Chroma
Qdrant
Pinecone
FAISS
...
```

In practical conversations, people sometimes use **vector store** and **vector database** interchangeably, but technically they're not always identical.

# 18. Vector Store Indexing

You may hear:

> "We need to create an index."

An index is a data structure that allows the system to perform similarity searches efficiently.

Without efficient indexing:

```text
Query
 ↓
Compare against every vector
 ↓
Very expensive
```

With an appropriate vector index:

```text
Query
 ↓
Vector Index
 ↓
Candidate vectors
 ↓
Top K
```

---

# 19. Exact vs Approximate Search

This is an important advanced concept.

## Exact Nearest Neighbor Search

Compare the query against every vector.

```text
Query
 ↓
Vector 1
Vector 2
Vector 3
...
Vector 1,000,000
```

Potentially expensive at scale.

## Approximate Nearest Neighbor (ANN)

Use an index to efficiently find likely nearest neighbors.

```text
Query
 ↓
ANN Index
 ↓
Candidate vectors
 ↓
Top K
```

You trade some search exactness for much better performance.

---

# 20. Important ANN Algorithms

You don't need to implement these initially, but know their names.

### HNSW

**Hierarchical Navigable Small World**

Very common in vector search.

Conceptually:

```text
Query
 ↓
Graph-based index
 ↓
Navigate toward nearest vectors
 ↓
Top K
```

### IVF

**Inverted File Index**

Partitions vectors into clusters and searches relevant clusters instead of the entire dataset.

### Product Quantization

Compresses vectors to reduce memory usage and accelerate search.

# 27. Retriever vs Vector Store

Another common interview question.

### Vector Store

Stores and searches vectors.

```text
Vector Store
 ↓
similarity_search()
```

### Retriever

Provides a standardized retrieval interface.

```text
Retriever
 ↓
Relevant Documents
```

A vector store can be converted into a retriever:

```python
retriever = vector_store.as_retriever(
    search_kwargs={"k": 3}
)
```

Then:

```python
documents = retriever.invoke(
    "What is overfitting?"
)
```

So:

```text
Vector Store
      ↓
   Retrieval
      ↓
  Retriever
```

A retriever can also be backed by other retrieval mechanisms, so **retriever ≠ vector store**.

# 32. Common Problems With Vector Stores

### 1. Bad Chunking

If chunks are too large:

```text
Too much irrelevant context
```

If too small:

```text
Important context gets separated
```

---

### 2. Poor Embeddings

If the embedding model doesn't represent your domain well:

```text
Poor vectors
 ↓
Poor retrieval
 ↓
Poor RAG answers
```

---

### 3. Wrong K

If:

```text
k = 1
```

you may miss important context.

If:

```text
k = 50
```

you may send too much irrelevant context to the LLM.

---

### 4. Duplicate Documents

Duplicates can cause:

```text
Same content
 ↓
Multiple retrieved results
```

---

### 5. Metadata Problems

Poor metadata makes filtering and source attribution harder.

# 34. Interview Questions

## Beginner

### Q1. What is a vector store?

> A vector store is a storage and retrieval system designed to store embeddings and perform similarity searches over them.

### Q2. Why do we need a vector store?

> To efficiently find semantically similar content from a collection of embedded documents.

### Q3. What is an embedding?

> A numerical representation of data that captures useful semantic or contextual information.

### Q4. What is similarity search?

> A search technique that finds vectors closest or most relevant to a query vector according to a chosen distance or similarity metric.

---

# 35. Intermediate Interview Questions

### Q5. What is the difference between an embedding model and a vector store?

> The embedding model converts text/data into vectors. The vector store stores those vectors and enables efficient similarity retrieval.

### Q6. What is Top-K retrieval?

> Returning the K most relevant documents or chunks for a query.

Example:

```python
k = 5
```

means retrieving the top 5 candidates.

### Q7. What is ANN?

> Approximate Nearest Neighbor search is a technique for finding near-neighbor vectors efficiently without exhaustively comparing the query against every vector.

### Q8. What is HNSW?

> HNSW is a graph-based approximate nearest-neighbor indexing algorithm commonly used to efficiently search high-dimensional vectors.
