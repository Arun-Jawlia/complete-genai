# Embedding Models in GenAI

> **An embedding model converts text, images, audio, or other data into numerical vectors that capture their semantic meaning.**

In simple words:

```text
Human-readable data
        ↓
   Embedding Model
        ↓
Numerical Vector
        ↓
[0.12, -0.45, 0.78, 0.21, ...]
```

The important idea is:

> **Similar meanings → similar vectors.**

This is one of the most important concepts for **RAG, semantic search, vector databases, recommendation systems, and semantic text splitting**.

---

# 1. What Is an Embedding?

An **embedding** is a numerical representation of data in a high-dimensional vector space.

For example:

```text
"Dog is an animal"
        ↓
Embedding Model
        ↓
[0.21, 0.45, -0.18, 0.73, ...]
```

Another sentence:

```text
"A dog is a type of animal"
        ↓
Embedding Model
        ↓
[0.24, 0.43, -0.16, 0.70, ...]
```

The vectors should be relatively close because the two sentences have similar meanings.

---

# 2. Why Do We Need Embeddings?

Computers cannot directly compare the meaning of sentences like humans do.

Consider:

```text
Sentence A:
I love programming.

Sentence B:
Software development is something I enjoy.
```

A simple keyword comparison might see very few common words.

But semantically:

```text
"I love programming"
        ≈
"I enjoy software development"
```

An embedding model converts both into vectors so that semantic similarity can be calculated mathematically.

# 4. What Is a Vector?

A vector is simply a list of numbers.

For example:

```python
[0.12, 0.45, -0.32, 0.78]
```

Real embedding models usually produce vectors with hundreds or thousands of dimensions.

For example:

```text
Text
 ↓
Embedding Model
 ↓
[0.12, -0.32, 0.55, 0.91, ...]
       ↑
Many dimensions
```

You normally **don't interpret individual dimensions manually**.

Instead, the entire vector represents the data's learned semantic characteristics.

---

# 5. Embedding Space

Imagine a 2D simplified embedding space:

```text
              Animals
                 ↑
        🐕 Dog   🐈 Cat
                 |
                 |
                 |
─────────────────┼────────────→
                 |
       🚗 Car    |     🚌 Bus
                 |
              Vehicles
```

Similar concepts are placed closer together.

For example:

```text
Dog ↔ Cat
```

should be closer than:

```text
Dog ↔ Database
```

The actual embedding space has many dimensions, not just two.

---

# 6. Semantic Similarity

Suppose:

```text
A = "I love Python programming."

B = "Python is my favorite programming language."
```

Their vectors might be:

```text
A → [0.2, 0.8, 0.4, ...]
B → [0.3, 0.7, 0.5, ...]
```

They are close.

But:

```text
C = "The weather is cold today."
```

might be represented differently:

```text
C → [-0.6, 0.1, 0.9, ...]
```

Therefore:

```text
Similarity(A, B) → High
Similarity(A, C) → Low
```

---

# 7. Cosine Similarity

One of the most common ways to compare embeddings is **cosine similarity**.

The idea is:

```text
Vector A
     ↘
      angle
     ↗
Vector B
```

If two vectors point in similar directions:

```text
Similarity → High
```

If they point in very different directions:

```text
Similarity → Low
```

The cosine similarity formula is:

[
\text{cosine similarity}
========================

\frac{A \cdot B}
{|A||B|}
]

For normalized vectors, this can be closely related to their dot product.

---

# 8. Example of Similarity

Suppose:

```text
A = [1, 2]
B = [2, 4]
```

They point in the same direction.

Therefore:

```text
Cosine Similarity ≈ 1
```

Meaning:

> Very similar direction.

Now:

```text
A = [1, 2]
B = [-1, -2]
```

They point in opposite directions.

```text
Cosine Similarity ≈ -1
```

The exact similarity interpretation depends on the embedding model and normalization, but the general principle is:

> **The closer the vectors are in direction, the more semantically similar they tend to be.**

# 17. Why Vector Databases?

Examples include:

* Pinecone
* Weaviate
* Milvus
* Qdrant
* Chroma
* FAISS as a vector similarity-search library/index rather than a full database

```text
Query
 ↓
Embedding
 ↓
Vector DB
 ↓
Nearest vectors
```

# 25. Are Embedding Values Human-Interpretable?

Usually, no.

You might see:

```text
[0.12, -0.34, 0.88, ...]
```

You generally cannot say:

```text
Dimension 1 = programming
Dimension 2 = Python
Dimension 3 = AI
```

The representation is distributed across many dimensions.

What matters is the **relationship between vectors**.

---

# 26. Dense vs Sparse Representations

Embedding models usually produce **dense vectors**.

Example:

```text
Dense:
[0.12, 0.45, -0.22, 0.81, ...]
```

Sparse representations have many zeros:

```text
Sparse:
[0, 0, 0, 1, 0, 0, 0, 1, ...]
```

Traditional keyword approaches such as TF-IDF are commonly represented as sparse vectors.

Modern semantic embeddings are generally dense.

# 28. Embeddings vs One-Hot Encoding

### One-Hot Encoding

Suppose vocabulary:

```text
["cat", "dog", "car"]
```

Then:

```text
cat → [1,0,0]
dog → [0,1,0]
car → [0,0,1]
```

The vectors don't naturally encode semantic similarity.

### Embeddings

```text
cat → [0.8, 0.2, 0.7, ...]
dog → [0.7, 0.3, 0.8, ...]
car → [-0.2, 0.9, -0.1, ...]
```

Now:

```text
cat ↔ dog
```

can be represented as more similar than:

```text
cat ↔ car
```

# 32. Important Interview Questions

## Beginner

### Q1. What is an embedding?

**Answer:**

An embedding is a numerical vector representation of data that captures useful semantic relationships.

---

### Q2. What is an embedding model?

**Answer:**

An embedding model converts data such as text into numerical vectors that can be compared using mathematical similarity measures.

---

### Q3. Why are embeddings used in RAG?

**Answer:**

They allow documents and user queries to be represented as vectors so that semantically relevant document chunks can be retrieved using vector similarity.

---

### Q4. What is semantic similarity?

**Answer:**

Semantic similarity measures how closely two pieces of content are related in meaning. With embeddings, this can be estimated by comparing their vectors.

---

# 33. Intermediate Interview Questions

### Q5. What is cosine similarity?

**Answer:**

Cosine similarity measures the angle between two vectors. Vectors pointing in similar directions generally have higher similarity.

---

### Q6. What is embedding dimension?

**Answer:**

Embedding dimension is the number of numerical values in the output vector.

For example:

```text
[0.1, 0.2, 0.3, ..., 0.8]
```

If it contains 1536 values, it is a 1536-dimensional embedding.

---

### Q7. What is the difference between an embedding model and a vector database?

**Answer:**

The embedding model converts data into vectors, while the vector database stores and efficiently searches those vectors.

---

### Q8. Why do we embed both documents and queries?

**Answer:**

Documents are embedded during ingestion and the user's query is embedded at search time. Their vectors can then be compared to find semantically relevant documents.

# 34. Scenario-Based Questions

### Q9. Your RAG system retrieves irrelevant documents. What could be wrong?

Possible causes include:

```text
Poor embedding model
Incorrect chunking
Poor query formulation
Wrong similarity metric
Poor vector database configuration
Insufficient metadata filtering
Low-quality source documents
```

Don't assume the embedding model is always the problem.

---

### Q10. Your embedding API is becoming expensive. What can you do?

Possible approaches:

```text
Use a lower-cost embedding model
Cache embeddings
Avoid re-embedding unchanged documents
Batch embedding requests
Remove duplicate documents
Optimize chunking
Consider a local embedding model
```

---

### Q11. Can two embedding models be mixed in the same vector index?

Generally, you should **not casually mix embeddings from incompatible models** in the same index.

Why?

Because different models can produce vectors with different:

* Dimensions
* Vector spaces
* Training objectives
* Semantic representations

If you change the embedding model, you will usually need to **re-embed the stored documents**.
