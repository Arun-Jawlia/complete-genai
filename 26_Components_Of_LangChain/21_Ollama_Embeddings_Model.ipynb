# 1. What Is Ollama?

### Simple Definition

> **Ollama is a tool/runtime that makes it easy to download, run, manage, and serve open-weight AI models on your own computer.**

Instead of:

```text
Your Application
       ↓
Internet
       ↓
OpenAI API
       ↓
GPT Model
       ↓
Response
```

you can build:

```text
Your Application
       ↓
Ollama
       ↓
Local AI Model
       ↓
Response
```

The model runs on your machine when using local models.

---

# 2. What Problem Does Ollama Solve?

Without a tool like Ollama, running an LLM locally can involve:

* Downloading model weights
* Understanding model formats
* Setting up inference libraries
* Managing GPU/CPU execution
* Handling tokenization
* Managing model memory
* Building an API layer

Ollama simplifies much of this.

Instead of manually configuring everything:

```text
Model
 ↓
Runtime
 ↓
GPU
 ↓
Inference
 ↓
API
```

you can often do:

```bash
ollama run gemma3
```

Ollama provides CLI commands, a local API, libraries, and integrations. ([Ollama][2])

---

# 3. Ollama Is NOT an LLM

This is a very important interview point.

Ollama ≠ LLM.

For example:

```text
Ollama
   ↓
Runtime / model manager
   ↓
Runs
   ├── Gemma
   ├── Qwen
   ├── DeepSeek
   ├── gpt-oss
   └── other supported models
```

Think of it like:

```text
Docker → runs containers

Ollama → runs/manages LLMs
```

That's not a perfect technical analogy, but it's useful for understanding the role.

# 27. Model Size Matters

You will see names like:

```text
7B
14B
32B
70B
```

The `B` generally refers to **billions of parameters**.

Conceptually:

```text
Smaller Model
↓
Less RAM
↓
Faster
↓
Lower resource requirements
```

versus:

```text
Larger Model
↓
More RAM/VRAM
↓
Potentially stronger capabilities
↓
More compute required
```

But parameter count alone does **not** determine quality.

---

# 28. Quantization

This is an important concept when working with Ollama.

You may see:

```text
Q4
Q5
Q8
```

Quantization reduces the numerical precision used to represent model weights.

Conceptually:

```text
Full Precision
      ↓
Quantization
      ↓
Smaller Model
      ↓
Less Memory
      ↓
Faster Local Inference
```

There is usually a tradeoff:

```text
Lower precision
→ smaller/faster
→ potentially some quality loss

Higher precision
→ larger/slower
→ potentially better fidelity
```

Ollama model variants can use different quantization levels; the exact available variants depend on the model. ([Ollama][7])

---

# 29. Hardware Matters

Before downloading a huge model, consider:

```text
RAM
VRAM
GPU
CPU
Disk
```

For example, a model that requires tens of GB of memory isn't a sensible choice for a laptop with 8 GB RAM.

A useful starting rule:

```text
Small laptop
→ Small models

Good GPU
→ Medium models

High-end GPU / workstation
→ Larger models
```

# 30. CPU vs GPU

Ollama can run models using available hardware.

Conceptually:

```text
                 Ollama
                    │
          ┌─────────┴─────────┐
          ↓                   ↓
         CPU                 GPU
          │                   │
      Inference           Inference
```

GPU inference is generally much faster for suitable workloads, but exact performance depends heavily on the model, quantization, hardware, memory bandwidth, and workload.

# 33. Ollama Context Window

The context window determines how much input/context the model can process for a request.

For example:

```text
User Query
+
System Prompt
+
Conversation
+
Retrieved RAG Documents
```

all consume context.

Ollama's FAQ currently states a default context window of 4096 tokens, which can be overridden through configuration. ([Ollama][8])

For RAG and coding applications, you may need a larger context window, but increasing it also increases memory requirements.

# 50. Temperature

You may see:

```text
PARAMETER temperature 0.2
```

Temperature controls randomness in generation.

Conceptually:

```text
Low temperature
→ More deterministic
→ Good for extraction, coding, factual tasks

Higher temperature
→ More varied/creative
→ Good for brainstorming/writing
```

It doesn't magically make the model more intelligent.

# 55. Important Interview Questions

## Beginner

### Q1. What is Ollama?

**Answer:**

> Ollama is a runtime and tooling platform that makes it easy to download, run, manage, and serve open models locally through a CLI and API.

### Q2. Is Ollama an LLM?

**Answer:**

> No. Ollama is the runtime/tooling layer. Models such as Gemma, Qwen, DeepSeek-R1, and gpt-oss run through it.

### Q3. Why use Ollama?

> To run supported models locally, build applications without relying entirely on hosted LLM APIs, and expose local models through an API.

---

## Intermediate

### Q4. How does LangChain work with Ollama?

```text
LangChain
 ↓
ChatOllama
 ↓
Ollama API
 ↓
Local Model
 ↓
Response
```

### Q5. What is the difference between Ollama and OpenAI?

> OpenAI provides hosted models through its APIs, while Ollama can run supported open models locally. Ollama also now provides cloud models, so the distinction is not strictly "Ollama = local" anymore. ([Ollama][3])

### Q6. What is a Modelfile?

> A Modelfile is a blueprint used to create/customize an Ollama model by defining a base model, system prompt, parameters, template, and other configuration. ([Ollama][10])

```text
Smaller model
Better quantization
GPU acceleration
Fewer concurrent requests
Shorter context
More capable hardware
Different model
```

### Q10. You need an embedding model. Should you use a chat model?

No.

Use an embedding-specific model such as:

```text
embeddinggemma
```

or:

```text
nomic-embed-text
```

rather than using a general chat model for your vector embeddings. Ollama's CLI explicitly supports these embedding models. ([Ollama][5])
