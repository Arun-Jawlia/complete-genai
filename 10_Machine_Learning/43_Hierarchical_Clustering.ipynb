# `What is Hierarchical Clustering?`
---

**Hierarchical Clustering** is an **Unsupervised Machine Learning algorithm** used to group similar data points into clusters by creating a **hierarchy (tree-like structure)**.

Unlike K-Means, it **does not require the number of clusters (K) to be specified before training**.

> **Hierarchical Clustering builds a tree of clusters, where similar data points are merged together step by step.**

---

# `Why Do We Need Hierarchical Clustering?`

Sometimes we don't know:

- How many clusters exist.
- Which data points belong together.

Hierarchical Clustering helps us discover the natural grouping in the data.

---

# Real-Life Example

Suppose a university has student information:

- Marks
- Attendance
- Assignment Score

Initially, the university doesn't know how many groups exist.

Hierarchical Clustering automatically groups students based on similarity.

Possible groups:

- Excellent Students
- Average Students
- Weak Students

---

# `Why is it called Hierarchical?`

The algorithm creates clusters in **levels (hierarchy)**.

It starts with many small clusters and gradually combines them into larger clusters.

The final structure looks like a **family tree**.

---

# `Types of Hierarchical Clustering`

There are **two types**:

## `1. Agglomerative Hierarchical Clustering (Bottom-Up)`

This is the **most commonly used** approach.

### Working

- Start with every data point as its own cluster.
- Merge the two closest clusters.
- Repeat until all points belong to one cluster.

> **Bottom-Up Approach**

---

## `2. Divisive Hierarchical Clustering (Top-Down)`

### Working

- Start with all data points in one cluster.
- Keep splitting the cluster into smaller clusters.
- Continue until every point becomes its own cluster.

> **Top-Down Approach**

---

# `Agglomerative Clustering (Most Important)`

## Step 1

Initially,

```
A   B   C   D   E

↓

5 Clusters
```

# `Workflow`

```text
Each Point = One Cluster
        ↓
Find Closest Clusters
        ↓
Merge Them
        ↓
Repeat
        ↓
Single Large Cluster
```

---

# `Measuring Similarity`

To merge clusters,

the algorithm first calculates the distance between them.

Common distance metrics:

- Euclidean Distance
- Manhattan Distance
- Cosine Distance

Most commonly,

**Euclidean Distance** is used.

---

# `Linkage Methods`

When comparing clusters,

Hierarchical Clustering uses different linkage methods.

---

## 1. Single Linkage

Measures the **minimum distance** between two clusters.

```
Nearest Point

Cluster A  •────• Cluster B
```

### Best For

- Long chain-shaped clusters

---

## 2. Complete Linkage

Measures the **maximum distance** between clusters.

```
Farthest Point
```

Produces compact clusters.

---

## 3. Average Linkage

Calculates the **average distance** between all points in two clusters.

Provides balanced clustering.

---

## 4. Ward Linkage ⭐

Most commonly used.

Instead of distance,

it minimizes the increase in **variance** after merging.

Produces compact and well-balanced clusters.

---

# `Dendrogram`

## What is a Dendrogram?

A **Dendrogram** is a tree diagram that shows how clusters are merged.

It helps determine the **optimal number of clusters**.

---

# Example Dendrogram

```code
          ───────
         |       |
     ────        |
    |   |        |
   A    B      ──┴──
               |    |
               C   ─┴─
                  |  |
                  D  E
```

Height represents the distance between merged clusters.

---

# `Choosing the Number of Clusters`

Draw a horizontal line across the dendrogram.

The number of vertical branches it intersects equals the optimal number of clusters.

Example

```code
Horizontal Line

──────────────

↓

Cuts 3 branches

↓

K = 3
```

# `Hierarchical Clustering vs K-Means`

| K-Means | Hierarchical Clustering |
|----------|------------------------|
| Need K beforehand | No need to specify K initially |
| Uses Centroids | Uses Distance Between Clusters |
| Fast | Slower |
| Best for Large Datasets | Best for Small & Medium Datasets |
| Uses Elbow Method | Uses Dendrogram |

---

# `Advantages`

- No need to specify K initially.
- Easy to visualize using a dendrogram.
- Produces a hierarchy of clusters.
- Can identify nested clusters.

---

# `Limitations`

- Slow for large datasets.
- Memory intensive.
- Sensitive to noise and outliers.
- Once clusters are merged, they cannot be split again.

**Answer:**

**Answer:**
