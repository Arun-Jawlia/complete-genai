# `K-Means Clustering`
---

## `What is K-Means Clustering?`

**K-Means Clustering** is an **Unsupervised Machine Learning algorithm** used to group similar data points into **K clusters**.

The algorithm automatically discovers hidden groups in the data without using any target labels.

> **K-Means = Divide similar data points into K groups (clusters).**

---

# `Why is it called K-Means?`

The name consists of two parts:

### K

Represents the **number of clusters** we want to create.

Example:

```
K = 3

↓

Create 3 clusters
```

---

### Means

Represents the **centroid (center point)** of each cluster.

The centroid is calculated as the **mean (average)** of all data points in that cluster.

Hence the name:

**K + Means = K-Means**

---

---

# `Goal of K-Means`

The objective is to:

- Group similar data together.
- Separate dissimilar data.
- Minimize the distance between data points and their cluster centroid.

---

# `What is a Cluster?`

A **cluster** is a group of similar data points.

Example

```code
Students

↓

Cluster 1

Programmers

------------------

Cluster 2

Football Players

------------------

Cluster 3

Artists
```

People inside a cluster are similar to each other.

---

# `What is a Centroid?`

A **centroid** is the center point of a cluster.

It represents the average location of all data points inside that cluster.

Example:

```code
        ●

    ●   X   ●

       ●
```

`X` = Centroid

---

# `K-Means Intuition`

## Example

Suppose we have customer data based on:

- Annual Income
- Spending Score

Initially, all customers are mixed together.

```code
● ● ● ● ●

● ● ● ● ●

● ● ● ● ●
```

Suppose we choose:

```
K = 3
```

The algorithm randomly places three centroids.

```
X      X      X
```

Each customer is assigned to the nearest centroid.

After assignment:

```
Cluster A

Cluster B

Cluster C
```

The centroid moves to the center of its assigned cluster.

Again, customers are reassigned.

Again, centroids move.

This process repeats until the centroids stop moving.

This is called **convergence**.

---

# `How K-Means Works (Step-by-Step)`

## `Step 1: Choose the Number of Clusters (K)`

Decide how many clusters you want.

Example:

```
K = 3
```

---

## `Step 2: Initialize K Centroids`

Randomly place K centroids.

Example

```
X

X

X
```

Initially, these are chosen randomly.

---

## `Step 3: Calculate Distance`

For every data point,

calculate the distance to every centroid.

Usually,

**Euclidean Distance** is used.

Formula

\[
$Distance=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}$
\]

---

## `Step 4: Assign Data Points`

Assign every point to the nearest centroid.

Example

```
Customer A

↓

Nearest to Centroid 1

↓

Cluster 1
```

---

## `Step 5: Update Centroids`

For each cluster,

calculate the new mean.

This new mean becomes the new centroid.

---

## `Step 6: Repeat`

Again,

- Calculate distance
- Assign points
- Update centroid

Repeat until:

- Centroids no longer move.
- Cluster assignments stop changing.

This is called **convergence**.

---

# `K-Means Workflow`

```code
Choose K
     ↓
Initialize K Centroids
     ↓
Calculate Distance
     ↓
Assign Data Points
     ↓
Update Centroids
     ↓
Repeat
     ↓
Centroids Stop Moving
     ↓
Final Clusters
```

# `Choosing the Right Value of K`

One of the biggest questions is:

> **How many clusters should we create?**

This is determined using the **Elbow Method**.

---

# `Elbow Method`

## What is the Elbow Method?

The Elbow Method helps us find the **optimal value of K**.

It plots:

```
K

vs

WCSS
```

---

# `What is WCSS?`

**WCSS** stands for:

**Within Cluster Sum of Squares**

It measures how close the data points are to their centroid.

Lower WCSS means:

- Better clustering

---

Choose the value of **K** where the graph bends like an elbow.

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is K-Means Clustering?

**Answer:**
K-Means is an unsupervised machine learning algorithm that groups similar data points into K clusters.

---

### Q2. Why is it called K-Means?

**Answer:**

* **K** = Number of clusters.
* **Means** = Mean (centroid) of each cluster.

---

### Q3. What is a centroid?

**Answer:**
A centroid is the center point (average) of all data points in a cluster.

---

### Q4. Which distance metric is commonly used in K-Means?

**Answer:**
**Euclidean Distance**.

---

### Q5. What is convergence in K-Means?

**Answer:**
Convergence occurs when the centroids stop changing and the cluster assignments remain the same.

---

## Intermediate Questions

### Q6. What is WCSS?

**Answer:**
**Within Cluster Sum of Squares (WCSS)** measures the total squared distance between data points and their cluster centroid. Lower WCSS indicates better clustering.

---

### Q7. What is the Elbow Method?

**Answer:**
The Elbow Method helps determine the optimal number of clusters by plotting WCSS against different values of K and identifying the "elbow" point.

---

### Q8. What are the limitations of K-Means?

**Answer:**

* Requires K beforehand.
* Sensitive to outliers.
* Sensitive to initial centroid placement.
* Assumes spherical clusters.

---

### Q9. Does K-Means require labeled data?

**Answer:**
No. It works on **unlabeled data**.

---

## Scenario-Based Questions

### Q10. An e-commerce company wants to divide customers into different spending groups without predefined labels. Which algorithm would you recommend?

**Answer:**
**K-Means Clustering**, because it automatically groups similar customers.

---

### Q11. How would you choose the value of K?

**Answer:**
Use the **Elbow Method** to find the optimal number of clusters.

---

### Q12. Why might K-Means perform poorly with many outliers?

**Answer:**
Because outliers can significantly shift the centroid, leading to poor cluster assignments.
