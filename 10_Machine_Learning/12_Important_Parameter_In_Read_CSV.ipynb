# `Important Parameters in read_csv()`
---

This notebook explains important pandas `read_csv()` parameters using separate markdown and code cells.


## Setup

Import pandas before using `pd.read_csv()`.


````python
import pandas as pd
````

## Basic Syntax

**Definition:** `pd.read_csv()` reads a CSV file and returns a pandas DataFrame.

**Syntax:**

```python
pd.read_csv("data.csv")
```


````python
df = pd.read_csv("../datasets/titanic.csv")
df.head()
````

## 1. `sep`

**Definition:** Specifies the delimiter used in the file. The default delimiter is comma `,`.

- Comma-separated file
`df = pd.read_csv("data.csv", sep=",")`

- Semicolon-separated file
`df = pd.read_csv("data.csv", sep=";")`

- Tab-separated file
`pd.read_csv("../datasets/katalog_gempa_v2.tsv", sep="\t")`


````python
# Tab-separated file
df = pd.read_csv("../datasets/katalog_gempa_v2.tsv", sep="\t")
````

````python
df.head()
````

## 2. `header`

**Definition:** Specifies which row contains column names.


````python
# First row as header
df = pd.read_csv("../datasets/titanic.csv", header=1)
df.head()
````

````python
# No header in file
df = pd.read_csv("../datasets/titanic.csv", header=None)
df.head()
````

## 3. `names`

**Definition:** Assigns custom column names while reading the file.


````python
df = pd.read_csv("../datasets/student_depression_dataset.csv")
df.columns
````

````python
df = pd.read_csv(
    "../datasets/student_depression_dataset.csv",
    header=None,
    names=['Student ID', 'Sex', 'Age', 'City', 'Profession', 'Academic Pressure',
       'Work Pressure', 'CGPA', 'Study Satisfaction', 'Job Satisfaction',
       'Sleep Duration', 'Dietary Habits', 'Degree',
       'Have you ever had suicidal thoughts ?', 'Work/Study Hours',
       'Financial Stress', 'Family History of Mental Illness', 'Depression']
)
df.head()
````

## 4. `usecols`

**Definition:** Reads only selected columns from the CSV file.


````python
# Select columns by name
df = pd.read_csv(
    "../datasets/student_depression_dataset.csv",
    usecols=["Age", "Gender"]
)
df.head()
````

````python
# Select columns by index position
df = pd.read_csv(
    "../datasets/student_depression_dataset.csv",
    usecols=[0, 5]
)
df.head()
````

## 5. `index_col`

**Definition:** Specifies a column to use as the DataFrame index.


````python
df = pd.read_csv(
    "../datasets/titanic.csv",
    index_col="PassengerId"
)
df.head()
````

## 7. `squeeze`: convert into series

````python
pd.read_csv("../datasets/titanic.csv", usecols=['Survived']).squeeze()
````

## 6. `dtype`

**Definition:** Specifies the data type of one or more columns.


````python
pd.read_csv(
    "../datasets/student_depression_dataset.csv",
    dtype={
        "Age": 'int32',
        "id": 'int32',
    }
).info()
````

## 7. `nrows`

**Definition:** Reads only the first `N` rows from the CSV file.


````python
pd.read_csv(
    "../datasets/student_depression_dataset.csv",
    nrows=100
)

````

## 8. `skiprows`

**Definition:** Skips selected rows while reading the CSV file.


````python
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
)
````

````python
# Skip first 2 rows
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    skiprows=2
)
````

````python
# Skip specific rows
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    skiprows=[1, 3, 5]
)
````

## 9. `skipfooter`

**Definition:** Skips rows from the bottom of the file. Use `engine="python"` with this parameter.


````python
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    skipfooter=2,
    engine="python"
)
````

## 10. `na_values`

**Definition:** Treats specified values as missing values, or `NaN`.


````python
pd.read_csv(
    "../datasets/titanic.csv",
    na_values=['NaN', "No", '']
)
````

````python
pd.read_csv(
    "../datasets/titanic.csv",
    na_values=["NaN", "Null", "?"]
)
````

## 11. `keep_default_na`

**Definition:** Controls whether pandas should use its default list of missing-value markers.


````python
pd.read_csv(
    "../datasets/titanic.csv",
    keep_default_na=False
)
````

## 12. `parse_dates`

**Definition:** Converts selected columns into datetime format while reading the file.


````python
pd.read_csv(
    "../datasets/katalog_gempa_v2.tsv",
    parse_dates=["datetime"],
    sep='\t'
)
````

````python
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    parse_dates=['Date']
).info()
````

## 13. `dayfirst`

**Definition:** Handles date formats where the day comes before the month, such as `25-12-2026`.


````python
df = pd.read_csv(
    "data.csv",
    parse_dates=["Date"],
    dayfirst=True
)
````

## 14. `encoding`

**Definition:** Specifies the file encoding used to read text correctly.


````python
pd.read_csv(
    "../datasets/titanic.csv",
    encoding="utf-8"
)
````

````python
pd.read_csv(
    "../datasets/katalog_gempa_v2.tsv",
    encoding="latin1"
)
````

## 15. `thousands`

**Definition:** Handles thousands separators in numeric columns.


````python
pd.read_csv(
    "../datasets/titanic.csv",
    thousands=","
)
````

## 17. `Convertors`

````python
def station(value):
    if value == 'Q':
        return 'Queenstown'
    
    return value
````

````python
pd.read_csv("../datasets/titanic.csv", converters={'Embarked': station})
````

## 16. `decimal`

**Definition:** Specifies the decimal separator used in numeric values.


````python
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    decimal=","
)
````

## 17. `true_values` / `false_values`

**Definition:** Converts selected text values into Boolean `True` and `False`.


````python
pd.read_csv(
    "../datasets/titanic.csv",
    true_values=["Yes"],
    false_values=["No"]
)
````

## 18. `chunksize`

**Definition:** Reads large files in smaller chunks instead of loading the full file at once.


````python
df = pd.read_csv(
    "../datasets/International_Education_Costs.csv",
    chunksize=100
)

for d in df:
    print(d.shape)
````

## 19. `compression`

**Definition:** Reads compressed files such as ZIP or GZIP files.


````python
df = pd.read_csv(
    "data.csv.zip",
    compression="zip"
)
````

````python
df = pd.read_csv(
    "data.csv.gz",
    compression="gzip"
)
````

## 20. `engine`

**Definition:** Specifies the parser engine. Common values are `"c"` and `"python"`.


````python
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    engine="c"
)
````

````python
pd.read_csv(
    "../datasets/Chocolate Sales.csv",
    engine="python"
)
````

## 21. `low_memory`

**Definition:** Controls memory optimization while reading large datasets. Use `low_memory=False` for more consistent type detection.


````python
df = pd.read_csv(
    "data.csv",
    low_memory=False
)
````

## 22. `on_bad_lines`

**Definition:** Handles malformed rows in the CSV file.


````python
pd.read_csv(
    "../datasets/titanic.csv",
    on_bad_lines="skip"
)
````

## 23. `quotechar`

**Definition:** Specifies the character used for quoted strings.


````python
df = pd.read_csv(
    "data.csv",
    quotechar='"'
)
````

## 24. `comment`

**Definition:** Ignores lines that start with the specified comment character.


````python
df = pd.read_csv(
    "data.csv",
    comment="#"
)
````
