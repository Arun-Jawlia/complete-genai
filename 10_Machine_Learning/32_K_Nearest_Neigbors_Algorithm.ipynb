# `K-Nearest Neighbor ( KNN )`
---

## `What is KNN?`

**K-Nearest Neighbors (KNN)** is a **supervised machine learning algorithm** used for both:

- **Classification**
- **Regression**

It works by finding the **nearest data points** to a new input and using them to make a prediction.

### Scikit-learn Classes

- `KNeighborsClassifier` → for classification
- `KNeighborsRegressor` → for regression

---

# `Why is KNN called a Lazy Learner?`

KNN does not build a complex model during training.

It simply stores the training data and makes predictions when new data arrives.

So, it is called a **lazy learning algorithm**.

---

# `How KNN Works`

## `Step 1: Choose the value of k`

`k` is the number of nearest neighbors used for prediction.

### Example

If `k = 3`, the model looks at the 3 nearest data points.

---

## `Step 2: Calculate distance`

KNN finds how close the new point is to all training points.

### Common Distance Metrics

#### `1. Euclidean Distance`

Used most often.

Formula:

$d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$.

It measures straight-line distance.

---

#### `2. Manhattan Distance`

Formula:

$d = |x_2 - x_1| + |y_2 - y_1|$.

It measures distance by moving along grid-like paths.

---

## `Step 3: Find the nearest neighbors`

After calculating distances, KNN selects the `k` closest points.

---

## `Step 4: Make prediction`

### For Classification

The model takes a **majority vote**.

Example:

- 2 neighbors → Class A
- 1 neighbor → Class B

Final prediction → **Class A**

---

### For Regression

The model takes the **average** of the nearby values.

Example:

- Neighbor house prices = 40, 50, 60
- Predicted price = 50

---

# Example: House Price Prediction

Suppose we want to predict the price of a house using nearby houses.

Training data:

| Area | Bedrooms | Price |
|------|----------|-------|
| 1000 | 2 | 40 |
| 1200 | 3 | 45 |
| 1500 | 3 | 60 |

For a new house, KNN finds the nearest houses and predicts the price based on them.

---

# `Choosing the Right Value of K`

## If `k` is too small

- Model becomes very sensitive to noise
- Can cause **overfitting**

Example:
- `k = 1` may memorize training data too closely

---

## If `k` is too large

- Model becomes too simple
- Can cause **underfitting**

Example:
- `k = 50` may ignore local patterns

---

## Best Practice

Use **fine-tuning** to find the best value of `k`.

Common method:

- Try multiple `k` values
- Evaluate performance
- Choose the best one

---

# `Advantages of KNN`

- Easy to understand
- Simple to implement
- Works for both classification and regression
- Useful for small datasets
- No explicit training phase

---

# `Limitations of KNN`

## 1. Sensitive to Outliers

Outliers can affect the distance calculation and wrong predictions may happen.

---

## 2. Not Good for Large Datasets

KNN must compare the new point with many training points, so it becomes slow on big datasets.

---

## 3. Sensitive to Missing Values

Missing data can affect distance computation.

---

## 4. Needs Feature Scaling

If one feature has much larger values than others, it can dominate the distance.

Example:

- Age: 25
- Salary: 500000

Without scaling, Salary may dominate the result.

---

# When Should We Use KNN?

Use KNN when:

- Dataset is small or medium-sized
- Features are properly scaled
- Data is not highly noisy
- You want a simple baseline model

---

# `Use Cases of KNN`

## 2. Image Recognition

KNN can be used in handwritten digit recognition.

Example:
