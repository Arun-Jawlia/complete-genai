## `Classification Algorithms`

Classification is a supervised machine learning task where the target/output is a category or class label instead of a continuous number. Examples include predicting whether a student is depressed or not, whether an email is spam or not, or whether a customer will churn.

Common classification algorithms:

1. Logistic Regression: predicts the probability of a class.
2. Decision Tree Classifier: splits data using decision rules.
3. Random Forest Classifier: combines many decision trees for stronger predictions.
4. XGBoost Classifier: uses boosted trees for high-performance classification.
5. AdaBoost Classifier: improves weak classifiers step by step.
6. CatBoost Classifier: works well on tabular and categorical data.
7. K-Neighbors Classifier: predicts based on nearby data points.
8. Support Vector Classifier: separates classes using a decision boundary.
9. Bernoulli Naive Bayes: works well with binary features.
10. Multinomial Naive Bayes: commonly used for count-based text data.
11. Bagging Classifier: averages multiple models trained on random samples.
12. Stacking Classifier: combines multiple models using a final meta-model.

Evaluation metrics:

1. Accuracy Score: percentage of correct predictions.
2. Precision: how many predicted positives are actually positive.
3. Recall: how many actual positives were correctly found.
4. F1-Score: balance between precision and recall.
5. Confusion Matrix: table showing correct and incorrect class predictions.

## Feature Selection and Feature Encoding 

```python
st = st[['Gender', 'Age', 'City', 'Profession', 'Academic Pressure', 'CGPA', 
    'Study Satisfaction', 'Sleep Duration', 'Dietary Habits', 'Work/Study Hours',
       'Financial Stress', 'Family History of Mental Illness', 'Depression']]
```

```python
st.info()
```

## Label Encoding 

```python
from sklearn.preprocessing import LabelEncoder
l = LabelEncoder()
```

```python
st.head()
```

```python
st['Gender'] = l.fit_transform(st['Gender'])
st['City'] = l.fit_transform(st['City'])
st['Profession'] = l.fit_transform(st['Profession'])
st['Sleep Duration'] = l.fit_transform(st['Sleep Duration'])
st['Dietary Habits'] = l.fit_transform(st['Dietary Habits'])
st['Family History of Mental Illness'] = l.fit_transform(st['Family History of Mental Illness'])
```

```python
st.head()
```

```python
st.info()
```

```python
st['Financial Stress'].value_counts()

# check for values in financial stress
```

```python
# there are some field with ?, we have to  remove
st = st[st['Financial Stress'] != '?']
```

```python
st['Financial Stress'] = st['Financial Stress'].astype('float32')
```

## Separating Input and Output Data

```python
st.columns
```

```python
X = st[['Gender', 'Age', 'City', 'Profession', 'Academic Pressure', 'CGPA',
       'Study Satisfaction', 'Sleep Duration', 'Dietary Habits',
       'Work/Study Hours', 'Financial Stress',
       'Family History of Mental Illness',]]
```

```python
Y = st['Depression']
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
```

```python
X = ss.fit_transform(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

# `11. Stacking Classifier -->  Ensemble Learning`
- we can choose the specific models to come up with the Final Class

```python
from sklearn.ensemble import StackingClassifier, RandomForestClassifier, BaggingClassifier, AdaBoostClassifier

# initial estimators
estimators = [('rf', RandomForestClassifier()), ('bc', BaggingClassifier())]

stack = StackingClassifier(estimators= estimators, final_estimator= AdaBoostClassifier())
```

```python
stack.fit(X_train, Y_train)
```

```python
y_predict = stack.predict(X_test)
accuracy_score(Y_test, y_predict)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
```

```python
param_grid = {'n_estimators': [100,200], "max_depth": [4,6,8]}
grid = GridSearchCV(RandomForestRegressor(), param_grid, cv=5)
```

```python
grid.fit(X_train, Y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
y_predict = grid.predict(X_test)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor

param_grid = {'n_estimators': [100,200], "max_depth": [4,6,8]}
random = RandomizedSearchCV(RandomForestRegressor(), param_distributions=param_grid , n_iter= 10, cv=5)

```

```python
random.fit(X_train, Y_train)
```

```python
print(random.best_params_)
print(random.best_score_)
```
