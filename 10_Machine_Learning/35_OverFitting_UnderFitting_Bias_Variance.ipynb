# `OverFitting, UnderFitting, Bias and Variance`
---

## `Introduction`

When we train a Machine Learning model, the main goal is not just to perform well on the training data, but also to perform well on new, unseen data.

A good model should:

- Learn the real pattern from data
- Avoid memorizing noise
- Generalize well on testing data

Two common problems in ML are:

- **Underfitting**
- **Overfitting**

These are closely related to:

- **Bias**
- **Variance**

---

# `Understanding the Goal`

## What do we want?

We want a model that is:

- Not too simple
- Not too complex
- Able to learn the real pattern
- Able to perform well on both training and testing data

This is called a **Generalized Model**.

> **`Generalized Model = Low Bias + Low Variance`**

---

# `1. Underfitting`

## Definition

**Underfitting** happens when the model is too simple to learn the pattern from the training data.

The model does not learn enough, so it performs poorly on both:

- Training data
- Testing data

### Simple Meaning

> The model is not able to understand the data properly.

---

## Example: Training a Dog

Suppose you are training a dog:

- You give a command: **"Sit"**
- But the dog does not understand the command properly
- It keeps behaving incorrectly even after training

This is similar to **underfitting**.

The dog has not learned enough.

---

## Signs of Underfitting

- Low training accuracy
- Low testing accuracy
- High training error
- High testing error

---

## Reasons for Underfitting

1. **Model is too simple**
   - Example: using a linear model for non-linear data

2. **Wrong feature selection**
   - Important features are missing

3. **Less training time**
   - The model did not get enough time to learn patterns

---

## How to Avoid Underfitting

1. Use a **more complex model**
2. Train for **more time**
3. Add **better features**
4. Improve **feature engineering**

---

# `2. Overfitting`

## Definition

**Overfitting** happens when the model learns the training data too well, including noise and unnecessary details.

The model performs very well on training data but poorly on testing data.

### Simple Meaning

> The model memorizes the data instead of learning the real pattern.

---

## Example: Training a Dog

Suppose you teach a dog to sit using only one exact voice tone and one exact hand gesture.

The dog learns that exact pattern very well.

But if you change your voice or hand movement slightly, the dog fails.

This is similar to **overfitting**.

The dog memorized the training pattern instead of understanding the command.

---

## Signs of Overfitting

- Very high training accuracy
- Low testing accuracy
- Low training error
- High testing error

---

## Reasons for Overfitting

1. **Too many features**
2. **Too complex model**
3. **Too much training on small data**
4. **Noise in data**
5. **Model memorizes input-output pairs**

---

## How to Avoid Overfitting

1. Use a **simpler model**
2. Collect **more data**
3. Remove unnecessary features
4. Use **regularization**
5. Use **cross-validation**
6. Use **early stopping** in deep learning

---

# `3. Bias`

## Definition

**Bias** is the error caused by a model making overly simple assumptions.

It tells us how wrong the model is during training.

### High Bias

- Model is too simple
- Model cannot learn the pattern properly
- Training error is high

### Low Bias

- Model learns the training data better
- Training error is low

---

## In Simple Words

> Bias is the training error caused by a model being too simple.

---

# `4. Variance`

## Definition

**Variance** is the error caused by the model changing too much when given different training data.

It tells us how unstable the model is.

### High Variance

- Model is too sensitive to training data
- Model memorizes patterns and noise
- Testing error is high

### Low Variance

- Model gives stable predictions
- Model generalizes well

---

## In Simple Words

> Variance is the testing error caused by a model being too sensitive to the training data.

---

# `5. Relationship Between Bias and Variance`

| Situation | Bias | Variance | Result |
|----------|------|----------|--------|
| Too Simple Model | High | Low | Underfitting |
| Too Complex Model | Low | High | Overfitting |
| Balanced Model | Low | Low | Generalized Model |

---

# `6. Why Underfitting Has High Bias`

Underfitting means the model is too simple.

So it makes a lot of mistakes even on training data.

That means:

- Training error is high
- Bias is high

---

# `7. Why Overfitting Has High Variance`

Overfitting means the model learns training data too closely.

It performs very well on training data but fails on testing data.

That means:

- Training error is low
- Testing error is high
- Variance is high

---

# `8. Generalized Model`

## Definition

A generalized model is a model that performs well on both training and testing data.

It learns the actual pattern without memorizing noise.

### Characteristics

- Low Bias
- Low Variance
- Good training accuracy
- Good testing accuracy

---

# `12. Key Takeaways`

- **Underfitting** means the model is too simple and has **high bias**.
- **Overfitting** means the model is too complex and has **high variance**.
- **Bias** is training error caused by simple assumptions.
- **Variance** is testing error caused by sensitivity to training data.
- The goal of Machine Learning is to build a **generalized model**.
- A good model should have **low bias** and **low variance**.
