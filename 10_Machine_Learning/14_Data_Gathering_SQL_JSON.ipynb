## `2. connect()`
- Function which can help us with connecting to our database --> connect()
- Parameter:
  - host --> server --> localhost
  - username
  - password
  - connect

```python
data = mysql.connector.connect(host='localhost', user='root', password = '', database = 'world')
```

```python
data
```

## `2. read_sql_query()`
- sql query
- database connection variable

```python
df = pd.read_sql_query('SELECT * FROM city', data)
```

```python
df.head()
```

```python
pd.read_sql_query("SELECT * FROM city WHERE Population > 1000000 ", data)
```

```python
ind = pd.read_sql_query("SELECT * FROM city WHERE CountryCode like 'Ind' ", data)
```

```python
ind
```

```python
pd.read_sql_query("SELECT * FROM country", data)
```
