# `Support Vector Machine`
---

## `What is SVM?`

**Support Vector Machine (SVM)** is a **supervised machine learning algorithm** used for both:

- **Classification**
- **Regression**

Its main goal is to find the **best separating boundary** between different classes.

> **SVM finds the optimal hyperplane that separates classes with the maximum margin.**

---

# `Why is SVM important?`

SVM is useful because it:

- Works well for classification problems
- Can handle linear and non-linear data
- Tries to create a strong decision boundary
- Performs well on smaller and medium-sized datasets

---

# `Key Terms in SVM`

## 1. Hyperplane

A **hyperplane** is the decision boundary that separates different classes.

### In different dimensions:

- **2D** → line
- **3D** → plane
- **4D and above** → hyperplane

### Simple meaning

> It is the boundary used to divide classes.

---

## 2. Margin

The **margin** is the distance between the hyperplane and the nearest data points of each class.

### Goal

SVM tries to make this margin as large as possible.

> **Maximum margin = better separation**

---

## 3. Support Vectors

**Support vectors** are the data points closest to the hyperplane.

These points are important because they help define the position of the boundary.

---

## 4. Marginal Planes

Marginal planes are imaginary boundaries that pass through the support vectors.

- One plane on each side of the hyperplane
- SVM tries to maximize the distance between these planes

---

# `Goal of SVM`

SVM has two main goals:

1. Find a hyperplane that separates classes correctly
2. Maximize the distance between the marginal planes

---

# `Types of Margin in SVM`

## `1. Hard Margin SVM`

### Definition

Hard Margin SVM tries to classify every training point correctly.

### Characteristics

- No misclassification allowed
- Works only when data is perfectly separable
- Sensitive to outliers

### Limitation

Not practical for real-world noisy data.

---

## `2. Soft Margin SVM`

### Definition

Soft Margin SVM allows some misclassification.

### Characteristics

- More flexible
- Works with noisy or overlapping data
- Better for real-world datasets

### Why needed?

Because real-world data is often not perfectly separable.

# `Kernels in SVM`

## What is a Kernel?

A **kernel** is used when the data is **not linearly separable**.

If a straight line cannot separate the classes, SVM uses a kernel to transform the data into a higher-dimensional space.

---

## `1. Linear Kernel`

```python
SVC(kernel="linear")
```

### When to use?

- When data is linearly separable

### Meaning

A straight line or hyperplane can separate the classes.

---

## `2. Polynomial Kernel`

```python
SVC(kernel="poly")
```

### When to use?

- When the relationship is not linear
- When curved boundaries are needed

### Meaning

It creates a polynomial decision boundary.

---

## 3.`RBF Kernel`

```python
SVC(kernel="rbf")
```

### When to use?

- When the relationship between data points is complex

### Meaning

RBF stands for **Radial Basis Function**.

It is one of the most commonly used kernels.

---

## 4. `Sigmoid Kernel`

```python
SVC(kernel="sigmoid")
```

### When to use?

- Less commonly used
- Produces an S-shaped decision boundary

---

# `Important SVM Parameters`

## `1. C (Regularization)`

### Definition

`C` controls how much the model should try to classify training points correctly.

### High C

- Focuses more on correct classification
- May lead to overfitting
- Very small margin

### Low C

- Allows some misclassification
- May lead to underfitting
- Larger margin

### Typical values

```text
0.001, 0.01, 0.1, 1, 10, 100, 1000
```

### Simple meaning

> High C = less tolerance for errors  
> Low C = more tolerance for errors

---

## `2. Gamma`

### Definition

`gamma` controls how much influence a single training point has.

### High Gamma

- Each point has a strong influence
- More complex boundary
- May lead to overfitting

### Low Gamma

- Each point has less influence
- Simpler boundary
- May lead to underfitting

### Typical values

```text
0.001, 0.01, 0.1, 1, 10, 100, 1000
```

### Simple meaning

> High gamma = very close focus on individual points  
> Low gamma = broader focus on data points
