## What is Naive Bayes?

**Naive Bayes** is a **supervised machine learning algorithm** used for **classification**.

It is based on two ideas:

- **Naive** → assumes all features are independent of each other
- **Bayes** → uses **Bayes' Theorem** to calculate probability

> Naive Bayes is called "naive" because it assumes features do not influence each other.

---

# Why is Naive Bayes Important?

Naive Bayes is:

- Simple and fast
- Works well for text data
- Useful in NLP and classification tasks
- Good for high-dimensional datasets
- Easy to implement

---

# Core Idea

Naive Bayes predicts the class of an input by calculating the probability of each class.

It chooses the class with the **highest probability**.

Example:

- Email → Spam or Not Spam
- Review → Positive or Negative
- News article → Sports, Politics, Business

---

# Naive + Bayes

## `1. Naive`

### Meaning

Features are assumed to be **independent** of each other.

### Example

If we are predicting whether an email is spam:

- Word "free"
- Word "win"
- Word "offer"

Naive Bayes assumes these words are independent.

---

## `2. Bayes`

### Meaning

Bayes part uses **conditional probability** to calculate the likelihood of an event.

### Bayes' Theorem

\[
P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}
\]

Where:

- **P(A|B)** = Probability of A given B
- **P(B|A)** = Probability of B given A
- **P(A)** = Prior probability of A
- **P(B)** = Prior probability of B

---

# `Independent Event vs Dependent Event`

## Independent Event

Two events are independent if one event does not affect the other.

### Example: Dice Roll

When rolling a dice, the probability of getting 1 is:

\[
\frac{1}{6}
\]

Every roll is independent of the previous roll.

### Key Point

- One event does not change the probability of another event.

---

## Dependent Event

Two events are dependent if one event affects the other.

### Example: Balls in a Box

Suppose there are:

- 4 black balls
- 3 white balls

Total balls = 7

### `Probability of picking a black ball`

\[
P(\text{black}) = \frac{4}{7}
\]

If one ball is removed, the probability changes.

### Key Point

- The probability changes based on the previous event.

---

# `Why Does Naive Bayes Assume Independence?`

The algorithm becomes much simpler and faster when features are independent.

In real-world data, features may not always be perfectly independent, but Naive Bayes still works surprisingly well in many cases.

---

# `Types of Naive Bayes`

There are 3 common types of Naive Bayes algorithms.

---

## `1. Gaussian Naive Bayes`

### Definition

Used when input features are **continuous numerical values**.

### Example Use Cases

- Medical diagnosis
- Numeric prediction-related classification
- Any dataset with continuous features

### Python

```python
from sklearn.naive_bayes import GaussianNB
```

---

## `2. Bernoulli Naive Bayes`

### Definition

Used when features are **binary**.

### Example Use Cases

- Spam detection
- Email filtering
- Binary text presence/absence features

### Example

- Word present = 1
- Word absent = 0

---

## `3. Multinomial Naive Bayes`

### Definition

Used when features are based on **word counts or frequencies**.

### Example Use Cases

- Sentiment analysis
- Text classification
- News categorization
- Document classification

### Example

Count of positive words vs negative words

---

### NLP Use Cases

- Language classification

# How Naive Bayes Works

1. Take input features
2. Calculate class probabilities
3. Apply Bayes' Theorem
4. Assume features are independent
5. Choose the class with the highest probability

---

# `Advantages of Naive Bayes`

- Simple and fast
- Works well for text data
- Good baseline model
- Handles high-dimensional data well
- Performs well even with limited training data

---

# `Limitations of Naive Bayes`

- Assumes feature independence
- Real-world features may be correlated
- Not always accurate for complex problems
- Probability estimates may be over-simplified

---

# `Naive Bayes vs Other Classification Algorithms`

| Naive Bayes | Logistic Regression |
|-------------|----------------------|
| Based on probability | Based on sigmoid and probability |
| Assumes feature independence | Does not assume independence |
| Very fast | Also effective for classification |
| Works great for text data | Good for structured data |
