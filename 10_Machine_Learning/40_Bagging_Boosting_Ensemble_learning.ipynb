# `Ensemble Learning`
---

## Definition

**Ensemble Learning** is a technique where multiple machine learning models are combined to produce a better result than a single model.

Instead of trusting one model, we combine the predictions of many models.

In Simple term, Ensemble learning is like taking opinions from multiple experts instead of relying on a single expert.


There is two types of Ensemble Learning
1. Bagging  : Random Forest, 
2. Boosting : XgBoost , CatBoost, AdaBoost, Gradient Boost

With Both Bagging and Boosting we can have capability to solve Both Regression and Classification

## Bagging ( BootStraped)
- In Bagging Ensemble Learning , we have multiple independent models in parrallel by training them on different bootstrapped ( random sample with replacement) subset of data and then it aggregates their output
- advantage: Reduce Overfitting and Variance, can handle high dimensions and large datasets, simple to implement
- Disadvantage: it does not reduce bias value
- Algorithm: Random Forest Algo

## Boosting:
- AdaBoost, XgBoost, GradientBoost, CatBoost
- In boosting Ensemble learning, boosting builds model sequentially where model learns to correct error of the previous model
- The final prediction is the weighted sum of all model
- Boosting focus on reducing underfitting
- Working of boosting: in case of boosting, 
    - train a weak learner
    - identify and focus on misclassify or poorly predicted sample data 
    - train the Next model on updated datasets
    - Repeat for many learners
    - Combine all learners with weighted majority vote classification or weighted sum ( Regression )
- Algorithm: AdaBoost, XgBoost, CatBoost, Gradient Boost

# `Ensemble Learning: Bagging and Boosting`

## `What is Ensemble Learning?`

**Ensemble Learning** is a Machine Learning technique where **multiple models are combined** to make a better prediction than a single model.

Instead of relying on one model, Ensemble Learning combines the predictions of several models to improve:

- Accuracy
- Stability
- Generalization
- Robustness

> **Ensemble Learning = Many models working together to produce a better result.**

---

# `Why Do We Need Ensemble Learning?`

A single Machine Learning model may suffer from:

- Overfitting
- Underfitting
- High variance
- High bias

Ensemble Learning helps overcome these problems by combining multiple models.

---

# `Real-Life Analogy`

Imagine a patient visits a hospital.

Instead of asking **one doctor**, the hospital asks **10 experienced doctors**.

Each doctor gives their opinion.

The final decision is based on the combined opinion.

This usually leads to a more reliable diagnosis.

This is exactly how Ensemble Learning works.

# `Types of Ensemble Learning`

There are two major types:

1. Bagging (Bootstrap Aggregating)
2. Boosting

---

# `1. Bagging (Bootstrap Aggregating)`

## Definition

**Bagging** trains **multiple models independently and in parallel** using different random subsets of the training data.

The final prediction is obtained by combining the predictions of all models.

---

# How Bagging Works

```code
Original Dataset
        │
        ▼
Bootstrap Sampling
        │
        ├─────────────┐
        ▼             ▼
 Decision Tree 1   Decision Tree 2
        │             │
        ▼             ▼
 Decision Tree 3   Decision Tree 4
        │
        ▼
Combine Predictions
        │
        ▼
Final Prediction
```

---

# Bootstrap Sampling

Bagging uses **Bootstrap Sampling**.

It means:

- Randomly select data
- Sampling is done **with replacement**
- One record may appear multiple times
- Some records may not appear

---

# Final Prediction in Bagging

## Classification

Uses **Majority Voting**

Example

```code
Tree 1 → Yes

Tree 2 → Yes

Tree 3 → No

Tree 4 → Yes

Final Prediction

↓

Yes
```

---

## Regression

Uses **Average Prediction**

Example

```code
Tree 1 → 52

Tree 2 → 50

Tree 3 → 55

Average

↓

52.3
```

---

# Advantages of Bagging

- Reduces overfitting
- Reduces variance
- Improves stability
- Works well with high-variance models

`# 2. Boosting`
---

## Definition

**Boosting** trains models **sequentially**, where each new model focuses on correcting the mistakes made by the previous model.

Instead of training independently, each model learns from previous errors.

---

# How Boosting Works

```code
Original Data
      │
      ▼
Model 1
      │
Mistakes
      ▼
Model 2
      │
Mistakes
      ▼
Model 3
      │
Mistakes
      ▼
Final Strong Model
```

---

# Main Idea of Boosting

Every new model tries to improve the previous one.

Weak learners are combined to create a strong learner.

---

# Advantages of Boosting

- High accuracy
- Reduces bias
- Learns complex relationships
- Excellent performance on structured/tabular data

---

# Limitations of Boosting

- Can overfit if not tuned
- Slower than Bagging
- Sensitive to noisy data

---

# `Popular Boosting Algorithms`

## `1. AdaBoost (Adaptive Boosting)`

### Definition

AdaBoost trains multiple weak learners sequentially.

After each model:

- Incorrectly classified samples receive higher importance.
- The next model focuses more on those difficult samples.

### Best For

- Simple classification problems

---

## `2. Gradient Boosting`

### Definition

Each new model learns from the **errors (residuals)** of the previous model.

Instead of directly learning the target, it learns to reduce prediction errors.

### Best For

- Regression
- Classification

---

## `3. XGBoost (Extreme Gradient Boosting)`

### Definition

An optimized and faster implementation of Gradient Boosting.

### Features

- Regularization
- Parallel processing
- Missing value handling
- High accuracy

### Industry Usage

One of the most widely used algorithms in Machine Learning competitions and production systems.

---

## `4. LightGBM`

### Definition

A fast Gradient Boosting algorithm developed by Microsoft.

### Features

- Faster training
- Lower memory usage
- Efficient for very large datasets

---

## `5. CatBoost`

### Definition

A Gradient Boosting algorithm developed by Yandex.

### Features

- Handles categorical features automatically
- Requires minimal preprocessing
- Excellent for datasets with many categorical variables

---

# `Bagging vs Boosting`

| Bagging | Boosting |
|----------|-----------|
| Models are trained independently | Models are trained sequentially |
| Reduces variance | Reduces bias |
| Parallel training | Sequential training |
| Less prone to overfitting | Can overfit if not tuned |
| Faster | Slower |
| Example: Random Forest | Example: XGBoost |
