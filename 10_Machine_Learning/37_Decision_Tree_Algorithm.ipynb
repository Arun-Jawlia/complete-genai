# `Decision Tree Algorithm`

## `What is a Decision Tree?`

A **Decision Tree** is a **supervised machine learning algorithm** used for both:

- **Classification**
- **Regression**

It makes predictions by asking a series of **yes/no questions** in a tree-like structure.

> **Decision Tree = Flowchart of decisions**

---

## `Why is it called a Tree?`

Because the model starts from one point and keeps splitting into smaller parts, just like a tree branches out.

- **Top node** → root
- **Middle nodes** → decisions
- **Bottom node** → final result

---

# `Decision Tree Structure`

## 1. Root Node

- The **starting node**
- Represents the **entire dataset**
- The first split happens here

---

## 2. Decision Node

- A node where a **split happens based on a feature**
- Each decision node asks a question like:
  - Is credit score > 700?
  - Is age > 30?
  - Is salary > 50,000?

---

## 3. Leaf Node

- The **final node**
- Also called the **terminal node**
- Gives the **final prediction**

Examples:

- Yes / No
- Approved / Rejected
- Class A / Class B

---

## 4. Branch

- The line connecting nodes
- Shows the outcome of a decision

Example:

- True branch
- False branch

---

## 5. Pruning

- Removing unnecessary branches
- Helps reduce **overfitting**

> **Pruning makes the tree simpler and more general**

---

# `How a Decision Tree Works`

1. Start at the **root node**
2. Choose the **best feature** to split the data
3. Split the data into smaller groups
4. Repeat the process for each child node
5. Stop when:
   - the node becomes pure
   - maximum depth is reached
   - no useful split is left
6. Predict using the **leaf node**

---

# `Decision Tree for Classification and Regression`

## Classification Tree

Used when the target is a **category**

Examples:

- Loan Approved / Not Approved
- Spam / Not Spam
- Disease / No Disease

---

## Regression Tree

Used when the target is a **continuous value**

Examples:

- House Price
- Salary
- Sales Forecast

---

# `How Does the Tree Decide the Best Split?`

The tree chooses the best feature using criteria such as:

- **Gini Impurity**
- **Entropy**
- **Information Gain**

These help the model split the data in a way that reduces confusion.

---

# `Pure Split vs Impure Split`

## Pure Split

A node is **pure** when it contains only one class.

Example:

- All samples are **Yes**
- All samples are **No**

This is a good split.

---

## Impure Split

A node is **impure** when it contains a mix of classes.

Example:

- 6 Yes
- 4 No

This is not a pure split.

---

# `Entropy`

## What is Entropy?

**Entropy** measures the **uncertainty** or **messiness** in the data.

- High entropy → more mixed classes
- Low entropy → more pure classes

> Entropy tells us how disordered the data is.

---

## Entropy Formula

```text
Entropy = - Σ(p_i log2 p_i)
```

Where:

* `p_i` = probability of class `i`

---

## Meaning of Entropy

| Condition  | Entropy |
| ---------- | ------- |
| Pure node  | 0       |
| Mixed node | High    |

---

# `Gini Impurity`

## What is Gini Impurity?

**Gini Impurity** measures how likely a randomly chosen sample would be misclassified.

> Lower Gini = better purity

---

## Gini Formula

```text
Gini = 1 - Σ(p_i²)
```

Where:

* `p_i` = probability of class `i`

---

## Meaning of Gini Impurity

| Condition  | Gini         |
| ---------- | ------------ |
| Pure node  | 0            |
| Mixed node | Higher value |

---

# Entropy vs Gini Impurity

| Metric        | Meaning                                | Best For                                        |
| ------------- | -------------------------------------- | ----------------------------------------------- |
| Entropy       | Measures uncertainty                   | Information gain based splitting                |
| Gini Impurity | Measures misclassification probability | Fast splitting, commonly used in large datasets |

---

# `Information Gain`

## What is Information Gain?

**Information Gain** tells us how much uncertainty is reduced after a split.

> It helps the tree decide which feature to split first.

### Idea

* High Information Gain → good split
* Low Information Gain → weak split

---

## Information Gain Formula

```text
Information Gain = Entropy(before split) - Entropy(after split)
```

---

# Why is Information Gain Important?

It helps answer:

* Which feature should be used first?
* Which split reduces uncertainty the most?

# `Why Use Decision Trees?`

* Easy to understand
* Easy to visualize
* Works for both classification and regression
* Handles numerical and categorical features
* No need for feature scaling

---

# `Limitations of Decision Trees`

* Can overfit easily
* Sensitive to small changes in data
* Can grow too large if not controlled
* Not always the most accurate model by itself

---

# `How to Reduce Overfitting`

Use:

* **Pruning**
* **Max depth**
* **Min samples split**
* **Min samples leaf**
