# **`Exploratory Data Analysis ( EDA )`**
---

## `Introduction`

Exploratory Data Analysis (EDA) is one of the most important steps in the Machine Learning pipeline.

It helps us understand the dataset before training a model by analyzing its structure, quality, and relationships between variables.

> **EDA = Understanding your data before building a Machine Learning model.**

---

# `Why is EDA Required in Machine Learning?`

Traditional Machine Learning algorithms **cannot automatically understand which features are important**.

Therefore, it is the responsibility of the Data Scientist to:

- Explore every feature (column)
- Detect data quality issues
- Understand feature relationships
- Select important features
- Remove unnecessary features

Without EDA, the model may learn incorrect patterns and produce poor predictions.

---

# `Machine Learning vs Deep Learning`

## Machine Learning (ML)

### Characteristics

- Requires manual feature extraction.
- Requires Exploratory Data Analysis (EDA).
- Requires Feature Engineering.
- Performs well on structured/tabular data.
- Usually needs less data than Deep Learning.

### Workflow

```
Raw Data
    ↓
EDA
    ↓
Feature Engineering
    ↓
Machine Learning Algorithm
    ↓
Prediction
```

## Deep Learning (DL)

### Characteristics

- Automatically learns features from raw data.
- Little or no manual feature engineering for many tasks.
- Little or no manual feature extraction for images, audio, and text.
- Works best with very large datasets.
- Learns hierarchical representations automatically.

### Workflow

```
Raw Images
      ↓
Deep Learning Model
      ↓
Automatic Feature Learning
      ↓
Prediction
```

---

# `Example: Cat vs Dog Image Classifier`

Suppose we want to classify images as either a **Cat** or a **Dog**.

## Machine Learning Approach

Before training the model, we must manually extract features such as:

- Color
- Shape
- Texture
- Edge information
- Image size

These extracted features are then provided to the ML algorithm.

```
Image
   ↓
Feature Extraction
   ↓
Machine Learning Model
   ↓
Cat / Dog
```

---

## Deep Learning Approach

Deep Learning models learn these features automatically.

```text
Image
   ↓
Deep Neural Network
   ↓
Cat / Dog
```

No manual feature extraction is required for image classification tasks.

---

# `Why Doesn't Deep Learning Need Much EDA?`

Deep Learning models, especially Convolutional Neural Networks (CNNs), can automatically learn useful features from raw images.

Instead of manually identifying edges or shapes, the network discovers them during training.

However, **basic EDA is still recommended** to:

- Check missing or corrupted data.
- Verify class balance.
- Remove duplicate images.
- Detect labeling errors.

> **Note:** While feature engineering is largely automated in Deep Learning, data quality checks and basic EDA are still important.

---

# `Deep Learning is Inspired by the Human Brain`

Deep Learning is inspired by the structure and functioning of the human brain.

It consists of multiple interconnected artificial neurons that process information layer by layer.

This structure is called an **Artificial Neural Network (ANN)**.

---

# `Perceptron`

## Definition

A **Perceptron** is the simplest artificial neuron and serves as the building block of Artificial Neural Networks.

Multiple perceptrons form layers, and multiple layers create a Deep Neural Network.

---

# `How Deep Learning Learns Features`

Deep Learning learns features hierarchically.

### First Layer

Learns simple features:

- Edges
- Lines
- Corners

---

### Second Layer

Combines edges to identify:

- Curves
- Shapes
- Patterns

---

### Third Layer

Combines shapes to recognize:

- Eyes
- Nose
- Ears
- Fur patterns

---

### Deeper Layers

Recognize complete objects.

Example:

```text
Edges
   ↓
Shapes
   ↓
Body Parts
   ↓
Complete Animal
   ↓
Cat
```

This automatic learning process is called **Hierarchical Feature Learning**.

---

# `Why Does Machine Learning Need EDA?`

Machine Learning algorithms cannot automatically determine:

- Which features are useful.
- Which features are irrelevant.
- Which values are outliers.
- Which variables have strong relationships.

Therefore, Data Scientists perform EDA before training.

---

# Objectives of EDA

EDA helps us:

- Understand the dataset.
- Check data quality.
- Detect missing values.
- Detect duplicate records.
- Detect outliers.
- Understand feature distributions.
- Study relationships between variables.
- Validate assumptions.
- Prepare data for feature engineering.

---

# Detecting Outliers

## `What is an Outlier?`

An outlier is a value that is significantly different from the rest of the data.

### Example

```
Student Marks

78
82
81
79
80
5
83
84
```

The value **5** is an outlier because it is far from the other observations.

Outliers can negatively affect many Machine Learning algorithms.

EDA helps identify and handle them before model training.

---

# `EDA Helps Validate Assumptions`

Many Machine Learning algorithms assume that the data satisfies certain conditions.

EDA is used to verify these assumptions before applying the algorithm.

### Common Assumptions

- No excessive missing values.
- No significant outliers.
- Correct data types.
- Reasonable feature distributions.
- Meaningful relationships between variables.

---

|------------------|---------------|
| Manual feature extraction | Automatic feature extraction |
| Requires EDA | Basic EDA still recommended, but less feature engineering |
| Requires feature engineering | Learns features automatically |

| Faster training | Longer training time |
| Better for structured/tabular data | Better for images, audio, and text |

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is EDA?

**Answer:**
Exploratory Data Analysis (EDA) is the process of analyzing and understanding a dataset before building a Machine Learning model.

---

### Q2. Why is EDA important?

**Answer:**
EDA helps identify missing values, outliers, duplicate records, feature distributions, and relationships between variables, improving data quality and model performance.

---

### Q3. Why does Machine Learning require EDA?

**Answer:**
Most Machine Learning algorithms cannot automatically identify important features or detect data quality issues, so EDA is needed to prepare the data.

---

### Q4. What is feature engineering?

**Answer:**
Feature engineering is the process of creating, selecting, or transforming input features to improve a model's performance.

---

### Q5. What is a perceptron?

**Answer:**
A perceptron is the simplest artificial neuron and the basic building block of Artificial Neural Networks.

---

## Intermediate Questions

### Q6. Does Deep Learning require EDA?

**Answer:**
Deep Learning performs automatic feature extraction, so it requires much less feature engineering. However, basic EDA—such as checking for missing values, duplicates, corrupted data, and class imbalance—is still recommended.

---

### Q7. Why is Deep Learning better for image classification?

**Answer:**
Deep Learning models automatically learn hierarchical features such as edges, shapes, textures, and complete objects from raw images.

---

### Q8. What is hierarchical feature learning?

**Answer:**
It is the process where deep neural networks learn simple features (edges), combine them into complex features (shapes), and eventually recognize complete objects.

---

## Scenario-Based Questions

### Q9. You are building a house price prediction model using tabular data. Should you perform EDA?

**Answer:**
Yes. For structured data, EDA is essential to understand feature distributions, detect outliers, handle missing values, and select relevant features.

---

### Q10. You are building a Cat vs Dog image classifier using a CNN. Do you need manual feature extraction?

**Answer:**
No. CNNs automatically learn image features such as edges, shapes, and textures. However, you should still perform basic EDA to ensure data quality.
