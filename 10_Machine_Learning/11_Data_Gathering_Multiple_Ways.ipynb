## Important Parameters and Ways of Data Gathering

Data gathering means collecting data from different sources before analysis or model building. In machine learning, data can come from files, databases, APIs, websites, forms, logs, sensors, or cloud storage.

Below are the most important parameters and methods, with definition, syntax, and code examples.


### 1. CSV / TSV Files using `pd.read_csv()`

**Definition:** Reads tabular data from a CSV, TSV, or text file into a pandas DataFrame.

**Syntax:**

```python
pd.read_csv(filepath_or_buffer, sep=",", header=0, names=None, index_col=None, usecols=None, nrows=None, encoding=None)
```

**Important parameters:**

| Parameter | Definition | Example |
|---|---|---|
| `filepath_or_buffer` | Path or URL of the file | `"Datasets/titanic.csv"` |
| `sep` | Separator between values | `sep=","`, `sep="\t"` |
| `header` | Row number to use as column names | `header=0` |
| `names` | Custom column names | `names=["id", "name"]` |
| `index_col` | Column to make DataFrame index | `index_col="PassengerId"` |
| `usecols` | Read selected columns only | `usecols=["Name", "Age"]` |
| `nrows` | Read first N rows only | `nrows=5` |
| `encoding` | File encoding | `encoding="utf-8"` |


````python
import pandas as pd

# Basic CSV reading
df = pd.read_csv("Datasets/titanic.csv")
df.head()
````

````python
# Important read_csv parameters in one place
titanic_small = pd.read_csv(
    "Datasets/titanic.csv",
    sep=",",
    header=0,
    index_col="PassengerId",
    usecols=["PassengerId", "Survived", "Pclass", "Name", "Sex", "Age"],
    nrows=10,
    encoding="utf-8"
)

titanic_small
````

### 4. SQL Database using `pd.read_sql()`

**Definition:** Reads data from a SQL database using a SQL query.

**Syntax:**

```python
pd.read_sql(sql, con)
```

**Important parameters:** `sql` is the query, and `con` is the database connection.


````python
import sqlite3

# Create a small temporary database in memory
connection = sqlite3.connect(":memory:")

employees = pd.DataFrame({
    "emp_id": [1, 2, 3],
    "name": ["Amit", "Neha", "Rahul"],
    "salary": [500000, 800000, 700000]
})

employees.to_sql("employees", connection, index=False, if_exists="replace")

# Gather data using SQL query
high_salary = pd.read_sql(
    "SELECT * FROM employees WHERE salary >= 700000",
    connection
)

high_salary
````

### 5. API Data using `requests`

**Definition:** API data gathering means sending a request to a server and receiving data, usually in JSON format.

**Syntax:**

```python
requests.get(url, params=None, headers=None, timeout=10)
```

**Important parameters:** `url`, `params`, `headers`, `timeout`.


````python
import requests

url = "https://api.github.com/repos/pandas-dev/pandas"

response = requests.get(url, timeout=10)
data = response.json()

api_df = pd.DataFrame([{
    "repository": data.get("full_name"),
    "stars": data.get("stargazers_count"),
    "forks": data.get("forks_count"),
    "language": data.get("language")
}])

api_df
````

### 7. Manual Data Gathering using Dictionary / List

**Definition:** Small datasets can be created manually using Python dictionaries or lists. This is useful for practice, testing, and demos.

**Syntax:**

```python

```


````python
student_data = {
    "name": ["Amit", "Neha", "Rahul"],
    "age": [21, 22, 20],
    "marks": [85, 92, 78]
}

students_df = pd.DataFrame(student_data)
students_df
````
