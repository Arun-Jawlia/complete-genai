## `What is a Machine Learning Pipeline?`

### Definition

A **Machine Learning Pipeline** is a sequence of steps that automates the complete Machine Learning workflow, from raw data to model prediction.

In simple words:

> **A Machine Learning Pipeline connects all preprocessing steps and the Machine Learning model into one workflow.**

Instead of manually performing each step, a pipeline executes them automatically in the correct order.

---

# `Why Do We Need a Machine Learning Pipeline?`

Without a pipeline, we have to manually:

- Clean the data
- Encode categorical variables
- Scale numerical features
- Train the model
- Predict new data

This is:

- Time-consuming
- Error-prone
- Difficult to maintain

A pipeline automates all these tasks.

# `Complete Example`

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

preprocessor = ColumnTransformer([
    ("num", num_pipeline, numerical_columns),
    ("cat", OneHotEncoder(), categorical_columns)
])

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression())
])

pipeline.fit(X_train, y_train)
```

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is a Machine Learning Pipeline?

**Answer:**
A Machine Learning Pipeline is a sequence of automated steps that preprocess data, train a model, and make predictions in a single workflow.

---

### Q2. Why do we use a Pipeline?

**Answer:**
To automate preprocessing and model training, reduce code duplication, improve maintainability, and prevent data leakage.

---

### Q3. Which Scikit-learn class is used to create a Pipeline?


---

### Q4. What preprocessing steps are commonly included in a Pipeline?

**Answer:**

* Missing Value Imputation
* Feature Encoding
* Feature Scaling

---

### Q5. Can ColumnTransformer be used inside a Pipeline?

**Answer:**
Yes. `ColumnTransformer` is often used within a `Pipeline` to apply different preprocessing techniques to different feature types before model training.

---

## Intermediate Questions

### Q6. What is data leakage, and how does a Pipeline help prevent it?

**Answer:**
Data leakage occurs when information from the test set is unintentionally used during training, leading to overly optimistic performance. A Pipeline ensures that preprocessing is **fit only on the training data** and then applied to the test data, preventing leakage.

---

### Q7. Why are Pipelines considered production-ready?

**Answer:**
They encapsulate preprocessing and model training in a single reusable object, making the workflow easier to test, deploy, and maintain consistently.

---

### Q8. Can a Pipeline contain multiple preprocessing steps?

**Answer:**
Yes. A Pipeline can include several preprocessing steps (imputation, encoding, scaling) followed by a machine learning model.

---

## Scenario-Based Questions

### Q9. You have missing values, categorical variables, and numerical variables in your dataset. How would you build a preprocessing workflow?

**Answer:**
Use a **ColumnTransformer** to:

* Impute and scale numerical features.
* Encode categorical features.

Then place the `ColumnTransformer` inside a `Pipeline` with the chosen machine learning model.

---

### Q10. Why is a Pipeline recommended before deploying a model?

**Answer:**
Because it ensures that every new input undergoes the same preprocessing steps used during training, resulting in consistent and reliable predictions.

### `1. First Step: Separate data into Multiple Data Categories`

````python
Numeric_Data = ['Age', 'Balance', 'Day', 'Duration', 'Campaign', 'Pdays', 'Previous']

Ordinal_Data = ['Education']

Nominal_Data = ['Job', 'Marital', 'Default', 'Housing', 'Loan', 'contact', 'Month', 'poutcome']
````

````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.preprocessing import StandardScaler
````

### `2. Second Step: Create Pipeline`

````python
Numeric_Pipelinie = Pipeline(steps=[('imputer', SimpleImputer(strategy='mean')), ('scaling', StandardScaler())])
````

````python
Ordinal_Pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent')), ('encoding', OrdinalEncoder(categories=[['unknown', 'primary', 'secondary', 'tertiary']])) ,('scaling', StandardScaler())])
````

````python
Nominal_Pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent')), ('encoding', OneHotEncoder(handle_unknown='ignore'))])
````

### `Third Step: Combine Pipeline Transformer with their respective data`

````python
from sklearn.compose import ColumnTransformer
````

````python
features = ColumnTransformer(transformers=[('Numeric Transformation', Numeric_Pipelinie, Numeric_Data), ('Ordinal Transformation', Ordinal_Pipeline, Ordinal_Data), ('Nominal Transformation', Nominal_Pipeline, Nominal_Data)])
````

### `4. Create an Model with the respective Transformers and along with their Algorithm`

````python
from sklearn.linear_model import LogisticRegression
````

````python
model = Pipeline(steps=[('features', features), ('classifier', LogisticRegression())])
````

````python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    OrdinalEncoder,
    LabelEncoder
)

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report
)

# Load Dataset
df = pd.read_csv("../Datasets/Bank_Marketing_Original_Data.csv")

# Features & Target
X = df.drop("TARGET", axis=1)
y = df["TARGET"]

# Encode Target
le = LabelEncoder()
y = le.fit_transform(y)

# Column Groups
numerical_cols = [
    "Age",
    "Balance",
    "Day",
    "Duration",
    "Campaign",
    "Pdays",
    "Previous"
]

nominal_cols = [
    "Job",
    "Marital",
    "Default",
    "Housing",
    "Loan",
    "contact",
    "poutcome"
]

ordinal_cols = [
    "Education",
    "Month"
]

# Numerical Pipeline
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Nominal Pipeline
nom_pipeline = Pipeline([
    ("encoder",
     OneHotEncoder(
         drop="first",
         handle_unknown="ignore"))
])

# Ordinal Pipeline
education_order = [
    "unknown",
    "primary",
    "secondary",
    "tertiary"
]

month_order = [
    "jan","feb","mar","apr",
    "may","jun","jul","aug",
    "sep","oct","nov","dec"
]

ord_pipeline = Pipeline([
    ("encoder",
     OrdinalEncoder(
         categories=[
             education_order,
             month_order
         ]))
])

# Column Transformer
preprocessor = ColumnTransformer([
    ("num", num_pipeline, numerical_cols),
    ("nom", nom_pipeline, nominal_cols),
    ("ord", ord_pipeline, ordinal_cols)
])

# Complete Pipeline
pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", RandomForestClassifier(random_state=42))
])

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train
pipeline.fit(X_train, y_train)

# Prediction
y_pred = pipeline.predict(X_test)

# Evaluation
print("Accuracy :", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````
