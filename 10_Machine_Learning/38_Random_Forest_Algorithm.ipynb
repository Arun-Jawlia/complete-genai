# `Random Forest Algorithm`
---

## `What is Random Forest?`

**Random Forest** is a **supervised machine learning algorithm** used for both:

- **Classification**
- **Regression**

It is an **ensemble learning algorithm** that combines multiple **Decision Trees** to make a better prediction.

> **Random Forest = Collection of Multiple Decision Trees**

Instead of relying on one decision tree, Random Forest builds many trees and combines their predictions.

---

# `Why Do We Need Random Forest?`

A single Decision Tree has one major problem:

- It can **overfit** the training data.

Random Forest solves this problem by combining many decision trees.

This results in:

- Better accuracy
- Less overfitting
- More stable predictions

---

# `What is Ensemble Learning?`

## Definition

**Ensemble Learning** is a technique where multiple machine learning models are combined to produce a better result than a single model.

Instead of trusting one model, we combine the predictions of many models.

---

# `Why is it Called "Random Forest"?`

- **Random** → Each tree is trained using random samples and random features.
- **Forest** → A collection of many Decision Trees.

So,

```code
Many Decision Trees

↓

Random Forest
```

---

# `How Random Forest Works`

## `Step 1: Create Multiple Random Datasets`

Random Forest creates multiple datasets by randomly selecting samples from the original dataset.

This technique is called **Bootstrap Sampling**.

Example

Original Dataset

```code
1000 Records
```

↓

Random Samples

```code
Dataset 1

Dataset 2

Dataset 3

...

Dataset N
```

Each dataset may contain duplicate records because sampling is done **with replacement**.

---

# `Step 2: Build Multiple Decision Trees`

Each random dataset is used to train one Decision Tree.

Example

```code
Dataset 1

↓

Tree 1

Dataset 2

↓

Tree 2

Dataset 3

↓

Tree 3
```

---

# `Step 3: Random Feature Selection`

Unlike a normal Decision Tree, Random Forest does **not** consider all features at every split.

Instead, it randomly selects a subset of features.

Example

Suppose we have:

- Age
- Salary
- Experience
- Education
- City

At one split, Random Forest may use only:

- Salary
- Education

At another split:

- Age
- City

This randomness creates diverse trees.

---

# `Step 4: Make Predictions`

Each tree makes its own prediction.

Then Random Forest combines them.

---

# `Random Forest for Classification`

Each tree votes for a class.

Example

```code
Tree 1 → Yes

Tree 2 → Yes

Tree 3 → No

Tree 4 → Yes

Tree 5 → No
```

Final Prediction

```
Yes
```

This is called **Majority Voting**.

---

# Random Forest for Regression

Each tree predicts a numerical value.

Example

```code
Tree 1 → 40

Tree 2 → 42

Tree 3 → 41

Tree 4 → 39

Tree 5 → 43
```

Final Prediction

```
Average

=

41
```

---

# `Bootstrap Sampling`

## Definition

Bootstrap Sampling means selecting samples **with replacement**.

### Example

Original Data

```
A B C D E
```

Random Sample

```
A C C D E
```

Notice:

- C appears twice.
- B is missing.

This is completely normal.

---

# Feature Randomness

Random Forest introduces randomness in two ways:

1. Random sampling of data
2. Random selection of features

This reduces correlation between trees and improves generalization.

---

# `Important Parameters`

## `1. n_estimators`

### Definition

Number of Decision Trees.

Example

```python
RandomForestClassifier(n_estimators=100)
```

Higher values generally improve performance but increase training time.

---

## `2. max_depth`

Controls the maximum depth of each tree.

Smaller depth

↓

Less overfitting

Larger depth

↓

More complex trees

---

## `3. max_features`

Controls how many features are considered at each split.

Common values

- sqrt
- log2
- integer value

---

## `4. random_state`

Used for reproducibility.

Example

```python
random_state=42
```

---

# `Advantages of Random Forest`

- High accuracy
- Reduces overfitting
- Handles large datasets
- Works with numerical and categorical features
- Handles missing values reasonably well
- Robust to noise and outliers
- Provides feature importance

---

# `Limitations`

- Slower than a single Decision Tree
- Uses more memory
- Less interpretable
- Training can be expensive for very large forests

---

# `Random Forest vs Decision Tree`

| Decision Tree | Random Forest |
|---------------|---------------|
| Single tree | Multiple trees |
| More prone to overfitting | Less prone to overfitting |
| Faster | Slower |
| Easier to interpret | Harder to interpret |
| Lower accuracy | Usually higher accuracy |

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is Random Forest?

**Answer:**
Random Forest is a supervised ensemble learning algorithm that combines multiple Decision Trees to improve prediction accuracy.

---

### Q2. Is Random Forest used for classification or regression?

**Answer:**
It is used for both **classification** and **regression**.

---

### Q3. Why is it called Random Forest?

**Answer:**
Because it builds a **forest of Decision Trees** using **randomly selected samples** and **random subsets of features**.

---

**Answer:**
Bootstrap Sampling is the process of creating random datasets by sampling the original data **with replacement**.

### Q5. How does Random Forest make predictions?

**Answer:**

* **Classification:** Majority voting.
* **Regression:** Average of predictions from all trees.

---

**Answer:**

**Answer:**
It specifies the number of Decision Trees in the forest.

### Q8. What is `max_features`?

**Answer:**
It controls the number of features randomly selected at each split in a tree.

### Q9. Why does Random Forest randomly select features?

**Answer:**
To make the trees less correlated, increasing diversity and improving overall model performance.
