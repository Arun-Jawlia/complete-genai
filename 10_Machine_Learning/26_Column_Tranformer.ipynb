# ColumnTransformer in Scikit-learn

## `What is ColumnTransformer?`

### Definition

**ColumnTransformer** is a preprocessing utility in **Scikit-learn** that allows us to apply **different preprocessing techniques to different columns** of a dataset in a single step.

In simple words:

> **ColumnTransformer lets you preprocess multiple columns using different transformations at the same time.**

---

# `Why Do We Need ColumnTransformer?`

Real-world datasets contain different types of columns.

Example:

| Column | Data Type | Required Preprocessing |
|---------|-----------|------------------------|
| Age | Numerical | Median Imputation + Scaling |
| Salary | Numerical | Standardization |
| Gender | Categorical | One-Hot Encoding |
| Education | Ordinal | Ordinal Encoding |

Instead of preprocessing each column separately, **ColumnTransformer** performs all transformations together.

# Syntax

```python

ct = ColumnTransformer(
    transformers=[
        ("Name", Transformer, Columns)
    ]
)
```

# Code

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    OneHotEncoder,
    OrdinalEncoder,
    StandardScaler
)

from sklearn.impute import SimpleImputer

ct = ColumnTransformer(

transformers=[

(
"num",

Pipeline([
("imputer",SimpleImputer(strategy="median")),
("scaler",StandardScaler())
]),

["Age","Salary"]

),

(
"gender",

OneHotEncoder(drop="first"),

["Gender"]

),

(
"education",

OrdinalEncoder(categories=[
["High School",
 "Bachelor",
 "Master",
 "PhD"]
]),

["Education"]

)

],

remainder="passthrough"

)
```

# 2. `Interview Questions & Answers`

## Beginner Questions

### Q1. What is ColumnTransformer?

**Answer:**
ColumnTransformer is a Scikit-learn utility that applies different preprocessing techniques to different columns of a dataset in a single step.

---

### Q2. Why do we use ColumnTransformer?

**Answer:**
Because different columns often require different preprocessing methods, such as scaling numerical features and encoding categorical features.

---

### Q3. Which module contains ColumnTransformer?

**Answer:**

```python
from sklearn.compose import ColumnTransformer
```

---

### Q4. What does `remainder="passthrough"` do?

**Answer:**
It keeps all columns that are **not explicitly transformed** unchanged in the output.

---

### Q5. What does `remainder="drop"` do?

**Answer:**
It removes all columns that are **not specified** in the transformers list. This is the default behavior.

---

## Intermediate Questions

### Q6. Can ColumnTransformer apply multiple transformations?

**Answer:**
Yes. It can apply different preprocessing techniques (such as imputation, scaling, one-hot encoding, and ordinal encoding) to different sets of columns simultaneously.

---

### Q7. Why is ColumnTransformer commonly used with Pipeline?

**Answer:**
Using `ColumnTransformer` with `Pipeline` ensures that preprocessing and model training occur together. This keeps the workflow organized and helps prevent **data leakage**.

---

### Q8. Give a real-world example of using ColumnTransformer.

**Answer:**
In the Titanic dataset:

* **Age** → Median Imputation + StandardScaler
* **Fare** → StandardScaler
* **Sex** → OneHotEncoder
* **Embarked** → OneHotEncoder

All these transformations can be combined into a single `ColumnTransformer`.

````python
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    OneHotEncoder,
    OrdinalEncoder,
    StandardScaler
)
````

````python
#load Dataset
df = pd.read_csv("../Datasets/Bank_Marketing_Original_Data.csv")

df.head()
````

````python
# separate Features and Target
X = df.drop("TARGET", axis=1)

y = df["TARGET"]
````

````python
# Numerical Columns
numerical_cols = [
    "Age",
    "Balance",
    "Day",
    "Duration",
    "Campaign",
    "Pdays",
    "Previous"
]
````

````python
# Nominal Columns: No Order
nominal_cols = [
    "Job",
    "Marital",
    "Default",
    "Housing",
    "Loan",
    "contact",
    "poutcome"
]
````

````python
# Ordinal Columns : Order
ordinal_cols = [
    "Education",
    "Month"
]
````

````python
# Ordinal Pipeline

education_order = [
    "unknown",
    "primary",
    "secondary",
    "tertiary"
]

month_order = [
    "jan","feb","mar","apr",
    "may","jun","jul","aug",
    "sep","oct","nov","dec"
]
````

````python
ct = ColumnTransformer(transformers = [
    ('scale', StandardScaler(), numerical_cols), 
    ('ordinal', OrdinalEncoder(categories = [education_order, month_order]), ordinal_cols), 
    ('onehot', OneHotEncoder(drop='first'), nominal_cols)
    ],
    remainder='passthrough'
    )
````

````python
X_transformed = ct.fit_transform(X)
````

````python
X_transformed.shape
````

````python
feature_names = ct.get_feature_names_out()

print(feature_names)
````

````python
X_final = pd.DataFrame(
    X_transformed,
    columns=feature_names
)

X_final.head()
````

⚠️ One Important Note

This code assumes there are no missing values. If your Bank Marketing Original Data contains missing values (e.g., "unknown" is treated as a category, not a missing value), the code will work as is. If there are actual NaN values, you'll need a SimpleImputer, which is typically combined with a Pipeline inside a ColumnTransformer.

For an introductory lecture on ColumnTransformer, however, the above version is the standard approach without using Pipeline.
