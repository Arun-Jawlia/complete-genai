# Feature Scaling

## `What is Feature Scaling?`

### Definition

**Feature Scaling** is the process of transforming numerical features so that they have a similar range or scale.

In simple words:

> **Feature Scaling ensures that all numerical features contribute equally during model training.**

---

# `Why Do We Need Feature Scaling?`

Different numerical features often have different ranges.

### Example

| Age | Salary |
|-----|---------|
| 25 | 500000 |
| 30 | 750000 |
| 40 | 1000000 |

Here,

- Age ranges from **20–60**
- Salary ranges from **50,000–10,00,000**

Since Salary has much larger values, many Machine Learning algorithms may give it more importance than Age.

Feature Scaling solves this problem.

---

# `Example Without Feature Scaling`

Suppose we want to predict whether a customer will buy a product.

Features:

| Age | Salary |
|-----|---------|
| 25 | 500000 |

Distance Calculation

```code
Age Difference = 5

Salary Difference = 500000
```

The Salary dominates the distance calculation.

As a result, Age contributes very little.

---

# `Example With Feature Scaling`

After Scaling

| Age | Salary |
|-----|---------|
| 0.25 | 0.45 |
| 0.30 | 0.60 |

Now both features are on a similar scale and contribute fairly to the model.

---

# `When is Feature Scaling Performed?`

Feature Scaling is performed **after Feature Encoding** and **before Model Training**.


# `Types of Feature Scaling`

There are two main Feature Scaling techniques.

1. Standardization (Z-Score Normalization)
2. Normalization (Min-Max Scaling)

---

# `1. Standardization (StandardScaler)`

## Definition

Standardization transforms data so that:

- Mean = 0
- Standard Deviation = 1

### Formula

\[
Z = \frac{X - \mu}{\sigma}
\]

Where:

- X = Original Value
- μ = Mean
- σ = Standard Deviation

---

## Example

Original Data

| Age |
|------|
| 20 |
| 25 |
| 30 |
| 35 |

After Standardization

| Age |
|------|
| -1.34 |
| -0.45 |
| 0.45 |
| 1.34 |

Notice:

- Mean becomes 0
- Values can be negative or positive

---

## Python

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

---

## When to Use?

Use StandardScaler when:

- Data is approximately normally distributed.
- Algorithms use distance or gradients.

Examples:

- Logistic Regression
- Linear Regression
- SVM
- KNN
- Neural Networks
- PCA

---

# `2. Normalization (MinMaxScaler)`

## Definition

Normalization scales values between **0 and 1**.

Mostly used in Neural Network

### Formula

\[
X' = \frac{X - X_{min}}{X_{max} - X_{min}}
\]

---

## Example

Original Data

| Age |
|------|
| 20 |
| 25 |
| 30 |
| 35 |

After Normalization

| Age |
|------|
| 0.00 |
| 0.33 |
| 0.67 |
| 1.00 |

---

## Python

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
```

---

## When to Use?

Use MinMaxScaler when:

- Features have different ranges.
- Neural Networks perform better with values between 0 and 1.
- Pixel values in images.

---

# `Comparison`

| Feature | StandardScaler | MinMaxScaler |
|----------|----------------|--------------|
| Range | No fixed range | 0 to 1 |
| Mean | 0 | Not fixed |
| Standard Deviation | 1 | Not fixed |
| Sensitive to Outliers | Yes | Yes |
| Negative Values | Yes | No |

---

# `Which ML Algorithms Need Feature Scaling?`

## Scaling Required ✅

- K-Nearest Neighbors (KNN)
- Support Vector Machine (SVM)
- Logistic Regression
- Linear Regression
- Neural Networks
- PCA
- K-Means Clustering

Reason:

These algorithms depend on **distance, gradients, or optimization**.

---

## Scaling Not Required ❌

- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost

Reason:

These algorithms split data based on feature values, not distances.

# StandardScaler vs MinMaxScaler

| StandardScaler | MinMaxScaler |
|----------------|--------------|
| Mean = 0 | Values between 0 and 1 |
| Better for normally distributed data | Better for bounded ranges |
| Can produce negative values | No negative values |

# Memory Trick

### Think "Distance"

If the algorithm calculates **distance**, **gradient**, or **similarity**, perform Feature Scaling.

Examples:

- KNN ✅
- SVM ✅
- K-Means ✅

If the algorithm builds **trees**, scaling is usually unnecessary.

Examples:

- Decision Tree ❌
- Random Forest ❌
- XGBoost ❌

---

# Key Takeaways

- Feature Scaling transforms numerical features to a common scale.
- It prevents large-valued features from dominating smaller ones.
- StandardScaler standardizes data (Mean = 0, Std = 1).
- MinMaxScaler normalizes data to the range [0, 1].
- Scale only numerical features.
- Distance-based algorithms require scaling.
- Tree-based algorithms generally do not require scaling.

## `Complete Code on Standarization`

````python
# Load Dataset
df = pd.read_csv("../Datasets/Bank_Marketing_Original_Data.csv")

# Numerical Columns
numerical_cols = [
    "Age",
    "Balance",
    "Day",
    "Duration",
    "Campaign",
    "Pdays",
    "Previous"
]

# Feature Scaling
scaler = StandardScaler()

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

# Display Dataset
print(df.head())
````

## `Complete Code on Normalization`

````python
from sklearn.preprocessing import MinMaxScaler

numerical_cols = [
    "Age",
    "Balance",
    "Day",
    "Duration",
    "Campaign",
    "Pdays",
    "Previous"
]

scaler = MinMaxScaler()

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print(df.head())
````

## `Feature Scaling with Column Transformer`

````python
from sklearn.compose import ColumnTransformer

numerical_cols = [
    "Age",
    "Balance",
    "Day",
    "Duration",
    "Campaign",
    "Pdays",
    "Previous"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("scaler", StandardScaler(), numerical_cols)
    ],
    remainder="passthrough"
)

scaled_data = preprocessor.fit_transform(df)

scaled_data
````
