# `Types of Machine Learning`
---

Machine Learning can be broadly classified into three main types:

```text
Machine Learning
      │
      ├── Supervised Learning
      │
      ├── Unsupervised Learning
      │
      └── Reinforcement Learning
```

Some books also include:

```text
Semi-Supervised Learning
Self-Supervised Learning
```

However, the three major types are:

1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning

---

# 1. `Supervised Machine Learning`
---

## Definition

Supervised Learning is a type of Machine Learning in which the model is trained using **labeled data**.

In supervised learning, both:

```text
Input (X)
```

and

```text
Output (Y)
```

are provided to the algorithm during training.

The goal of the algorithm is to learn the relationship (mapping) between input and output so that it can accurately predict outputs for new unseen inputs.

---

# `Why is it Called "Supervised" Learning?`

The word **Supervised** means that learning takes place under guidance.

Just as a teacher teaches a student by providing questions and correct answers, in supervised learning we provide:

```text
Input + Correct Output
```

to the machine.

The machine learns from these examples.

### Human Learning Example

A teacher shows:

```text
2 × 2 = 4
3 × 3 = 9
4 × 4 = 16
```

The student learns the pattern.

Similarly:

```text
Input → Output

2 → 4
3 → 9
4 → 16
```

The machine learns:

```text
Output = Input²
```

---

# Basic Idea of Supervised Learning

The machine tries to learn:

```text
Input → Output
```

relationship.

### Example

Training Data:

| Hours Studied | Marks |
|--------------|--------|
| 2 | 30 |
| 4 | 50 |
| 6 | 70 |
| 8 | 90 |

The machine learns:

```text
More Study Hours
        ↓
Higher Marks
```

After training:

```text
Hours = 7
```

The model predicts:

```text
Marks ≈ 80
```

---

# `Components of Supervised Learning`

## 1. Input Features (X)

Input variables used to make predictions.

### Example

For house price prediction:

```text
Area
Number of Rooms
Location
Age of House
```

These are called Features.

---

## 2. Output Label (Y)

The target value we want to predict.

### Example

```text
House Price
```

---

## 3. Training Data

A collection of inputs and outputs.

Example:

| Area | Price |
|--------|--------|
| 1000 | 50 Lakh |
| 1200 | 60 Lakh |
| 1500 | 75 Lakh |

---

## 4. Learning Algorithm

The algorithm discovers patterns between:

```text
Input Features
        ↓
Output Label
```

Examples:

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)

---

# Workflow of Supervised Learning

```text
Training Data
(Input + Output)
        ↓
Learning Algorithm
        ↓
Pattern Learning
        ↓
Model Creation
        ↓
Prediction
```

---

# Step-by-Step Working

## Step 1: Collect Data

Example:

| Hours Studied | Marks |
|--------------|--------|
| 2 | 30 |
| 4 | 50 |
| 6 | 70 |
| 8 | 90 |

---

## Step 2: Train the Model

The algorithm analyzes:

```text
2 → 30
4 → 50
6 → 70
8 → 90
```

and learns the relationship.

---

## Step 3: Create Model

The learned pattern is stored in a model.

```text
Training Data
        ↓
Learning
        ↓
Model
```

---

## Step 4: Prediction

New Input:

```text
Hours Studied = 7
```

Model predicts:

```text
Marks ≈ 80
```

# 2. **`Unsupervised Machine Learning`**
---

## Definition

Unsupervised Learning is a type of Machine Learning in which the model is trained using **unlabeled data**.

In Unsupervised Learning:

```text
Input Data Available
Output Data Not Available
```

The machine is given only input data and must discover hidden patterns, structures, relationships, or groups on its own.

---

# `Why is it Called "Unsupervised"?`

In Supervised Learning, a teacher provides:

```text
Question + Correct Answer
```

But in Unsupervised Learning:

```text
Question Only
```

No correct answer is provided.

The machine must learn by itself.

Therefore, it is called:

```text
Unsupervised Learning
```

---

# Supervised vs Unsupervised Learning

## Supervised Learning

```text
Input + Output
      ↓
Learn Relationship
      ↓
Prediction
```

Example:

| Hours Studied | Marks |
|--------------|--------|
| 2 | 30 |
| 4 | 50 |
| 6 | 70 |

The machine learns:

```text
Hours → Marks
```

---

## Unsupervised Learning

```text
Input Only
     ↓
Find Hidden Patterns
     ↓
Create Groups/Relationships
```

Example:

| Age | Salary |
|------|--------|
| 22 | 25000 |
| 24 | 28000 |
| 45 | 90000 |
| 48 | 95000 |

No output column is provided.

The machine tries to identify natural groupings.

---

# Goal of Unsupervised Learning

The main goal is to discover:

- Hidden patterns
- Similarities
- Differences
- Relationships
- Groups (Clusters)

within the data.

---

# `How Unsupervised Learning Works`

Suppose we have customer data:

| Customer | Age | Income |
|-----------|------|---------|
| A | 22 | 25000 |
| B | 24 | 28000 |
| C | 45 | 90000 |
| D | 48 | 95000 |

The algorithm observes:

```text
A and B are similar
C and D are similar
```

It automatically forms groups:

```text
Group 1 → Young Customers

Group 2 → High Income Customers
```

No human tells the algorithm how to group them.

---

# Workflow of Unsupervised Learning

```text
Unlabeled Data
       ↓
ML Algorithm
       ↓
Pattern Discovery
       ↓
Grouping / Relationships
```

---

# Real-Life Example

Imagine entering a new classroom.

You do not know anyone.

After some observation, you notice:

```text
Group 1 → Students interested in Sports

Group 2 → Students interested in Coding

Group 3 → Students interested in Music
```

Nobody tells you these groups.

You discover them by observing patterns.

This is exactly what Unsupervised Learning does.

# 3. `Semi-Supervised Machine Learning`
---

## Definition

Semi-Supervised Learning is a type of Machine Learning that uses **both labeled and unlabeled data** for training.

It lies between:

```text
Supervised Learning
        and
Unsupervised Learning
```

### Formula

```text
Semi-Supervised Learning
      =
Small Amount of Labeled Data
      +
Large Amount of Unlabeled Data
```

---

# Why Do We Need Semi-Supervised Learning?

In the real world:

### Labeled Data

```text
Expensive
Time Consuming
Requires Human Effort
```

### Unlabeled Data

```text
Cheap
Easy to Collect
Available in Huge Quantity
```

For example:

Suppose a company has:

```text
10,00,000 Images
```

Labeling every image as:

```text
Dog
Cat
Bird
Car
```

would require a lot of human effort.

Instead:

```text
1000 Labeled Images
+
9,99,000 Unlabeled Images
```

can be used together.

This approach is called:

```text
Semi-Supervised Learning
```

---

# Position of Semi-Supervised Learning

```text
Machine Learning
       │
       ├── Supervised Learning
       │      (Only Labeled Data)
       │
       ├── Unsupervised Learning
       │      (Only Unlabeled Data)
       │
       └── Semi-Supervised Learning
              (Both Types of Data)
```

---

# Data Used in Semi-Supervised Learning

## Example Dataset

### Labeled Data

| Image | Label |
|---------|---------|
| Dog Image | Dog |
| Cat Image | Cat |
| Dog Image | Dog |

---

### Unlabeled Data

| Image |
|---------|
| Image 1 |
| Image 2 |
| Image 3 |
| Image 4 |

No labels are provided.

---

The algorithm learns from both datasets.

---

# How Semi-Supervised Learning Works

## Step 1

Train the model using a small amount of labeled data.

```text
Labeled Data
      ↓
Initial Model
```

---

## Step 2

Use the trained model to predict labels for unlabeled data.

```text
Unlabeled Data
      ↓
Predicted Labels
```

---

## Step 3

Select highly confident predictions.

Example:

```text
Image 1 → Dog (99%)

Image 2 → Cat (98%)
```

These predictions are treated as new labeled examples.

---

## Step 4

Retrain the model using:

```text
Original Labeled Data
+
Newly Labeled Data
```

This improves performance.

---

# Workflow

```text
Small Labeled Dataset
           +
Large Unlabeled Dataset
           ↓
Semi-Supervised Algorithm
           ↓
Pattern Learning
           ↓
Model
           ↓
Prediction
```

---

# Example: Dog and Cat Classification

Suppose we have:

### Labeled Data

```text
Dog Image → Dog
Cat Image → Cat
Dog Image → Dog
```

---

### Unlabeled Data

```text
Image A
Image B
Image C
Image D
```

---

Then it predicts:

```text
Image A → Dog

Image B → Cat

Image C → Dog
```

The model uses these new labels to improve itself.

```text
100 Spam Emails

100 Normal Emails
```

---

### Unlabeled Emails

```text
10,00,000 Emails
```

Labeling all emails manually is impossible.

The system:

```text
Uses 200 labeled emails
        ↓
Learns patterns
        ↓
Labels remaining emails
        ↓
Improves model
```

This is Semi-Supervised Learning.

# Real-Life Example: Medical Diagnosis

Suppose a hospital has:

```text
1000 Diagnosed X-Ray Images
```

and

```text
10,00,000 Unlabeled X-Ray Images
```

Doctors cannot label all images.

Semi-Supervised Learning uses:

```text
Small Labeled Data
+
Large Unlabeled Data
```

to train a better model.

---

# Why Not Use Only Supervised Learning?

Supervised Learning requires:

```text
Large Amount of Labeled Data
```

But labeling data is expensive.

Examples:

- Medical Images

- Satellite Images

require experts for labeling.

---

# Why Not Use Only Unsupervised Learning?

Unsupervised Learning:

```text
No Labels Available
```

The model may not learn the exact target task.

Adding a small amount of labeled data provides guidance.

# Advantages of Semi-Supervised Learning

---

## 2. Better Accuracy

## 3. Uses Large Amounts of Available Data

## 4. Suitable for Large Datasets

Works well when collecting labels is difficult.

# Disadvantages of Semi-Supervised Learning

## 2. More Complex

More difficult to implement than Supervised Learning.

---

## 3. Depends on Initial Labeled Data

Poor-quality labeled data can reduce performance.

# 4. `Reinforcement Learning (RL)`

## Definition

Reinforcement Learning (RL) is a type of Machine Learning in which an **Agent** learns by interacting with an **Environment** and receives **Rewards** or **Penalties** based on its actions.

The goal of the agent is to learn the best actions that maximize the total reward over time.

---

# Why is it Called Reinforcement Learning?

The word **Reinforcement** means:

```text
Strengthening a behavior through rewards
and discouraging a behavior through penalties.
```

Just like humans and animals learn from experience:

```text
Good Action
      ↓
Reward
      ↓
Repeat Action
```

```text
Bad Action
      ↓
Penalty
      ↓
Avoid Action
```

Similarly, a machine learns through rewards and punishments.

---

# Real-Life Example: Teaching a Dog

Suppose you are training a dog.

### Situation

```text
Sit
```

Dog sits correctly.

### Result

```text
Reward (Biscuit)
```

The dog learns:

```text
Sitting = Reward
```

and repeats the behavior.

---

### Wrong Action

Dog jumps on the sofa.

### Result

```text
No Reward
```

or

```text
Penalty
```

The dog gradually learns to avoid that behavior.

This is the basic idea behind Reinforcement Learning.

---

# Main Components of Reinforcement Learning

There are four important components.

```text
Agent
Environment
Action
Reward
```

---

## 1. Agent

The learner or decision-maker.

Examples:

- Robot
- Self-Driving Car
- Chess Program
- Game AI

```text
Agent = Learner
```

---

## 2. Environment

The world in which the agent operates.

Examples:

- Chess Board
- Road
- Video Game
- Robot's Surroundings

```text
Environment = World
```

---

## 3. Action

Any move made by the agent.

Examples:

### Self-Driving Car

```text
Move Left
Move Right
Accelerate
Brake
```

### Chess

```text
Move Pawn
Move Queen
Move King
```

---

## 4. Reward

Feedback received after performing an action.

### Positive Reward

```text
+10
+50
+100
```

means good action.

### Negative Reward

```text
-10
-50
-100
```

means bad action.

---

# Working of Reinforcement Learning

## Basic Workflow

```text
Agent
   ↓
Performs Action
   ↓
Environment Responds
   ↓
Reward / Penalty
   ↓
Learning
```

---

## Detailed Flow

```text
Current State
      ↓
Agent Chooses Action
      ↓
Environment Changes
      ↓
Reward Received
      ↓
Agent Learns
      ↓
Next Action
```

This cycle repeats many times.

---

# Key Idea of Reinforcement Learning

Unlike Supervised Learning:

```text
No Correct Output Given
```

The agent learns through trial and error.

---

## Supervised Learning

```text
Input
 ↓
Correct Output Available
 ↓
Learn
```

Example:

```text
Dog Image → Dog
Cat Image → Cat
```

---

## Reinforcement Learning

```text
Input
 ↓
Action
 ↓
Reward / Penalty
 ↓
Learn
```

No correct answer is provided directly.

---

# Trial and Error Learning

RL is based on:

```text
Try
 ↓
Observe Result
 ↓
Learn
 ↓
Improve
```

The agent gradually discovers the optimal behavior.

---

# Exploration vs Exploitation

One of the biggest challenges in RL.

---

## Exploration

Trying new actions.

Example:

```text
Maybe this path is shorter.
```

---

## Exploitation

Using known good actions.

Example:

```text
Follow the best path discovered earlier.
```

---

A good RL agent balances both.

---

# One-Line Summary

```text
Reinforcement Learning is a type of Machine Learning in which an agent learns the best actions by interacting with an environment and receiving rewards or penalties, with the goal of maximizing long-term rewards.
```

# Key Points

- Reinforcement Learning learns through trial and error.
- An Agent interacts with an Environment.
- Actions lead to Rewards or Penalties.
- The objective is to maximize cumulative rewards.
- No labeled data is required.
- Widely used in robotics, self-driving cars, gaming, and recommendation systems.
- It is inspired by how humans and animals learn from experience.
