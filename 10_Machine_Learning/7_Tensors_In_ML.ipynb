# `Tensors in Machine Learning`
---

- A **Tensor** is a mathematical and computational data structure used to represent and store data in one or more dimensions. 
- It is a generalized form of a scalar, vector, and matrix that can hold numerical values arranged in multiple dimensions.

- In Deep Learning and Artificial Intelligence, tensors are the fundamental units of data. 
- Every type of data such as numbers, images, text, audio, and videos is ultimately converted into tensors before being processed by neural networks.

- A tensor can be thought of as a multi-dimensional container of numbers that enables computers to efficiently perform mathematical operations such as addition, multiplication, transformation, and optimization during model training and prediction.

- Tensors provide a standardized format through which Deep Learning frameworks like TensorFlow and PyTorch represent data and perform computations.

---

## Academic Definition

- A Tensor is a multi-dimensional mathematical object that represents data arranged along one or more axes and is used to perform numerical computations in Machine Learning and Deep Learning models.

---

## Simple Definition

- A Tensor is a multi-dimensional array of numerical values used by Deep Learning models to store, process, and manipulate data.

---

## Why Do We Need Tensors?

Computers cannot directly understand:

```text
Images
Audio
Videos
Text
Documents
```

They only understand numbers.

Therefore:

```text
Image
  ↓
Numbers
  ↓
Tensor
```

```text
Audio
  ↓
Numbers
  ↓
Tensor
```

```text
Text
  ↓
Numbers
  ↓
Tensor
```

The tensor acts as a container that stores these numerical representations.

---

## Importance of Tensors in Deep Learning

Tensors allow Deep Learning models to:

- Store large amounts of data
- Process multi-dimensional data
- Perform mathematical operations efficiently
- Train Neural Networks
- Handle images, audio, text, and videos

Without tensors, modern Deep Learning would not be possible.
Every Deep Learning model follows the workflow:

```text
Real World Data
        ↓
Tensor Conversion
        ↓
Neural Network
        ↓
Prediction
```

Example:

```text
Dog Image
      ↓
Tensor
      ↓
Deep Learning Model
      ↓
Dog
```

# `Types of Tensors`
---

A Tensor can be classified based on the number of dimensions it contains.

```text
Tensor
│
├── 0-D Tensor (Scalar)
├── 1-D Tensor (Vector)
├── 2-D Tensor (Matrix)
├── 3-D Tensor
└── N-D Tensor
```

As the number of dimensions increases, the tensor becomes capable of representing more complex data.

---

# 1. 0-D Tensor (Scalar)

## Definition

A 0-D Tensor, also called a **Scalar**, contains only a single numerical value.

It has:

```text
No Rows
No Columns
Only One Value
```

### Example

```python
5
```

or

```python
100
```

### Representation

```text
5
```

### Shape

```text
()
```

### Dimension

```text
0-D
```

### Real-Life Example

```text
Temperature = 35°C

Age = 25

Salary = 50000
```

These are single values and are represented as Scalars.

---

# 2. 1-D Tensor (Vector)

## Definition

A 1-D Tensor, also called a **Vector**, is a collection of numbers arranged in a single dimension.

It contains:

```text
One Axis
Multiple Values
```

### Example

```python
[10, 20, 30, 40]
```

### Representation

```text
10  20  30  40
```

### Shape

```text
(4,)
```

Meaning:

```text
4 Elements
```

### Dimension

```text
1-D
```

### Real-Life Example

A student's marks:

```python
[80, 75, 90, 85, 88]
```

---

# 3. 2-D Tensor (Matrix)

## Definition

A 2-D Tensor, also called a **Matrix**, contains rows and columns.

It is similar to a table.

### Example

```python
[
 [1, 2, 3],
 [4, 5, 6]
]
```

### Representation

```text
1  2  3
4  5  6
```

### Shape

```text
(2,3)
```

Meaning:

```text
2 Rows
3 Columns
```

### Dimension

```text
2-D
```

### Real-Life Example

Student Marks Table:

| Student | Marks |
|----------|--------|
| A | 80 |
| B | 75 |
| C | 90 |

This tabular data can be represented as a Matrix.

---

# 4. 3-D Tensor

## Definition

A 3-D Tensor consists of multiple matrices stacked together.

It introduces a third dimension.

### Example

```python
[
 [
  [1, 2],
  [3, 4]
 ],

 [
  [5, 6],
  [7, 8]
 ]
]
```

### Visualization

```text
Matrix 1

1 2
3 4

Matrix 2

5 6
7 8
```

### Shape

```text
(2,2,2)
```

Meaning:

```text
2 Matrices
2 Rows
2 Columns
```

### Dimension

```text
3-D
```

### Real-Life Example

A colored image:

```text
Height
Width
RGB Channels
```

Example Shape:

```text
(224,224,3)
```

where:

```text
224 = Height
224 = Width
3 = RGB Channels
```

---

# 5. N-D Tensor (Higher-Dimensional Tensor)

## Definition

A tensor having more than three dimensions is called an N-D Tensor.

These tensors are commonly used in Deep Learning.

### Examples

```text
4-D Tensor
5-D Tensor
6-D Tensor
...
N-D Tensor
```

---

## Example of 4-D Tensor

Suppose we have:

```text
100 Images
```

Each image:

```text
224 × 224 × 3
```

Tensor Shape:

```text
(100,224,224,3)
```

Meaning:

```text
100 Images
224 Height
224 Width
3 RGB Channels
```

---

## Example of 5-D Tensor

Video Data:

```text
Batch Size
Frames
Height
Width
Channels
```

Shape:

```text
(32,100,224,224,3)
```

Meaning:

```text
32 Videos
100 Frames
224 Height
224 Width
3 Channels
```
