## `What is API`
- API (Application Programming Interface) allows AI systems to access real-time data from external applications and services.
- APIs are one of the most important data sources for modern AI applications.
- Brigde between let the apps share the information among each other
- Access tto the database of an app or website

# Why APIs?

APIs provide live data from external systems.

Examples:

- Weather Data
- Stock Market Data
- Maps & Location Data
- Payment Data
- Social Media Data
- News Data
- E-commerce Data

Without APIs, AI systems cannot access real-time information.

# API Data Gathering Workflow

```code
External Service
        ↓
       API
        ↓
   JSON Response
        ↓
   AI Application
        ↓
 Analysis / Action
```

```markdown
# APIs in RAG Systems

RAG = Retrieval Augmented Generation

Data Sources:

- APIs
- Databases
- PDFs
- Websites
- Knowledge Bases

API-based RAG enables access to live information.

# RAG Workflow using APIs

```text
User Question
↓
Retriever
↓
API Call
↓
JSON Data
↓
LLM
↓
Generated Answer

```
The LLM uses retrieved information instead of relying only on training data.

# Example: Weather RAG

User:
"What is today's weather in Delhi?"

↓

Weather API

↓

{
   "temperature": 35,
   "humidity": 60
}

↓

LLM generates:

"Today's temperature in Delhi is 35°C with 60% humidity."
