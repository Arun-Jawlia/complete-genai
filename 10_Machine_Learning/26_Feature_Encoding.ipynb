# Feature Encoding
---

## `What is Feature Encoding?`

### Definition

**Feature Encoding** is the process of converting **categorical (text) data into numerical values** so that Machine Learning algorithms can understand and process it.

In simple words:

> **Feature Encoding converts text labels into numbers.**

---

# `Why is Feature Encoding Needed?`

Most Machine Learning algorithms perform mathematical calculations.

They cannot directly understand text values such as:

| Gender |
|---------|
| Male |
| Female |

The algorithm understands numbers, not words.

Therefore, we convert text into numerical values.

---

# Example

Before Encoding

| Gender |
|---------|
| Male |
| Female |
| Male |
| Female |

↓

After Encoding

| Gender |
|---------|
| 1 |
| 0 |
| 1 |
| 0 |

Now the Machine Learning model can use this feature.

---

# `When Do We Use Feature Encoding?`

Whenever a dataset contains **categorical columns**.

Examples

- Gender
- City
- Department
- Country
- Education
- Product Category

---

# `Types of Categorical Data`

There are **two types** of categorical variables.

## `1. Nominal Data`

### Definition

Categories **do not have any order or ranking**.

### Examples

- Gender
- City
- Country
- Blood Group
- Color

Example

```
Red
Blue
Green
Yellow
```

There is **no ranking** between these values.

---

## `2. Ordinal Data`

### Definition

Categories have a **meaningful order or ranking**.

### Examples

Education

```
High School
Bachelor
Master
PhD
```

Customer Rating

```
Poor
Average
Good
Excellent
```

These values have a natural order.

---

# `Common Encoding Techniques`

## `1. Label Encoding`

### What is it?

Each category is assigned a unique integer.

### Example

Before

| Size |
|------|
| Small |
| Medium |
| Large |

After

| Size |
|------|
| 0 |
| 1 |
| 2 |

---

### When to Use?

✅ Best for **Ordinal Data**.

Because the order matters.

---

### Python Example

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Size"] = encoder.fit_transform(df["Size"])
```

---

# 2. `One-Hot Encoding`

### What is it?

Creates a separate binary column for each category.

### Example

Before

| Color |
|--------|
| Red |
| Blue |
| Green |

↓

After

| Red | Blue | Green |
|-----|------|--------|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |

---

### Why?

One-Hot Encoding prevents the model from assuming an order between categories.

---

### When to Use?

✅ Best for **Nominal Data**.

Examples

- Gender
- City
- Country
- Department

---

### Python Example

```python
pd.get_dummies(df, columns=["Color"])
```

or

```python
from sklearn.preprocessing import OneHotEncoder
```

---

# 3. `Ordinal Encoding`

### What is it?

Assigns numbers while preserving the natural order.

### Example

Education

| Education |
|------------|
| High School |
| Bachelor |
| Master |
| PhD |

↓

| Education |
|------------|
| 1 |
| 2 |
| 3 |
| 4 |

---

### When to Use?

Only when categories have a meaningful ranking.

---

# 4. `Frequency Encoding`

### What is it?

Replace each category with its frequency.

### Example

| City |
|------|
| Delhi |
| Mumbai |
| Delhi |
| Pune |

Frequency

Delhi = 2

Mumbai = 1

Pune = 1

↓

| City |
|------|
| 2 |
| 1 |
| 2 |
| 1 |

---

# 5. `Target Encoding`

### What is it?

Replace each category with the average value of the target variable.

### Example

Loan Approval

| City | Loan Approved |
|------|---------------|
| Delhi | 1 |
| Delhi | 0 |
| Mumbai | 1 |

Target Mean

Delhi → 0.5

Mumbai → 1.0

---

### Used For

High-cardinality categorical features.

---

# `Comparison of Encoding Techniques`

| Technique | Best For | Creates New Columns? |
|------------|----------|----------------------|
| Label Encoding | Ordinal Data | ❌ No |
| One-Hot Encoding | Nominal Data | ✅ Yes |
| Ordinal Encoding | Ordered Categories | ❌ No |
| Frequency Encoding | High-cardinality Features | ❌ No |
| Target Encoding | Advanced ML Tasks | ❌ No |

---

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is Feature Encoding?

**Answer:**
Feature Encoding is the process of converting categorical (text) data into numerical values so Machine Learning algorithms can process it.

---

### Q2. Why is Feature Encoding required?

**Answer:**
Most Machine Learning algorithms perform mathematical computations and cannot directly understand text values.

---

### Q3. What are the two types of categorical data?

**Answer:**

* **Nominal Data:** No order (e.g., Gender, City).
* **Ordinal Data:** Has a natural order (e.g., Education Level, Customer Rating).

---

### Q4. Which encoding technique is best for nominal data?

**Answer:**
**One-Hot Encoding**, because it does not introduce any artificial order.

---

### Q5. Which encoding technique is best for ordinal data?

**Answer:**
**Ordinal Encoding** or **Label Encoding**, because they preserve the natural order.

---

## Intermediate Questions

### Q6. Why is One-Hot Encoding preferred over Label Encoding for nominal data?

**Answer:**
Label Encoding assigns numerical values (e.g., Red = 0, Blue = 1, Green = 2), which may incorrectly imply an order. One-Hot Encoding avoids this by creating separate binary columns for each category.

---

### Q7. What is the drawback of One-Hot Encoding?

**Answer:**
It can create many new columns when a feature has many unique categories, increasing the dataset's dimensionality.

---

### Q8. What is Target Encoding?

**Answer:**
Target Encoding replaces each category with the average value of the target variable. It is useful for high-cardinality categorical features but should be applied carefully to avoid data leakage.

---

## Scenario-Based Questions

### Q9. You have a **Gender** column with values *Male* and *Female*. Which encoding technique would you use?

**Answer:**
**One-Hot Encoding**, because Gender is a nominal variable with no natural order.

---

### Q10. You have an **Education Level** column with values *High School*, *Bachelor*, *Master*, and *PhD*. Which encoding technique is appropriate?

**Answer:**
**Ordinal Encoding**, because the categories have a meaningful order.

# `Step 2: Apply Label Encoding (Binary Columns)`

## 🎯 Interview Tip

When asked **"Which encoding technique would you choose?"**, always follow this decision process:

* **Binary categorical (Yes/No)** → **Label Encoding**
* **Ordinal categorical (has order, e.g., Education, Month)** → **Ordinal Encoding**
* **Nominal categorical (no order, e.g., Job, Marital, Contact)** → **One-Hot Encoding**

This is the approach commonly used in production Machine Learning pipelines.

# Complete Code

```python
import pandas as pd

from sklearn.preprocessing import (
    LabelEncoder,
    OrdinalEncoder,
    OneHotEncoder
)

# ----------------------------
# Label Encoding
# ----------------------------

binary_cols = [
    "default",
    "housing",
    "loan",
    "TARGET"
]

le = LabelEncoder()

for col in binary_cols:
    df[col] = le.fit_transform(df[col])

# ----------------------------
# Ordinal Encoding
# ----------------------------

education_order = [[
    "unknown",
    "primary",
    "secondary",
    "tertiary"
]]

education_encoder = OrdinalEncoder(categories=education_order)

df[["education"]] = education_encoder.fit_transform(df[["education"]])

month_order = [[
    "jan","feb","mar","apr",
    "may","jun","jul","aug",
    "sep","oct","nov","dec"
]]

month_encoder = OrdinalEncoder(categories=month_order)

df[["month"]] = month_encoder.fit_transform(df[["month"]])

# ----------------------------
# One Hot Encoding
# ----------------------------

nominal_cols = [
    "job",
    "marital",
    "contact",
    "poutcome"
]

ohe = OneHotEncoder(
    drop="first",
    sparse_output=False
)

encoded = ohe.fit_transform(df[nominal_cols])

encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(nominal_cols)
)

df = df.drop(columns=nominal_cols)

df = pd.concat(
    [df, encoded_df],
    axis=1
)

print(df.head())
```

## 🚀 Interview Tip

When using **Scikit-learn**, choose the encoder based on the data type:

| Data Type                                 | Scikit-learn Encoder                                                        |
| ----------------------------------------- | --------------------------------------------------------------------------- |
| Binary (Yes/No)                           | `LabelEncoder` *(or map `{yes:1, no:0}` for feature columns in production)* |
| Ordinal (Education, Month)                | `OrdinalEncoder`                                                            |
| Nominal (Job, Marital, Contact, Poutcome) | `OneHotEncoder`                                                             |

**Note:** In production ML pipelines, `LabelEncoder` is generally intended for encoding the **target variable (`y`)**. For binary feature columns (`X`), `OrdinalEncoder` or simple mapping is often preferred. However, many beginner courses use `LabelEncoder` for binary feature columns as well.
