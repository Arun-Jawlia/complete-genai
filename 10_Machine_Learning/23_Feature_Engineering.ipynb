# `Feature Engineering`
---

## What is Feature Engineering?

**Feature Engineering** is the process of creating, selecting, modifying, and transforming features (columns) in a dataset to improve the performance of a Machine Learning model.

In simple words:

> **Feature Engineering means preparing and improving the input data so that the Machine Learning model can learn better.**

---

# Why is Feature Engineering Important?

Machine Learning models learn only from the features (input columns) we provide.

If the features are:

- Incomplete
- Incorrect
- Irrelevant
- Poorly formatted

then the model will produce poor predictions.

Good features → Better Model Performance 🚀

---

# When is Feature Engineering Performed?

Feature Engineering is performed **after EDA (Exploratory Data Analysis)** and **before training the Machine Learning model**.

### ML Pipeline

```
Business Problem
        ↓
Data Collection
        ↓
Data Cleaning
        ↓
EDA
        ↓
Feature Engineering
        ↓
Model Training
        ↓
Evaluation
```

---

# Goals of Feature Engineering

- Improve data quality
- Increase model accuracy
- Reduce noise
- Make data understandable for ML algorithms
- Create meaningful features

---

# Common Feature Engineering Techniques

## 1. Handling Missing Values

### Why?

Machine Learning algorithms cannot properly learn from missing values.

### Techniques

- Mean Imputation
- Median Imputation
- Mode Imputation
- Forward Fill
- Backward Fill
- KNN Imputer

### Example

| Age |
|------|
| 20 |
| NaN |
| 25 |

↓

| Age |
|------|
| 20 |
| 22.5 |
| 25 |

---

## 2. Handling Outliers

### Why?

Outliers can negatively affect model performance.

### Techniques

- Remove Outliers
- Cap (Winsorization)
- Log Transformation

### Example

Salary

```
25000
27000
26000
29000
5000000   ← Outlier
```

---

## 3. Encoding Categorical Variables

### Why?

Machine Learning models work with numbers, not text.

### Example

Before

| Gender |
|---------|
| Male |
| Female |

↓

After

| Gender |
|---------|
| 0 |
| 1 |

### Common Methods

- Label Encoding
- One-Hot Encoding

---

## 4. Feature Scaling

### Why?

Different numerical features may have different ranges.

Example

Age = 25

Salary = 500000

The model may give more importance to Salary because of its larger values.

Scaling brings features to a similar range.

### Common Methods

- Standardization
- Normalization

---

## 5. Feature Selection

### Why?

Not every feature is useful.

Removing irrelevant features:

- Improves accuracy
- Reduces training time
- Prevents overfitting

### Example

House Price Prediction

Useful:

- Area
- Bedrooms
- Location

Not Useful:

- House ID

---

## 6. Feature Creation

### Why?

Sometimes creating new features improves model performance.

### Example

Existing Features

- Date of Birth

New Feature

- Age

Another Example

Existing Features

- Length
- Width

New Feature

- Area = Length × Width

---

## 7. Feature Transformation

### Why?

Some algorithms work better when data follows a normal distribution.

### Techniques

- Log Transformation
- Square Root Transformation
- Box-Cox Transformation

### Q7. What is the difference between Feature Selection and Feature Creation?

**Answer:**

* **Feature Selection** removes unnecessary features.
* **Feature Creation** creates new features from existing ones.

## `Family Type`

`````python
def famType(num):
    if num >=1 and num <=3 :
        return 'Small Family'
    elif num > 3 and num <=5:
        return "Medium Family"
    else:
        return 'Large Family'
`````

`````python
df['Family Type'] = df['Family Size'].apply(famType)
`````

# Feature Engineering on Titanic Dataset

## Step 1: Understand the Features

| Column      | Type                  | Action                               |

---

# Step 2: Handle Missing Values

## Age

Missing Values

* `titanic.csv` → **86**
* `titanic2.csv` → **177**

### Recommended

Since **Age** is numerical and slightly skewed,

✅ Fill using **Median**

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

---

## Fare

Missing Values

* `titanic.csv` → **1**
* `titanic2.csv` → **0**

### Recommended

Fill with Median

```python
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
```

---

## Embarked

Missing Values

* `titanic.csv` → **0**
* `titanic2.csv` → **2**

Categorical variable

Fill with Mode

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

---

## Cabin

Missing Values

* `titanic.csv` → **327**
* `titanic2.csv` → **687**

More than **75% missing**

### Recommended

Either

* Drop Cabin

OR

Create a new feature

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

or

```python
df["Deck"] = df["Cabin"].str[0]
```

## Interview Answer

**Q: What feature engineering techniques would you apply to the Titanic dataset?**

**Answer:**

1. Handle missing values (`Age` → Median, `Fare` → Median, `Embarked` → Mode).
2. Handle `Cabin` by dropping it or extracting the deck information.
3. Create new features such as `FamilySize`, `IsAlone`, `Title`, and optionally `Deck`.
4. Drop unnecessary columns like `PassengerId`, `Name` (after extracting `Title`), and `Ticket`.
5. Encode categorical variables (`Sex`, `Embarked`, `Title`, `Deck`).
6. Scale numerical features (`Age`, `Fare`, `FamilySize`) if required by the chosen algorithm.
7. Detect and handle outliers in numerical columns like `Age` and `Fare`.

This workflow reflects a typical feature engineering pipeline used for the Titanic dataset in interviews, competitions, and production ML projects.
