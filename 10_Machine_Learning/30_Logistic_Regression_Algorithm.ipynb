# `Logistic Regression`
---

## Introduction

Machine Learning algorithms are commonly used for two types of problems:

- **Regression** → Predicts continuous values
- **Classification** → Predicts categories

---

# `Regression vs Classification`

## Regression

Regression is used when the output is a **continuous numerical value**.

### Examples

- House price prediction
- Salary prediction
- Temperature prediction

---

## `Classification`

Classification is used when the output is a **category or class**.

### `Types of Classification`

1. **Binary Classification**  
   - Two classes only  
   - Example: True / False, Yes / No, Spam / Not Spam

2. **Multiclass Classification**  
   - More than two classes  
   - Example: Dog / Cat / Horse

3. **Ordinal Classification**  
   - Categories with order  
   - Example: Good / Medium / Bad

---

# `What is Logistic Regression?`

## Definition

**Logistic Regression** is a **supervised machine learning algorithm** used for **classification**, especially **binary classification**.

It predicts the **probability** that an input belongs to a particular class.

### Output

- 0 or 1
- Yes or No
- True or False

---

# `Why Not Use Linear Regression for Classification?`

Linear Regression predicts continuous values, but classification needs category output.

### Problem with Linear Regression for Classification

If we use Linear Regression for class labels:

- The output may go below 0 or above 1
- It cannot directly represent probability
- It does not give a clean decision for class membership

So we shift from Linear Regression to Logistic Regression.

---

# `Why Logistic Regression Works Better`

Logistic Regression:

- Converts output into probability
- Uses a threshold to decide the final class
- Works well for classification problems

---

# `Probability in Logistic Regression`

Logistic Regression predicts a probability value between **0 and 1**.

### Example

If the model says:

- Probability of spam = 0.92

Then the email is very likely spam.

- Probability of spam = 0.15

Then the email is likely not spam.

---

# `Threshold / Decision Boundary`

## Definition

A **threshold** is the cut-off value used to convert probability into final class output.

### Common Threshold

- If probability >= 0.5 → Class = 1
- If probability < 0.5 → Class = 0

### Example

- 0.8 → 1
- 0.3 → 0

---

# `What is the Sigmoid Function?`

## Definition

The **sigmoid function** converts any real number into a value between **0 and 1**.

This is why it is used in Logistic Regression.

---

## Sigmoid Formula

$y = \frac{1}{1 + e^{-x}}$

Where:

- `x` = input value
- `e` = Euler’s number
- `y` = output probability

---

##` Why is it called Squashing?`

The sigmoid function is also called a **squashing function** because it “squashes” any large or small input into a value between 0 and 1.

### Example

- Very large positive input → output close to 1
- Very large negative input → output close to 0

---

# How Logistic Regression Works`

```code
Input Features
      ↓
Linear Combination
      ↓
Sigmoid Function
      ↓
Probability
      ↓
Threshold / Decision Boundary
      ↓
Final Class (0 or 1)
```

---

# `General Workflow`

1. Take input features
2. Calculate weighted sum
3. Apply sigmoid function
4. Get probability
5. Apply threshold
6. Predict final category

---

# `Logistic Regression Equation`

First, the model calculates:

\[
z = w_1x_1 + w_2x_2 + ... + b
\]

Then sigmoid is applied:

\[
y = \frac{1}{1 + e^{-z}}
\]

- Patient has disease or not

# Types of Logistic Regression

## `1. Binary Logistic Regression`

Used when there are **two classes**.

### Examples

- 0 or 1
- Yes or No
- True or False

---

## `2. Multinomial Logistic Regression`

Used when there are **more than two categories**.

### Example

- Dog
- Cat
- Horse

---

## `3. Ordinal Logistic Regression`

Used when categories have a **natural order**.

### Example

Customer feedback:

- Bad
- Medium
- Good

These categories are ordered, so ordinal regression is useful.

---

# `What are Embeddings?`

## Definition

**Embeddings** are numerical representations of text, words, or categories.

They are commonly used in NLP and deep learning.

### Example

The word:

- "Amit"

can be converted into a numerical vector so that a machine can understand it.

### Why are Embeddings Useful?

- They convert text into numbers
- They capture meaning and similarity
- They help models process language data better

---

# `Logistic Regression vs Linear Regression`

| Linear Regression | Logistic Regression |
|------------------|--------------------|
| Predicts continuous values | Predicts categories |
| Used for regression problems | Used for classification problems |
| Output can be any number | Output is converted into probability |
| No sigmoid function | Uses sigmoid function |

---

# `Interview Tips`

### Remember

- **Linear Regression** → Predicts continuous values
- **Logistic Regression** → Predicts categories
- **Binary Classification** → Two classes
- **Multiclass Classification** → More than two classes
- **Ordinal Classification** → Ordered categories
- **Sigmoid Function** → Converts output into probability between 0 and 1
- **Threshold** → Converts probability into final class

---

# Key Takeaways

- Logistic Regression is a supervised learning algorithm for classification.
- It is mainly used for binary classification.
- It predicts probability, not direct class output.
- Sigmoid function converts values into a range between 0 and 1.
- Threshold is used to decide the final class.
- Logistic Regression can be binary, multinomial, or ordinal.
- It is widely used in spam detection, disease prediction, marketing, and fraud detection.
