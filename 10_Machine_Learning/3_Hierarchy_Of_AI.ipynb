# `# Hierarchy of Artificial Intelligence`
---

Artificial Intelligence is a broad field that contains several subfields.

The relationship between these technologies can be represented as:

```text
Artificial Intelligence (AI)
        │
        ▼
Machine Learning (ML)
        │
        ▼
Deep Learning (DL)
        │
        ├── Natural Language Processing (NLP)
        │
        └── Generative AI (GenAI)
```

Or:

```text
AI
 └── ML
      └── DL
           ├── NLP
           └── GenAI
```

---

# 1. Artificial Intelligence (AI)

## Definition

Artificial Intelligence (AI) is a field of Computer Science that aims to create machines capable of performing tasks that normally require human intelligence.

AI systems can:

- Learn
- Reason
- Solve Problems
- Make Decisions
- Understand Language
- Recognize Images
- Adapt to New Situations

### Examples

- Self-driving cars
- Virtual assistants
- Expert systems
- Recommendation systems
- Robotics

---

# 2. Machine Learning (ML)

## Definition

Machine Learning is a subset of Artificial Intelligence that allows machines to learn patterns from data without being explicitly programmed.

Instead of writing rules manually:

```python
if condition:
    do_something()
```

we provide data and allow the machine to discover patterns automatically.

### Workflow

```text
Training Data
      ↓
Machine Learning Algorithm
      ↓
Pattern Learning
      ↓
Model
      ↓
Prediction
```

### Examples

- House price prediction
- Spam email detection
- Product recommendation
- Fraud detection

---

# Why ML is a Subset of AI

```text
AI Goal:
Create intelligent machines.

ML Solution:
Teach machines using data.
```

Machine Learning is one of the most important techniques used to achieve Artificial Intelligence.

---

# 3. Deep Learning (DL)

## Definition

Deep Learning is a subset of Machine Learning that uses Artificial Neural Networks with multiple layers to learn complex patterns from large amounts of data.

Deep Learning is inspired by the structure of the human brain.

### Structure

```text
Input Layer
      ↓
Hidden Layer
      ↓
Hidden Layer
      ↓
Hidden Layer
      ↓
Output Layer
```

The multiple hidden layers make it "Deep" Learning.

---

## Why Deep Learning?

Traditional Machine Learning often requires manual feature extraction.

Deep Learning automatically learns features from raw data.

### Example

Image Recognition:

Machine Learning:

```text
Human extracts features
      ↓
Model learns
```

Deep Learning:

```text
Image
  ↓
Neural Network
  ↓
Features learned automatically
  ↓
Prediction
```

---

# Applications of Deep Learning

- Image Recognition
- Speech Recognition
- Face Recognition
- Language Translation
- Chatbots
- Autonomous Vehicles
- Medical Diagnosis

---

# 4. Natural Language Processing (NLP)

## Definition

Natural Language Processing (NLP) is a field that enables computers to understand, process, analyze, and generate human language.

NLP combines:

- Artificial Intelligence
- Machine Learning
- Deep Learning
- Linguistics

---

## Goal of NLP

Allow computers to communicate with humans using natural language.

### Human Language Examples

```text
English
Hindi
French
Spanish
Tamil
Punjabi
```

NLP helps machines understand these languages.

---

## NLP Applications

- Chatbots
- Language Translation
- Speech Recognition
- Sentiment Analysis
- Text Classification
- Question Answering
- Voice Assistants

### Examples

- Siri
- Alexa
- Google Assistant
- ChatGPT

---

# 5. Generative AI (GenAI)

## Definition

Generative AI is a subset of Deep Learning that can generate new content such as:

- Text
- Images
- Audio
- Video
- Code

Instead of only making predictions, GenAI creates new content.

---

## Traditional AI vs Generative AI

### Traditional AI

```text
Input
   ↓
Prediction
```

Example:

```text
Email
   ↓
Spam or Not Spam
```

---

### Generative AI

```text
Prompt
   ↓
Generate New Content
```

Example:

```text
Write a poem
        ↓
AI generates a poem
```

---

# Examples of Generative AI

## Text Generation

Prompt:

```text
Write a story about a robot.
```

Output:

```text
A complete story generated by AI.
```

---

## Image Generation

Prompt:

```text
Create an image of a tiger riding a bicycle.
```

Output:

```text
AI-generated image.
```

---

## Code Generation

Prompt:

```text
Write a Python program for Binary Search.
```

Output:

```python
def binary_search(arr, target):
    pass
```

Generated automatically by AI.

---

# Why GenAI is a Subset of Deep Learning

Generative AI models are built using advanced Deep Learning architectures such as:

- Neural Networks
- Transformers
- Large Language Models (LLMs)

Examples:

- ChatGPT
- Gemini
- Claude
- Llama

Without Deep Learning, modern Generative AI would not be possible.

---

# Complete Hierarchy

```text
Artificial Intelligence (AI)
│
├── Machine Learning (ML)
│     │
│     └── Deep Learning (DL)
│            │
│            ├── Natural Language Processing (NLP)
│            │
│            └── Generative AI (GenAI)
│
└── Other AI Techniques
      ├── Expert Systems
      ├── Robotics
      ├── Search Algorithms
      └── Knowledge Representation
```

---

# Real-World Example

Consider ChatGPT:

```text
Artificial Intelligence
        ↓
Machine Learning
        ↓
Deep Learning
        ↓
Natural Language Processing
        ↓
Generative AI
        ↓
ChatGPT
```

ChatGPT:

- Uses AI principles.
- Learns using Machine Learning.
- Is built using Deep Learning.
- Understands language through NLP.
- Generates content using Generative AI.

---

# Key Points

- Artificial Intelligence (AI) is the broadest field.
- Machine Learning (ML) is a subset of AI.
- Deep Learning (DL) is a subset of Machine Learning.
- NLP enables machines to understand and process human language.
- Generative AI (GenAI) creates new content such as text, images, audio, video, and code.
- Modern GenAI systems are built using Deep Learning techniques.
- ChatGPT is an example of AI + ML + DL + NLP + GenAI working together.
