## Agentic AI Movie Ticket Booking System

### Understanding How AI Agents Work

### Real World Scenario

User says:

"Book 2 tickets for Final Destination movie on Sunday at 5 PM in my nearest PVR."

Instead of only showing information, the AI Agent performs the entire task on behalf of the user.

This is the difference between a traditional AI Application and an Agentic AI System.

# `Traditional AI Application`

## User Query

Show me Final Destination movie timings near me.

## AI Response

PVR Saket

- 5:00 PM
- 7:30 PM
- 9:45 PM

The AI only provides information.

The user must:

1. Open browser
2. Visit movie booking website
3. Select movie
4. Select seats
5. Make payment
6. Confirm booking

### Workflow

User
↓
AI Application
↓
Information

# `Agentic AI System`

## User Query

Book 2 tickets for Final Destination movie on Sunday at 5 PM in nearest PVR.

The AI behaves like a Human Assistant.

### Workflow

User
↓
AI Agent
↓
Find Movie
↓
Find Nearest PVR
↓
Check Availability
↓
Select Seats
↓
Make Payment
↓
Confirm Booking

The Agent performs the task automatically.

# `Step 1: User Input`

Natural Language Input

Book 2 tickets for Final Destination movie on Sunday at 5 PM in nearest PVR.

This is unstructured text.

AI cannot directly perform actions using raw text.

The text must first be converted into structured data.


# `Step 2: NLP (Natural Language Processing)`

NLP extracts useful entities from the user's request.

Extracted Information:

- Movie Name
- Number of Tickets
- Date
- Time
- Theatre Preference
- Location

Workflow:

Text
↓
NLP
↓
Structured Data
↓
JSON


```code
user_request = {
    "movie_name": "Final Destination",
    "tickets": 2,
    "date": "Sunday",
    "time": "5:00 PM",
    "theatre": "PVR",
    "location": "nearest"
}

user_request

```

# Why JSON?

JSON (JavaScript Object Notation) is the universal language for data exchange.

Every system understands JSON:

- APIs
- Databases
- AI Models
- AI Agents
- Web Applications
- Mobile Applications

JSON acts as the communication bridge between systems.

```code
booking_request = {
    "intent": "book_movie_ticket",
    "movie_name": "Final Destination",
    "tickets": 2,
    "theatre": "PVR",
    "preferred_time": "5:00 PM",
    "date": "Sunday",
    "location": "nearest"
}

booking_request

```

# `Step 3: Planner Agent`

The Planner Agent is the brain of the Agentic AI system.

Its responsibility is to:

1. Understand the task
2. Break the task into smaller subtasks
3. Assign subtasks to specialized agents

Task:

Book Movie Ticket

Subtasks:

1. Find nearest PVR
2. Check movie timings
3. Check seat availability
4. Select seats
5. Complete payment
6. Confirm booking

# `Step 4: Specialized Agents`

Multiple agents work together.

Examples:

1. Location Agent
2. Movie Search Agent
3. Seat Selection Agent
4. Payment Agent
5. Booking Agent

This architecture is called Multi-Agent Architecture.

```code
location_agent_output = {
    "theatre": "PVR Select City Walk"
}

location_agent_output

```

```code
movie_agent_output = {
    "movie": "Final Destination",
    "show_time": "5:00 PM"
}

movie_agent_output
```

```code
seat_agent_output = {
    "seat_1": "F12",
    "seat_2": "F13"
}

seat_agent_output
```


```code
payment_agent_output = {
    "payment_status": "success"
}

payment_agent_output```

# Complete Agent Workflow
```text
User
↓
OpenAI API
↓
Large Language Model (LLM)
↓
Planner Agent
↓
--------------------------------
Location Agent
Movie Agent
Seat Agent
Payment Agent
Booking Agent
--------------------------------
↓
Booking Confirmation

The Planner Agent coordinates all agents.
```

# Role of OpenAI API

The LLM does not directly book tickets.

The LLM performs:

- Understanding
- Reasoning
- Planning
- Decision Making

Then it calls external tools and APIs.

Workflow:
```code
User Request
↓
OpenAI API
↓
LLM
↓
Generate JSON
↓
Call APIs
↓
Execute Actions
↓
Return Results
```
