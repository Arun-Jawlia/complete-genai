# Types of Machine Learning

## What is Machine Learning?

### Definition

**Machine Learning (ML)** is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed.

> **Machine Learning = Learning from Data + Making Predictions**

---

# `Types of Machine Learning`

Machine Learning is mainly divided into **4 types**:

1. Supervised Learning
2. Unsupervised Learning
3. Semi-Supervised Learning
4. Reinforcement Learning

---

# `1. Supervised Learning`

## Definition

Supervised Learning is a type of Machine Learning where the model learns using **labeled data**.

### What is Labeled Data?

Labeled data contains:

- Input Features (X)
- Correct Output/Target (Y)

The model learns the relationship between **X** and **Y**.

### Workflow

```
Labeled Data

(X, Y)

↓

Machine Learning Algorithm

↓

Trained Model

↓

Prediction
```

---

## Example

### House Price Prediction

| Area (sq.ft) | Bedrooms | Price |
|--------------|----------|--------|
| 1000 | 2 | ₹40L |
| 1500 | 3 | ₹60L |
| 2000 | 4 | ₹90L |

Here,

**Input (X)**

- Area
- Bedrooms

**Output (Y)**

- House Price

The model learns to predict the price of a new house.

---

## Applications

- House Price Prediction
- Email Spam Detection
- Disease Prediction
- Credit Card Fraud Detection
- Student Marks Prediction
- Loan Approval Prediction

---

## `Types of Supervised Learning`

### A. Regression

Predicts **continuous numerical values**.

Examples

- House Price
- Salary
- Temperature

Popular Algorithms

- Linear Regression
- Decision Tree Regressor
- Random Forest Regressor

---

### B. Classification

Predicts **categories or classes**.

Examples

- Spam / Not Spam
- Yes / No
- Cat / Dog

Popular Algorithms

- Logistic Regression
- Decision Tree
- Random Forest
- SVM
- Naive Bayes

---

# `2. Unsupervised Learning`

## Definition

Unsupervised Learning uses **unlabeled data**.

The model tries to discover hidden patterns or groups without knowing the correct output.

### Workflow

```
Input Data

↓

Machine Learning Algorithm

↓

Hidden Patterns / Clusters
```

---

## Example

Customer Segmentation

Customers are grouped based on:

- Age
- Income
- Spending Score

No labels are provided.

The algorithm automatically forms customer groups.

---

## Applications

- Customer Segmentation
- Market Basket Analysis
- Recommendation Systems
- Anomaly Detection
- Topic Modeling

---

## Popular Algorithms

- K-Means Clustering
- DBSCAN
- Hierarchical Clustering
- PCA (Dimensionality Reduction)

---

# `3. Semi-Supervised Learning`

## Definition

Semi-Supervised Learning uses a combination of:

- Small amount of labeled data
- Large amount of unlabeled data

The model first learns from labeled examples and then uses unlabeled data to improve its performance.

---

## Workflow

```
Small Labeled Data

+

Large Unlabeled Data

↓

Machine Learning Algorithm

↓

Prediction
```

---

## Example

Medical Image Classification

- 500 X-ray images are labeled.
- 20,000 X-ray images are unlabeled.

The model uses both datasets for training.

---

## Applications

- Medical Diagnosis
- Face Recognition
- Speech Recognition
- Image Classification

---

# `4. Reinforcement Learning`

## Definition

Reinforcement Learning (RL) is a type of Machine Learning where an **agent learns by interacting with an environment**.

The agent receives:

- Reward for good actions.
- Penalty for bad actions.

Its goal is to maximize the total reward over time.

---

## Workflow

```code
Environment

↓

Agent

↓

Action

↓

Reward / Penalty

↓

Learning
```

---

## Example

A Robot Learning to Walk

- Takes a step → Reward if balanced.
- Falls down → Penalty.
- Repeats the process until it learns to walk.

---

## Applications

- Self-Driving Cars
- Robotics
- Game Playing (Chess, Go)
- Recommendation Systems
- Resource Optimization

---

# `Linear Regression Algorithm`

## What is Linear Regression?

### Definition

**Linear Regression** is a **Supervised Machine Learning algorithm** used to predict **continuous numerical values**.

It finds the **best-fit straight line** that describes the relationship between input features and the target variable.

> **Linear Regression predicts a numerical value by fitting a straight line through the data.**

---

## Why is it called "Linear"?

Because it assumes a **linear relationship** between the input (`X`) and the output (`Y`).

Example:

As the area of a house increases, its price generally increases in a roughly linear way.

---

## Mathematical Equation

\[
Y = mX + c
\]

Where:

- **Y** = Predicted Output
- **X** = Input Feature
- **m** = Slope (change in Y for a unit change in X)
- **c** = Intercept (value of Y when X = 0)

---

## How Linear Regression Works

```
Training Data

↓

Find Best-Fit Line

↓

Learn Relationship

↓

Predict New Values
```

---

## Example

House Price Prediction

| Area (sq.ft) | Price (₹ Lakhs) |
|--------------|-----------------|
| 1000 | 40 |
| 1500 | 60 |
| 2000 | 80 |

If a new house has an area of **1800 sq.ft**, the model predicts its price using the learned line.

---

## Real-Life Applications

- House Price Prediction
- Salary Prediction
- Sales Forecasting
- Stock Price Trend Analysis
- Demand Forecasting

---

## Advantages

- Easy to understand.
- Fast to train.
- Highly interpretable.
- Works well for linear relationships.

---

## Limitations

- Assumes a linear relationship.
- Sensitive to outliers.
- May underperform on complex, non-linear datasets.

# Interview Tips

### Remember

**Supervised Learning**
- Uses labeled data.
- Solves Regression and Classification problems.

**Unsupervised Learning**
- Uses unlabeled data.
- Finds hidden patterns.

**Semi-Supervised Learning**
- Uses both labeled and unlabeled data.

**Reinforcement Learning**
- Learns through rewards and penalties.

**Linear Regression**
- Supervised algorithm.
- Used for Regression.
- Predicts continuous numerical values.

---

# Key Takeaways

- Machine Learning has **4 major types**.
- Supervised Learning uses labeled data.
- Unsupervised Learning finds hidden patterns in unlabeled data.
- Semi-Supervised Learning combines labeled and unlabeled data.
- Reinforcement Learning learns by trial and error using rewards.
- Linear Regression is a supervised regression algorithm that predicts continuous values by fitting a straight line.

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is Machine Learning?

**Answer:**
Machine Learning is a branch of AI that enables computers to learn patterns from data and make predictions without being explicitly programmed.

---

### Q2. What are the four types of Machine Learning?

**Answer:**

1. Supervised Learning
2. Unsupervised Learning
3. Semi-Supervised Learning
4. Reinforcement Learning

---

### Q3. What is the difference between Supervised and Unsupervised Learning?

**Answer:**

* **Supervised Learning** uses labeled data (input + correct output).
* **Unsupervised Learning** uses unlabeled data and discovers hidden patterns or clusters.

---

### Q4. What is Linear Regression?

**Answer:**
Linear Regression is a supervised learning algorithm used to predict continuous numerical values by modeling a linear relationship between input features and the target variable.

---

### Q5. Is Linear Regression used for Classification?

**Answer:**
No. Linear Regression is used for **Regression** problems, not Classification.

---

## Intermediate Questions

### Q6. Give examples of Regression and Classification.

**Answer:**

* **Regression:** House Price Prediction, Salary Prediction.
* **Classification:** Spam Detection, Disease Diagnosis, Loan Approval.

---

### Q7. When should you use Semi-Supervised Learning?

**Answer:**
When only a small portion of the data is labeled but a large amount of unlabeled data is available, such as in medical imaging.

---

### Q8. Give a real-world example of Reinforcement Learning.

**Answer:**
A self-driving car learns to navigate roads by receiving rewards for safe driving and penalties for unsafe actions.
