# Types of Machine Learning

## What is Machine Learning?

### Definition

**Machine Learning (ML)** is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed.

> **Machine Learning = Learning from Data + Making Predictions**

---

# `Types of Machine Learning`

Machine Learning is mainly divided into **4 types**:

1. Supervised Learning
2. Unsupervised Learning
3. Semi-Supervised Learning
4. Reinforcement Learning

---

# `1. Supervised Learning`

## Definition

Supervised Learning is a type of Machine Learning where the model learns using **labeled data**.

### What is Labeled Data?

Labeled data contains:

- Input Features (X)
- Correct Output/Target (Y)

The model learns the relationship between **X** and **Y**.

### Workflow

```
Labeled Data

(X, Y)

↓

Machine Learning Algorithm

↓

Trained Model

↓

Prediction
```

---

## Example

### House Price Prediction

| Area (sq.ft) | Bedrooms | Price |
|--------------|----------|--------|
| 1000 | 2 | ₹40L |
| 1500 | 3 | ₹60L |
| 2000 | 4 | ₹90L |

Here,

**Input (X)**

- Area
- Bedrooms

**Output (Y)**

- House Price

The model learns to predict the price of a new house.

---

## Applications

- House Price Prediction
- Email Spam Detection
- Disease Prediction
- Credit Card Fraud Detection
- Student Marks Prediction
- Loan Approval Prediction

---

## `Types of Supervised Learning`

### A. Regression

Predicts **continuous numerical values**.

Examples

- House Price
- Salary
- Temperature

Popular Algorithms

- Linear Regression
- Decision Tree Regressor
- Random Forest Regressor

---

### B. Classification

Predicts **categories or classes**.

Examples

- Spam / Not Spam
- Yes / No
- Cat / Dog

Popular Algorithms

- Logistic Regression
- Decision Tree
- Random Forest
- SVM
- Naive Bayes

---

# `2. Unsupervised Learning`

## Definition

Unsupervised Learning uses **unlabeled data**.

The model tries to discover hidden patterns or groups without knowing the correct output.

### Workflow

```
Input Data

↓

Machine Learning Algorithm

↓

Hidden Patterns / Clusters
```

---

## Example

Customer Segmentation

Customers are grouped based on:

- Age
- Income
- Spending Score

No labels are provided.

The algorithm automatically forms customer groups.

---

## Applications

- Customer Segmentation
- Market Basket Analysis
- Recommendation Systems
- Anomaly Detection
- Topic Modeling

---

## Popular Algorithms

- K-Means Clustering
- DBSCAN
- Hierarchical Clustering
- PCA (Dimensionality Reduction)

---

# `3. Semi-Supervised Learning`

## Definition

Semi-Supervised Learning uses a combination of:

- Small amount of labeled data
- Large amount of unlabeled data

The model first learns from labeled examples and then uses unlabeled data to improve its performance.

---

## Workflow

```
Small Labeled Data

+

Large Unlabeled Data

↓

Machine Learning Algorithm

↓

Prediction
```

---

## Example

Medical Image Classification

- 500 X-ray images are labeled.
- 20,000 X-ray images are unlabeled.

The model uses both datasets for training.

---

## Applications

- Medical Diagnosis
- Face Recognition
- Speech Recognition
- Image Classification

---

# `4. Reinforcement Learning`

## Definition

Reinforcement Learning (RL) is a type of Machine Learning where an **agent learns by interacting with an environment**.

The agent receives:

- Reward for good actions.
- Penalty for bad actions.

Its goal is to maximize the total reward over time.

---

## Workflow

```code
Environment

↓

Agent

↓

Action

↓

Reward / Penalty

↓

Learning
```

---

## Example

A Robot Learning to Walk

- Takes a step → Reward if balanced.
- Falls down → Penalty.
- Repeats the process until it learns to walk.

---

## Applications

- Self-Driving Cars
- Robotics
- Game Playing (Chess, Go)
- Recommendation Systems
- Resource Optimization

---

# `Linear Regression Algorithm`

## What is Linear Regression?

### Definition

**Linear Regression** is a **Supervised Machine Learning algorithm** used to predict **continuous numerical values**.

It finds the **best-fit straight line** that describes the relationship between input features and the target variable.

> **Linear Regression predicts a numerical value by fitting a straight line through the data.**

---

## Why is it called "Linear"?

Because it assumes a **linear relationship** between the input (`X`) and the output (`Y`).

Example:

As the area of a house increases, its price generally increases in a roughly linear way.

---

## Mathematical Equation

\[
Y = mX + c
\]

Where:

- **Y** = Predicted Output
- **X** = Input Feature
- **m** = Slope (change in Y for a unit change in X)
- **c** = Intercept (value of Y when X = 0)

---

## How Linear Regression Works

```
Training Data

↓

Find Best-Fit Line

↓

Learn Relationship

↓

Predict New Values
```

---

## Example

House Price Prediction

| Area (sq.ft) | Price (₹ Lakhs) |
|--------------|-----------------|
| 1000 | 40 |
| 1500 | 60 |
| 2000 | 80 |

If a new house has an area of **1800 sq.ft**, the model predicts its price using the learned line.

---

## Real-Life Applications

- House Price Prediction
- Salary Prediction
- Sales Forecasting
- Stock Price Trend Analysis
- Demand Forecasting

---

## Advantages

- Easy to understand.
- Fast to train.
- Highly interpretable.
- Works well for linear relationships.

---

## Limitations

- Assumes a linear relationship.
- Sensitive to outliers.
- May underperform on complex, non-linear datasets.

# `Linear Regression`
---

## `What is Linear Regression?`

### Definition

**Linear Regression** is a **Supervised Machine Learning algorithm** used to predict **continuous numerical values** by finding the best-fit straight line between input features and the target variable.

In simple words:

> **Linear Regression predicts a numerical value by learning a straight-line relationship between input and output.**

---

# `Why is it called Linear Regression?`

It is called **Linear** because it assumes a **linear (straight-line) relationship** between the independent variable(s) and the dependent variable.

Example:

```
Area of House ↑

↓

House Price ↑
```

As the area increases, the price generally increases.

---

# `When Should We Use Linear Regression?`

Use Linear Regression when:

- Target variable is **continuous**
- Relationship between variables is approximately linear

Examples

- House Price Prediction
- Salary Prediction
- Sales Forecasting
- Temperature Prediction
- Stock Price Trend (simple cases)

---

# `Types of Linear Regression`

## `1. Simple Linear Regression`

Uses **one independent variable** to predict the target.

### Example

```
Area

↓

House Price
```

Equation

\[
Y = mX + c
\]

---

## `2. Multiple Linear Regression`

Uses **multiple independent variables** to predict the target.

### Example

```
Area
Bedrooms
Bathrooms
Parking

↓

House Price
```

Equation

\[
Y = b_0 + b_1X_1 + b_2X_2 + ... + b_nX_n
\]

---

# `Important Terminology`

## Independent Variable (Feature)

Also called:

- Input
- Predictor
- Feature
- X

Examples

- Area
- Age
- Experience

---

## Dependent Variable (Target)

Also called:

- Output
- Target
- Label
- Y

Examples

- House Price
- Salary
- Sales

---

# `Linear Regression Equation`


`Y = mX + c`


Where:

| Symbol | Meaning |
|---------|----------|
| Y | Predicted Value |
| X | Input Feature |
| m | Slope |
| c | Intercept |

---

# `Understanding the Equation`

## X (Independent Variable)

The input value.

Example

```
Area = 1500 sq.ft
```

---

## Y (Dependent Variable)

The value we want to predict.

Example

```
House Price = ₹60 Lakhs
```

---

## m (Slope)

Slope tells us **how much Y changes when X increases by one unit.**

Example

```
Slope = 5000
```

Means

Every additional **1 sq.ft**

↓

House Price increases by **₹5,000**

---

## c (Intercept)

Intercept is the predicted value of **Y when X = 0**.

Example

If

```code
X = 0

House Price = ₹10 Lakhs
```

Then

```
Intercept = ₹10 Lakhs
```

---

# `How Does Linear Regression Work?`

```code
Training Data

↓

Find Best-Fit Line

↓

Learn Relationship

↓

Predict New Values
```

The algorithm tries to find the straight line that minimizes prediction errors.

---

# Best-Fit Line

Example Dataset

| Area | Price |
|------|-------|
|1000|40|
|1500|60|
|2000|80|

Best Fit Line

```
Price

90 |                         ●
80 |                      ●
70 |
60 |                  ●
50 |
40 |              ●
30 |
   ---------------------------------

      1000   1500   2000

             Area
```

The line represents the relationship learned by the model.

---

```
Y = 5000X + 1000000
```

House Area

Prediction

```
Price

= 5000 × 1800 + 1000000

= ₹1,00,00,000
```

# Assumptions of Linear Regression

1. Linear relationship between X and Y.
2. No multicollinearity (for multiple regression).
3. Errors are independent.
4. Constant variance of errors (Homoscedasticity).
5. Errors are approximately normally distributed.

---

# Evaluation Metrics

Common metrics for Linear Regression:

- MAE (Mean Absolute Error)
- MSE (Mean Squared Error)
- RMSE (Root Mean Squared Error)
- R² Score (Coefficient of Determination)
