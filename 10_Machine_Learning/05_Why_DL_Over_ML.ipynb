# `Why Did We Shift from Machine Learning (ML) to Deep Learning (DL)?`

---

Before understanding Deep Learning, we must understand the limitations of traditional Machine Learning.

---

# Machine Learning (ML)

## Definition

Machine Learning is a subset of Artificial Intelligence that uses statistical techniques to learn patterns between input and output data.

### Workflow

```text
Data
 ↓
ML Algorithm
 ↓
Pattern Learning
 ↓
Prediction
```

The algorithm learns the relationship between:

```text
Input → Output
```

---

# How Machine Learning Works

Suppose we want to identify whether an image contains a dog or a cat.

### Input

```text
Image
```

### Output

```text
Dog / Cat
```

Machine Learning cannot directly understand images.

Therefore, we need to manually extract important features.

---

# Feature Engineering in Machine Learning

Before feeding data into an ML model, humans create features manually.

### Example

For Dog vs Cat Classification

Possible features:

```text
Ear Length
Tail Length
Face Shape
Body Size
Color
Weight
```

Workflow:

```text
Image
 ↓
Human Extracts Features
 ↓
Feature Vector
 ↓
ML Algorithm
 ↓
Prediction
```

---

# Problem with Machine Learning

Feature Engineering is:

- Time consuming
- Expensive
- Domain dependent
- Requires experts

### Example

For different problems, different features are needed.

#### Face Recognition

Features:

```text
Eye Distance
Face Shape
Nose Length
```

#### Spam Detection

Features:

```text
Word Frequency
Email Length
```

#### Speech Recognition

Features:

```text
Audio Characteristics
Pitch
Frequency
```

Humans must create features manually.

This becomes difficult for complex problems.

---

# Another Limitation of ML

Machine Learning works well with:

```text
Small Data
Medium Data
Structured Data
```

Examples:

- House Price Prediction
- Sales Forecasting
- Customer Churn Prediction

However, when data becomes very large:

```text
Millions of Images
Billions of Words
Thousands of Hours of Audio
```

Traditional ML struggles to capture complex patterns.

---

# Need for Deep Learning

Researchers wanted a system that could:

- Learn automatically
- Discover features automatically
- Handle huge amounts of data
- Solve highly complex problems

This led to the development of Deep Learning.

---

# What is Deep Learning?

Deep Learning is a subset of Machine Learning that uses Artificial Neural Networks with multiple layers to learn patterns from data.

### Workflow

```text
Data
 ↓
Deep Learning Model
 ↓
Pattern Learning
 ↓
Prediction
```

Unlike ML:

```text
No Manual Feature Engineering Required
```

---

# Inspiration Behind Deep Learning

Deep Learning is inspired by the most intelligent system on Earth:

```text
Human Brain
```

The human brain consists of:

```text
Neurons
```

connected with one another.

```text
Neuron
  ↓
Neuron
  ↓
Neuron
```

Information flows through these connections.

Scientists tried to replicate this behavior computationally.

---

# Artificial Neural Network (ANN)

Artificial Neural Networks are inspired by biological neurons.

### Human Brain

```text
Neuron
 ↕
 Neuron
 ↕
 Neuron
```

### Neural Network

```text
Perceptron
 ↕
 Perceptron
 ↕
 Perceptron
```

Perceptrons are artificial neurons.

---

# Structure of a Neural Network

A neural network consists of layers.

```text
Input Layer
      ↓
Hidden Layer
      ↓
Hidden Layer
      ↓
Hidden Layer
      ↓
Output Layer
```

Each layer learns different patterns from the data.

---

# Why Is It Called Deep Learning?

Because there are multiple hidden layers.

```text
Input
 ↓
Layer 1
 ↓
Layer 2
 ↓
Layer 3
 ↓
Layer 4
 ↓
Output
```

More hidden layers:

```text
More Learning Capacity
More Complex Pattern Detection
```

Hence the term:

```text
Deep Learning
```

---

# Biggest Advantage of Deep Learning

## Automatic Feature Extraction

Machine Learning:

```text
Image
 ↓
Human Creates Features
 ↓
ML Model
 ↓
Prediction
```

Deep Learning:

```text
Image
 ↓
Deep Learning Model
 ↓
Automatic Feature Detection
 ↓
Prediction
```

No human intervention is required.

The model automatically learns:

- Edges
- Shapes
- Patterns
- Objects
- Relationships

from raw data.

---

# Example: Cat vs Dog Classification

## Machine Learning

```text
Image
 ↓
Extract Ear Size
Extract Tail Length
Extract Face Shape
 ↓
ML Model
 ↓
Dog / Cat
```

Humans create features.

---

## Deep Learning

```text
Image
 ↓
Neural Network
 ↓
Feature Learning
 ↓
Dog / Cat
```

Features are automatically learned.

---

# Data Requirement in Deep Learning

Machine Learning:

```text
Less Data
Works Well
```

Deep Learning:

```text
More Data
Better Performance
```

The more data we provide:

```text
More Learning
Better Patterns
Higher Accuracy
```

---

# Relationship Between Data and Performance

```text
More Data
      ↓
More Learning
      ↓
Better Model
      ↓
Better Prediction
```

This is one reason why Deep Learning became extremely successful after the rise of Big Data.

---

# Why Deep Learning Outperformed Machine Learning

## Machine Learning

```text
Manual Features
Limited Complexity
Smaller Datasets
```

## Deep Learning

```text
Automatic Features
Complex Pattern Learning
Massive Datasets
```

Deep Learning can learn patterns that humans may not even notice.

# ML vs DL Comparison

| Machine Learning | Deep Learning |
|------------------|--------------|
| Subset of AI | Subset of ML |
| Uses statistical algorithms | Uses neural networks |
| Manual feature engineering required | Automatic feature extraction |
| Works with smaller datasets | Performs best with large datasets |
| Less computational power | High computational power |
| Easier to interpret | Harder to interpret |
| Suitable for structured data | Suitable for structured and unstructured data |
| Lower training time | Higher training time |
| Limited complexity handling | Learns highly complex patterns |

---

# Why We Shifted from ML to DL

1. Machine Learning requires manual feature engineering.
2. Deep Learning automatically extracts features.
3. Deep Learning can learn highly complex patterns.
4. Deep Learning performs better on images, audio, and text.
5. Deep Learning improves significantly with large datasets.
6. Deep Learning powers modern AI systems such as:
   - Self-Driving Cars
   - ChatGPT
   - Image Generation
   - Speech Recognition
   - Recommendation Systems

---

# One-Line Summary

```text
We shifted from Machine Learning to Deep Learning because Deep Learning automatically learns features from large amounts of data using neural networks, making it capable of solving complex real-world problems that traditional Machine Learning struggles with.
```
