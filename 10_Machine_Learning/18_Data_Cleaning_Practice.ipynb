# convert Data Type

```python
df.info()
```

```python
df['Heart Rate (BPM)'] = df['Heart Rate (BPM)'].astype("float32")
```

```python
df['Blood Oxygen Level (%)'] = df['Blood Oxygen Level (%)'].astype("float32")
```

```python
# Convert Step Count to a numeric float type for consistent analysis.
df['Step Count'] = df['Step Count'].astype("float32")
```

```python
# Confirm the updated data types and non-null counts after conversions.
df.info()
```

```python
# Check how many values are missing in Blood Oxygen Level.
df['Blood Oxygen Level (%)'].isnull().sum()
```

```python
# Calculate the mean blood oxygen level to use for imputing missing values.
blood_oxygen_mean = df['Blood Oxygen Level (%)'].mean()
blood_oxygen_mean
```

```python
# Fill missing blood oxygen values with the column mean.
df['Blood Oxygen Level (%)'] = df['Blood Oxygen Level (%)'].fillna(blood_oxygen_mean)
```

```python
# Verify that Blood Oxygen Level no longer has missing values.
df['Blood Oxygen Level (%)'].isnull().sum()
```

# Step Count

Clean the `Step Count` column by checking for missing values and removing rows where step count is unavailable.


```python
# Check missing values in Step Count before cleaning.
df['Step Count'].isnull().sum()
```

```python
# Drop rows where Step Count is missing because this feature is important for activity analysis.
df.dropna(subset=['Step Count'], inplace=True)
```

```python
# Confirm that missing Step Count values have been removed.
df['Step Count'].isnull().sum()
```

## Sleep Duration

Clean the `Sleep Duration (hours)` column by removing invalid entries, converting it to a numeric type, and filtering unrealistic outliers.


```python
# Review unique sleep duration values and their frequencies.
df['Sleep Duration (hours)'].value_counts()
```

```python
# Remove rows with invalid text entries in the sleep duration column.
df = df[df['Sleep Duration (hours)'] != 'ERROR']
```

```python
# Recheck the values after removing invalid entries.
df['Sleep Duration (hours)'].value_counts()
```

```python
# Convert Sleep Duration to a numeric float type for calculations.
df['Sleep Duration (hours)'] = df['Sleep Duration (hours)'].astype("float32")
```

```python
# Check the maximum sleep duration before outlier removal.
df['Sleep Duration (hours)'].max()
```

```python
# Remove Outliers

# Keep realistic sleep durations and remove unusually high values.
```

```python
# Filter out sleep durations of 12 hours or more as potential outliers.
df = df[df['Sleep Duration (hours)'] < 12.0]
```

```python
# Confirm the maximum sleep duration after outlier removal.
df['Sleep Duration (hours)'].max()
```

# Activity Level

Clean the `Activity Level` column by removing rows where activity information is missing.


```python
# Check missing values in Activity Level.
df['Activity Level'].isnull().sum()
```

```python
# Drop rows where Activity Level is missing.
df.dropna(subset=['Activity Level'], inplace=True)
```

```python
# Confirm that Activity Level has no missing values.
df['Activity Level'].isnull().sum()
```

# Stress Level

Clean the `Stress Level` column by removing invalid categories, converting the column to numeric format, and filling missing values.


```python
# Check the current data type of Stress Level.
df['Stress Level'].dtype
```

```python
# Review Stress Level values before cleaning.
df['Stress Level'].value_counts()
```

```python
# Check missing values in Stress Level.
df['Stress Level'].isnull().sum()
```

```python
# Remove invalid text category from Stress Level.
df = df[df['Stress Level'] != 'Very High']
```

```python
# Confirm the remaining Stress Level values.
df['Stress Level'].value_counts()
```

```python
# Convert Stress Level to a numeric float type.
df['Stress Level'] = df['Stress Level'].astype('float32')
```

```python
# Fill missing stress values with a neutral score of 5.
df['Stress Level'] = df['Stress Level'].fillna(5)
```

```python
# Verify that Stress Level no longer has missing values.
df['Stress Level'].isnull().sum()
```

```python
# Final check for missing values across all columns after cleaning.
df.isnull().sum()
```
