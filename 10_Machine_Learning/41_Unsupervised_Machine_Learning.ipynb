# `What is Unsupervised Machine Learning?`

## Definition

**Unsupervised Machine Learning** is a type of machine learning where the model learns patterns from data **without knowing the correct output (labels).**

Unlike supervised learning, there is **no target column** available.

The model tries to discover hidden structures, patterns, or relationships within the data.

> **Unsupervised Learning = Learning from unlabeled data**

---

# `Supervised vs Unsupervised Learning`

| Supervised Learning | Unsupervised Learning |
|----------------------|-----------------------|
| Has input (X) and output (Y) | Has only input (X) |
| Learns from labeled data | Learns from unlabeled data |
| Predicts output | Finds hidden patterns |
| Example: House Price Prediction | Example: Customer Segmentation |

---

# `Why Do We Use Unsupervised Learning?`

Sometimes, we have a lot of data but **no labels**.

Example:

A company has customer information:

- Age
- Salary
- Spending
- City

But it doesn't know:

- Premium Customer
- Regular Customer
- New Customer

The algorithm automatically groups similar customers.

---

# `What is Clustering?`

## Definition

**Clustering** is an Unsupervised Learning technique used to group similar data points into clusters.

Data points inside the same cluster are more similar to each other than to points in other clusters.

> **Clustering = Grouping similar data together**

---

# `Goal of Clustering`

The main objective is:

- Maximize similarity within a cluster.
- Minimize similarity between different clusters.

Example:

```code
Customers

↓

Cluster 1

↓

High Income

↓

High Spending

---------------------

Cluster 2

↓

Medium Income

↓

Medium Spending

---------------------

Cluster 3

↓

Low Income

↓

Low Spending
```

---

# `How Clustering Works`

Suppose we have customer data:

| Customer | Age | Salary |
|-----------|-----|---------|
| A | 25 | 30K |
| B | 27 | 32K |
| C | 55 | 90K |
| D | 58 | 95K |

The algorithm notices:

- A and B are similar.
- C and D are similar.

So it creates two clusters automatically.

# `Types of Clustering Algorithms`

## `1. K-Means Clustering ⭐`

Most popular clustering algorithm.

- Fast
- Simple
- Efficient

Requires the number of clusters (`K`) before training.

---

## `2. Hierarchical Clustering`

Creates a tree-like hierarchy of clusters.

Useful for understanding relationships between clusters.

---

## `3. DBSCAN`

Groups dense regions together.

Can detect:

- Arbitrary-shaped clusters
- Outliers

---

# `K-Means Clustering (Introduction)`

## How K-Means Works

1. Choose the number of clusters (`K`).
2. Initialize K centroids randomly.
3. Assign each point to the nearest centroid.
4. Update the centroid.
5. Repeat until centroids stop changing.

---

# Example of K-Means

Suppose we have customers:

```code
Customer Data

↓

K = 3

↓

Cluster 1

↓

Students

-----------------

Cluster 2

↓

Working Professionals

-----------------

Cluster 3

↓

Business Owners
```

# `Advantages of Clustering`

- Finds hidden patterns
- No labeled data required
- Helps in customer segmentation
- Useful for recommendation systems
- Easy to understand

---

# `Limitations`

- Difficult to evaluate accuracy
- Choosing the right number of clusters can be challenging
- Sensitive to noisy data (especially K-Means)
- Different algorithms work better for different datasets
