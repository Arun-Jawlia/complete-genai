# Introduction to LangChain and LangGraph

## `What is LangChain?`

### Definition

**LangChain** is an open-source framework used to build **AI applications** by connecting (or chaining) Large Language Models (LLMs) with other components such as prompts, tools, memory, databases, and APIs.

The name **LangChain** comes from:

- **Lang** → Language Model
- **Chain** → Connecting multiple operations together

> **LangChain = Language Model + Chain**

---

---

# `Why Do We Need LangChain?`

A single LLM can answer questions, but real-world AI applications require multiple steps.

For example, an AI assistant may need to:

1. Receive a user query.
2. Search a database or the web.
3. Retrieve relevant information.
4. Send the information to an LLM.
5. Generate a final response.

Instead of manually writing all these steps, **LangChain** allows us to connect them into a workflow (chain).

---

# Example: Automated Text Summarization

### Without LangChain

```
Input Text
      ↓
ChatGPT
      ↓
Summarized Text
```

This works well for simple tasks.

---

### With LangChain

```
Input Text
      ↓
Preprocessing
      ↓
LLM
      ↓
Post-processing
      ↓
Final Summary
```

Here, multiple steps are connected to build a complete AI application.

---

# `What is Chaining?`

### Definition

A **chain** is a sequence of operations where the output of one step becomes the input for the next step.

### Example

```
User Input
      ↓
Prompt Template
      ↓
LLM
      ↓
Output Formatter
      ↓
Final Response
```

Each component works together to complete the task.

---

# `What is LangGraph?`

### Definition

**LangGraph** is a framework built on top of LangChain that helps create **complex AI workflows** using graph-based execution.

Instead of a simple linear chain, LangGraph allows multiple paths, decisions, loops, and interactions.

> LangGraph is ideal for building AI Agents.

---

# `LangChain vs LangGraph`

| LangChain | LangGraph |
|------------|-----------|
| Uses linear chains | Uses graph-based workflows |
| Suitable for simple AI applications | Suitable for complex AI applications |
| Easier to learn | More advanced |
| Sequential execution | Supports branching and looping |
| Best for chatbots, summarizers, Q&A | Best for AI Agents and multi-step reasoning |

---

# `Handling Missing Values in Machine Learning`

## Introduction

In real-world datasets, missing values are very common. Before training a Machine Learning model, these missing values must be handled properly because they can negatively impact model performance.

Instead of manually cleaning data every time, we can automate the process using tools such as **SimpleImputer**.

---

# Automatic Data Cleaning

## What is Automatic Data Cleaning?

Automatic Data Cleaning is the process of automatically detecting and handling data quality issues such as missing values, instead of manually fixing them.

### Benefits

- Saves time
- Reduces manual effort
- Produces consistent results
- Essential for production ML pipelines
- Can be integrated into preprocessing pipelines

---

# `What is SimpleImputer?`

## Definition

**SimpleImputer** is a preprocessing class in **Scikit-learn** used to automatically replace missing values with a specified strategy.

It is commonly used during the data preprocessing stage.

### Import

```python
from sklearn.impute import SimpleImputer
```

---

# Common Imputation Strategies

| Strategy | Used For | Example |
|----------|-----------|----------|
| Mean | Numerical Data | Missing Age → Average Age |
| Median | Numerical Data (with outliers) | Missing Salary → Median Salary |
| Most Frequent | Categorical Data | Missing Gender → Male/Female |
| Constant | Any Data Type | Replace with "Unknown" or 0 |

---

# Types of Missing Data

Understanding *why* data is missing is important because the handling strategy depends on the type of missingness.

There are **three main types** of missing data.

---

# 1. `Missing Completely At Random (MCAR)`

## Definition

Data is missing completely by chance.

There is **no relationship** between the missing value and any other variable in the dataset.

> The missing values occur randomly.

---

## Characteristics

- No identifiable pattern
- Random occurrence
- Least problematic type of missing data

---

## Examples

### Example 1

A data entry operator accidentally skips some rows while entering customer ages.

| Name | Age |
|------|-----|
| Arun | 25 |
| Ravi | NaN |
| Priya | 30 |

Reason:
- Human error

---

### Example 2

A system glitch causes random sensor readings to be lost.

---

## How to Handle MCAR?

Common methods:

- Mean Imputation
- Median Imputation
- Mode Imputation
- SimpleImputer
- Delete rows (if very few values are missing)

---

# 2. `Missing At Random (MAR)`

## Definition

Missing values are related to **another observed variable**, but **not** to the missing value itself.

> There is a recognizable pattern in the missing data.

---

## Characteristics

- Missingness depends on another column.
- A pattern exists in the dataset.

---

## Example

### Healthcare Dataset

Suppose we have:

| Gender | Weight |
|----------|--------|
| Male | 72 |
| Female | NaN |
| Male | 80 |
| Female | NaN |

Here, weight values are missing mainly for females.

The missingness depends on the **Gender** column.

Therefore, this is **Missing At Random (MAR)**.

---

## More Examples

- Income missing only for retired people.
- Blood pressure missing only for elderly patients.
- Salary missing only in one department.

---

## How to Handle MAR?

Possible techniques:

- Group-wise Mean/Median Imputation
- SimpleImputer
- Predictive Imputation
- KNN Imputer
- Iterative Imputer

---

# 3. `Missing Not At Random (MNAR)`

## Definition

Missing values are related to the **missing value itself**.

People intentionally or naturally avoid providing certain information.

> The reason for missingness is hidden within the missing value.

---

## Characteristics

- Deliberate missing values
- Most difficult type to handle
- Can introduce bias if ignored

---

## Examples

### High Income Earners

People with very high salaries may choose not to disclose their income.

| Person | Monthly Income |
|---------|----------------|
| A | ₹50,000 |
| B | NaN |
| C | ₹60,000 |

Reason:
- High-income individuals intentionally hide their salary.

---

### Other Examples

- Patients hide medical conditions.
- People skip age on social media.
- Customers refuse to share phone numbers.
- Survey participants avoid sensitive questions.

---

## How to Handle MNAR?

Possible approaches:

- Collect additional data
- Domain expert consultation
- Model-based imputation
- Create a "Missing" category
- Understand the business context before imputing

---

# Comparison of Missing Data Types

| Feature | MCAR | MAR | MNAR |
|----------|------|------|-------|
| Pattern Exists? | ❌ No | ✅ Yes | ✅ Yes |
| Depends on Other Variables? | ❌ No | ✅ Yes | Sometimes |
| Depends on Itself? | ❌ No | ❌ No | ✅ Yes |
| Easy to Handle? | ✅ Yes | ⚠️ Moderate | ❌ Difficult |

---

# SimpleImputer Example

```python
from sklearn.impute import SimpleImputer
import numpy as np

data = [[25],
        [30],
        [np.nan],
        [28]]

imputer = SimpleImputer(strategy="mean")

clean_data = imputer.fit_transform(data)

print(clean_data)
```

### Output

```text
[[25.]
 [30.]
 [27.67]
 [28.]]
```

The missing value is replaced with the **mean** of the available values.

# Key Takeaways
