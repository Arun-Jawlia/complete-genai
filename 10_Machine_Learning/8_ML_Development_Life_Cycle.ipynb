# **`Machine Learning Development Life Cycle  ( MLDLC )`**
---

## Introduction

Machine Learning Development Life Cycle (MLDLC) is a systematic process used to build, train, deploy, and maintain Machine Learning models.

Just as Software Development follows SDLC (Software Development Life Cycle), Machine Learning projects follow MLDLC.

The purpose of MLDLC is to transform raw data into a trained Machine Learning model that can make accurate predictions and solve real-world problems.

---

## Why Do We Need MLDLC?

Without a proper process:

- Data may be incorrect.
- Models may perform poorly.
- Predictions may be inaccurate.
- Deployment becomes difficult.
- Business objectives may not be achieved.

Therefore, a structured Machine Learning Life Cycle is required.

---

## `Machine Learning Development Life Cycle`

```text
1. Problem Definition
          ↓
2. Data Collection / Gathering
          ↓
3. Data Preprocessing
          ↓
4. Exploratory Data Analysis (EDA)
          ↓
5. Feature Engineering
          ↓
6. Model Selection
          ↓
7. Model Training
          ↓
8. Model Evaluation
          ↓
9. Model Deployment
          ↓
10. Monitoring and Maintenance
```

---

## `Overview of All Stages`

### 1. Problem Definition

Understand the business problem and define the objective.

Solution could be supervised / unsupervised or regression or classification

### 2. Data Collection / Gathering

Collect relevant data from different sources.

### 3. Data Preprocessing

Clean and prepare data for analysis.

### 4. Exploratory Data Analysis (EDA)

Analyze data and discover patterns.

### 5. Feature Engineering

Create useful features for the model.

### 6. Model Selection

Choose the appropriate Machine Learning algorithm.

### 7. Model Training

Train the model using training data.

### 8. Model Evaluation

Measure model performance using evaluation metrics.

### 9. Model Deployment

Deploy the model into a production environment.

### 10. Monitoring and Maintenance

Monitor performance and retrain when necessary.

---

## Real-World Example

Suppose a company wants to predict house prices.

### Problem

### Input Features

```text
Area
Number of Rooms
Location
Age of House
```

### Output

```text
House Price
```

The Machine Learning Life Cycle helps build a model capable of predicting house prices accurately.

---

## Key Points

- MLDLC stands for Machine Learning Development Life Cycle.
- It is a step-by-step process for building ML solutions.
- The life cycle starts with problem definition and ends with monitoring.
- Following MLDLC improves model quality and reliability.
- Every successful ML project follows these stages.

# `Step 3: Data Cleaning & Preprocessing`
---

## Definition

Data Preprocessing is the process of cleaning, transforming, and preparing raw data before it is used for training a Machine Learning model.

The data collected from real-world sources is usually:

- Incomplete
- Noisy
- Inconsistent
- Incorrect
- Unstructured

Therefore, before feeding data into a Machine Learning algorithm, it must be cleaned and transformed into a suitable format.

---

## Why Do We Need Data Preprocessing?

Machine Learning algorithms expect:

```text
Clean Data
Consistent Data
Meaningful Data
```

However, real-world data often contains problems.

Example:

| Age | Salary | City |
|------|---------|------|
| 25 | 30000 | Delhi |
| ? | 45000 | Mumbai |
| 30 | ? | Delhi |
| -5 | 60000 | Chennai |

Problems:

```text
Missing Values
Invalid Values
Inconsistent Data
```

If such data is used directly:

```text
Poor Data
      ↓
Poor Learning
      ↓
Poor Predictions
```

---

## Objectives of Data Preprocessing

The main objectives are:

- Remove errors
- Handle missing values
- Standardize data
- Convert data into numerical format
- Improve data quality

---

## Major Steps in Data Preprocessing

```text
Raw Data
    ↓
Data Cleaning
    ↓
Handling Missing Values
    ↓
Handling Duplicates
    ↓
Encoding Categorical Data
    ↓
Feature Scaling
    ↓
Processed Data
```

---

## 1. Data Cleaning

### Definition

Data Cleaning is the process of identifying and correcting errors in the dataset.

---

### Example

Raw Data:

| Age |
|------|
| 25 |
| -10 |
| 30 |

Problem:

```text
Age = -10
```

which is invalid.

After Cleaning:

| Age |
|------|
| 25 |
| 30 |

---

## 2. Handling Missing Values

### Definition

Missing values occur when information is not available for certain records.

---

### Example

| Age | Salary |
|------|---------|
| 25 | 30000 |
| 30 | ? |
| ? | 45000 |

---

### Methods to Handle Missing Values

#### Remove Records

Delete rows containing missing values.

---

#### Replace with Mean

Example:

```text
Age Values:

20
25
30
35
```

Mean:

```text
(20+25+30+35)/4 = 27.5
```

Missing values can be replaced with:

```text
27.5
```

---

#### Replace with Median

Suitable when outliers exist.

---

#### Replace with Mode

Used for categorical data.

Example:

```text
Delhi
Mumbai
Delhi
Delhi
```

Mode:

```text
Delhi
```

---

## 3. Handling Duplicate Data

### Definition

Duplicate records occur when the same row appears multiple times.

---

### Example

| ID | Name |
|----|------|
| 1 | Arun |
| 1 | Arun |
| 2 | Amit |

Duplicate:

```text
ID = 1
```

One duplicate record should be removed.

---

## 4. Handling Inconsistent Data

### Example

```text
Male
male
MALE
```

All represent the same category.

Standardized Version:

```text
Male
Male
Male
```

---

## 5. Encoding Categorical Data

## Problem

Machine Learning algorithms cannot understand text directly.

Example:

| City |
|--------|
| Delhi |
| Mumbai |
| Chennai |

Algorithms understand numbers.

---

### Solution

Convert categories into numbers.

Example:

| City | Encoded Value |
|--------|--------------|
| Delhi | 0 |
| Mumbai | 1 |
| Chennai | 2 |

---

### Label Encoding

```text
Delhi → 0
Mumbai → 1
Chennai → 2
```

---

### One-Hot Encoding

```text
Delhi   → [1,0,0]
Mumbai  → [0,1,0]
Chennai → [0,0,1]
```

---

## 6. Feature Scaling

### Definition

Feature Scaling is the process of bringing numerical values into a common range.

---

### Why Is It Needed?

Example:

| Age | Salary |
|------|----------|
| 25 | 30000 |
| 30 | 60000 |

Notice:

```text
Age Range = 20–50

Salary Range = 30,000–100,000
```

Salary values are much larger.

The algorithm may give more importance to salary.

---

### Solution

Scale values to a similar range.

---

### Normalization

Range:

```text
0 to 1
```

Example:

```text
30000 → 0.3
60000 → 0.6
```

---

### Standardization

Transforms data to:

```text
Mean = 0
Standard Deviation = 1
```

Widely used in Machine Learning.

---

## 7. Feature Selection

### Definition

Feature Selection is the process of selecting only important features.

---

### Example

House Price Prediction

Available Features:

```text
Area
Bedrooms
Location
Owner's Favorite Color
```

The feature:

```text
Owner's Favorite Color
```

may not affect house price.

It can be removed.

---

## Data Preprocessing Workflow

```text
Raw Data
     ↓
Data Cleaning
     ↓
Handle Missing Values
     ↓
Remove Duplicates
     ↓
Encode Categories
     ↓
Feature Scaling
     ↓
Feature Selection
     ↓
Clean Dataset
```

---

## Example

### Raw Data

| Age | Salary | City |
|------|---------|------|
| 25 | 30000 | Delhi |
| ? | 45000 | Mumbai |
| 30 | ? | Delhi |

---

### After Preprocessing

| Age | Salary | City |
|------|---------|------|
| 25 | 30000 | 0 |
| 27.5 | 45000 | 1 |
| 30 | 37500 | 0 |

The dataset is now ready for analysis and model training.

---

## Output of Data Preprocessing

The output of this stage is:

```text
Clean Dataset
```

which is:

- Complete
- Consistent
- Numerical
- Ready for Machine Learning

---

## Key Points

- Data Preprocessing is the third stage of MLDLC.
- It converts raw data into clean and usable data.
- Missing values, duplicates, and inconsistencies are handled.
- Categorical data is converted into numerical form.
- Feature Scaling ensures all features are on a similar scale.
- The output of this stage is a clean dataset ready for analysis and model building.

# `Step 4: Exploratory Data Analysis (EDA)`
---

## Definition

Exploratory Data Analysis (EDA) is the process of analyzing, understanding, and visualizing data to discover patterns, relationships, trends, anomalies, and insights before building a Machine Learning model.

EDA helps us answer questions such as:

- What does the data look like?
- What patterns exist in the data?
- Are there any missing values?
- Are there any outliers?
- Which features are important?
- What relationships exist between variables?

EDA is often called:

```text
Data Investigation Phase
```

because we investigate the dataset before training a model.

---

## Why Do We Need EDA?

Without understanding data:

```text
Wrong Assumptions
        ↓
Poor Model
        ↓
Poor Predictions
```

EDA helps us:

- Understand data structure
- Discover hidden patterns
- Detect anomalies
- Identify relationships
- Select important features

---

## Objectives of EDA

The main objectives are:

- Understand dataset characteristics
- Identify data quality issues
- Detect missing values
- Detect outliers
- Find feature relationships
- Generate insights for model building

---

## Workflow of EDA

```text
Clean Dataset
       ↓
Data Understanding
       ↓
Statistical Analysis
       ↓
Visualization
       ↓
Pattern Discovery
       ↓
Insights
```

---

## Example Dataset

Suppose we have:

| Area | Bedrooms | Price |
|--------|----------|--------|
| 1000 | 2 | 50 |
| 1200 | 3 | 60 |
| 1500 | 4 | 75 |
| 1800 | 4 | 90 |

Before building a model, we need to understand:

```text
How Area affects Price?
How Bedrooms affect Price?
Are there any unusual values?
```

This is the purpose of EDA.

# `Step 5: Feature Engineering`
---

## Definition

Feature Engineering is the process of creating, transforming, extracting, and selecting the most useful features from raw data to improve the performance of a Machine Learning model.

It is one of the most important stages in the Machine Learning Development Life Cycle because the quality of features directly affects the quality of predictions.

A common statement in Machine Learning is:

> Better Features → Better Model → Better Predictions

---

## What is a Feature?

A Feature is an individual measurable property or characteristic of data that is used as input to a Machine Learning model.

### Example

House Price Prediction Dataset:

| Area | Bedrooms | Age | Price |
|--------|----------|------|--------|
| 1000 | 2 | 5 | 50 Lakh |

Features:

```text
Area
Bedrooms
Age
```

Target Variable:

```text
Price
```

---

## Why Do We Need Feature Engineering?

Raw data often contains:

- Irrelevant information
- Redundant information
- Missing relationships
- Complex patterns

Feature Engineering helps convert raw data into meaningful information.

---

## Example

Suppose we have:

| Date of Birth |
|--------------|
| 15-08-2000 |

For prediction, Date of Birth may not be useful directly.

We can create:

```text
Age = Current Year - Birth Year
```

Now:

```text
Age
```

becomes a useful feature.

This process is called Feature Engineering.

---

## Objectives of Feature Engineering

- Improve model accuracy
- Reduce noise
- Highlight important patterns
- Simplify learning
- Improve prediction performance

---

## Types of Feature Engineering

```text
Feature Engineering
        │
        ├── Feature Creation
        │
        ├── Feature Transformation
        │
        ├── Feature Extraction
        │
        └── Feature Selection
```

---

## 1. Feature Creation

### Definition

Feature Creation is the process of generating new features from existing features.

---

### Example

Dataset:

| Length | Width |
|---------|--------|
| 10 | 5 |

Create:

```text
Area = Length × Width
```

New Dataset:

| Length | Width | Area |
|---------|--------|------|
| 10 | 5 | 50 |

The newly created feature may provide more useful information.

---

## Real-Life Example

Customer Dataset:

| Date of Birth |
|--------------|
| 15-08-2000 |

Create:

```text
Age
```

from Date of Birth.

---

## 2. Feature Transformation

### Definition

Feature Transformation converts existing features into a more suitable form.

---

### Example

Salary:

```text
10000
50000
1000000
```

Large variations may affect learning.

Transformation:

```text
Log(Salary)
```

This reduces extreme differences.

---

## Common Transformations

### Scaling

```text
0 → 1 Range
```

### Standardization

```text
Mean = 0
Standard Deviation = 1
```

### Log Transformation

Used for skewed data.

---

## 3. Feature Extraction

### Definition

Feature Extraction is the process of extracting important information from raw data and converting it into features.

---

## Example: Image Recognition

Raw Image:

```text
Millions of Pixels
```

Extract Features:

```text
Edges
Shapes
Patterns
Textures
```

These features are then used by the model.

---

## Example: Audio Processing

Raw Audio:

```text
Voice Signal
```

Feature Extraction:

```text
MFCC
(Mel Frequency Cepstral Coefficients)
```

Output:

```text
Numerical Features
```

which can be used for speech recognition.

---

## Example: Text Processing

Text:

```text
I Love Machine Learning
```

Feature Extraction:

```text
Bag of Words
TF-IDF
Word Embeddings
```

Converted into numerical vectors.

---

## Important Note

In Traditional Machine Learning:

```text
Feature Extraction
```

is often performed manually.

In Deep Learning:

```text
Feature Extraction
```

is performed automatically by neural networks.

This is one major reason why Deep Learning became popular.

---

## 4. Feature Selection

### Definition

Feature Selection is the process of selecting only the most important features and removing unnecessary ones.

---

## Example

House Price Dataset:

| Area | Bedrooms | Location | Favorite Color | Price |
|--------|----------|-----------|---------------|--------|

Important Features:

```text
Area
Bedrooms
Location
```

Unimportant Feature:

```text
Favorite Color
```

Removing irrelevant features:

- Reduces complexity
- Improves speed
- Improves model performance

---

## Benefits of Feature Selection

### Faster Training

Less data means faster computation.

---

### Better Accuracy

Irrelevant features can confuse the model.

---

### Reduced Overfitting

The model focuses only on meaningful information.

# `Step 7: Model Training`
---

## Definition

Model Training is the process of teaching a Machine Learning algorithm to learn patterns and relationships from the training data.

During training, the algorithm analyzes the input data and corresponding outputs, identifies hidden patterns, and creates a mathematical model capable of making predictions on unseen data.

In simple words:

> Model Training is the process through which a Machine Learning algorithm learns from data.

---

## Why Do We Need Model Training?

Before training:

```text
Data Available
       ↓
No Knowledge
```

After training:

```text
Data Available
       ↓
Pattern Learning
       ↓
Knowledge Acquired
```

The trained model can then make predictions on new data.

---

## Example

Training Data:

| Hours Studied | Marks |
|--------------|--------|
| 2 | 30 |
| 4 | 50 |
| 6 | 70 |
| 8 | 90 |

The algorithm observes:

```text
2 → 30
4 → 50
6 → 70
8 → 90
```

and learns:

```text
More Study Hours
        ↓
Higher Marks
```

---

## What Happens During Training?

The model tries to discover:

```text
Input
   ↓
Relationship
   ↓
Output
```

Example:

```text
Area
Bedrooms
Location
        ↓
House Price
```

The algorithm learns how input features affect the target variable.

---

## Training Dataset

The data used for learning is called:

```text
Training Data
```

Example:

| Area | Price |
|--------|--------|
| 1000 | 50 |
| 1200 | 60 |
| 1500 | 75 |

The model learns patterns from this dataset.

---

## Train-Test Split

Before training, the dataset is usually divided into two parts.

### Training Set

Used to teach the model.

---

### Testing Set

Used to evaluate the model.

---

## Example

Dataset:

```text
1000 Records
```

Split:

```text
Training Data = 80%
Testing Data = 20%
```

or

```text
800 Records → Training

200 Records → Testing
```

Visualization:

```text
Dataset
    ↓
 ┌───────────┬───────────┐
 │ Training  │ Testing   │
 │    80%    │    20%    │
 └───────────┴───────────┘
```

---

## Why Do We Use Train-Test Split?

If we train and test on the same data:

```text
Model Memorizes Data
```

which may produce misleading results.

Testing on unseen data checks whether the model has actually learned patterns.

## Learning Process

During training:

```text
Input Features
        ↓
Algorithm
        ↓
Pattern Discovery
        ↓
Model
```

The algorithm repeatedly adjusts itself to improve predictions.

## Example: House Price Prediction

is a parameter learned during training.

### Examples

Unlike parameters:

```text
Hyperparameters
```

### Example

Actual Price:

Predicted Price:

Error:

```text
₹5 Lakh
```

The loss function calculates this error.

Visualization:

# `Step 8: Model Evaluation`
---

## Definition

Model Evaluation is the process of measuring how well a trained Machine Learning model performs on unseen data.

After training, we need to answer:

```text
Has the model learned correctly?
Can it make accurate predictions?
Will it work on new data?
```

Model Evaluation helps determine whether the model is good enough for deployment.

---

## Why Do We Need Model Evaluation?

A model may perform very well on training data but poorly on new data.

Example:

```text
Training Accuracy = 99%

Testing Accuracy = 60%
```

This indicates:

```text
Overfitting
```

Therefore, evaluating the model on unseen data is essential.

---

## Goal of Model Evaluation

The main objectives are:

- Measure model performance
- Compare different models
- Detect overfitting and underfitting
- Select the best model
- Ensure good generalization

---

## Evaluation Workflow

```text
Training Data
       ↓
Model Training
       ↓
Trained Model
       ↓
Testing Data
       ↓
Performance Metrics
       ↓
Model Evaluation
```

---

## Example

Suppose we trained a model to predict house prices.

Testing Data:

| Actual Price | Predicted Price |
|-------------|----------------|
| 50 | 48 |
| 60 | 62 |
| 75 | 73 |

The evaluation process measures how close predictions are to actual values.

---

## Types of Evaluation Metrics

The metrics depend on the type of problem.

```text
Regression Problems
        ↓
MAE
MSE
RMSE
R² Score

Classification Problems
        ↓
Accuracy
Precision
Recall
F1 Score
Confusion Matrix
```

---

## Evaluation for Classification Problems

Classification examples:

```text
Spam Detection
Disease Prediction
Dog vs Cat Classification
Fraud Detection
```

---

## Accuracy

### Definition

Accuracy measures the percentage of correct predictions made by the model.

Formula:

```text
Accuracy =
Correct Predictions
-------------------
Total Predictions
```

---

### Example

Out of:

```text
100 Predictions
```

the model predicts:

```text
90 Correctly
```

Accuracy:

```text
90%
```

---

## Limitation of Accuracy

Consider:

```text
1000 Emails

990 Not Spam
10 Spam
```

Model predicts:

```text
All Emails = Not Spam
```

Accuracy:

```text
990/1000 = 99%
```

Looks excellent but the model never identifies spam emails.

Therefore, accuracy alone is not enough.

---

## Confusion Matrix

### Definition

A Confusion Matrix is a table used to evaluate classification models.

It shows:

```text
Correct Predictions
Wrong Predictions
```

---

### Structure

| | Predicted Positive | Predicted Negative |
|---|---|---|
| Actual Positive | TP | FN |
| Actual Negative | FP | TN |

---

### Terms

#### True Positive (TP)

Model predicts:

```text
Positive
```

and actual value is also:

```text
Positive
```

Example:

```text
Spam Email → Predicted Spam
```

---

#### True Negative (TN)

Model predicts:

```text
Negative
```

and actual value is also:

```text
Negative
```

Example:

```text
Normal Email → Predicted Normal
```

---

#### False Positive (FP)

Model predicts:

```text
Positive
```

but actual value is:

```text
Negative
```

Example:

```text
Normal Email → Predicted Spam
```

---

#### False Negative (FN)

Model predicts:

```text
Negative
```

but actual value is:

```text
Positive
```

Example:

```text
Spam Email → Predicted Normal
```

---

## Precision

### Definition

Precision measures how many predicted positive cases are actually positive.

Formula:

```text
Precision =
TP
---------
TP + FP
```

---

### Example

Suppose:

```text
Predicted Spam Emails = 100

Actually Spam = 90
```

Precision:

```text
90%
```

---

### Use Case

Important when:

```text
False Positives are costly.
```

Examples:

- Spam Detection
- Fraud Detection

---

## Recall

### Definition

Recall measures how many actual positive cases were correctly identified.

Formula:

```text
Recall =
TP
---------
TP + FN
```

---

### Example

Suppose:

```text
Actual Spam Emails = 100

Detected Spam Emails = 95
```

Recall:

```text
95%
```

---

### Use Case

Important when:

```text
Missing positives is dangerous.
```

Examples:

- Cancer Detection
- Disease Prediction

---

## F1 Score

### Definition

F1 Score is the balance between Precision and Recall.

Formula:

```text
2 × Precision × Recall
----------------------
Precision + Recall
```

---

### Why Use F1 Score?

If:

```text
Precision High
Recall Low
```

or

```text
Precision Low
Recall High
```

F1 Score provides a balanced evaluation.

---

## Evaluation for Regression Problems

Regression examples:

```text
House Price Prediction
Salary Prediction
Temperature Prediction
```

---

## MAE (Mean Absolute Error)

### Definition

Measures the average absolute difference between actual and predicted values.

Example:

| Actual | Predicted |
|----------|------------|
| 100 | 90 |
| 200 | 210 |

Errors:

```text
10
10
```

MAE:

```text
(10 + 10) / 2

= 10
```

---

## MSE (Mean Squared Error)

### Definition

Measures average squared error.

Example:

Errors:

```text
10
10
```

Squared Errors:

```text
100
100
```

MSE:

```text
(100 + 100)/2

= 100
```

---

## RMSE (Root Mean Squared Error)

### Definition

Square root of MSE.

Advantages:

```text
Easy to Interpret
Same Unit as Output
```

---

## R² Score

### Definition

Measures how well the model explains the variation in data.

Range:

```text
0 to 1
```

Interpretation:

```text
0 → Poor Model

1 → Perfect Model
```

Example:

```text
R² = 0.90
```

Meaning:

```text
90% of variation explained by model.
```

---

## Detecting Underfitting and Overfitting

### Underfitting

Training Accuracy:

```text
Low
```

Testing Accuracy:

```text
Low
```

Meaning:

```text
Model Too Simple
```

---

### Overfitting

Training Accuracy:

```text
Very High
```

Testing Accuracy:

```text
Low
```

Meaning:

```text
Model Memorized Data
```

---

### Good Fit

Training Accuracy:

```text
High
```

Testing Accuracy:

```text
High
```

Meaning:

```text
Model Learned Patterns Correctly
```

---

## Model Comparison

Example:

| Model | Accuracy |
|----------|----------|
| Logistic Regression | 88% |
| Decision Tree | 91% |
| Random Forest | 95% |

Best Model:

```text
Random Forest
```

because it performs better on testing data.

---

## Output of Model Evaluation

At the end of this stage, we obtain:

```text
Performance Metrics
```

such as:

```text
Accuracy
Precision
Recall
F1 Score
MAE
RMSE
R² Score
```

These metrics help determine whether the model is ready for deployment.

---

## Workflow

```text
Trained Model
       ↓
Testing Data
       ↓
Predictions
       ↓
Performance Metrics
       ↓
Evaluation
       ↓
Model Selection
```

---

## Key Points

- Model Evaluation measures how well a trained model performs on unseen data.
- Classification models use Accuracy, Precision, Recall, F1 Score, and Confusion Matrix.
- Regression models use MAE, MSE, RMSE, and R² Score.
- Evaluation helps detect overfitting and underfitting.
- Multiple models can be compared using evaluation metrics.
- Only well-performing models should be deployed.
- The output of this stage is a validated model ready for deployment.

# `Step 10: Monitoring and Maintenance`

## Definition

Monitoring and Maintenance is the final stage of the Machine Learning Development Life Cycle (MLDLC).

After deployment, the Machine Learning model must be continuously monitored to ensure that it continues to perform accurately and reliably in real-world environments.

Over time, data patterns may change, causing the model's performance to degrade. Therefore, regular maintenance and retraining are necessary.

In simple words:

> Monitoring and Maintenance is the process of continuously tracking, evaluating, improving, and updating a deployed Machine Learning model.

---

## Why Do We Need Monitoring?

A model may perform well during testing but fail after deployment.

Reason:

```text
Real World Data
≠
Training Data
```

The environment changes continuously.

Examples:

- Customer behavior changes
- Market conditions change
- User preferences change
- New products are introduced
- Fraud patterns evolve

As a result:

```text
Model Accuracy
      ↓
Decreases Over Time
```

---

## Example

Suppose a spam detection model was trained in 2024.

Training Data:

```text
Win Lottery
Free Prize
Claim Reward
```

By 2026, spammers may start using new words.

Example:

```text
Crypto Giveaway
NFT Reward
Digital Bonus
```

The model may fail to detect these new spam patterns.

Therefore:

```text
Monitoring
      +
Retraining
```

becomes necessary.

---

## Objectives of Monitoring

The main objectives are:

- Track model performance
- Detect performance degradation
- Identify errors
- Detect data drift
- Detect concept drift
- Retrain when necessary

---

## Monitoring Workflow

```text
Deployed Model
        ↓
Real User Data
        ↓
Performance Monitoring
        ↓
Issue Detection
        ↓
Model Improvement
        ↓
Retraining
```

---

## What Should Be Monitored?

### Prediction Accuracy

Check whether prediction accuracy remains acceptable.

Example:

```text
Initial Accuracy = 95%

Current Accuracy = 80%
```

This indicates a problem.

---

### Response Time

Monitor how quickly predictions are generated.

Example:

```text
Expected Time = 1 Second

Current Time = 10 Seconds
```

Performance optimization may be required.

---

### Error Rate

Track failed predictions and system errors.

Example:

```text
Prediction Failure
API Failure
Server Failure
```

---

### Resource Usage

Monitor:

```text
CPU Usage
Memory Usage
Storage Usage
Network Usage
```

---

## Data Drift

### Definition

Data Drift occurs when the characteristics of input data change over time.

---

## Example

Training Data:

```text
Customer Age

20–40 Years
```

Current Data:

```text
Customer Age

50–70 Years
```

Input distribution has changed.

This is:

```text
Data Drift
```

---

## Visualization

```text
Old Data
      ↓
Model Trained
      ↓
New Data Different
      ↓
Performance Drops
```

---

## Concept Drift

### Definition

Concept Drift occurs when the relationship between input and output changes.

---

## Example

Before:

```text
High Income
      ↓
High Purchase Probability
```

After:

```text
High Income
      ↓
Low Purchase Probability
```

Customer behavior has changed.

This is:

```text
Concept Drift
```

---

## Data Drift vs Concept Drift

| Data Drift | Concept Drift |
|------------|---------------|
| Input data changes | Relationship changes |
| Feature distribution changes | Prediction logic changes |
| Easier to detect | Harder to detect |
| Example: Age distribution changes | Example: Customer behavior changes |

---

## Model Retraining

### Definition

Retraining is the process of training the model again using updated data.

---

## Why Retraining Is Needed?

Because:

```text
New Data
     ↓
New Patterns
     ↓
Old Model Becomes Outdated
```

Retraining allows the model to learn the latest patterns.

---

## Retraining Workflow

```text
New Data
      ↓
Data Collection
      ↓
Preprocessing
      ↓
Model Training
      ↓
New Model
      ↓
Deployment
```

---

## Example

Old Model:

```text
Accuracy = 95%
```

After one year:

```text
Accuracy = 80%
```

Retraining with new data:

```text
Accuracy = 94%
```

Performance improves again.

---

## Continuous Learning

Modern AI systems often follow:

```text
Monitor
   ↓
Collect New Data
   ↓
Retrain Model
   ↓
Deploy Updated Model
   ↓
Monitor Again
```

This creates a continuous improvement cycle.

---

## Real-World Examples

### Netflix

Monitors:

```text
User Viewing Behavior
```

and updates recommendation models regularly.

---

### Amazon

Monitors:

```text
Customer Purchases
```

and retrains recommendation systems.

---

### Google Maps

Monitors:

```text
Traffic Patterns
```

and continuously updates route prediction models.

---

### Fraud Detection Systems

Monitor:

```text
New Fraud Techniques
```

and retrain models frequently.

---

## Output of Monitoring and Maintenance

The output of this stage is:

```text
Updated and Reliable Model
```

that continues to perform effectively in changing environments.

---

## Complete Workflow

```text
Deployed Model
       ↓
Performance Monitoring
       ↓
Detect Issues
       ↓
Collect New Data
       ↓
Retraining
       ↓
Updated Model
       ↓
Redeployment
```

---

## Complete Machine Learning Development Life Cycle

```text
1. Problem Definition
          ↓
2. Data Collection
          ↓
3. Data Preprocessing
          ↓
4. Exploratory Data Analysis (EDA)
          ↓
5. Feature Engineering
          ↓
6. Model Selection
          ↓
7. Model Training
          ↓
8. Model Evaluation
          ↓
9. Model Deployment
          ↓
10. Monitoring and Maintenance
```

---

## Key Points

- Monitoring and Maintenance is the final stage of MLDLC.
- Deployed models must be continuously monitored.
- Model performance may decrease over time due to changing data.
- Data Drift occurs when input data changes.
- Concept Drift occurs when the relationship between inputs and outputs changes.
- Retraining helps the model learn new patterns.
- Continuous monitoring ensures long-term model reliability.
- Monitoring and Maintenance make Machine Learning a continuous process rather than a one-time activity.
