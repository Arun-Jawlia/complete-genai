# `Confusion Matrix, Accuracy, Precision, Recall and F-Score`
---

## `Introduction`

After training a **classification model**, we need to evaluate how well it performs.

Unlike regression, where we use metrics like **MAE, MSE, and RMSE**, classification models are evaluated using:

- Confusion Matrix
- Accuracy
- Precision
- Recall
- F1-Score

These metrics help us understand where the model is making correct and incorrect predictions.

---

# `Example Problem`

Suppose we build a model to detect whether an email is **Spam** or **Not Spam**.

Actual Labels:

```
Spam
Not Spam
Spam
Spam
Not Spam
```

Predicted Labels:

```
Spam
Spam
Spam
Not Spam
Not Spam
```

Some predictions are correct, while others are incorrect.

To summarize these predictions, we use a **Confusion Matrix**.

---

# `What is a Confusion Matrix?`

## Definition

A **Confusion Matrix** is a table used to evaluate the performance of a classification model.

It compares:

- Actual values
- Predicted values

and counts:

- Correct predictions
- Incorrect predictions

---

# Confusion Matrix Structure

| Actual \ Predicted | Positive | Negative |
|-------------------|----------|----------|
| Positive | True Positive (TP) | False Negative (FN) |
| Negative | False Positive (FP) | True Negative (TN) |

---

# `Understanding the Four Terms`

## `1. True Positive (TP)`

### Definition

The model correctly predicts the **positive class**.

Example:

Actual = Disease

Predicted = Disease

✅ Correct Prediction

---

## `2. True Negative (TN)`

### Definition

The model correctly predicts the **negative class**.

Example:

Actual = No Disease

Predicted = No Disease

✅ Correct Prediction

---

## `3. False Positive (FP)`

### Definition

The model predicts positive when the actual class is negative.

Also called:

**Type I Error**

Example:

Actual = No Disease

Predicted = Disease

❌ Wrong Prediction

---

## `4. False Negative (FN)`

### Definition

The model predicts negative when the actual class is positive.

Also called:

**Type II Error**

Example:

Actual = Disease

Predicted = No Disease

❌ Wrong Prediction

---

# Simple Example

Suppose 100 patients were tested.

| | Predicted Disease | Predicted Healthy |
|---|---:|---:|
| Actual Disease | 45 | 5 |
| Actual Healthy | 10 | 40 |

Here,

TP = 45

FN = 5

FP = 10

TN = 40

---

# `Accuracy`

## Definition

Accuracy measures the percentage of total predictions that are correct.

Formula:

\[
$Accuracy=\frac{TP+TN}{TP+TN+FP+FN}$
\]

Example:

TP = 45

TN = 40

FP = 10

FN = 5

Accuracy

\[
=$\frac{45+40}{45+40+10+5}$
=$\frac{85}{100}$
=85\%
\]

---

# When Should We Use Accuracy?

Accuracy is useful when the dataset is **balanced**.

Example:

- 50% Positive
- 50% Negative

---

# Limitation of Accuracy

Accuracy can be misleading for **imbalanced datasets**.

Example:

1000 patients

- Healthy = 990
- Disease = 10

Model predicts everyone as healthy.

Accuracy = 99%

But it fails to detect any disease.

So we need Precision and Recall.

---

# `Precision`

## Definition

Precision tells us:

> **Out of all the predicted positive cases, how many were actually positive?**

Formula:

\[
$Precision=\frac{TP}{TP+FP}$
\]

Example:

TP = 45

FP = 10

Precision

\[
=$\frac{45}{45+10}$
=81.8\%
\]

---

# When is Precision Important?

Precision is important when **False Positives are costly**.

Examples:

- Spam Detection
- Fraud Detection

We don't want genuine emails or genuine transactions to be incorrectly flagged.

---

# Recall

## Definition

Recall tells us:

> **Out of all the actual positive cases, how many did the model correctly identify?**

Formula:

\[
$Recall=\frac{TP}{TP+FN}$
\]

Example:

TP = 45

FN = 5

Recall

\[
=$\frac{45}{45+5}$
=90\%
\]

---

# When is Recall Important?

Recall is important when **False Negatives are dangerous**.

Examples:

- Disease Detection
- Cancer Detection

Missing a patient who actually has the disease can be very serious.

---

# Precision vs Recall

| Precision | Recall |
|-----------|--------|
| Focuses on predicted positives | Focuses on actual positives |
| Reduces False Positives | Reduces False Negatives |

---

# `F1-Score`

## Definition

F1-Score combines **Precision** and **Recall** into a single metric.

It is the **harmonic mean** of Precision and Recall.

Formula:

\[
$F1=\frac{2\times Precision\times Recall}{Precision+Recall}$
\]

---

# Why Use F1-Score?

Sometimes:

- Precision is high
- Recall is low

or

- Recall is high
- Precision is low

F1-Score balances both.

---

# When Should We Use F1-Score?

Use F1-Score when:

- Dataset is imbalanced
- Precision and Recall are both important

Examples:

- Medical Diagnosis
- Fraud Detection
- Spam Detection

---

# `Summary Table`

| Metric | Formula | Best Used When |
|---------|----------|----------------|
| Accuracy | (TP+TN)/(Total) | Balanced dataset |
| Precision | TP/(TP+FP) | False Positives are costly |
| Recall | TP/(TP+FN) | False Negatives are costly |
| F1-Score | Harmonic Mean of Precision & Recall | Need balance between Precision & Recall |

---

# `Scikit-learn Implementation`

## Import

```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

---

- Missing a sick patient (FN) is dangerous.

---

- High Precision is preferred.
- Genuine emails should not be marked as spam (FP).

---

---

# Key Takeaways

- A Confusion Matrix summarizes the performance of a classification model.
- TP, TN, FP, and FN form the basis of all classification metrics.
- Accuracy is suitable for balanced datasets.
- Precision is important when minimizing False Positives.
- Recall is important when minimizing False Negatives.
- F1-Score balances Precision and Recall, especially for imbalanced datasets.

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is a Confusion Matrix?

**Answer:**
A Confusion Matrix is a table that compares actual and predicted classes to evaluate a classification model.

---

### Q2. What are TP, TN, FP, and FN?

**Answer:**

* **TP (True Positive):** Correctly predicted positive.
* **TN (True Negative):** Correctly predicted negative.
* **FP (False Positive):** Incorrectly predicted positive.
* **FN (False Negative):** Incorrectly predicted negative.

---

### Q3. What is Accuracy?

**Answer:**
Accuracy is the percentage of correct predictions out of all predictions.

---

### Q4. What is Precision?

**Answer:**
Precision measures how many predicted positive cases are actually positive.

---

### Q5. What is Recall?

**Answer:**
Recall measures how many actual positive cases are correctly identified by the model.

---

### Q6. What is F1-Score?

**Answer:**
F1-Score is the harmonic mean of Precision and Recall. It balances both metrics into a single score.

---

## `Intermediate Questions`

### Q7. When should Accuracy not be used?

**Answer:**
Accuracy should not be relied on for **imbalanced datasets**, as it can be misleading.

---

### Q8. When is Precision more important than Recall?

**Answer:**
When **False Positives** are costly, such as in spam detection or fraud detection.

---

### Q9. When is Recall more important than Precision?

**Answer:**
When **False Negatives** are costly, such as in disease or cancer detection.

---

### Q10. Why is F1-Score useful?

**Answer:**
Because it balances Precision and Recall, making it suitable for imbalanced datasets.

---

## `Scenario-Based Questions`

### Q11. Which metric would you prioritize for cancer detection?

**Answer:**
**Recall**, because missing a patient with cancer (False Negative) is very dangerous.

---

### Q12. Which metric is most important for spam detection?

**Answer:**
**Precision**, because marking legitimate emails as spam (False Positive) should be minimized.

---

### Q13. Which metric would you use for an imbalanced fraud detection dataset?

**Answer:**
**F1-Score**, since both Precision and Recall are important.
