## First Step Of The Machine Learning Development Life Cycle


---
## Case Study: `Netflix` Revenue Down From `2024` To `2025`

**Level:** `Senior Data Scientist` / `Senior ML Engineer`
---


## 1. Business Problem

### Scenario

You are a `Senior Data Scientist` at `Netflix`.

Management asks:

> `Netflix` revenue is lower in `2025` compared to `2024`. Find the reason and suggest ways to increase revenue.

Before jumping into ML algorithms, first understand the business problem.

### Revenue Formula

`Revenue = (Number of Subscribers * Subscription Price) + Advertisement Revenue`

### Possible Reasons

1. Customer churn increased
2. New customer acquisition decreased
3. Ad impressions decreased
4. Users downgraded plans
5. Content quality decreased
6. Competition became stronger
7. Inflation and economic slowdown affected spending
8. Regional cancellations increased


## 2. Convert Business Problem To ML Problem

### Conversion Flow

`Business Problem -> Data Problem -> Mathematical Problem -> Machine Learning Problem`

### Example Business Question

`Why is Netflix revenue down?`

We cannot directly train a model on this question. We must convert it into measurable ML problems.

### Problem 1: Churn Prediction

Predict which users are likely to leave `Netflix`.

### Target Variable

| User Status | Target Value |
|---|---:|
| Leaves `Netflix` | `1` |
| Stays on `Netflix` | `0` |

### ML Type

- `Classification`
- `Supervised Learning`

### Problem 2: Revenue Forecasting

Predict next month's revenue.

### ML Type

- `Regression`
- `Supervised Learning`

### Problem 3: User Segmentation

Group users based on behavior.

### ML Type

- `Clustering`
- `Unsupervised Learning`


## 3. Define Business Goal

### Current Situation

| Metric | Value |
|---|---:|
| Previous churn rate | `8%` |
| Current churn rate | `6%` |

A previous data scientist reduced churn from `8%` to `6%`.

### New Goal

Reduce churn from `6%` to `4%`.

### Business KPI

Decrease churn by `2%`.

### ML Goal

Identify users who are likely to leave `Netflix`.


## 4. Understand The Current Solution

Before building a model, ask:

> How is the problem currently solved?

### Existing Solutions

1. Email notifications
2. Personalized recommendations
3. Discount offers
4. New content alerts

### Research Competitors

- `Disney+`
- `Amazon Prime`
- `Hulu`
- `YouTube Premium`

### Questions To Ask

- How do top companies solve churn?
- What features do they use?
- What models do they use?


## 5. Data Collection

### Data Sources

1. User database
2. Logs
3. Data warehouse
4. Streaming events

### ETL Pipeline

`ETL = Extract -> Transform -> Load`


## 6. ETL Process

### Extract

Collect data from:

- User database
- Viewing history
- Search logs
- Payment systems

### Transform

1. Remove duplicates
2. Handle missing values
3. Handle outliers
4. Create features

### Load

Store data into:

1. Data warehouse
2. Feature store
3. Training dataset


## 7. Features For Churn Prediction

### Possible Features

| Feature | Why It Matters |
|---|---|
| `Watch Time` | Low watch time may indicate low engagement |
| `Browsing Time` | High browsing without watching may show dissatisfaction |
| `Search With No Result Found` | Users may not find desired content |
| `Content Left Midway` | Users may dislike content quality or relevance |
| `Recommendation Click Rate` | Low clicks may show poor personalization |
| `Days Since Last Login` | Long inactivity often indicates churn risk |
| `Subscription Type` | Plan type can influence churn behavior |
| `Payment Failure` | Failed payments can lead to involuntary churn |
| `Number Of Sessions` | Fewer sessions may indicate reduced engagement |
| `Device Type` | Behavior can differ across mobile, TV, and web |

These features indicate engagement level. Lower engagement often leads to churn.


## 8. Define Target Variable

### Target Column

`Churn`

### Example

| User | Churn |
|---|---:|
| `A` | `1` |
| `B` | `0` |
| `C` | `1` |

### Meaning

- `1` = Left `Netflix`
- `0` = Stayed on `Netflix`


## 9. Type Of Machine Learning Problem

### Input Features

- `Watch Time`
- `Browsing Time`
- `Search History`
- `Payment History`
- `Recommendation Clicks`

### Output

`Probability of Churn`

### Example Predictions

| User | Churn Probability | Risk Level |
|---|---:|---|
| `User_A` | `92%` | `High Risk` |
| `User_B` | `15%` | `Low Risk` |
| `User_C` | `80%` | `High Risk` |


## 10. Business Actions

### For High-Risk Users

1. Offer a discount
2. Recommend popular content
3. Send reminder emails
4. Offer free trial extensions
5. Notify users about new releases

The model should not stop at prediction. It should help the business take useful action.


## 11. Evaluation Metrics

### Key Question

How do we know if the model is good?

### Accuracy

`Accuracy = Correct Predictions / Total Predictions`

### Precision

Among users predicted to churn, how many actually churned?

### Recall

Among users who actually churned, how many did we identify?

### F1 Score

`F1 Score` balances `Precision` and `Recall`.


## 12. Business Metrics

Model metrics are not enough. Business metrics matter.

### Example

| Stage | Churn Rate |
|---|---:|
| Before model | `6%` |
| After model | `4.5%` |

### Business Success Indicators

1. Revenue increased
2. Retention increased
3. Customer satisfaction increased


## 13. Online Vs Batch Prediction

### Batch Prediction

`Database -> ML Model -> Daily Prediction`

### Example

Every night, predict churn for all users.

### Advantages

1. Simple
2. Cheap
3. Easy to maintain

### Disadvantage

- Not real time

### Online Prediction

`Streaming Data -> Kafka -> ML Model -> Real-Time Prediction`

### Example

If a user suddenly becomes inactive, the model immediately updates the churn probability.

### Advantages

1. Real time
2. Better user experience

### Disadvantages

1. Expensive
2. Complex infrastructure


## 14. Assumption Checking

A `Senior ML Engineer` always checks assumptions.

### Example Markets

`Netflix` operates in:

- `India`
- `USA`
- `Europe`
- `Japan`

User behavior differs across regions.

### Example Behavior Difference

| Region | Common Device Pattern |
|---|---|
| `India` | Mostly mobile |
| `USA` | Mostly TV |

### Important Question

Does the model work globally, or only for one region?

### Validate By Segment

1. Country wise
2. Device wise
3. Age group wise
4. Subscription wise


## 15. Root Cause Analysis

Suppose analysis shows multiple reasons for revenue decline.

### Reason 1: Churn Increased

**Solution:** Build a churn prediction model.

### Reason 2: Advertisement Revenue Decreased

**Solutions:**

1. Increase active users
2. Improve ad targeting
3. Increase watch time

### Reason 3: Inflation

**Solutions:**

1. Optimize pricing
2. Use regional pricing strategy

### Reason 4: Poor Recommendation System

**Solution:** Improve the recommendation engine.


## 16. Interview Framework

Whenever an interviewer asks about `Revenue Down`, `Sales Down`, `Engagement Down`, `Profit Down`, or `Retention Down`, use this framework.

### Step-By-Step Framework

1. Understand the business problem
2. Define the KPI
3. Convert the business problem to an ML problem
4. Identify the problem type: `Classification`, `Regression`, or `Clustering`
5. Understand the current solution
6. Research competitor solutions
7. Collect data
8. Build the ETL pipeline
9. Perform feature engineering
10. Select the model
11. Define evaluation metrics
12. Choose online or batch prediction
13. Deploy the model
14. Validate assumptions
15. Give business recommendations


## Senior Data Scientist Answer Template

### Interviewer Question

`Netflix` revenue is down. What is your approach?

### Answer

1. Understand the business problem.
2. Identify KPIs: `Revenue`, `Churn`, `Watch Time`, and `Ad Revenue`.
3. Convert the business problem into an ML problem, such as `Churn Prediction`.
4. Define the problem type as `Classification` and `Supervised Learning`.
5. Collect data using an `ETL` pipeline.
6. Create features such as `Watch Time`, `Search History`, and `Recommendation Click Rate`.
7. Train the model.
8. Evaluate using `Precision`, `Recall`, and `F1 Score`.
9. Deploy using `Batch Prediction` or `Online Prediction`.
10. Use predictions to reduce churn.
11. Measure business impact, such as reducing churn from `6%` to `4%`.
12. Validate the model globally across regions and user segments.
13. Increase revenue through better retention, better recommendations, and better ad targeting.


## Quick Notes

### Problem Statement

`Netflix -> Senior Data Scientist -> 2024-2025 -> Why is revenue down?`

You need to explain why revenue decreased and suggest how to increase it.

### Convert The Problem

Convert the business problem into a mathematical and ML problem.

### Possible Revenue Decline Reasons

1. Ad revenue decreased, so increase active customers and improve ad targeting.
2. Inflation affected users, so consider pricing optimization or regional pricing.
3. Churn increased, so build an ML churn prediction model.

### ML Problem Type

Choose the correct problem type:

- `Classification`
- `Regression`
- `Supervised Learning`
- `Unsupervised Learning`
- `Reinforcement Learning`

### Current Solution

Understand the current churn-reduction solution. In startups, also research how top MNCs solve the same problem.

A previous data scientist reduced churn from `8%` to `6%`. Our goal is to reduce it from `6%` to `4-5%`.

### Data Collection

Get data from databases using the `ETL` technique.

`ETL = Extract -> Transform -> Load`

### Useful Features

1. `Watch Time`
2. `Browsing Time`
3. `Search With No Result Found`
4. `Content Left In The Middle`
5. `Clicked On Recommendation Or Not`

### Prediction

For `100M` users, predict the probability that each user will leave the platform.

### Metrics To Measure

- If the model predicts `95%` churn probability, check whether the user actually leaves.
- If the user does not leave, analyze whether the model needs improvement.
- Check whether model predictions are accurate.

### Online Vs Batch

- `Batch`: Run predictions periodically on a server and retrain with new user data.
- `Online`: Use streaming data and update predictions in real time.
