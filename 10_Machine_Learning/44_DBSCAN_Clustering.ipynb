# `DBSCAN Clustering`
---

## What is DBSCAN?

**DBSCAN (Density-Based Spatial Clustering of Applications with Noise)** is an **Unsupervised Machine Learning algorithm** used to group data points based on **density**.

Unlike K-Means and Hierarchical Clustering, DBSCAN can:

- Discover clusters of **any shape**
- Detect **outliers (noise)**
- Does **not require specifying the number of clusters (K)** beforehand

> **DBSCAN groups together points that are closely packed and labels isolated points as noise (outliers).**

---

# `Why Do We Need DBSCAN?`

K-Means has some limitations:

- Need to choose K before training
- Doesn't work well for irregular-shaped clusters
- Sensitive to outliers

DBSCAN solves these problems.

# `Main Idea Behind DBSCAN`

DBSCAN assumes:

> **If many points are close together, they belong to the same cluster.**

If a point is far away from every cluster,

it is treated as an **outlier**.

---

# `Important Parameters`

DBSCAN mainly uses **two parameters**.

## `1. eps (ε)`

### Definition

`eps` is the **maximum distance** between two data points for them to be considered neighbors.

Think of it as a **search radius**.

---

### Small eps

- Very small neighborhood
- Creates many small clusters
- Many points become noise

---

### Large eps

- Larger neighborhood
- More points join clusters
- Different clusters may merge together

---

## `2. min_samples`

### Definition

The minimum number of neighboring points required to form a **dense region (cluster).**

Example

```
min_samples = 5

↓

At least 5 nearby points

↓

Cluster can be formed.
```

---

# `Types of Points in DBSCAN`

## 1. Core Point

A point having at least `min_samples` neighbors inside the `eps` radius.

Core points form the heart of a cluster.

---

## 2. Border Point

A point that has fewer than `min_samples` neighbors but lies within the `eps` radius of a core point.

Border points belong to a cluster but cannot create one.

---

## 3. Noise (Outlier)

A point that is neither:

- Core Point
- Border Point

These points do not belong to any cluster.

# `How DBSCAN Works (Step-by-Step)`

## Step 1

Choose:

- eps
- min_samples

---

## Step 2

Select any point.

---

## Step 3

Find all neighboring points within the eps radius.

---

## Step 4

If the point has at least `min_samples` neighbors,

it becomes a **Core Point**.

---

## Step 5

Expand the cluster by connecting all reachable core and border points.

---

## Step 6

Repeat for all remaining points.

---

## Step 7

Points that do not belong to any cluster are labeled as **Noise**.

---

# `DBSCAN Workflow`

```code
Choose eps & min_samples
          ↓
Select a Point
          ↓
Find Neighbors
          ↓
Core Point?
      ↓        ↓
    Yes        No
     ↓          ↓
Create      Noise /
Cluster     Border Point
     ↓
Expand Cluster
     ↓
Repeat
```

# `Choosing eps and min_samples`

There is no fixed rule.

Common practice:

- Start with small eps.
- Increase gradually.
- Use domain knowledge.
- Use a **K-Distance Graph** to estimate a good eps value.

---

# `Advantages of DBSCAN`

- No need to specify the number of clusters.
- Detects outliers automatically.
- Works well for irregular-shaped clusters.
- Robust to noise.
- Suitable for spatial data.

---

# `Limitations`

- Choosing eps can be difficult.
- Doesn't perform well when cluster densities vary significantly.
- Performance decreases in high-dimensional data.
- Sensitive to parameter selection.

---

# `DBSCAN vs K-Means`

| K-Means | DBSCAN |
|----------|---------|
| Need K before training | No need for K |
| Uses Centroids | Uses Density |
| Cannot detect outliers | Detects outliers automatically |
| Best for spherical clusters | Best for arbitrary-shaped clusters |
| Sensitive to outliers | Robust to outliers |

---

# `DBSCAN vs Hierarchical Clustering`

| Hierarchical | DBSCAN |
|--------------|---------|
| Builds a cluster hierarchy | Builds density-based clusters |
| Uses Dendrogram | Uses Density |
| Doesn't identify noise explicitly | Detects noise naturally |
| Better for small datasets | Better for spatial and noisy datasets |

**Answer:**

**Answer:**
Because it forms clusters based on the density of data points rather than a predefined number of clusters.

**Answer:**
Many points may not have enough neighbors, resulting in many small clusters or noise points.

---

### Q8. What happens if `eps` is too large?

**Answer:**
Different clusters may merge together into one large cluster.
