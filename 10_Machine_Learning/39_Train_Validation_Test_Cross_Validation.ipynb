# `Train Data, Validation Data, Test Data & Cross Validation`

## Introduction

Before building a Machine Learning model, the collected data is divided into different parts.

This helps us:

- Train the model
- Tune the model
- Evaluate the model fairly

The dataset is generally divided into:

1. Train Data
2. Validation Data
3. Test Data

---

# Complete Machine Learning Workflow

```code
Data Gathering
        ↓
Data Preprocessing
(Data Cleaning + Feature Engineering + Feature Selection)
        ↓
Train-Test Split
        ↓
Train Model
        ↓
Hyperparameter Tuning (Validation)
        ↓
Final Model Selection
        ↓
Test on Unseen Data
        ↓
Deploy Model
```

---

# `Why Do We Split Data?`

If we train and test the model on the same data, the model may simply memorize the training examples.

As a result:

- Training accuracy becomes very high.
- Real-world performance becomes poor.

To avoid this, we split the dataset into separate parts.

---

# `1. Train Data`

## Definition

**Train Data** is the portion of the dataset used to teach the model the relationship between input features and output labels.

The model **learns patterns** from this data.

> **Training Data is used to fit the model.**

---

## Purpose of Training Data

- Learn patterns
- Learn mathematical relationships
- Build the Machine Learning model

---

## Example

### Spam Detection

Training Data:

| Email | Spam |
|--------|------|
| Win ₹1,00,000 | Yes |
| Meeting at 5 PM | No |
| Claim your reward | Yes |

The model learns which words or patterns indicate spam.

---

# `2. Validation Data`

## Definition

**Validation Data** is a portion of the training data used to **tune hyperparameters** and select the best model.

The model **does not learn** from this data.

Instead, it checks how well different model configurations perform.

---

# Why Do We Need Validation Data?

Many Machine Learning algorithms have **hyperparameters**.

These values are chosen **before training**.

Example:

- SVM → `C`, `gamma`, `kernel`
- KNN → `k`
- Decision Tree → `max_depth`
- Random Forest → `n_estimators`

Validation data helps us find the best values.

---

# `Hyperparameter Tuning Example (SVM)`

Suppose we are training an SVM model.

### Case 1

```
C = 0.001
```

Result:

- Large margin
- Many misclassified points
- Underfitting

---

### Case 2

```
C = 100
```

Result:

- Very small margin
- Almost every training point classified correctly
- Overfitting

---

### Goal

Find the value of `C` that gives the best performance on **validation data**.

---

# `Hyperparameter Tuning`

## Definition

Hyperparameter tuning is the process of trying different hyperparameter values and selecting the combination that gives the best validation performance.

Examples of hyperparameters:

- `k` in KNN
- `C` in SVM
- `gamma` in SVM
- `max_depth` in Decision Tree
- `n_estimators` in Random Forest

---

# Manual vs Automated Hyperparameter Tuning

### Manual

Try values one by one.

Example:

```
C = 0.001
C = 0.01
C = 0.1
C = 1
C = 10
```

Choose the best result.

---

### Automated

Scikit-learn provides automated techniques such as:

- Grid Search (`GridSearchCV`)
- Random Search (`RandomizedSearchCV`)

These automatically test multiple combinations and select the best hyperparameters.

---

# Validation Data Helps Detect Overfitting

A model may perform well on training data but poorly on unseen data.

Validation data acts like an early check.

Example:

```
Training Accuracy = 99%

Validation Accuracy = 80%
```

This suggests the model is likely overfitting.

---

# `3. Test Data`

## Definition

**Test Data** is completely unseen data used only after the final model has been selected.

It provides an unbiased estimate of how the model will perform in the real world.

> **The model never learns from test data.**

---

## Purpose of Test Data

- Evaluate the final model
- Measure real-world performance
- Estimate model generalization before deployment

---

---

# `Cross Validation`

## What is Cross Validation?

Instead of using one fixed validation set, **Cross Validation** repeatedly changes the validation data so that every sample gets a chance to be validated.

This provides a more reliable estimate of model performance.

---

# `Why Use Cross Validation?`

Suppose your validation data is not representative.

The model may appear good or bad simply because of that particular split.

Cross Validation reduces this problem.

Benefits:

- Better model evaluation
- Better hyperparameter tuning
- Uses data more efficiently
- Reduces dependence on one validation split

---

# `K-Fold Cross Validation`

The most common type is **K-Fold Cross Validation**.

### Steps

1. Divide the training data into **K equal folds**.
2. Choose one fold as the validation set.
3. Train on the remaining folds.
4. Repeat until every fold has been used as validation once.
5. Compute the average performance.

---

## `Example: 5-Fold Cross Validation`

```text
Fold 1 → Validation
Fold 2 → Train
Fold 3 → Train
Fold 4 → Train
Fold 5 → Train

↓

Repeat

Fold 2 → Validation

↓

Repeat

Fold 3 → Validation

...

Until all 5 folds have been used.
```

The final score is the **average** of all five runs.

---

# `Difference Between Train, Validation, and Test Data`

| Dataset | Model Learns? | Purpose |
|----------|---------------|---------|
| Train | ✅ Yes | Learn patterns |
| Validation | ❌ No | Tune hyperparameters and select the best model |
| Test | ❌ No | Final unbiased evaluation |

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is Train Data?

**Answer:**
Train data is the portion of the dataset used to teach the model the relationship between input features and output labels.

---

### Q2. What is Validation Data?

**Answer:**
Validation data is used to tune hyperparameters and select the best-performing model. The model does not learn from this data.

---

### Q3. What is Test Data?

**Answer:**
Test data is completely unseen data used to evaluate the final model before deployment.

---

### Q4. Why do we split the dataset?

**Answer:**
To avoid overfitting and to fairly evaluate how well the model generalizes to unseen data.

---

### Q5. What is Hyperparameter Tuning?

**Answer:**
Hyperparameter tuning is the process of finding the best hyperparameter values (such as `k`, `C`, or `max_depth`) to improve model performance.

---

## Intermediate Questions

### Q6. What is Cross Validation?

**Answer:**
Cross Validation is a technique where different parts of the training data are used as validation in multiple rounds to obtain a more reliable estimate of model performance.

---

### Q7. What is K-Fold Cross Validation?

**Answer:**
In K-Fold Cross Validation, the training data is divided into K folds. Each fold is used once as validation while the remaining folds are used for training.

---

### Q8. What is the difference between model parameters and hyperparameters?

**Answer:**

* **Parameters** are learned during training (e.g., weights).
* **Hyperparameters** are set before training (e.g., `k`, `C`, `gamma`, `max_depth`).

---

### Q9. Name two automated hyperparameter tuning techniques.

**Answer:**

* `GridSearchCV`
* `RandomizedSearchCV`

---

## Scenario-Based Questions

### Q10. Your model achieves 99% training accuracy but only 82% validation accuracy. What does this indicate?

**Answer:**
The model is likely **overfitting** because it performs much better on training data than on unseen validation data.

---

### Q11. Why shouldn't test data be used during hyperparameter tuning?

**Answer:**
Because it would bias the evaluation. Test data must remain completely unseen until the final evaluation.

---

### Q12. Which dataset is most similar to a real-world deployment scenario?

**Answer:**
**Test Data**, because the model has never seen it before.
