# `Metrics in Regression: RMSE, MAE, MSE`
---

- ## MSE: Mean Squared Error
- ## RMSE: Root Mean Squared Error
- ## MAE: Mean Absolute Error

## `What are Regression Metrics?`

Regression metrics are used to measure how well a regression model predicts **continuous numerical values**.

They tell us how far the predicted values are from the actual values.

---

# `Why Do We Need Regression Metrics?`

A regression model may look good, but we need a clear way to check its performance.

Regression metrics help us:

- compare predictions with actual values
- measure prediction errors
- choose the best model
- understand how accurate the model is

---

# `Common Regression Metrics`

The most common regression metrics are:

1. **MAE** → Mean Absolute Error
2. **MSE** → Mean Squared Error
3. **RMSE** → Root Mean Squared Error

---

# `1. MAE - Mean Absolute Error`

## Definition

MAE is the average of the **absolute differences** between actual and predicted values.

### Formula

$MAE = \frac{1}{n}\sum_{i=1}^{n} |y_i - \hat{y}_i|$.

Where:

- \(y_i\) = actual value
- \(\hat{y}_i\) = predicted value
- \(n\) = number of observations

---

## Why Use MAE?

- Easy to understand
- Gives error in the same unit as the target
- Treats all errors equally

---

## Example

Actual values: 10, 20, 30  
Predicted values: 12, 18, 33

Absolute errors:

- |10 - 12| = 2
- |20 - 18| = 2
- |30 - 33| = 3

MAE:


$\frac{2+2+3}{3} = 2.33


So, the average prediction error is **2.33 units**.

---

## Limitation

MAE does not punish large errors heavily.

---

# `2. MSE - Mean Squared Error`

## Definition

MSE is the average of the **squared differences** between actual and predicted values.

### Formula

\[
$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$
\]

---

## Why Use MSE?

- Penalizes large errors more strongly
- Useful when large mistakes are very bad

---

## Example

Actual values: 10, 20, 30  
Predicted values: 12, 18, 33

Squared errors:

- (10 - 12)² = 4
- (20 - 18)² = 4
- (30 - 33)² = 9


\[
$\frac{4+4+9}{3} = 5.67
\]

So, the average squared error is **5.67**.

---

## Limitation

MSE is harder to interpret because the error is squared.

Example:
- If target is in rupees, MSE is in rupees²

---

# `3. RMSE - Root Mean Squared Error`

## Definition

RMSE is the square root of MSE.

### Formula

\[
$RMSE = \sqrt{MSE}$
\]

or

\[
$RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$
\]

---

## Why Use RMSE?

- Easy to interpret
- Same unit as the target variable
- Penalizes large errors more than MAE

---

## Example

If MSE = 5.67

\[
RMSE = \sqrt{5.67} \approx 2.38
\]

So, the typical prediction error is about **2.38 units**.

---

# `Comparison of MAE, MSE, and RMSE`

| Metric | Full Form | Meaning | Penalizes Large Errors? | Unit |
|--------|-----------|---------|--------------------------|------|
| MAE | Mean Absolute Error | Average absolute error | No | Same as target |
| MSE | Mean Squared Error | Average squared error | Yes | Squared unit |
| RMSE | Root Mean Squared Error | Square root of MSE | Yes | Same as target |

---

# `Which Metric Should We Use?`

## Use MAE when:

- You want simple interpretation
- All errors should be treated equally
- Outliers should not dominate the score

---

## Use MSE when:

- Large errors should be punished more
- You want the model to reduce big mistakes

---

## Use RMSE when:

- You want an error measure in the same unit as the target
- You want to penalize larger errors more than smaller ones

---

# `Real-Life Example`

## House Price Prediction

Actual prices:

- 50
- 60
- 70

Predicted prices:

- 52
- 58
- 80

Here:

- MAE tells us the average absolute prediction error
- MSE increases the effect of the large error in the last prediction
- RMSE gives a more interpretable error in the original unit

---

# Important Note

A good regression model should have:

- **Low MAE**
- **Low MSE**
- **Low RMSE**

Lower values mean better predictions.

---

# Interview Tips

### Remember

- **MAE** → average absolute error
- **MSE** → average squared error
- **RMSE** → square root of MSE

### Easy Memory Trick

- **MAE** = “How far are we on average?”
- **MSE** = “How badly do big errors hurt?”
- **RMSE** = “MSE but easier to interpret”

# `Key Takeaways`

- Regression metrics measure prediction error in regression problems.
- MAE is simple and interpretable.
- MSE penalizes large errors more.
- RMSE is the square root of MSE and is easy to understand.
- Lower metric values mean better model performance.

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What are regression metrics?

**Answer:**
Regression metrics are used to measure how close the predicted values are to the actual values in a regression problem.

---

### Q2. What is MAE?

**Answer:**
MAE stands for Mean Absolute Error. It is the average of the absolute differences between actual and predicted values.

---

### Q3. What is MSE?

**Answer:**
MSE stands for Mean Squared Error. It is the average of the squared differences between actual and predicted values.

---

### Q4. What is RMSE?

**Answer:**
RMSE stands for Root Mean Squared Error. It is the square root of MSE.

---

### Q5. Which metric is easiest to interpret?

**Answer:**
MAE and RMSE are easier to interpret because they are in the same unit as the target variable.

---

## Intermediate Questions

### Q6. Why is MSE useful?

**Answer:**
MSE penalizes large errors more heavily, so it is useful when big mistakes are especially undesirable.

---

### Q7. What is the difference between MAE and RMSE?

**Answer:**

* **MAE** treats all errors equally.
* **RMSE** gives more weight to large errors because it is based on squared differences.

---

### Q8. Why is RMSE preferred over MSE in many cases?

**Answer:**
Because RMSE is in the same unit as the target variable, making it easier to interpret.

---

## Scenario-Based Questions

### Q9. Your model has one very large prediction error. Which metric will reflect this more strongly?

**Answer:**
**MSE** and **RMSE** will reflect it more strongly because they square the errors.

---

### Q10. If you want a simple average error, which metric should you use?

**Answer:**
**MAE**, because it gives the average absolute error directly.
