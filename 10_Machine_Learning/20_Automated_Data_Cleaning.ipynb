### `Forward Fill and Backward Fill in Missing Value Imputation`
- After identifying missing values in a dataset, one way to handle them is by using the values from nearby rows instead of statistical measures like mean or median.

- Two commonly used techniques are:

    - Forward Fill (FFill)
    - Backward Fill (BFill)

- These methods are especially useful for **time-series data** and datasets where neighboring values are related.

---

### ` **Forward Fill (FFill)** `: replaces a missing value with the **last valid (non-missing) value** that appears before it.
#### Original Data

| Day | Temperature |
|-----|-------------|
| Monday | 30 |
| Tuesday | NaN |
| Wednesday | NaN |
| Thursday | 32 |
| Friday | 31 |

#### After Forward Fill

| Day | Temperature |
|-----|-------------|
| Monday | 30 |
| Tuesday | 30 |
| Wednesday | 30 |
| Thursday | 32 |
| Friday | 31 |

The missing values are filled using the previous available value.

---

### `**Backward Fill**` (BFill): replaces a missing value with the **next valid (non-missing) value**.

## How It Works

### Original Data

| Day | Temperature |
|-----|-------------|
| Monday | 30 |
| Tuesday | NaN |
| Wednesday | NaN |
| Thursday | 32 |
| Friday | 31 |

### After Backward Fill

| Day | Temperature |
|-----|-------------|
| Monday | 30 |
| Tuesday | 32 |
| Wednesday | 32 |
| Thursday | 32 |
| Friday | 31 |

The missing values are filled using the next available value.

---

### `Forward Fill vs Backward Fill`

| Feature | Forward Fill (FFill) | Backward Fill (BFill) |
|----------|----------------------|-----------------------|
| Uses | Previous value | Next value |
| Direction | Top → Bottom | Bottom → Top |
| Best For | Time-series, sensor data | Missing values at the beginning or when future values are reliable |
| Pandas Function | `ffill()` | `bfill()` |

---

### When Should You Use Forward Fill?

Use Forward Fill when:

- Data is sequential or time-based.
- The previous value is expected to remain valid until a new value appears.

### Examples

- Stock prices
- Temperature readings
- Sensor measurements
- Inventory levels

---

### When Should You Use Backward Fill?

Use Backward Fill when:

- The next value can reasonably represent the missing value.
- Missing values occur before the first recorded observation.

### Examples

- Scheduled events
- Future-known data
- Survey responses where later values apply

---

### Advantages

#### Forward Fill

- Simple and fast.
- Preserves trends in time-series data.
- Easy to implement.

#### Backward Fill

- Useful when future observations are more appropriate.
- Effective for leading missing values.

---

### Limitations

### Forward Fill

- Can propagate outdated values.
- Not suitable if values change rapidly.

### Backward Fill

- Uses future information, which may not be available in real-time applications.
- May introduce data leakage in predictive models if applied incorrectly.

# `Different Techniques of Imputation in Machine Learning`

## `What is Imputation?`

### Definition

**Imputation** is the process of replacing missing values in a dataset with estimated or meaningful values instead of removing the data.

> **Goal:** Preserve as much information as possible while improving data quality.

---

# Why is Imputation Important?

Missing values can:

- Reduce model accuracy
- Cause errors during model training
- Introduce bias
- Reduce the amount of usable data

Imputation helps create a complete dataset for Machine Learning models.

---

# `Types of Imputation Techniques`

There are two broad categories:

## 1. `Simple Imputation`

Uses simple statistical measures or nearby values.

### Techniques

- Mean Imputation
- Median Imputation
- Mode (Most Frequent) Imputation
- Constant Value Imputation
- Forward Fill (FFill)
- Backward Fill (BFill)

---

## 2. `Advanced Imputation`

Uses Machine Learning or statistical algorithms to estimate missing values.

### Techniques

- KNN Imputer
- Iterative Imputer (MICE)
- Regression Imputation
- Random Forest Imputation
- Multiple Imputation
- Deep Learning-based Imputation

---

# 1.` Mean Imputation`

## Definition

Replace missing numerical values with the **average (mean)** of the column.

### Formula

\[
\text{Mean} = \frac{\text{Sum of Values}}{\text{Number of Values}}
\]

### Example

| Age |
|------|
| 20 |
| 25 |
| NaN |
| 35 |

Mean = (20 + 25 + 35) / 3 = **26.67**

After Imputation:

| Age |
|------|
| 20 |
| 25 |
| 26.67 |
| 35 |

### Best For

- Numerical data
- Normally distributed data

### Limitation

Sensitive to outliers.

---

# 2. `Median Imputation`

## Definition

Replace missing values with the **middle value** after sorting the data.

### Example

Values:

20, 25, 30, 1000

Median = **27.5**

### Best For

- Numerical data with outliers
- Skewed distributions

### Advantage

Less affected by extreme values than the mean.

---

# 3. `Mode (Most Frequent) Imputation`

## Definition

Replace missing values with the most frequently occurring value.

### Example

| Gender |
|---------|
| Male |
| Female |
| Male |
| NaN |

Mode = **Male**

After Imputation:

| Gender |
|---------|
| Male |
| Female |
| Male |
| Male |

### Best For

Categorical variables.

---

# 4. `Constant Value Imputation`

## Definition

Replace missing values with a fixed value.

### Examples

- 0
- Unknown
- Not Available
- Missing

### Example

| City |
|------|
| Delhi |
| NaN |
| Mumbai |

After:

| City |
|------|
| Delhi |
| Unknown |
| Mumbai |

---

# 5. `Forward Fill (FFill)`

## Definition

Replace missing values with the **previous valid value**.

### Example

| Value |
|--------|
| 10 |
| NaN |
| NaN |
| 15 |

After:

| Value |
|--------|
| 10 |
| 10 |
| 10 |
| 15 |

### Best For

- Time-series data
- Sensor readings

---

# 6. `Backward Fill (BFill)`

## Definition

Replace missing values with the **next valid value**.

### Example

| Value |
|--------|
| 10 |
| NaN |
| NaN |
| 15 |

After:

| Value |
|--------|
| 10 |
| 15 |
| 15 |
| 15 |

---

# 7. `KNN Imputer`

## Definition

KNN (K-Nearest Neighbors) Imputer fills missing values by looking at similar data points.

### How It Works

1. Find the K nearest neighbors.
2. Calculate the average (or weighted average).
3. Replace the missing value.

### Example

If similar customers have ages:

24, 25, 26

Missing age = **25**

### Best For

- Medium-sized datasets
- Correlated features

---

# 8. `Iterative Imputer (MICE)`

## Definition

Iterative Imputer predicts missing values using other features in the dataset.

Each feature is modeled as a function of the remaining features.

### Example

Missing Salary can be predicted using:

- Age
- Education
- Experience

### Advantage

More accurate than simple imputation.

---

# 9. `Regression Imputation`

## Definition

Uses a regression model to predict missing values.

### Example

Predict house price using:

- Area
- Bedrooms
- Location

Missing house price is estimated using the regression model.

---

# 10. `Random Forest Imputation`

## Definition

Uses a Random Forest model to estimate missing values.

### Advantages

- Handles nonlinear relationships.
- High prediction accuracy.
- Works well with complex datasets.

---

# 11. `Multiple Imputation`

## Definition

Creates multiple versions of the dataset by filling missing values in different ways.

The results are then combined to reduce uncertainty.

### Best For

- Statistical analysis
- Medical research
- Scientific studies

---

# Comparison of Imputation Techniques

| Technique | Data Type | Handles Outliers | Complexity |
|------------|-----------|------------------|------------|
| Mean | Numerical | ❌ No | Easy |
| Median | Numerical | ✅ Yes | Easy |
| Mode | Categorical | N/A | Easy |
| Constant | Any | N/A | Easy |
| Forward Fill | Sequential | N/A | Easy |
| Backward Fill | Sequential | N/A | Easy |
| KNN | Numerical | ✅ Moderate | Medium |
| Iterative | Mixed | ✅ Yes | High |
| Regression | Numerical | Moderate | High |
| Random Forest | Mixed | ✅ Yes | High |
| Multiple Imputation | Mixed | ✅ Yes | Very High |

---

# Choosing the Right Technique

| Situation | Recommended Technique |
|------------|----------------------|
| Normal numerical data | Mean |
| Numerical data with outliers | Median |
| Categorical data | Mode |
| Time-series data | Forward Fill |
| Leading missing values | Backward Fill |
| Correlated features | KNN Imputer |
| High accuracy required | Iterative Imputer |
| Complex relationships | Random Forest Imputer |

# 2. `Interview Questions & Answers`

## Beginner Questions

### Q1. What is imputation?

**Answer:**
Imputation is the process of replacing missing values with estimated or meaningful values to create a complete dataset.

---

### Q2. Why do we use imputation?

**Answer:**
To handle missing data, improve model performance, and avoid losing valuable information by deleting rows or columns.

---

### Q3. Which imputation technique is best for categorical data?

**Answer:**
**Mode (Most Frequent) Imputation.**

---

### Q4. When should you use median instead of mean?

**Answer:**
Use the **median** when the data contains outliers or has a skewed distribution.

---

## Intermediate Questions

### Q5. What is the difference between KNN Imputer and SimpleImputer?

**Answer:**

* **SimpleImputer** replaces missing values using statistics like mean, median, or mode.
* **KNN Imputer** uses similar records (nearest neighbors) to estimate missing values.

---

### Q6. What is Iterative Imputer?

**Answer:**
It predicts missing values using machine learning models built from the other features in the dataset, making it more accurate than simple methods.

---

### Q7. Which imputation technique is commonly used for time-series data?

**Answer:**
**Forward Fill (`ffill`)** and **Backward Fill (`bfill`)**, depending on the context.

---

## Scenario-Based Questions

### Q8. A salary dataset contains extreme outliers. Which imputation technique would you choose?

**Answer:**
**Median Imputation**, because it is robust to outliers.

---

### Q9. Your dataset contains customer gender with some missing values. Which method is appropriate?

**Answer:**
**Mode Imputation**, as gender is a categorical variable.

---

### Q10. You need high-quality imputation for a healthcare dataset with correlated features. Which technique would you use?

**Answer:**
**Iterative Imputer** or **KNN Imputer**, as they leverage relationships between features for more accurate estimates.
