# Model Building Lifecycle and Data Cleaning

## Introduction

Machine Learning is not just about training a model. Before building a model, we must understand the business problem, collect data, explore it, clean it, and then train the model.

A common rule in AI and ML is:

> **Garbage In → Garbage Out (GIGO)**

If the input data is poor quality, the model's predictions will also be poor.

# Validity vs Accuracy

## Validity

Checks whether data follows predefined rules.

### Example

Age = -5

This is invalid because age cannot be negative.

---

## Accuracy

Checks whether data reflects the real-world value correctly.

### Example

Actual Age = 25

Stored Age = 45

Valid?

* Yes (positive value)

Accurate?

* No (incorrect age)

---

### Interview Point

| Validity           | Accuracy                            |
| ------------------ | ----------------------------------- |
| Follows rules      | Reflects reality                    |
| Age = -5 → Invalid | Age = 45 instead of 25 → Inaccurate |

---

# Stemming and Lemmatization

## Stemming

Reduces a word to its root form by removing suffixes.

### Examples

| Word      | Stem |
| --------- | ---- |
| Playing   | Play |
| Played    | Play |
| Playingly | Play |

May produce non-dictionary words.

Example:

```text
Studies → Studi
```

---

## Lemmatization

Converts a word into its meaningful dictionary form.

### Examples

| Word    | Lemma |
| ------- | ----- |
| Running | Run   |
| Better  | Good  |
| Studies | Study |

Produces meaningful words.

---

## Stemming vs Lemmatization

| Stemming                 | Lemmatization        |
| ------------------------ | -------------------- |
| Faster                   | Slower               |
| Rule-based               | Context-based        |
| May create invalid words | Produces valid words |
| Less accurate            | More accurate        |
