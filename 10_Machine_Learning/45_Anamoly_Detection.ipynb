# `Anomaly Detection`
---
## `What is Anomaly Detection?`

**Anomaly Detection** is a Machine Learning technique used to identify **rare, unusual, or abnormal data points** that are significantly different from the majority of the data.

These unusual data points are called:

- Anomalies
- Outliers
- Exceptions
- Abnormal observations

> **Anomaly Detection = Finding data points that do not behave like the rest of the data.**

---

# `Why Do We Need Anomaly Detection?`

In real-world datasets, not every data point follows the normal pattern.

Some points may indicate:

- Fraud
- Cyber attacks
- Machine failures
- Medical emergencies
- Network intrusions

Finding these abnormal points early helps prevent major losses.

---

# `Normal Data vs Anomaly`

Example:

Suppose the daily transaction amount of a customer is:

```code
₹500
₹700
₹650
₹550
₹600
₹45,000  ← Anomaly
₹750
```

Here,

₹45,000 is very different from the normal spending pattern.

This is an **anomaly**.

---

# `Simple Definition`

An anomaly is a data point that:

- Is significantly different from other observations.
- Does not follow the normal pattern.
- Occurs very rarely.

# `Types of Anomalies`

## `1. Point Anomaly`

A single data point is abnormal.

Example:

```
10
12
11
13
120
14
```

120 is an anomaly.

---

## `2. Contextual Anomaly`

A value is abnormal only in a specific context.

Example:

Temperature

```code
40°C

↓

Summer

Normal

-------------

40°C

↓

Winter

Anomaly
```

---

## `3. Collective Anomaly`

A group of data points together forms an anomaly.

Example:

One failed login is normal.

1000 failed logins in one minute may indicate a cyber attack.

---

# `How Does Anomaly Detection Work?`

## Step 1

Collect data.

---

## Step 2

Learn what **normal behavior** looks like.

---

## Step 3

Compare new observations with the normal pattern.

---

## Step 4

If a point is significantly different,

mark it as an anomaly.

---

# `Methods for Anomaly Detection`

## `1. Statistical Methods`

Uses statistical measures such as:

- Mean
- Standard Deviation
- Z-Score

Useful for normally distributed data.

---

## `2. Distance-Based Methods`

Measures the distance between data points.

Points far away from others are treated as anomalies.

Example:

- K-Nearest Neighbors (KNN)

---

## `3. Density-Based Methods`

Detects areas with low data density.

Example:

**DBSCAN**

- Dense regions → Normal
- Sparse regions → Anomalies

---

## `4. Machine Learning Methods`

Uses algorithms to learn normal behavior.

Examples:

- Isolation Forest ⭐
- One-Class SVM
- Local Outlier Factor (LOF)

---

# `Popular Algorithms`

## 1. Isolation Forest ⭐

### Idea

Instead of grouping normal points,

it tries to **isolate anomalies**.

Anomalies are easier to isolate because they are few and different.

### Advantages

- Fast
- Handles large datasets
- Widely used in industry

---

## 2. One-Class SVM

Learns the boundary around normal data.

Anything outside the boundary is treated as an anomaly.

Best for:

- Small datasets

---

## 3. Local Outlier Factor (LOF)

Compares the local density of a point with the density of its neighbors.

If the density is much lower,

the point is an anomaly.

---

## 4. DBSCAN

Points not belonging to any cluster are labeled as:

```
-1
```

These are considered anomalies.

---

# `Anomaly Detection Workflow`

```code
Collect Data
      ↓
Data Preprocessing
      ↓
Learn Normal Pattern
      ↓
Choose Algorithm
      ↓
Predict New Data
      ↓
Normal or Anomaly
```

# `Anomaly Detection vs Classification`

| Classification | Anomaly Detection |
|---------------|-------------------|
| Predicts predefined classes | Detects unusual observations |
| Requires labeled data | Often works with unlabeled data |
| Multiple classes | Normal vs Abnormal |
| Example: Spam Detection | Example: Fraud Detection |
