# `Feature Engineering and Feature Selection`

```python
# Import the core libraries used for data handling, visualization, and modeling.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
# Load the international education cost dataset into a pandas DataFrame.
df = pd.read_csv('../datasets/International_Education_Costs.csv')
```

## `Feature Engineering`

Feature engineering creates or transforms columns so the model has more useful inputs for learning patterns.

- Feature engineering uses domain knowledge and data transformations to create more informative features from raw columns.
- It can improve model accuracy, reduce noise, and make relationships easier for algorithms to learn.
- In this notebook, `Total Cost` is created as a simple combined expense feature.

```python
# Create the target expense column by combining tuition and rent.
df['Total Cost'] = df['Tuition_USD'] + df['Rent_USD']
```

```python
# Display the full DataFrame after creating the Total Cost column.
df
```

```python
# Check the available columns after feature engineering.
df.columns
```

## `Feature Selection`

- Feature selection is the process of choosing the most relevant input columns for model training.
- It removes unnecessary, duplicate, or noisy columns so the model can train faster and focus on useful patterns.
- Here, we keep location, university, program, duration, living-cost, fee, insurance, and total-cost columns.

```python
# Keep only the columns needed for the predictive model.
df = df[['Country', 'City', 'University', 'Program', 'Level', 'Duration_Years', 'Living_Cost_Index', 'Visa_Fee_USD',
       'Insurance_USD','Total Cost']]
```

```python
# Preview the selected features before encoding categorical columns.
df.head()
```

## `Label Encoding or Feature Encoding`

- Label encoding converts categorical text values into numeric labels so machine learning algorithms can process them.
- Each unique category receives an integer value, usually starting from 0.
- For example, countries such as `India`, `USA`, and `UK` are converted into numeric codes.
- Label encoding is simple and useful for quick demos, but for many real projects, one-hot encoding may be safer for non-ordered categories.

```python
# Review the data before applying label encoding.
df.head()
```

```python
# Import LabelEncoder to convert categorical text columns into numeric labels.
from sklearn.preprocessing import LabelEncoder
```

```python
# Create one encoder object and reuse it column by column.
le = LabelEncoder()
```

```python
# Encode country names as numeric labels.
df['Country']  = le.fit_transform(df['Country'])
```

```python
# Confirm that the Country column has been converted into numeric labels.
df.head()
```

```python
# Encode the remaining categorical columns as numeric labels.
df['City'] = le.fit_transform(df['City'])
```

```python
# Encode university names as numeric labels.
df['University'] = le.fit_transform(df['University'])
```

```python
# Encode program names as numeric labels.
df['Program'] = le.fit_transform(df['Program'])
```

```python
# Encode education levels as numeric labels.
df['Level'] = le.fit_transform(df['Level'])
```

```python
# Preview the fully encoded dataset.
df.head()
```

## `Separate Input and Output Data`

- Input data, often written as `X`, contains the features used to make predictions.
- Output data, often written as `y`, contains the target value the model learns to predict.
- In this notebook, the target is `Total Cost`.

```python
# Separate independent variables/features from the target variable.
x = df.iloc[:,:9]
y = df['Total Cost']
```

```python
# Display the input feature matrix.
x
```

```python
# Display the target/output values.
y
```

## `Feature Scaling`

- Feature scaling standardizes or normalizes numeric input columns so they share a comparable scale.
- Scaling prevents large-value columns from dominating distance-based or gradient-based models only because of their magnitude.
- Example: a living cost index and a visa fee may have very different numeric ranges.

Common techniques:
- Standardization, also called Z-score scaling
- Normalization, also called Min-Max scaling
- Robust scaling, which is less affected by outliers

```python
# StandardScaler converts each feature to a mean near 0 and standard deviation near 1.
from sklearn.preprocessing import StandardScaler

s = StandardScaler()
```

```python
# Fit the scaler on the feature data and transform it into scaled numeric values.
X = s.fit_transform(x)
```

```python
# Store the target column separately using the conventional uppercase name used below.
Y = df['Total Cost']
```

```python
# Display the scaled feature array that will be used for model training.
X
```

## `Train-Test Split`

- A train-test split divides the dataset into two parts: training data and testing data.
- The model learns patterns from the training set.
- The testing set is kept separate so we can evaluate how well the model performs on unseen data.
- A common split is 70-80% for training and 20-30% for testing.
- `random_state` keeps the split reproducible so results stay consistent across runs.

```python
# Import train_test_split to divide features and target into training and testing sets.
from sklearn.model_selection import train_test_split
```

```python
# Use 80% of the data for training and 20% for testing.
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size= 0.2, random_state = 42 )
```

```python
# Check how many rows and columns are available in the testing feature set.
X_test.shape
```

```python
# Check the number of target values in the training set.
Y_train.shape
```

```python
# Check the number of target values in the testing set.
Y_test.shape
```

Once the model learns from historical data, it can make predictions on new, unseen data.

## `1. Linear Regression`

Linear Regression establishes a linear relationship between input features and the target variable.

### Formula

\[
Y = mX + c
\]

Where:
- **Y** = Predicted value
- **X** = Input feature
- **m** = Slope (Coefficient)
- **c** = Intercept

### Types

### a) Simple Linear Regression
Uses only one independent variable.

**Example:**

Predicting salary based on years of experience.

### b) Multiple Linear Regression
Uses multiple independent variables.

**Example:**

Predicting total cost using:
- Quantity
- Unit Price
- Shipping Cost
- Discount

### Advantages
- Easy to understand
- Fast training
- Highly interpretable

### Disadvantages
- Assumes linear relationships
- Sensitive to outliers


```python
# Create a Linear Regression model.
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
# Train the Linear Regression model on the training data.
lr.fit(X_train, Y_train)
```

```python
# Predict total cost for the test data.
y_predict = lr.predict(X_test)
```

```python
# Import Mean Absolute Error to measure the average prediction error in dollars.
from sklearn.metrics import mean_absolute_error
```

```python
# Calculate the Linear Regression model's Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

```python
# Import Mean Squared Error to penalize larger prediction errors more strongly.
from sklearn.metrics import mean_squared_error
```

```python
# Calculate the Linear Regression model's Mean Squared Error.
mean_squared_error(Y_test, y_predict)
```

## `2. Ridge Regression ( L2 Regularization )`

Ridge Regression is a regularized version of Linear Regression. It adds a penalty to reduce overfitting and keep model coefficients more stable.

Adds a penalty based on the square of coefficients.

#### Formula

\[
Loss = RSS + \lambda \sum \beta^2
\]

### Advantages
- Reduces overfitting
- Handles multicollinearity

```python
# Create a Ridge Regression model with L2 regularization strength alpha=1.0.
from sklearn.linear_model import Ridge
r = Ridge( alpha = 1.0 )
```

```python
# Train the Ridge Regression model.
r.fit(X_train, Y_train)
```

```python
# Predict total cost using the Ridge model.
y_predict = r.predict(X_test)
```

```python
# Evaluate the Ridge model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

## `3. Lasso Regression`

Lasso Regression is another regularized linear model. It can shrink less useful feature coefficients toward zero, which may help with feature selection.

Adds a penalty based on the absolute value of coefficients.

#### Formula

\[
Loss = RSS + \lambda \sum |\beta|
\]

### Advantages
- Performs feature selection
- Removes irrelevant features

```python
# Create a Lasso Regression model with L1 regularization.
from sklearn.linear_model import Lasso
l = Lasso(alpha=0.01)
```

```python
# Train the Lasso Regression model.
l.fit(X_train, Y_train)
```

```python
# Predict total cost using the Lasso model.
y_predict = l.predict(X_test)
```

```python
# Evaluate the Lasso model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

## `4. Elastic Net Regression`

ElasticNet combines Ridge and Lasso penalties, balancing coefficient stability with feature selection behavior.

Combination of:
- Ridge Regression

### Advantages

```python
# Train the ElasticNet model.
e.fit(X_train, Y_train)
```

```python
# Predict total cost using the ElasticNet model.
y_predict = e.predict(X_test)
```

```python
# Evaluate the ElasticNet model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

Random Forest is an ensemble method that trains many decision trees and combines their predictions to improve stability and accuracy.

Collection of multiple Decision Trees.

### Working
1. Create many trees
2. Train each tree on random samples
3. Average all predictions

### Advantages
- High accuracy
- Less overfitting

### Disadvantages
- Slower than a single tree

```python
# Create a Random Forest Regressor with 100 decision trees.
from sklearn.ensemble import RandomForestRegressor
forest = RandomForestRegressor(n_estimators = 100)
```

## `10. K-Nearest Neighbors Regression`

K-Nearest Neighbors Regression predicts a value by looking at the nearest data points and averaging their target values.

Predicts values based on the nearest data points.

### Working
1. Find K nearest neighbors
2. Calculate average target value
3. Use that average as prediction

### Example

If nearest neighbors have costs:500 ,550, 600

Prediction: (500 + 550 + 600) / 3 = 550

### Advantages
- Simple algorithm
- No training phase

### Disadvantages
- Slow on large datasets
- Sensitive to feature scaling

```python
# Create a K-Nearest Neighbors Regressor with 5 neighbors.
from sklearn.neighbors import KNeighborsRegressor

KNN = KNeighborsRegressor(n_neighbors=5)
```

```python
# Train the KNN model.
KNN.fit(X_train, Y_train)
```

```python
# Predict total cost using the KNN model.
y_predict = KNN.predict(X_test)
```

```python
# Evaluate the KNN model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

## `Support Vector Regression`

Support Vector Regression tries to fit a boundary that predicts numeric values while allowing a small margin of error.

SVR tries to fit a line within a specified margin of error.

Unlike Linear Regression, SVR focuses on minimizing prediction errors within a tolerance boundary.

### Advantages
- Effective for complex datasets
- Works well with non-linear data

### Disadvantages
- Computationally expensive
- Requires feature scaling

```python
# Create a Support Vector Regressor using the RBF kernel for non-linear patterns.
from sklearn.svm import SVR

svr = SVR(kernel='rbf')
```

```python
# Train the Support Vector Regressor.
svr.fit(X_train, Y_train)
```

```python
# Predict total cost using the SVR model.
y_predict = svr.predict(X_test)
```

```python
# Evaluate the SVR model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

## `Bagging Regressor`

Bagging, short for bootstrap aggregating, trains multiple models on different random samples of the training data and averages their predictions. It often reduces variance and makes predictions more stable.

```python
# Create a Bagging Regressor with 100 base estimators.
from sklearn.ensemble import BaggingRegressor

bag  = BaggingRegressor(n_estimators= 100)
```

```python
# Train the Bagging Regressor.
bag.fit(X_train, Y_train)
```

```python
# Predict total cost using the Bagging model.
y_predict = bag.predict(X_test)
```

```python
# Evaluate the Bagging model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

## `Stacking Regressor`

Stacking trains multiple base models first, then uses a final estimator to learn how to combine their predictions. It is useful when different models capture different patterns in the same dataset.

```python
# Import the ensemble and tree models needed for stacking.
from sklearn.ensemble import RandomForestRegressor, StackingRegressor, BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
```

```python
# Define the base models that will provide predictions to the final model.
estimators = [('rf', RandomForestRegressor()), ('br', BaggingRegressor())]
```

```python
# Create a Stacking Regressor with a Decision Tree as the final estimator.
stack = StackingRegressor(estimators=estimators, final_estimator= DecisionTreeRegressor())
```

```python
# Train the Stacking Regressor.
stack.fit(X_train, Y_train)
```

```python
# Predict total cost using the Stacking model.
y_predict = stack.predict(X_test)
```

```python
# Evaluate the Stacking model using Mean Absolute Error.
mean_absolute_error(Y_test, y_predict)
```

## `Classification Algorithms`

Classification is a supervised machine learning task where the target/output is a category or class label instead of a continuous number. Examples include predicting whether a student is depressed or not, whether an email is spam or not, or whether a customer will churn.

Common classification algorithms:

1. Logistic Regression: predicts the probability of a class.
2. Decision Tree Classifier: splits data using decision rules.
3. Random Forest Classifier: combines many decision trees for stronger predictions.
4. XGBoost Classifier: uses boosted trees for high-performance classification.
5. AdaBoost Classifier: improves weak classifiers step by step.
6. CatBoost Classifier: works well on tabular and categorical data.
7. K-Neighbors Classifier: predicts based on nearby data points.
8. Support Vector Classifier: separates classes using a decision boundary.
9. Bernoulli Naive Bayes: works well with binary features.
10. Multinomial Naive Bayes: commonly used for count-based text data.
11. Bagging Classifier: averages multiple models trained on random samples.
12. Stacking Classifier: combines multiple models using a final meta-model.

Evaluation metrics:

1. Accuracy Score: percentage of correct predictions.
2. Precision: how many predicted positives are actually positive.
3. Recall: how many actual positives were correctly found.
4. F1-Score: balance between precision and recall.
5. Confusion Matrix: table showing correct and incorrect class predictions.
