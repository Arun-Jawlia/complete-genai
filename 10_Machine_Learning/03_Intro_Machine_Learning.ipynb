# `Machine Learning (ML)`
---

## Definition

Machine Learning (ML) is a field of Computer Science and a subset of Artificial Intelligence (AI) that uses statistical techniques to give computer systems the ability to learn from data without being explicitly programmed.

Instead of manually writing all rules, we provide data to the machine and allow it to discover patterns automatically.

---

# `Traditional Programming vs Machine Learning`

## Traditional Programming

In traditional programming, rules are written explicitly by the programmer.

### Workflow

```text
Input + Program (Rules)
            ↓
         Output
```

### Example

```python
x = int(input())
print(x * x)
```

The programmer explicitly tells the computer:

```text
Output = x²
```

---

## `Machine Learning`

In Machine Learning, we do not explicitly tell the machine the rule.

Instead, we provide examples.

### Workflow

```text
Input Data + Output Data
            ↓
     ML Algorithm
            ↓
      Learn Pattern
            ↓
         ML Model
```

The machine automatically discovers the relationship between input and output.

---

# `Two Important Components of Machine Learning`

Machine Learning requires two things:

## 1. `Data`

Data contains examples from which the machine learns.

### Example Dataset

| Input (x) | Output (y) |
|------------|------------|
| 1 | 1 |
| 2 | 4 |
| 3 | 9 |
| 4 | 16 |
| 5 | 25 |

The machine observes:

```text
1 → 1
2 → 4
3 → 9
4 → 16
5 → 25
```

---

## 2. `Algorithm`

An algorithm is the learning procedure that discovers patterns from data.

The algorithm analyzes:

```text
Input
   ↓
Relationship
   ↓
Output
```

and tries to answer:

```text
How is Output related to Input?
```

---

# `Pattern Between Input and Output`

The most important goal of Machine Learning is to find patterns.

Consider:

```text
Input → Output

1 → 1
2 → 4
3 → 9
4 → 16
5 → 25
```

The algorithm observes the examples and discovers:

```text
Output = Input × Input
```

or

```text
Output = x²
```

This discovered relationship is called a **Pattern**.

---

# `Role of Machine Learning Algorithm`

The Machine Learning algorithm performs the following tasks:

```text
Training Data
        ↓
Analyze Data
        ↓
Find Hidden Pattern
        ↓
Create Model
```

The model stores the learned knowledge.

---

# Training Phase

Initially, the algorithm is provided with data.

### Training Data

```text
1 → 1
2 → 4
3 → 9
4 → 16
5 → 25
```

The algorithm learns:

```text
Output = x²
```

This learning process is called **Training**.

---

# Testing Phase

After training, we provide new data that the algorithm has never seen before.

### New Input

```text
6
```

The model predicts:

```text
36
```

### New Input

```text
7
```

The model predicts:

```text
49
```

If predictions are accurate, we know that the algorithm has successfully learned the pattern.

---

# Complete Machine Learning Workflow

```text
Data
 ↓
ML Algorithm
 ↓
Pattern Learning
 ↓
Model Creation
 ↓
New Input
 ↓
Prediction
 ↓
Output
```

---

# Why Are Both Data and Algorithm Required?

Machine Learning cannot work without either component.

## Case 1: Data Without Algorithm

```text
Data Available
No Learning Procedure
```

Result:

```text
No Pattern Discovery
```

The machine cannot learn.

---

## Case 2: Algorithm Without Data

```text
Learning Procedure Available
No Examples Available
```

Result:

```text
Nothing To Learn
```

The machine cannot learn.

---

## Therefore

Machine Learning requires:

```text
Data + Algorithm
```

Both are equally important.

---

# `Why Do We Need Machine Learning?`

Many real-world problems are too complex to solve using traditional programming.

Writing rules manually becomes impossible.

---

## Problem 1: Spam Email Detection

Traditional Programming:

```text
Write thousands of rules
for every spam email pattern.
```

Very difficult.

Machine Learning:

```text
Provide thousands of:

Spam Emails
Not Spam Emails
```

The machine learns patterns automatically.

---

## Problem 2: Face Recognition

Can we manually write rules for:

- Every face shape
- Every eye shape
- Every hairstyle
- Every lighting condition

No.

Machine Learning learns these patterns automatically from images.

---

## Problem 3: Language Translation

Traditional programming cannot easily handle:

```text
English → Hindi
English → French
English → Spanish
```

because language contains countless rules and exceptions.

Machine Learning learns translation patterns from large datasets.

---

## Problem 4: Recommendation Systems

Examples:

- Netflix Recommendations
- Amazon Recommendations
- YouTube Recommendations

Writing explicit rules for every user is impossible.

Machine Learning learns user preferences automatically.

---

No need to manually write all rules.

---

## 2. Improves With More Data

More data generally leads to better performance.

---

## 3. Handles Complex Problems

Can solve problems where rule-writing is difficult.

# Key Points

- Machine Learning is a subset of Artificial Intelligence.
- Machine Learning uses statistical techniques to learn from data.
- Machine Learning does not require explicit programming of every rule.
- Data and Algorithms are the two fundamental components of Machine Learning.
- The ML algorithm discovers patterns between inputs and outputs.
- Training is the process of learning patterns from data.
- Testing checks whether the learned pattern works on unseen data.
- Machine Learning is required for complex problems where manual rule-writing is impractical or impossible.
