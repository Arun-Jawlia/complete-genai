# `# Hierarchy of Artificial Intelligence`
---

Artificial Intelligence is a broad field that contains several subfields.

The relationship between these technologies can be represented as:

```text
Artificial Intelligence (AI)
        │
        ▼
Machine Learning (ML)
        │
        ▼
Deep Learning (DL)
        │
        ├── Natural Language Processing (NLP)
        │
        └── Generative AI (GenAI)
```

Or:

```text
AI
 └── ML
      └── DL
           ├── NLP
           └── GenAI
```

---

# 1. Artificial Intelligence (AI)

## Definition

Artificial Intelligence (AI) is a field of Computer Science that aims to create machines capable of performing tasks that normally require human intelligence.

AI systems can:

- Learn
- Reason
- Solve Problems
- Make Decisions
- Understand Language
- Recognize Images
- Adapt to New Situations

### Examples

- Self-driving cars
- Virtual assistants
- Expert systems
- Recommendation systems
- Robotics

---

# 2. Machine Learning (ML)

## Definition

Machine Learning is a subset of Artificial Intelligence that allows machines to learn patterns from data without being explicitly programmed.

Used for General Purpose like Image Classification

No Rule based learnign will be there

Instead of writing rules manually:

```python
if condition:
    do_something()
```

we provide data and allow the machine to discover patterns automatically.

### Workflow

```text
Training Data
      ↓
Machine Learning Algorithm  --> Statiscal Algo --> Mathematical Operation 
      ↓
Pattern Learning
      ↓
Model
      ↓
Prediction --> Output
```

### Examples

- House price prediction
- Spam email detection
- Product recommendation
- Fraud detection

---

# Why ML is a Subset of AI

```text
AI Goal:
Create intelligent machines.

ML Solution:
Teach machines using data.
```

Machine Learning is one of the most important techniques used to achieve Artificial Intelligence.

---

# 3. Deep Learning (DL)

## Definition

Deep Learning is a subset of Machine Learning that uses Artificial Neural Networks with multiple layers to learn complex patterns from large amounts of data.

Deep Learning is inspired by the structure of the human brain.

### Structure

```text
Input Layer
      ↓
Hidden Layer
      ↓
Hidden Layer
      ↓
Hidden Layer
      ↓
Output Layer
```

The multiple hidden layers make it "Deep" Learning.

---

## Why Deep Learning?

Traditional Machine Learning often requires manual feature extraction.

Deep Learning automatically learns features from raw data.

### Example

Image Recognition:

Machine Learning:

```text
Human extracts features
      ↓
Model learns
```

Deep Learning:

```text
Image
  ↓
Neural Network
  ↓
Features learned automatically
  ↓
Prediction
```

---

# Applications of Deep Learning

- Image Recognition
- Speech Recognition
- Face Recognition
- Language Translation
- Chatbots
- Autonomous Vehicles
- Medical Diagnosis

# 4. `Natural Language Processing (NLP)`

## Definition

Natural Language Processing (NLP) is a field that enables computers to understand, process, analyze, and generate human language.

Tools which can convert the text into numeric

NLP combines:

- Artificial Intelligence
- Machine Learning
- Deep Learning
- Linguistics

---

## Goal of NLP

Allow computers to communicate with humans using natural language.

### Human Language Examples

```text
English
Hindi
French
Spanish
Tamil
Punjabi
```

NLP helps machines understand these languages.

---

## NLP Applications

- Chatbots
- Language Translation
- Speech Recognition
- Sentiment Analysis
- Text Classification
- Question Answering
- Voice Assistants

### Examples

- Siri
- Alexa

- ChatGPT

# 5. Generative AI (GenAI)

## Definition

Generative AI is a subset of Deep Learning that can generate new content such as:

- Text
- Images
- Audio
- Video
- Code

Instead of only making predictions, GenAI creates new content.

---

## Traditional AI vs Generative AI

### Traditional AI

```text
Input
   ↓
Prediction
```

Example:

```text
Email
   ↓
Spam or Not Spam
```

---

### Generative AI

```text
Prompt
   ↓
Generate New Content
```

Example:

```text
Write a poem
        ↓
AI generates a poem
```

# `Difference Between Machine Learning (ML) and Natural Language Processing (NLP)`

| Machine Learning (ML) | Natural Language Processing (NLP) |
|----------------------|-----------------------------------|
| ML is a subset of AI. | NLP is a subfield of AI focused on human language. |
| Works mainly with numerical data. | Works mainly with text and language data. |
| Input is usually numbers/features. | Input is usually text or speech. |
| Output can be prediction, classification, or regression. | Output can be translation, summarization, sentiment analysis, etc. |
| Goal is to learn patterns from data. | Goal is to understand and process human language. |
| Examples: House price prediction, spam detection. | Examples: ChatGPT, Google Translate, Siri. |

---

# Why Do We Need NLP?

Machine Learning algorithms cannot directly understand text.

### ML Requirement

```text
Input → Numbers
Output → Numbers
```

Example:

```text
Age → Salary
Marks → Grade
Area → House Price
```

ML works naturally with numerical data.

---

# Problem with Text

Consider:

```text
"I love Machine Learning"
```

A machine cannot directly understand this text.

Machines understand:

```text
0, 1, 2, 3, 4...
```

not words.

---

# Role of NLP

NLP converts human language into numerical form so that Machine Learning and Deep Learning models can process it.

### Workflow

```text
Text
 ↓
NLP Processing
 ↓
Numbers (Vectors)
 ↓
ML / DL Model
 ↓
Prediction
```

---

# Example

### Input Text

```text
"I love AI"
```

### NLP Conversion

```text
[0.45, 0.82, 0.13]
```

### ML Model

```text
Positive Sentiment
```

---

# Audio Processing

For audio data, we first convert sound into numerical features.

One popular technique is:

## MFCC

**MFCC = Mel Frequency Cepstral Coefficients**

### Workflow

```text
Audio
 ↓
MFCC
 ↓
Numerical Features
 ↓
ML / DL Model
```

# `Why Shift from ML to NLP?`

Traditional ML:

```text
Numbers → Prediction
```

But most real-world data is:

- Text
- Speech
- Documents
- Emails
- Social Media Posts

To process this data, we need NLP techniques.

### Therefore

```text
Text
 ↓
NLP
 ↓
Numbers
 ↓
Machine Learning
 ↓
Prediction
```
