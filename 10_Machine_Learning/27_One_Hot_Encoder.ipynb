# `One Hot Encoder`
---

## `What is One-Hot Encoding?`

### Definition

**One-Hot Encoding (OHE)** is a feature encoding technique that converts each category of a categorical variable into a separate binary (0 or 1) column.

In simple words:

> **One category = One new column**

Each row will have:

- **1** → Category is present
- **0** → Category is absent

---

# `Why Do We Need One-Hot Encoding?`

Machine Learning algorithms cannot understand text values like:

| Color |
|--------|
| Red |
| Blue |
| Green |

They only understand numbers.

One-Hot Encoding converts these categories into binary columns without introducing any order.

---

# Example

## Before Encoding

| Color |
|--------|
| Red |
| Blue |
| Green |
| Blue |

---

## After One-Hot Encoding

| Color_Red | Color_Blue | Color_Green |
|------------|------------|-------------|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |
| 0 | 1 | 0 |

---

# `Why is it called "One-Hot"?`

Because **only one column contains 1**, while all other columns contain 0.

Example

Red

```
Red     = 1
Blue    = 0
Green   = 0
```

Only **one** value is "hot" (1).

---

# `When Should We Use One-Hot Encoding?`

Use One-Hot Encoding when the feature is **Nominal**.

### Nominal Data

Categories have **no order**.

Examples

- Gender
- City
- Job
- Department
- Country
- Color

---

# `When Should We NOT Use One-Hot Encoding?`

Do **not** use it for **Ordinal Data**.

Example

Education

- Primary
- Secondary
- Higher Secondary
- Graduation

These values have an order.

Use **Ordinal Encoding** instead.

# `One-Hot Encoding in Python`

## Method 1: Using Pandas

```python
import pandas as pd

df = pd.get_dummies(df, columns=["City"])
```

---

## `Method 2: Using Scikit-Learn`

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder()

encoded = encoder.fit_transform(df[["City"]])
```

---

# Drop First Column

Sometimes we use

```python
pd.get_dummies(df, drop_first=True)
```

## Why?

To avoid the **Dummy Variable Trap** (multicollinearity), where one encoded column can be perfectly predicted from the others.

### Example

Original

| Gender |
|---------|
| Male |
| Female |

Without `drop_first`

| Male | Female |
|------|---------|
| 1 | 0 |
| 0 | 1 |

With `drop_first=True`

| Male |
|------|
| 1 |
| 0 |

The missing category (**Female**) is implied when `Male = 0`.

---

# Advantages

- Simple to understand.
- No artificial ordering of categories.
- Works well with most Machine Learning algorithms.
- Suitable for nominal categorical data.

---

# Disadvantages

### High Cardinality Problem

If a feature has many unique categories, One-Hot Encoding creates many new columns.

### Example

City

```
1000 unique cities

↓

1000 new columns
```

This increases memory usage and model complexity.

---

---

# One-Hot Encoding vs Label Encoding

| One-Hot Encoding | Label Encoding |
|------------------|----------------|
| Creates new columns | Keeps one column |
| Best for Nominal Data | Best for Ordinal/Binary Data |
| No ordering introduced | Numbers may imply an order |
| Higher memory usage | Lower memory usage |

---

# When to Use Which?

| Data | Encoding |
|------|----------|
| Gender | One-Hot Encoding |
| City | One-Hot Encoding |
| Job | One-Hot Encoding |
| Department | One-Hot Encoding |
| Country | One-Hot Encoding |
| Education Level | Ordinal Encoding |
| Customer Rating | Ordinal Encoding |

---

# Interview Tips

### Remember

Nominal Data

↓

One-Hot Encoding

Ordinal Data

↓

---

# Memory Trick

**No Order → One-Hot Encoding**

**Has Order → Ordinal Encoding**

---

# Key Takeaways

- One-Hot Encoding converts each category into a separate binary column.
- It is best suited for nominal categorical variables.
- It avoids introducing false numerical relationships.
- `pd.get_dummies()` and `OneHotEncoder` are the two most common implementations in Python.
- Use `drop_first=True` to avoid the Dummy Variable Trap in linear models.
