```python
tokenizer = AutoTokenizer.from_pretrained("google/gemma-3-1b-it")
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embedding = OpenAIEmbeddings(model='text-embedding-3-small')
```

```python
embeddings = embedding.embed_query("Hello how are you")
```

```python
len(embeddings)
```

By default: Dimension is 1536

```python
new_embed = OpenAIEmbeddings(model='text-embedding-3-small', dimensions=100)
```

```python
new_embed.embed_query("Hello how are you")
```
