# Self-Attention in Transformers

## Definition

**Self-Attention** is the core mechanism of the Transformer architecture. It enables every token in a sentence to **look at (attend to) all other tokens**, determine which ones are important, and use that information to build a richer contextual representation.

> **Definition:**
> **Self-Attention** is a mechanism that calculates how much attention each token should pay to every other token in the same input sequence.

It answers the question:

> **"Which other words should I focus on to understand the meaning of this word?"**

---

# Where Does Self-Attention Fit?

The Transformer pipeline looks like this:

```text
Input Text
      ↓
Tokenizer
      ↓
Token IDs
      ↓
Embeddings
      ↓
Positional Encoding
      ↓
Self-Attention
      ↓
Feed Forward Network
      ↓
Output
```

---

# Why Do We Need Self-Attention?

Consider the sentence:

```text
The animal didn't cross the street because it was tired.
```

Question:

> What does **"it"** refer to?

Possible answers:

* Street ❌
* Animal ✅

To answer correctly, the model must understand that **"it"** is related to **"animal"**, even though they are several words apart.

Self-Attention helps the Transformer discover this relationship.

---

# Another Example

Sentence:

```text
Arun loves Deep Learning because he enjoys solving problems.
```

Question:

Who is **"he"**?

Humans instantly know:

```text
he → Arun
```

A Transformer learns this using **Self-Attention**.

# Intuition Behind Self-Attention

Suppose the sentence is:

```text
I love Deep Learning
```

When processing the word:

```text
Learning
```

The Transformer asks:

* Should I pay attention to **I**?
* Should I pay attention to **love**?
* Should I pay attention to **Deep**?

It computes an **attention score** for each word.

Example:

| Word     | Attention Score |
| -------- | --------------: |
| I        |            0.05 |
| love     |            0.20 |
| Deep     |            0.70 |
| Learning |            0.05 |

Since **Deep** has the highest score, the representation of **Learning** is influenced more by **Deep**.

# How Self-Attention Works

Suppose we have:

```text
I love AI
```

---

## Step 1: Token Embeddings

After tokenization and embeddings:

```text
I

↓

Embedding Vector
```

```text
love

↓

Embedding Vector
```

```text
AI

↓

Embedding Vector
```

---

## Step 2: Create Three Vectors

For every token, the model creates **three different vectors**.

These are called:

* Query (Q)
* Key (K)
* Value (V)

Every token has its own Q, K, and V vectors.

---

# What are Query, Key and Value?

## Query (Q)

Represents:

> "What information am I looking for?"

Think of it as asking a question.

---

## Key (K)

Represents:

> "What information do I contain?"

Think of it as an identity card.

---

## Value (V)

Represents:

> "What information should I send if someone pays attention to me?"

This is the actual content that gets shared.

---

# Library Analogy

Imagine a library.

You want a book about AI.

### Query

Your search request.

```text
"I need an AI book."
```

---

### Key

Each book's title and keywords.

```text
Artificial Intelligence
Machine Learning
Data Science
```

---

### Value

The actual content of the selected book.

---

The Query searches all Keys.

The best matching Key returns its Value.

This is exactly how Self-Attention operates.

---

# Step 3: Calculate Attention Scores

Each Query is compared with every Key.

Example:

For the word:

```text
Learning
```

The model compares:

```text
Learning(Q)

↓

I(K)

↓

Love(K)

↓

Deep(K)

↓

Learning(K)
```

This produces similarity scores.

Example:

| Compared With | Score |
| ------------- | ----: |
| I             |   1.2 |
| Love          |   2.8 |
| Deep          |   7.5 |
| Learning      |   3.0 |

Higher score = More relevant.

---

# Step 4: Softmax

The raw scores are converted into probabilities using the **Softmax** function.

Example:

Before Softmax:

```text
[1.2, 2.8, 7.5, 3.0]
```

↓

After Softmax:

```text
[0.04, 0.12, 0.74, 0.10]
```

These probabilities add up to **1**.

---

# Step 5: Weighted Sum

Now each **Value vector** is multiplied by its attention weight.

Example:

| Token    | Weight |
| -------- | ------ |
| I        | 0.04   |
| Love     | 0.12   |
| Deep     | 0.74   |
| Learning | 0.10   |

The weighted Value vectors are added together to create a new representation for the current token.

This new representation contains contextual information from the entire sentence.

---

# Visual Flow

```text
Sentence

↓

Embeddings

↓

Generate

Q
K
V

↓

Q × K

↓

Attention Scores

↓

Softmax

↓

Attention Weights

↓

Weighted Sum of V

↓

Context-Aware Representation
```

---

# Mathematical Formula

The attention calculation is:

```text
Attention(Q,K,V)

=

Softmax

(

QKᵀ
──────
 √dk

)

×

V
```

Where:

* **Q** = Query Matrix
* **K** = Key Matrix
* **V** = Value Matrix
* **dk** = Dimension of Key vectors

### Why divide by √dk?

Without scaling, the dot-product values can become very large for high-dimensional vectors, making the Softmax output extremely peaked and harder to train.

Scaling by √dk keeps the values in a stable range.

> **Interview Tip:** This is why the mechanism is called **Scaled Dot-Product Attention**.

---

# Why Is It Called "Self" Attention?

Because the model attends **within the same sequence**.

Example:

```text
I love AI
```

Every token can attend to:

* I
* love
* AI

Since the attention happens within the input itself, it is called **Self-Attention**.

---

# Benefits of Self-Attention

* Captures long-range dependencies.
* Understands context effectively.
* Processes tokens in parallel.
* Handles long sentences better than RNNs.
* Forms the foundation of modern LLMs.

---

# Self-Attention vs RNN

| RNN                              | Self-Attention                         |
| -------------------------------- | -------------------------------------- |
| Sequential processing            | Parallel processing                    |
| Struggles with long dependencies | Captures long-range relationships well |
| Slower training                  | Faster training                        |
| Information may fade             | Direct interaction between all tokens  |

---

# Limitations

* Computational cost grows with sequence length (quadratic complexity, **O(n²)**).
* Requires more memory for very long inputs.
* Long-context optimization remains an active research area.

# Self-Attention in Transformer Architecture

## What is Self-Attention?

**Self-Attention** is the core mechanism of the Transformer architecture. It allows **every token in a sequence to look at every other token** and decide **how important each token is** when creating its own representation.

Unlike RNNs and LSTMs, which process words one at a time, Self-Attention processes **all words simultaneously**, enabling Transformers to capture long-range dependencies efficiently.

---

# Why Do We Need Self-Attention?

Consider the sentence:

```text
The animal didn't cross the road because it was tired.
```

When processing the word:

```text
it
```

The model must determine:

> What does **"it"** refer to?

Possible candidates:

* animal ✅
* road ❌

Self-Attention allows **"it"** to focus more on **"animal"** than on unrelated words.

---

# Another Example

Sentence:

```text
The cat sat on the mat because it was soft.
```

While processing:

```text
it
```

The model attends to:

```text
mat
```

because "it" refers to the mat.

# Where Self-Attention Fits in a Transformer

```text
Input Text
      │
      ▼
Tokenization
      │
      ▼
Embeddings
      │
      ▼
Positional Encoding
      │
      ▼
Multi-Head Self-Attention
      │
      ▼
Feed Forward Network
      │
      ▼
Output
```

# Intuition

Suppose the sentence is:

When processing the word:

```text
Learning
```

The model doesn't only look at:

```text
Learning
```

It also examines:

```text
I

love

Deep
```

and decides:

* How important is **I**?
* How important is **love**?

# Visual Representation

# Step-by-Step Working

Suppose the sentence is

```text
I love AI
```

---

## Step 1: Tokenization

```text
["I", "love", "AI"]
```

---

## Step 2: Embedding

Each token becomes a vector.

```text
I

↓

[0.2,0.4,0.1]

love

↓

[0.7,0.8,-0.2]

AI

↓

[-0.3,0.9,0.6]
```

---

## Step 3: Generate Query, Key, and Value

Each embedding is multiplied by three different trainable weight matrices.

```
Query (Q)

Key (K)

Value (V)
```

For every token:

```text
Embedding

↓

Query

↓

Key

↓

Value
```

---

# What are Query, Key, and Value?

Think of them like a search engine.

Suppose you're searching a library.

### Query (Q)

"What information am I looking for?"

---

### Key (K)

"What information do I contain?"

---

### Value (V)

"What information should I provide if selected?"

---

# Example

Sentence

```text
The animal was tired because it ran.
```

For the token:

```text
it
```

Query asks:

```text
Who does "it" refer to?
```

Each other word provides:

```
Key → "Here's what I represent."

Value → "Here's my information."
```

The Query is compared with all Keys to decide where attention should go.

---

# Step 4: Compute Attention Scores

Every Query is compared with every Key.

Mathematically:

[
\text{Score}=QK^T
]

Example:

```
Query("AI")

×

Key("I")

↓

0.2
```

```
Query("AI")

×

Key("love")

↓

0.8
```

```
Query("AI")

×

Key("AI")

↓

1.5
```

Higher scores indicate greater relevance.

---

# Step 5: Scale the Scores

Large dot products can make the Softmax distribution overly peaked, leading to unstable gradients.

To stabilize training, divide by:

[
\sqrt{d_k}
]

where:

* (d_k) = dimension of the key vectors.

The scaled scores are:

[
\frac{QK^T}{\sqrt{d_k}}
]

---

# Step 6: Apply Softmax

Softmax converts the scores into probabilities.

Suppose the scores are:

| Word | Score |
| ---- | ----: |
| I    |     1 |
| love |     3 |
| AI   |     6 |

After Softmax:

| Word | Attention Weight |
| ---- | ---------------: |
| I    |             0.01 |
| love |             0.12 |
| AI   |             0.87 |

The weights sum to **1**.

---

# Step 7: Weighted Sum of Values

Each attention weight multiplies its corresponding Value vector.

```
0.01 × Value(I)

+

0.12 × Value(love)

+

0.87 × Value(AI)

↓

New Representation
```

The resulting vector now contains contextual information from the entire sentence.

---

# Complete Attention Formula

[
\boxed{
\text{Attention}(Q,K,V)
=======================

\text{Softmax}
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)
V
}
]

This is the fundamental equation behind Self-Attention.

---

# Workflow

```text
Embedding
      │
      ▼
Generate Q, K, V
      │
      ▼
Q × Kᵀ
      │
      ▼
Scale
      │
      ▼
Softmax
      │
      ▼
Attention Weights
      │
      ▼
Multiply with Values
      │
      ▼
Context Vector
```

---

# Numerical Example

Suppose the attention weights are:

```
Dog

↓

0.70
```

```
Cat

↓

0.20
```

```
Car

↓

0.10
```

The context vector becomes:

```
0.70 × Value(Dog)

+

0.20 × Value(Cat)

+

0.10 × Value(Car)
```

The output emphasizes information from **Dog** because it has the highest attention weight.

# Why Self-Attention is Powerful

Unlike RNNs:

```
Word 1

↓

Word 2

↓

Word 3

↓

Word 4
```

information must flow through intermediate steps.

With Self-Attention:

```
Word 1 ↔ Word 4

Word 2 ↔ Word 3

Word 1 ↔ Word 3

Word 2 ↔ Word 4
```

Every token can interact directly with every other token in a **single layer**.

---

# Advantages

* Captures long-range dependencies effectively.
* Processes all tokens in parallel during training.
* Learns contextual representations dynamically.
* Scales better than recurrent architectures for large models.

---

# Limitations

* Standard self-attention has **quadratic time and memory complexity** with respect to sequence length, making very long sequences expensive.
* Requires significant computational resources for large contexts.
* Variants such as sparse, linear, and local attention have been developed to improve efficiency.

# PyTorch Example

```python
import torch
import torch.nn as nn

embed_dim = 128
num_heads = 8

attention = nn.MultiheadAttention(
    embed_dim=embed_dim,
    num_heads=num_heads,
    batch_first=True
)

x = torch.randn(32, 20, 128)

output, weights = attention(x, x, x)

print(output.shape)
print(weights.shape)
```

Output:

```
torch.Size([32, 20, 128])
torch.Size([32, 20, 20])
```

Explanation:

* **32** → Batch size
* **20** → Sequence length
* **128** → Embedding dimension
* **20 × 20** → Attention weights between every pair of tokens

---

# TensorFlow/Keras Example

```python
import tensorflow as tf
from tensorflow.keras.layers import MultiHeadAttention

attention = MultiHeadAttention(
    num_heads=8,
    key_dim=64
)

x = tf.random.normal((32, 20, 128))

output = attention(
    query=x,
    value=x,
    key=x
)

print(output.shape)
```

Output:

```
(32, 20, 128)
```

# Self-Attention vs RNN
| Feature                 | RNN/LSTM               | Self-Attention                                                      |
| ----------------------- | ---------------------- | ------------------------------------------------------------------- |
| Processing              | Sequential             | Parallel                                                            |
| Long-Range Dependencies | Harder                 | Easy                                                                |
| Training Speed          | Slower                 | Faster (parallelizable)                                             |
| Context                 | Mostly previous tokens | Entire sequence (or previous tokens only when using causal masking) |
| Scalability             | Limited                | Excellent                                                           |

> **Important:** In encoder models like BERT, self-attention can attend to all tokens. In decoder-only models like GPT, **causal (masked) self-attention** prevents a token from attending to future tokens during generation.

# Interview Summary

> **Self-Attention** is the core mechanism of the Transformer architecture. It enables every token to attend to every other relevant token in the sequence by generating **Query (Q), Key (K), and Value (V)** vectors. Attention scores are computed using the dot product of Queries and Keys, scaled by (\sqrt{d_k}), normalized with Softmax, and then used to compute a weighted combination of the Value vectors. This produces context-aware representations for every token. Self-attention allows Transformers to model long-range dependencies efficiently and process sequences in parallel, which is a key reason they outperform RNNs and LSTMs on many language tasks.
