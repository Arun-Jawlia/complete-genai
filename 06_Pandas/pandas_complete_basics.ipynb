# Pandas Basics

## What is Pandas?

It helps us work with data in **tabular format** (rows and columns), just like Excel sheets or databases.

With Pandas, we can:
- Organize data
- Clean messy data
- Analyze large datasets
- Filter and transform information easily

Pandas is one of the most important libraries used in:
- Data Science
- Machine Learning
- Artificial Intelligence
- Data Analytics

# What is a Series?

A Series is a **one-dimensional labeled array** in Pandas.

It is similar to a single column in Excel.


````
import pandas as pd

data = [10, 20, 30, 40]

series = pd.Series(data)

print(series)
````

# What is a DataFrame?

A DataFrame is a **two-dimensional table** in Pandas.

It contains:
- Rows
- Columns

It is similar to an Excel sheet or SQL table.


````
import pandas as pd

students = {
    "Name": ["Arun", "Rahul", "Priya"],
    "Marks": [90, 85, 95],
    "City": ["Delhi", "Mumbai", "Pune"]
}

df = pd.DataFrame(students)

print(df)
````

# Basic Pandas Operations

Pandas provides many useful operations:
- Viewing data
- Selecting columns
- Filtering rows
- Finding statistics


````
import pandas as pd

students = {
    "Name": ["Arun", "Rahul", "Priya"],
    "Marks": [90, 85, 95],
    "City": ["Delhi", "Mumbai", "Pune"]
}

df = pd.DataFrame(students)

# View first rows
print(df.head())

# Select column
print(df["Name"])

# Filter data
print(df[df["Marks"] > 88])

# Statistics
print(df["Marks"].mean())
````
