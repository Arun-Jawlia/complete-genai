# `Series in Pandas`
---

- Pandas Series is a column of data in a Table like in Excel or Google Sheets with labels called as indexes attached to each value.
- A Series is a **one-dimensional labeled array** in Pandas.

```python
# pip install pandas
```

```python
# pip install numpy
```

```python
import pandas as pd
import numpy as np
```

```python
pd.Series(data = [10,20,30]) # if no indexes value given, it will take by default 
```

```python
pd.Series(data = [10,20,30], index=["Bread", "Milk", "Butter"]) # Add Index Values
```

```python
pd.Series(data = [10,20,30], index=["Bread", "Milk", "Butter"], name="Price of Grocery Item") # Add Name
```

## `Series:`

- 1D arraya that can hold any type of data like integers, float, strings (object), boolean or mix datatype

Data Type:
1. Integers: 
2. Float
3. Strings
4. Boolean
5. Dates or Time
6. Mixed Types

Creating a Series in Multiple Ways
1. List
2. NumPy Arrays
3. Dictonary
4. Custom Index
5. Scalar Value

### `Create a Series From List`

```python
country_list = ['India', 'America', "Russia", "Iran", "Pakistan"]
pd.Series(country_list)
```

```python
batsman_run = [48,55,62,100,99,98,65,78,92]
pd.Series(batsman_run, dtype='int32')
```

```python
subject_names = ["Maths", "Chemistry", "Biology", "Physics", 'English']
subject_marks = [92,82,68,60,83]

pd.Series(data=subject_marks, index=subject_names,name = "Subject Marks Data", dtype='int32')
```

## `Creating an Series using NumPy Array`

```python
values = np.arange(10,20)
pd.Series(values)
```

## `Create Series from Dictonary`
- Dictionary is another data structure that stores the data in a key value pair
- It Stores the data in a key value pair so we don't have to explicitly mention the index

```python
data_items = {
    "Maths":100,
    "Biology":90,
    "Chemistry":80,
    "Physics":70,
    "English":60,
}

pd.Series(data_items) # already in key-value pair so we don't need to mention data or value explicitly
```

## `Create Series using Custom Index`

```python
values = [1,2,3,4,5]
pd.Series(data = values, index=['a','b','c','d','e'])
```

## `Create Series using Scalar Value`

```python
pd.Series(4, index=['a','b','c','d']) # same value for all indexes
```
