```python
df_two.head(1)
```

## 2. `tail()`
- provides the Botom Preview of Data as the Bottom 5 Rows of the Dataframe or Series

```python
df_two
```

## 3. `sample()`:
- provide any custom or random row from the dataset for us to preview it

```python
df_two.sample()
```

```python
df_two.sample(5) # any 5 random rows from data set
```

## 4. `value_count()`
- provide the idea about the count of the values ( categorical values )
- if i am having a column where the categorical value count I want then use value_count()
- eg. any name repeated, show how many time repeated

```python
youtube = pd.read_csv('youtube_2025_dataset_202504061614.csv')
print(youtube)
```

```python
yt_name = youtube['Youtuber Name']
print(yt_name)
print('\n')
print(yt_name.value_counts())
```

```python
avg = youtube['Avg Video Length (min)']
print(avg)
print(type(avg))
print("\n")
avg.value_counts() # show how many times repeated a value in dataset
```

## 5. `sort_values()`: 
- sort the values in ascending manner
- useful for the numerical column
- sort the value in descending manner , passing ascending = False in the sort_values(ascending=False)

```python
avg.sort_values() # youtube video length in ascending manner
```

```python
avg.sort_values(ascending=False)
```

### `Method Channing` :
- use multiple methods together to get the required value in a way the first method's output  will be input for the next one and so on..

```python
avg.sort_values(ascending=False).head(1)
```

```python
avg.sort_values(ascending=False).head(1).values[0]
```

## 5. `inplace` : in order to change the data permanent

```python
avg.sort_values(ascending=False)
```

```python
avg_copy = avg.copy()
```

```python
avg_copy.sort_values(inplace=True) # now it is permanent change
```

```python
avg_copy
```
