# `Working of Bidirectional RNN (BiRNN)`
---


A **Bidirectional Recurrent Neural Network (BiRNN)** processes a sequence in **both forward (left → right)** and **backward (right → left)** directions simultaneously. Unlike a traditional RNN, which only uses previous context, a BiRNN uses **both previous and future context** to make predictions.

It consists of **two separate RNNs**:

* **Forward RNN:** Processes the sequence from left to right.
* **Backward RNN:** Processes the sequence from right to left.

The outputs of both RNNs are then **combined** (usually by concatenation) to form the final hidden representation.

---

# Architecture

```text
                  Forward RNN

x1 ----> x2 ----> x3 ----> x4
 |         |         |         |
hf1      hf2      hf3      hf4


                 Backward RNN

x1 <---- x2 <---- x3 <---- x4
 |         |         |         |
hb1      hb2      hb3      hb4


        Final Hidden State

h1 = [hf1 ; hb1]

h2 = [hf2 ; hb2]

h3 = [hf3 ; hb3]

h4 = [hf4 ; hb4]
```

---

# Step-by-Step Working

Suppose the input sentence is:

```text
Deep Learning is Amazing
```

The words are:

```text
x1 = Deep
x2 = Learning
x3 = is
x4 = Amazing
```

---

# Step 1: Input Sequence

The sequence is fed into **two RNNs simultaneously**.

```text
Deep → Learning → is → Amazing
```

---

# Step 2: Forward Pass (Left → Right)

The first RNN processes the sentence from **left to right**.

```text
Deep → Learning → is → Amazing
```

Hidden states are computed one by one.

### At Time Step 1

Input:

```text
Deep
```

Previous hidden state:

```text
h0 = 0
```

Compute

[
$\overrightarrow{h_1}=f(x_1,h_0)$
]

Now the model remembers:

```text
Deep
```

---

### At Time Step 2

Input

```text
Learning
```

Compute

[
$\overrightarrow{h_2}=f(x_2,\overrightarrow{h_1})$
]

Now the memory contains

```text
Deep

Learning
```

---

### At Time Step 3

Input

```text
is
```

Memory becomes

```text
Deep

Learning

is
```

---

### At Time Step 4

Input

```text
Amazing
```

The final forward hidden state has seen the **entire sequence** from left to right.

---

# Forward Hidden States

```text
hf1

↓

hf2

↓

hf3

↓

hf4
```

Each hidden state contains information from **all previous words**.

---

# Step 3: Backward Pass (Right → Left)

The second RNN processes the sentence in reverse.

```text
Amazing → is → Learning → Deep
```

---

### Time Step 4

Input

```text
Amazing
```

Compute

[
$\overleftarrow{h_4}=f(x_4,h_0)$
]

---

### Time Step 3

Input

```text
is
```

Compute

[
$\overleftarrow{h_3}=f(x_3,\overleftarrow{h_4})$
]

Memory now contains

```text
Amazing

is
```

---

### Time Step 2

Input

```text
Learning
```

Memory becomes

```text
Amazing

is

Learning
```

---

### Time Step 1

Input

```text
Deep
```

The backward RNN has now seen the complete sentence from right to left.

---

# Backward Hidden States

```text
hb1

↑

hb2

↑

hb3

↑

hb4
```

Each hidden state contains information from **future words** in the original sequence.

---

# Step 4: Combine Both Hidden States

For every word, the forward and backward hidden states are combined.

Usually by concatenation.

For example,

For word

```text
Learning
```

Forward hidden state

```text
hf2
```

Backward hidden state

```text
hb2
```

Final hidden state

[
$h_2=[hf_2;hb_2]$
]

Example

Forward

```text
[0.2,0.8]
```

Backward

```text
[0.6,0.1]
```

Combined

```text
[0.2,0.8,0.6,0.1]
```

This representation now contains **both past and future context**.

---

# Step 5: Prediction

The combined hidden state is passed to the output layer.

```code
Combined Hidden State

↓

Dense Layer

↓

Softmax / Sigmoid

↓

Prediction
```

Depending on the task, the prediction could be:

* Sentiment
* Named Entity
* POS Tag
* Class Label
* Translation Output

---

# Complete Workflow

```code
             Input Sentence

      Deep Learning is Amazing
                |
                |
     ----------------------------
     |                          |
Forward RNN              Backward RNN
Left → Right            Right → Left
     |                          |
hf1 hf2 hf3 hf4       hb1 hb2 hb3 hb4
     |                          |
     -------- Concatenate --------
                |
      Final Hidden States
                |
          Dense Layer
                |
           Prediction
```

---

# Mathematical Representation

### Forward Hidden State

[
$\overrightarrow{h_t}=f(x_t,\overrightarrow{h_{t-1}})$
]

---

### Backward Hidden State

[
$\overleftarrow{h_t}=f(x_t,\overleftarrow{h_{t+1}})$
]

---

### Final Hidden State

[
$h_t=[\overrightarrow{h_t};\overleftarrow{h_t}]$
]

where:

* $(x_t)$ = current input
* $(\overrightarrow{h_t})$ = forward hidden state
* $(\overleftarrow{h_t})$ = backward hidden state
* $([,;,])$ = concatenation

---

# Example: Sentiment Analysis

Sentence:

```text
The movie was not good
```

Suppose the model is processing the word:

```text
good
```

### Forward RNN sees

```text
The movie was not
```

### Backward RNN sees

```text
(end of sentence)
```

If the sentence were:

```text
The movie was good but predictable
```

then while processing **good**, the backward RNN would also capture:

```text
but predictable
```

Combining both directions gives a richer understanding of the word's role in the sentence.

---

# Advantages

* Captures both past and future context.
* Improves understanding of ambiguous words.
* Produces richer hidden representations.
* Often achieves better accuracy than a unidirectional RNN on sequence-labeling tasks.

---

# Limitations

* Requires the complete input sequence before processing.
* Cannot be used for real-time next-word prediction or autoregressive text generation.
* Approximately doubles computation and parameters compared to a single-direction RNN.
* Still suffers from the sequential nature of RNNs, making it slower than Transformer-based models.

---

# Interview Summary

> A **Bidirectional RNN (BiRNN)** uses two separate RNNs: one processes the sequence from left to right, while the other processes it from right to left. At each time step, the forward and backward hidden states are combined—typically by concatenation—to form a context-rich representation. This allows the model to leverage information from both past and future words, making it well suited for tasks such as sentiment analysis, named entity recognition, POS tagging, and speech recognition. However, because it needs the entire sequence before processing, it is not suitable for next-word prediction or other autoregressive generation tasks.
