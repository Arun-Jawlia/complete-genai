# Implementation of Simple RNN vs Bidirectional RNN (BiRNN)

The implementation difference between a **Simple RNN** and a **Bidirectional RNN (BiRNN)** is minimal. The main distinction is that a BiRNN wraps the recurrent layer with a **Bidirectional** wrapper (in Keras) or sets `bidirectional=True` (in PyTorch).

---

# 1. Simple RNN Implementation (TensorFlow/Keras)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

model = Sequential([
    SimpleRNN(
        units=64,
        input_shape=(100, 50)
    ),
    Dense(1, activation='sigmoid')
])

model.summary()
```

### Explanation

```python
input_shape = (100, 50)
```

* **100** → Sequence length (time steps)
* **50** → Features per time step

Example:

```text
Sentence

I Love Deep Learning
```

After embedding

```text
(100 time steps, 50-dimensional embedding)
```

---

## Architecture

```text
Input
   │
   ▼
Simple RNN (64 Units)
   │
   ▼
Dense
   │
   ▼
Prediction
```

---

# 2. Bidirectional RNN Implementation (TensorFlow/Keras)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Bidirectional

model = Sequential([
    Bidirectional(
        SimpleRNN(64),
        input_shape=(100, 50)
    ),
    Dense(1, activation='sigmoid')
])

model.summary()
```

---

## What Changed?

Simple RNN:

```python
SimpleRNN(64)
```

Bidirectional RNN:

```python
Bidirectional(SimpleRNN(64))
```

The `Bidirectional` wrapper automatically creates:

* One RNN running **forward** (left → right)
* One RNN running **backward** (right → left)

Their outputs are concatenated.

---

## Architecture

```text
                   Input
                     │
         -------------------------
         │                       │
         ▼                       ▼
 Forward RNN              Backward RNN
         │                       │
         --------Concatenate-------
                    │
                    ▼
                 Dense
                    │
                    ▼
               Prediction
```

---

# Output Shape Difference

Suppose:

```python
SimpleRNN(64)
```

### Simple RNN

Output:

```text
64 Features
```

### Bidirectional RNN

Forward RNN:

```text
64 Features
```

Backward RNN:

```text
64 Features
```

After concatenation:

```text
128 Features
```

Therefore, the next layer receives **128 features** instead of **64**.

# Model Summary Comparison

### Simple RNN

```text
Input

↓

SimpleRNN (64)

↓

Dense(1)
```

---

### Bidirectional RNN

```text
Input

↓

Forward RNN (64)

+

Backward RNN (64)

↓

Concatenate

↓

Dense(1)
```

# Example: Sentiment Analysis

Sentence:

```text
The movie was really good
```

### Simple RNN

Processes:

```text
The → movie → was → really → good
```

Uses only previous words.

---

### Bidirectional RNN

Forward:

```text
The → movie → was → really → good
```

Backward:

```text
good → really → was → movie → The
```

The final representation for each word includes information from both directions.

# Using LSTM

### Simple LSTM

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

model = Sequential([
    LSTM(128, input_shape=(100, 50)),
    Dense(1, activation='sigmoid')
])
```

---

### Bidirectional LSTM

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Bidirectional, LSTM, Dense

model = Sequential([
    Bidirectional(
        LSTM(128),
        input_shape=(100, 50)
    ),
    Dense(1, activation='sigmoid')
])
```

---

# PyTorch Implementation

### Simple RNN

```python
import torch.nn as nn

model = nn.RNN(
    input_size=50,
    hidden_size=64,
    num_layers=1,
    batch_first=True
)
```

---

### Bidirectional RNN

```python
import torch.nn as nn

model = nn.RNN(
    input_size=50,
    hidden_size=64,
    num_layers=1,
    batch_first=True,
    bidirectional=True
)
```

The only additional argument is:

```python
bidirectional=True
```

# Comparison Table

| Feature                      | Simple RNN   | Bidirectional RNN             |
| ---------------------------- | ------------ | ----------------------------- |
| Processing Direction         | Left → Right | Left → Right and Right → Left |
| Future Context               | ❌ No         | ✅ Yes                         |
| Number of RNNs               | 1            | 2                             |
| Hidden State Size (64 units) | 64           | 128 (64 + 64)                 |
| Parameters                   | Fewer        | More                          |
| Training Speed               | Faster       | Slower                        |
| Accuracy                     | Lower        | Generally Higher              |
| Real-Time Prediction         | ✅ Yes        | ❌ No (requires full sequence) |

# Interview Tip

The implementation change is extremely small:

**Simple RNN**

```python
SimpleRNN(64)
```

**Bidirectional RNN**

```python
Bidirectional(SimpleRNN(64))
```

The same pattern applies to LSTM and GRU:

* `Bidirectional(LSTM(...))`
* `Bidirectional(GRU(...))`

The wrapper creates two recurrent networks that process the sequence in opposite directions and concatenates their outputs, allowing the model to leverage both past and future context.
