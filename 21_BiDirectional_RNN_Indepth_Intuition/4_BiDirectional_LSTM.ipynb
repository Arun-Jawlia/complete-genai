# Bidirectional LSTM (BiLSTM)

A **Bidirectional LSTM (BiLSTM)** is an extension of the standard **LSTM** that processes a sequence in **both forward (left → right)** and **backward (right → left)** directions.

It combines:

* **LSTM's ability to learn long-term dependencies**
* **Bidirectional processing to capture both past and future context**

As a result, BiLSTM often performs better than a unidirectional LSTM on tasks where the **entire input sequence is available**.

---

# Why Do We Need Bidirectional LSTM?

## Problem with Standard LSTM

A standard LSTM reads a sequence only from **left to right**.

Example:

```text
The movie was not good
```

While processing **good**, the LSTM knows:

```text
The movie was not
```

But it cannot use words that appear **after "good"**.

For many NLP tasks, future words also provide useful context.

---

# Example

Sentence:

```text
I deposited money in the bank yesterday.
```

The word **bank** could mean:

* Financial institution
* River bank

The words before and after **bank** help determine the correct meaning.

A BiLSTM uses information from **both directions**.

---

# Architecture

A Bidirectional LSTM consists of **two LSTM networks**.

1. **Forward LSTM**

   * Reads left → right

2. **Backward LSTM**

   * Reads right → left

Their outputs are concatenated.

```text
                 Input Sequence

x1 → x2 → x3 → x4 → x5

Forward LSTM

hf1 → hf2 → hf3 → hf4 → hf5

Backward LSTM

hb1 ← hb2 ← hb3 ← hb4 ← hb5

Final Output

[hf ; hb]
```

---

# Working of Bidirectional LSTM

Suppose the sentence is:

```text
Deep Learning is Amazing
```

Sequence:

```text
x1 = Deep
x2 = Learning
x3 = is
x4 = Amazing
```

---

## Step 1: Forward LSTM

The first LSTM reads

```text
Deep → Learning → is → Amazing
```

It computes

```text
hf1

↓

hf2

↓

hf3

↓

hf4
```

Each hidden state contains information from all **previous words**.

---

## Step 2: Backward LSTM

The second LSTM reads

```text
Amazing → is → Learning → Deep
```

It computes

```text
hb4

↓

hb3

↓

hb2

↓

hb1
```

Each hidden state contains information from **future words** in the original sequence.

---

## Step 3: Concatenate Hidden States

For each position

```text
Deep

↓

[hf1 ; hb1]

Learning

↓

[hf2 ; hb2]

is

↓

[hf3 ; hb3]

Amazing

↓

[hf4 ; hb4]
```

These combined vectors become the final representation for each word.

---

# Mathematical Representation

Forward LSTM

[
\overrightarrow{h_t}=LSTM(x_t,\overrightarrow{h_{t-1}})
]

Backward LSTM

[
\overleftarrow{h_t}=LSTM(x_t,\overleftarrow{h_{t+1}})
]

Final Hidden State

[
h_t=[\overrightarrow{h_t};\overleftarrow{h_t}]
]

---

# Complete Workflow

```text
Input Sentence

Deep Learning is Amazing

        |

        |

-------------------------------

|                             |

Forward LSTM            Backward LSTM

Left → Right            Right → Left

|                             |

hf1 hf2 hf3 hf4       hb1 hb2 hb3 hb4

|                             |

------ Concatenate ------------

             |

      Final Hidden State

             |

        Dense Layer

             |

        Prediction
```

---

# Example

Sentence

```text
The food was not good
```

Forward LSTM knows

```text
The

↓

food

↓

was

↓

not
```

Backward LSTM knows

```text
(end)

↓

good
```

When predicting sentiment, both directions help the model understand that **"not good"** indicates a negative sentiment.

# TensorFlow/Keras Implementation

### Standard LSTM

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

model = Sequential([
    LSTM(64, input_shape=(100, 50)),
    Dense(1, activation='sigmoid')
])

model.summary()
```

---

### Bidirectional LSTM

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Bidirectional, LSTM, Dense

model = Sequential([
    Bidirectional(
        LSTM(64),
        input_shape=(100, 50)
    ),
    Dense(1, activation='sigmoid')
])

model.summary()
```

Only one change:

```python
LSTM(64)
```

↓

```python
Bidirectional(LSTM(64))
```

---

# PyTorch Implementation

### Standard LSTM

```python
import torch.nn as nn

model = nn.LSTM(
    input_size=50,
    hidden_size=64,
    batch_first=True
)
```

---

### Bidirectional LSTM

```python
import torch.nn as nn

model = nn.LSTM(
    input_size=50,
    hidden_size=64,
    batch_first=True,
    bidirectional=True
)
```

The only additional argument is:

```python
bidirectional=True
```

# Advantages

* Learns long-term dependencies.
* Uses both past and future context.
* Higher accuracy than a unidirectional LSTM on many sequence-labeling tasks.
* Excellent for contextual understanding.

---

# Disadvantages

* Roughly doubles the number of computations and parameters compared to a single LSTM of the same hidden size.
* Slower training and inference.
* Requires the entire input sequence before processing.
* Not suitable for autoregressive next-word generation.

---

# Applications

BiLSTMs are commonly used for:

1. Sentiment Analysis
2. Named Entity Recognition (NER)
3. Part-of-Speech (POS) Tagging
4. Speech Recognition
5. Machine Translation (encoder)
6. Text Classification
7. Question Answering
8. Medical Text Processing

# LSTM vs Bidirectional LSTM

| Feature                | LSTM         | Bidirectional LSTM            |
| ---------------------- | ------------ | ----------------------------- |
| Direction              | Left → Right | Left → Right and Right → Left |
| Uses Past Context      | ✅            | ✅                             |
| Uses Future Context    | ❌            | ✅                             |
| Long-Term Memory       | ✅            | ✅                             |
| Context Understanding  | Good         | Better                        |
| Number of LSTMs        | 1            | 2                             |
| Hidden Size (64 units) | 64           | 128 (after concatenation)     |
| Training Speed         | Faster       | Slower                        |
| Accuracy               | Good         | Often Better                  |
| Real-Time Prediction   | ✅ Yes        | ❌ No                          |

---

# BiRNN vs BiLSTM vs BiGRU

| Feature                | BiRNN              | BiLSTM                           | BiGRU                   |
| ---------------------- | ------------------ | -------------------------------- | ----------------------- |
| Memory Mechanism       | Basic hidden state | Cell state + hidden state        | Hidden state with gates |
| Long-Term Dependencies | Poor               | Excellent                        | Very Good               |
| Number of Gates        | 0                  | 3                                | 2                       |
| Training Speed         | Fast               | Slowest                          | Faster than BiLSTM      |
| Number of Parameters   | Few                | Most                             | Fewer than BiLSTM       |
| Performance            | Lowest             | Highest (traditional RNN family) | Comparable to BiLSTM    |

# Interview Summary

> **Bidirectional LSTM (BiLSTM)** consists of two LSTMs: one processes the sequence from left to right, while the other processes it from right to left. The hidden states from both directions are concatenated to produce a richer representation for each time step. Because it captures both past and future context while retaining LSTM's ability to learn long-term dependencies, BiLSTM is widely used in tasks such as sentiment analysis, named entity recognition, speech recognition, and machine translation encoders. However, it requires the full input sequence and is therefore not suitable for real-time next-word prediction or autoregressive language generation.
