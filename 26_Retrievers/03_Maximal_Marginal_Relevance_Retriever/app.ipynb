```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_openai import OpenAIEmbeddings
# from langchain_chroma import Chroma
# from langchain_core.documents import Document
from dotenv import load_dotenv
```

```python
# embeddings = OpenAIEmbeddings() # due to api cost, I am using HuggingFaceEmbeddings
```

```python
load_dotenv()
```

```python
embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
query = 'What are the Renewabel Enery Source?'
```

```python
docs = [
    'Solar power converst sunlight into electricity.',
    'Solar panels are widely used in residential homes.',
    'Wind Turbines harness wind energy for electricity.',
    'Hydopower generates energy from flowing water.',
    'Geothermal enery comes from heat inside the Earth.'
]
```

```python
vector_store = FAISS.from_texts(docs, embeddings)
```

```python
retriever = vector_store.as_retriever(search_type='mmr', search_kwards={'k': 3, 'lambda_mult': 1})
```

Note: lambda mult: add diversity and it should be between 0 and 1

```python
results = retriever.invoke(query)
```

```python
print("MMR Results")
for i in results:
    print("-", i.page_content)
```

```python
retriever = vector_store.as_retriever(search_type='mmr', search_kwards={'k': 3, 'lambda_mult': 0.5})
```

Note: lambda mult: add diversity and it should be between 0 and 1

```python
results = retriever.invoke(query)
```

```python
print("MMR Results")
for i in results:
    print("-", i.page_content)
```
