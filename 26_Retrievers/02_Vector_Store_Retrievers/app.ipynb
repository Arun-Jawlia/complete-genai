# Vector Store Retrievers

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
documents = [Document(page_content = 'Deep Learning Unsupervised Learning Technique is GAN which helps in GenAI'),
Document(page_content = 'Deep Learning uses Human Brain Neural Network as the Working Model'),
Document(page_content = 'Deep learing is subfield of Machine Learning')
]
```

```python
vector_store = Chroma.from_documents(documents= documents, embedding = embeddings, collection_name = 'my_collection')
```

```python
retirever = vector_store.as_retriever(search_kwargs = {"k": 1})
```

```python
query='How Deep Learning helps in achieving GenAI'
```

```python
answer = retirever.invoke(query)
```

```python
for i, doc  in enumerate(answer):
    print(f'\n Result{i+1}')
    print(doc.page_content)
```
