```python
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
```

```python
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings, HuggingFaceEndpoint
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_core.documents import Document
import os
```

```python
embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
query = 'How do I reset my Router Password'
```

```python
docs = [
    '''The router has multiple features including firewall, guest wifi, and VPN.
     To reset the router password: Log into 192.168.0.1, go to admin Setting -> Security --> Change Password.
      You can also restore factory setting if the password is lost '''
]
```

```python
vectorStore = FAISS.from_texts(docs, embeddings)
```

```python
# LLM = ChatOpenAI(model='gpt-3.5-turbo', temperature=0)
```

```python
LLM = HuggingFaceEndpoint(
    repo_id="Qwen/Qwen2.5-0.5B-Instruct",
    task="conversational",
    max_new_tokens=256,
    temperature=0.2,
)
```

```python
base_retriever = vectorStore.as_retriever()
```

```python
compressor = LLMChainExtractor.from_llm(LLM)
```

```python
compression_retriever = ContextualCompressionRetriever(base_compressor=compressor, base_retriever=base_retriever)
```

```python
results = compression_retriever.invoke(query)
```

```python
print('\n Contextual Compression Results:')
for i in results:
    print(f'i-{i.page_content.strip()}')
```
