# RNN Architecture (Recurrent Neural Network Architecture)
---

# Introduction

A **Recurrent Neural Network (RNN)** is designed to process **sequential data** by maintaining a **hidden state (memory)**.

Unlike Artificial Neural Networks (ANNs), which process every input independently, an RNN carries information from previous time steps while processing the current input.

The key idea behind RNN architecture is:

> **Current Output = Current Input + Previous Memory**

---

# Why Do We Need an RNN Architecture?

Consider the sentence:

```
I love Deep Learning
```

Each word depends on the previous words.

```
I

↓

I love

↓

I love Deep

↓

I love Deep Learning
```

To understand the sentence correctly, the model must remember previous words.

This is exactly what the RNN architecture is designed to do.

---

# Basic RNN Cell

The smallest building block of an RNN is called an **RNN Cell**.

It receives:

- Current Input (xₜ)
- Previous Hidden State (hₜ₋₁)

It produces:

- Current Hidden State (hₜ)
- Current Output (yₜ)

```
            Previous Hidden State (hₜ₋₁)
                     │
                     ▼
Current Input (xₜ) ─► [ RNN Cell ] ─► Output (yₜ)
                     │
                     ▼
             Current Hidden State (hₜ)
```

---

# Components of an RNN Architecture

## 1. Input Layer (x)

The input at each time step.

Examples:

Text

```
Word

↓

Embedding Vector
```

Speech

```
Audio Frame
```

Stock Prediction

```
Today's Price
```

---

## 2. Hidden Layer (Hidden State)

The hidden layer stores information from previous time steps.

It acts as the **memory** of the RNN.

```
Previous Memory

+

Current Input

↓

Hidden State
```

---

## 3. Output Layer

Produces predictions.

Examples

- Next word
- Sentiment
- Translation
- Stock price
- Weather forecast

---

# How Hidden State Flows

Suppose we have

```
I

Love

AI
```

Initially

```
h₀ = 0
```

---

### Time Step 1

```
Input

I

+

h₀

↓

RNN Cell

↓

h₁
```

---

### Time Step 2

```
Love

+

h₁

↓

RNN Cell

↓

h₂
```

---

### Time Step 3

```
AI

+

h₂

↓

RNN Cell

↓

h₃
```

Notice that every hidden state carries information from all previous words.

---

# Unrolled RNN Architecture

An RNN is easier to understand when it is "unrolled" across time.

```
          h₀
           │
           ▼
x₁ ─► [RNN] ─► h₁ ─► y₁
           │
           ▼
x₂ ─► [RNN] ─► h₂ ─► y₂
           │
           ▼
x₃ ─► [RNN] ─► h₃ ─► y₃
```

Each box represents the **same RNN cell**.

---

# Why Are They the Same Cell?

The network is **not creating a new neural network** for every word.

Instead,

```
One Cell

↓

Reuse

↓

Reuse

↓

Reuse
```

The same weights are shared.

This is called **weight sharing**.

---

# Weight Sharing

The following matrices remain the same at every time step:

```
Wxh

(Input → Hidden)

Whh

(Hidden → Hidden)

Why

(Hidden → Output)
```

Only the **inputs** and **hidden states** change.

---

# Mathematical Representation

## Hidden State

```
hₜ = tanh(Wxh·xₜ + Whh·hₜ₋₁ + b)
```

Where:

- **xₜ** → Current input
- **hₜ₋₁** → Previous hidden state
- **Wxh** → Input-to-hidden weights
- **Whh** → Hidden-to-hidden weights
- **b** → Bias
- **tanh** → Activation function

---

## Output

```
yₜ = softmax(Why·hₜ + c)
```

For binary classification, the output layer often uses a **sigmoid** activation instead of softmax.

---

# Data Flow Inside an RNN

```
Current Input

+

Previous Hidden State

↓

Matrix Multiplication

↓

Activation Function

↓

New Hidden State

↓

Prediction

↓

Pass Memory Forward
```

---

# Internal Working of an RNN Cell

```
          Previous Hidden
                 │
                 ▼
Input ─► Matrix Addition ─► tanh ─► Hidden State
                 │
                 ▼
             Output Layer
                 │
                 ▼
              Prediction
```

---

# Types of RNN Architectures

Different tasks require different input-output structures.

## 1. One-to-One

```
Input

↓

Output
```

Example:

- Image Classification

(Note: This is a standard neural network pattern rather than a sequence-processing RNN.)

---

## 2. One-to-Many

```
Image

↓

RNN

↓

Word 1

↓

Word 2

↓

Word 3
```

Example:

- Image Captioning

---

## 3. Many-to-One

```
Word

↓

Word

↓

Word

↓

Sentiment
```

Examples:

- Sentiment Analysis
- Spam Detection

---

## 4. Many-to-Many (Same Length)

```
English

↓

French
```

Examples:

- POS Tagging
- Named Entity Recognition

---

## 5. Many-to-Many (Different Length)

```
English Sentence

↓

Encoder

↓

Decoder

↓

French Sentence
```

Examples:

- Machine Translation
- Text Summarization

This encoder-decoder architecture was widely used before Transformers became dominant.

# Advantages of RNN Architecture

- Processes sequential data naturally.
- Maintains memory using hidden states.
- Shares weights across time steps.
- Handles variable-length sequences.
- Suitable for NLP, speech, and time-series applications.

---

# Limitations of Basic RNN Architecture

- Vanishing gradients
- Exploding gradients
- Difficulty remembering long sequences
- Sequential processing limits parallel computation

These limitations motivated:

- LSTM
- GRU

Later,

Transformers became the dominant architecture for many NLP applications.

---

# Comparison: ANN vs RNN Architecture

| Feature | ANN | RNN |
|---------|-----|-----|
| Memory | ❌ No | ✅ Yes |
| Sequence Processing | ❌ No | ✅ Yes |
| Weight Sharing | ❌ No | ✅ Yes |
| Variable-Length Input | Limited | Yes |
| Hidden State | No | Yes |
| Best For | Tabular Data | Sequential Data |

# Interview Points

- An RNN cell receives the current input and previous hidden state.
- The hidden state acts as the memory of the network.
- The same RNN cell and weight matrices are reused across all time steps.
- The hidden state is computed using `hₜ = tanh(Wxh·xₜ + Whh·hₜ₋₁ + b)`.
- Different RNN architectures include one-to-many, many-to-one, and many-to-many configurations.
- Basic RNNs inspired more advanced architectures such as LSTMs, GRUs, and later Transformers.

---

# Key Takeaways

- The RNN architecture is specifically designed for sequential data.
- Its core innovation is the hidden state, which stores information across time.
- Weight sharing makes RNNs efficient and capable of processing sequences of different lengths.
- The unrolled architecture helps visualize how information flows from one time step to the next.
- Understanding the RNN architecture is essential before studying LSTMs, GRUs, sequence-to-sequence models, and Transformers.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. What is the architecture of an RNN?

**Answer:**
An RNN architecture consists of an input layer, a recurrent hidden layer (hidden state), and an output layer. The hidden state carries information from previous time steps, enabling sequence modeling.

---

### 2. What is an RNN cell?

**Answer:**
An RNN cell is the basic computational unit that receives the current input and the previous hidden state, computes a new hidden state, and optionally produces an output.

---

### 3. What is the hidden state?

**Answer:**
The hidden state is the memory of the RNN that stores information from previous inputs and passes it to the next time step.

---

### 4. Why does an RNN reuse the same cell?

**Answer:**
The same cell shares its learned weights across all time steps, reducing the number of parameters and allowing the network to process sequences of varying lengths.

---

### 5. What are the three main weight matrices in a basic RNN?

**Answer:**

* **Wxh** – Input → Hidden
* **Whh** – Hidden → Hidden
* **Why** – Hidden → Output

---

## Intermediate Questions

### 6. What is an unrolled RNN?

**Answer:**
An unrolled RNN is a visualization where the same RNN cell is shown repeatedly across time steps to illustrate how information flows through a sequence.

---

### 7. Why is weight sharing important?

**Answer:**
Weight sharing reduces the number of trainable parameters, improves generalization, and enables the same model to process sequences of different lengths.

---

### 8. What is the role of the hidden state equation?

**Answer:**
It combines the current input with the previous hidden state to compute the updated memory that is passed to the next time step.

---

### 9. What is the difference between many-to-one and many-to-many RNNs?

**Answer:**

* **Many-to-One:** Multiple inputs produce a single output (e.g., sentiment analysis).
* **Many-to-Many:** Multiple inputs produce multiple outputs (e.g., machine translation or sequence labeling).

---

### 10. Why are LSTMs and GRUs preferred over basic RNNs?

**Answer:**
They include gating mechanisms that help preserve long-term information and reduce the vanishing gradient problem, making them more effective for long sequences.

---

## Scenario-Based Questions

### Q1. You're building a sentiment analysis model that classifies an entire review as positive or negative. Which RNN architecture fits best?

**Answer:**
A **many-to-one** architecture because it processes the entire sequence and produces one final prediction.

---

### Q2. You're building an image captioning system. Which RNN architecture is suitable?

**Answer:**
A **one-to-many** architecture, where a single image is used to generate a sequence of words describing it.

---

### Q3. Why does an RNN not need a separate network for every word in a sentence?

**Answer:**
Because the same RNN cell and shared weights are reused at every time step, making the model efficient and capable of handling variable-length sequences.
