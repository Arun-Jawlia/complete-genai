# Why Recurrent Neural Networks (RNNs) are Required?

## Introduction

Before the introduction of **Recurrent Neural Networks (RNNs)**, most neural networks such as **Artificial Neural Networks (ANNs)** and **Convolutional Neural Networks (CNNs)** treated every input independently.

However, many real-world problems involve **sequential data**, where the order of information matters.

Examples:

- Sentences
- Speech
- Time-series data
- Stock prices
- Weather forecasting
- Music
- Videos

To understand such data, we need a model that can **remember previous information**.

This requirement led to the development of **Recurrent Neural Networks (RNNs).**

---

# What is Sequential Data?

## Definition

Sequential data is data where **the order of elements is important**.

Example

```
"I love Deep Learning"
```

Changing the order changes the meaning.

```
"Learning Deep love I"
```

The sentence no longer makes sense.

---

# Why Can't ANN Handle Sequential Data?

ANN assumes every input is independent.

Example

```
x1

x2

x3

↓

ANN

↓

Output
```

The ANN has **no memory**.

It cannot remember previous inputs.

---

## Example

Sentence

```
The movie was not good.
```

If the model looks at only one word at a time,

```
good
```

it may predict **Positive**.

But if it remembers

```
not
```

the prediction becomes

```
Negative
```

Context matters.

---

# Why Can't CNN Handle Sequential Data?

CNN is excellent for:

- Images
- Object Detection
- Face Recognition
- Medical Imaging

because images have **spatial relationships**.

Example

```
Pixel

↓

Convolution

↓

Feature Extraction
```

CNN focuses on neighboring pixels, not on remembering previous time steps.

Although specialized CNN variants can process sequences, standard CNNs are not designed to maintain memory across long sequences.

---

# The Problem

Traditional Neural Networks

```
Input

↓

Prediction

↓

Forget Everything
```

There is **no memory**.

Each prediction is independent.

---

# Why Do We Need Memory?

Imagine reading a story.

Sentence 1

```
Rahul bought a dog.
```

Sentence 2

```
It is very playful.
```

Question

Who is **"It"**?

Without memory,

the model cannot know.

With memory,

```
It

↓

Dog
```

---

# What is an RNN?

## Definition

A **Recurrent Neural Network (RNN)** is a neural network designed to process **sequential data** by maintaining information from previous time steps.

Unlike ANN,

RNN has a **memory mechanism**.

---

# Basic RNN Architecture

```
Input x₁

↓

Hidden State h₁

↓

Output y₁

↓

Input x₂

↓

Hidden State h₂

↓

Output y₂

↓

Input x₃

↓

Hidden State h₃

↓

Output y₃
```

The hidden state carries information from one step to the next.

---

# Hidden State

The hidden state acts as the memory of the network.

```
Previous Information

+

Current Input

↓

Hidden State

↓

Next Prediction
```

Instead of forgetting everything,

the network remembers useful information.

---

# RNN Working

Suppose the sentence is

```
I

love

AI
```

Step 1

```
Input

I

↓

Memory

I
```

---

Step 2

```
Input

love

+

Previous Memory

↓

I love
```

---

Step 3

```
Input

AI

+

Previous Memory

↓

I love AI
```

The final prediction uses both the current input and the accumulated context.

# Why RNN Became Important

Before RNN,

models struggled to understand sequences.

RNN introduced the idea of:

```
Previous Knowledge

+

Current Input

↓

Better Prediction
```

This significantly improved performance on sequential tasks.

---

# Limitations of Basic RNN

Although RNN introduced memory,

it struggles with very long sequences due to problems such as:

- Vanishing gradients
- Exploding gradients
- Difficulty remembering long-term dependencies

These limitations led to more advanced architectures like:

- LSTM (Long Short-Term Memory)
- GRU (Gated Recurrent Unit)

Today, many NLP tasks are handled by Transformer-based models, which are even better at capturing long-range dependencies.

---

# Interview Points

- RNN is designed for sequential data.
- It introduces a hidden state that acts as memory.
- ANN cannot remember previous inputs.
- CNN is designed for spatial data such as images.
- RNN processes one element at a time while carrying forward information.
- LSTM and GRU were developed to overcome limitations of basic RNNs.

---

# Key Takeaways

- RNNs were introduced because many real-world problems involve sequential data.
- The hidden state allows RNNs to remember previous information.
- RNNs are useful for NLP, speech recognition, and time-series forecasting.
- Basic RNNs have limitations with long sequences, leading to LSTM, GRU, and later Transformer models.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. Why is an RNN required?

**Answer:**
RNNs are required to process sequential data where previous information influences the current prediction. They introduce a memory mechanism through the hidden state.

---

### 2. What is sequential data?

**Answer:**
Sequential data is data in which the order of elements matters, such as text, speech, time-series data, and videos.

---

### 3. Why can't an ANN handle sequential data effectively?

**Answer:**
ANNs treat each input independently and have no memory of previous inputs.

---

### 4. What is the hidden state in an RNN?

**Answer:**
The hidden state is the internal memory of the RNN that carries information from previous time steps to future time steps.

---

### 5. Name some applications of RNNs.

**Answer:**

* Machine Translation
* Speech Recognition
* Sentiment Analysis
* Time-Series Forecasting
* Video Analysis

---

## Intermediate Questions

### 6. Why isn't a CNN typically used for language modeling?

**Answer:**
Standard CNNs are designed to learn spatial patterns in images rather than maintain sequential memory across time steps.

---

### 7. How does an RNN remember previous information?

**Answer:**
By passing the hidden state from one time step to the next, combining previous context with the current input.

---

### 8. What are the limitations of a basic RNN?

**Answer:**

* Vanishing gradients
* Exploding gradients
* Difficulty learning long-term dependencies

---

### 9. Which architectures improved upon the basic RNN?

**Answer:**

* LSTM (Long Short-Term Memory)
* GRU (Gated Recurrent Unit)

---

### 10. Why are Transformers now preferred for many NLP tasks?

**Answer:**
Transformers process sequences in parallel and capture long-range dependencies more effectively than RNNs, making them the dominant architecture for modern large language models.

---

## Scenario-Based Questions

### Q1. You're building a model to predict tomorrow's stock price using historical prices. Which architecture would you choose?

**Answer:**
An RNN-based architecture such as LSTM or GRU is suitable because stock prices are sequential time-series data. Transformer-based time-series models may also be considered for complex problems.

---

### Q2. Why would an ANN struggle with language translation?

**Answer:**
Because translation depends on word order and context, while an ANN processes inputs independently without memory.

---

### Q3. You're building a voice assistant. Why is sequence modeling important?

**Answer:**
Speech is a sequence of sounds where each word depends on previous sounds and context. A sequence model can capture these dependencies to improve recognition accuracy.
