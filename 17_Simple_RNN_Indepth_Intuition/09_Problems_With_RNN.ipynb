# Problems with Recurrent Neural Networks (RNN)

# Introduction

Recurrent Neural Networks (RNNs) were one of the first neural network architectures designed to process **sequential data**.

Unlike Artificial Neural Networks (ANNs), RNNs can remember previous information using a **hidden state**, making them suitable for tasks like:

- Language Translation
- Sentiment Analysis
- Speech Recognition
- Time Series Forecasting

However, despite their ability to process sequences, **basic RNNs have several major limitations**.

These limitations eventually led to the development of **LSTM, GRU, and Transformers**.

---

# Why Do We Need to Study RNN Problems?

Understanding the limitations of RNN helps answer an important question:

> **If RNNs can process sequences, why were LSTMs, GRUs, and Transformers invented?**

The answer lies in the shortcomings of basic RNNs.

---

# Problem 1: Vanishing Gradient

## What is it?

During Backpropagation Through Time (BPTT), gradients are propagated backward through multiple time steps.

If the gradients become very small after repeated multiplication, they eventually approach zero.

As a result, the network stops learning from earlier inputs.

---

## Mathematical Intuition

Suppose the gradient at every time step is

```
0.5
```

After four time steps

```
0.5 × 0.5 × 0.5 × 0.5

=

0.0625
```

After many time steps

```
≈ 0
```

The earlier layers receive almost no gradient.

---

## Example

Sentence

```
I grew up in France.
Now I speak fluent ______.
```

The correct prediction is

```
French
```

To predict "French",

the model must remember

```
France
```

which appeared much earlier.

A basic RNN often forgets this information due to the vanishing gradient problem.

---

## Impact

- Cannot learn long-term dependencies.
- Slow or no learning for early time steps.
- Poor performance on long sequences.

---

# Problem 2: Exploding Gradient

## What is it?

Sometimes gradients become larger after repeated multiplication.

Instead of shrinking,

they grow rapidly.

---

## Mathematical Intuition

Suppose every derivative is

```
2
```

After five time steps

```
2 × 2 × 2 × 2 × 2

=

32
```

After many time steps

```
≈ Very Large
```

Weight updates become unstable.

---

## Impact

- Training becomes unstable.
- Loss oscillates or diverges.
- Model parameters become extremely large.

---

## Solution

Gradient Clipping

```
Large Gradient

↓

Clip

↓

Safe Gradient
```

---

# Problem 3: Short-Term Memory

## What is it?

Basic RNNs can remember only recent information.

They struggle to retain information from far earlier in the sequence.

---

## Example

```
John was born in Canada.

...

After many sentences...

He became a scientist.
```

To understand

```
He
```

the model must remember

```
John
```

Basic RNNs often fail on long documents.

---

## Real-Life Analogy

Imagine reading a 500-page novel.

If you forget the first chapter,

the ending becomes difficult to understand.

Basic RNN behaves similarly.

---

# Problem 4: Sequential Computation

## What is it?

Each hidden state depends on the previous hidden state.

```
h₁

↓

h₂

↓

h₃

↓

h₄
```

Therefore,

```
h₂
```

cannot be computed before

```
h₁
```

is completed.

---

## Impact

- Cannot fully parallelize computations.
- Training becomes slower.
- GPU utilization is limited.

---

## Comparison

ANN

```
All inputs

↓

Processed together
```

Transformer

```
All tokens

↓

Processed simultaneously
```

RNN

```
One token

↓

Next token

↓

Next token
```

---

# Problem 5: Slow Training

Because computations happen one time step after another,

RNN training is slower than Transformer training.

Longer sequences require more computation.

---

# Problem 6: High Memory Consumption

During BPTT,

the hidden state at every time step must be stored.

```
Time1

↓

Store h₁

Time2

↓

Store h₂

Time3

↓

Store h₃
```

All hidden states are required during backpropagation.

For long sequences,

memory usage increases significantly.

---

# Problem 7: Difficulty Learning Long-Term Dependencies

Suppose

```
The movie released in 1995 ...

...

Several paragraphs later ...

It won an Oscar.
```

To understand

```
It
```

the model must remember

```
movie
```

from much earlier.

Basic RNNs often fail to capture such long-range relationships.

---

# Problem 8: Limited Context

Basic RNN compresses all previous information into a single hidden state.

```
Sentence

↓

Hidden State

↓

Prediction
```

As the sequence grows,

the hidden state may not be sufficient to retain all important information.

---

# Problem 9: Exposure Bias (During Sequence Generation)

During training,

the model often receives the correct previous token.

During inference,

it must use its own previous prediction.

If one prediction is wrong,

future predictions may also become incorrect.

---

## Example

Correct sentence

```
I love deep learning.
```

Model predicts

```
I love sleep learning.
```

The incorrect word

```
sleep
```

affects subsequent predictions.

---

# Summary Table

| Problem | Description | Effect |
|----------|-------------|--------|
| Vanishing Gradient | Gradient becomes very small | Cannot learn long-term dependencies |
| Exploding Gradient | Gradient becomes very large | Unstable training |
| Short-Term Memory | Forgets older information | Poor performance on long sequences |
| Sequential Computation | Processes one step at a time | Slow training |
| High Memory Usage | Stores hidden states | More memory required |
| Long-Term Dependency | Cannot relate distant information | Lower accuracy |
| Limited Context | Single hidden state bottleneck | Information loss |
| Exposure Bias | Uses own predictions during inference | Error accumulation |

# How Were These Problems Solved?

| Problem | Solution |
|----------|----------|
| Vanishing Gradient | LSTM, GRU |
| Exploding Gradient | Gradient Clipping |
| Long-Term Dependency | LSTM, GRU |
| Sequential Processing | Transformer |
| Limited Context | Attention Mechanism |
| Context Bottleneck | Attention + Transformer |

# Interview Points

- Vanishing gradient is the most important limitation of basic RNNs.
- Exploding gradients can be controlled using Gradient Clipping.
- Basic RNNs struggle with long-term dependencies.
- Sequential computation prevents efficient parallelization.
- LSTM and GRU improve memory retention using gating mechanisms.
- Transformers replace recurrence with self-attention, enabling parallel processing and better long-range dependency modeling.

---

# Key Takeaways

- RNNs introduced memory into neural networks but have important limitations.
- Vanishing and exploding gradients make training difficult.
- Basic RNNs cannot reliably learn long-term dependencies.
- Sequential computation slows training.
- LSTM, GRU, Attention, and Transformers were developed to overcome these limitations.

# 2. Interview Questions & Answers

### 1. What is the biggest limitation of a basic RNN?

**Answer:**
The biggest limitation is the **vanishing gradient problem**, which prevents the network from learning long-term dependencies.

---

### 2. What causes the vanishing gradient problem?

**Answer:**
Repeated multiplication of small derivatives during Backpropagation Through Time causes gradients to shrink toward zero.

---

### 3. What is the exploding gradient problem?

**Answer:**
It occurs when repeated multiplication of large derivatives causes gradients to become extremely large, leading to unstable training.

---

### 4. How can exploding gradients be prevented?

**Answer:**
Using **Gradient Clipping**, which limits the magnitude of gradients before updating the weights.

---

### 5. Why are RNNs slower than Transformers?

**Answer:**
RNNs process one time step after another because each hidden state depends on the previous one, while Transformers process all tokens in parallel using self-attention.

---

### 6. Why do basic RNNs struggle with long-term dependencies?

**Answer:**
Because information from earlier time steps gradually fades due to the vanishing gradient problem and the limited capacity of a single hidden state.

---

### 7. Which architectures were developed to overcome RNN limitations?

**Answer:**
LSTM, GRU, Attention Mechanisms, and Transformer architectures.

---

### 8. Is a basic RNN still used in industry?

**Answer:**
Basic RNNs are rarely used for modern large-scale NLP tasks. They are mainly studied for educational purposes, while LSTMs, GRUs, and Transformers are preferred in production.

# 4. 2-Minute Revision

RNNs were the first deep learning models capable of handling sequential data through recurrent connections and hidden states. However, they face major challenges such as vanishing and exploding gradients, short-term memory, difficulty learning long-range dependencies, sequential computation, and high memory usage during BPTT. These limitations inspired more advanced architectures like LSTM, GRU, Attention Mechanisms, and Transformers, which are now widely used in modern NLP and sequence modeling.
