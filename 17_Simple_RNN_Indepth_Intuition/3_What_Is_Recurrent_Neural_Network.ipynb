# Recurrent Neural Networks (RNN) – Detailed Explanation with Vectorization

# Introduction

After understanding the limitations of Artificial Neural Networks (ANNs), researchers developed **Recurrent Neural Networks (RNNs)** to solve problems involving **sequential data**.

Unlike ANNs, RNNs have **memory**, allowing them to use information from previous inputs while processing the current input.

This makes RNNs suitable for:

- Natural Language Processing (NLP)
- Speech Recognition
- Time-Series Forecasting
- Machine Translation
- Music Generation
- Video Processing

---

# What is an RNN?

## Definition

A **Recurrent Neural Network (RNN)** is a neural network architecture designed to process sequential data by maintaining a **hidden state (memory)** that carries information from one time step to the next.

Unlike ANN,

```
Current Output

depends only on

Current Input
```

In RNN,

```
Current Output

depends on

Current Input

+

Previous Memory
```

---

# Why is RNN Called "Recurrent"?

The word **Recurrent** means **repeating**.

The same neural network is reused for every time step.

Instead of creating a different neural network for every word,

the **same weights** are reused repeatedly.

```
Word 1

↓

Same Network

↓

Word 2

↓

Same Network

↓

Word 3

↓

Same Network
```

This sharing of weights allows RNNs to process sequences of arbitrary length.

---

# Sequential Example

Sentence

```
I love AI
```

Sequence

```
x₁ = I

x₂ = love

x₃ = AI
```

Each word arrives one after another.

---

# Unrolled RNN

Instead of showing one repeated cell,

we "unroll" the network.

```
            h₀
             │
             ▼
x₁ ───► [RNN] ───► h₁ ───► y₁
             │
             ▼
x₂ ───► [RNN] ───► h₂ ───► y₂
             │
             ▼
x₃ ───► [RNN] ───► h₃ ───► y₃
```

Where

- x = input
- h = hidden state (memory)
- y = output

# Components of an RNN

## 1. Input Vector (x)

Represents the current input.

Examples

Text

```
"I"
```

Image Sequence

```
Frame 1
```

```
Today's Price
```

## 2. Hidden State (h)

The hidden state stores information from previous time steps.

It acts as the memory.

```
Past Knowledge

↓

Hidden State

↓

Future Prediction
```

---

## 3. Output (y)

The prediction produced at each time step.

Examples

# Weight Matrices in RNN

An RNN learns three main sets of weights.

## 1. Input → Hidden

```
Wxh
```

Connects

```
Input

↓

Hidden
```

Shape

```
(hidden_size × input_size)
```

Example

```
4 × 3
```

---

## 2. Hidden → Hidden

```
Whh
```

Connects

```
Previous Hidden

↓

Current Hidden
```

This matrix gives the RNN its memory.

Shape

```
4 × 4
```

---

## 3. Hidden → Output

```
Why
```

Connects

```
Hidden

↓

Output
```

Shape depends on the output size.

---

# Hidden State Equation

This is the most important equation in an RNN.

```
hₜ = tanh(Wxh·xₜ + Whh·hₜ₋₁ + b)
```

Where:

- **xₜ** = current input vector
- **hₜ₋₁** = previous hidden state
- **Wxh** = input-to-hidden weight matrix
- **Whh** = hidden-to-hidden (recurrent) weight matrix
- **b** = bias vector
- **tanh** = activation function

This equation means:

```
Current Input

+

Previous Memory

↓

Linear Transformation

↓

Activation

↓

New Memory
```

---

# Output Equation

After computing the hidden state,

the output is

```
yₜ = softmax(Why·hₜ + c)
```

For binary classification, the output layer often uses a **sigmoid** activation instead of softmax.

---

# Numerical Example (Simplified)

Suppose

```
Input Vector

x₁

=

[1
0
0]
```

Previous Hidden

```
h₀

=

[0
0
0
0]
```

The RNN computes

```
Hidden State

=

tanh(

Wxh × x₁

+

Whh × h₀

)
```

The resulting hidden state might be

```
h₁

=

[0.41
0.20
0.82
0.35]
```

For the next word,

```
love
```

the model computes

```
h₂

=

tanh(

Wxh × x₂

+

Whh × h₁

)
```

Notice that **h₁** carries information from the previous word.

---

# Why Weight Sharing Matters

Instead of learning separate weights for every word,

RNN uses the same matrices:

```
Wxh

Whh

Why
```

for every time step.

Benefits:

- Fewer parameters
- Can process variable-length sequences
- Better generalization

---

# Backpropagation Through Time (BPTT)

Training an RNN uses a variation of backpropagation called **Backpropagation Through Time (BPTT)**.

```
Forward Pass

↓

Compute Loss

↓

Unroll Through Time

↓

Propagate Errors Backward

↓

Update Shared Weights
```

Because the same weights are reused across time, gradients from every time step contribute to updating them.

# Limitations of Basic RNN

- Vanishing gradients
- Exploding gradients
- Difficulty remembering long-term dependencies
- Sequential computation limits parallelism

These limitations motivated the development of:

- LSTM
- GRU

Later, **Transformers** became the dominant architecture for many NLP tasks because they model long-range relationships more effectively and allow parallel processing.

# Key Takeaways

- RNN extends ANN by introducing memory through the hidden state.
- Each time step combines the current input with information from previous steps.
- Vectorization represents inputs, hidden states, and weights as vectors and matrices for efficient computation.
- Weight sharing allows the same RNN cell to process sequences of any length.
- Understanding basic RNNs is essential before learning LSTMs, GRUs, and Transformers.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. What is an RNN?

**Answer:**
An RNN (Recurrent Neural Network) is a neural network designed for sequential data that maintains a hidden state (memory) across time steps.

---

### 2. Why is it called a Recurrent Neural Network?

**Answer:**
Because the same network (with the same weights) is applied repeatedly at each time step in a sequence.

---

### 3. What is the hidden state?

**Answer:**
The hidden state is the memory vector that stores information from previous inputs and passes it to future time steps.

---

### 4. What are the three main weight matrices in an RNN?

**Answer:**

* **Wxh**: Input → Hidden
* **Whh**: Hidden → Hidden (recurrent)
* **Why**: Hidden → Output

---

### 5. What activation function is commonly used in the hidden state of a basic RNN?

**Answer:**
The **tanh** activation function is commonly used, although other activations may also be used in some implementations.

---

## Intermediate Questions

### 6. Why is vectorization important in RNNs?

**Answer:**
Vectorization allows inputs, hidden states, and weights to be represented as matrices and vectors, enabling efficient computation using optimized linear algebra operations.

---

### 7. What does the equation `hₜ = tanh(Wxh·xₜ + Whh·hₜ₋₁ + b)` represent?

**Answer:**
It computes the current hidden state by combining the current input, the previous hidden state, and a bias, followed by the tanh activation.

---

### 8. Why does an RNN share weights across time steps?

**Answer:**
Weight sharing reduces the number of parameters, improves generalization, and allows the same model to process sequences of varying lengths.

---

### 9. What is Backpropagation Through Time (BPTT)?

**Answer:**
BPTT is the training algorithm for RNNs that computes gradients by unrolling the network across time steps and propagating errors backward through the sequence.

---

### 10. Why are word embeddings preferred over one-hot vectors in modern NLP?

**Answer:**
Embeddings are dense, lower-dimensional representations that capture semantic relationships between words, unlike sparse one-hot vectors.

---

## Scenario-Based Questions

### Q1. You're building a next-word prediction model. Why does the hidden state matter?

**Answer:**
The hidden state carries information from previous words, enabling the model to use context when predicting the next word.

---

### Q2. Your RNN performs poorly on long paragraphs. What could be the reason?

**Answer:**
Basic RNNs struggle with long-term dependencies because of vanishing gradients. LSTMs, GRUs, or Transformers are generally better choices for long sequences.

---

### Q3. Why are the same weights reused for every word in a sentence?

**Answer:**
Using shared weights allows the model to apply the same learned transformation to every time step, making it efficient and capable of handling variable-length sequences.
