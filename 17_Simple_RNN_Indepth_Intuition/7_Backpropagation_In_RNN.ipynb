# Backpropagation in RNN (Backpropagation Through Time - BPTT)

# Introduction

In the previous notebook, we learned how an RNN performs **Forward Propagation** by processing one input at a time and updating its hidden state.

However, after making predictions, an important question arises:

> **How does the RNN learn from its mistakes?**

The answer is **Backpropagation Through Time (BPTT).**

Just like an Artificial Neural Network (ANN) uses **Backpropagation** to update its weights, an RNN also updates its weights.

The difference is:

- **ANN → Backpropagation through Layers**
- **RNN → Backpropagation through Time**

---

# Recap: Backpropagation in ANN

Suppose we are predicting whether a student gets placed.

Features

```
IQ | CGPA | Qualification
```

↓

ANN

↓

Prediction

```
Placed
```

Suppose

```
Actual Output = 1
```

Prediction

```
0.75
```

Loss

```
Loss = Actual − Prediction

= 1 − 0.75

= 0.25
```

The network then:

```
Compute Loss

↓

Backpropagate Error

↓

Update Weights

↓

Better Prediction
```

The error flows **backward layer by layer**.

```
Output Layer

↓

Hidden Layer

↓

Input Layer
```

---

# Why Can't Normal Backpropagation Be Used in RNN?

An RNN is different because every prediction depends on previous hidden states.

Example

```
Namaste

↓

Mera

↓

Name

↓

Arun

↓

Hai
```

Each word affects the next hidden state.

Therefore,

the error must travel not only through layers,

but also through **time**.

---

# What is Backpropagation Through Time (BPTT)?

## Definition

**Backpropagation Through Time (BPTT)** is the training algorithm used for RNNs.

It computes gradients by **unrolling the RNN across time steps** and propagating errors backward from the last time step to the first.

Instead of only moving backward through layers,

it also moves backward through **time**.

---

# Why is it Called "Through Time"?

Suppose

```
x₁

↓

x₂

↓

x₃

↓

x₄
```

During forward propagation,

information moves

```
Time 1

↓

Time 2

↓

Time 3

↓

Time 4
```

During training,

the error moves

```
Time 4

↑

Time 3

↑

Time 2

↑

Time 1
```

Hence the name

```
Backpropagation

Through

Time
```

# Example Dataset

Suppose we want to perform **Sentiment Analysis**.

| Text | Sentiment |
|------|----------|
| cat mat rat | Positive (1) |
| mat mat cat | Negative (0) |

---

# Step 1: Convert Words into Vectors

Vocabulary

```
cat

mat

rat
```

One-Hot Encoding

```
cat

[1 0 0]

mat

[0 1 0]

rat

[0 0 1]
```

Example input

```
cat mat rat
```

becomes

```
x₁ = [1 0 0]

x₂ = [0 1 0]

x₃ = [0 0 1]
```

---

# Step 2: Forward Propagation

The RNN is unfolded across time.

```code
         h₀
          │
          ▼
x₁ ─► [RNN] ─► h₁
          │
          ▼
x₂ ─► [RNN] ─► h₂
          │
          ▼
x₃ ─► [RNN] ─► h₃
          │
          ▼
      Prediction
```

The hidden states are computed as:

```
h₁ = f(Wxh·x₁ + Whh·h₀)
```

```
h₂ = f(Wxh·x₂ + Whh·h₁)
```

```
h₃ = f(Wxh·x₃ + Whh·h₂)
```

where:

- **Wxh** → Input → Hidden weights
- **Whh** → Hidden → Hidden weights
- **f** → Activation function (usually tanh)

Notice that **h₂ depends on h₁**, and **h₃ depends on h₂**.

---

# Step 3: Compute Prediction

For a sentiment classification task,

the final hidden state is passed to the output layer.

```
h₃

↓

Output Layer

↓

ŷ
```

Example

```
Actual Sentiment

1
```

Prediction

```
0.82
```

---

# Step 4: Calculate Loss

The loss measures how different the prediction is from the actual value.

For binary classification, we typically use **Binary Cross-Entropy (BCE)** instead of a simple subtraction.

The BCE loss for one sample is:

\[
$L = -\left[y \log(\hat{y}) + (1-y)\log(1-\hat{y})\right]$
\]

Where:

- **y** = Actual label
- **ŷ** = Predicted probability

If

```
Actual = 1

Prediction = 0.82
```

the loss will be small.

If

```
Prediction = 0.15
```

the loss will be much larger.

> The expression `Loss = y_actual - y_prediction` is useful only for understanding the idea of prediction error. During training, deep learning models use differentiable loss functions such as Binary Cross-Entropy or Categorical Cross-Entropy.

---

# Step 5: Backpropagation Through Time

Now,

the error starts moving backward.

```
Loss

↑

Output Layer

↑

h₃

↑

h₂

↑

h₁
```

Unlike ANN,

the gradients are propagated across every previous time step.

---

# Step 6: Updating the Shared Weights

The RNN has three shared weight matrices.

```
Wxh

Whh

Why
```

Since the same weights are used at every time step,

their gradients are accumulated from **all time steps**.

Example

```
Gradient at t₁

+

Gradient at t₂

+

Gradient at t₃

↓

Total Gradient

↓

Update Wxh
```

The same process updates

```
Whh

Why
```

This is why RNN learning depends on the entire sequence.

---

# Complete BPTT Workflow

```code
Input Sequence

↓

Forward Propagation

↓

Prediction

↓

Loss Calculation

↓

Unroll Network

↓

Backpropagate Through Time

↓

Compute Gradients

↓

Update Shared Weights

↓

Repeat
```

---

# Why Does BPTT Become Difficult?

Imagine a sentence with 200 words.

```
Word₁

↓

Word₂

↓

...

↓

Word₂₀₀
```

During BPTT,

the gradient must travel all the way back.

Problems arise:

- Vanishing Gradient
- Exploding Gradient

---

# Vanishing Gradient

Suppose the gradient becomes smaller at every step.

```
1

↓

0.5

↓

0.25

↓

0.12

↓

0.06

↓

0.03
```

Eventually,

it becomes almost zero.

The network forgets early information.

---

# Exploding Gradient

Sometimes the gradient becomes larger.

```
1

↓

3

↓

9

↓

27

↓

81
```

The updates become unstable,

making training difficult.

---

# Why Were LSTMs Introduced?

Because basic RNNs struggle with long sequences.

LSTM introduces **gates** that help preserve important information and reduce the vanishing gradient problem.

GRUs solve a similar problem using a simpler gating mechanism.

---

# ANN vs RNN Backpropagation

| ANN | RNN |
|------|------|
| Backpropagation through Layers | Backpropagation through Time |
| Error flows layer by layer | Error flows across time steps |
| Independent inputs | Sequential inputs |
| Standard Backpropagation | BPTT |

# Interview Points

- RNNs are trained using **Backpropagation Through Time (BPTT)**.
- BPTT unrolls the network across time and propagates gradients backward through the sequence.
- The same weight matrices are reused at every time step, so gradients from all time steps are accumulated before updating the weights.
- Binary Cross-Entropy is commonly used for binary classification tasks.
- Vanishing and exploding gradients are major challenges in training basic RNNs.
- LSTMs and GRUs were developed to mitigate these problems.

---

# Key Takeaways

- BPTT is the learning algorithm for RNNs.
- Unlike ANN, the error propagates backward through both layers and time.
- Shared weights receive gradient contributions from every time step.
- Long sequences make gradient propagation difficult, leading to vanishing and exploding gradients.
- Understanding BPTT is essential before studying LSTM, GRU, and gradient clipping.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. What is BPTT?

**Answer:**
BPTT (Backpropagation Through Time) is the training algorithm used for Recurrent Neural Networks. It computes gradients by propagating errors backward through the unrolled sequence.

---

### 2. Why is it called "Through Time"?

**Answer:**
Because the error is propagated backward across time steps rather than only across network layers.

---

### 3. How is BPTT different from standard backpropagation?

**Answer:**
Standard backpropagation propagates errors through layers, while BPTT propagates errors through both layers and time steps in an unrolled RNN.

---

### 4. Which loss function is commonly used for binary sentiment classification?

**Answer:**
Binary Cross-Entropy (BCE).

---

### 5. Why are the same weights updated at every time step?

**Answer:**
Because RNNs use shared weights (Wxh, Whh, Why) across all time steps, and gradients from each time step are combined before updating them.

---

## Intermediate Questions

### 6. What is an unrolled RNN?

**Answer:**
An unrolled RNN is a visualization where the recurrent cell is expanded across time steps to show how information and gradients flow through the sequence.

---

### 7. What causes the vanishing gradient problem?

**Answer:**
Repeated multiplication of small derivatives during backpropagation causes gradients to shrink toward zero, making it difficult to learn long-term dependencies.

---

### 8. What causes the exploding gradient problem?

**Answer:**
Repeated multiplication of large derivatives causes gradients to grow rapidly, leading to unstable updates.

---

### 9. How do LSTMs improve upon basic RNNs?

**Answer:**
LSTMs use gating mechanisms to selectively retain or forget information, making it easier to learn long-term dependencies.

---

### 10. Why is BPTT computationally expensive?

**Answer:**
Because the network must be unrolled across all time steps, requiring additional memory and computation to store hidden states and compute gradients.
