# Introduction

Forward propagation is the process by which an RNN processes an input sequence from **left to right**, updating its memory (hidden state) and producing outputs.

Unlike an ANN, which processes one input independently, an RNN remembers previous information using its **hidden state**.

The forward pass answers the question:

> **How does an RNN convert a sequence of words into meaningful predictions?**

---

# Example Sentence

We will use the sentence:

```
Namaste Mera Name Arun Hai
```

The sequence is:

```
x₁ = Namaste
x₂ = Mera
x₃ = Name
x₄ = Arun
x₅ = Hai
```

Each word is processed one after another.

---

# Step 1: Convert Words into Vectors

Neural networks cannot understand text.

Therefore, every word is converted into a vector.

Example Vocabulary

| Word | Index |
|-------|------:|
| Namaste | 0 |
| Mera | 1 |
| Name | 2 |
| Arun | 3 |
| Hai | 4 |

### One-Hot Encoding

```
Namaste

[1 0 0 0 0]

Mera

[0 1 0 0 0]

Name

[0 0 1 0 0]

Arun

[0 0 0 1 0]

Hai

[0 0 0 0 1]
```

> **Note:** In modern NLP, these one-hot vectors are usually converted into **word embeddings** before being fed into the RNN.

---

# Step 2: RNN Architecture

```
             h₀ (Initial Memory)
                    │
                    ▼
x₁ ─► [RNN Cell] ─► h₁ ─► y₁
                    │
                    ▼
x₂ ─► [RNN Cell] ─► h₂ ─► y₂
                    │
                    ▼
x₃ ─► [RNN Cell] ─► h₃ ─► y₃
                    │
                    ▼
x₄ ─► [RNN Cell] ─► h₄ ─► y₄
                    │
                    ▼
x₅ ─► [RNN Cell] ─► h₅ ─► y₅
```

Notice:

- Same RNN Cell
- Same Weights
- Memory flows forward

---

# Step 3: Weight Matrices

The RNN learns three weight matrices.

```
Wxh

Input

↓

Hidden
```

```
Whh

Previous Hidden

↓

Current Hidden
```

```
Why

Hidden

↓

Output
```

These weights are **shared** across every time step.

---

# Step 4: Initial Hidden State

Initially,

the RNN has no previous memory.

Therefore,

```
h₀ = [0 0 0 0]
```

This is called the **initial hidden state**.

---

# Step 5: Forward Propagation Equation

The most important equation in an RNN is the hidden state update:

\[
$h_t = \tanh(W_{xh}x_t + W_{hh}h_{t-1} + b)$
\]

Here:

- **xₜ** = Current word vector
- **hₜ₋₁** = Previous hidden state (memory)
- **Wxh** = Input → Hidden weights
- **Whh** = Hidden → Hidden weights
- **b** = Bias
- **tanh** = Activation function

### Intuition

```
Current Word

+

Previous Memory

↓

Weighted Sum

↓

tanh

↓

New Memory
```

---

# Step 6: Output Equation

Once the hidden state is computed, the output is calculated:

\[
$y_t = \text{softmax}(W_{hy}h_t + c)$
\]

Where:

- **Why (or Why)** = Hidden → Output weights
- **c** = Output bias
- **Softmax** = Produces probabilities across all possible output classes

For binary classification, a **sigmoid** activation is often used instead of softmax.

## Time Step 1

### Input

```
Namaste
```

```
x₁

↓

RNN
```

Previous Hidden

```
h₀ = 0
```

The RNN computes

```
h₁ = tanh( Wxh·x₁ + Whh·h₀ + b )
```

Since

```
h₀ = 0
```

only the input contributes initially.

Suppose

```
h₁

=

[0.21 0.63 0.14 0.88]
```

Memory after first word:

```
Namaste
```

---

## Time Step 2

Current Input

```
Mera
```

Previous Memory

```
Namaste
```

Now

```
h₂ = tanh( Wxh·x₂ + Whh·h₁ + b )
```

The hidden state now contains information from:

```
Namaste

+

Mera
```

Suppose

```
h₂ = [0.45 0.78 0.32 0.61]
```

---

## Time Step 3

Input

```
Name
```

Memory

```
Namaste

Mera
```

Computation

```
h₃ = tanh( Wxh·x₃ + Whh·h₂ + b )
```

Now memory contains

```
Namaste

↓

Mera

↓

Name
```

---

## Time Step 4

Input

```
Arun
```

Computation

```
h₄ = tanh( Wxh·x₄ + Whh·h₃ + b )
```

Memory becomes

```
Namaste

↓

Mera

↓

Name

↓

Arun
```

---

## Time Step 5

Input

```
Hai
```

Computation

```
h₅ = tanh( Wxh·x₅ + Whh·h₄ + b )
```

Now the final hidden state contains contextual information from the **entire sentence**:

```
Namaste

↓

Mera

↓

Name

↓

Arun

↓

Hai
```

This final hidden state can be used for tasks such as sentence classification, translation, or text generation.

# Why is the Hidden State Important?

Suppose the model sees

```
Arun
```

Without memory,

it only knows

```
Arun
```

With hidden state,

it knows

```
Namaste

↓

Mera

↓

Name

↓

Arun
```

The prediction becomes much more meaningful because it includes context.

# Complete Architecture

```code
                 h₀
                  │
                  ▼
Namaste ─► [RNN] ─► h₁ ─► y₁
                  │
                  ▼
Mera     ─► [RNN] ─► h₂ ─► y₂
                  │
                  ▼
Name     ─► [RNN] ─► h₃ ─► y₃
                  │
                  ▼
Arun     ─► [RNN] ─► h₄ ─► y₄
                  │
                  ▼
Hai      ─► [RNN] ─► h₅ ─► y₅
```

The same RNN cell is reused for every word.

---

# Why Weight Sharing?

Instead of learning different weights for every word, the RNN reuses:

- **Wxh**
- **Whh**
- **Why**

Benefits:

- Fewer parameters
- Better generalization
- Ability to process variable-length sequences

---

# Interview Points

- Forward propagation processes the sequence from left to right.
- Every hidden state depends on the current input and the previous hidden state.
- The hidden state acts as the network's memory.
- The same weight matrices are reused at every time step.
- The final hidden state contains contextual information from the complete sequence.
- During training, this forward pass is followed by **Backpropagation Through Time (BPTT)** to update the shared weights.

---

# Key Takeaways

- Forward propagation is the process of computing hidden states and outputs sequentially.
- Each new hidden state combines the current word with previously learned context.
- Weight sharing makes RNNs efficient and capable of handling sequences of varying lengths.
- The hidden state is the core mechanism that allows RNNs to remember past information.
- Understanding the forward pass is essential before learning BPTT, LSTM, and GRU.

# 2. Interview Questions & Answers

### 1. What happens during forward propagation in an RNN?

**Answer:**
The RNN processes one input at a time, combines it with the previous hidden state, computes a new hidden state, generates an output, and passes the hidden state to the next time step.

---

### 2. Why is the hidden state important?

**Answer:**
The hidden state acts as memory, carrying contextual information from previous inputs so the model can make better predictions.

---

### 3. What is the hidden state equation?

**Answer:**

[
$h_t = \tanh(W_{xh}x_t + W_{hh}h_{t-1} + b)$
]

It combines the current input with the previous hidden state to create the new memory.

---

### 4. Why are the same weights used at every time step?

**Answer:**
Weight sharing reduces the number of trainable parameters, improves generalization, and allows the model to process sequences of arbitrary length.

---

### 5. What is the purpose of the softmax layer?

**Answer:**
Softmax converts the output scores into a probability distribution over all possible output classes.

# 3. 30-Second Revision

* Key equations:

  * $(h_t = \tanh(W_{xh}x_t + W_{hh}h_{t-1} + b))$
  * $(y_t = \text{softmax}(W_{hy}h_t + c))$
* The hidden state is passed to the next time step.
* The final hidden state summarizes the entire sequence.

---

# 4. 2-Minute Revision

* RNN forward propagation begins with an initial hidden state, usually initialized to zeros.
* Each input word is converted into a vector and combined with the previous hidden state using shared weight matrices.
* The resulting hidden state captures all information seen so far in the sequence.
* An output is generated from the hidden state at each time step.
* After the forward pass, the network is trained using Backpropagation Through Time (BPTT) to improve the shared weights.
