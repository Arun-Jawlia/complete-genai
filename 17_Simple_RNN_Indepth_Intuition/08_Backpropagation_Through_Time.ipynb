# `Backpropagation Through Time`

# Introduction

We already know that an RNN performs **Forward Propagation** by processing one input at a time and storing information in the hidden state.

But after making predictions, a question arises:

> **How does an RNN learn from its mistakes?**

The answer is **Backpropagation Through Time (BPTT).**

BPTT is simply the **backpropagation algorithm specially designed for sequential neural networks (RNNs).**

---

# What is Backpropagation Through Time?

## Definition

Backpropagation Through Time (BPTT) is the training algorithm used for Recurrent Neural Networks.

Instead of propagating the error only through network layers, it propagates the error **backward through both layers and time steps**.

---

# Why "Through Time"?

In ANN

```
Input

↓

Hidden

↓

Output
```

Error moves

```
Output

↑

Hidden

↑

Input
```

Only through layers.

---

In RNN

```
Time 1

↓

Time 2

↓

Time 3

↓

Time 4
```

During training

```
Time 4

↑

Time 3

↑

Time 2

↑

Time 1
```

The error travels backward through time.

Hence,

**Backpropagation Through Time (BPTT).**

# Step 1 : Unroll the RNN

Although an RNN is a single recurrent cell,

during training it is expanded across time.

```
Time1      Time2      Time3

x1 ─► RNN ─► RNN ─► RNN

      h1      h2      h3
```

Notice

These are **not different networks.**

It is the **same network** repeated over time.

Each hidden state is computed as

```
h1 = tanh(Wxh·x1 + Whh·h0)
```

```
h2 = tanh(Wxh·x2 + Whh·h1)
```

```
h3 = tanh(Wxh·x3 + Whh·h2)
```

where

- Wxh → Input → Hidden
- Whh → Hidden → Hidden

These weights are shared across all time steps.

Now

```
Loss

↑

Output Layer

↑

h3

↑

h2

↑

h1
```

Notice

# Step 6 : Chain Rule

Suppose

```
Whh

↓

h1

↓

h2

↓

h3

↓

Loss
```

The gradient becomes

```
∂L
───
∂Whh
```

```
∂L     ∂h3     ∂h2     ∂h1
── × ──── × ──── × ─────
∂h3    ∂h2     ∂h1    ∂Whh
```

This is the **Chain Rule**.

Every previous hidden state contributes to the gradient.

```
Whh
```

```
Time1

Time2

Time3
```

```
Gradient(Time1)

+

Gradient(Time2)

+

Gradient(Time3)
```

# Step 8 : Gradient Descent

Once gradients are computed

the weights are updated.

```
Wnew = Wold − η × Gradient
```

For example

```
Whh = Whh − η × ∂L/∂Whh
```

Similarly

```
Wxh = Wxh − η × ∂L/∂Wxh
```

```
Why = Why − η × ∂L/∂Why
```

# Interview Points

- BPTT is the training algorithm for RNNs.
- It unrolls the RNN across time.
- Error flows backward through all previous time steps.
- Shared weights receive gradient contributions from every time step.
- Chain Rule is used to compute gradients.
- Gradient Descent updates Wxh, Whh and Why.
- Vanishing and Exploding gradients are major limitations of BPTT.

---

# Key Takeaways

- BPTT is an extension of standard backpropagation for sequence models.
- The RNN is unrolled across time before gradients are computed.
- The same recurrent weights are reused and updated using accumulated gradients.
- BPTT enables RNNs to learn sequential dependencies but becomes inefficient for very long sequences.
- LSTM and GRU were introduced to overcome the limitations of BPTT in basic RNNs.

# 2. Interview Questions & Answers

### 1. What is Backpropagation Through Time (BPTT)?

**Answer:**
BPTT is the algorithm used to train Recurrent Neural Networks. It computes gradients by unrolling the RNN across time and propagating errors backward through each time step.

---

### 2. Why is BPTT needed instead of normal backpropagation?

**Answer:**
Because each hidden state in an RNN depends on previous hidden states. Therefore, gradients must flow backward through time, not just through layers.

---

### 3. Why are gradients summed across time steps?

**Answer:**
The same recurrent weight matrices (such as **Whh**) are reused at every time step. Each usage contributes to the final loss, so their gradient contributions are accumulated.

---

### 4. What is the role of the chain rule in BPTT?

**Answer:**
The chain rule computes how changes in earlier hidden states and weights influence the final loss by multiplying derivatives along the computational path.

---

### 5. Why does BPTT suffer from vanishing gradients?

**Answer:**
Repeated multiplication of small derivatives (typically less than 1) causes gradients to shrink exponentially as they move backward through long sequences.

---

### 6. Why does BPTT suffer from exploding gradients?

**Answer:**
Repeated multiplication of large derivatives (greater than 1) causes gradients to grow rapidly, leading to unstable training.

---

### 7. How are the weights updated after BPTT?

**Answer:**
Using Gradient Descent:

[
$W_{\text{new}} = W_{\text{old}} - \eta \frac{\partial L}{\partial W}$
]

where (\eta) is the learning rate.

---

# 3. 30-Second Revision

* BPTT = Backpropagation Through Time.
* The RNN is unrolled across time before training.
* Error flows from the last time step to the first.
* Shared weights (**Wxh**, **Whh**, **Why**) receive gradients from all time steps.
* Gradients are computed using the chain rule.
* Weights are updated using Gradient Descent.
* Main limitations: **vanishing** and **exploding gradients**.

---

# 4. 2-Minute Revision

* During forward propagation, an RNN processes the sequence and computes hidden states.
* The prediction is compared with the target to compute the loss.
* During BPTT, the RNN is unrolled across time, and the error is propagated backward through every hidden state.
* Since the same weights are reused across all time steps, gradients from each time step are accumulated before updating the weights.
* Gradient Descent updates the weights using the accumulated gradients.
* Long sequences make BPTT computationally expensive and can lead to vanishing or exploding gradients, motivating architectures like LSTM and GRU.

## 💡 Interview Insight

The core mathematical idea behind BPTT is **not** that gradients simply move backward through time. The key idea is:

* The **same recurrent weight matrix (`W_{hh}`)** is reused at every time step.
* Therefore, the **total gradient for that weight is the sum of its contributions from every time step**.
* Those contributions are computed using the **chain rule** across the unrolled computation graph.
* Because the chain rule repeatedly multiplies Jacobians across time, gradients can **shrink toward zero (vanishing gradients)** or **grow without bound (exploding gradients)**. This mathematical behavior is the main reason architectures like **LSTMs** and **GRUs** were developed.
