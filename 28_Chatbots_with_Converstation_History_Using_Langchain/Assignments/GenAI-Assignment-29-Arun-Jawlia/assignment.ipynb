# PART 1 — Q&A Chatbot using OpenAI
## Task 1: OpenAI Setup

```python
from langchain_openai import ChatOpenAI
```

```python
llm_openai = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

print("OpenAI model initialized successfully.")
```

## Task 2: Basic OpenAI Q&A Chatbot

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful general-purpose Q&A assistant. "
        "Answer questions clearly and accurately."
    ),
    ("human", "{question}")
])
```

```python
def ask_openai(question):
    messages = prompt.invoke({
        "question": question
    })

    response = llm_openai.invoke(messages)

    return response.content
```

```python
ask_openai("What is Python?")
```

```python
ask_openai("What is the difference between AI and machine learning?")
```

```python
ask_openai("Explain REST API in simple words.")
```

```python
ask_openai("What is the purpose of Docker")
```

```python
ask_openai("Explain object-oriented programming with an example.")
```

## Task 3: Multi-Turn Q&A — Optional

```python
from langchain_core.messages import HumanMessage, AIMessage
```

```python
chat_history = []
```

```python
def chat_openai(question):
    messages = [
        (
            "system",
            "You are a helpful general-purpose Q&A assistant. "
            "Answer questions clearly and accurately."
        )
    ]

    messages.extend(chat_history)
    messages.append(HumanMessage(content=question))

    response = llm_openai.invoke(messages)
    chat_history.append(HumanMessage(content=question))
    chat_history.append(AIMessage(content=response.content))

    return response.content
```

```python
chat_openai("What is Python?")
```

```python
chat_openai("What are its main advantages?")
```

```python
chat_openai("Give me a simple example.")
```

## Task 6: Compare OpenAI vs Ollama

```python
question = "Explain what an API is in simple words."
```

```python
print(ask_openai(question))
```

```python
print(ask_ollama(question))
```

1. Response Quality
OpenAI:
- Generally produces strong responses across many general-purpose tasks.
- Usually follows instructions consistently.
- Quality depends on the specific OpenAI model being used.
Ollama:
- Quality depends heavily on the local model.
- llama3 can handle general Q&A effectively.
- Smaller local models may provide less detailed or less accurate answers for difficult questions.

2. Latency

OpenAI: User → Internet → OpenAI Server → Response

Latency depends on:
- Internet connection
- API/server response time
- Model being used

Ollama: User → Local Computer → Model → Response

Latency depends mainly on:
- CPU/GPU
- RAM
- Model size
- Quantization

3. Cost
- OpenAI: API usage is generally billed according to the model's pricing and token usage.
- Ollama: The software itself can run models locally without per-token API charges, but you still incur the cost of your own hardware and electricity.

4. Privacy
- OpenAI: The question is sent to the OpenAI API for processing, so your application's data handling depends on the API/service configuration and applicable policies.
- Ollama: When running a model locally, prompts and responses can remain on your machine rather than being sent to an external model API.

# Task 7: Model Switch Logic

```python
def get_answer(question, model_type="openai"):
    
    messages = prompt.invoke({
        "question": question
    })

    if model_type == "openai":
        response = llm_openai.invoke(messages)

    elif model_type == "ollama":
        response = llm_ollama.invoke(messages)

    else:
        return "Invalid model type. Choose 'openai' or 'ollama'."

    return response.content
```

```python
get_answer("What is Python?", "openai")
```

```python
get_answer("What is Python?", "ollama")
```

# Task 8: Build Simple App — CLI

```python
model_type = input("Choose model (openai/ollama): ").lower()

if model_type not in ["openai", "ollama"]:
    print("Invalid model. Please choose openai or ollama.")
else:
    while True:
        question = input("\nYou: ")

        if question.lower() == "exit":
            print("Goodbye!")
            break

        answer = get_answer(question, model_type)

        print("\nAssistant:", answer)
```

## Task 9: Conceptual Questions

1. When to prefer OpenAI models?
- OpenAI models are useful when you want a managed API, strong general-purpose performance, and don't want to manage model infrastructure yourself.

2. When to prefer open-source models?
- Open-source/local models are useful when you need local inference, greater control over the model, or want to keep prompts and responses within your own infrastructure.

3. Trade-offs in production systems
OpenAI:
- Managed infrastructure
- API dependency
- Usage-based cost
- Internet connection required
Ollama/local model:
- More infrastructure responsibility
- Hardware requirements
- Local deployment
- No per-request model API charge
The actual choice depends on requirements such as latency, privacy, infrastructure, model capability, and operating cost.

4. Cost and scalability consi
- OpenAI: Cost generally increases with API usage and token consumption. Scaling primarily involves handling API traffic and controlling usage.
- Ollama: There is no per-token API bill for local inference, but you need to provide CPU/GPU, RAM, storage, and electricity. Scaling to many concurrent users may require additional machines or GPU infrastructure.
