```python
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import AIMessage, SystemMessage, HumanMessage
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain_community.document_loaders import WebBaseLoader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
```

```python
from langchain_classic.chains import create_retrieval_chain   
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
```

```python
load_dotenv()
```

```python
llm = ChatOpenAI(model='gpt-4o')
```

```python
loader= WebBaseLoader('https://www.amazon.in/iPhone-Pro-Max-256-Promotion/dp/B0FQFV2XKS')
```

```python
docs = loader.load()
```

```python
docs
```

```python
docs[0].page_content
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
```

```python
splits = text_splitter.split_documents(docs)
```

```python
splits
```

```python
len(splits)
```

```python
vector= Chroma.from_documents(documents=splits, embedding=OpenAIEmbeddings())
```

```python
retriever = vector.as_retriever()
```

```python
retriever
```

```python
system_data = '''
You are a Helpful Questin Answering System. 
Answer the query from the provided context.
If you dont know the anser directly say Thank you and I dont know.
Keep the anwer concise. '{context}'
'''
```

```python
prompt = ChatPromptTemplate.from_messages([
    ('system', system_data),
    ('human', '{input}')
])
```

```python
question_answer_chain = create_stuff_documents_chain(llm, prompt)
```

```python
rag_chain = create_retrieval_chain(retriever, question_answer_chain)
```

```python
response = rag_chain.invoke({'input':'What is the price of iphone 17 Pro Max ?'})
```

```python
from langchain_classic.chains import create_history_aware_retriever
from langchain_core.prompts import MessagesPlaceholder
```

```python
context_system = '''Given a chat History and the latest user question,
 which might reference context in chat history the user and dont answer without looking into the chat history'''
```

```python
context_prompt = ChatPromptTemplate.from_messages([
    ('system', 'Reframe the query as a standalone query based on past context or past history'),
    MessagesPlaceholder('chat_history'),
    ('human', '{input}')
])
```

```python
qa_prompt = ChatPromptTemplate.from_messages([
    ('system', context_system),
    MessagesPlaceholder('chat_history'),
    ('human', '{input} \n\n Context: \n {context}')
])
```

```python
history_aware_retriever =create_history_aware_retriever(llm, retriever, context_prompt)
```

```python
question_answer_chain = create_stuff_documents_chain(llm, qa_prompt)
```

```python
rag_chain = create_retrieval_chain(history_aware_retriever, question_answer_chain)
```

```python
chat_history=[]
```

```python
question = 'What is the price of iphone 17 Pro Max ?'
```

```python
response = rag_chain.invoke({'input':question, "chat_history": chat_history})
```

```python
from langchain_core.messages import HumanMessage, AIMessage
```

```python
chat_history.extend([
    HumanMessage(content=question),
    AIMessage(content=response['answer'])
    ]
)
```
