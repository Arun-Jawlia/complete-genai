# Keywords

## Keywords in Python are reserved words that have predefined meanings and cannot be used as identifiers (like variable names, function names, etc.).
- False, True, None,
- and, or, not,
- if, elif, else,
- for, while, break, continue,
- def, return, lambda,
- class,
- try, except, finally, raise,
- import, from, as,
- with,
- pass, yield,
- global, nonlocal,
- assert, del, in, is
