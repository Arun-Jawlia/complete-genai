# Variable Naming Convention
- must be descriptive
- must be start with letter or underscore ( _ ) and can contain letter , number and underscore ( _name, _Age, Address, address, a1234bcde )
- case sensitve ( name is not equal Name) name != Name
  

```python
# Check what type of variable
age = 27
is_Boolean = True
is_float = 0.6
is_string = 'abvljkdg'

print("type of age" ,type(age))
print("type of is_true", type(is_Boolean))
print('type of float', type(is_float))
print("type of String", type(is_string))
```

# input()
input() function can take the input from the user side and then can store that input into a variable.

```python
name = input('Enter your name')
print("Your Name",name)
```

### input function only takes the input in a string format until and unless we are not going to explicitly tkae the input in integer or other format.

```python
age = float(input('Enter your age: '))
print(age, type(age))
```

```python
value = bool(input("Enter a value 0 or 1: "))
print(value, type(value))
```

```python
## Building a Simple Calculator

value_1 = int(input("Enter your first value:"))
value_2 = int(input("Enter your second value:"))

print(f"Addtion of {value_1} and {value_2} ", value_1 + value_2)
print(f"Substract of {value_1} and {value_2} ", value_1 - value_2)
print(f"Multiply of {value_1} and {value_2} ", value_1 * value_2)
print(f"Division of {value_1} and {value_2} ", value_1 / value_2)
```
