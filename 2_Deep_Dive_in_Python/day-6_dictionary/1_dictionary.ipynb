# Dictionary

### A dictionary (dict) is a built-in data structure that stores data in key–value pairs (format).

"arun" : "123456789"
"keshav" : "987654321"

A dictionary is:

- Mutable (can be changed)
- Unordered (before Python 3.7), insertion-ordered (Python 3.7+)
- Stores unique keys
- Keys must be immutable (e.g., int, str, tuple)

### Syntax for Dictionary
```
{
    "key_1": value_1,
    "key_2": value_2,
    "key_3": value_3,
}
```

## Dictionary Method
1. keys()
2. values()
3. items()
4. get()
5. update()
6. setDefault()
7. pop() and popitem()
8. copy()
9. fromkeys()

````python
# demo dictionary
dict3 = {
    'name': 'Arun',
    'fruit': "Apple",
    'phone': '987456321',
    'email': 'arun@gmail.com',
    "list_items":[1,2,3,4,5],
    "float":8.9,
    "Boolean": True
    }

dict3
````

````python
# keys() --> method in dictionary which return all the keys present inside our dictionary
print(dict3.keys())
print(type(dict3.keys))

# values --> method in dictionary which return all the values present inside our dictionary
print(dict3.values())
print(type(dict3.values))

# items() --> method in dictionary which return all the key value pairs in list format
print(dict3.items())
print(type(dict3.items))
````

````python
# get() --> get the value
print(dict3.get("name"))
print(dict3.get("fruit"))

#There is another method to access
print(dict3["name"])

print(dict3)
# update() --> to add new key-value in dictionary
dict3.update({"new_key": "new value"})
print(dict3)


# dictionary["name"]      # KeyError if not exist
# dictionary.get("name")  # Returns None

# print(dict3["hello"]) # KeyError: 'hello'
print(dict3.get("hello")) # None
````

````python
demo_dict = {"name":"Arun", "age": 26}
demo_dict2 = demo_dict

demo_dict["name"] = "Keshav Jawlia"
print(demo_dict)
print(demo_dict2)

# If I change in One Dictionary, then it will reflect the changes in another one thats why we use shallo copy
````

````python
'''
Shallow copy ise being used for creating a copy of the original dictionary in a way such that the changes we are going tho make with the 
original dictionary won't be getting reflected inot our copied dictioanry
'''
demo_dict = {"name":"Arun", "age": 26}
demo_dict2 = demo_dict.copy()

demo_dict["name"] = "Keshav Jawlia"
print(demo_dict)
print(demo_dict2)
````

## Nested Dictionary - dictionary inside dictionary

### I want to store the information for multiple students like their name, marks and grade

````python
nested_students_dict = {
    "student1":{
        "name":"Arun",
        "marks": 93,
        "grade": "C+" 
    },
    "student2":{
        "name":"Gourav",
        "marks": 95,
        "grade": "B+" 
    },
    "student3":{
        "name":"keshav",
        "marks": 99,
        "grade": "A+" 
    },
}
````

````python
nested_students_dict
````

````python
# Iterate through the nested dictionary

for outer_key, outer_value in nested_students_dict.items():
    print(outer_key, outer_value)
    for  key_1, value_1 in outer_value.items():
        print(f"Inner key is {key_1} and Inner value is {value_1}")
````

## Dictionary Comprehension
- helps us with using loops statement along with conditional statement inside our dictionary

### Syntax for Dictionary Comprehension
- {express for item in iterable if condition}

````python
squares = {x**2 for x in range(11)}
# {0:0, 1:1, 2:4, 3:9, 4:16}
print(squares)
````

````python
squares = {x:x**2 for x in range(11)}
# {0:0, 1:1, 2:4, 3:9, 4:16}
print(squares)
````

````python
even_squares = {x**2 for x in range(11) if x%2==0}
# {0:0, 1:1, 2:4, 3:9, 4:16}
print(even_squares)
````
