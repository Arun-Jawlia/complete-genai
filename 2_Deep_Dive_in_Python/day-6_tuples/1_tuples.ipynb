# Tuples ( Data Structure )
---

### Tuples are second type of Data Structure which can stored the data in a ordered way just like list but Tuples are not mutable.

- ordered --> items keep their position
- Immutable --> can't be change after creation
- allow duplicates
- can store multiple data types  ---> capable of storing the data of any type

```python
# Way to Creating a 
# Basic tuple
t = (1, 2, 3)

# Mixed data types
t = (1, "hello", 3.5)

# Single element tuple (important!)
t = (5,)   # comma is required
```

```python
a = tuple()
print(type(a))
print(a)
```

```python
tuple_values = (1,2,3,4,5)
print(tuple_values)
print(type(tuple_values))
```

```python
# Tuples are like List only which cna store the data inside it in a ordered way but are going to immutable.

list = [1,2,3,4,5,]
print(list)
list[1]= 1000
print(list)


t = (1,2,3,4,5)
# print(t)
# t[1] = 1000 # TypeError: 'tuple' object does not support item assignment
print(t)
```

```python
## Tuples are going to be ordered and are going to be index
tuple_data = (1,2,3,3,4,4,5,5,5,)
tuple_data[:]
```

```python
tuple_data[3:]
```

```python
tuple_data[::-1]
```

```python
tuple_data[2:6]
```

## Methods in Tuples

### Two method
- count()
- index()

```python
# count() --> Returns how many times a value appears
tuple_data = (1,2,3,3,4,4,5,5,5,)
tuple_data.count(10)
```

```python
# index() --> Returns first occurrence index
tuple_data.index(5)
```

## Packing and Unpacking Of Tuples

```python
# Packing of Tuples
a = 1,2,3,4,15,26,37

print(a)
type(a)
```

```python
a = 1,2,3,"hello", 4.5,True
print(a)
```

```python
# Unpacking of Tuples

# a,b,c,d,e = (1, "Hello", True, 4.5, 90)
a,b,c,d,e = 1, "Hello", True, 4.5, 90

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# Unpacking the value using a*

# a, b, c= (1,20,300, 4,5,6,7,8) #ValueError: too many values to unpack (expected 3, got 8)

first, *mid, last = (1,20,300, 4,5,6,7,8)
print(first)
print(mid)
print(last)
```
