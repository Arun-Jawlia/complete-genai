# `Standard Libraries in Python`
---
- Python standard library is a vast collection of Modules and Packages and come bundled/preinstalled with Python and it provides us with a vast range of functionalities.
These libraries provide ready-made functionality for:

1. file handling
2. operating system interaction
3. networking
4. mathematics
5. data processing
6. dates & time
7. JSON handling
8. multithreading
9.  databases
10. and much more


Definition: A set of modules and packages that are bundled with Python and available without installing anything externally.

### Why Standard Libraries are Important

Without standard libraries:

1. developers would rewrite common functionality repeatedly
2. development would become slow
3. applications would become inconsistent

Python standard libraries:

1. save time
2. reduce code duplication
3. improve reliability
4. provide optimized implementations

### 1. `math` : having functionalities for performing mathematical calculations

```python
from math import *
num = int(input("Enter the number to calculate the square root: "))
print(sqrt(num))

print(sqrt(25))
print(factorial(5))
print(pi)
```

| Function      | Purpose     |
| ------------- | ----------- |
| `sqrt()`      | square root |
| `factorial()` | factorial   |
| `pow()`       | power       |
| `ceil()`      | round up    |
| `floor()`     | round down  |

### 3. `statistics` Module : Used for statistical calculations.

```python
import statistics

data = [1, 2, 3, 4, 5]

print(statistics.mean(data))
print(statistics.median(data))
```

### 3. csv library : most of the time we are going to work with csv files sot that why knowledge with csv library functionality matters.

```python
import csv
```

```python
with open('example.txt', mode = 'w', newline="") as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(["Amit", 27])
```

```python
with open("example.txt", 'r') as file:
    content = csv.reader(file)
    for row in content:
        print(row)
```
