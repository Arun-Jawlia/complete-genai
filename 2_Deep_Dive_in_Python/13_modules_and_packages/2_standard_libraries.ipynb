```python
from math import *
num = int(input("Enter the number to calculate the square root: "))
print(sqrt(num))

print(sqrt(25))
print(factorial(5))
print(pi)
```

## `File and Data Handling Libraries`

### 1. `csv`: most of the time we are going to work with csv files sot that why knowledge with csv library functionality matters.

```python
import csv
```

```python
with open('example.txt', mode = 'w', newline="") as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(["Amit", 27])
```

```python
with open("example.txt", 'r') as file:
    content = csv.reader(file)
    for row in content:
        print(row)
```

### 2. `json`
Used for JSON data handling.

JSON is heavily used in:

- APIs
- web applications
- configuration files

We we get json data, we get in a JSON string format so by using json library we can convert our string data inot dictionary format for serializing our data.

```python
import json

data = {"name": "Arun", "age": 22}

json_data = json.dumps(data)

print(json_data)
print(type(json_data))
```

```python
python_data = json.loads(json_data)

print(python_data)
print(type(python_data))
```

## `Date and Time Libraries`

datetime: Used for
- current date/time
- formatting
- calculations

### 1. `datetime`

```python
from datetime import datetime

now = datetime.now()

print(now)
```

```python
print(now.strftime("%d-%m-%Y"))
```

### 2. `timedelta`

```python
from datetime import timedelta
now = datetime.now()
yesterday_time = now - timedelta(days=1)
print(yesterday_time)
```

### 3. `time` : provide the current time and if you want your interpreter to sleep for seconds you can do it by using time library

```python
import time

print(time.time())
```

```python
## sleep: if you are writing a piece of code and you want your interpreter to sleep for a number of seconds then you can make the use of sleep function

print(time.time())
time.sleep(4)
print(time.time())
```

## `Operating System Library`: Used for interacting with the OS. 

```python
import os 
```

### 1. `getcwd()`: tell us the current working directory/folder

```python
print(os.getcwd())  # get current working directory
print(os.listdir)
```

### 2. `mkdir()` : make directory function which can help us out with creating the directory / folder

```python
os.mkdir('New_folder')
os.mkdir("new_directory")
```

| Function    | Purpose           |

## `Regular Expression`

Regular expression is used to retrieve info based on a specific pattern or if the pattern matches

```python
import re

text = "Hello my name is Arun and my password is arun@123"

pattern = r'\d+'
match = re.search(pattern, text)
print(match)
print(match.group())
```
