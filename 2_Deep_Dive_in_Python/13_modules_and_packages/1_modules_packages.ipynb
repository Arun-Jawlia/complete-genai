# `Modules and Packages`
---

- In Python, modules and packages are used to organize code into reusable and manageable components.
- They are fundamental concepts for writing scalable Python applications.

## `What is Module`

- A Module in Python is simply a file containing python code ( functions, classes or variables ) that can be imported and used in other programs.

- A module is simply a Python file ` (.py) ` containing:
    1. functions
    2. classes
    3. variables
    4. executable code

### Importing a Module

````python
# math module

import math 

print(math.sqrt(16))
print(math.add(10,20))
print(math.pi)
````

### Different Ways to Import 

1. Import Entire Module
   - Avoids naming conflicts.
2. Import Specific Function
   - Shorter syntax.
3. Import with Alias
4. Import Everything  # `Not recommended because it pollutes namespace.`

````python
# 1. Import Entire Module

import math
print(math.sqrt(16))
````

````python
# 2. Import Specific Function

from math import sqrt
print(sqrt(25))
````

````python
# 3. Import with Alias

import numpy as np
````

````python
# 4. Import Everything 

from math import *
````

## `What is Package`
- A package is a collection of related modules organized in directories.
- It helps structure large projects.

A package is a directory containing:
- multiple modules
- subpackages
- `__init__.py`

Packages help organize large projects hierarchically.

### Why Packages are Needed

Imagine a huge application:

- 100+ modules

Without packages:

- impossible organization
- naming conflicts
- difficult imports

Packages solve this.

### Package Structure

```
ecommerce/
│
├── main.py
│
├── products/
│   ├── __init__.py
│   ├── inventory.py
│   └── pricing.py
│
├── users/
│   ├── __init__.py
│   ├── auth.py
│   └── profile.py
│
└── payments/
    ├── __init__.py
    └── gateway.py
```

- mypackage → package
- math_utils.py → module
- `__init__.py` → tells Python this directory is a package

### Understanding `__init__.py`

This file:

- marks folder/directory as package
- can initializes package
- can contain package-level code
- simplify imports

### Difference between Module and Package

| Feature  | Module            | Package                 |
| -------- | ----------------- | ----------------------- |
| Meaning  | Single `.py` file | Collection of modules   |
| Purpose  | Reuse code        | Organize large projects |
| Example  | `math.py`         | `numpy`                 |
| Contains | Functions/classes | Multiple modules        |
