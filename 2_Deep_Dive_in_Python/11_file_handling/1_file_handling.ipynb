# File Handling in Python
---

- File handling in Python is how you create, read, write, and manage files on disk. 
- It allows your programs to store data permanently on a storage device rather than losing it once the program finishes execution.
- It’s a fundamental I/O (input/output) operation used in logging, data processing, configuration management, etc.

- RAM Memory --> Information will be their temporary
- File --> Information will be their Permanently

### Use cases of file in AI/Ml

1. Natural Language Processing (NLP): Text Data --> Text Format File
    - Data Category: Text Data (Sequential)
    - Common Formats: .txt, .csv, .json, .mdStorage 
    - Type: Text Format

2. Machine Learning (Image Classification) : Image Classifiers --> png or jpeg --> Binary format file
    - Example: Cat/Dog Image Classifier
    - Common Formats: .png, .jpeg, .jpg, .bmp
    - Storage Type: Binary Format

3. Deep Learning (Object Detection) : Object Classifiers --> png or jpeg --> Binary format file
    - Example: Multi-Object Classifiers
    - Common Formats: .png, .jpeg, .tiff, .webp
    - Storage Type: Binary Format

### Types of Files
1. Text File --> .txt, .docx, .csv, .json
2. Binary File --> .png, .mp4, .mp3

### File Handling Mode

| Mode  | Meaning                       |
| ----- | ----------------------------- |
| `"r"` | Read (default)                |
| `"w"` | Write (overwrite)             |
| `"a"` | Append                        |
| `"r+"`| read first and then write     |
| `"w+"`| write first and then read     |
| `"x"` | Create (fails if file exists) |
| `"b"` | Binary mode                   |
| `"t"` | Text mode (default)           |



### File Methods

| Method         | Description            |
| -------------- | ---------------------- |
| `read()`       | Read entire file       |
| `readline()`   | Read one line          |
| `readlines()`  | Read all lines as list |
| `write()`      | Write string           |
| `writelines()` | Write list of strings  |
| `seek()`       | Move cursor            |
| `tell()`       | Get cursor position    |
| `close()`      | Close file             |


### 1. Read the content using `read()` from my `example.txt` file in the system

```python
f =  open('example.txt',"r")
content = f.read()
print(content)
f.close()
```

### 2. Write the content using `write()` into the file `example.txt` 
    -  write() --> it overwrite the whole content into the file 
    -  append() --> it adds the data from the end of the file

```python
f = open('example.txt', "w")
f.write("Hello!  this is example of write() into a file ")
f.close()
```

### 3. Appending into the file using `append()`

```python
f = open("example.txt", "a")
f.write("\nThis is append mode text append into the end of the file")
f.close()
```

## `with()` Statement
- Always prefer with → it automatically closes the file (no resource leak).
- used for first of all opening the file as a object and there is not requirement for closing the file as well

```python
with open('example.txt','r') as f:
    content = f.read()
    print(content)
```

```python
with open('example.txt', "w") as file:
    file.write("Hello!  this is example of write() into a file ")
```

```python
with open('example.txt', "a") as file:
    file.write("\nThis is append mode text append into the end of the file")
```
