# Conditional Statements
---

- Conditional statements allow a program to make decisions and execute code based on conditions (True/False).
- They control the flow of execution.

### Why do we use?
- Execute logic based on conditions
- Build decision-making systems
- Handle real-world scenarios

### Basis Syntax
```
if condition:
    # code runs if condition is True
```

````python
# Example - Eligibility for driving on meeting a certain condition

# Conditional statements --> if statements where if a condition is meeting then a block of code gets executed

age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible for driving")
else:
    print("Sorry!, you are not eligible for driving")
````

## Types of Conditional Statements

1. if statement
2. if-else statement
3. if-elif-else statement

````python
# if Statement

num = int(input("Enter your age: "))

if num >= 10:
    print("Num is greater than 10")
````

````python
# if-else Statement

arun_height = int(input("Enter what is your height Arun: "))
keshav_height = int(input("Enter what is your height Keshav: "))

if arun_height > keshav_height:
    print("Arun height is greater than Keshav")
else:
    print("Keshav height is greater than Arun")
````

````python
# if-elif-else Statement ---> Used for multiple conditions
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
````

### Nested Conditional Statement --> if statement inside the another if statement

````python
age = 25
has_license = True

if age >= 18:
  if has_license:
    print("You can drive")
  else:
    print("You need a license")
else:
  print("You are too young to drive")
````

````python
score = 85
attendance = 90
submitted = True

if score >= 60:
  if attendance >= 80:
    if submitted:
      print("Pass with good standing")
    else:
      print("Pass but missing assignment")
  else:
    print("Pass but low attendance")
else:
  print("Fail")
````

````python
username = "Emil"
password = "python123"
is_active = True

if username:
  if password:
    if is_active:
      print("Login successful")
    else:
      print("Account is not active")
  else:
    print("Password required")
else:
  print("Username required")
````

````python
score = 92
extra_credit = 5

if score >= 90:
  if extra_credit > 0:
    print("A+ grade")
  else:
    print("A grade")
elif score >= 80:
  print("B grade")
else:
  print("C grade or below")
````

````python
# Whether if any Year is a leap year or not
year = int(input("Enter any 4 digit year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0: 
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
````

## Logical Operators in Conditions

| Operator | Meaning              |
| -------- | -------------------- |
| `and`    | Both conditions true |
| `or`     | At least one true    |
| `not`    | Reverse condition    |


## Comparison Operators

| Operator | Meaning          |
| -------- | ---------------- |
| `==`     | Equal            |
| `!=`     | Not equal        |
| `>`      | Greater          |
| `<`      | Less             |
| `>=`     | Greater or equal |
| `<=`     | Less or equal    |
