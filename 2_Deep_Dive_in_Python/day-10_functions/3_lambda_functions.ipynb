# Lambda Functions
---

- Definition: A lambda function in Python is a small, anonymous function defined using the `lambda` keyword
- `One Liner Functions` 

### Key Characteristics
1. Anonymous → no function name required
2. Any Number of Arguments
3. Single expression only ( no multiple statements ) / One logic
4. Automatically returns the result ( no return keyword needed )
5. Often used for short operations, temporary tasks as As Higher Order Functions

### Syntax
- `lambda arguments: expression`

- lambda keyword: let the python interpreter know that the current function is our lambda function
- arguments → inputs (like parameters in a normal function)
- expression → a single expression that gets evaluated and returned

### Examples

```python
# Lambda Function to calculate the sum of two numbers
add = lambda a, b: a + b
print(add(3, 5))   # Output: 8

# Function definition
def add(a, b):
    return a + b

print(add(4,5))

# Square of numbers
squares = lambda x: x ** 2
print(squares(2))

def squares(n):
    return n ** 2

print(squares(3))
print(type(squares)) #<class 'function'>

product = lambda x,y,z: x*y*z

print(product(10,100,9))
```

### 1. `map()` : applies a given function to all items in an iterable or a list and returns an Map Object / Apply a function to all items in a list:
    - used for transforming the data or applying an operation on all items of the iterable
    - help us out with avoiding unnecessary lines of code and we can apply the operations on all values or all items in an iterable and this is sign of good programmer

- Syntax of map() --> `map(function, iterable)`

```python
# Squaring all items of a list
def square(list_items):
    squared_list = []
    for item in list_items:
        square_item = item**2
        squared_list.append(square_item)
    
    return squared_list

print(square([1,2,3,4,5]))

# Using map() function

squares = list(map(lambda x: x**2, [5,4,3,2,1]))
print(squares)
```

```python
# Map function with basic function

def squared_number(num):
    return num ** 2


squares = list(map(squared_number,[1,2,3,4,5]))
print(squares)
```

```python
# Map functions with Multiple Iterables
# While mapping the values of one iterable with another iterable we can actually apply the operations in between the values of the two iterables in a sync way
# [1,2,3,4], [5,6,7,8]  ---> Addition ---> [6,8,10,12]

num_1 = [1,2,3,4,5]
num_2 = [6,7,8,9,10]

addition = list(map(lambda x,y: x+y, num_1,num_2))
print(addition)
```

```python
# map() to convert all string values items in an iterable to an integer

integers = list(map(int,['1','2','3','4','5']))
print(integers)
```

```python
# map() to convert each and every item to Uppercase

employees_name = ["amit",'arun','rakhi','gourav','keshav']

name_to_uppercase = list(map(str.upper, employees_name))
print(name_to_uppercase)
```

```python
# Retrieve the values inside a dictionary using map()
employees = [
    {
        "name":"Arun",
        "age":24,
        "qualification":"Bteach"
    },
    {
        "name":"Gourav",
        "age":27,
        "qualification":"bcom"
    },
    {
        "name":"Rakhi",
        "age":25,
        "qualification":"ma"
    },
    ]

def name_retrievel(list_item):
    return list_item['name']


names = list(map(name_retrievel, employees))
print(names)
```

### 2. filter() : retreive only those items which are meeting a certain condition in an iterable
    - The filter function constructs an Iterator from elements of an iterable for which a condition remains True or which a function return True Values
    - Filter elements based on a condition
  
- Syntax of `filter()` : `list(filter(func, iterable))`

```python
# Retrieve only those values which are even

def even(value):
    return value % 2 == 0

num = [1,2,3,4,5,6,9,11,12,16,17,18]
print(list(filter(even,num)))

# Using lambda function

even_nums = list(filter(lambda x: x%2==0 , num))
print(even_nums)
```

```python
# Retreive value which are greater than 5 

nums = [1,3,4,5,6,7,8,9,100]
print(list(filter(lambda x: x > 5 , nums)))

# all those values greater than 5 and even as well
print(list(filter(lambda x: x > 5 and x % 2 == 0 , nums)))
```

### sorted() : Sort using a custom key:

```python
#
students = [("Arun", 25), ("Rahul", 20), ("Neha", 22)]
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)
# [('Rahul', 20), ('Neha', 22), ('Arun', 25)]
```
