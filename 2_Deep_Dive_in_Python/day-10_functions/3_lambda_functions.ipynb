# Lambda Functions
---

- Definition: A lambda function in Python is a small, anonymous function defined using the `lambda` keyword
- `One Liner Functions` 

### Key Characteristics
1. Anonymous → no function name required
2. Any Number of Arguments
3. Single expression only ( no multiple statements ) / One logic
4. Automatically returns the result ( no return keyword needed )
5. Often used for short operations, temporary tasks as As Higher Order Functions

### Syntax
- `lambda arguments: expression`

- lambda keyword: let the python interpreter know that the current function is our lambda function
- arguments → inputs (like parameters in a normal function)
- expression → a single expression that gets evaluated and returned

### Examples

```python
# Lambda Function to calculate the sum of two numbers
add = lambda a, b: a + b
print(add(3, 5))   # Output: 8

# Function definition
def add(a, b):
    return a + b

print(add(4,5))

# Square of numbers
squares = lambda x: x ** 2
print(squares(2))

def squares(n):
    return n ** 2

print(squares(3))
print(type(squares)) #<class 'function'>

product = lambda x,y,z: x*y*z

print(product(10,100,9))
```
