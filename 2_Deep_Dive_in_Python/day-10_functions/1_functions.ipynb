# Functions
---

- Defintion: reusable block of code that perform a specific task and can be exectued whenever we need.
- Functions helps in organizing code , Reusing code and improve readability.

### Why Use Functions?
- Avoid code repetition (DRY principle)
- Improve readability and structure
- Easier debugging and testing
- Reusability across programs

### Syntax of Function
- start with a `def` keyword
- function_name
- provide parameter
- logic statement

```python
def function_name(parameters): 
    ''' Docstring '''
    #Function body statement
    return value
```

### # Defining a function: using def keyword. A Function might take input in the form of parameters 

```python

def greet(): 
    print("Hello World!")
```

### Calling the function

```python

def greet(): 
    print("Hello World!")

greet() # calling the function
```

### Function Arguments 
- arguments are the values passed inside the parenthesis of the function. A function can have any no. of arguments separated by comma.

```python
def greet(name, age, location): 
    print(f"my name is {name}, age is {age}, live in {location}")

greet("Arun", "26", "Delhi") # calling the function
```

```python
# Create a function that take input form the user and will figure out that the number is even or odd

def even_or_odd(num):
    if num % 2 == 0:
        print(f"{num} is a even number.")
    else:
        print(f"{num} is a odd number.")

n = int(input("Enter the number: ")) #  ValueError: invalid literal for int() with base 10: '4j'
result = even_or_odd(n) # ValueError: invalid literal for int() with base 10: '4j'
```

```python
def even_or_odd_instance(num):
    if isinstance(num, int):
        if num % 2 == 0:
            print(f"{num} is an even number.")
        else:
            print(f"{num} is an odd number.")
    elif isinstance(num,str):
        print("Invalid data type")
    else:
        print(f"{num} is not a valid value for this function")


even_or_odd_instance(5) #5 is an odd number.
even_or_odd_instance('s') #Invalid data type
even_or_odd_instance(14.4) #14.4 is not a valid value for this function
```
