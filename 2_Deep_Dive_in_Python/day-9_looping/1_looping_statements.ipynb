# Looping Statements

- Looping statements allow a program to execute a block of code repeatedly until a condition is met or for a sequence of values.
- In Short, They help automate repetitive tasks.
- Can be used for iterating over a series of numbers or through iterables or data structures or through a collection of data.

### Why Do We use Loops?
- Avoid writing repetitive code
- Iterate over collections (list, tuple, string, etc.)
- Process large datasets efficiently


### Types of Loops in Python
1. for loop → used for iterating over a sequence
2. while loop → runs until a condition becomes False


---

## Range()

* `range()` is used to generate a sequence of numbers.
* It is commonly used in loops like `for`.

### Syntax:

```python
range(start, stop, step)
```

### Parameters:

* **start** → starting value (optional, default is `0`)
* **stop** → ending value (**not included**)
* **step** → difference between numbers (optional, default is `1`)

### Example:

```python
for i in range(1, 5):
    print(i)
```

**Output:**

```python
1
2
3
4
```

### Example with step:

```python
for i in range(1, 10, 2):
    print(i)
```

**Output:**

```python
1
3
5
7
9
```

💡 **Important:** `stop` value is always excluded.
