# Introduction

Artificial Intelligence (AI) is a broad field that aims to make machines perform tasks that normally require human intelligence.

Over time, AI evolved into more specialized fields:

```
Artificial Intelligence (AI)

        ↓

Machine Learning (ML)

        ↓

Deep Learning (DL)

        ↓

Generative AI (GenAI)
```

Each is a subset of the previous one.

---

# Relationship Between AI, ML, DL & GenAI

```
Artificial Intelligence
│
├── Machine Learning
│     │
│     ├── Deep Learning
│     │      │
│     │      ├── Generative AI
```

Think of it like nested circles:

- AI is the biggest field.
- ML is a subset of AI.
- DL is a subset of ML.
- GenAI is a subset of DL.

# 1. Artificial Intelligence (AI)

## Definition

**Artificial Intelligence (AI)** is the branch of computer science that enables machines to perform tasks that normally require human intelligence, such as learning, reasoning, decision-making, perception, and language understanding.

AI is the overall goal of making machines behave intelligently.

---

## Goal of AI

To enable machines to:

- Think logically
- Learn from experience
- Solve problems
- Understand language
- Recognize images
- Make decisions

---

## Examples

- Self-driving cars
- Virtual assistants (Siri, Alexa)
- Recommendation systems
- Face recognition
- Chess-playing systems
- Fraud detection

---

## AI Engineer

An **AI Engineer** designs and integrates AI capabilities into software products.

Typical responsibilities:

- Build AI-powered applications.
- Integrate AI models into web/mobile products.
- Deploy AI solutions.
- Connect AI models with APIs and databases.
- Optimize AI systems for production.

Example:

```
AI Chatbot

↓

LLM

↓

Backend API

↓

Database

↓

Frontend Application
```

---

# 2. Machine Learning (ML)

## Definition

**Machine Learning (ML)** is a subset of AI where computers learn patterns from data instead of being explicitly programmed.

Instead of writing rules manually, we allow the model to learn from examples.

---

## Intuition

Think of teaching a child.

You show:

Input

↓

Correct Answer

↓

Child learns the pattern.

Similarly,

Machine Learning learns from:

```
Input (Features)

↓

Output (Labels)

↓

Learning Algorithm

↓

Prediction
```

---

## Characteristics

- Learns from historical data.
- Uses statistical techniques.
- Makes predictions.
- Performs classification and regression.
- Works well on structured/tabular datasets.

---

## Common ML Algorithms

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)
- Naive Bayes

---

## Limitations of Traditional ML

Traditional ML often requires:

- Manual feature engineering.
- Domain expertise.
- Structured data.

As datasets become larger and more complex (images, audio, text), traditional ML may struggle.

This is why **Deep Learning** became popular.

> **Note:** Traditional ML does not inherently perform poorly on large datasets. In fact, many ML algorithms scale well. The main limitation is that they often rely on manual feature engineering and may not perform as well as Deep Learning on highly unstructured data such as images, speech, and natural language.

---

## ML Engineer

An ML Engineer typically:

- Collects and cleans data.
- Performs feature engineering.
- Trains ML models.
- Evaluates model performance.
- Deploys prediction models.

---

# 3. Deep Learning (DL)

## Definition

**Deep Learning (DL)** is a subset of Machine Learning that uses multi-layer Artificial Neural Networks (ANNs) to automatically learn complex patterns from data.

It is inspired by the structure of the human brain.

---

## Why Deep Learning?

Instead of manually creating features,

Deep Learning automatically learns them.

```
Raw Data

↓

Neural Network

↓

Feature Learning

↓

Prediction
```

---

## Popular Neural Networks

### Artificial Neural Network (ANN)

Used mainly for:

- Tabular data
- Classification
- Regression

---

### Convolutional Neural Network (CNN)

Designed for image-related tasks.

Applications:

- Image Classification
- Object Detection
- Face Recognition
- Medical Imaging
- Self-Driving Cars

---

### Recurrent Neural Network (RNN)

Designed for sequential data.

Applications:

- Machine Translation
- Text Classification
- Speech Recognition
- Language Modeling

Variants:

- LSTM
- GRU
- Bidirectional RNN

---

## Deep Learning Engineer

Responsibilities:

- Build neural network models.
- Train CNNs, RNNs, LSTMs, Transformers.
- Optimize deep learning models.
- Work with GPUs.
- Deploy deep learning systems.

---

# 4. Generative AI (GenAI)

## Definition

**Generative AI (GenAI)** is a subset of Deep Learning that generates new content rather than only analyzing existing data.

It can create:

- Text
- Images
- Videos
- Music
- Audio
- Code

---

# Two Types of AI Models

## 1. Discriminative Models

Purpose:

Predict or classify existing data.

Examples:

- Spam Detection
- Disease Prediction
- Image Classification
- Sentiment Analysis

Question answered:

> **"What is this?"**

---

## 2. Generative Models

Purpose:

Generate completely new content.

Examples:

- ChatGPT
- Image Generation
- Music Generation
- Code Generation

Question answered:

> **"Create something new."**

---

# How GenAI Works

Generative AI models are trained on extremely large datasets.

These datasets may include:

- Books
- Articles
- Websites
- Research papers
- Source code
- Images
- Videos

The model learns patterns and generates new content based on those patterns.

---

# Large Foundation Models

Foundation Models are large pre-trained models that can be adapted to many downstream tasks.

Examples:

### Large Language Models (LLMs)

Generate text.

Examples:

- GPT
- Llama
- Gemini

---

### Large Image Models (LIMs)

Generate images.

Examples include diffusion-based image generation models.

---

# What is RAG?

## Retrieval-Augmented Generation (RAG)

RAG combines:

```
Your Private Data

+

Knowledge Retrieval

+

Foundation Model

↓

More Accurate Response
```

Instead of relying only on what the model learned during training, RAG retrieves relevant external information before generating an answer.

This reduces hallucinations and improves factual accuracy.

---

## Real-World Example

### AI Doctor Assistant

Without RAG:

```
User Question

↓

LLM

↓

Possible Hallucination
```

With RAG:

```
User Question

↓

Hospital Database

↓

Relevant Medical Documents

↓

LLM

↓

Accurate Answer
```

Frameworks commonly used:

- LangChain
- LlamaIndex

# 2. Interview Questions & Answers

## Beginner Questions

### Q1. What is the relationship between AI, ML, DL, and GenAI?

**Answer:** AI is the parent field. ML is a subset of AI, DL is a subset of ML, and GenAI is a subset of DL focused on generating new content.

### Q2. What is the difference between Machine Learning and Deep Learning?

**Answer:** ML often requires manual feature engineering, while DL automatically learns features using neural networks.

### Q3. What is Generative AI?

**Answer:** Generative AI is a branch of Deep Learning that creates new content such as text, images, code, audio, and videos.

### Q4. What is an LLM?

**Answer:** An LLM (Large Language Model) is a foundation model trained on massive text datasets to understand and generate human-like language.

### Q5. What is RAG?

**Answer:** Retrieval-Augmented Generation (RAG) combines external knowledge retrieval with an LLM to generate more accurate and up-to-date responses.

---

## Intermediate Questions

### Q1. Why is Deep Learning preferred for image and language tasks?

**Answer:** Deep Learning automatically learns hierarchical features from unstructured data, making it more effective than traditional ML for images, audio, and text.

### Q2. What is the difference between discriminative and generative models?

**Answer:** Discriminative models classify or predict labels from existing data, while generative models create new content based on learned patterns.

### Q3. Why are Foundation Models important?

**Answer:** They are pre-trained on large datasets and can be fine-tuned or prompted for a wide range of downstream tasks.

### Q4. Why can LLMs hallucinate?

**Answer:** They generate responses based on learned probability distributions rather than verifying facts, which can lead to plausible but incorrect outputs.

### Q5. Why is RAG commonly used with LLMs?

**Answer:** RAG retrieves relevant external information before generation, improving factual accuracy and reducing hallucinations.

---

## Scenario-Based Questions

### Q1. You need to build a spam email classifier. Which approach would you choose?

**Answer:** Traditional Machine Learning or a lightweight Deep Learning model, since the task is classification rather than content generation.

### Q2. You're developing an AI assistant that answers questions from company documents. Which architecture is appropriate?

**Answer:** An LLM combined with RAG so the model can retrieve and use company-specific information.

### Q3. Your project requires automatic object detection in self-driving cars. Which Deep Learning model is suitable?

**Answer:** CNN-based object detection models such as YOLO or Faster R-CNN.

### Q4. A client wants an AI system that writes marketing copy. Which technology would you recommend?

**Answer:** Generative AI using an LLM, optionally enhanced with prompt engineering and RAG if company-specific information is needed.

### Q5. You have a small structured dataset for predicting house prices. Should you use GenAI?

**Answer:** No. A traditional Machine Learning regression model is more appropriate and computationally efficient.
