# `Evolution of LLMS`
---

### Introduction
- Language Model - understanding of language data -> text
- 1967 - Eliza
- 1970 - Shidlu
- 1980 - Xcalibu
- 1988 - RNN - No long term context
- 1997 - LSTM - Long Term content
- 2014 - GRU - less complext structure then LSTM and able to maintain long term context
- 2017 - Transformer
- 2018 - BERT (Google), GPT-1(OpenAI)
- 2019 - GPT-2, ROBETA, XLNET
- 2020 - GPT-3 
- 2021 - GPT 3.5 (45 TBs)
- 2022 - ChatGPT (SFT + RLHF) --> PaLM , Instruct GPT
- 2023 - LLAMA (Open soruce model introduced by Meta), Falcon, GPT-4, PaLM 2
- 2024 - OpenAI (Multimodal -> generate text , image etc), GPT 4.0, Google gemini Pro 1.5, LLAMA 3, Anthropic (Claude), Google (Gemma 2)
- 2025 - Sora AI (video), GPT 5
