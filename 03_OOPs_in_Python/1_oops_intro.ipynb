---

- OOPs (Object-Oriented Programming System) is a programming paradigm where programs are designed using objects and classes.
- Instead of writing everything in functions only, OOP helps organize code into reusable and real-world structures.
- Python is an object-oriented programming language.

- In Object Oriented Programming we are trying to structure the code in such a way that can match up with the real world scenario by using Programming.
- In OOPs we are going to learn about classes and objects
- By a class we mean to provide a blueprint that must remain the same for each and every single object for us.
- These objects can have different functionalities from each other and different characteristics from each other as well but they must follow the similar blueprint.
- By a blueprint or a class what we mean to have a means of providing the functioalities and the characteristices to our objects.

## `Main Concepts of OOPs`

### `Class: ` A class is a blueprint or template for creating objects.

````python
class Car:
    pass

class Avengers:
    pass

class Animals:
    pass

# Car, Avengers and Animars ---> Class
````

### `Object: ` An object is an instance of a class.

````python
class Car:
    pass

class Avengers:
    pass

car1 = Car()
car2 = Car()

avenger1 = Avengers()
avenger2 = Avengers()

# car1 , cars2 --> object of class Car
# avenger1 , avenger2 --> object of class Avengers
````

### Examples:

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

# Creating object
s1 = Student("Arun", 21)

# Calling method
s1.display()
````

### `Understanding the above code: `

1. `class Student`: create a class named Student
2. `__init__()`: 
   1. it is called a constructor
   2. It runs automatically when an object is created
   3. Used to initialize object data.
3. `self`: refers to the current object. Store the value inside the object
4. `display`: methods of object
5. `s1 = Student("Arun", 21)` : initialize or creating an object
6. `s1.display()`: calling the Student's methods

````python
# Example

class Avengers: # class
    # constructure
    def __init__(self, name, age, weapon, nationality): 
        self.name = name
        self.age = age
        self.weapon = weapon
        self.nationality = nationality
    
    # method 1
    def intro(self):
        return f"My name is {self.name}"

    # method 2
    def about(self):
        return f"I am {self.name} and age is {self.age}. My super power is {self.weapon} and from {self.nationality}"

# creating an object
a1 = Avengers('Iron Man', 57, "Mark-46", "American")
a2 = Avengers("Captain America", 40, "Titanium Shield", "Africa")

# calling the methods
print(a1.intro())
print(a1.about()) ## #### kmfjmfjm
print(a2.intro())
print(a2.about()) 
````

### `Why Use OOPs?`
1. Code Reusability : You can reuse classes again and again.
2. Better Organization : Large projects become easier to manage.
3. Reacl world Modeling : You can represent real-world entities easily.
4. Easy Maintenance : Code becomes modular and easier to debug.

### `Four Pillars of OOPs`

| Concept       | Meaning                                 |
| ------------- | --------------------------------------- |
| Encapsulation | Wrapping data and methods together      |
| Abstraction   | Hiding unnecessary details              |
| Inheritance   | Acquiring properties from another class |
| Polymorphism  | One method behaving differently         |
