# `Magic Methods or Dunder Methods`
---
- Special methods in Python that allow us to customize how objects behave.
- also called as Dunder Methods because they have double underscores `__` before and after the name. eg.`__init__`, `__name__`  
- Provide special behaviour to objects

### Categories of Magic Methods

Magic methods are divided into many categories:

- Constructor & Initialization Methods
- String Representation Methods
- Arithmetic Operator Methods
- Comparison Operator Methods
- Container Methods
- Iteration Methods
- Callable Methods
- Destructor Methods
- Context Manager Methods
- Attribute Access Methods

### `Example of Magic Methods`
| Magic Method | Purpose               |
| ------------ | --------------------- |
| `__init__`   | Constructor           |
| `__str__`    | String representation |
| `__len__`    | Length behavior       |
| `__add__`    | `+` operator          |
| `__sub__`    | `-` operator          |
| `__mul__`    | `*` operator          |
| `__eq__`     | `==` comparison       |
| `__gt__`     | `>` comparison        |
| `__lt__`     | `<` comparison        |


### `Why Magic Methods Exist?`

Magic methods allow objects to behave like:

- built-in types
- numbers
- strings
- lists

They enable:

- operator overloading
- custom behavior
- Pythonic syntax

### Whye to use Magic Methods
1. make objects behave like built-in types eg lists, intergers
2. Improve readability by allowing basic operation object1 + object2 
3. Simplifies code by replacing verbose function calls with natural syntax

### `__str__`
- what is returned when print(object) is used

```python
class Avenger:
    def __init__(self, name, strength):
        self.name = name
        self.strength = strength

    def __add__(self, other):
        return Avenger(self.name + "&" + other.name, self.strength + other.strength )
        
    def __str__(self):
        return f"{self.name} with combined strength {self.strength}"
        
ironman = Avenger("Ironman", 80)
thor= Avenger("Thor", 100)
team =  ironman + thor
print(team)
```

## Constructor & Initialization Magic Methods
- `__new__`: To get called in an object's instantiation.
- `__init__`: To get called by the `__new__` method.
- `__del__`: It is the destructor.

### 1. `__init__()` : Constructor
- Automatically called when object is created.
- Used to initialize object data.

```python
class Student:

    def __init__(self, name):
        self.name = name

s = Student("Arun")

print(s.name)
```

Internal Working: Python internally does: `s.__init__("Arun")`
- Used to:

- initialize variables
- assign object data
- setup object state

### 2. `__new__()`
- Responsible for:
- creating object in memory
- Called before __init__().

```python
class Test:

    def __new__(cls):
        print("Creating Object")
        return super().__new__(cls)

    def __init__(self):
        print("Initializing Object")

t = Test()
```

### 2. Numeric magic methods
- `__trunc__(self)`: Implements behavior for math.trunc()
- `__round__(self,n)`: Implements behavior for the built-in round()
- `__abs__(self)`: Implements behavior for the built-in abs()
- `__neg__(self)`: Implements behavior for negation
- `__pos__(self)`: Implements behavior for unary positive

### 3. Arithmetic operators
- `__add__(self, other)`: Implements behavior for the + operator (addition).
- `__sub__(self, other)`: Implements behavior for the - operator (subtraction).
- `__mul__(self, other)`: Implements behavior for the * operator (multiplication).
- `__truediv__(self, other)`: Implements behavior for the / operator (true division).
- `__mod__(self, other)`: Implements behavior for the % operator (modulus).

## 4. String Representation Magic Methods
- `__str__(self)`: Defines behavior for when str() is called on an instance of your class.
- `__repr__(self)`: To get called by built-int repr() method to return a machine readable representation of a type.
- `__hash__(self)`: It has to return an integer, and its result is used for quick key comparison in dictionaries.
- `__bool__(self)`: Returns True or False to determine the truth value of an object, used in conditions and bool() calls.
- `__dir__(self)`: This method to return a list of attributes of a class.

## 5. Comparison magic methods
- `__eq__(self, other)`: Defines behavior for the equality operator, ==.
- `__ne__(self, other)`: Defines behavior for the inequality operator, !=.
- `__lt__(self, other)`: Defines behavior for the less-than operator, <.
- `__gt__(self, other)`: Defines behavior for the greater-than operator, >.
- `__ge__(self, other)`: Defines behavior for the greater-than-or-equal-to operator, >=.

### 1. `__str__()` — String Representation
- Controls what gets printed when using:

```python
# Without __str__
class Student:
    pass

s = Student()

print(s) # <__main__.Student object at 0x000002...>
```

```python
# with __str__
class Student:

    def __str__(self):
        return "Student Object"

s = Student()

print(s) #Student Object
```

### 3. `__len__()` — Length Behavior

```python
# len(object)
```

```python
class Book:

    def __len__(self):
        return 500

b = Book()

print(len(b)) # 500
```
