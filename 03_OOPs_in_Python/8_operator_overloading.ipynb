# `Operator Overloading`
---
- allows us to define how operators ( +, -, * , %, /, etc...) works with user defined objects
- In Python, operators are not limited to: integers, floats , strings. 
- Operators can also work with: user-defined objects
- This feature is called: Operator Overloading

## Why is Operator Overloading useful ?
    - makes custom objects behave like built-in types ( custom objects)
    - Improves code readability by using natural syntax ( object1 + object2) instead of object1.add(object) ( readability )

###  Arithmetic operators
- `__add__(self, other)`: for + operator (addition).
- `__sub__(self, other)`: for - operator (subtraction).
- `__mul__(self, other)`: for * operator (multiplication).
- `__truediv__(self, other)`: for / operator (true division).
- `__mod__(self, other)`: for % operator (modulus).
- `__eq__(self,other)`: for == operator (equal)
- `__gt__(self,other)`: for > operator (greater than)
- `__lt__(self,other)`: for <> operator (less than)

```python
# example

class Vehicle:
    def __init__(self, name, type):
        self.name = name
        self.type = type

    def __add__(self, other):
        return Vehicle( self.name + " & " + other.name ,self.type + other.type)

    def __sub__(self, other):
        return Vehicle(self.name, self.type - other.type)

    def __str__(self):
        return f"{self.name} with Power {self.type}"

v1 = Vehicle('BMW', 100)
v2 = Vehicle("Audi", 3000)
v = v1 + v2
print(v)

```

## `Example 1: Overloading +`

```python
class Vector:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x,
                      self.y + other.y)

    def show(self):
        print(f"({self.x}, {self.y})")


v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2

v3.show()
```

How It Works Internally
- When Python sees: v1 + v2

- It actually does:: v1.__add__(v2)

### Example 2: `Overloading -`

```python
class Number:

    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return Number(self.value - other.value)

    def show(self):
        print(self.value)


n1 = Number(20)
n2 = Number(5)

result = n1 - n2

result.show()
```

### Example 3: `Overloading *`

```python
class Multiply:

    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        return Multiply(self.value * other.value)

    def show(self):
        print(self.value)


a = Multiply(5)
b = Multiply(4)

c = a * b

c.show()
```

## Example 4: `Overloading ==`

```python
class Student:

    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks


s1 = Student(90)
s2 = Student(90)

print(s1 == s2)
```

Note: Without `__eq__`, Python compares memory addresses.

## Example 5: `Overloading < and >`

```python
class Box:

    def __init__(self, weight):
        self.weight = weight

    def __lt__(self, other):
        return self.weight < other.weight

    def __gt__(self, other):
        return self.weight > other.weight


b1 = Box(10)
b2 = Box(20)

b3 = Box(10)
b4 = Box(20)

print(b1 < b2)
print(b3 > b4)
```

## Example 6: `__str__`

```python
class Car:

    def __init__(self, brand):
        self.brand = brand

    def __str__(self):
        return f"Car brand is {self.brand}"


c = Car("BMW")

print(c)
```

| Operator | Magic Method   |
| -------- | -------------- |
| `+`      | `__add__`      |
| `-`      | `__sub__`      |
| `*`      | `__mul__`      |
| `/`      | `__truediv__`  |
| `//`     | `__floordiv__` |
| `%`      | `__mod__`      |
| `**`     | `__pow__`      |
| `==`     | `__eq__`       |
| `!=`     | `__ne__`       |
| `<`      | `__lt__`       |
| `>`      | `__gt__`       |
| `<=`     | `__le__`       |
| `>=`     | `__ge__`       |


```python
class AIModel:
    def __init__(self, name, accuracy):
        self.name = name
        self.accuracy = accuracy

    def __gt__(self, other):
        return self.accuracy > other.accuracy

    def __lt__(self, other):
        return self.accuracy < other.accuracy

    def __eq__(self, other):
        return self.accuracy == other.accuracy

gpt4 = AIModel("GPT-4",98)
bard = AIModel("Google Bard", 96)

print(gpt4 > bard)
print(gpt4 < bard)
print(gpt4 == bard)
```

### Practical Example: Bank Account

```python
class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def __add__(self, other):
        return BankAccount(self.balance + other.balance)

    def __str__(self):
        return f"Balance = {self.balance}"


a1 = BankAccount(5000)
a2 = BankAccount(3000)

total = a1 + a2

print(total)
```

### Advantages of Operator Overloading
- Cleaner syntax
- More readable code
- Mathematical modeling
- Makes custom classes behave like built-in types
  
### Disadvantages
- Can confuse readers if overloaded badly
- Too much overloading reduces clarity
- Incorrect implementation causes bugs
