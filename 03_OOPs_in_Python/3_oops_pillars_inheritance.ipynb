## `Inheritance`
- Fundamental concept in OOPs because of which the child class is able to inherit the properties, attributes and functionalities of the parent class.
- Inheritance allows one class to acquire: properties and methods from another class.

| Term         | Meaning                          |
| ------------ | -------------------------------- |
| Parent Class | Base class / Superclass          |
| Child Class  | Derived class / Subclass         |
| Inheritance  | Acquiring properties from parent |

- eg. Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparents and definitely in this hierarchy the grandchild can have some features different from the parents and grandparent as well

Benefits of Inheritance
1. code reuse
2. avoids duplication
3. better organization
4. hierarchical structure


### 3 Types of Inheritance
1. Single Inheritance
2. Multilevel Inheritance
3. Multiple Inheritance

### 1. `Single Inheritance`
- The Features or the Attributes from one parent class will be getting inherited into one child class.
- Parent Class ( Properties --> Attributes and Functionalities) ------> Child Class
- One child inherits one parent.

```python
class A:
    pass

class B(A):
    pass
```

```python
class Avenger:
    def __init__(self, name, power):
        self.name = name
        self.power = power

    def introduction(self):
        print(f"Hello my name is {self.name}")

class IronMan(Avenger):
    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon = weapon
    
    def weapon_details(self):
        print(f"My powerfull weapon is {self.weapon}")


iron_man = IronMan("Iron Man", "Genius Intellect", "Mark 46 Iron Man Suit")
iron_man.introduction()
iron_man.weapon_details()
```

### 2. `Multilevel Inheritance`:
- The properties or methods and attributres of the grandparent calss can be inherited by the parent class and then the properties of the parent class will be inherited by the child class.
- In a proper heirarchy, the Inheritance process is going to work for us.

Multilevel Inheritance is a type of inheritance where:
- one class inherits from another class
- and another class inherits from that derived class and forming an inheritance chain.


`Grandparent(Properties) --> Parent(Properties) --> Child`
The son inherits:
- properties from father
- and indirectly from grandfather

Similarly in Python:

- child class gets access to all ancestor class members.

```python
class GrandParent:
    def __init__(self, nationality, baldness, anger , diabetic):
        self.nationality = nationality
        self.baldness = baldness
        self.anger = anger
        self.diabetic = diabetic

    def introduction(self):
        print(f"My Nationality is {self.nationality} and I am {self.diabetic}")

class Parent(GrandParent):
    def __init__(self, nationality, baldness, anger , diabetic, intelligence ):
        super().__init__(nationality, baldness, anger , diabetic)
        self.intelligence = intelligence

    def intelligent(self):   
        print(f"About my intelligence is {self.intelligence}") 


class Child(Parent):
    def __init__(self, nationality, baldness, anger , diabetic, intelligence, charm ):
        super().__init__(nationality, baldness, anger , diabetic, intelligence)
        self.charm = charm

    def charmed(self):
        print(f"About my charm, whether I am charm or not. Answer is {self.charm}")


parent1 = Parent("Indian", "No", "Yes", "Yes", "Yes")
parent1.intelligent()
parent1.introduction() # inherit from GrandParent class

child1 = Child("American", "Yes", "No", "Yes", "Yes", "Yes")
child1.introduction() # Child class inherit from Grandparent class
child1.intelligent() # Child class inherit from Parent class
child1.charmed()

```

```python
class Grandfather:

    def land(self):
        print("Grandfather owns land")

class Father(Grandfather):

    def house(self):
        print("Father owns house")

class Son(Father):

    def bike(self):
        print("Son owns bike")

s = Son()

s.land() # inherit from GrandFather
s.house() # inherit from Father
s.bike() 
```

```python
class Animal:

    def eat(self):
        print("Animal eats food")

class Dog(Animal):

    def bark(self):
        print("Dog barks")

class Puppy(Dog):

    def weep(self):
        print("Puppy weeps")

p = Puppy()

p.eat()
p.bark()
p.weep()
```

```python
# Level 1: Base Class (Grandparent)
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    
    def move(self):
        print(f"The {self.brand} is moving.")

# Level 2: Derived Class (Parent)
class LandVehicle(Vehicle):
    def __init__(self, brand, wheels):
        super().__init__(brand)
        self.wheels = wheels
    
    def drive(self):
        print(f"Driving on {self.wheels} wheels.")

# Level 3: Derived Class (Child)
class Car(LandVehicle):
    def __init__(self, brand, wheels, fuel_type):
        super().__init__(brand, wheels)
        self.fuel_type = fuel_type
    
    def display_info(self):
        print(f"This is a {self.brand} car with {self.wheels} wheels running on {self.fuel_type}.")

# Creating an instance of the bottom-level class
my_car = Car("Toyota", 4, "Petrol")

# Accessing methods from all levels
my_car.move()          # From Grandparent
my_car.drive()         # From Parent
my_car.display_info()  # From Child
```

### Note: Method Overriding in Multilevel Inheritance : Child class can override ancestor methods.

```python
class A:

    def show(self):
        print("Class A")

class B(A):

    def show(self):
        print("Class B")

class C(B):
    pass

c = C()

c.show()
```

### Advantages of Multilevel Inheritance
1. Code Reusability : Reuse methods from multiple ancestor classes.

2. Reduces Redundancy: No repeated code.

3. Better Hierarchical Structure: Represents real-world relationships clearly.

4. Easy Maintenance: Changes in parent affect descendants automatically.

### Disadvantages
1. Complex Debugging: Long inheritance chains can confuse beginners.

2. Tight Coupling: Changes in parent may affect many child classes.

3. Reduced Readability: Too many inheritance levels make code harder to understand.

### `3. Multiple Inheritance: `

- Multiple Inheritance is a type of inheritance where: One child class inherits from more than one parent class.

- The child class gets access to:
    1. variables
    2. methods

from all parent classes.


- Real World Analogy
Suppose a child inherits:

  - intelligence from mother
  - business skills from father

Similarly in programming:

child class inherits features from multiple classes.

```python
# Basic Syntax

class Parent1:
    pass

class Parent2:
    pass

class Child(Parent1, Parent2):
    pass
```

```python
class Father:

    def money(self):
        print("Father gives money")

class Mother:

    def care(self):
        print("Mother gives care")

class Child(Father, Mother):

    def fun(self):
        print("Child enjoys")

c = Child()

c.money()
c.care()
c.fun()
```

```python
class GrandFather:

    def ancestoral_land(self):
        print("GrandFather gives ancestoral land")

class Father:

    def unlimited_money(self):
        print("Father gives unlimited money")

class Child(GrandFather, Father):

    def enjoying_both(self):
        print("I am enjoying land from Grandfather and unlimited money from Father")

c = Child()

c.ancestoral_land()
c.unlimited_money()
c.enjoying_both()
```

```python
class Camera:

    def click(self):
        print("Taking Photo")

class Phone:

    def call(self):
        print("Calling")

class Smartphone(Camera, Phone):

    def internet(self):
        print("Using Internet")

s = Smartphone()

s.click()
s.call()
s.internet()
```

### Why Use Multiple Inheritance?

Used when a class needs:

- features from multiple classes
- combined functionality

### Method Overridding in Multiple Inheritance

```python
class A:

    def show(self):
        print("A")

class B:

    def show(self):
        print("B")

class C(A, B):

    def show(self):
        print("C")

c = C()

c.show()
```

### Advantages of Multiple Inheritance
1. Code Reusability: Reuse code from multiple classes.

2. Modular Design: Separate features into different classes.

3. Flexibility: Combine functionalities easily.

### Disadvantages of Multiple Inheritance
1. Complexity: Code becomes difficult to understand.

2. Diamond Problem: Method ambiguity may occur.

3. Difficult Debugging: MRO confusion in large projects.

### Best Practices
- Use Multiple Inheritance Carefully
Good for:
1. mixins
2. reusable utility classes

Avoid:
1. deep complicated hierarchies
