# `Polymorphism`
---

The word Polymorphism comes from:

- Poly → Many
- Morph → Forms

So polymorphism means `many forms` in Greek: `One thing having many forms.`

In Object-Oriented Programming:

- The same method/function/operator can behave differently in different situations.
- Polymorphism allows object  of Different classes as if they are the object for the same class and it enables resuability and makes program more flexible

### Real World Analogy

Think about a person.
One person can behave as:

1. Teacher in school
2. Father at home
3. Friend socially

Same person → different behavior.
- That is polymorphism.

### Another Example

A button on phone:

- short press → screen on/off
- long press → restart phone

Same button → different actions.

### `Why Use Polymorphism?`

Polymorphism provides:

1.. flexibility
2. code reusability
3. extensibility
4. cleaner interfaces

It allows writing generalized code.

## Types of Polymorphism in Python
1. Method Overriding --> ( Runtime Polymorhism)
2. Operator Overloading ( Compile Time Polymorphism)
3. Operator Overloading

### `1. Method Overriding`
- When a child class provides its own implementation of a parent class method.
- Same method name: different behavior

- A child class provide , a specific implementation of a method that has already been defined into the parent class
- This allows Dynamic Selection at Runtime

```python
class Animal:

    def sound(self):
        print("Animal makes sound")

class Dog(Animal):

    def sound(self):
        print("Dog barks")

d = Dog()

d.sound()  # Dog overrides sound method from Animal
```

```python
class Grandparent: # Parent Class
    def speak(self):
        print("I am  Grandparent")

class Parent(Grandparent): #Child Class
    def speak(self):
        print("I am Parent")


class Child(Grandparent):
    def speak(self):
        print("I am the Child ")



parent1 = Parent()
parent1.speak()


child1 = Child()
child1.speak()

# flow: Child --> Parent --> GrandParent
```

```python
class Shape:
    def __init__(self):
        return "This method returns the area of the shape"

class Rectangle(Shape):
    def __init__(self,width, height):
        self.width = width
        self.height = height

    def area(self):
        return f"area of rectange: {self.width * self.height}"

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return f"area of circle : {self.radius * self.radius * 3.14} "


rect1 = Rectangle(100,200)
print(rect1.area())
circle1 = Circle(5)
print(circle1.area())
```

###  Access Parent Method Using super()

```python
class Animal:

    def sound(self):
        print("Animal sound")

class Dog(Animal):

    def sound(self):
        super().sound()
        print("Dog barks")

d = Dog()

d.sound()
```

### `Method Overloading`
- Method overloading means: Same method name with different parameters.

`Important Point`: Python does NOT support true method overloading directly like:
- Java
- C++

Because Python replaces old methods with latest one.

```python
class Test:

    def add(self, a):
        print(a)

    def add(self, a, b):
        print(a + b)

t1 = Test()
t1.add(10,20) # Only second method survives.
```

### `Operator Overloading`
- Python operators can behave differently depending on object type.

```python
print(5 + 3) # Addition for integers.

print("Hello " + "World") # Concatenation for strings.
```
