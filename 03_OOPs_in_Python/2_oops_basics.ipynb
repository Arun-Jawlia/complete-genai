# `Object Oriented Programming Basics`
---

```python
class Cartoon:
    def __init__(self,name, show, catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase

    def introduction(self):
        return f"My name is {self.name}"
    
    def speak(self):
        return f"{self.name} in his show {self.show} says {self.catchphrase}"

tom = Cartoon('Tom', "Tom and Jerry", "Kill the Jerry")
print(tom.introduction())
print(tom.speak())

mickey_mouse = Cartoon("Mickey Mouse", "Mickey Mouse The Great", "Help me Help me")
print(mickey_mouse.introduction())
print(mickey_mouse.speak())
```

## `Object Oriented Programming Terminologies`

1. Attributes --> stands for the characteristices that a specific object is going to have
2. Methods --> Functionality that a specific object is going to have
3. self --> it refers to the current object of the class
4. constructor --> being referred by `__init__()` and gets automatically called as soon as we are going to create an object and responsible for associating the attributes to your current object

```python
class Car:
    color = "Blue"
    type = "electric"

car1 = Car()
print(car1.color)
print(car1.type)
```

### Types of Constructors:
1. Default Constructor : it doesn't have the parameters inside it and it is only having the self keyword inside it. It assign a default value to all the object
2. Parameterized Constructor : it consist of parameters inside it so that every object can have a specific attribute associated to it. It allows us to pass parameter while creating an object.

### `Default Constructor :` 
- A constructor that does not take extra parameters except self.

```python
# Default Constructor

class Dog:
    def __init__(self):
        self.name = "Pitbull"
        self.color = "Black-Brown"

    def show_info(self):
        return f"This dog is {self.name} and color of the dog is {self.color}"
    

dog1 = Dog()
print(dog1.name)
print(dog1.color)
print(dog1.show_info())

dog2 = Dog()
print(dog2.name)
print(dog2.color)
print(dog2.show_info())

# Each and every object have the default values
```

### `Parameterized Constructor :`
- A constructor that accepts parameters.
- Values are passed during object creation:

```python
# Parameterized Constructor

class Dog:
    def __init__(self, name,breed, color):
        self.name = name
        self.breed = breed
        self.color = color

    def show_info(self):
        return f"This dog is {self.name}and its breed is {self.breed} and color of the dog is {self.color}"
    

dog1 = Dog("Bantu", "Pitbull", "Black")
print(dog1.name)
print(dog1.breed)
print(dog1.color)
print(dog1.show_info())

dog2 = Dog("Moti", "Siberian Husky", "White")
print(dog2.name)
print(dog2.breed)
print(dog2.color)
print(dog2.show_info())
```

## `Types of Methods`

### 1. `Instance Method`
- Instance methods operate on Object Attributes
- use self keyword to access and modify Object Attributes
- These methods are called using an object
- Require `self` 

Characteristics
1. accesses instance variables
2. object dependent
3. called using object

```python
class Avengers:
    def __init__(self,name, power):
        self.name = name
        self.power = power

    def show_details(self):
        print(f"My name is {self.name} and Super Power is {self.power}")

iron_man = Avengers("Iron Man", "Genius Intellect")
iron_man.show_details()

captain_america = Avengers("Captain America", "Titanium Shield")
captain_america.show_details()
        
```

### 2. `Class Method`
- Method of whole class or the whole blueprint which will remain the same for all the objects than such kind of Method is being referred as a Class method.
- In Class Method, These methods work with the class level attributes
- They are going to use `cls` instead of self
- They are called using the class name as well
- for specific that the current method is a class method we have to use the abbreviation / decorator as `@classmethod`


Class method accesses class-level data.
- Characteristics
    - works with class variables
    - no object required
    - called using class name

```python
class Avengers:
    
    origin = "Avenger Team"

    def __init__(self, name, power):
        self.name = name
        self.power = power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")
    
    def self_info(self):
        print(f"My name is {self.name} and Super Power is {self.power}")
        

iron_man = Avengers("Iron Man", "Genius Intellect")
iron_man.origin_info()
iron_man.self_info()

```

```python
class Student:

    school = "ABC School"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
```

### 3. `Static Method`
- don't use `self` or `cls` decorators
- work like normal functions but belong to a class
- designed by the abbrevation `@staticmethod`
- also be called by the classname also

Static methods behave like normal functions but belong to class namespace.
- Characteristics
    1. utility/helper methods
    2. independent of object and class data
    3. improves code organization

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b

print(Math.add(5, 3))
```

```python
class Avengers:

    # Static Method
    @staticmethod
    def mission():
        print("All Avengers must protect the Earth")
        

iron_man = Avengers()
iron_man.mission()
```

### Example using Instance, Class and Static Method

```python
class Avengers:

    # Static Method
    @staticmethod
    def mission():
        print("All Avengers must protect the Earth")

    origin = "Avenger"
    
    def __init__(self, name, power):
        self.name = name
        self.power = power

    # Class Method
    @classmethod
    def origin_info(cls):
        print(f"Origin is {cls.origin}")

    # Instance Method
    def show_details(self):
        print(f"My name is {self.name} and power is {self.power}")
        

iron_man = Avengers("Iron Man", "Genius Intellect")
iron_man.mission()
iron_man.origin_info()
iron_man.show_details()
```

```python
class Employee:

    company = "Google"

    def __init__(self, name):
        self.name = name

    # Instance Method
    def show_name(self):
        print(self.name)

    # Class Method
    @classmethod
    def show_company(cls):
        print(cls.company)

    # Static Method
    @staticmethod
    def greet():
        print("Welcome")

e1 = Employee("Arun")

e1.show_name()

Employee.show_company()

Employee.greet()
```

### `Difference Between Method Types`

| Feature                   | Instance Method | Class Method   | Static Method   |
| ------------------------- | --------------- | -------------- | --------------- |
| Uses                      | Object data     | Class data     | Utility work    |
| First Parameter           | `self`          | `cls`          | None            |
| Decorator                 | No              | `@classmethod` | `@staticmethod` |
| Access Instance Variables | Yes             | No             | No              |
| Access Class Variables    | Yes             | Yes            | No              |
| Object Required           | Yes             | No             | No              |
