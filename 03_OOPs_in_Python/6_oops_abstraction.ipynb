# `Abstraction`
---
- Abstraction is one of the four pillars of OOP.
- Concept of hiding the implementation details and only showing the necessary features of the object.
- This helps in reducing programming complexity and efforts
- Allows us to work with an object with what an object does rather than how it does it.

- It means:
    - `Hiding internal implementation details and showing only essential features to the user.`

- The user only knows: what an object does
- not: how it does it internally.

### `ATM Example`

You can:

- withdraw money
- deposit money

But you don't know:

- server communication
- database queries
- authentication algorithms

Only essential functionality is exposed.

### `Why Abstraction is Needed?`

Without abstraction:

- systems become too complex
- users need to understand unnecessary details

Abstraction:

- reduces complexity
- improves usability
- increases maintainability

## `Abstraction in Python`
Python implements abstraction using:
1. Abstract Class
2. Abstract Method

### 1. `What is an Abstract Class?`
- A class that can't be instantiated or that means for an abstract class we can't create the object of this class.

- An abstract class is a class that:
    1. cannot be instantiated directly
    2. is meant to be inherited

- It acts like a blueprint.


### 2. `What is an Abstract Method?`
- Abstract method just because it can't accessed by the object of the abstract class itslef thats why we have to call this abstract method in out child clas  and the implementation of the abstract method is not provided inside the abstract class but must be provided in the child class.

- An abstract method:
    - has declaration only
    - no implementation (usually)

- Child classes MUST implement it.
- defined by `@abstractmethod`

```python
from abc import ABC, abstractmethod # ABC -- Abstract Class

class Vehicle(ABC): # Creates abstract class.

    @abstractmethod 
    def start(self): # Defines abstract method.
        pass 
    
    @abstractmethod
    def stop(self): # Defines abstract method.
        pass

class Car(Vehicle):

    def start(self):
        print("The car start with an ignition key ")

    def stop(self):
        print("The car stops with the brakes")

class ElectricScooter(Vehicle):

    def start(self):
        print("The electric scooter start with power button")

    def stop(self):
        print("The electric scooter stops with an electric brake")

# v1 = Vehicle() # TypeError: Can't instantiate abstract class Vehicle with abstract method stop
c1 = Car()
c1.start()
c1.stop()

e1 = ElectricScooter()
e1.start()
e1.stop()
```

### `Note:` : What Happens if Child Does Not Implement Method? like Start and Stop Method
- TypeError: Can't instantiate abstract class ElectricScooter with abstract method start
- TypeError: Can't instantiate abstract class ElectricScooter with abstract method stop

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass

class UPI(Payment):

    def pay(self):
        print("Payment through UPI")

class Card(Payment):

    def pay(self):
        print("Payment through Card")

u = UPI()
c = Card()

u.pay()
c.pay()
```

### `Benefits of Abstraction`
1. Reduces Complexity: Users interact with simple interfaces.

2. Improves Security: Internal logic remains hidden.

3. Better Maintainability: Internal implementation can change without affecting users.

4. Encourages Standardization: All child classes follow same method structure.`

### Abstract Class Can Have Normal Methods Too

Abstract classes can contain:

- abstract methods
- normal methods
- constructors
- variables

```python
from abc import ABC, abstractmethod

class Person(ABC):

    def details(self):
        print("This is normal method")

    @abstractmethod
    def profession(self):
        pass

class Teacher(Person):

    def profession(self):
        print("Teaching")

t = Teacher()

t.details()
t.profession()
```

| Concept           | Meaning                       |
| ----------------- | ----------------------------- |
| Abstraction       | Hiding implementation details |
| Abstract Class    | Blueprint class               |
| Abstract Method   | Method without implementation |
| `ABC`             | Base class for abstraction    |
| `@abstractmethod` | Declares abstract methods     |
