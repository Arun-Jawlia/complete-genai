# `Encapsulation`
---

- Encapsulation is one of the four pillars of OOP.
- Wrapping data (variables) and methods (functions) together into a single unit called a class.
- It also means: Restricting direct access to data components's which is a means of preventing accidental inference , misuse and controlling it through methods.
- It also means : Restricting direct access to data and controlling it through methods.
Simple Meaning

Encapsulation is:

1. data hiding
2. data protection
3. controlled access to data


### Real World Analogy: ATM Machine Example

When using an ATM:

- you can withdraw money
- deposit money
- check balance

But you cannot directly access:

- bank database
- server logic
- internal calculations

Internal details are hidden and protected.

That is encapsulation.

### Another Example

Think of a capsule medicine.

Inside capsule:

- many ingredients are wrapped together

Similarly in OOP:

data + methods are wrapped inside class.

### `Advantages of Encapsulation`
1. Data Security: Protects important data.

2. Controlled Access: Only authorized operations allowed.

3. Better Flexibility: Internal implementation can change without affecting users.

4. Easy Maintenance: Code becomes modular.

5. Prevents Accidental Changes: Protects object state.

`Disadvantages of Encapsulation`
1. Extra Code: Need getter/setter methods.

2. Slight Complexity: May feel difficult initially.

### Encapsulation in Python

Python achieves encapsulation using:
1. classes
2. access specifiers

### `Access Specifiers of Encapsulation in Python`:
1. Public Members
2. Protected Members
3. Private Members

### 1. `Public Members`
- Accessible from Anywhere
- Attributes are accessible outside of the class
- These members are accessible by the derived class also
- public member attribute is directly declared without any specific keyword

```python
class Student:

    def __init__(self):
        self.name = "Arun"

s = Student()

print(s.name)
```

```python
class Bank_Details:
    def __init__(self, balance, name, email):
        self.balance = balance
        self.name = name
        self.email = email

    def bankdetails(self):
        print(f"My name is {self.name} and email is {self.email} and Bank balance is {self.balance}")

myAcc = Bank_Details(100000, 'Arun', "arunkumar@gmail.com")
print(myAcc.name)
print(myAcc.email)
print(myAcc.balance)

# balance, bankdetails, email, name are Accessible from AnyWhere or outside the class
```

```python
dir(myAcc)
```

### 2. `Private Members`
- Not Accesible outside the class and not even in the derived class 
- in order to have the access over these private members or to bring some modifications on these members we use `getter and setter` methods.

- Private Members are declared by using `__double_underscore` before the name `__anyvariable or method` eg. __balance , __password

Private members are used to:

1. hide sensitive data
2. prevent direct modification
3. enforce controlled access

```python
class Bank_Details:
    def __init__(self, balance, name, email, phone):
        self.name = name
        self.email = email
        self.__balance = balance
        self.__phone = phone

myAcc = Bank_Details(100000, 'Arun', "arunkumar@gmail.com", "9718653508")
print(myAcc.name)
print(myAcc.email)
# print(myAcc.__balance) # AttributeError: 'Bank_Details' object has no attribute '__balance'
# print(myAcc.__phone) # AttributeError: 'Bank_Details' object has no attribute '__phone'

# balance, bankdetails, email, name are Accessible from AnyWhere or outside the class
```

```python
dir(myAcc)
```

### Getter and Setter Methods

Encapsulation often uses:

1. getters : Used to read private data.
2. setters : Used to modify private data safely.

to control access.

Why Use Getter and Setter?
They help:

1. validate data
2. secure data
3. prevent invalid modification

```python
class Bank_Details:
    def __init__(self, balance, name, email, phone):
        self.name = name
        self.email = email
        self.__balance = balance
        self.__phone = phone

    # getter method
    def show_balance(self):
        print(f"My Bank Balance is {self.__balance}")
    
    # setter method
    def add_balance(self, amount):
        if not amount >= 0:
            print("Add a valid amount")
            return 
        new_balance = self.__balance + int(amount)
        self.__balance = new_balance
    

myAcc = Bank_Details(2000, 'Arun', "arunkumar@gmail.com", "9718653508")
print(myAcc.name)
print(myAcc.email)
myAcc.add_balance(1000) 
myAcc.show_balance()
myAcc.add_balance(4000)
myAcc.show_balance()

# balance, bankdetails, email, name are Accessible from AnyWhere or outside the class
```

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):

        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def show_balance(self):
        print("Balance:", self.__balance)

acc = BankAccount(10000)

acc.deposit(5000)

acc.withdraw(3000)

acc.show_balance()
```

```python
class User:
    def __init__(self, email, password, name):
        self.name = name
        self.__email = email
        self.__password = password

    def my_info(self):
        print(f"Name: {self.__email}")
    
u1 = User("arun@gmail.com", "12344", "Arun")
print(u1.name)
# print(u1.__email) # AttributeError: 'User' object has no attribute '__email'
# print(u1.__password) #AttributeError: 'User' object has no attribute '__password'

u1.my_info()
```

### 3. `Protected Members`
- Data which is not accessible outside the class But the protected data is going to be accessible by the derived class or child class.
- Denoted by `_attribute` or single underscore

`Important Point`

Python does NOT strictly enforce protection.

Protected members are mainly:

- a convention
- a warning for developers

Meaning:

“This member should not be accessed directly outside the class unless necessary.”

### Real World Analogy

Think of a company's internal document.

- Employees inside company can access it.
- Public users should not access it directly.

Protected members are similar:

- accessible internally
- not intended for outsiders

```python
class GrandParent:
    def __init__(self, name, locker_key):
        self.name = name
        self._locker_key = locker_key
    
    # getter method
    def locker_info(self):
        print(f"The locker key info is {self._locker_key}") # AttributeError: 'GrandParent' object has no attribute 'locker_key'

class Parent(GrandParent):
    def __init__(self, name, locker_key):
        super().__init__(name, locker_key)

    def locker_reveals(self):
        print(f"Locker key information is {self._locker_key}") # AttributeError: 'Parent' object has no attribute 'locker_key'


gp = GrandParent('Amit', "902022")
# print(gp.locker_key) #AttributeError: 'GrandParent' object has no attribute 'locker_key'

print(gp._locker_key) # Allowed by not Recommended, it is only for internal use
gp.locker_info()

child = Parent("Arun", "123456")
child.locker_reveals()

```

Note: because _locker_key is only conventionally protected.

```python
# Bank Account System

class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    # getter method
    def get_balance(self):
        print(f"My bank balance is {self.__balance}")

    
    def deposit(self,amount):
        if amount >=0:
            self.__balance += amount

            print(f"Deposited Amount successfully and Your total Balance is {self.__balance}")
        else:
            print("Desposited amount must be positive")

    def withdraw(self,amount):
        if self.__balance >= amount:
            self.__balance -= amount
            print(f"WithDraw amount is {amount} and Your Remaining Balance is {self.__balance}")
        else:
            print("Insufficient Fund")
        
user1 = BankAccount("Arun Jawlia", 100000)
# print(user1.__balance) #AttributeError: 'BankAccount' object has no attribute '__balance'
user1.get_balance()
user1.deposit(1000)
user1.deposit(1000)
user1.deposit(1000)
user1.withdraw(50000)
```

### `Protected vs Private Members`
| Feature                   | Protected                  | Private                 |
| ------------------------- | -------------------------- | ----------------------- |
| Syntax                    | `_name`                    | `__name`                |
| Accessible in Child Class | Yes                        | No (directly)           |
| Accessible Outside Class  | Yes                        | No                      |
| Enforced Restriction      | No                         | Partial (name mangling) |
| Purpose                   | Internal + inheritance use | Stronger data hiding    |


### `Difference Between Public, Protected, and Private`
| Type      | Symbol   | Accessible Outside Class |
| --------- | -------- | ------------------------ |
| Public    | `name`   | Yes                      |
| Protected | `_name`  | Yes (but discouraged)    |
| Private   | `__name` | No                       |
