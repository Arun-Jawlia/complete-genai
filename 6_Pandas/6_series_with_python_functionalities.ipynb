## 13. `Boolean Indexing` in Series
- to perform the Important Analysis on the column values, where in a specific index values if it missing a condition that we get True over their else we get False
- Return --> True --> Condition is True
- Return --> False --> Condition is Not True

```python
avg
```

```python
# which youtube are having average video lenght greater than 24 minutes

avg > 24
```

```python
# only get the youtube having 24 minutes average length
avg[avg > 24]
# avg[avg > 24].size
```

```python
dataframe[dataframe > 120]
```

## 14. `Plotting Graph in Pandas`
- for plotting graphs in Pandas, we require `plot()` function
- kind --> it decides what kind of graph you want to plot like bar, pipe 
- Syntax: df.plot()

```python
avg.plot()
```

```python
dataframe.sort_values(ascending=False).head(15).plot(kind='bar')
```

```python
dataframe.sort_values(ascending=False).sample(15).plot(kind='bar')
```

```python
dataframe.sort_values(ascending=False).sample(15).plot(kind='pie')
```
