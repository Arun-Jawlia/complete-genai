## Mathematical Functions in Pandas

## 1. `sum()`: sum of all values present in the series

```python
subs.sum()
```

## 2. `product()`: product of all values in the series

```python
subs.product()
```

## 3. `sort_values()`: sort the series

```python
subs.sort_values()
```

## 4. `mean()`: provides us average values of the data column 

```python
subs.mean()
```

## 5. `median()`: provide the idea about the average of the data itself
- eg. we need to sort the value in Ascending manner and then the middle value after the sorting the data is median

```python
subs.median()
```

## 6. `mode()`: provide us with the idea which value is getting repeated maximum number of times.

```python
subs.mode()
```

## 7. `std()` : stands for Standard Deviation
- if i need to get the idea that how the vlaues inside the series are distributed from the average or mean value then we can come up with the standard deviation

```python
avg.std()
```

## 8. `variance()`: provide the insight about the distribution of the data from the mean value and variance is the square of the standard deviation. 
- standard_deviation = |mean - v_1| + |mean - v_2|..../ total count of values
- variance = std**2

```python
avg
```

```python
avg.var()
```

## 10. `max()` : maximum value in the Series

## 11. `describe()` : provide the idea about the summary of the important mathematical calculations for the data  (VV Important)
- min: minimum value
- max: maximum value
- mean: average value
- std: standard deviation
- 25%: 25% of the values out of my all values are less than this specific value.
- 50%: 50% of the values out of my all values are less than this specific value.
- 75%: 75% of the values out of my all values are less than this specific value.

eg. JEE mains : 25% --> my score in JEE Mains was greater than 25% of the total applicants who applied in JEE

# `Indexing in Series`

---

## `Indexing`
To have the access over the certain value or values in the Series.

Indexing is done by
1. Positive Indexing ( just like list in Python ): Start from 0 to forward direction
2. No Negative Indexing in Series

```python
print(avg[0])
print(avg[1])
```

```python
# print(avg[-1]) # ValueError: -1 is not in range
```

## `Slicing` : 
- In Series, wew can get the certain values over a given specific range and we can do it using 
- Syntax: [initial_value, final_value + 1]

```python
avg[7:17]
```

I want the values skipped as a certain values

```python
avg[2:21:2] # [start, end, skip]
```

### Reverse values in a Series

```python
avg[::-1]
```
