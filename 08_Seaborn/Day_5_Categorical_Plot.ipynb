- Categorical plots are used to visualize relationships involving one or more **categorical columns**.

They help answer questions like:

- Which category has the highest value?
- How is data distributed across categories?
- How do numerical values vary between categories?
- Are there outliers within categories?

---

### Why Use Categorical Plots?

Suppose we have:

| Gender | Salary |
|----------|---------|
| Male | 50000 |
| Female | 45000 |
| Male | 60000 |

Questions:

- Average salary by gender?
- Salary distribution by gender?
- Number of males vs females?

Categorical plots help answer these questions.

### `What are Categorical Plots in Seaborn?`

Categorical plots are visualizations used to analyze categorical variables and their relationship with numerical variables. Seaborn provides countplot, barplot, boxplot, violinplot, stripplot, swarmplot, and pointplot for comparing categories, distributions, outliers, and aggregated statistics.


--- 

## 1. `Bar Plot`
- Shows an aggregated statistic (mean by default) of a Numerical Value for each category .
- Syntax: `sns.barplot(data=df,x="Gender",y="Salary")`
- Parameter
    - x --> categorical column
    - y --> Numerical column
    - data --> dataframe
    - hue --> optional --> for further split by Group 
- Default Aggregation : Mean
- Use Cases

    - Average Salary by Department
    - Average CGPA by Branch

- Analysis Type: Categorical + Numerical

````python
gaps.head()
````

````python
sns.barplot(data = gaps, x='continent', y = 'lifeExp')
````

````python
sns.barplot(data = gaps, x='continent', y = 'gdpPercap')
````

````python
sns.barplot(data = student_data, x='Dietary Habits', y = 'Depression')
````

````python
sns.barplot(data= tips, x='sex', y = 'total_bill', hue='smoker')
````

---

## 2. `Count Plot`: 
- Shows the frequency (count) / occurence of each category, As, How many number of times a category is getting occured.
- Syntax: `sns.countplot(data=df, x="Gender")`
- Parameter
    - x --> categorical column
    - data --> dataframe
- Use Cases

    - Gender Distribution
    - Department Distribution
    - City Distribution

- Analysis Type: Categorical Column Only

````python
sns.countplot(data= gaps, x = 'continent')
````

````python
sns.countplot(data= tips, x = 'day')
````

````python
sns.countplot(data= student_data, x = 'Dietary Habits')
````

---


## 3. `Box Plot`: 
- Shows data distribution, spread and outliers of a Numerical Variables including Media, Quartiles, minimum, maximum, outliers for each category.

- Syntax: `sns.boxplot(data=df,x="Gender",y="Salary")` 
- Parameters 
  - x --> Category Variable
  - y --> Numerical Variable
  - data --> dataframe

- Shows

    - Minimum
    - Q1
    - Median
    - Q3
    - Maximum
    - Outliers

- Use Cases

    - Identifying / Detection Outlier only
    - Distribution Comparison

- Analysis Type: Categorical + Numerical

````python
sns.boxplot(data= student_data, x = 'CGPA')
````

````python
sns.boxplot(data = tips, x = 'total_bill')
````

````python
sns.boxplot(data= gaps, x = 'continent', y = 'lifeExp')
````

---

## 4. `Violin Plot`: 
- Shows distribution + density of data.
- Combination of Box Plot and KDE Plot and shows the Distribution Shape and Summary Statistics

- Syntax: `sns.violinplot(data=df,x="Gender",y="Salary")`
- Parameters 
  - x --> Category Variable
  - y --> Numerical Variable
  - data --> dataframe

- Advantage: More informative than Box Plot because it also shows distribution shape.

- Analysis Type: Categorical + Numerical


````python
sns.violinplot(data= tips ,x="total_bill")
````

````python
sns.violinplot(data= gaps ,x="continent", y = 'lifeExp')
````

---

### 5. Strip Plot: 
- Displays every observation as a point.

- Syntax: `sns.stripplot(data=df,x="Gender",y="Salary")`

- Advantage

    - Shows actual data points
    - Good for small datasets

- Analysis Type: Categorical + Numerical

````python
sns.stripplot(data = tips, x = 'sex', y = 'total_bill')
````

---


### 6. Swarm Plot: 
- Shows individuals data points while avoiding overlapping so we can see the density so we can see the density and  individual points
- Similar to Strip Plot but prevents overlapping points.

- Syntax: `sns.swarmplot(data=df,x="Gender",y="Salary")`
- Parameters 
  - x --> Category Variable
  - y --> Numerical Variable
  - data --> dataframe

- Advantage: Better visualization when many points exist.

- Analysis Type: Categorical + Numerical

````python
sns.swarmplot(data = chocolate_data, x = 'Country', y = 'Boxes Shipped')
````

---


### 7. Point Plot: 
- Shows aggregated values using points connected by lines.

- Syntax: sns.pointplot(data=df,x="Department",y="Salary")`

- Advantage: Easy comparison between category means.

- Analysis Type: Categorical + Numerical

````python
tips.head()
````

````python
sns.pointplot(data = tips, x = 'sex', y = 'total_bill')
````
