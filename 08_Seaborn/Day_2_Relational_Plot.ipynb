## `Relational Plot`

---

### Definition

* A **Relational Plot** is used to analyze the relationship between **two or more numerical variables**.
* It helps us understand how changes in one numerical column affect another numerical column.
* Mainly used for **Bivariate Analysis** (**Numerical vs Numerical**).
* Helps identify:

  * Relationship
  * Correlation
  * Trends
  * Patterns

### Example
- Experience ──► Salary
- Study Hours ──► Marks
- Height ──► Weight

---

### Types of Relational Plots

#### 1. Line Plot

* Used for **Time Series Analysis**.
* Shows how a value changes with respect to time.
* X-axis → Time
* Y-axis → Variable that changes over time

Syntax: `sns.lineplot(data=df,x="Month",y="Sales")`


---

#### 2. Scatter Plot

* Used to analyze the relationship between two numerical variables.
* Shows how one numerical variable affects another.
* Helps identify correlation between variables.

 Syntax: `sns.scatterplot(data=df,x="Age",y="Salary")`


---

### Common Parameters

1.  `hue`

* Groups data using colors based on another column.

2. `style`

* Changes marker or line style based on another column.

3.  `size`

* Changes marker size based on another column.

```python
size="Experience"
```

## Figure-Level Function

### `sns.relplot()`

* Figure-level function for relational analysis.
* Can create both Line Plot and Scatter Plot.

1.  Line Plot: `sns.relplot(data=df,x="Month",y="Sales",kind="line")`

2. Scatter Plot: `sns.relplot(data=df,x="Age",y="Salary",kind="scatter")`

---

### Syntax
1.  Axes-Level
- Line Plot:  `sns.lineplot(data=df, x="", y="")` 
- Scatter Plot: `sns.scatterplot(data=df, x="", y="")`
  
2. Figure-Level: `sns.relplot(data=df,x="",y="",kind="line")`  kind = 'line' or 'scatter'

# `Facet Plot`
---

- A Facet Plot is a visulization technique where the same plot is repeated for different subsets of data. Instead of displaying everything in one graph, - Seaborn creates multiple small graphs (facets) based on the values of a categorical column.

Think of it as:

- "Split the dataset into groups and draw the same chart for each group."


`What is a Facet Plot in Seaborn?`

- A Facet Plot is a visualization technique that splits data into multiple subsets based on categorical variables and draws the same type of plot for each subset. 
- In Seaborn, faceting is commonly achieved using relplot() with the row, col, and col_wrap parameters, allowing easy comparison across different categories.

### Common Faceting Parameters 
1. col: Creates facets column-wise.
   - Syntax: `sns.relplot(data=df,x="Age",y="Depression Score",col="Gender")`

2. row: Creates facets row-wise.
   - Syntax: `sns.relplot(data=df,x="Age",y="Depression Score",row="Gender")`


3. col_wrap: Wraps columns after a specified number.
    - Syntax: `sns.relplot(data=df,x="Age",y="Depression Score",col="Department",col_wrap=3)`
    - Without col_wrap, all plots would appear in one long row.

4. height: Controls the height of each facet.
   - Synatx: `sns.relplot(data=df,x="Age",y="Depression Score",col="Gender",height=4)`
   - Each subplot height = 4 inches.

4. aspect: Controls width-to-height ratio.
    - Syntax: `sns.relplot(data=df,x="Age",y="Depression Score",col="Gender",aspect=1.5)`
    - Width = Height × 1.5

    - If: height = 4
    - aspect = 1.5

    - Then:
    - Width = 4 × 1.5 = 6 inches
