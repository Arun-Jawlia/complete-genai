## Import 

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.runnables import RunnableLambda
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
```

```python
load_dotenv()
```

```python
HF_TOKEN = os.getenv("HF_TOKEN")

if not HF_TOKEN:
    print("Token is missing")
```

```python
MODEL = "meta-llama/Llama-3.1-8B-Instruct"
```

```python


client = InferenceClient(
    model=MODEL,
    token=HF_TOKEN
)
```

# TASK 1

```python
prompt = 'Explain Huggingface in Simple terms'

response = client.chat_completion(
    messages=[
        {
            "role":"user",
            "content": prompt
        }
    ],
    max_tokens=100
)


```

```python
print("response", response.choices[0].message['content'])
```

Observation:
- model gives clear explainatino

# TASK 2: 

```python
model_name = "mistralai/Mistral-7B-Instruct-v0.2"

llm = HuggingFaceEndpoint(
    repo_id=model_name,
    huggingfacehub_api_token=HF_TOKEN,
    max_new_tokens=100,
)
```

```python
prompt = PromptTemplate.from_template("Answer the following questin in simple lanugage: \n\n {question}")
```

```python
chain = prompt | llm
```

```python
question_1 = 'What is Langchain'
question_2 = 'What is Rag? why do we use?'
question_3 = 'What is Vector Database? and Name any two vector database'
question_4  = 'How do fine tune a model'
```

```python
result1 = chain.invoke({
    'question': question_1
})

print(question_1)
print(result1)
```

```python
result2 = chain.invoke({
    'question': question_2
})
print(question_2)
print(result2)
```

```python
result3 = chain.invoke({
    'question': question_3
})
print(question_3)
print(result3)
```

```python
result4 = chain.invoke({
    'question': question_4
})
print(question_4)
print(result4)
```

# TASK 3

```python
normal_prompt = PromptTemplate.from_template("Explain {topic} in simple language")
```

```python
normal_chain = normal_prompt | llm
```

```python
topic = "Transformers"
```

```python

normal_result = normal_chain.invoke({"topic": topic})
print(normal_result)
```

```python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI teacher names Alex"),
    ("human","Explain {topic} in simple language")
])
```

```python
chat_chain = chat_prompt | llm
```

```python
chat_result = chat_chain.invoke({"topic":topic}) 
```

# Comparison

```python
print("NORMAL ")
print(normal_result)
```

```python

print("CHAT")
print(chat_result)
```

Observation
- PromptTemplate is useful for creating a normal text prompt
- ChatPromptTemplate allows us to define different messages roles such as system and human, gives more control over how the model should behave
