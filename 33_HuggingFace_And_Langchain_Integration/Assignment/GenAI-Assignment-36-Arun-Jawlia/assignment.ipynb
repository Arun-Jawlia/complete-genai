## Import 

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.runnables import RunnableLambda
```

```python
load_dotenv()
```

```python
HF_TOKEN = os.getenv("HF_TOKEN")

if not HF_TOKEN:
    print("Token is missing")
```

```python
MODEL = "meta-llama/Llama-3.1-8B-Instruct"
```

```python


client = InferenceClient(
    model=MODEL,
    token=HF_TOKEN
)
```

# TASK 1

```python
prompt = 'Explain Huggingface in Simple terms'

response = client.chat_completion(
    messages=[
        {
            "role":"user",
            "content": prompt
        }
    ],
    max_tokens=100
)


```

```python
print("response", response.choices[0].message['content'])
```

# TASK 2: 

```python
def hugging_face_llm(prompt:str)->str:
    prompt = str(prompt)
    response = client.chat_completion(
        messages = [
            {
                "role":'user',
                "content":prompt
            }
        ],
        max_tokens=200
    )
    return response.choices[0].message.content
```

```python
huggingface_llm = RunnableLambda(hugging_face_llm)
```

```python
prompt_template = PromptTemplate.from_template("Explain this {question} in easy language")
```

```python
chain = prompt_template | huggingface_llm
```

```python
question_1 = 'What is Langchain'
question_2 = 'What is Rag? why do we use?'
question_3 = 'What is Vector Database? and Name any two vector database'
question_4  = 'How do fine tune a model'
```

```python
result1 = chain.invoke({
    'question': question_1
})

print(question_1)
print(result1)
```

```python
result2 = chain.invoke({
    'question': question_2
})
print(question_2)
print(result2)
```

```python
result3 = chain.invoke({
    'question': question_3
})
print(question_3)
print(result3)
```

```python
result4 = chain.invoke({
    'question': question_4
})
print(question_4)
print(result4)
```
