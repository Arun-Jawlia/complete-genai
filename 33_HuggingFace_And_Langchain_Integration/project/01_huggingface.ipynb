```python
import os
from huggingface_hub import InferenceClient

HF_TOKEN = os.getenv("HF_TOKEN")

print("Token loaded:", bool(HF_TOKEN))
print("Token prefix:", HF_TOKEN[:7] if HF_TOKEN else None)

client = InferenceClient(
    api_key=HF_TOKEN,
    provider="auto",
)

response = client.chat.completions.create(
    model="deepseek-ai/DeepSeek-V3-0324",
    messages=[
        {
            "role": "user",
            "content": "What is 2 + 2?",
        }
    ],
)

print(response.choices[0].message.content)
```

huggingface_hub: libarary let you connect to Huggingface model endpoints through python which are running on the huggingface server.

Inference client: key class used to send text/chat requests to a hosted model and get responses from models hosted on Huggingface-server.

```python
HF_TOKEN = os.getenv("HF_TOKEN")
```

```python
# HF_TOKEN
```

```python
HF_MODEL = 'deepseek-ai/DeepSeek-V3-0324'
```

```python
client = InferenceClient(
    model=HF_MODEL,
    token=HF_TOKEN
    )
```

```python
# InferenceClient: Initialize a connectin to the specific Model Hosted on huggingface server 
## Mistral 7b: chat style 
```

```python
prompt = 'What is Machine Learning?'
```

```python
response = client.chat_completion(
    model=HF_MODEL,
    messages=[
        {
            'role':"user",
            "content":prompt
        },
    ],
    max_tokens = 150,
    temperature=0.7
)
```

```python
print(response.choices[0].message.content)
```
