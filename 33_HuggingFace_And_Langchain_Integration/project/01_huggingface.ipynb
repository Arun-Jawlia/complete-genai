```python
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

huggingface_hub: libarary let you connect to Huggingface model endpoints through python which are running on the huggingface server.

Inference client: key class used to send text/chat requests to a hosted model and get responses from models hosted on Huggingface-server.

```python
HF_TOKEN = os.getenv("HF_TOKEN")
```

```python
HF_TOKEN
```

```python
HF_MODEL = 'Qwen/Qwen3-0.6B'
```

```python
client = InferenceClient(
    model=HF_MODEL,
    token=HF_TOKEN
    )
```

```python
# InferenceClient: Initialize a connectin to the specific Model Hosted on huggingface server 
## Mistral 7b: chat style 
```

```python
prompt = 'What is Machine Learning?'
```

```python
response = client.chat_completion(
    model=HF_MODEL,
    messages=[
        {
            'role':"user",
            "content":prompt
        },
    ],
    max_tokens = 150,
    temperature=0.7
)
```

```python
print(response)
```
