# `NumPy Statistical Functions`
---

```python
import numpy as np
```

## 1. `mean()`
- mean of all values inside the array and matrix
- Syntax: np.mean()

```python
arr = np.arange(1,20)
print("mean", np.mean(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Mean of Matrix: ", np.mean(mat))
print("\n")
print("Mean of Columns: ", np.mean(mat, axis= 0))
print("\n")
print("Mean of Rows: ", np.mean(mat, axis = 1))
```

## 2. `std()`
- provides with the standard deviation of all valeus of the matrix or array
- Syntax: np.std(arr)

```python
arr = np.arange(1,20)
print("Standard Deviation", np.std(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Standard Deviation of Matrix: ", np.std(mat))
print("\n")
print("Standard Deviation of Columns: ", np.std(mat, axis= 0))
print("\n")
print("Standard Deviation of Rows: ", np.std(mat, axis = 1))
```

## 3. `median()`
- help us out calculate the Median of all values in the Array or Matrix
- Syntax: `np.median(arr)`

```python
arr = np.arange(1,20)
print("Median of an array: ", np.median(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Median of Matrix: ", np.median(mat))
print("\n")
print("Median of Columns: ", np.median(mat, axis= 0))
print("\n")
print("Median of Rows: ", np.median(mat, axis = 1))
```

## 4. `var()`
- calculating the variance of all values inside the array or matrix
- Syntax: np.var()

```python
arr = np.arange(1,20)
print("Variance of an array: ", np.median(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Variance of Matrix: ", np.var(mat))
print("\n")
print("Variance of Columns: ", np.var(mat, axis= 0))
print("\n")
print("Variance of Rows: ", np.var(mat, axis = 1))
```

## `Indexing and Slicing in NumPy`

Indexing --> lists[,size-1] --> to have the access over the value

```python
one_darray = np.arange(1,15)
print(one_darray)
print(one_darray[0])
print(one_darray[2])
print(one_darray[2])
print(one_darray[-1])
```

```python
two_darray = np.round(np.random.random((5,5))*100).astype("int32")
print(two_darray)
print('\n col')
print(two_darray[0])
```

```python
print(two_darray[0,0])
print(two_darray[4,4])
# two_darray[5,5]IndexError: index 5 is out of bounds for axis 0 with size 5
```

```python
three_darray = np.arange(1,9).reshape(2,2,2)
print(three_darray)
```

```python
three_darray = np.array([[[1,2,3],[4,5,6]], [[11,12,13],[14,15,16]], [[21,22,23],[24,25,26]]])
print(three_darray)
print(three_darray[0,1,2])
print(three_darray[2,1,1])
# Indexing --> [z-axis, x-axis, y-axis]
# Indexing --> [z-axis, row, col]
```

## `Slicing`
- similar approach just like the lists
- 2D array : [row : row_2, col_1: col_2]

```python
two_darray = np.arange(1,17).reshape(4,4)
print(two_darray)
print(two_darray[:,2])
print("---")
print(two_darray[0,:])
print("---")
print(two_darray[2:4])
print("---")
print(two_darray[2:4, 3:4])
```

```python
print(two_darray[1:3]) # row 1 to 3 so row 1 and 2 print
print(two_darray[1:3, 1:3])
```

## `Iterateing over an Array`

```python
array = np.arange(1,10)

for i in array:
    print(i)
```

```python
array = np.arange(1,10).reshape(3,3)
print(array)

for i in array:
    for j in array:
        print(array[i][j])
```

```python
array = np.arange(1,28).reshape(3,3,3)
# print(array)

for i in array:
    print(i)
```

## `nditer()`: n dimension iterate function
- help us out with iterating over every value in a nd array
- Synatx: np.nditer()


1D array --> value by value on iteration

2D array --> row by row on iteration

3D array --> 2d array by 2d array

```python
array = np.arange(1,28).reshape(3,3,3)

for i in np.nditer(array):
    print(i)
```

## `Stacking and Splitting`

### `Stacking`: joining the Arrays Together
- Horizontal Stacking: Row By Row Joining is done
  - Syntax: `np.hstack()`
- Vertical Stacking: Column By Column Joining is done
  - Syntax: `np.vstack()` 

Note: Size of Arrays must be same

```python
mat_1 = np.arange(1,13).reshape(3,4)
mat_2 = np.arange(13,25).reshape(3,4)

print(mat_1)
print(mat_2)
```

```python
horizontal_stacking = np.hstack((mat_1, mat_2))
print(horizontal_stacking)
```

```python
vertical_stacking = np.vstack((mat_1, mat_2))
print(vertical_stacking)
```

## `Splitting`
- Horizontal Split --> np.hsplit(arr, number of parts you want to be splitted at)
- Vertical Split --> np.vsplit(arr, number of parts)

```python
print("Original Array")
print(vertical_stacking)
np.hsplit(vertical_stacking,2)
```

```python
print("Original Array")
print(vertical_stacking)
print(np.vsplit(vertical_stacking,2))
```
