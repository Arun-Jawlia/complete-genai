```python
import numpy as np
```

## Scalar Operations: 
- A scalar value wiht be multiplied with the array or a numerice value  ( eg 1, 4 )

```python
arr = np.random.random((3,3))*100

print(arr)
```

```python
new_arr = np.round(arr).astype('int32')
print(new_arr)
```

```python
# Scalar Operations with Our Matrix or Array

addition_operation = new_arr + 15 # Addition 
print("Add", addition_operation)

print('\n')

substract_operation = new_arr - 5 # substract 5 from all the value of matrix
print(substract_operation)

print('\n')

multiplication_operation = new_arr * 5 # multiplication 5 from all the value of matrix
print(multiplication_operation)

print('\n')

division_operation = new_arr / 5 # division 5 from all the value of matrix
print(division_operation)
```

```python
# We can appply scalar operation with the array in terms of arithmetic operations
```

```python
arr = np.random.random((3,3))*100

print(arr > 50)
print("\n")
print(arr < 50)
```

```python
# Vector Operations - Operations will be in between the given numbers of matrices

arr_1 = np.arange(1,17).reshape(4,4)
arr_2 = np.arange(17,33).reshape(4,4)

print(arr_1)
print("\n")
print(arr_2)
print("\n Addition")
print(arr_1 + arr_2)
print("\n Substraction")
print(arr_1 - arr_2)
print("\n Multiplication")
print(arr_1 * arr_2)
```

## Important Functions in NumPy for Machine Learning and DAta Science

1. `max()`
   - To calculate the max value inside a Matrix or an Array
   - Syntax: `np.max(array)`

```python
arr = np.arange(1,10)
print(np.max(arr))
```

```python
max_arr = np.arange(1,17).reshape(4,4) 
print(np.max(max_arr)) # from the matrix
```

```python
max_arr = np.round(np.random.random((4,4))*100).astype('int32')
print(max_arr)
print(np.max(max_arr))
```

axis --> provide us with the idea about any operation value inside a specific row or column or we can get a specific value row wise or column wise

axix =  0 --> column wise operation
axix =  1 --> row wise operation

```python
print(np.max(max_arr, axis = 0)) # max from columns
print("\n")
print(np.max(max_arr, axis = 1)) # max from rows
```

## 2. `min()`
- used for calculating the minimum value for the Matrix or an Array
- Syntax: `np.min(array)`

```python
arr = np.arange(1,10)
print(np.min(arr))
```

```python
min_arr = np.round(np.random.random((3,3))*100).astype("int32")
print(min_arr)
print("Minimum of matrix", np.min(min_arr))
```

```python
print(np.min(min_arr, axis = 0)) # min from columns
print("\n")
print(np.min(min_arr, axis = 1)) # min from rows
```

## 3. `sum()`
- calculates the sum of all of the values of a given matrix
- Syntax: np.sum(arr)

```python
sum_arr = np.arange(1,20).sum()
print(sum_arr)
```

```python
min_matrix = np.round(np.random.random((3,3))*100).astype("int32")
print(min_matrix)
print(np.sum(min_matrix))
```

```python
sum_matrix_row = np.round(np.random.random((3,3))*100).astype("int32") # row
print(sum_matrix_row)
print(np.sum(sum_matrix_row, axis = 1))
```

```python
sum_matrix_col = np.round(np.random.random((3,3))*100).astype("int32") # column
print(sum_matrix_col)
print(np.sum(sum_matrix_col, axis = 0))
```

## 4. `prod()`
- calculating the product of all values inside the array or matrix
- Syntax: `np.prod(arr)`

```python
arr = np.arange(1,10)
print(np.prod(arr))
```

```python
prod_matrix = np.round(np.random.random((3,3))*100).astype("int32") 
print(prod_matrix)
print('\n')
print(np.prod(prod_matrix))
```

```python
prod_matrix = np.round(np.random.random((3,3))*100).astype("int32") 
print(prod_matrix)
print('\n')
print(np.prod(prod_matrix))
print('\n')
print(np.prod(prod_matrix, axis = 0)) # column wise product
print('\n')
print(np.prod(prod_matrix, axis = 1)) # row wise product
```
