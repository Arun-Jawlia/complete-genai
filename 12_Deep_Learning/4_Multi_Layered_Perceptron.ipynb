# Multi-Layer Perceptron (MLP)

---

# `What is a Multi-Layer Perceptron (MLP)?`

## Definition

A **Multi-Layer Perceptron (MLP)** is a neural network formed by connecting **multiple perceptrons** together in multiple layers.

It is one of the simplest forms of a **Deep Neural Network (DNN)**.

> **Interview Definition**
>
> A Multi-Layer Perceptron (MLP) is a feed-forward neural network consisting of an input layer, one or more hidden layers, and an output layer that can learn **non-linear relationships** using activation functions.

---

# `Why Do We Need Multi-Layer Perceptron?`

A **single perceptron** can only learn a **linear relationship**.

Mathematically,

```
y = wx + b
```

or

```
y = mx + c
```

Both equations represent a **straight line**.

---

## Visualization

### Single Perceptron

```text
           Linear Data

      ●        ●

   ●

----------------------------

                  ●

                     ●
```

A straight line can separate the data.

✅ Single Perceptron works.

---

### Non-Linear Data

```text
           ○ ○ ○ ○

        ○         ○

           ● ●

        ○         ○

           ○ ○ ○
```

No straight line can separate these points.

❌ Single Perceptron fails.

---

# `Solution`

Instead of using **one perceptron**, we combine **multiple perceptrons**.

```text
          Perceptron

               +

          Perceptron

               +

          Perceptron

               +

          Perceptron

               │

               ▼

      Multi-Layer Perceptron
```

Multiple perceptrons together can learn highly complex patterns.

---

# `Architecture of an MLP`

```text
                Input Layer

         IQ          CGPA

          ●            ●
           \          /
            \        /
             \      /

        Hidden Layer 1

        ●    ●    ●    ●

             ││││

        Hidden Layer 2

        ●    ●    ●

             │││

         Output Layer

          Placed?

            ●
```

Unlike a single perceptron, MLP contains **one or more hidden layers**.

---

# `Why Hidden Layers?`

Hidden layers automatically learn features from data.

Each layer learns increasingly complex information.

```text
Input Data

      │

      ▼

Hidden Layer 1

Simple Features

↓

Hidden Layer 2

Intermediate Features

↓

Hidden Layer 3

Complex Features

↓

Output
```

---

# `Example (IQ & CGPA)`

Suppose we want to predict whether a student gets placed.

## Dataset

| IQ | CGPA | Placed |
|----|------|---------|
| 120 | 8.9 | ✅ |
| 85 | 7.0 | ❌ |
| 105 | 9.1 | ✅ |
| 95 | 6.2 | ❌ |

---

## Single Perceptron

The equation is

```
z = w₁(IQ) + w₂(CGPA) + b
```

Suppose,

```
w₁ = 0.05

w₂ = 0.6

b = -8
```

For a student,

```
IQ = 120

CGPA = 8.9
```

Compute

```
z

=

0.05 × 120

+

0.6 × 8.9

-

8

=

6

+

5.34

-

8

=

3.34
```

The activation function predicts

```
Placed
```

This works only if the relationship is approximately **linear**.

---

# `What if Placement Depends on Complex Rules?`

Suppose companies use rules like:

```text
IF

IQ is High

AND

CGPA is High

↓

Placed

--------------------

IF

IQ is Medium

BUT

Projects are Excellent

↓

Placed

--------------------

IF

Communication is Good

AND

Problem Solving is Excellent

↓

Placed
```

These relationships are **non-linear**.

A single equation

```
y = mx + c
```

cannot model all these conditions.

---

# `How MLP Solves This?`

Instead of one equation,

MLP creates **many equations**.

---

## Hidden Layer

Neuron 1

```
z₁

=

w₁x₁

+

w₂x₂

+

b₁
```

Neuron 2

```
z₂

=

w₃x₁

+

w₄x₂

+

b₂
```

Neuron 3

```
z₃

=

w₅x₁

+

w₆x₂

+

b₃
```

Each neuron learns a **different pattern**.

---

## Output Layer

The output neuron combines all hidden neurons.

```
Output

=

v₁z₁

+

v₂z₂

+

v₃z₃

+

b
```

This allows the network to model highly complex decision boundaries.

---

# `Mathematical Flow`

```text
Input

x₁

x₂

     │

     ▼

Neuron 1

z₁=w₁x₁+w₂x₂+b₁

     │

Neuron 2

z₂=w₃x₁+w₄x₂+b₂

     │

Neuron 3

z₃=w₅x₁+w₆x₂+b₃

     │

     ▼

Output Neuron

y=v₁z₁+v₂z₂+v₃z₃+b
```

Notice that **each neuron learns a different mathematical function**, and together they approximate complex relationships.

---

# `Role of Activation Function`

Without an activation function,

multiple linear equations remain **linear**.

Mathematically,

```
Linear

+

Linear

+

Linear

=

Linear
```

Therefore,

adding more perceptrons **without an activation function does not increase learning power**.

---

# `Activation Function Makes MLP Powerful`

Each hidden neuron applies an activation function.

```text
Weighted Sum

↓

Activation Function

↓

Non-Linearity

↓

Next Layer
```

This enables the network to learn complex patterns.

---

# `Sigmoid Activation Function`

The sigmoid function maps any real value into the range **0 to 1**.

:contentReference[oaicite:0]{index=0}

Mathematical Formula

\[
\sigma(z)=\frac{1}{1+e^{-z}}
\]

---

## Sigmoid Curve

```text
Output

1 |                         ●●●●

  |                     ●●

0.5|----------------●----------------

  |             ●

0 |______●___________________________

          Input (z)
```

### Properties

- Output lies between **0 and 1**
- Smooth and differentiable
- Useful for binary classification
- Historically used in early neural networks

> **Modern Note:** Hidden layers in today's deep networks usually use **ReLU** instead of Sigmoid because it trains faster and reduces the vanishing gradient problem. Sigmoid is still common in the **output layer** for binary classification.

---

# `Why Activation Function is Important?`

Without Activation Function

```text
Input

↓

Linear Equation

↓

Linear Equation

↓

Linear Equation

↓

Still Linear
```

---

With Activation Function

```text
Input

↓

Linear

↓

Sigmoid/ReLU

↓

Linear

↓

Sigmoid/ReLU

↓

Complex Decision Boundary
```

Now the network can solve non-linear problems.

# `What are Parameters?`

A neural network learns **parameters** during training.

Parameters are:

```text
Parameters

=

Weights

+

Biases
```

---

## Example

Suppose one neuron has

```
2 Inputs

↓

2 Weights

+

1 Bias

=

3 Parameters
```

---

Suppose a hidden layer has

```
100 Neurons

↓

Each has

11 Parameters

↓

Total

1100 Parameters
```

Deep Learning models have millions or billions of such parameters.

When people say

```
Llama 7B

or

Llama 70B
```

# `Single Perceptron vs Multi-Layer Perceptron`

| Single Perceptron | Multi-Layer Perceptron |
|-------------------|-------------------------|
| One Layer | Multiple Hidden Layers |
| Linear Decision Boundary | Non-Linear Decision Boundary |
| Solves Simple Problems | Solves Complex Problems |
| Step Activation | ReLU, Sigmoid, Tanh, Softmax |
| Few Parameters | Millions/Billions of Parameters |
| Cannot solve XOR | Can solve XOR |

# `Interview Tips ⭐`

- A single perceptron learns only **linear** relationships.
- MLP introduces hidden layers to learn **non-linear** relationships.
- Activation functions introduce **non-linearity**.
- Parameters = **Weights + Biases**.
- Llama 7B = **7 Billion Parameters**.
- Llama 70B = **70 Billion Parameters**.
- More parameters generally increase model capacity but also computational cost.

# `Key Takeaways`

- A Multi-Layer Perceptron (MLP) is a network of multiple perceptrons.
- Hidden layers enable hierarchical feature learning.
- Activation functions (ReLU, Sigmoid, etc.) make MLPs capable of modeling non-linear relationships.
- Combining multiple neurons creates complex decision boundaries that a single perceptron cannot learn.
- Large Language Models (LLMs) like Llama are essentially very large neural networks with billions of learnable parameters.

# `2. Interview Questions & Answers`

### Beginner Questions

**1. Why do we need a Multi-Layer Perceptron?**
A single perceptron can only model linear relationships. MLPs use hidden layers and activation functions to learn complex, non-linear patterns.

**2. What is the role of hidden layers?**
Hidden layers automatically learn intermediate and high-level features from the input data.

**3. What are parameters in a neural network?**
Parameters are the learnable values in a neural network: **weights + biases**.

---

### Intermediate Questions

**4. Why can't stacking linear layers without activation functions improve learning?**
Because the composition of linear functions is still a linear function. Activation functions introduce the non-linearity needed to learn complex decision boundaries.

**5. Why is ReLU preferred over Sigmoid in hidden layers?**
ReLU is computationally efficient, reduces the vanishing gradient problem, and usually speeds up training.

**6. What does Llama 70B mean?**
It indicates that the model has approximately **70 billion learnable parameters (weights and biases)**.

---

### Scenario-Based Questions

**7. You have a dataset where the classes form concentric circles. Would a single perceptron work?**
No. The data is non-linearly separable. An MLP with non-linear activation functions is required.

**8. A colleague removes all activation functions from an MLP. What happens?**
The network effectively becomes a single linear transformation, losing its ability to learn non-linear relationships.

# `4. 2-Minute Revision`

A Multi-Layer Perceptron (MLP) extends the single perceptron by adding one or more hidden layers. Each neuron computes a weighted sum, adds a bias, and applies an activation function. Hidden layers learn different mathematical representations, and activation functions introduce non-linearity, allowing the network to model complex relationships that a single perceptron cannot. The learnable components of an MLP are its **weights and biases**, collectively called **parameters**. Modern LLMs like Llama contain billions of these parameters, enabling them to model highly complex language patterns.
