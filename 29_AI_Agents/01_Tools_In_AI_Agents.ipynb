# Tools in AI Agents

Tools are one of the **most important concepts in AI Agents**.

A normal LLM can **generate text**, but an AI Agent can **take actions** by using tools.

A simple way to think about it:

> **LLM = Brain**
> **Tool = Ability**
> **Agent = Brain + Reasoning + Tools + Actions**

For example:

* LLM alone → "The weather in Delhi is probably hot."
* LLM + weather tool → "The current temperature in Delhi is 34°C."
* LLM + weather tool + calculator → "It's 34°C, which is 7°C warmer than yesterday."
* LLM + tools + reasoning → "It's too hot right now, so I recommend going out after 6 PM."

---

# 1. What is a Tool in an AI Agent?

A **tool** is an external function, API, service, database, or capability that an AI agent can invoke to perform a specific task.

For example:

```text
User
 ↓
AI Agent
 ↓
"Need current weather"
 ↓
Weather Tool
 ↓
Weather API
 ↓
Current weather data
 ↓
AI Agent
 ↓
Final Answer
```

The important point is:

**The LLM does not necessarily perform the actual operation itself.**

Instead, it decides:

> "I need to use this tool."

The tool performs the operation and returns the result.

---

# 2. Why Do AI Agents Need Tools?

LLMs have several limitations.

## Problem 1: LLM knowledge can be outdated

Suppose you ask:

> What is the current price of Bitcoin?

An LLM's training knowledge isn't inherently a live market feed.

A **financial/market tool** can retrieve the current value.

---

## Problem 2: LLMs are not calculators

Suppose:

```text
Calculate:

(12893 × 782) / 37
```

An LLM can reason about arithmetic, but using a calculator tool gives a deterministic result.

```text
Agent
 ↓
Calculator Tool
 ↓
Exact calculation
 ↓
Agent
```

---

## Problem 3: LLMs cannot directly access your database

Suppose you ask:

> Show me my last 10 orders.

The LLM doesn't automatically know your MongoDB/PostgreSQL data.

You can provide a tool:

```python
get_orders(user_id)
```

The agent can invoke it.

---

## Problem 4: LLMs cannot directly perform actions

Consider:

> Send an email to John.

The LLM can generate the email content.

But actually sending it requires an email tool:

```text
send_email(
    recipient="john@example.com",
    subject="Meeting",
    body="..."
)
```

---

# 3. Tool vs Function

These terms are closely related but have a slightly different emphasis.

### Function

A function is a piece of executable code.

```python
def add(a, b):
    return a + b
```

### Tool

A tool is a function/capability that is **exposed to an AI agent**, usually with a description and a defined input schema so the model can decide when and how to call it.

For example:

```python
def calculator(expression: str):
    ...
```

When registered as an agent tool, the model may see something conceptually like:

```text
Tool: calculator

Description:
Evaluate a mathematical expression.

Input:
expression: string
```

The LLM can then decide:

```text
"I need calculator."
```

and provide:

```json
{
  "expression": "25 * 40"
}
```

---

# 4. Tool Calling

The mechanism by which an LLM requests a tool to execute is commonly called **tool calling** or **function calling**.

The basic flow is:

```text
User
  ↓
LLM
  ↓
Does the task require a tool?
  ↓
Yes
  ↓
Tool Call
  ↓
Tool executes
  ↓
Tool Result
  ↓
LLM
  ↓
Final Response
```

For example:

```text
User:
What is 125 × 48?

        ↓

Agent:
I should use calculator.

        ↓

Calculator Tool:
125 × 48 = 6000

        ↓

Agent:
The answer is 6000.
```

---

# 5. Tool Calling Is Not the Same as the Tool

This distinction is important.

### Tool

The actual capability:

```python
calculator()
```

### Tool calling

The LLM's request to invoke that capability:

```text
Call calculator with:
expression = "125 * 48"
```

### Tool execution

The actual execution:

```python
calculator("125 * 48")
```

### Tool result

```text
6000
```

# 6. Anatomy of a Tool

A good AI tool normally has several components.

```text
Tool
├── Name
├── Description
├── Input Schema
├── Function
├── Execution
└── Output
```

Example:

```python
def get_weather(city: str):
    ...
```

Conceptually:

```text
Name:
get_weather

Description:
Get the current weather for a city.

Input:
city: string

Output:
weather information
```

The **description is extremely important**.

The LLM uses the description to determine when the tool should be used.

# 25. Tool Calling Loop

Real agents often work in a loop.

```text
             ┌───────────────┐
             │     User      │
             └───────┬───────┘
                     ↓
             ┌───────────────┐
             │      LLM      │
             └───────┬───────┘
                     ↓
             Need a tool?
                /       \
              No         Yes
              ↓           ↓
          Final       Tool Call
          Answer          ↓
                     Tool Executes
                          ↓
                     Tool Result
                          ↓
                         LLM
                          ↓
                    Need another tool?
                      /       \
                    Yes       No
                     ↓         ↓
                   Tool      Answer
```

This is one of the fundamental patterns behind agentic systems.

# 28. Built-in Tools in LangChain

Since you are learning LangChain, this is particularly important.

LangChain provides a tool abstraction that allows functions and external capabilities to be exposed to models/agents.

Conceptually:

```python
from langchain_core.tools import tool

@tool
def calculator(expression: str) -> str:
    """Calculate a mathematical expression."""
    return str(eval(expression))
```

The important part is:

```python
@tool
```

It converts the function into a tool representation that can be provided to an agent/model.

The docstring is also important:

```python
"""Calculate a mathematical expression."""
```

because the model uses the tool metadata to understand the tool's purpose.

# 30. Structured Tools

For complex tools, structured inputs are better.

For example:

```python
def search_products(
    category: str,
    min_price: float,
    max_price: float
):
    ...
```

The model can provide:

```json
{
    "category": "laptop",
    "min_price": 500,
    "max_price": 1000
}
```

This is safer and more reliable than asking the model to construct an arbitrary string.

---

# 31. Tool Schema

A tool should clearly define its input.

For example:

```json
{
  "name": "search_products",
  "description": "Search products by category and price range",
  "parameters": {
    "type": "object",
    "properties": {
      "category": {
        "type": "string"
      },
      "min_price": {
        "type": "number"
      },
      "max_price": {
        "type": "number"
      }
    },
    "required": [
      "category"
    ]
  }
}
```

This schema gives the model a structured interface.

# 37. Tool Design Best Practices

When creating custom tools:

### 1. Keep tools focused

Bad:

```text
do_everything()
```

Better:

```text
get_customer()
get_orders()
create_order()
cancel_order()
```

---

### 2. Give tools clear names

Bad:

```text
process()
```

Better:

```text
search_customer_orders()
```

---

### 3. Write precise descriptions

Bad:

```text
"""Get data."""
```

Better:

```text
"""Retrieve all orders placed by a customer
within a specified date range."""
```

---

### 4. Use structured input

Prefer:

```json
{
    "customer_id": 123,
    "start_date": "2026-08-01",
    "end_date": "2026-08-20"
}
```

over:

```text
"123 from August 1 to August 20"
```

---

### 5. Validate input

Never blindly trust model-generated arguments.

```python
if amount <= 0:
    raise ValueError("Amount must be positive")
```

---

### 6. Return useful errors

Instead of:

```text
Error
```

return:

```text
Customer ID 123 was not found.
```

This gives the agent useful information for recovery.

# 39. Tools vs Agents

Another important distinction:

### Tool

Decides what to do and which tools to use.

```text
Agent
 ├── calculator
 ├── web_search
 ├── database
 └── email
```

Therefore:

> **Tool = capability**
> **Agent = decision-maker that can use capabilities**

# 40. Tools vs Chains

In LangChain terminology:

### Chain

Usually follows a predefined workflow.

```text
Input
 ↓
Prompt
 ↓
LLM
 ↓
Output Parser
 ↓
Output
```

### Agent

Can dynamically choose actions/tools.

```text
Input
 ↓
Agent
 ↓
Choose Tool A
 ↓
Result
 ↓
Reason
 ↓
Choose Tool B
 ↓
Result
 ↓
Final Answer
```

This dynamic decision-making is what makes agents more flexible.

# 45. Interview Definition

If an interviewer asks:

### "What are tools in AI agents?"
