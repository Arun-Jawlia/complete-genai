# Introduction

Before building any Deep Learning model, we must decide several important components.

Think of building a Deep Learning model like constructing a building.

```text
               Build a House
                    │
      ┌─────────────┼─────────────┐
      │             │             │
Foundation      Materials      Design

                    ↓

          Build Deep Learning Model

                    │
      ┌─────────────┼─────────────┐
      │             │             │
 Data        Neural Network    Training
```

A Deep Learning model is **not just a Neural Network**.

It consists of several interconnected components.

---

# `Complete Deep Learning Pipeline`

```text
                    Dataset
                       │
                       ▼
              Data Preprocessing
                       │
                       ▼
           Feature Selection/Input
                       │
                       ▼
        Design Neural Network Architecture
                       │
        (ANN / CNN / Transformer)
                       │
                       ▼
             Initialize Parameters
             (Weights & Biases)
                       │
                       ▼
            Forward Propagation
                       │
                       ▼
               Calculate Loss
                       │
                       ▼
              Backpropagation
                       │
                       ▼
                  Optimizer
                       │
                       ▼
             Update Parameters
                       │
                       ▼
           Repeat for Many Epochs
                       │
                       ▼
               Trained Model
```

# `Step 5: Loss Function`

## Definition

A Loss Function measures **how wrong the model's prediction is**.

```text
Prediction

↓

Compare

↓

Actual Value

↓

Loss
```

Example

```text
Actual

Cat

Prediction

Dog

↓

High Loss
```

Examples of Loss Functions

- Mean Squared Error (Regression)
- Binary Cross Entropy
- Categorical Cross Entropy

---

# `Step 6: Backpropagation`

Backpropagation computes how much each parameter contributed to the loss.

```text
Prediction

↓

Loss

↓

Backpropagation

↓

Gradients
```

Backpropagation tells the optimizer **how to update weights**.

---

# `Step 7: Optimizer`

The optimizer updates the parameters to reduce loss.

```text
Loss

↓

Optimizer

↓

Update Weights

↓

Better Prediction
```

Popular Optimizers

- Gradient Descent
- Stochastic Gradient Descent (SGD)
- Adam
- AdaGrad

# `Gradient Descent`

Gradient Descent is the most fundamental optimization algorithm.

```text
          Loss

           ▲

           │\

           │ \

           │  \

           │   \

           │    ●

           │   /

           │  /

           │ /

───────────┼──────────────────► Weights

      Gradient Descent

Moves Towards Minimum Loss
```

Goal

```text
High Loss

↓

Low Loss

↓

Best Model
```

---

# `Relationship Between Backpropagation and Optimizer`

```text
Forward Propagation

↓

Prediction

↓

Loss Function

↓

Backpropagation

↓

Gradient

↓

Optimizer

↓

Update Weights

↓

Repeat
```

### Easy Way to Remember

- **Loss Function** → Measures error.
- **Backpropagation** → Calculates gradients.
- **Optimizer** → Updates weights.

# `Interview Tips ⭐`

- Number of **input neurons = number of input features**.
- Binary classification → 1 output neuron.
- Multi-class classification → Multiple output classes (often implemented with a probability distribution over classes).
- Parameters = Weights + Biases.
- Hyperparameters are **not learned** during training.
- Loss measures error.
- Backpropagation computes gradients.
- Optimizer updates parameters.
- GPT is based on the **Transformer architecture** with Attention and MLP layers.

---

# `Key Takeaways`

- Building a Deep Learning model involves much more than designing a neural network.
- Choose the correct input layer based on the number of features.
- Select the output layer according to the prediction task.
- Learn parameters (weights and biases) during training.
- Use a loss function to measure prediction error.
- Backpropagation computes gradients, while optimizers update the parameters.
- Hyperparameter tuning helps improve model performance.
- Modern LLMs like GPT and Llama are built using stacked Transformer blocks with billions of learnable parameters.

# `2. Interview Questions & Answers`

## Beginner Questions

### 1. What are the basic requirements for building a Deep Learning model?

**Answer:**

* Dataset
* Input features
* Neural network architecture
* Weights and biases
* Loss function
* Optimizer
* Backpropagation
* Hyperparameters

---

### 2. How do you decide the number of input neurons?

**Answer:**
The number of input neurons should equal the number of input features.

---

### 3. What are parameters in a neural network?

**Answer:**
Parameters are the learnable values of the model: **weights and biases**.

---

### 4. What is the purpose of a loss function?

**Answer:**
It measures the difference between the predicted output and the actual output.

---

## Intermediate Questions

### 5. Explain the difference between a parameter and a hyperparameter.

| Parameter                 | Hyperparameter                              |
| ------------------------- | ------------------------------------------- |
| Learned during training   | Set before training                         |
| Examples: weights, biases | Examples: learning rate, epochs, batch size |

---

### 6. What is the relationship between backpropagation and an optimizer?

**Answer:**

* **Backpropagation** computes gradients of the loss with respect to each parameter.
* **Optimizer** uses those gradients to update the parameters and reduce the loss.

---

### 7. Why is Adam more popular than Gradient Descent?

**Answer:**
Adam adapts the learning rate for each parameter, converges faster, and works well on a wide variety of deep learning problems.

---

## Scenario-Based Questions

### 8. You have a dataset with 15 input features. How many input neurons are needed?

**Answer:**
15 input neurons.

---

### 9. You're building a language model like ChatGPT. Which architecture would you choose?

**Answer:**
A Transformer architecture with Attention mechanisms and stacked Multi-Layer Perceptrons (MLPs).

---

### 10. Your model's loss is not decreasing during training. What components would you investigate?

**Answer:**

* Learning rate
* Optimizer choice
* Loss function
* Data quality
* Model architecture
* Hyperparameters
* Gradient issues (e.g., vanishing/exploding gradients)

# `4. 2-Minute Revision`

A Deep Learning model requires defining the input and output layers, selecting an appropriate architecture (ANN, CNN, Transformer, or Diffusion), initializing learnable parameters (weights and biases), and training them through forward propagation, loss computation, backpropagation, and optimization. The number of input neurons equals the number of input features, while the output layer depends on the prediction task. Hyperparameters such as learning rate and batch size control the training process but are not learned. Large Language Models like GPT and Llama are Transformer-based networks containing billions of learnable parameters.
