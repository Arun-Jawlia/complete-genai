# Backpropagation Example (Step-by-Step with IQ, CGPA & Salary Prediction)

---

# `Introduction`

In the previous chapter, we learned the theory of **Backpropagation**.

Now, let's understand **how Backpropagation actually works** using a simple regression example.

We will predict the **Salary Package (LPA)** using

- IQ
- CGPA

---

# `Dataset`

| IQ | CGPA | Package (LPA) |
|----|------|---------------|
| 80 | 8 | 3 |
| 60 | 9 | 5 |
| 70 | 9 | 5 |
|120 | 7 |11 |

Input Features

```text
IQ

CGPA
```

Target

```text
Package (LPA)
```

Since the output is a **continuous value**,

this is a **Regression Problem**.

Loss Function

```
Mean Squared Error (MSE)
```

---

# `Neural Network Architecture`

```text
                 Input Layer

          IQ               CGPA

           ●                 ●
          / \               / \
         /   \             /   \

      w11     w12      w21     w22

       \       /        \      /

        ▼     ▼          ▼    ▼

         Hidden Layer (2 Neurons)

          H1             H2

           ●-------------●

           |             |

         w31           w32

           |             |

           ▼             ▼

              Output Layer

                  ŷ

          Predicted Package
```

Every connection has

- Weight
- Bias

Example

```text
Hidden Neuron

↓

Weighted Sum

↓

Bias

↓

Activation

↓

Output
```

---

# Step 1 — Initialize Weights & Biases

Initially,

the neural network knows nothing.

So,

we assign **random (or simple) initial values**.

For understanding,

let

```text
All Weights = 1

All Biases = 1
```

Example

```text
w11 = 1

w12 = 1

w21 = 1

w22 = 1

w31 = 1

w32 = 1

Biases

b1 = 1

b2 = 1

b3 = 1
```

These values are **not correct**.

Training will improve them.

---

# Step 2 — Select One Training Sample

Backpropagation works on one training sample at a time (or a batch of samples).

Choose

| IQ | CGPA | Actual Package |
|----|------|----------------|
|80|8|3|

```text
Input

IQ

=

80

CGPA

=

8

Actual Package

=

3 LPA
```

---

# Step 3 — Forward Propagation

The input travels from **left to right**.

```text
Input

↓

Hidden Layer

↓

Output Layer

↓

Prediction
```

Suppose the network predicts

```text
ŷ

=

18 LPA
```

Actual value

```text
y

=

3 LPA
```

Clearly,

```text
Prediction

>

Actual
```

The model has made a large error.

---

# Step 4 — Calculate Loss

Since this is a regression problem,

we use **Mean Squared Error (MSE)**.

Formula


$\boxed{
L=(y-\hat{y})^2
}$


Substitute the values

```text
Actual

=

3

Prediction

=

18
```

$\[
L=(3-18)^2
\]$

$\[
=(-15)^2
\]$

$\[
L=225
\]$

Loss = **225**

This is a **large loss**, indicating the prediction is far from the actual value.

---

# Step 5 — What Should the Model Do?

The model asks:

```text
Why is the loss so large?

↓

Which weight caused it?

↓

Which bias caused it?

↓

How much should each parameter change?
```

The Loss Function **cannot answer these questions**.

This is where **Backpropagation** begins.

---

# Step 6 — Apply Backpropagation

Backpropagation sends the error **backward**.

```text
Prediction

↓

Loss

↓

Output Layer

↓

Hidden Layer

↓

Input Layer
```

Every neuron receives a **gradient**.

Gradient means

```text
How much did I contribute

to the loss?
```

---

# Chain Rule

Suppose

```text
IQ

↓

Hidden Layer

↓

Output

↓

Loss
```

The IQ node does **not directly affect** the loss.

It affects

Hidden Layer

↓

Output

↓

Loss

So,

Backpropagation applies the **Chain Rule** to calculate how each weight indirectly influences the final loss.

```text
Loss

↓

Output Layer

↓

Hidden Layer

↓

Input Layer
```

The blame is shared **layer by layer**.

---

# Step 7 — Compute Gradients

The gradient tells us

```text
How much

Loss

changes

when

Weight

changes.
```

Mathematically


$\boxed{
\frac{\partial L}{\partial W}
}$


Similarly,

for the bias

\[
$\boxed{
\frac{\partial L}{\partial b}
}$
\]

A large gradient means the parameter strongly influenced the loss and should be updated more.

---

# Step 8 — Update the Weights

Once gradients are available,

the optimizer updates every weight.

Formula

\[
$\boxed{
W_{new}=W_{old}-\eta\frac{\partial L}{\partial W}
}$
\]

Where

- $W_{old}$ = Current weight
- $W_{new}$ = Updated weight
- \(\eta\) = Learning Rate
* $\boxed{\frac{\partial L}{\partial W}}$ = Gradient

Suppose

```text
Current Weight

=

1

Learning Rate

=

0.1

Gradient

=

2
```

Then

\[
$W_{new}=1-(0.1\times2)=0.8$
\]

The weight decreases because the prediction was too high.

---

# Step 9 — Update the Bias

Bias is updated in exactly the same way.

Formula


$\boxed{
b_{new}=b_{old}-\eta\frac{\partial L}{\partial b}
}$

Example

```text
Current Bias

=

1

Learning Rate

=

0.1

Gradient

=

1.5
```

\[
$b_{new}=1-(0.1\times1.5)=0.85$
\]

---

# `Step 10 — Repeat the Process`

The updated weights are used again.

```text
Updated Weights

↓

Forward Pass

↓

New Prediction

↓

Loss

↓

Backpropagation

↓

Update Again
```

This process repeats for **every training sample** across many **epochs**.

Gradually,

```text
Prediction

18

↓

12

↓

7

↓

4

↓

3.1

↓

3
```

The prediction becomes closer to the actual value.

---

# `Complete Backpropagation Flow`

```text
Dataset

↓

Select One Training Sample

↓

Initialize Weights & Biases

↓

Forward Propagation

↓

Prediction (ŷ)

↓

Calculate Loss

↓

Backpropagation

↓

Compute Gradients

↓

Update Weights

↓

Update Biases

↓

Repeat

↓

Better Prediction
```

---

# `Why Does the Formula Use a Minus Sign?`

```text
Weight

↓

Prediction Too High

↓

Loss High

↓

Reduce Weight

↓

Prediction Decreases

↓

Loss Reduces
```

The minus sign moves the weight in the direction that reduces the loss.

# `Interview Tips ⭐`

- Backpropagation starts **after the loss is computed**.
- The forward pass predicts; the backward pass computes gradients.
- Regression problems commonly use **MSE Loss**.
- Gradients indicate how much each parameter contributed to the error.
- Weights and biases are updated using the learning rate.
- The process repeats until the loss becomes sufficiently small.

---

# `Key Takeaways`

- Backpropagation is a step-by-step procedure for reducing prediction error.
- Training starts with randomly initialized weights and biases.
- A forward pass generates predictions, and the loss function measures the error.
- Backpropagation computes gradients using the Chain Rule.
- The optimizer updates weights and biases to minimize the loss.
- Repeating this cycle over many epochs enables the neural network to learn accurate predictions.

# `2. Interview Questions & Answers`

## Beginner Questions

### 1. Why are weights initialized randomly?

**Answer:**
Because the neural network has no prior knowledge before training. Random initialization helps different neurons learn different features.

---

### 2. Which loss function is used in this example?

**Answer:**
Mean Squared Error (MSE), because predicting salary (LPA) is a regression problem.

---

### 3. When does Backpropagation begin?

**Answer:**
Backpropagation starts **after the loss has been computed** from the model's prediction.

---

### 4. What does a gradient tell us?

**Answer:**
A gradient indicates how much a small change in a weight or bias will affect the loss.

---

## Intermediate Questions

### 5. Why do we update both weights and biases?

**Answer:**
Both weights and biases influence the model's predictions. Updating both allows the network to better fit the training data.

---

### 6. Why is the learning rate important?

**Answer:**
The learning rate controls the size of each parameter update. Too small leads to slow learning, while too large can make training unstable.

---

### 7. Why is the Chain Rule required?

**Answer:**
Because hidden-layer weights influence the loss indirectly through later layers. The Chain Rule computes these indirect effects so gradients can be propagated backward.

---

## Scenario-Based Questions

### 8. Your model predicts 18 LPA when the actual salary is 3 LPA. What happens next?

**Answer:**
The loss becomes large, backpropagation computes gradients, and the optimizer updates the weights and biases to reduce future prediction errors.

---

### 9. After many epochs, your predictions become much closer to the actual values. What does this indicate?

**Answer:**
It indicates that the weights and biases have been successfully optimized, resulting in a lower loss and a better-performing model.

---

### 10. If the gradients are very large, what might happen?

**Answer:**
The parameter updates may become too large, causing unstable training. Techniques such as learning rate tuning or gradient clipping can help address this.
