# Introduction

In the previous chapter, we learned about **Gradient Descent**, which updates the weights to reduce the loss.

Now the question is:

> **When should the weights be updated?**

Should we

- update after seeing **all data**?
- update after **every data point**?
- or update after **a small group of data points**?

Based on this, Gradient Descent is divided into three types.

Batch Gradient Descent uses the **entire training dataset** to compute the loss and update the weights **once per epoch**.

---

# Batch Gradient Descent vs Stochastic Gradient Descent

| Batch Gradient Descent | Stochastic Gradient Descent |
|-------------------------|-----------------------------|
| Uses the entire dataset | Uses one sample at a time |
| Batch Size = Dataset Size | Batch Size = 1 |
| One update per epoch | One update per sample |
| Stable updates | Noisy updates |
| Requires more memory | Memory efficient |

# `Speed Comparison`

Training speed (processing one epoch)

```text
Batch GD

>

Mini Batch GD

>

Stochastic GD
```

Reason:

- Batch operations are highly optimized on CPUs/GPUs.
- Mini-batches also benefit from parallel computation.
- SGD performs an update after every sample, introducing more update overhead.

---

# `Convergence Comparison`

Convergence (reaching a good solution)

```text
Stochastic GD

>

Mini Batch GD

>

Batch GD
```

Why?

- SGD updates weights frequently, helping it explore the loss surface and often escape shallow local minima.
- Mini Batch GD balances stability and exploration.
- Batch GD makes fewer updates and can converge more slowly in practice.

# `Interview Tips ⭐`

- **Batch Size** = Number of samples processed before one weight update.
- Batch Gradient Descent updates weights **once per epoch**.
- Stochastic Gradient Descent updates weights **after every sample**.
- Mini Batch Gradient Descent updates weights **after every mini-batch**.
- TensorFlow and PyTorch use **Mini Batch Gradient Descent** by default.
- Mini Batch GD is the most commonly used approach in modern Deep Learning because it balances computational efficiency and convergence.

---

# `Key Takeaways`

- Gradient Descent can be performed using the entire dataset, one sample, or a mini-batch of samples.
- Batch size determines how frequently the model updates its parameters.
- Batch Gradient Descent is stable but performs infrequent updates.
- Stochastic Gradient Descent performs frequent updates but produces noisier gradients.
- Mini Batch Gradient Descent combines the strengths of both methods and is the industry standard.
````

---

# `2. Interview Questions & Answers`

## Beginner Questions

### 1. What is batch size?

**Answer:**
Batch size is the number of training samples processed before the model performs one weight update.

---

### 2. What is Batch Gradient Descent?

**Answer:**
Batch Gradient Descent computes the loss using the entire training dataset and updates the weights once per epoch.

---

### 3. What is Stochastic Gradient Descent?

**Answer:**
Stochastic Gradient Descent updates the weights after processing each individual training sample (batch size = 1).

---

### 4. What is Mini Batch Gradient Descent?

**Answer:**
Mini Batch Gradient Descent processes a small batch of samples before updating the weights, combining the advantages of Batch GD and SGD.

---

## Intermediate Questions

### 5. If a dataset has 100 samples and the batch size is 20, how many weight updates occur in one epoch?

**Answer:**
[
100 \div 20 = 5
]

So there are **5 weight updates** in one epoch.

---

### 6. Why is Mini Batch Gradient Descent preferred in practice?

**Answer:**
It balances memory usage, computational efficiency, and stable convergence, making it well-suited for GPU-based training.

---

### 7. Which frameworks use Mini Batch Gradient Descent by default?

**Answer:**
Popular frameworks such as **TensorFlow** and **PyTorch** use Mini Batch Gradient Descent by default.

---

## Scenario-Based Questions

### 8. You have a dataset of 10,000 samples and batch size 100. How many updates occur in one epoch?

**Answer:**
[
10000 \div 100 = 100
]

So there are **100 updates per epoch**.

---

### 9. What happens when the batch size is 1?

**Answer:**
The training process becomes **Stochastic Gradient Descent**, updating the weights after every sample.

---

### 10. What happens when the batch size equals the entire dataset?

**Answer:**
The training process becomes **Batch Gradient Descent**, updating the weights once after processing all samples in an epoch.

---

# `3. 30-Second Revision`

* **Batch Size** = Number of samples before one weight update.
* **Batch GD** → Entire dataset → One update per epoch.
* **SGD** → One sample → One update per sample.
* **Mini Batch GD** → Small batch → One update per batch.
* TensorFlow and PyTorch use **Mini Batch Gradient Descent** by default.

---

# `4. 2-Minute Revision`

The three types of Gradient Descent differ in **when** they update the model parameters. Batch Gradient Descent processes the entire dataset before updating the weights, making it stable but infrequent. Stochastic Gradient Descent updates the weights after every individual sample, allowing frequent learning but producing noisier updates. Mini Batch Gradient Descent processes small groups of samples before updating the weights, providing a practical balance between speed, memory efficiency, and convergence. This is why modern deep learning frameworks such as TensorFlow and PyTorch use Mini Batch Gradient Descent as the default training strategy.

---
