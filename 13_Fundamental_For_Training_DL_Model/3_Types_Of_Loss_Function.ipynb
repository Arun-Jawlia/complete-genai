# Introduction

In Deep Learning, the **Loss Function** measures the error for **one training example**, while the **Cost Function** measures the average error over the **entire dataset**.

```text
                Training Dataset

        ┌─────────┬─────────┬─────────┐
        │ Sample1 │ Sample2 │ Sample3 │
        └─────────┴─────────┴─────────┘
              │         │         │
              ▼         ▼         ▼
          Loss₁      Loss₂     Loss₃
              │         │         │
              └─────────┼─────────┘
                        ▼
                 Average Error
                        │
                        ▼
                  Cost Function
```

---

# Loss Function vs Cost Function

## `Loss Function`

### Definition

A **Loss Function** calculates the error for **one training example**.

```text
One Training Sample

↓

Prediction

↓

Actual Value

↓

Loss
```

Example

Suppose

```text
Actual Price

=

₹20 Lakhs

Prediction

=

₹18 Lakhs
```

Loss

```
20 - 18

=

2 Lakhs
```

This is the error for **one sample**.

---

# `Cost Function`

## Definition

A **Cost Function** calculates the **average loss over all training samples**.

```text
Sample 1

↓

Loss = 2

-------------------

Sample 2

↓

Loss = 5

-------------------

Sample 3

↓

Loss = 1

-------------------

Average

↓

Cost Function
```

---

# `Mathematical Difference`

### Loss Function

For a single example,

```text
Loss

=

Actual

-

Prediction
```

More generally,

```
Loss = L(y,\hat{y})
```

---

### Cost Function

The Cost Function is the average of all individual losses.

For Mean Squared Error,

\[
J=$\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y_i})^2$
\]

Where,

- **n** = Number of training samples
- **y** = Actual value
- **ŷ** = Predicted value

---

# `Difference Between Loss and Cost`

| Loss Function | Cost Function |
|---------------|---------------|
| Error for one sample | Average error for all samples |
| Used for individual predictions | Used to evaluate the whole model |
| Calculated many times | One overall value for the dataset |

---

# `1. Regression Loss Functions`

Regression predicts **continuous numerical values**.

Examples

- House Price Prediction
- Salary Prediction
- Temperature Prediction
- Stock Price Prediction

---

## Mean Squared Error (MSE)

### Definition

Mean Squared Error measures the average of the squared differences between actual and predicted values.

Formula

\[
$\boxed{MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y_i})^2}$
\]

---

## Working of MSE

```text
Actual Value

↓

Prediction

↓

Difference

↓

Square Error

↓

Average

↓

MSE
```

---

### Example

| Actual | Predicted | Error | Error² |
|---------|-----------|--------|---------|
| 10 | 8 | 2 | 4 |
| 20 | 18 | 2 | 4 |
| 30 | 29 | 1 | 1 |

```
MSE

=

(4+4+1)

/

3

=

3
```

---

## Why Do We Square the Error?

```text
Errors

+2

-2

↓

Square

↓

4

4
```

Advantages

- Removes negative signs
- Penalizes large errors more heavily
- Smooth and differentiable

---

## Mean Absolute Error (MAE)

Instead of squaring,

MAE takes the absolute value of the error.

Formula

\[
$\boxed{MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y_i}|}$
\]

---

## Working of MAE

```text
Actual

↓

Prediction

↓

Difference

↓

Absolute Value

↓

Average

↓

MAE
```

---

### Example

| Actual | Predicted | Error | Absolute Error |
|---------|-----------|--------|----------------|
| 10 | 8 | 2 | 2 |
| 20 | 18 | 2 | 2 |
| 30 | 29 | 1 | 1 |

```
MAE

=

(2+2+1)

/

3

=

1.67
```

---

## MSE vs MAE

| MSE | MAE |
|------|-----|
| Squares the error | Uses absolute error |
| Penalizes large errors heavily | Treats all errors equally |
| Smooth gradients | Less sensitive to outliers |
| Common in Deep Learning | Easy to interpret |

# `2. Classification Loss Functions`

# `Binary Classification`

Example

```text
Email

↓

Spam ?

↓

Yes / No
```

Only **two classes** exist.

```text
Output Layer

       ●

Probability

0 to 1
```

For binary classification, we typically use:

- **1 Output Neuron**
- **Sigmoid Activation**
- **Binary Cross Entropy Loss**

---

# `Binary Cross Entropy (Log Loss)`

## Definition

Binary Cross Entropy measures the difference between the predicted probability and the actual binary label.

Formula

\[
$\boxed{
L=-\left[y\log(\hat{y})+(1-y)\log(1-\hat{y})\right]
}$
\]

Where,

- **y** = Actual label (0 or 1)
- **ŷ** = Predicted probability

---

## `Binary Classification Pipeline`

```text
Input

↓

Neural Network

↓

Sigmoid Function

↓

Probability (0–1)

↓

Binary Cross Entropy

↓

Loss
```

---

# `Why Sigmoid?`

The sigmoid function converts any real number into a probability.

```text
Input

-∞ ------------------ +∞

↓

Sigmoid

↓

0 -------------------- 1
```

Output Range

```
0 ≤ Probability ≤ 1
```

This makes it ideal for binary classification.

---

# `Why is Binary Cross Entropy Better Than MSE?`

Suppose

```text
Actual

=

1

Prediction

=

0.01
```

```text
MSE

↓

Small Gradient

↓

Slow Learning
```

```text
Binary Cross Entropy

↓

Large Penalty

↓

Faster Learning
```

Binary Cross Entropy provides stronger learning signals when predictions are confidently wrong.

---

# `Is Binary Cross Entropy Differentiable?`

✅ Yes.

Binary Cross Entropy is differentiable, making it suitable for **Backpropagation**.

```text
Prediction

↓

Loss

↓

Derivative

↓

Gradient

↓

Backpropagation

↓

Update Weights
```

Differentiability allows gradient-based optimization algorithms such as Gradient Descent and Adam to work effectively.

---

# `GPT and Loss Function`

GPT predicts the next token from a very large vocabulary.

```text
Current Sentence

↓

Transformer

↓

Softmax Layer

↓

Probability

↓

Cross Entropy Loss
```

Unlike binary classification,

GPT uses **Categorical Cross Entropy** (implemented with Softmax) because there are **many possible output classes (tokens)**.

# `Interview Tips ⭐`

- **Loss Function** → Error for one sample.
- **Cost Function** → Average error over the dataset.
- Regression → MSE or MAE.
- Binary Classification → Sigmoid + Binary Cross Entropy.
- Multi-Class Classification → Softmax + Categorical Cross Entropy.
- Binary Cross Entropy is differentiable and works well with gradient-based optimization.

---

# `Key Takeaways`

- Loss measures the error for a single training example; cost is the average loss across all examples.
- MSE and MAE are commonly used for regression tasks.
- Binary Cross Entropy is the standard loss for binary classification.
- Sigmoid activation outputs probabilities between 0 and 1, making it suitable for Binary Cross Entropy.
- GPT uses **Softmax + Categorical Cross Entropy** because it predicts one token from a large vocabulary.

# `2. Interview Questions & Answers`

## Beginner Questions

### 1. What is the difference between a Loss Function and a Cost Function?

**Answer:**

* **Loss Function** measures the error for a single training sample.
* **Cost Function** is the average of the losses over the entire training dataset.

---

### 2. Which loss function is commonly used for regression?

**Answer:**
Mean Squared Error (MSE) and Mean Absolute Error (MAE).

---

### 3. Which loss function is used for binary classification?

**Answer:**
Binary Cross Entropy (also called Log Loss).

---

### 4. Why is Sigmoid used with Binary Cross Entropy?

**Answer:**
Sigmoid converts the output into a probability between **0 and 1**, which Binary Cross Entropy requires.

---

## Intermediate Questions

### 5. Why is MSE not preferred for binary classification?

**Answer:**
MSE can produce smaller gradients when predictions are very wrong, leading to slower learning. Binary Cross Entropy provides stronger gradients and faster convergence.

---

### 6. What is the difference between Binary Cross Entropy and Categorical Cross Entropy?

| Binary Cross Entropy | Categorical Cross Entropy                     |
| -------------------- | --------------------------------------------- |
| Two classes          | Multiple classes                              |
| Uses Sigmoid         | Uses Softmax                                  |
| Spam vs Not Spam     | GPT token prediction, ImageNet classification |

---

### 7. Why is Binary Cross Entropy differentiable?

**Answer:**
Its mathematical form is smooth and differentiable, allowing gradients to be computed during backpropagation.

---

## Scenario-Based Questions

### 8. You're building a house price prediction model. Which loss function would you choose?

**Answer:**
Mean Squared Error (MSE) or Mean Absolute Error (MAE), because the target is a continuous value.

---

### 9. You're building a cat-vs-dog classifier. Which activation and loss functions would you use?

**Answer:**

* **Output Activation:** Sigmoid
* **Loss Function:** Binary Cross Entropy

---

### 10. You're building a next-word prediction model like GPT. Which loss function would you use?

**Answer:**
Categorical Cross Entropy (with a Softmax output layer), since the model predicts one token from a large vocabulary.

# `4. 2-Minute Revision`

Loss Functions quantify the prediction error for individual training examples, while the Cost Function averages these errors across the dataset. For regression tasks such as house price or salary prediction, Mean Squared Error (MSE) and Mean Absolute Error (MAE) are commonly used. For binary classification tasks like spam detection or cat-vs-dog classification, Binary Cross Entropy combined with a Sigmoid output layer is the standard choice. Multi-class tasks, including next-token prediction in GPT, use Softmax with Categorical Cross Entropy. Choosing the correct loss function is essential for stable and effective model training.
