# What is Multi-Class Classification?

## Definition

A **Multi-Class Classification** problem is one where the model must choose **exactly one class from three or more possible classes**.

Unlike Binary Classification (2 classes), Multi-Class Classification deals with **multiple output categories**.

# `Softmax Activation Function`

## Why Do We Need Softmax?

The neural network produces raw scores called **logits**.

Example

```text
Coffee

2.5

Tea

1.3

Juice

0.8

Soft Drink

0.4
```

These are **not probabilities**.

Softmax converts them into probabilities.

---

## Softmax Formula

\[
$\boxed{
P_i=\frac{e^{z_i}}{\sum_{j=1}^{n}e^{z_j}}
}$
\]

Where

- $z_i$ = Logit for class *i*
- $P_i$ = Probability of class *i*

---

## Softmax Diagram

```text
Raw Scores

↓

Coffee = 2.5

Tea = 1.3

Juice = 0.8

Soft Drink = 0.4

↓

Softmax

↓

Coffee = 0.62

Tea = 0.18

Juice = 0.12

Soft Drink = 0.08

↓

Total

0.62+0.18+0.12+0.08

=

1
```

### Properties of Softmax

- Converts logits into probabilities.
- Output values are between **0 and 1**.
- The probabilities always sum to **1**.
- Used in the output layer for multi-class classification.

# `Categorical Cross Entropy`

## Definition

Categorical Cross Entropy measures the difference between the **predicted probability distribution** and the **true one-hot encoded distribution**.

Formula


$\boxed{
L=-\sum_{i=1}^{n}y_i\log(\hat{y_i})
}$



where

- $y_i$ = Actual one-hot encoded label
- $hat{y_i}$ = Predicted probability
---

```text
Input

↓

Neural Network

↓

Softmax

↓

Probability Distribution

↓

Compare

↓

One-Hot Label

↓

Categorical Cross Entropy

↓

Loss
```

```text
Input

↓

Neural Network

↓

Softmax

↓

Predicted Probabilities

↓

Class Index

↓

Sparse Categorical Cross Entropy

↓

Loss
```

---

# `Why Use Sparse Categorical Cross Entropy?`

Advantages

✅ No One-Hot Encoding required.

✅ Uses less memory.

✅ Faster for datasets with many classes.

✅ Commonly used when the number of categories is very large.

Examples

- GPT Vocabulary (50,000+ tokens)
- Language Translation
- ImageNet (1000 classes)

---

# `Categorical vs Sparse Categorical Cross Entropy`

| Categorical Cross Entropy | Sparse Categorical Cross Entropy |
|----------------------------|----------------------------------|
| Requires One-Hot Encoding | No One-Hot Encoding |
| Labels are vectors | Labels are integer class IDs |
| Slightly more memory | More memory efficient |
| Same mathematical objective | Same mathematical objective |
| Uses Softmax | Uses Softmax |

# `GPT Training Pipeline`

```text
Input Sentence

↓

Transformer

↓

Attention

↓

Multi-Layer Perceptrons

↓

Softmax

↓

Probability of Every Token

↓

Categorical / Sparse Categorical Cross Entropy

↓

Loss

↓

Backpropagation

↓

Optimizer

↓

Update Parameters
```

In practice, GPT-style language models usually use **Sparse Categorical Cross Entropy (or an equivalent indexed cross-entropy implementation)** because the vocabulary contains tens of thousands of tokens, making integer class labels much more memory efficient than one-hot vectors.

# `Interview Tips ⭐`

- Multi-Class Classification predicts **one class from many**.
- Softmax is the standard activation function for the output layer.
- Categorical Cross Entropy requires One-Hot Encoding.
- Sparse Categorical Cross Entropy uses integer labels instead of One-Hot Encoding.
- GPT predicts one token from a very large vocabulary using a Softmax output layer and cross-entropy loss.

---

# `Key Takeaways`

- Multi-Class Classification handles problems with more than two classes.
- Softmax converts logits into probabilities that sum to one.
- Categorical Cross Entropy compares predicted probabilities with one-hot encoded labels.
- Sparse Categorical Cross Entropy is a memory-efficient alternative that uses integer labels.
- Large language models like GPT rely on Softmax and cross-entropy loss for next-token prediction.

# `2. Interview Questions & Answers`

## Beginner Questions

### 1. What is Multi-Class Classification?

**Answer:**
It is a classification problem where the model predicts one class from three or more possible classes.

---

### 2. Which activation function is used for Multi-Class Classification?

**Answer:**
**Softmax**.

---

### 3. Which loss function is used for Multi-Class Classification?

**Answer:**
**Categorical Cross Entropy** or **Sparse Categorical Cross Entropy**.

---

### 4. What is One-Hot Encoding?

**Answer:**
It represents each class as a binary vector where only the correct class is `1` and all other classes are `0`.

Example:

```
Dog

↓

[0 1 0 0]
```

---

## Intermediate Questions

### 5. Why is Softmax preferred over Sigmoid for Multi-Class Classification?

**Answer:**
Softmax produces a probability distribution across all classes, ensuring the probabilities sum to **1**, whereas Sigmoid treats each output independently.

---

### 6. What is the difference between Categorical and Sparse Categorical Cross Entropy?

| Categorical Cross Entropy                 | Sparse Categorical Cross Entropy                  |
| ----------------------------------------- | ------------------------------------------------- |
| Uses One-Hot Encoded labels               | Uses integer class labels                         |
| Higher memory usage                       | Lower memory usage                                |
| Suitable for small to medium class counts | Ideal for very large vocabularies or many classes |

---

### 7. Why is Sparse Categorical Cross Entropy commonly used in GPT?

**Answer:**
GPT has a vocabulary containing tens of thousands of tokens. Storing labels as integer indices is far more memory-efficient than storing one-hot vectors.

---

## Scenario-Based Questions

### 8. You're building a handwritten digit recognizer (0–9). Which activation and loss functions would you choose?

**Answer:**

* **Activation:** Softmax
* **Loss:** Categorical Cross Entropy (or Sparse Categorical Cross Entropy if labels are integers)

---

### 9. You're building a language model like GPT. Which output activation and loss function are appropriate?

**Answer:**

* **Output Activation:** Softmax
* **Loss Function:** Sparse Categorical Cross Entropy (or an equivalent indexed cross-entropy implementation)

---

### 10. You have 100,000 output classes. Which loss function is more practical?

**Answer:**
**Sparse Categorical Cross Entropy**, because it avoids one-hot encoding and significantly reduces memory usage.

# `4. 2-Minute Revision`

Multi-Class Classification is used when a model must choose one class from several possible categories. The output layer uses the **Softmax** activation function to convert logits into a probability distribution whose values sum to one. Training commonly uses **Categorical Cross Entropy**, which compares the predicted probability distribution with a one-hot encoded target label. For tasks with very large numbers of classes—such as GPT's vocabulary—**Sparse Categorical Cross Entropy** is preferred because it uses integer class labels instead of one-hot vectors, making training more memory-efficient while optimizing the same objective.
