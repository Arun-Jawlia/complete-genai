# What is Prompting and Prompt Template

## 1. What is Prompting?

**Prompting** is the process of giving instructions or input to an AI model to get the desired output.

A **prompt** tells the LLM:

* What task to perform
* What information to use
* How to respond
* What format the output should have

### Simple Example

```text
Prompt:
Explain Retrieval-Augmented Generation in simple terms.
```

The LLM processes this instruction and generates an appropriate response.

### Real-World Example

Suppose we are building a **customer support application**.

```text
Prompt:
You are a customer support assistant.
Answer the customer's question politely and concisely.

Customer Question:
{question}
```

If the user asks:

```text
My order has not arrived yet.
```

The model can generate:

```text
I'm sorry for the delay. Please provide your order ID so I can help
you check the delivery status.
```

---

## 2. What is Prompt Engineering?

**Prompt Engineering** is the practice of designing effective prompts to get reliable and useful outputs from an LLM.

A good prompt usually specifies:

1. **Role** – Who should the AI act as?
2. **Task** – What should it do?
3. **Context** – What information should it consider?
4. **Constraints** – What should it avoid?
5. **Output Format** – How should the answer be structured?

### Example

Instead of:

```text
Explain Python.
```

A better prompt is:

```text
You are a Python instructor.

Explain Python decorators to a beginner.

Requirements:
- Use simple language.
- Give one practical example.
- Explain the example step-by-step.
- Keep the answer under 300 words.
```

This gives the model much clearer instructions.

---

# 3. What is a Prompt Template?

A **Prompt Template** is a reusable prompt containing **variables/placeholders**.

Instead of writing a new prompt every time, we create a template and dynamically insert values into it.

### Normal Prompt

```text
Explain Python decorators in simple terms.
```

### Prompt Template

```text
Explain {topic} in simple terms.
Give one practical example.
```

Here:

```text
{topic}
```

is a **variable**.

If:

```text
topic = "Python decorators"
```

the final prompt becomes:

```text
Explain Python decorators in simple terms.
Give one practical example.
```

---

# 4. Why Use Prompt Templates?

Prompt templates are useful because they make prompts:

* **Reusable**
* **Dynamic**
* **Maintainable**
* **Consistent**
* **Easy to integrate into applications**

### Example

A single template can handle:

```text
topic = "RAG"
topic = "LangChain"
topic = "Vector Database"
topic = "Embeddings"
```

without changing the application code.

---

# 5. Prompt Template in LangChain

LangChain provides `ChatPromptTemplate` for creating prompts that can be dynamically populated with variables.

### Basic Example

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple terms."
)

final_prompt = prompt.invoke({
    "topic": "RAG"
})

print(final_prompt)
```

The `{topic}` variable is replaced with `"RAG"`.

# 10. Create Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    """
    You are an expert computer science teacher.

    Explain the following topic in simple terms:

    Topic: {topic}

    Requirements:
    - Give a clear definition.
    - Explain the concept step-by-step.
    - Give a practical example.
    - Mention important interview points.
    """
)
```

Here:

```text
{topic}
```

is a dynamic variable.

---

# 11. Connect Prompt with Model

LangChain allows us to create a **chain** using the pipe operator `|`.

```python
chain = prompt | model
```

The flow becomes:

```text
Input
  ↓
Prompt Template
  ↓
OpenAI Chat Model
  ↓
Response
```

This is an example of **LangChain Expression Language (LCEL)**.

---

# 12. Invoke the Chain

```python
response = chain.invoke({
    "topic": "Vector Database"
})

print(response.content)
```

The prompt template receives:

```python
{
    "topic": "Vector Database"
}
```

and produces a formatted prompt.

The OpenAI model then generates the response.

---

# 13. Complete Application

```python
from dotenv import load_dotenv

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# Load environment variables
load_dotenv()

# Create OpenAI Chat Model
model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

# Create Prompt Template
prompt = ChatPromptTemplate.from_template(
    """
    You are an expert computer science teacher.

    Explain the following topic in simple terms:

    Topic: {topic}

    Requirements:
    - Give a clear definition.
    - Explain the concept step-by-step.
    - Give a practical example.
    - Mention important interview points.
    """
)

# Create Chain
chain = prompt | model

# Take user input
topic = input("Enter a topic: ")

# Invoke chain
response = chain.invoke({
    "topic": topic
})

# Print response
print("\nAI Response:\n")
print(response.content)
```

### Example Input

```text
Enter a topic: RAG
```

### Application Flow

```text
"RAG"
  ↓
{topic}
  ↓
ChatPromptTemplate
  ↓
Formatted Prompt
  ↓
ChatOpenAI
  ↓
AIMessage
  ↓
response.content
```
