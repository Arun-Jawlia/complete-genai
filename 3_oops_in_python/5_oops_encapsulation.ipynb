# `Encapsulation`
---

- Wrapping data (variables) and methods (functions) together into a single unit called a class.
- It also means: Restricting direct access to data components's which is a means of preventing accidental inference and misuse and controlling it through methods.
Simple Meaning

Encapsulation is:

1. data hiding
2. data protection
3. controlled access to data


### Real World Analogy: ATM Machine Example

When using an ATM:

- you can withdraw money
- deposit money
- check balance

But you cannot directly access:

- bank database
- server logic
- internal calculations

Internal details are hidden and protected.

That is encapsulation.

### `Access Specifiers of Encapsulation in Python`:
1. Public Members
2. Protected Members
3. Private Members

### 1. `Public Members`
- Accessible from Anywhere or Attributes are accessible outside of the class
- These members are accessible by the derived class
- public member attribute is directly declared without any specific keyword

```python
class Student:

    def __init__(self):
        self.name = "Arun"

s = Student()

print(s.name)
```

```python
class Bank_Details:
    def __init__(self, balance, name, email):
        self.balance = balance
        self.name = name
        self.email = email

    def bankdetails(self):
        print(f"My name is {self.name} and email is {self.email} and Bank balance is {self.balance}")

myAcc = Bank_Details(100000, 'Arun', "arunkumar@gmail.com")
print(myAcc.name)
print(myAcc.email)
print(myAcc.balance)

# balance, bankdetails, email, name are Accessible from AnyWhere or outside the class
```

```python
dir(myAcc)
```

### 2. `Private Members`
- Not Accesible outside the class and not even in the derived class 
- in order to have the access over these private members or to bring some modifications on these members we use `getter and setter` methods.

- Private Members are declared by using double underscore before the name `__anyvariable or method`

```python
class Bank_Details:
    def __init__(self, balance, name, email, phone):
        self.name = name
        self.email = email
        self.__balance = balance
        self.__phone = phone

    def __bankdetails(self):
        print(f"My name is {self.name} and email is {self.email} and Bank balance is {self.balance}")

myAcc = Bank_Details(100000, 'Arun', "arunkumar@gmail.com", "9718653508")
print(myAcc.name)
print(myAcc.email)
# print(myAcc.__balance) # AttributeError: 'Bank_Details' object has no attribute '__balance'
# print(myAcc.__phone) # AttributeError: 'Bank_Details' object has no attribute '__phone'

# balance, bankdetails, email, name are Accessible from AnyWhere or outside the class
```

```python
dir(myAcc)
```

### Getter and Setter Methods

Encapsulation often uses:

1. getters : Used to read private data.
2. setters : Used to modify private data safely.

to control access.

Why Use Getter and Setter?
They help:

1. validate data
2. secure data
3. prevent invalid modification

```python
class Bank_Details:
    def __init__(self, balance, name, email, phone):
        self.name = name
        self.email = email
        self.__balance = balance
        self.__phone = phone

    # getter method
    def show_balance(self):
        print(f"My Bank Balance is {self.__balance}")
    
    # setter method
    def add_balance(self, amount):
        if not amount >= 0:
            print("Add a valid amount")
            return 
        new_balance = self.__balance + int(amount)
        self.__balance = new_balance
    

myAcc = Bank_Details(2000, 'Arun', "arunkumar@gmail.com", "9718653508")
print(myAcc.name)
print(myAcc.email)
myAcc.add_balance(1000) 
myAcc.show_balance()
myAcc.add_balance(4000)
myAcc.show_balance()

# balance, bankdetails, email, name are Accessible from AnyWhere or outside the class
```

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):

        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def show_balance(self):
        print("Balance:", self.__balance)

acc = BankAccount(10000)

acc.deposit(5000)

acc.withdraw(3000)

acc.show_balance()
```
