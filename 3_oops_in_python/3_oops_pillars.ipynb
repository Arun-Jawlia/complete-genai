## `Inheritance`
- Fundamental concept in OOPs because of which the child class is able to inherit the properties, attributes and functionalities of the parent class.
- Inheritance allows one class to acquire: properties and methods from another class.

| Term         | Meaning                          |
| ------------ | -------------------------------- |
| Parent Class | Base class / Superclass          |
| Child Class  | Derived class / Subclass         |
| Inheritance  | Acquiring properties from parent |

- eg. Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparents and definitely in this hierarchy the grandchild can have some features different from the parents and grandparent as well

Benefits of Inheritance
1. code reuse
2. avoids duplication
3. better organization
4. hierarchical structure


### 3 Types of Inheritance
1. Single Inheritance
2. Multilevel Inheritance
3. Multiple Inheritance

### 1. `Single Inheritance`
- The Features or the Attributes from one parent class will be getting inherited into one child class.
- Parent Class ( Properties --> Attributes and Functionalities) ------> Child Class
- One child inherits one parent.

```python
class A:
    pass

class B(A):
    pass
```

```python
class Avenger:
    def __init__(self, name, power):
        self.name = name
        self.power = power

    def introduction(self):
        print(f"Hello my name is {self.name}")

class IronMan(Avenger):
    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon = weapon
    
    def weapon_details(self):
        print(f"My powerfull weapon is {self.weapon}")


iron_man = IronMan("Iron Man", "Genius Intellect", "Mark 46 Iron Man Suit")
iron_man.introduction()
iron_man.weapon_details()
```
