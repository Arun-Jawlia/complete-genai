## Data Exploration Before Plotting Graphs

### `What is Data Exploration?`
- Data Exploration is the process of understanding our dataset
- before performing analysis or creating visualizations.

### During data exploration, we identify:
- Number of rows and columns
- Data types of columns
- Missing values
- Duplicate values
- Numerical columns
- Categorical columns

### Why Identify Numerical and Categorical Columns Before Plotting?

- Different types of data require different types of graphs.
- Choosing the correct graph helps us gain meaningful insights
- and avoid misleading visualizations.

### Therefore, before plotting any graph, we first identify:
1. Numerical Columns
2. Categorical Columns

---

## `Categorical Columns`: 
- contain labels, groups, or categories rather than measurable numeric values.
  
Characteristics
- Represent classes or groups.
- Arithmetic operations are not meaningful.
- Used for classification and grouping.
- Usually contain text values.

Real-World Examples
- Gender → Male, Female
- City → Delhi, Mumbai, Chennai
- Department → HR, IT, Finance
- Blood Group → A+, B+, O+, AB+

Questions We Can Answer:
- How many males and females are there?
- Which department has the most employees?
- Which product category generates the highest sales?

Common Graphs for Categorical Data:
- Bar Plot
- Pie Chart
- Count Plot


### Types of Categorical Data

#### A. `Nominal Data1`: No ordering exists.

Examples:
- Apple
- Samsung
- Nokia
- Google

#### B. `Ordinal Data`: Categories have an order.

| Rating    |
| --------- |
| Poor      |
| Average   |
| Good      |
| Excellent |


## `Numerical Columns`: 
- contain quantitative values that can be measured and used in mathematical calculations.

### Characteristics
- Arithmetic operations are possible.
- Can calculate:
    - Sum
    - Average
    - Median
    - Variance
    - Standard Deviation

Real-World Examples
- Age
- Salary
- Marks
- Height
- Weight
- Temperature
- Number of Sales

### Questions We Can Answer:
- What is the average age?
- What is the highest salary?
- How does sales change over time?

### Common Graphs for Numerical Data:
- Histogram
- Line Plot
- Scatter Plot
- Box Plot

---

## `Comparison between Categorical and Numerical columns`
| Feature                 | Categorical          | Numerical            |
| ----------------------- | -------------------- | -------------------- |
| Data Type               | Labels/Groups        | Numbers              |
| Mathematical Operations | Not Meaningful       | Meaningful           |
| Example                 | Male, Female         | 18, 25, 30           |
| Purpose                 | Classification       | Measurement          |
| Visualization           | Bar Chart, Pie Chart | Histogram, Line Plot |

## `Univariate, Bivariate, and Multivariate Analysis`

---
## 1. `Univariate Analysis`:
- Uni = One, Univariate Analysis means analyzing only one column (variable) at a time. 
- Analysis or will be creating visualization for a givne unique
- Purpose:
   - Understand the distribution of data
   - Find minimum and maximum values
   - Find central tendency (mean, median, mode)
   - Detect outliers
   - Understand frequency of categories
  
- Examples:
   - Age of students
   - Salary of employees
   - Gender distribution
   - Product categories

   Dataset Example: Age--> 18,20,23,25,28,30

- Questions We Can Answer:
   - What is the average age?
   - What is the age distribution?
   - Which age occurs most frequently?

-  Common Graphs:
   - Numerical Column:
   - Histogram
   - Box Plot

- Categorical Column:
   - Bar Plot
   - Pie Chart

## 2.  `Bivariate Analysis`:
- Bi = Two
- bivariate Analysis means analyzing the relationship between two columns.

- Purpose:
   - Understand relationships between two variables
   - Compare categories
   - Find correlations
   - Identify patterns
   
- Types:
   1. Numerical vs Numerical
   2. Numerical vs Categorical
   3. Categorical vs Numerical
   4. Categorical vs Categorical

- Example:

   | Age   |   Salary |
   |-------|----------|
   | 20    |   25000  |
   | 22    |   30000  |
   | 25    |   40000  |
   | 28    |   50000  |

- Questions We Can Answer:
  - Does salary increase with age?
  - Is there a relationship between age and salary?

-  Another Example:
   |Gender |   Salary |
   |-------|----------|
   |Male   |   40000  |
   |Female |   45000  |

- Questions:
   - Which gender earns more on average?
   - How does salary vary across genders?

- `Common Graphs`:

1. Numerical vs Numerical:
- Scatter Plot
- Line Plot

2. Categorical vs Numerical:
- Bar Plot
- Box Plot

3. Categorical vs Categorical:
- Grouped Bar Plot
- Stacked Bar Plot

---


## 3. `Multivariate Analysis`:
   - Multi = More Than Two
   - Multivariate Analysis means analyzing three or more variables simultaneously.

   Purpose
      1. Discover complex relationships.
      2. Build predictive models.
      3. Understand interactions among variables.

- Example:

    | Age |   Salary  |  Experience |   Gender |
    |-----|-----------|-------------|----------|
    | 22  |   30000   |  1          |   Male   |
    | 25  |   45000   |  3          |   Female |
    | 30  |   70000   |  7          |   Male   |

- Questions We Can Answer:
   - How do Age and Experience affect Salary?
   - Does Gender influence Salary?
   - Which factors contribute most to Salary?

- Common Graphs:
   - Pair Plot
   - Heatmap
   - Bubble Chart
   - 3D Scatter Plot
   - Correlation Matrix
  
---

| Analysis Type | Number of Columns | Example                             |
| ------------- | ----------------- | ----------------------------------- |
| Univariate    | 1                 | Age Distribution                    |
| Bivariate     | 2                 | Experience vs Salary                |
| Multivariate  | 3 or More         | Age, Experience, Education → Salary |

## `Functions`

### 1. `plt.figure()`
- parameter --> figsize --> (length, breadth) --> figure dimension of our graph
- Synatx: plt.figure(figsize=(10,6))

```python
plt.figure(figsize=(10,6))
plt.plot(x,y)
```

### 2. `plt.title()`
- title() --> takes the value as the Title of your plot that what it is representing as what insight we are getting through this graph
- Parameter --> Color --> What color the Title is going to have

```python
plt.figure(figsize=(10,4))
plt.title("Count of Population as per Gender", color='blue')
plt.plot(x,y)
```

### 3. `plt.ylabel()`
- represent that What Y axis is representing to us

```python
plt.figure(figsize=(10,4))
plt.title("Count of Population as per Gender", color='blue')
plt.ylabel("count of Population")
plt.plot(x,y)
```

### 4. `plt.xlabel()`
- represent that What X axis is representing to us

```python
plt.figure(figsize=(10,4))
plt.title("Count of Population as per Gender", color='blue')
plt.ylabel("count of Population")
plt.xlabel('Gender')
plt.plot(x,y)
```

### 5. `plt.grid()` : helps to bring grid on our graph

```python
plt.figure(figsize=(10,4))
plt.title("Count of Population as per Gender", color='blue')
plt.ylabel("count of Population")
plt.xlabel('Gender')
plt.grid()
plt.plot(x,y)
```

### 6. `plt.show()`: helps in representing the chart 

```python
plt.figure(figsize=(10,4))
plt.title("Count of Population as per Gender", color='blue')
plt.ylabel("count of Population")
plt.xlabel('Gender')
plt.plot(x,y)
plt.grid()
plt.show()
```

### 7. `plt.xlim()`: X values --> [10,1000] --> [100,1000] --> [700,800]
- it puts a limitation that how much data mush be represented on the X-Axis

### 8. `plt.ylim()`: X values --> [500,900] --> [600,700]
- it puts a limitation that how much data mush be represented on the Y-Axis

## Customization For Line 
Parameters:
1. color --> decide the color of line. eg. red, gree, blue, or hash code of color
2. marker --> how point of intersection appears eg. o - (circle), ^ - (arrow), s ( square) , +, d, h , p
3. linestyle --> define the style of the line. eg. `'-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'`
4. linewidth --> the width of the line eg. 1, 1.5
5. markersize --> change the size of the marker eg.  6, 10
6. label --> label that what this specific line is representing but is only helpful while plotting multiple lines in a single visualization

```python
plt.figure(figsize=(10,4))
plt.title("Count of Population as per Gender", color='blue')
plt.ylabel("count of Population")
plt.xlabel('Gender')
plt.plot(x,y, color='Red', marker='o', linestyle='--', linewidth=3, markersize=10)
plt.grid()
plt.show()
```
