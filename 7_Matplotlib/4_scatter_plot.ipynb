# `Scatter Plot`
---

Difference between Line Plot vs Scatter Plot
- Line Plot ---> Time Series Anaylysis --> How values are varying as per Time
- Scatter Plot ---> Userd for Bivariate Analysis ---> Two Numerical Columns , we check that based on the changes in one Numerical Column Value --> How it affects the other Numerical Column Values.

eg. age--> 10,15,20,25,30,40,50...
cholestrol level --> 50,60,75,65,80...
- Scatter plot --> will figure out that based on the increment in Age Value --> If chloestrol level is Increasing or Decreasing or is steady
- Scatter plot --> figures out the correlation between the two Numerical Values

```python
# Packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Age Vs Average CGPA

```python
student_data = df.groupby('Age')['CGPA'].mean().head(10)
```

```python
x = student_data.index
```

```python
y = student_data.values
```

## Function to create Scatter Plot
- Syntax: `plt.scatter(x, y)`

```python
plt.figure(figsize=(10,5))
plt.title("Average CGPA variation as per AGE")
plt.xlabel("Age")
plt.ylabel("Average CGPA")
plt.scatter(x,y, color='red', marker="o")
plt.grid()
```

# Example 3

```python
chocolate_data = pd.read_csv('Chocolate Sales.csv')
```

```python
chocolate_data.sample(5)
```

```python
chocolate_data.info()
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].str.strip("$")
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].str.replace(",", '')
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].astype("float").astype("int32")
```

```python
chocolate_data['Amount']
```

```python
chocolate_data['Boxes Shipped'] = chocolate_data['Boxes Shipped'].astype('int32')
```

```python
x = chocolate_data['Boxes Shipped'] 
y = chocolate_data['Amount']
```

```python
plt.figure(figsize=(10,5))
plt.title("Total Amount Variation as Per Total Boxes Shipped")
plt.xlabel("Number of Boxes")
plt.ylabel("Amount")
plt.scatter(x,y, color='red', marker="o")
```
