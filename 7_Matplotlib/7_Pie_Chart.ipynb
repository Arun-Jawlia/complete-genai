# `Pie Chart`
---

Pie chart is being used for Bivariate Analysis and is being used for relating Categorical Numerical Columns But in Pie Chart , we will look to the Percentage Distribution ( Univariate Analysis ) and Bivariate Analysis( Categorical - Numerical) as in terma of Category by Category Percentage Distribution in Aggregation of Numeric Data.

- Pie Chart ---> Bivariate Analysis
- Categorical-Numerical Column
- Numerical Aggregation as per categories
- Percentage Contribution of Categories in Total Total Numeric Value

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
student_data = pd.read_csv("Student_depression_dataset.csv")
```

```python
chocolate_data = pd.read_csv("Chocolate Sales.csv")
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].str.strip("$")
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].str.replace(",", '')
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].astype("float").astype("int32")
```

```python
chocolate_data['Date'] = pd.to_datetime(chocolate_data['Date'], format='%d/%m/%Y')
```

```python
chocolate_data['Date']= chocolate_data['Date'].dt.month_name()
```

```python
chocolate_data['Date']= chocolate_data['Date'].astype("category")
```

### Univariate Analysis --> Categorical Column value data contirbution in percentage

```python
data = student_data['Dietary Habits'].value_counts()
```

```python
data
```

```python
plt.figure(figsize=(12,6))
plt.pie(data, labels = data.index, autopct='%0.1f%%', colors=['red', 'yellow', 'blue', 'black'])
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.pie(data, labels = data.index, autopct='%0.1f%%', explode=[0.1,0,0,0])
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.pie(data, labels = data.index, autopct='%0.1f%%', explode=[0.1,0,0,0], shadow=True)
plt.legend()
plt.show()
```

## Chocolate Sales

```python
d1 = chocolate_data.groupby('Country')['Amount'].sum()
```

```python
plt.figure(figsize=(12,6))
plt.title("Percentage Contribution of Countries in Total Sales")
plt.pie(d1, labels = d1.index, autopct='%0.1f%%')
plt.legend()
plt.show()
```
