# `Bar Plot`
---

Bar Plot --> being used for Categorical Columns - Numerical Columns

#### Line Chart ---> Line chart was being used for representing Time Series Analysis
#### Scatter Plot ---> Scatter Plot is used to represent the corelation in between Two Numerical
Columns (Such that How on changes in the first numerical column it affects the variation in the
second Numerical Column)
#### Bar Plot ---> Bar Plot is being used for Categorical - Numerical Columns
### Data---> Sales in Multiple Countries ---> group the rows on the basis of Country Name & will
come up with the Total Sum of Sales
#### 1000 records ---> 80 records ---> India ---> group all 80 rows or records --> Sales value
associated with these 80 rows ---> Aggregation Operation (sum(), count(), mean())
#### Categorical Column ----> Numerical Column Aggregation
#### groupby --> is something we will using Everytime I
#### Univariate & Bivariate Both Analysis will be done in Bar Plot
#### Univariate ---> Categorical Column Data Distribution (how the data in Multiple Categories
is Distributed )
#### Use case ---> Aggregate Analysis of Groups |

## Plotting a Bar Plot
- Syntax: `plt.bar(x,y, color)`

### `Example 1`

## `Example 4`

### Gender --> Equal Distribution vs Academic Pressure

```python
data = pd.read_csv('student_depression_dataset.csv')
```

```python
data['Gender'].value_counts()
```

```python
Male = data[data['Gender']== 'Male'].head(12354)
```

```python
Female = data[data['Gender']== 'Female']
```

```python
Female
```

## `concat()` : function joins two dataset or Multiple dataset by using concat()
- Syntax: `pd.concat([df_1, df_2, df_3])` 

```python
df = pd.concat([Male, Female])
```

```python
df['Gender'].value_counts()
```

```python
data = df.groupby('Gender')['Academic Pressure'].sum()
```

```python
x = data.index
y = data.values
```

```python
plt.figure(figsize=(4,4))
plt.title("Gender based Academic Pressure")
plt.xlabel("Gender")
plt.ylabel("Total Academic Pressure")
plt.bar(x,y, color='red', width=0.5)
plt.show()
```

## Chocolate Sales Data
---

```python
chocolate_data = pd.read_csv('Chocolate Sales.csv')
```

```python
chocolate_data.head()
```

### Country Based Total Sales

```python
chocolate_data['Amount'] = chocolate_data['Amount'].str.strip("$")
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].str.replace(",", '')
```

```python
chocolate_data['Amount'] = chocolate_data['Amount'].astype("float").astype("int32")
```

```python
data = chocolate_data.groupby("Country")['Amount'].sum()
```

```python
x = data.index
y = data.values
```

```python
plt.figure(figsize=(10,4))
plt.title("Country Based Total Sales")
plt.xlabel("Country")
plt.ylabel("Total Sales")
plt.bar(x,y, color='green', width=0.5)
plt.show()
```
