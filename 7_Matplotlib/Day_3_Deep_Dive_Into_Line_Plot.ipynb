## `groupy()` Function
- Dataset: country --> sales info
- groupby()---> Dataset ---> Country ---> Sales info

Country |Sales|
--------|------|
| India | 150  |
| China | 5000 |
| USA   | 1000 |  
| India | 500  |
| USA'  | 4000 |

Total Sales I am getting Country wise
- group 1 --> 'India' ---> 5150
- group 2---> 'China' ----> 5000
- group 3---->'USA' ----> 5000

groupby --> helps to group the rows based on row values and the values associateed with the group, onto those values some mathematical function can be applied like sum(), mean() etc 

- Sytax: df.groupby("categorical column")['Numerical Column"].mathematical_funcation()
- df.groupby("Country")['Sales'].sum()


## `Line Plot ( Age vs CGPA )`

```python
df['Age'].value_counts().head(10)
```

```python
data = df.groupby("Age")['CGPA'].mean().head(10)
```

```python
x = data.index
y = data.values

plt.figure(figsize=(10,5))
plt.title("CGPA Distribution as per Age")
plt.xlabel("Age")
plt.ylabel("CGPA")
plt.plot(x,y, color='red', linestyle='--', linewidth=2, marker='o', markersize=6)
plt.grid() 
plt.show()
```

```python
sales = pd.read_csv("Chocolate Sales.csv")
```

```python
sales['Sales Person'] = sales['Sales Person'].astype('category') # to change datatype of Sales Person
```

```python
sales['Product'] = sales['Product'].astype("category")
```

## To convert a Date Column to datetime
- Use `pd.to_datetime()` 
- Extract year value: `df['column_name].dt.year()`
- Extract month value: `df['column_name].dt.month()`
- Extract month name: `df['column_name].dt.month_name()`

```python
sales['Date'] = pd.to_datetime(sales['Date'], format='%d/%m/%Y')
```

```python
sales['Date']= sales['Date'].dt.month_name()
```

```python
sales['Date']= sales['Date'].astype("category")
```

```python
sales['Date']
```

```python
sales['Amount'] = sales['Amount'].str.strip("$")
```

```python
sales['Amount'] = sales['Amount'].str.replace(",", '')
```

```python
sales['Amount'] = sales['Amount'].astype("float").astype("int32")
```

```python
sales['Amount']
```

```python
sales['Boxes Shipped'] = sales['Boxes Shipped'].astype('int32')
```

```python
sales.info()
```

```python
# Group Date by Amount
sales_data = sales.groupby('Date')["Amount"].sum()
```

```python
sales_data.index = pd.CategoricalIndex(sales_data.index, categories=['January', 'February','March',"April", "May", "June", "July", "August"], ordered=True)
```

```python
sales_data = sales_data.sort_index()
```

```python
x = sales_data.index
```

```python
x
```

```python
y =sales_data.values
```

```python
y
```

```python
plt.figure(figsize=(10,5))
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.plot(x,y, linewidth=2, linestyle='--', color="red", marker='o')
plt.grid()
plt.show()
```
