# Difference Between Numerical and Categorical Columns

In Data Analysis and Machine Learning, data is generally divided into two major types:

## 1. Categorical Columns

Categorical columns contain **labels, groups, or categories** rather than measurable numeric values.

### Characteristics

* Represent classes or groups.
* Arithmetic operations are not meaningful.
* Used for classification and grouping.
* Usually contain text values.

### Examples

| Student | Gender |
| ------- | ------ |
| 1       | Male   |
| 2       | Female |
| 3       | Male   |

| Student | Mobile Brand |
| ------- | ------------ |
| 1       | Apple        |
| 2       | Samsung      |
| 3       | Nokia        |

### Real-World Examples

* Gender → Male, Female
* City → Delhi, Mumbai, Chennai
* Department → HR, IT, Finance
* Blood Group → A+, B+, O+, AB+

### Types of Categorical Data

#### A. Nominal Data

No ordering exists.

Examples:

* Apple
* Samsung
* Nokia
* Google

#### B. Ordinal Data

Categories have an order.

Examples:

| Rating    |
| --------- |
| Poor      |
| Average   |
| Good      |
| Excellent |

Here:

Poor < Average < Good < Excellent

---

## 2. Numerical Columns

Numerical columns contain **quantitative values** that can be measured and used in mathematical calculations.

### Characteristics

* Arithmetic operations are possible.
* Can calculate:

  * Sum
  * Average
  * Median
  * Variance
  * Standard Deviation

### Examples

| Student | Age |
| ------- | --- |
| 1       | 18  |
| 2       | 19  |
| 3       | 20  |

| Product | Price |
| ------- | ----- |
| A       | 500   |
| B       | 1200  |
| C       | 2500  |

### Real-World Examples

* Age
* Salary
* Height
* Weight
* Temperature
* Number of Sales

---

## Comparison

| Feature                 | Categorical          | Numerical            |
| ----------------------- | -------------------- | -------------------- |
| Data Type               | Labels/Groups        | Numbers              |
| Mathematical Operations | Not Meaningful       | Meaningful           |
| Example                 | Male, Female         | 18, 25, 30           |
| Purpose                 | Classification       | Measurement          |
| Visualization           | Bar Chart, Pie Chart | Histogram, Line Plot |

---

# Univariate, Bivariate, and Multivariate Analysis

Analysis refers to studying data and discovering patterns.

---

# 1. Univariate Analysis

**Uni = One**

Univariate Analysis means analyzing **only one column (variable)** at a time.

### Purpose

* Understand distribution
* Find frequency
* Find central tendency
* Detect outliers

### Example

Age column:

| Age | Students |
| --- | -------- |
| 18  | 50       |
| 19  | 40       |
| 20  | 30       |

We are analyzing only the **Age** column.

### Questions Answered

* Which age occurs most?
* What is average age?
* What is age distribution?

### Common Visualizations

#### For Numerical Columns

* Histogram
* Box Plot
* Density Plot

#### For Categorical Columns

* Count Plot
* Bar Chart
* Pie Chart

### Example

```
Age Distribution of Students
```

Only one variable → Age

Therefore:

**Univariate Analysis**

---

# 2. Bivariate Analysis

**Bi = Two**

Bivariate Analysis means analyzing the relationship between **two columns**.

### Purpose

* Find relationships
* Find trends
* Find correlations
* Compare categories

---

## Example 1: Numerical vs Numerical

| Experience | Salary |
| ---------- | ------ |
| 1          | 30000  |
| 2          | 40000  |
| 3          | 55000  |

Question:

"Does salary increase with experience?"

### Visualization

* Scatter Plot
* Line Plot

Relationship between:

```
Experience ↔ Salary
```

---

## Example 2: Numerical vs Categorical

| Gender | Salary |
| ------ | ------ |
| Male   | 50000  |
| Female | 45000  |

Question:

"Do males and females earn different salaries?"

### Visualization

* Box Plot
* Bar Plot

Relationship between:

```
Gender ↔ Salary
```

---

## Example 3: Categorical vs Numerical

| Month | Sales |
| ----- | ----- |
| Jan   | 1000  |
| Feb   | 1200  |
| Mar   | 1500  |

Question:

"How do sales change month by month?"

### Visualization

* Bar Plot
* Line Plot

Relationship:

```
Month ↔ Sales
```

---

## Example 4: Categorical vs Categorical

| Gender | Purchased |
| ------ | --------- |
| Male   | Yes       |
| Female | No        |

Question:

"Is purchasing behavior related to gender?"

### Visualization

* Grouped Bar Chart
* Heatmap

Relationship:

```
Gender ↔ Purchased
```

---

## Types of Bivariate Analysis

### 1. Numerical ↔ Numerical

Examples:

* Age ↔ Salary
* Height ↔ Weight

Tools:

* Scatter Plot
* Correlation Matrix

---

### 2. Numerical ↔ Categorical

Examples:

* Salary ↔ Department
* Marks ↔ Subject

Tools:

* Box Plot
* Violin Plot

---

### 3. Categorical ↔ Categorical

Examples:

* Gender ↔ Purchased
* City ↔ Product Category

Tools:

* Count Plot
* Heatmap

---

# 3. Multivariate Analysis

**Multi = More Than Two**

Multivariate Analysis means analyzing **three or more variables simultaneously**.

### Purpose

* Discover complex relationships.
* Build predictive models.
* Understand interactions among variables.

### Example

| Age | Experience | Education | Salary |
| --- | ---------- | --------- | ------ |
| 22  | 1          | B.Tech    | 30000  |
| 25  | 3          | B.Tech    | 50000  |
| 30  | 6          | MBA       | 90000  |

Question:

"What factors affect salary?"

Variables:

* Age
* Experience
* Education
* Salary

More than two columns are involved.

Therefore:

**Multivariate Analysis**

---

### Common Visualizations

* Pair Plot
* Heatmap
* 3D Scatter Plot
* Correlation Matrix
* Bubble Chart

### Example

```
Age
   \
Experience -----> Salary
   /
Education
```

Salary depends on multiple factors.

---

# Quick Summary

| Analysis Type | Number of Columns | Example                             |
| ------------- | ----------------- | ----------------------------------- |
| Univariate    | 1                 | Age Distribution                    |
| Bivariate     | 2                 | Experience vs Salary                |
| Multivariate  | 3 or More         | Age, Experience, Education → Salary |

### Easy Memory Trick

* **Univariate** → One column → "Understand a variable"
* **Bivariate** → Two columns → "Find relationship"
* **Multivariate** → More than two columns → "Find complex patterns"
