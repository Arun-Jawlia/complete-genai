## `What is Matplotlib`
- Matplotlib is one of the most popular Data Visualization libraries in Python.
- It is used to create graphs, charts, and plots from data.
- It is widely used in data science, scientific computing, machine learning, and engineering.
- Matplotlib helps us represent numerical and categorical data visually,
- making it easier to understand patterns, trends, and relationships.

## `Developed By`: John D. Hunter (2003)

With Matplotlib, you can create:

- Line plots
- Bar charts
- Scatter plots
- Histograms
- Pie charts
- Heatmaps
- 3D plots
- Customized figures for publications and presentations

---

## `What is Data Visualization?`
- Data Visualization is the process of representing data in graphical form.
- Instead of looking at large tables of numbers, we can use charts and graphs
- to understand the data quickly and effectively.

### Example:
Without Visualization:
    - January  -> 100 Sales
    - February -> 120 Sales
    - March    -> 150 Sales

### With Visualization:: A Line Plot can instantly show that sales are increasing over time.
