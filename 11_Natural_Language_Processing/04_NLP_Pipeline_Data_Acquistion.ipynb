# `Data Augmentation`

## Definition

Data Augmentation generates new training samples from existing data to increase dataset size and improve model generalization.

> **Note:** Augmented samples are synthetic (artificially generated) but preserve the original meaning.

---

## Common Data Augmentation Techniques

### 1. Synonym Replacement

Replace words with their synonyms.

Example

Original

```text
I like to run.
```

Augmented

```text
I like to jog.
```

---

### 2. Bigram Flip (Word Reordering)

Swap neighboring words while preserving meaning where possible.

Example

Original

```text
I am walking on the road.
```

Possible Augmented

```text
I am on the road walking.
```

> Use carefully, as incorrect reordering can change the sentence meaning.

---

### 3. Back Translation

Translate text into another language and then translate it back.

Example

```text
English

↓

French

↓

English
```

Original

```text
I love this phone.
```

Back-translated

```text
I really love this phone.
```

---

### 4. Adding Noise

Introduce small changes without changing the overall meaning.

Examples

- Minor spelling variations
- Character swaps
- Random insertions
- Word deletions (carefully)

Purpose

Improve model robustness to noisy user input.

# `Step 2: Text Preparation`

---

## Definition

Text Preparation converts raw, noisy text into a clean and structured format suitable for NLP models.

It consists of:

1. Text Cleaning
2. Basic Preprocessing
3. Advanced Preprocessing

---

# 1. `Text Cleaning`

Raw text often contains:

- Emojis
- URLs
- HTML tags
- Spelling mistakes
- Special characters
- Short forms

These should be cleaned before training.

---

## A. Regular Expressions (Regex)

Regex is used to search, match, and replace text patterns.

Examples

- Remove URLs
- Remove phone numbers
- Remove HTML tags
- Remove extra spaces

Example

```text
Visit https://abc.com

↓

Visit
```

---

## B. Emoji Handling

Emojis carry sentiment.

Example

```text
😊

↓

Positive Emotion
```

Instead of removing emojis blindly, they can be converted into Unicode representations or descriptive text using Unicode normalization.

Example

```text
😊

↓

U+1F60A
```

---

## C. Spelling Correction

Correct spelling mistakes before training.

Example

```text
Machien Learning

↓

Machine Learning
```

Popular Library

```python
from textblob import TextBlob

TextBlob("machien").correct()
```

---

# 2. `Basic Preprocessing Techniques`

These are commonly applied before training an NLP model.

---

## A. Tokenization

### Definition

Tokenization splits text into smaller units called **tokens**.

Types

### Word Tokenization

```text
Hello, I am Arun.

↓

["Hello", "I", "am", "Arun"]
```

### Sentence Tokenization

```text
Hello. How are you?

↓

["Hello.", "How are you?"]
```

Tokenization is one of the most fundamental preprocessing steps in NLP.

---

## B. Stop Word Removal (Optional)

Stop words are frequently occurring words that often contribute little semantic meaning.

Examples

- a
- an
- the
- is
- am
- are

Example

Original

```text
I am learning NLP.
```

After Removal

```text
learning NLP
```

> **Important:** Stop-word removal is **task-dependent**.  
> For applications like **Automatic Speech Recognition (ASR)** or **Machine Translation**, stop words are often retained because they contribute to sentence structure and meaning.

---

## C. Stemming

Converts words into their root form by removing suffixes.

Example

```text
Dance

Dancing

Danced

↓

Dance
```

---

## D. Lemmatization

Converts words into their dictionary (base) form.

Example

```text
Running

↓

Run

Better

↓

Good
```

Lemmatization produces linguistically correct base words and is generally preferred over stemming.

---

## E. Removing Digits & Punctuation

Digits and punctuation are removed when they do not contribute to the meaning of the text.

Example

```text
Hello!!!

↓

Hello
```

---

## F. Lowercasing

Convert all text to lowercase to ensure consistency.

Example

```text
Habit

HABIT

habit

↓

habit
```

This prevents the model from treating them as different words.

---

## G. Language Detection

Identify the language before further processing.

Applications

- Machine Translation
- Multilingual Chatbots
- Cross-language Search

Example

```text
Bonjour

↓

French
```

---

# `3. Advanced Preprocessing Techniques`

These techniques help models better understand language structure and context.
1. Part-Of-Speech (POS) Tagging
2. Pasing
3. Coreference Resolution

---

## A. Part-of-Speech (POS) Tagging

Assign grammatical roles to each word.

Example

Sentence

```text
Hello, I am Amit.
```

Possible Tags

| Word | POS Tag |
|------|----------|
| Hello | Interjection |
| I | Pronoun |
| am | Verb |
| Amit | Proper Noun |

Applications

- Chatbots
- Machine Translation
- Information Extraction

---

## B. Parsing

Parsing analyzes the grammatical structure of a sentence.

It helps determine how words are related.

Example

```text
The cat chased the mouse.
```

Parsing identifies:

- Subject → The cat
- Verb → chased
- Object → the mouse

Applications

- Question Answering
- Grammar Checking
- Syntax Analysis

---

## C. Coreference Resolution

Coreference Resolution identifies when different words refer to the same entity.

Example

```text
Arun bought a laptop.

He loves it.
```

Resolved

- He → Arun
- It → Laptop

Applications

- Chatbots
- Document Understanding
- Question Answering

---

# `Best Practices`

- Always inspect your data before preprocessing.
- Do not remove stop words blindly; consider the task.
- Use lemmatization when preserving meaning is important.
- Preserve emojis if they contribute to sentiment.
- Choose preprocessing techniques based on your application rather than applying every technique by default.

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What is Data Acquisition in NLP?**
**Answer:** Data Acquisition is the process of collecting text or speech data that will be used to train and evaluate NLP models.

**2. What are common sources of NLP data?**
**Answer:** CSV files, databases, APIs, web scraping, PDFs, images (OCR), audio (ASR), and public datasets.

**3. What is Data Augmentation?**
**Answer:** It creates additional synthetic training examples from existing data using techniques like synonym replacement and back translation.

**4. What is Tokenization?**
**Answer:** Tokenization splits text into smaller units such as words or sentences, making it easier for NLP models to process.

**5. What is the difference between OCR and ASR?**
**Answer:** OCR extracts text from images or scanned documents, whereas ASR converts spoken language into text.

---

## `Intermediate Questions`

**1. Why is stop-word removal optional?**
**Answer:** Stop words can be removed for tasks like text classification but are often retained in tasks such as machine translation or speech recognition because they contribute to grammar and meaning.

**2. Why is lemmatization generally preferred over stemming?**
**Answer:** Lemmatization returns valid dictionary words and preserves meaning better than stemming, which may produce incomplete root forms.

**3. What is Coreference Resolution?**
**Answer:** It identifies when different words or pronouns refer to the same entity, improving document understanding and question answering.

**4. What is the role of POS Tagging?**
**Answer:** POS Tagging assigns grammatical categories (noun, verb, adjective, etc.) to words, helping models understand sentence structure.

---

## `Scenario-Based Questions`

**Q1. Your startup has only 500 labeled customer reviews. How would you improve the dataset?**
**Answer:** Use data augmentation techniques such as synonym replacement and back translation, collect more user feedback, or leverage transfer learning with pre-trained language models.

**Q2. You receive customer feedback as scanned PDFs. How do you prepare it for NLP?**
**Answer:** Use OCR to extract text from the PDFs, clean the extracted text, preprocess it, and then feed it into the NLP pipeline.

**Q3. You're building a multilingual chatbot. Why is language detection important?**
**Answer:** Language detection identifies the user's language so the chatbot can route the request to the appropriate language model or translation system.
