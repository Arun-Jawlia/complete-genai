# Text Representation in NLP (Text Vectorization / Text-to-Number Conversion)

> **Text Representation** is the process of converting human-readable text into numerical vectors so that Machine Learning and Deep Learning models can process, learn, and make predictions.

---

# `What is Text Representation?`

## Definition

**Text Representation** (also called **Text Vectorization** or **Feature Extraction from Text**) is the process of converting textual data into numerical form.

Since computers and ML algorithms understand **numbers**, not text, every word, sentence, or document must be represented as numerical vectors.

Simply put,

> **Text → Numbers → Machine Learning Model → Prediction**

---

# `Why Can't Machines Understand Text?`

Consider the sentence:

```text
I love Machine Learning.
```

For humans:

- We understand the meaning instantly.

For a machine:

- It only sees a sequence of characters.
- It cannot understand the semantic meaning directly.

Machine Learning algorithms perform mathematical operations such as:

- Addition
- Multiplication
- Matrix operations
- Distance calculations

These operations require **numeric input**.

---

# `Why Do We Need Text Vectorization?`

Suppose you want to build a **Sentiment Analysis** model.

Input

```text
This movie is amazing.
```

Desired Output

```text
Positive
```

The model cannot directly process the sentence.

It first needs numerical features that capture the important information.

---

# `What is Feature Extraction from Text?`

## Definition

Feature Extraction is the process of identifying and extracting useful information from text and converting it into numerical features.

The extracted features should preserve as much useful information as possible.

---

# `Why Do We Need Feature Extraction?`

The main objective is to generate **optimal features**.

## What are Optimal Features?

Optimal features are numerical representations that contain enough information for the model to correctly solve the task.

Example

Text

```text
I love this phone.
```

Useful Features

- Positive words
- Word frequency
- Context
- Semantic meaning

These help the model predict:

```text
Positive Sentiment
```

---

# `Why is Feature Extraction Difficult?`

Unlike images or tabular data, language is highly complex.

Challenges include:

- Synonyms
- Polysemy (multiple meanings)

Example

```text
Bank
```

Could mean:

- River bank
- Financial bank

---

- Context

Example

```text
Apple
```

Could mean:

- Fruit
- Company

---

- Word order

Example

```text
Dog bites man.
```

vs

```text
Man bites dog.
```

Same words, completely different meanings.

---

- Sarcasm

Example

```text
Great! Another meeting...
```

The actual sentiment may be negative.

---

# `Core Idea of Text Representation`

Whenever we convert text into numbers,

our goal is:

> **Preserve the semantic meaning of the text as much as possible.**

A good representation should allow the model to understand:

- Meaning
- Context
- Similarity
- Relationships between words

---

# `Text Representation Techniques`

There are multiple ways to convert text into numbers.

---

# 1. Ordinal Encoding

## Definition

Ordinal Encoding assigns ordered numerical values to categories.

Example

```text
Small

Medium

Large
```

↓

```text
1

2

3
```

### Important

Order matters.

### Is it used for text vectorization?

❌ No.

Ordinal Encoding is mainly used for **ordinal categorical data**, not for converting text into vectors.

---

# 2. Label Encoding

## Definition

Assigns an integer to each category.

Example

```text
Positive

Negative

Neutral
```

↓

```text
2

0

1
```

### Important

Mostly used for **target labels (y)** in supervised learning.

Not used for representing input text.

---

# 3. One-Hot Encoding

## Definition

Represents each unique word as a binary vector.

Only one position contains **1**, while all others are **0**.

---

### Example

Vocabulary

```text
I

Love

AI
```

Vectors

| Word | Vector |
|------|---------|
| I | [1,0,0] |
| Love | [0,1,0] |
| AI | [0,0,1] |

---

### Advantages

- Very simple
- Easy to understand
- No training required

---

### Limitations

- Huge sparse vectors
- No semantic meaning
- Similar words appear completely unrelated

Example

Dog

```text
[1,0,0]
```

Cat

```text
[0,1,0]
```

The vectors suggest no similarity even though the meanings are related.

---

# 4. Bag of Words (BoW)

## Definition

Bag of Words represents a document using the frequency of words.

It ignores:

- Grammar
- Word order
- Context

Only word counts matter.

---

### Example

Documents

```text
Doc1

I love AI

Doc2

I love NLP
```

Vocabulary

```text
I

Love

AI

NLP
```

BoW Matrix

| Document | I | Love | AI | NLP |
|-----------|---|------|----|-----|
| Doc1 | 1 | 1 | 1 | 0 |
| Doc2 | 1 | 1 | 0 | 1 |

---

### Advantages

- Simple
- Fast
- Effective for text classification

---

### Limitations

- Ignores word order
- No semantic understanding
- Produces sparse vectors

---

# 5. N-Grams

## Definition

An **N-Gram** is a sequence of N consecutive words.

---

## Types

### Unigram

```text
I

Love

NLP
```

---

### Bigram

```text
I Love

Love NLP
```

---

### Trigram

```text
I Love NLP
```

---

### Why N-Grams?

They preserve local word order.

Example

```text
New York
```

A Bigram keeps these words together.

---

### Limitations

- Vocabulary grows quickly
- High memory usage
- Still limited semantic understanding

---

# 6. TF-IDF (Term Frequency–Inverse Document Frequency)

## Definition

TF-IDF assigns weights to words based on:

- How frequently they appear in a document.
- How rare they are across all documents.

---

### Intuition

Common words such as:

- the
- is
- and

appear in almost every document.

Therefore, they receive lower weights.

Rare but informative words receive higher weights.

---

### Advantages

- Better than Bag of Words
- Reduces influence of common words
- Works well for classification and search

---

### Limitations

- Still ignores semantic meaning
- Ignores context
- Cannot understand synonyms

---

# 7. Custom Features

Sometimes domain-specific handcrafted features are useful.

Example (Sentiment Analysis)

Sentence

```text
I absolutely love this phone!
```

Possible custom features:

- Number of positive words
- Number of negative words
- Number of exclamation marks
- Sentence length
- Number of emojis

These handcrafted features can improve traditional ML models.

---

# 8. Word2Vec

## Definition

Word2Vec is a neural network-based technique that learns dense vector representations of words.

Unlike previous methods, it captures **semantic meaning**.

---

### Example

Word2Vec learns that:

King ≈ Queen

Doctor ≈ Nurse

Car ≈ Vehicle

because they appear in similar contexts.

---

### Advantages

- Dense vectors
- Captures semantics
- Better similarity representation

---

### Limitations

- Same vector for every occurrence of a word.

Example

```text
Bank
```

River Bank

↓

Same Vector

Financial Bank

↓

Same Vector

It cannot distinguish different meanings based on context.

---

# `Comparison of Text Representation Techniques`

| Technique | Captures Frequency | Captures Order | Captures Meaning | Sparse | Dense |
|------------|-------------------|----------------|------------------|--------|-------|
| One-Hot Encoding | ❌ | ❌ | ❌ | ✅ | ❌ |
| Bag of Words | ✅ | ❌ | ❌ | ✅ | ❌ |
| N-Grams | ✅ | Partial | ❌ | ✅ | ❌ |
| TF-IDF | ✅ | ❌ | ❌ | ✅ | ❌ |
| Word2Vec | ❌ | Partial | ✅ | ❌ | ✅ |
| Transformer Embeddings | ✅ | ✅ | ✅ | ❌ | ✅ |

---

# `When Should We Use Each Technique?`

| Technique | Best For |
|------------|----------|
| One-Hot Encoding | Small toy examples, teaching concepts |
| Bag of Words | Spam Detection, Text Classification |
| N-Grams | Search Engines, Language Models |
| TF-IDF | Search, Document Classification |
| Word2Vec | Semantic Similarity, Recommendation Systems |
| Transformer Embeddings | Chatbots, Question Answering, RAG, LLMs |

---

# `Best Practices`

- Start with simple techniques (BoW or TF-IDF) for classical ML models.
- Use Word2Vec or Transformer Embeddings when semantic understanding is required.
- Choose the representation based on the NLP task, dataset size, and computational resources.
- Remember that no single technique is optimal for every problem.

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What is Text Representation in NLP?**
**Answer:** It is the process of converting text into numerical vectors so that Machine Learning and Deep Learning models can process and learn from the data.

**2. Why is Feature Extraction necessary?**
**Answer:** ML and DL models operate on numerical data. Feature extraction transforms raw text into meaningful numerical representations while preserving important information.

**3. What is a Corpus?**
**Answer:** A Corpus is the complete collection of text documents used for NLP tasks.

**4. What is the difference between a Corpus and a Vocabulary?**
**Answer:** A Corpus contains all text documents (including repeated words), whereas a Vocabulary contains only the unique words present in the corpus.

**5. Is Ordinal Encoding used for text vectorization?**
**Answer:** No. Ordinal Encoding is intended for ordinal categorical variables where the order of categories is meaningful. It is not suitable for representing text.

---

## Intermediate Questions

**1. Why is One-Hot Encoding not ideal for large vocabularies?**
**Answer:** It creates very large, sparse vectors and fails to capture semantic relationships between words.

**2. Explain the difference between Bag of Words and TF-IDF.**
**Answer:** BoW represents documents using raw word frequencies, whereas TF-IDF assigns higher weights to informative words that are frequent in a document but rare across the corpus.

**3. What is the main limitation of Word2Vec?**
**Answer:** Word2Vec generates the same embedding for a word regardless of its context, so it cannot distinguish different meanings of ambiguous words such as "bank."

**4. Why are Transformer Embeddings preferred in modern NLP?**
**Answer:** Transformer embeddings are contextual, meaning they generate different representations for the same word depending on its surrounding context, leading to better language understanding.

---

## Scenario-Based Questions

**Q1. You're building a spam email classifier with a small dataset and a Logistic Regression model. Which text representation would you choose?**
**Answer:** TF-IDF is a strong choice because it is efficient, interpretable, and works well with traditional ML algorithms.

**Q2. You're building a chatbot that must understand context and semantics. Which representation is more suitable?**
**Answer:** Transformer-based contextual embeddings (e.g., BERT embeddings), as they capture context and semantic meaning.

**Q3. Why would Word2Vec outperform One-Hot Encoding in semantic similarity tasks?**
**Answer:** Word2Vec places semantically similar words close together in the vector space, whereas One-Hot Encoding treats every word as completely independent.
