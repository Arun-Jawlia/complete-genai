# `N-Grams and TF-IDF in NLP (Text Vectorization)`

> **N-Grams** and **TF-IDF (Term Frequency–Inverse Document Frequency)** are classical text vectorization techniques that improve upon the limitations of **Bag of Words (BoW)** by preserving local word order (N-Grams) or assigning importance to informative words (TF-IDF).

---

# Recap: Why Do We Need Better Techniques?

Earlier techniques like:

- One-Hot Encoding
- Bag of Words (BoW)

have several limitations:

- Ignore word order
- Ignore context
- Ignore semantic meaning
- Produce sparse vectors
- Suffer from Out-of-Vocabulary (OOV) issues

To overcome some of these limitations, we use:

- N-Grams
- TF-IDF

---

# `N-Grams`

## Definition

An **N-Gram** is a sequence of **N consecutive words (tokens)**.

Unlike Bag of Words, which treats every word independently, N-Grams preserve **local word order**.

> **N = Number of consecutive words**

---

# Types of N-Grams

| N | Name | Example |
|---|------|----------|
| 1 | Unigram | NLP |
| 2 | Bigram | Machine Learning |
| 3 | Trigram | I Love NLP |
| 4 | Four-gram | Deep Learning Model |

---

# `Why N-Grams are Better than BoW`

Consider two sentences:

Sentence 1

```text

```

```text

```

---

## Bag of Words

Vocabulary

```text

Movie

Very

Good

Not
```

Both sentences contain almost the same words.

The vectors are very similar.

BoW struggles to distinguish the different meanings.

---

Sentence 1

```text

very good
```


# Advantages of N-Grams

- Preserves local word order.
- Better than BoW for short phrases.
- Captures simple context.
- Easy to implement.
- Works well for search and document classification.

# `TF-IDF (Term Frequency–Inverse Document Frequency)`

## Definition

TF-IDF is a statistical technique that measures **how important a word is in a document relative to the entire corpus**.

Unlike Bag of Words, which only counts word frequency,

TF-IDF reduces the importance of common words and increases the importance of rare but informative words.

---

# Why Do We Need TF-IDF?

Consider these words:

```text
the

is

and
```

They appear in almost every document.

Although frequent,

they are not useful for distinguishing documents.

TF-IDF gives such words **lower weights**.

---

Words like:

```text
Cancer

Transformer

Quantum
```

may appear in only a few documents.

These are highly informative.

TF-IDF assigns them **higher weights**.

---

# TF-IDF Formula

```text
TF-IDF = TF × IDF
```

---

# `Term Frequency (TF)`

## Definition

Term Frequency measures how often a word appears **within the current document**.

Formula

```text
TF =

(Number of occurrences of the word)

/

(Total number of words in the document)
```

---

## Example

Sentence

```text
Hello Arun

I am Arun

Nice to meet you Arun

Hello
```

Suppose

Total Words = 12

Occurrences

Arun = 3

Hello = 2

Then

```text
TF(Arun)

=

3/12

=

0.25
```

```text
TF(Hello)

=

2/12

=

0.166
```

Higher TF means

the word is more important **within that document**.

---

# `Inverse Document Frequency (IDF)`

## Definition

IDF measures **how rare or unique a word is across the entire corpus**.

Formula

```text
IDF

=

log

(

Total Documents

/

Documents containing the word

)
```

In practice, libraries such as scikit-learn use a **smoothed version** of this formula to avoid division by zero.

---

# Example

Documents

```text
D1

People Love AI

D2

AI Love AI-ML

D3

AI-ML Love Love

D4

Love People People
```

Total Documents

```text
N = 4
```

---

### Word = Love

Appears in

D1

D2

D3

D4

Document Frequency

```text
DF = 4
```

IDF

```text
log(4/4)

=

log(1)

=

0
```

Love receives a very low weight because it appears everywhere.

---

### Word = People

Appears in

D1

D4

Document Frequency

```text
DF = 2
```

IDF

```text
log(4/2)

=

log(2)

≈ 0.693
```

Higher than "Love"

because it is rarer.

---

# Final TF-IDF Score

```text
TF-IDF

=

TF × IDF
```

Words that:

- appear frequently in one document
- but rarely in the corpus

receive the highest scores.

---

# Applications of TF-IDF

TF-IDF is widely used in:

- Search Engines

| Feature | N-Grams | TF-IDF |
|----------|----------|---------|
| Preserves Word Order | ✅ | ❌ |
| Uses Word Frequency | ✅ | ✅ |
| Gives Importance to Rare Words | ❌ | ✅ |

| OOV Problem | ✅ | ✅ |
| Best For | Local Context | Information Retrieval & Text Classification |

# `Why Modern NLP Moved Beyond TF-IDF`

Although N-Grams and TF-IDF improved over Bag of Words,

they still cannot understand:

- Context
- Semantics
- Synonyms
- Polysemy
- Long-range dependencies

This led to embedding-based methods:

- Word2Vec
- FastText
- GloVe
- BERT
- GPT
- Sentence Transformers

These methods learn dense vectors that preserve semantic relationships.

---

# `Interview Tips`

✅ N-Grams preserve local word order.

✅ Bigrams consist of two consecutive words; Trigrams consist of three consecutive words.

✅ TF-IDF assigns higher weights to words that are frequent in a document but rare across the corpus.

✅ TF-IDF is widely used in search engines and information retrieval.

✅ Neither N-Grams nor TF-IDF captures true semantic meaning.

---

# `Key Takeaways`

- N-Grams improve upon Bag of Words by preserving short word sequences.
- TF-IDF improves upon Bag of Words by weighting informative words more highly than common words.
- Both techniques still produce sparse vectors and do not capture semantic relationships.
- Modern NLP systems rely on dense embeddings and Transformer-based models to overcome these limitations.

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What is an N-Gram?**
An N-Gram is a sequence of **N consecutive words** used as a feature in text representation.

**2. What is the difference between Unigram, Bigram, and Trigram?**

* Unigram → 1 word
* Bigram → 2 consecutive words
* Trigram → 3 consecutive words

---

**3. What is TF-IDF?**
TF-IDF (Term Frequency–Inverse Document Frequency) measures how important a word is within a document relative to the entire corpus.

---

## Intermediate Questions

**4. Why is TF-IDF better than Bag of Words?**
Because TF-IDF reduces the weight of common words (e.g., "the", "is") and increases the weight of rare but informative words.

**5. Why does vocabulary size increase in N-Grams?**
As we combine consecutive words, the number of possible combinations grows rapidly, increasing the vocabulary size and computational cost.

---

## Scenario-Based Questions

**Q1. You're building a search engine for research papers. Would you choose BoW or TF-IDF? Why?**
**Answer:** TF-IDF, because it highlights rare and informative terms, making document retrieval and ranking more effective.

**Q2. Two sentences differ only by the word "not". Which technique better distinguishes them: BoW or Bigrams?**
**Answer:** Bigrams, because phrases like **"is not"** are treated as unique features, preserving local word order and improving distinction.
