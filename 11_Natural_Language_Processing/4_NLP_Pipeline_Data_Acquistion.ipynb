# NLP – Sentiment Analysis Pipeline

**Problem Statement:**
Train a sentiment analysis model using text data so it can make accurate predictions on unseen reviews.

---

# Step 1: Data Gathering

The quality of a machine learning model largely depends on the quality and quantity of the data used for training.

## 1. Available Data

### A. CSV Files

* Data is directly available in CSV format.
* This is the easiest scenario.
* Load the dataset and proceed with preprocessing.

**Example**

* Customer Reviews
* Movie Reviews
* Product Reviews
* Twitter Comments

---

### B. Database

Organizations often store data inside databases instead of CSV files.

#### Large Organizations

Typical data flow:

```
Applications
      │
      ▼
Data Engineering Team
      │
      ▼
ETL Pipeline
(Extract → Transform → Load)
      │
      ▼
Database
      │
      ▼
Data Warehouse
      │
      ▼
Machine Learning Team
```

The ML engineer receives cleaned and transformed data.

---

#### Small Startups

Small startups may not have a dedicated Data Engineering team.

Therefore, an ML Engineer may need to:

* Connect to databases
* Write SQL queries
* Extract data
* Convert into CSV/DataFrame
* Perform preprocessing

Having basic Data Engineering skills becomes valuable.

---

### C. Less Data

Machine learning models generally require large datasets.

If only a small amount of data is available, we can use **Data Augmentation**.

Data augmentation artificially generates new training samples while preserving the original meaning.

Although generated samples are synthetic (fake), they help improve model generalization.

Common augmentation techniques include:

### 1. Synonym Replacement

Replace words with their synonyms.

Example

```
Original:
I like to run.

Augmented:
I like to jog.
```

---

### 2. Bigram Flip

Swap nearby words while maintaining readability.

Example

```
Original:
I am walking on the road.

Augmented:
I am on the walking road.
```

*(Used carefully because excessive flipping may change sentence meaning.)*

---

### 3. Back Translation

Translate a sentence into another language and then translate it back.

Example

```
English
↓

French

↓

English

Original:
The product is amazing.

Back Translation:
This product is really amazing.
```

The meaning remains similar while sentence structure changes.

---

### 4. Adding Noise

Introduce controlled modifications.

Examples

* Minor spelling variations
* Typographical noise
* Random word insertion
* Random deletion

This improves model robustness.

# 1. Text Cleaning

Real-world text contains many inconsistencies.

Text cleaning removes unwanted information and improves data quality.

---

## A. Regular Expressions (Regex)

Regular Expressions help identify patterns inside text.

Examples

* Email IDs
* URLs
* Phone Numbers
* HTML Tags
* Extra Spaces

Regex is commonly used for pattern-based cleaning.

---

## B. Unicode Normalization

Humans understand emojis, but machines process numerical Unicode values.

Example

```
😊

↓

Unicode

↓

UTF-8 Representation
```

Unicode normalization converts text into a standardized machine-readable format.

---

## C. Spelling Correction

Users often make spelling mistakes.

Example

```
Original

I luv this produckt.

↓

Corrected

I love this product.
```

Popular library

```
TextBlob

correct()
```

Correct spelling improves prediction quality.

---

# 2. Basic Preprocessing Techniques

Some preprocessing techniques are applied to almost every NLP application.

Others depend on the specific problem.

---

## A. Tokenization (Almost Always Used)

Tokenization breaks text into smaller independent units called **tokens**.

Example

```
Hello, I am Arun Jawlia.
```

### Word Tokenization

```
["Hello",
"I",
"am",
"Arun",
"Jawlia"]
```

### Sentence Tokenization

```
Sentence 1

Sentence 2

Sentence 3
```

Tokenization is the foundation of almost every NLP pipeline.

---

## B. Stop Word Removal (Optional)

Stop words are extremely common words that usually contribute little semantic meaning.

Examples

```
a
an
the
is
am
are
of
to
```

Example

```
Original

The movie is very good.

↓

After Removal

movie good
```

Not every NLP application removes stop words.

For example:

* Machine Translation ❌
* Automatic Speech Recognition ❌
* Question Answering ❌

Whereas:

* Text Classification ✅
* Topic Modeling ✅
* Text Summarization (sometimes) ✅

---

## C. Stemming (Optional)

Stemming reduces words to their root form using simple heuristic rules.

Example

```
Dance

Dances

Dancing

↓

Dance
```

Another example

```
Pass

Passes

Passing

↓

Pass
```

The stem may not always be a valid English word.

---

## D. Lemmatization (Optional)

Lemmatization also converts words into their root form but uses vocabulary and grammar rules.

Example

```
Running

↓

Run

Better

↓

Good

Children

↓

Child
```

Compared to stemming, lemmatization produces linguistically correct root words.

---

## E. Lowercasing (Optional)

Convert all text into lowercase to reduce vocabulary size.

Example

```
Python

PYTHON

python

↓

python
```

---

## F. Removing Digits (Optional)

Example

```
Product123

↓

Product
```

Useful when numbers do not contribute to the prediction task.

---

## G. Removing Punctuation (Optional)

Example

```
Hello!!!

↓

Hello
```

Removes symbols such as:

```
.
,
!
?
:
;
(
)
```

---

## H. Language Detection (Optional)

Some datasets contain multiple languages.

Example

```
English

Hindi

French

Spanish
```

Language detection helps apply language-specific preprocessing techniques.
