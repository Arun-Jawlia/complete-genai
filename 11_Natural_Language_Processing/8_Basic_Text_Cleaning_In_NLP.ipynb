# Basic Text Cleaning in NLP

> **Text Cleaning** is the process of removing unwanted or noisy information from raw text before applying NLP algorithms.

---

# `What is Text Cleaning?`

## Definition

Text Cleaning is the **first preprocessing step** in an NLP pipeline where raw text is cleaned to make it suitable for Machine Learning or Deep Learning models.

Raw text collected from social media, websites, emails, reviews, or chats often contains:

- Spelling mistakes
- HTML tags
- URLs
- Emojis
- Numbers
- Special characters
- Multiple spaces
- Punctuation
- User mentions
- Hashtags

These unnecessary elements are called **noise**.

The goal of text cleaning is to remove or normalize this noise while preserving meaningful information.

---

# `Why is Text Cleaning Important?`

Imagine you are building a **Sentiment Analysis** model.

Raw Review:

```text
😍😍 OMG!!!! This phone is soooo amazing!!!! 🔥🔥🔥
Visit: https://amazon.in
```

Without cleaning, the model receives many irrelevant tokens.

After cleaning:

```text
this phone is so amazing
```

The cleaned text helps the model learn meaningful patterns.

# `Common Text Cleaning Techniques`

## 1. Lowercasing

### Definition

Convert every character into lowercase.

Example

Before

```text
Machine Learning
MACHINE LEARNING
machine learning
```

After

```text
machine learning
machine learning
machine learning
```

### Why?

Without lowercasing, the model may treat:

```text
Apple

APPLE

apple
```

as three different words.

---

# `Regular Expressions (Regex)`

## What is Regex?

**Regular Expression (Regex)** is a sequence of characters used to search, match, replace, or validate patterns in text.

Think of Regex as a **powerful search language**.

It is widely used in NLP for cleaning noisy text.

---

# Why Regex is Important?

Regex helps remove:

- URLs
- HTML Tags
- Emails
- Phone Numbers
- Digits
- Punctuation
- Multiple Spaces
- User Mentions
- Hashtags

---

# `Basic Regex Syntax`

| Symbol | Meaning | Example |
|---------|---------|----------|
| `.` | Any character | `a.c` → abc |
| `*` | Zero or more occurrences | `ab*` |
| `+` | One or more occurrences | `ab+` |
| `?` | Optional occurrence | `colou?r` → color, colour |
| `[]` | Character set | `[abc]` |
| `[^ ]` | Not in set | `[^0-9]` |
| `\d` | Digit | 0–9 |
| `\D` | Non-digit | A, B, @ |
| `\w` | Word character | a-z, A-Z, 0-9, _ |
| `\W` | Non-word character | !, @, # |
| `\s` | White space | Space, Tab |
| `\S` | Non-space | Any visible character |
| `^` | Start of string | `^Hello` |
| `$` | End of string | `World$` |

# Best Practices

✅ Lowercase almost every NLP dataset.

✅ Remove URLs unless URLs contain useful information.

✅ Keep emojis in sentiment analysis because they express emotions.

✅ Don't remove numbers if they are meaningful (prices, dates, years).

✅ Preserve hashtags if they contain semantic information.

---

# Common Mistakes

❌ Removing emojis in sentiment analysis.

❌ Removing all numbers blindly.

❌ Removing hashtags without checking their importance.

❌ Applying every cleaning step to every NLP problem.

---

# Interview Tips

✅ Text Cleaning removes noisy data.

✅ Regex is the most common tool for pattern matching.

✅ Regex is heavily used for removing URLs, HTML tags, emails, digits, and punctuation.

✅ Cleaning depends on the application. There is no universal cleaning pipeline.

# Key Takeaways

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What is Text Cleaning in NLP?**
**Answer:** Text Cleaning is the process of removing or normalizing noisy and irrelevant information from raw text before feature engineering and model training.

**2. Why do we use Regular Expressions (Regex)?**
**Answer:** Regex is used to search, match, validate, and replace text patterns such as URLs, emails, phone numbers, HTML tags, punctuation, and extra spaces.

**3. What is the difference between Text Cleaning and Text Preprocessing?**
**Answer:** Text Cleaning removes noise (e.g., URLs, HTML, punctuation), while Text Preprocessing prepares cleaned text for modeling through techniques such as tokenization, stop-word removal, stemming, and lemmatization.

**4. Should emojis always be removed?**
**Answer:** No. In sentiment analysis, emojis often carry emotional information and should be preserved or converted into descriptive text.

**5. Why is lowercasing commonly applied?**
**Answer:** Lowercasing ensures that words like `Apple`, `APPLE`, and `apple` are treated as the same token, reducing vocabulary size.

---

## Intermediate Questions

**1. What are the limitations of Regex?**
**Answer:** Regex is excellent for pattern-based cleaning but cannot understand language semantics or context. It should be combined with other NLP preprocessing techniques.

**2. When should numbers be retained in NLP?**
**Answer:** Numbers should be retained when they contain useful information, such as prices, years, dates, quantities, or identifiers relevant to the task.

**3. Explain a complete text cleaning workflow for social media data.**
**Answer:** Lowercasing → Remove HTML → Remove URLs → Handle emojis → Remove mentions → Handle hashtags → Remove punctuation → Normalize whitespace → Correct spelling (if required).
