# `Artificial Neural Network (ANN) Overview`
---

## What is an Artificial Neural Network (ANN)?

An **Artificial Neural Network (ANN)** is a **Deep Learning model** inspired by the structure and working of the **human brain**.

Just like the human brain consists of **billions of neurons** connected together, an ANN consists of **artificial neurons (perceptrons)** connected in multiple layers.

> **Definition:**
> An Artificial Neural Network (ANN) is a computational model made up of interconnected artificial neurons that learns patterns from data by adjusting weights and biases.

---

# Why Do We Need ANN?

Traditional Machine Learning algorithms require **manual feature engineering**.

For example:

Suppose we want to classify an image as **Cat** or **Dog**.

Using Machine Learning, we manually extract features like:

- Ear Shape
- Nose Shape
- Tail Length
- Fur Color

Then we train a classifier.

However, ANN automatically learns these important features directly from data.

---

# Human Brain vs Artificial Neural Network

| Human Brain | Artificial Neural Network |
|-------------|--------------------------|
| Biological Neurons | Artificial Neurons (Perceptrons) |
| Synapses | Weights |
| Brain learns from experience | ANN learns from data |
| Decision Making | Prediction |

# Architecture of ANN

```text
          Input Layer

     x1     x2     x3     x4

        \    |    |    /

         Hidden Layer

      ○      ○      ○

           Hidden Layer

        ○      ○

          Output Layer

             ○

        Cat / Dog
```

ANN mainly consists of **three types of layers**.

---

# 1. Input Layer

The Input Layer receives the input features.

Example:

House Price Prediction

Input:

- Area
- Bedrooms
- Bathrooms
- Parking

These values enter the network through the input layer.

---

# 2. Hidden Layer

This is where the actual learning happens.

Each neuron performs mathematical calculations.

The hidden layer extracts important patterns from the data.

Multiple hidden layers make the model **Deep Learning**.

---

# 3. Output Layer

Produces the final prediction.

Examples:

Binary Classification

```
Spam

Not Spam
```

Regression

```
House Price

₹75 Lakhs
```

Multiclass Classification

```
Cat

Dog

Horse
```

---

# What is a Perceptron?

A **Perceptron** is the smallest unit (artificial neuron) in an ANN.

Each perceptron receives inputs, performs a weighted calculation, adds a bias, applies an activation function, and produces an output.

---

# Mathematical Equation

Each neuron calculates:

\[
y = w_1x_1 + w_2x_2 + w_3x_3 + ... + b
\]

Where

- x = Inputs
- w = Weights
- b = Bias

---

# Example

Suppose

```
Age = 25

Salary = ₹80,000

Experience = 3 Years
```

Weights

```
Age

↓

0.2

Salary

↓

0.7

Experience

↓

0.5
```

Notice that **Salary has the highest weight**, meaning it is considered more important for the prediction.

---

# What are Weights?

Weights represent the **importance** of each input feature.

Example

Loan Approval

Features

- Credit Score
- Salary
- Age

Weights

```
Credit Score

↓

Highest Importance

Salary

↓

Medium Importance

Age

↓

Lower Importance
```

The ANN learns these weights automatically during training.

---

# What is Bias?

Bias is an additional parameter that helps the model make better predictions.

Think of bias as an **adjustment value**.

Without bias,

the model's prediction may always pass through the origin.

Bias shifts the prediction curve to better fit the data.

---

# Activation Function

## Why Do We Need an Activation Function?

Without an activation function,

ANN behaves like multiple Linear Regression models stacked together.

It cannot learn complex patterns.

Activation functions introduce **non-linearity**.

---

# Common Activation Functions

## 1. Sigmoid

Range

```
0 → 1
```

Used for

- Binary Classification

Example

Spam Detection

```
Probability = 0.92

↓

Spam
```

---

## 2. ReLU (Rectified Linear Unit)

Formula

```
f(x)=max(0,x)
```

Used in

- Hidden Layers

Advantages

- Fast
- Simple
- Most widely used

---

## 3. Tanh

Range

```
-1 → +1
```

Useful when data has negative values.

---

## 4. Softmax

Converts outputs into probabilities.

Used in

- Multiclass Classification

Example

```
Cat

↓

0.82

Dog

↓

0.12

Horse

↓

0.06
```

Prediction

```
Cat
```

---

# Forward Propagation

## What is Forward Propagation?

The process of moving information from the input layer to the output layer.

Workflow

```
Input

↓

Weights

↓

Activation Function

↓

Prediction
```

---

# Loss Function

The Loss Function measures how wrong the prediction is.

Example

Actual

```
Cat
```

Prediction

```
Dog
```

Loss

```
High
```

Goal

```
Minimize Loss
```

Common Loss Functions:

- Mean Squared Error (Regression)
- Binary Cross Entropy (Binary Classification)
- Categorical Cross Entropy (Multiclass Classification)

---

# Backpropagation

## What is Backpropagation?

Backpropagation is the process of updating weights and bias to reduce prediction error.

Steps

1. Make Prediction
2. Calculate Loss
3. Move Backward
4. Update Weights
5. Update Bias
6. Repeat

---

# Complete ANN Workflow

```text
Input Data
      ↓
Input Layer
      ↓
Hidden Layer
      ↓
Activation Function
      ↓
Output Layer
      ↓
Prediction
      ↓
Calculate Loss
      ↓
Backpropagation
      ↓
Update Weights & Bias
      ↓
Repeat
```

# 2. Interview Questions & Answers

### Beginner Questions

**Q1. What is an Artificial Neural Network (ANN)?**
**Answer:** An ANN is a deep learning model inspired by the human brain that learns patterns from data using interconnected artificial neurons.

---

**Q2. What are the three main layers of an ANN?**
**Answer:**

* Input Layer
* Hidden Layer(s)
* Output Layer

---

**Q3. What is a perceptron?**
**Answer:** A perceptron is the basic computational unit (artificial neuron) of an ANN that processes inputs using weights, bias, and an activation function.

---

**Q4. What is the role of weights?**
**Answer:** Weights determine the importance of each input feature during prediction.

---

**Q5. Why do we use activation functions?**
**Answer:** Activation functions introduce non-linearity, allowing the ANN to learn complex relationships.

---

### Intermediate Questions

**Q6. What is the difference between forward propagation and backpropagation?**

**Answer:**

* **Forward Propagation:** Computes the prediction.
* **Backpropagation:** Updates weights and bias to reduce prediction error.

---

**Q7. Why is ReLU preferred in hidden layers?**

**Answer:** ReLU is computationally efficient, helps reduce the vanishing gradient problem, and speeds up training.

---

**Q8. Which activation function is used for binary classification?**

**Answer:** **Sigmoid**.

---

**Q9. Which activation function is used for multiclass classification?**

**Answer:** **Softmax**.

---

### Scenario-Based Questions

**Q10. You are building a handwritten digit recognition system (digits 0–9). Which activation function would you use in the output layer?**

**Answer:** **Softmax**, because it predicts probabilities for multiple classes.

---

**Q11. You are developing an email spam detector. Which activation function is suitable for the output layer?**

**Answer:** **Sigmoid**, because it is a binary classification problem (Spam / Not Spam).

---

**Q12. Why is ANN preferred over traditional Machine Learning for image classification?**

**Answer:** Because ANN (especially deep neural networks) automatically learns important features from images, eliminating the need for manual feature engineering.
