# CBOW (Continuous Bag of Words) and Skip-Gram
---

## Introduction

**CBOW (Continuous Bag of Words)** and **Skip-Gram** are the **two architectures of the Word2Vec algorithm** used to learn **word embeddings**.

Both methods use a simple **Artificial Neural Network (ANN)** to learn dense vector representations of words.

The main difference lies in **what they predict**.

---

# `Why Do We Need CBOW and Skip-Gram?`

The goal of Word2Vec is **not just to predict words**.

Instead, it creates a **dummy prediction task** so that the neural network learns meaningful **word embeddings**.

These embeddings capture:

- Semantic meaning
- Word similarity
- Contextual relationships

Example:

```
King ≈ Queen

Car ≈ Bus

Doctor ≈ Hospital
```

---

# `1. CBOW (Continuous Bag of Words)`

## Definition

**CBOW predicts the target (center) word using the surrounding context words.**

> **Input = Context Words**  
> **Output = Center Word**

---

# Intuition

Imagine a sentence with one missing word.

Your task is to guess the missing word using the surrounding words.

---

# `Example`

Sentence:

```text
I love learning Machine Learning
```

Suppose the target word is:

```text
learning
```

Context words:

```text
I

love

Machine

Learning
```

Prediction:

```text
I + love + Machine + Learning

↓

learning
```

The model learns to predict **"learning"** from its surrounding words.

# `2. Skip-Gram`

## Definition

**Skip-Gram predicts the surrounding context words using the center word.**

> **Input = Center Word**  
> **Output = Context Words**

---

# Intuition

Instead of guessing the missing word,

the model starts with one word and predicts its neighboring words.

# `Softmax Activation`

Both CBOW and Skip-Gram use the **Softmax Activation Function**.

### Why?

Because the model predicts one word from the entire vocabulary.

Softmax converts the output into probabilities.

Example

```
Cat

↓

0.75

Dog

↓

0.20

Horse

↓

0.05
```

The highest probability is selected.

---

# `Loss Function`

Loss measures the prediction error.

```
Loss

=

Actual Word

−

Predicted Word
```

Goal:

**Minimize Loss**

---

# `Backpropagation`

If prediction is incorrect,

the network updates:

- Weights
- Bias

This process repeats until the loss becomes very small.

---

# `How to Improve Word2Vec Performance`

- Increase training data.
- Increase embedding dimensions (e.g., 100 → 300).
- Choose an appropriate window size.
- Train for more epochs.
- Use a larger vocabulary.
- Remove noisy text before training.

---

# `When Should We Use Which?`

## Use CBOW When

- Dataset is small.
- Faster training is required.
- Frequent words are more important.
- Limited computational resources.

---

## Use Skip-Gram When

- Dataset is very large.
- Better embeddings are required.
- Rare words are important.
- Accuracy is preferred over speed.

# `Key Takeaways`

- **CBOW** predicts the center word using surrounding context words.
- **Skip-Gram** predicts surrounding context words using the center word.
- Both are architectures of the Word2Vec algorithm.
- Both use Artificial Neural Networks and Softmax.

- CBOW is faster and suitable for smaller datasets.
- Skip-Gram is slower but produces better embeddings, especially for rare words and large datasets.

# `2. Interview Questions & Answers`

## Beginner Questions

### Q1. What is CBOW?

**Answer:**
CBOW (Continuous Bag of Words) is a Word2Vec architecture that predicts the **center word** from its surrounding **context words**.

---

### Q2. What is Skip-Gram?

**Answer:**
Skip-Gram is a Word2Vec architecture that predicts the **surrounding context words** from the **center word**.

---

### Q3. What is the main goal of CBOW and Skip-Gram?

**Answer:**
Their main goal is to learn meaningful **word embeddings (dense vectors)** by solving a word prediction task.

---

### Q4. Which activation function is commonly used in Word2Vec?

**Answer:**
**Softmax**, because the model predicts a word from the entire vocabulary.

---

### Q5. What is a window size?

**Answer:**
Window size defines how many neighboring words are considered as context during training.

---

## Intermediate Questions

### Q6. Why is CBOW faster than Skip-Gram?

**Answer:**
CBOW predicts one target word from multiple context words, requiring fewer computations than Skip-Gram, which predicts multiple context words from one target word.

---

### Q7. Which model works better for rare words?

**Answer:**
**Skip-Gram**, because it learns better representations for infrequent words.

---

### Q8. Why do CBOW and Skip-Gram use a neural network?

**Answer:**
The neural network learns weights that become meaningful word embeddings after training.

---

### Q9. What is the role of backpropagation in Word2Vec?

**Answer:**
Backpropagation updates the weights and biases to minimize prediction loss, improving the quality of the learned embeddings.

---

## Scenario-Based Questions

### Q10. You have a small text dataset and limited computing resources. Which Word2Vec architecture would you choose?

**Answer:**
**CBOW**, because it trains faster and performs well on smaller datasets.

---

### Q11. You are training on billions of words and need high-quality embeddings for rare words. Which architecture is more suitable?

**Answer:**
**Skip-Gram**, because it produces better embeddings for rare words on large datasets.

---

### Q12. Why isn't the word prediction task itself the final goal?

**Answer:**
The prediction task is a **dummy objective**. The real objective is to learn high-quality word embeddings that can be used in downstream NLP tasks like sentiment analysis, machine translation, and question answering.
