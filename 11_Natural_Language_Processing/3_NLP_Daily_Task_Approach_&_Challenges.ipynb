# `What are NLP Tasks?`

## Definition

An **NLP Task** is a specific problem related to understanding, processing, or generating human language that an AI system is designed to solve.

Examples include:
- Classifying text
- Translating languages
- Summarizing documents
- Detecting sentiment
- Answering questions

---

# Common NLP Tasks

## `1. Text / Document Classification
`
### Definition

Text Classification is the process of assigning predefined categories or labels to text documents.

It is a **Supervised Machine Learning Classification Problem**.

### Examples

- News Classification
- Email Spam Detection
- Document Categorization
- Resume Classification

Example

```text
Document 1 → Sports

Document 2 → Politics

Document 3 → Entertainment

Document 4 → Technology
```

### Applications

- News websites
- Gmail Spam Filter
- Customer support ticket routing
- Resume screening

---

## `2. Sentiment Analysis`

### Definition

Sentiment Analysis identifies the emotional tone or opinion expressed in text.

The sentiment is generally classified into:

- Positive 😊
- Negative 😠
- Neutral 😐

### Examples

Movie Review

> "The movie was fantastic."

Prediction

Positive

Twitter/X

> "Worst service ever."

Prediction

Negative

### Applications

- Product Reviews
- Social Media Monitoring
- Brand Reputation Analysis
- Customer Feedback

---

## `3. Information Retrieval (IR)`

### Definition

Information Retrieval is the process of finding the most relevant information from a large collection of documents.

### Example

Query

> "What was the JEE Advanced cutoff in 2024?"

Search Engine

↓

Retrieve relevant web pages

↓

Display ranked results

### Applications

- Google Search
- Bing Search
- Enterprise Search Systems
- Digital Libraries

---

## `4. Language Detection & Machine Translation`

### Language Detection

Automatically identifies the language of a text.

Example

```text
Bonjour

↓

French
```

### Machine Translation

Converts text from one language to another while preserving meaning.

Example

```text
English

Hello

↓

Hindi

नमस्ते
```

### Applications

- Google Translate
- DeepL
- Microsoft Translator

---

## `5. Next Word Prediction`

### Definition

Predict the next most likely word based on previous words.

Example

```text
I love playing ______

↓

Cricket
```

### Applications

- ChatGPT
- Gemini
- Gboard

## `6. Chatbots & Question Answering`

### Definition

Chatbots understand user questions and generate relevant responses.

### Pipeline

```text
User Question

↓

NLU

↓

Intent Understanding

↓

Backend Processing

↓

NLG

↓

Generated Response
```

### Examples

- ChatGPT

### Definition

Generate a concise version of a long document while preserving the main ideas.

### Types

### Applications

- Research Papers

- Meeting Notes

## `10. Speech-to-Text (Automatic Speech Recognition)`

### Definition

Convert spoken language into text.

```text
Speech

↓

Audio Signal

↓

Feature Extraction

↓

ASR Model

↓

Text
```

# NLP Approaches

Different NLP problems can be solved using different approaches.

---

# 1. `Heuristic (Rule-Based) Approach`

## Definition

A Heuristic Approach solves problems using manually designed rules instead of learning from data.

It is also called a **Rule-Based Approach**.

### Example

Sentiment Analysis

Rules

Positive Words

- Good
- Amazing
- Excellent

Negative Words

- Bad
- Terrible
- Poor

If:

Positive Words > Negative Words

↓

Positive Review

Otherwise

↓

Negative Review

---

## WordNet

WordNet is a large lexical database where words are connected through semantic relationships such as:

- Synonyms
- Antonyms
- Hypernyms
- Hyponyms

It helps rule-based NLP systems understand word meanings and relationships.

---

## Advantages

- Simple to implement
- Fast execution
- Easy to understand
- No training data required

## Disadvantages

- Difficult to scale
- Cannot handle ambiguity well
- Requires manually written rules
- Performs poorly on complex language

---

# `2. Machine Learning Approach`

## Definition

Instead of manually writing rules, Machine Learning learns patterns from labeled data.

### Workflow

```text
Text Data

↓

Preprocessing

↓

Feature Engineering

↓

Train ML Model

↓

Prediction
```

---

## Feature Engineering

Machine Learning models cannot understand text directly.

Text is converted into numerical vectors using:

- Bag of Words (BoW)
- TF-IDF
- Word2Vec
- GloVe

---

## Popular ML Algorithms

- Naive Bayes
- Logistic Regression
- Support Vector Machine (SVM)
- Latent Dirichlet Allocation (LDA)
- Hidden Markov Models (HMM)

---

## Advantages

- Better accuracy than rule-based systems
- Learns patterns automatically
- Easier to generalize

## Limitations

- Requires feature engineering
- Performance decreases on very large datasets
- Struggles with long sequential context

---

# `3. Deep Learning Approach`

## Definition

Deep Learning automatically learns features from raw text without manual feature engineering.

It is ideal for:

- Huge datasets
- Long text
- Sequential data
- Complex language understanding

---

## Why Deep Learning?

Unlike ML:

- No manual feature engineering
- Better context understanding
- Learns semantic relationships
- Scales to billions of words

---

# `Popular Deep Learning Models`

## 1. RNN (Recurrent Neural Network)

Designed for sequential data.

Advantages

- Understands sequence order

Limitation

- Suffers from the **Vanishing Gradient Problem**
- Cannot remember long-term context effectively

---

## 2. LSTM (Long Short-Term Memory)

Improved version of RNN.

Advantages

- Remembers long-term dependencies
- Maintains context over long sequences

Limitation

- Complex architecture
- Slower training

---

## 3. GRU (Gated Recurrent Unit)

A simplified version of LSTM.

Advantages

- Faster than LSTM
- Fewer parameters
- Good performance on many sequence tasks

---

## 4. BiLSTM (Bidirectional LSTM)

Processes sequences in both forward and backward directions.

Applications

- Named Entity Recognition (NER)
- Part-of-Speech Tagging
- Machine Translation

---

## 5. Transformers

The modern state-of-the-art NLP architecture.

Advantages

- Parallel processing
- Handles long-range dependencies
- Faster training
- Excellent contextual understanding

Examples

- GPT (Generative Pre-trained Transformer)
- BERT (Bidirectional Encoder Representations from Transformers)
- T5
- RoBERTa
- LLaMA

---

# `Challenges in NLP`

Natural language is highly complex and constantly evolving.

---

## 1. Ambiguity

A sentence can have multiple meanings.

Example

> "I saw the man with a telescope."

Who has the telescope?

- Me?
- The man?

---

## 2. Contextual Meaning

The same word may have different meanings in different contexts.

Example

> "I ran to the store because we ran out of milk."

Meaning 1

Ran = Move quickly

Meaning 2

Ran out = Finished

---

## 3. Slang & Idioms

Humans often use expressions whose meanings cannot be understood literally.

Examples

- Piece of cake → Very easy
- Pulling your leg → Joking
- Break a leg → Good luck

---

## 4. Sarcasm

The literal meaning is opposite to the intended meaning.

Example

> "That's just what I needed today!"

Depending on the tone, it could express satisfaction or frustration.

Detecting sarcasm remains a difficult NLP challenge.

---

## 5. Language Diversity

There are over **7,000 spoken languages** worldwide.

Challenges include:

- Different grammar rules
- Regional dialects
- Code-switching (mixing languages)
- Limited training data for low-resource languages

Example

> "Kal meeting hai, don't forget."

This sentence mixes Hindi and English, making processing more challenging.

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What are the common NLP tasks?**
**Answer:** Text classification, sentiment analysis, information retrieval, machine translation, language detection, text summarization, topic modeling, spell checking, speech recognition, and question answering.

**2. What is a heuristic approach in NLP?**
**Answer:** A rule-based approach where manually defined rules are used to solve language problems without learning from data.

**3. Why is feature engineering required in Machine Learning?**
**Answer:** ML algorithms cannot process raw text, so text must be converted into numerical features using techniques like BoW or TF-IDF.

**4. What is MFCC?**
**Answer:** **Mel-Frequency Cepstral Coefficients (MFCC)** is a feature extraction technique that converts speech signals into numerical representations for speech recognition models.

**5. Why are Transformers preferred over RNNs?**
**Answer:** Transformers process sequences in parallel, capture long-range dependencies more effectively, and train faster than RNN-based models.

---

## Intermediate Questions

**1. Compare Heuristic, Machine Learning, and Deep Learning approaches in NLP.**

| Heuristic             | Machine Learning              | Deep Learning                            |
| --------------------- | ----------------------------- | ---------------------------------------- |
| Rule-based            | Learns from labeled data      | Learns automatically from large datasets |
| No training           | Requires training             | Requires large-scale training            |
| Manual rules          | Manual feature engineering    | Automatic feature learning               |
| Best for simple tasks | Good for structured NLP tasks | Best for modern NLP and LLMs             |

**2. Why do RNNs struggle with long text?**
**Answer:** They suffer from the **Vanishing Gradient Problem**, making it difficult to retain long-term context.

**3. What problem does LSTM solve?**
**Answer:** LSTM introduces memory cells and gates that allow it to preserve important information over long sequences.

**4. What is Topic Modeling?**
**Answer:** An unsupervised NLP technique that identifies hidden themes or topics within a collection of documents.

---

## Scenario-Based Questions

**Q1. An e-commerce company wants to classify product reviews into Positive, Negative, or Neutral. Which NLP task and approach would you choose?**
**Answer:** Use **Sentiment Analysis** as the task and a Machine Learning or Transformer-based Deep Learning model depending on the dataset size and complexity.

**Q2. A company needs an AI assistant to summarize lengthy legal documents. Which NLP task is suitable?**
**Answer:** **Text Summarization**, preferably using Transformer models such as **BART** or **T5**.

**Q3. Why would a multilingual chatbot struggle with sarcasm or slang?**
**Answer:** Sarcasm and slang depend heavily on context, culture, and tone, making them challenging for NLP systems to interpret accurately.
