### `Handling Emojis`
- understand the meaning of emoji
- eg. spam detection / Topic modelling

````python
import re

def remove_emoji(text):
    # Fixed the \U000IFIFF typo to \U0001F1FF
    # Combined standard emoji blocks, transport, symbols, and flags
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # Emoticons
        u"\U0001F300-\U0001F5FF"  # Misc Symbols/Pictographs
        u"\U0001F680-\U0001F6FF"  # Transport/Map
        u"\U0001F1E0-\U0001F1FF"  # Regional Indicator Pairs (Flags)         # Misc Symbols / Dingbats
        "]+",
        flags=re.UNICODE
    )
    # Changed 'r' to '' to properly remove the emojis instead of inserting 'r'
    return emoji_pattern.sub(r'', text)
````

# `Advanced Text Cleaning & Linguistic Preprocessing in NLP`

> **Advanced Text Cleaning** goes beyond removing noise. It helps machines understand the grammatical structure, relationships, and meaning of language before feature engineering or model training.

---

# `What is Advanced Text Cleaning?`

## Definition

Advanced Text Cleaning (also called **Advanced Text Preprocessing** or **Linguistic Preprocessing**) consists of NLP techniques that improve a model's understanding of language by analyzing grammar, syntax, semantics, and context.

Unlike basic text cleaning, which removes unwanted characters or normalizes text, advanced preprocessing helps the model understand **what each word means, how words relate to each other, and what the entire sentence conveys**.

---

# `Basic Cleaning vs Advanced Cleaning`

| Basic Text Cleaning | Advanced Text Cleaning |
|---------------------|------------------------|
| Removes noise | Understands language |
| Removes URLs | Finds grammatical structure |
| Removes punctuation | Identifies entities |
| Lowercasing | Understands context |
| Regex operations | Understands sentence meaning |
| Pattern matching | Linguistic analysis |

---

Simple cleaning cannot answer questions like:

Sentence:

Only advanced NLP techniques can determine the correct meaning from context.

# `Major Advanced Text Cleaning Techniques`

## `1. Part-of-Speech (POS) Tagging`

### Definition

POS Tagging assigns a grammatical role to every word in a sentence.

Each word is labeled as:

- Noun
- Verb
- Pronoun
- Adjective
- Adverb
- Determiner
- Preposition
- Conjunction
- Interjection

---

### Example

Sentence

```text
Arun is learning NLP.
```

POS Tags

| Word | POS |
|------|------|
| Arun | Proper Noun |
| is | Verb |
| learning | Verb |
| NLP | Proper Noun |

---

### Another Example

Sentence

```text
The beautiful girl sings well.
```

| Word | POS |
|------|------|
| The | Determiner |
| beautiful | Adjective |
| girl | Noun |
| sings | Verb |
| well | Adverb |

---

### Applications

- Chatbots
- Machine Translation
- Grammar Checking
- Question Answering
- Information Extraction

---

## `2. Parsing`

### Definition

Parsing identifies the grammatical structure (syntax) of a sentence and determines how words are related.

It answers questions like:

- Who is the subject?
- What is the verb?
- What is the object?

---

### Types of Parsing

#### A. Dependency Parsing

Shows relationships between words.

Example

```text
The cat chased the mouse.
```

Relationships

```text
Subject → cat

Verb → chased

Object → mouse
```

---

#### B. Constituency Parsing

Breaks a sentence into grammatical phrases.

Example

```text
[The cat]
[chased]
[the mouse]
```

---

### Applications

- Grammar Checking
- Machine Translation
- Chatbots
- Information Extraction

---

## `3. Named Entity Recognition (NER)`

### Definition

NER identifies and classifies important entities in text.

Entities include:

- Person
- Organization
- Location
- Date
- Money
- Product
- Disease

---

### Example

Sentence

```text
Elon Musk founded SpaceX in California in 2002.
```

NER Output

| Word | Entity |
|------|---------|
| Elon Musk | Person |
| SpaceX | Organization |
| California | Location |
| 2002 | Date |

---

### Applications

- Resume Parsing
- Medical NLP
- Finance
- Search Engines
- News Analysis

---

## `4. Chunking (Shallow Parsing)`

### Definition

Chunking groups related words into meaningful phrases.

---

### Example

Sentence

```text
The quick brown fox jumps.
```

Chunks

```text
[Noun Phrase]

The quick brown fox

[Verb Phrase]

jumps
```

---

### Applications

- Information Extraction
- Question Answering
- Chatbots

---

## `5. Lemmatization`

### Definition

Lemmatization converts words into their correct dictionary (base) form using grammar rules.

---

### Examples

| Word | Lemma |
|------|--------|
| Running | Run |
| Studies | Study |
| Better | Good |
| Children | Child |

---

### Why POS Matters

Word

```text
meeting
```

Without POS

↓

meeting

As Verb

↓

meet

As Noun

↓

meeting

This is why POS tagging is often performed **before lemmatization**.

---

## `6. Word Sense Disambiguation (WSD)`

### Definition

Determines the correct meaning of a word based on context.

---

### Example

Sentence

```text
He sat on the bank.
```

Possible meanings

- River Bank
- Bank Account

The surrounding words determine the correct meaning.

---

### Another Example

```text
She deposited money in the bank.
```

Meaning

Financial Institution

---

## `7. Coreference Resolution`

### Definition

Identifies when multiple words refer to the same entity.

---

### Example

Sentence

```text
Arun bought a laptop.

He loves it.
```

Resolved

- He → Arun
- It → Laptop

---

### Why It Matters

Without coreference resolution, a machine cannot understand who "he" or "it" refers to.

---

### Applications

- Chatbots
- Document Understanding
- Legal Documents
- Healthcare

---

## `8. Negation Handling`

### Definition

Negation changes the meaning of a sentence.

---

### Example

Sentence

```text
The movie is good.
```

Positive

Sentence

```text
The movie is not good.
```

Negative

---

### Why It Matters

If the model ignores "not", sentiment analysis becomes incorrect.

---

# Best Practices

- Perform basic cleaning before advanced preprocessing.
- Do not remove words that affect meaning (e.g., "not").
- Use POS tagging before lemmatization for more accurate base forms.
- Preserve emojis when performing sentiment analysis.
- Apply only the techniques relevant to your NLP task to avoid unnecessary processing.

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What is Advanced Text Cleaning in NLP?**
**Answer:** It is the process of applying linguistic preprocessing techniques such as POS Tagging, Parsing, NER, and Coreference Resolution to help machines understand the meaning and structure of language.

**2. What is POS Tagging?**
**Answer:** POS Tagging assigns grammatical categories (noun, verb, adjective, etc.) to each word in a sentence.

**3. What is Named Entity Recognition (NER)?**
**Answer:** NER identifies and classifies entities such as people, organizations, locations, dates, and money in text.

**4. What is Parsing?**
**Answer:** Parsing determines the grammatical structure of a sentence and identifies relationships between words.

**5. What is Coreference Resolution?**
**Answer:** It resolves references such as pronouns (he, she, it) to the entities they represent.

---

## Intermediate Questions

**1. Why is POS Tagging performed before Lemmatization?**
**Answer:** The correct lemma often depends on the word's grammatical role. POS information enables more accurate lemmatization.

**2. Explain the difference between Parsing and Chunking.**
**Answer:** Chunking groups words into phrases (e.g., noun phrases), while Parsing analyzes the complete grammatical relationships within a sentence.

**3. What is Word Sense Disambiguation (WSD)?**
**Answer:** WSD identifies the correct meaning of a word with multiple possible meanings by analyzing its surrounding context.

**4. Why is Negation Handling important in Sentiment Analysis?**
**Answer:** Negation words such as "not" can completely reverse the sentiment of a sentence, making correct handling essential.

---

## Scenario-Based Questions

**Q1. A chatbot cannot determine who "he" refers to in a conversation. Which NLP technique is missing?**
**Answer:** Coreference Resolution.

**Q2. A legal document contains many names of organizations, dates, and locations. Which technique should be applied?**
**Answer:** Named Entity Recognition (NER).

**Q3. A sentiment analysis model classifies "The product is not good" as positive. What preprocessing improvement would you recommend?**
**Answer:** Implement proper Negation Handling so that the model captures the change in sentiment introduced by "not".
