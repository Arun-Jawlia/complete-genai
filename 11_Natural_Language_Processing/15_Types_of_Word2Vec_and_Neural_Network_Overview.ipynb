# `Word2Vec Techniques and Neural Network Overview`
---

## Introduction

**Word2Vec** is a technique used in **Natural Language Processing (NLP)** to convert words into **dense vector representations** called **embeddings**.

Instead of representing words as simple text, Word2Vec learns the meaning of words based on their context.

> **Word2Vec = Learning word embeddings from context**

---

# Why Do We Need Word2Vec?

Traditional text representation methods like one-hot encoding have limitations:

- Very sparse
- Do not capture meaning
- Do not capture similarity between words

### Example

Words like:

- king
- queen
- prince

should have related meanings, but one-hot encoding treats them as unrelated.

Word2Vec solves this problem by learning embeddings where similar words are close to each other in vector space.

---

# Google Pretrained Word2Vec

Google released a pretrained Word2Vec model trained on a huge corpus.

### Benefits

- Already learned useful embeddings
- Can be used directly in NLP applications
- Saves training time

### Important Idea

Word2Vec is often trained using a **dummy prediction task**.
The real goal is not the dummy task itself, but the **embeddings** learned as a by-product.

---

# Types of Word2Vec

There are **2 main techniques**:

1. **CBOW** (Continuous Bag of Words)
2. **Skip-Gram**

---

# 1. CBOW (Continuous Bag of Words)

## What is CBOW?

CBOW predicts the **center word** using the **context words** around it.

> **Input = context words**  
> **Output = target word**

### Simple Meaning

The model looks at nearby words and tries to predict the missing word.

---

## Example Sentence

```text
hello welcome to my AI course
```

If the window size is 2, the model may use words around the target word.

### Example

Context words:

- hello
- to

Target word:

- welcome

So the model learns:

```text
hello + to -> welcome
```

---

## Why is it called a Dummy Problem?

The purpose is not really to predict missing words for its own sake.

The actual goal is to train a neural network so that it learns **good embeddings**.

So CBOW is a **training trick** to learn word vectors.

---

## How CBOW Works

1. Take context words as input
2. Convert words into vectors
3. Pass vectors through a neural network
4. Predict the center word
5. Update weights using loss minimization
6. Repeat until embeddings improve

---

# 2. Skip-Gram

## What is Skip-Gram?

Skip-Gram is the **reverse of CBOW**.

It predicts the **context words** using the **center word**.

> **Input = target word**  
> **Output = context words**

---

## Example Sentence

```text
hello welcome to my AI course
```

If the target word is:

- welcome

Then the model tries to predict nearby words such as:

- hello
- to

So the model learns:

```text
welcome -> hello, to
```

---

## How Skip-Gram Works

1. Take the center word as input
2. Convert it into vector form
3. Pass it through a neural network
4. Predict nearby context words
5. Update weights using loss minimization
6. Repeat until embeddings improve

---

# CBOW vs Skip-Gram

| CBOW | Skip-Gram |
|------|-----------|
| Predicts target word | Predicts context words |
| Input = context | Input = center word |
| Faster | Slower |
| Better for smaller datasets | Better for larger datasets |
| Works well when context is important | Works well when word meaning is important |

---

# When to Use CBOW vs Skip-Gram

## Use CBOW when:

- Dataset is smaller
- You want faster training
- Context words are more important

## Use Skip-Gram when:

- Dataset is large
- You want better embeddings
- You want to learn richer word relationships

---

# Window Size

## What is Window Size?

Window size decides how many surrounding words are considered as context.

### Example

Sentence:

```text
hello welcome to my AI course
```

If window size = 2, the model will use nearby words within 2 positions.

### Why is it important?

- Small window size → local context
- Large window size → broader context

---

# Neural Network Overview

## Goal of Deep Learning

Deep Learning tries to build a mathematical system inspired by the **human brain**.

### Idea

- Human brain has neurons
- Neural networks have artificial neurons
- Deep learning uses multiple layers of neurons

> **Deep Learning = Collection of artificial neurons working together**

---

# Artificial Neural Network (ANN)

## What is ANN?

An **Artificial Neural Network (ANN)** is a network of connected artificial neurons that process input data and produce output.

### Layers of ANN

1. Input Layer
2. Hidden Layer(s)
3. Output Layer

---

# Perceptron

## What is a Perceptron?

A **Perceptron** is the simplest artificial neuron.

It performs a mathematical calculation:

```text
y = x1w1 + x2w2 + ... + b
```

Where:

- `x1, x2, ...` = input features
- `w1, w2, ...` = weights
- `b` = bias

---

# Weights and Bias

## Weights

Weights show the **importance** of each input.

- High weight → more importance
- Low weight → less importance

## Bias

Bias helps shift the output so the model can make better predictions.

> Bias is adjusted during training to reduce prediction error.

---

# Example: Cat vs Dog Classifier

Suppose the model is trying to classify images as:

- Cat
- Dog

The inputs may include features like:

- Nose shape
- Ear shape
- Eye shape
- Fur texture

Some features may be more important than others.

### Example

- Nose shape → high importance
- Eye shape → less clear
- Fur texture → useful supporting feature

The model learns weights for these features and uses them to predict the final class.

---

# Activation Function

## Why do we need it?

A neuron should not always produce a simple linear output.

Activation functions help decide:

- whether a neuron should activate
- how strongly it should contribute to the next layer

---

## Common Activation Functions

### 1. Sigmoid

- Output range: 0 to 1
- Useful for binary classification

### 2. Softmax

- Converts outputs into probabilities
- Useful for multiclass classification

### 3. Threshold Function

- Used in simple classification logic
- Produces output based on a cutoff value

---

# Softmax Activation

## What is Softmax?

Softmax converts raw outputs into probabilities.

### Example

For classes:

- Cat
- Dog
- Horse

Softmax gives probabilities for each class and the class with the highest probability becomes the final prediction.

> Softmax is commonly used in multiclass classification.

---

# Loss Function

## What is Loss?

Loss measures how far the prediction is from the actual output.

### Idea

```text
Loss = Actual Value - Predicted Value
```

The goal is to **minimize loss**.

---

# Backpropagation

## What is Backpropagation?

If the prediction is wrong, the network adjusts:

- weights
- bias

by going backward through the network.

This process is called **Backpropagation**.

> Backpropagation helps the model learn from mistakes.

---

# How Word2Vec Uses Neural Networks

Word2Vec uses a neural network to learn embeddings.

### Flow

```text
Input Words
      ↓
Neural Network
      ↓
Predicted Word / Context
      ↓
Loss Calculation
      ↓
Backpropagation
      ↓
Updated Embeddings
```

The final output is not just prediction accuracy.

The real output is the **learned word embeddings**.

---

# Dense Vector

A **dense vector** is a compact numerical representation of a word.

### Example

Instead of:

```text
hello = [0, 0, 1, 0, 0, 0, 0, 0]
```

Word2Vec produces something like:

```text
hello = [0.12, -0.44, 0.89, ...]
```

Dense vectors:

- are shorter
- store meaning
- capture similarity

---

# How to Improve Word2Vec Performance

To make Word2Vec better:

1. Increase training data
2. Increase embedding dimension
3. Increase window size carefully
4. Train for more epochs
5. Use a larger and better corpus
