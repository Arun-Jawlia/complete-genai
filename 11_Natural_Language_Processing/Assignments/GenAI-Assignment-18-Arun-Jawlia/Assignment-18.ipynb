## Task 1: Manula One-Hot Encoding

```python
sentences = df["final_clean_text"].dropna().head(10).tolist()

for sentence in sentences:
    print(sentence)
```

```python
# Create vocabulary
vocabulary = set()
for sentence in sentences:
    words = sentence.split()
    for word in words:
        vocabulary.add(word)
vocabulary = sorted(list(vocabulary))

print(vocabulary)
```

```python
one_hot_vectors = []

for sentence in sentences:
    words = sentence.split()
    vector = []
    for word in vocabulary:
        if word in words:
            vector.append(1)
        else:
            vector.append(0)
    
    one_hot_vectors.append(vector)
```

```python
for i in range(len(sentences)):
    print('vectos', one_hot_vectors[i])
```

```python
one_hot_matrix = pd.DataFrame(
    one_hot_vectors,
    columns=vocabulary
)

print(one_hot_matrix)
```

# Part 2: Bag of Words(BoW)

## Task 3: Bag of Words Representation

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
texts = df["final_clean_text"].fillna("")
```

```python
bow = CountVectorizer()
bow_matrix = bow.fit_transform(texts)
```

```python
vocabulary = bow.get_feature_names_out()

len(vocabulary)
```

```python
len(bow.vocabulary_)
```

```python
bow_matrix[:5].toarray()
```

```python
bow_df = pd.DataFrame(
    bow_matrix[:5,:10].toarray(),
    columns=vocabulary[:10]
)
```

```python
# Bow Dataframe 
bow_df
```

## Task 3: Calculate total Frequence

```python
word_frequency = bow_matrix.sum(axis=0)
word_frequency = word_frequency.A1
```

```python
frequency_df = pd.DataFrame({"word": vocabulary,"frequency": word_frequency})
frequency_df.head()
```

Note: Bag of Words represents each document using word counts. If a word appears multiple times in a document, its value will be higher. By adding the columns of the BoW matrix, we can calculate the total frequency of each word in the complete dataset. Therefore, BoW provides information about how often each word occurs.
