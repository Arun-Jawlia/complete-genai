## Task 1: Manula One-Hot Encoding

```python
sentences = df["final_clean_text"].dropna().head(10).tolist()

for sentence in sentences:
    print(sentence)
```

```python
# Create vocabulary
vocabulary = set()
for sentence in sentences:
    words = sentence.split()
    for word in words:
        vocabulary.add(word)
vocabulary = sorted(list(vocabulary))

print(vocabulary)
```

```python
one_hot_vectors = []

for sentence in sentences:
    words = sentence.split()
    vector = []
    for word in vocabulary:
        if word in words:
            vector.append(1)
        else:
            vector.append(0)
    
    one_hot_vectors.append(vector)
```

```python
for i in range(len(sentences)):
    print('vectos', one_hot_vectors[i])
```

```python
one_hot_matrix = pd.DataFrame(
    one_hot_vectors,
    columns=vocabulary
)

print(one_hot_matrix)
```

Note: Bag of Words represents each document using word counts. If a word appears multiple times in a document, its value will be higher. By adding the columns of the BoW matrix, we can calculate the total frequency of each word in the complete dataset. Therefore, BoW provides information about how often each word occurs.

# TASK 10

### Question: Difference between One-Hot Encoding and BoW
- One hot encoding represents whether a word is present or basent using 0 and 1
- Bag of words represents how many times a words appear in a document

### Why N-grams increase dimensionality
- N-Grams create combinations of consecutive words. Unigram contain individual words, while bigram and trigram contain two words and three words
- N-Grams create many additional features and increase the vocabulary size

### When to prefer TF-IDF over BoW
- TFIDF is preferred when common words are less useful and want to give more importance to words that are more specific to particular documents
- Bow only considers word frequncy, while TF-IDF also considerss how common a word is across the complete dataset

### Limitations of count based vectorization
- creates a large and sparse vocabulary
- does not understand the menaing of words
- does not understand relationships between words unless N-Grams are used
- synonyms are treated as different words
- require a lot of memory for larger datasets
