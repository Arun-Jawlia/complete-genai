```python
def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # Emoticons
        "\U0001F300-\U0001F5FF"  # Symbols & Pictographs
        "\U0001F680-\U0001F6FF"  # Transport
        "\U0001F1E0-\U0001F1FF"  # Flags
        "\U00002700-\U000027BF"
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )

    return emoji_pattern.sub('', text)
```

```python
data['review'] =data['review'].apply(remove_emoji)
```

## Task 4: Handling Stopwords

```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
english = stopwords.words("english")
```

```python
def stopwords_removal(text):
    li = []
    for i in text.split():
        if i in english:
            li.append('')
        else:
            li.append(i)
    
    x = li[:]
    li.clear()
    return " ".join(x)
```

```python
data['text_no_stopwords'] =data['clean_text_advanced'].apply(stopwords_removal) 
```

## Task 5: Hanlding Repeated Characters & Slang(Optional)

```python
def remove_repeated_characters(text):
    return re.sub(r'(.)\1{2,}', r'\1', text)
```

```python
slang_dict = {
    "u": "you",
    "ur": "your",
    "gr8": "great",
    "btw": "by the way",
    "idk": "i do not know",
    "lol": "laughing",
    "omg": "oh my god",
    "thx": "thanks",
    "pls": "please",
    "msg": "message",
    "b4": "before",
    "luv": "love"
}
```

```python
def replace_slang(text):
    words = text.split()

    new_words = [
        slang_dict[word] if word in slang_dict else word
        for word in words
    ]

    return " ".join(new_words)
```

```python
data['text_no_stopwords']
```

```python
data["text_no_stopwords"] =data["text_no_stopwords"].apply(remove_repeated_characters)
```

```python
data["text_no_stopwords"] = data["text_no_stopwords"].apply(replace_slang)
```

# Task 9: Final NLP Pipeline Creation

```python
import string
from nltk import pos_tag
from nltk.corpus import wordnet
```

```python
nltk.download("averaged_perceptron_tagger_eng")
```

```python
stop_words = set(stopwords.words("english"))
```

```python
def get_wordnet_pos(tag):

    if tag.startswith("J"):
        return wordnet.ADJ

    elif tag.startswith("V"):
        return wordnet.VERB

    elif tag.startswith("N"):
        return wordnet.NOUN

    elif tag.startswith("R"):
        return wordnet.ADV

    return wordnet.NOUN
```

```python
def nlp_preprocess(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+','',text)

    text = re.sub(r'\S+@\S+','',text)

    text = re.sub(r'<.*?>','',text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\d+','',text)
    text = re.sub(r'[^a-zA-Z\s]','',text)

    text = re.sub(r'\s+',' ',text).strip()

    tokens = word_tokenize(text)
    tokens = [
        word
        for word in tokens
        if word not in stop_words
    ]

    pos_tags = pos_tag(tokens)

    lemmatized_words = []

    for word, tag in pos_tags:
        wordnet_pos = get_wordnet_pos(tag)
        lemma = lemmatizer.lemmatize(
            word,
            pos=wordnet_pos
        )

        lemmatized_words.append(lemma)
    return " ".join(lemmatized_words)
```

```python
data['final_clean_text'] = df['review'].apply(nlp_preprocess)
```
