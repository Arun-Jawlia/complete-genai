### Task 2: Basic Text Cleaning

```python
def apply_lower_case(text):
    return text.lower()
```

```python
data['review'] = data['review'].apply(apply_lower_case)
```

```python
def punctuation_removal(text):
    return re.sub(r'[^\w\s]', '', text)
```

```python
data['review'] = data['review'].apply(punctuation_removal)
```

```python
punctuation = re.findall(r"[^\w\s]", data['review'][1])
punctuation[:30]
```

```python
def remove_html_tage(text):
    return re.sub(r"<.*?>", '', text)
```

```python
data['review'] = data['review'].apply(remove_html_tage)
```

```python
html = re.findall(r"<.*?>", data['review'][0])
html
```

```python
def remove_numbers(text):
    return re.sub(r"\d+", "", text)
```

```python
data['review'] = data['review'].apply(remove_numbers)
```

```python
def remove_extra_space(text):
    return re.sub(r"\s+", " ", text).strip()
```

```python
data['review'] = data['review'].apply(remove_extra_space)
```

## Part 2: Advanced Text Cleaning

### Task 3: Removing Noise

```python
def url_removal(text):
    return re.sub(r"https?://\S+|www\.\S+", '', text)
```

```python
data['review'] = data['review'].apply(url_removal)
```

```python
def email_removal(text):
    return re.sub(r"https?://\S+|www\.\S+", '', text)
```

```python
data['review'] = data['review'].apply(email_removal)
```

```python
def find_email(text):
    return re.findall(r"https?://\S+|www\.\S+", text)
```

```python
data['review'].apply(find_email)
```
