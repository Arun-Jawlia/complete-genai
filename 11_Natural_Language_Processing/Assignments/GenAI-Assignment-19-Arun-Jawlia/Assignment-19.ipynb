## TASK 1: Understanding Word Embeddings

1. What are word embeddings?
- Word embeddings are numerical representations of words in the form of vectors. The main idea is that words with similar meanings of similar usage shuld have similar vector.

2. Why one-hot encoding and BoW fail to capture semantics?
- One hot encoding represents every word using vector where only one positino is 1 and all other possitions are 0. the problesm is that these vectors don't tell us anything about the relationship between words.

3. How word emebddings solve these problems?
- Word Emebddings represents words ar dense numerical vectors learned from their context.
- Words that appear in similar contexts tend to get similar vector representations
- Unlike One-Hot Encoding, embeddings don't treat every word as completely unrelated. They allow mathematical operations and similarity calculations between words.

## TASK 2: Word2Vec Overview and techniques

1. Explain What Word2Vec is?
- Ans: Word2Vec is a popular technique for learning word embeddings from text. It was introduced by researchers at Google.
- Word2Vec learns the representation of a word by looking at the words that appear around it.

2. Explain the idea of prediction words from context
- Basic idea behind Word2Vec is that A word can be understood from the words that usually appear around it.
- eg: The cat drinks milk --> If cat is the target word, the surrounding words give information about the word.

3. Define 
   - Vocabulary: It is the collection of unique words present in the training dataset.
   - Context window: It defines how many words around a target word are considered its context
   - Embeddings Dimensions: It is the number of  numerical values used to represent each word.

## TASK 3: Types of Word2Vec Techniques

1. CBOW (Continuous Bag of Words): 
- CBOW predicts the target word using its surround context words
- CBOW is generally faster to train because its combine the surrounding context to predict the target word. it works well when the dataset is larget and contain many common words.

2. Skip-Gram
- Skip-Gram works in the opposite direction. It uses a target words to predict its surrounding context words.
- Skip-Gram usually takes more training time than CBOW because it can generate multiple target-context training pairs

## TASK 4: Neural Network intuition Behind Word2Vec

1. Input Layer Representation
- it is the word representation layer.
- Initially words can be represented using one hot vectors
- This input layers recieves all the representation

2. Hidden layer (embeddings layer)
- Input layer is connected to a hidden layer.
- This layer contains the actual embeddings representations.
- During training the weights connecting the input to this layer are updated.

3. Output layer
- The output layer contains the vocabulary words.

4. How weights become word embeddings
- At the beginning, the network weights are basically random. During training, the model repeatedly sees words and their contexts.
- The model makes prediction and calculate the error.
- Then it updates the weights to reduce that error.
- The process happens over many sentences and many training examples.

```python
import re
import nltk
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
cleaned_corpus = []

for sentence in corpus:
    sentence = sentence.lower()
    sentence = re.sub(r"[^a-zA-Z\s]", "", sentence)
    sentence = re.sub(r"\s+", " ", sentence).strip()
    cleaned_corpus.append(sentence)
print(cleaned_corpus[:5])
```

```python
text = " ".join(cleaned_corpus)
sentences = sent_tokenize(text)
print("Number of sentences:", len(sentences))
```

```python
tokenized_sentences = []
for sentence in sentences:
    words = word_tokenize(sentence)
    words = [word for word in words if word.isalpha()]
    tokenized_sentences.append(words)
```

```python
tokenized_sentences[:5]
```

## TASK 6:

```python
from gensim.models import Word2Vec
```

```python
model = Word2Vec(sentences=tokenized_sentences,vector_size=100,window=5,min_count=1,sg=0,workers=4,epochs=100)
```

```python
vocab_size = len(model.wv.key_to_index)
```

```python
vocab_size
```

```python
word = "python"
vector = model.wv[word]
print(vector)
```

```python
len(model.wv['python'])
```

```python
from gensim.models import word2vec
import time
```

```python
start_time = time.time()

skipgram_model = Word2Vec(sentences=tokenized_sentences,vector_size=100,window=5,min_count=1,sg=1,workers=4,epochs=100)
skipgram_time = time.time() - start_time
```

```python
print("Skip-Gram training time:", skipgram_time, "seconds")
print("Vocabulary size:", len(skipgram_model.wv))
```

```python
start_time = time.time()
cbow_model = Word2Vec(sentences=tokenized_sentences,vector_size=100,window=5,min_count=1,sg=0,workers=4,epochs=100)
cbow_time = time.time() - start_time
```

```python
print("CBOW training time:", cbow_time, "seconds")
print("Skip-Gram training time:", skipgram_time, "seconds")
```

```python
print("CBOW similar words:")
print(cbow_model.wv.most_similar("python", topn=5))
```

```python
print("\nSkip-Gram similar words:")
print(skipgram_model.wv.most_similar("python", topn=5))
```

# TASK 10: Observations and Limitations

1. Difference between CBOW and Skip-Gram in practice
- CBOW uses surrounding context words to predict the target word.
- Skip-Gram uses the target word to predict surrounding context words.
- In my experiment, the difference was small because the corpus contains only 40 sentences.

2. Advantages of Word2Vec orver TF-IDF
- Word2Vec creates dense vectors instead of sparse vectors.
- It captures contextual relationships between words.
- Similar words can have similar vector representations.
- Vector arithmetic can represent certain semantic relationships.
- The vectors can be used as features for other NLP tasks.

3. Limiations of Word2Vec
- Small corpus problem
- One vector per word
- Out-of-vocabulary words: If a word was not present during training, Word2Vec cannot provide its vector.
- Require huge data

4. Why Context still matters in modern NLP ( lead-in to transformers)
- Context is still fundamental in modern NLP.
- Word2Vec was an important step because it showed that the context in which a word appears can be used to learn meaningful representations.
