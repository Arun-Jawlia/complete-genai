# `Text Preprocessing in NLP: Tokenization, Stemming & Lemmatization`

> **Text Preprocessing** is the process of transforming raw text into a clean, structured, and machine-readable format before feature engineering and model training.

---

# `What is Text Preprocessing?`

## Definition

Text Preprocessing is one of the most important stages in the NLP pipeline. It prepares text data by applying techniques that help machines understand language more effectively.

Unlike **Text Cleaning**, which removes unwanted noise (URLs, HTML tags, emojis, etc.), **Text Preprocessing** focuses on transforming the cleaned text into a format suitable for Machine Learning or Deep Learning models.

---

# `Why is Text Preprocessing Important?`

Imagine you have the following sentence:

```text
I am Learning NLP, and I absolutely LOVE it!
```

A machine may treat:

- Learning
- learning
- LEARNING

as different words.

Similarly,

- running
- runs
- ran

represent the same concept but appear as different words.

Text preprocessing standardizes such variations, reducing vocabulary size and improving model performance.

---

# `Major Text Preprocessing Techniques`

This lecture covers the three most fundamental preprocessing techniques:

1. Tokenization
2. Stemming
3. Lemmatization

---

# `1. Tokenization`

## Definition

**Tokenization** is the process of breaking text into smaller meaningful units called **tokens**.

A token can be:

- Word
- Sentence
- Character
- Subword

Tokenization is almost always the **first preprocessing step** in NLP.

---

# Why Do We Need Tokenization?

Machines cannot process an entire paragraph efficiently.

Instead, we split the text into manageable pieces.

Example

Sentence

```text
I love Machine Learning.
```

After Tokenization

```python
["I", "love", "Machine", "Learning"]
```

Now each word can be processed independently.

---

# `Types of Tokenization`

## A. Word Tokenization

Splits text into individual words.

Example

```text
Hello, I am Arun.
```

↓

```python
["Hello", "I", "am", "Arun"]
```

---

## B. Sentence Tokenization

Splits a paragraph into individual sentences.

Example

```text
Hello everyone.
How are you?
I am learning NLP.
```

↓

```python
[
 "Hello everyone.",
 "How are you?",
 "I am learning NLP."
]
```

---

## C. Character Tokenization

Splits text into individual characters.

Example

```text
NLP
```

↓

```python
["N","L","P"]
```

Used less frequently but useful in tasks such as OCR and language modeling.

---

## D. Subword Tokenization

Breaks rare or unknown words into smaller meaningful units.

Example

```text
unbelievable
```

↓

```text
un + believe + able
```

Used in modern Transformer models like:

- BERT
- GPT
- LLaMA
- Gemini

Popular algorithms:

- Byte Pair Encoding (BPE)
- WordPiece
- SentencePiece

---

# Advantages of Tokenization

- Reduces processing complexity.
- Makes text machine-readable.
- Foundation for feature engineering.
- Essential for embeddings and Transformer models.

---

# Python Example

```python
from nltk.tokenize import word_tokenize

text = "I love NLP."

tokens = word_tokenize(text)

print(tokens)
```

Output

```python
['I', 'love', 'NLP', '.']
```

---

# `2. Stemming`

## Definition

**Stemming** reduces a word to its **stem (root)** by removing prefixes or suffixes using simple rules.

The resulting word **may not be a valid dictionary word**.

Think of stemming as **chopping off endings** to reduce words to a common form.

---

# Why Do We Need Stemming?

Different forms of the same word should be treated as one feature.

Example

```text
play

playing

played

plays
```

↓

```text
play
```

This reduces vocabulary size and improves learning.

---

# Examples

| Original Word | Stem |
|---------------|------|
| Playing | Play |
| Played | Play |
| Player | Player (may vary) |
| Studies | Studi |
| Running | Run |
| Happiness | Happi |

Notice that:

```text
Studies

↓

Studi
```

"Studi" is **not** a real English word.

This is one limitation of stemming.

---

# Common Stemming Algorithms

- Porter Stemmer
- Snowball Stemmer
- Lancaster Stemmer

---

# Python Example

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmer.stem("playing")
```

Output

```text
play
```

---

# Advantages of Stemming

- Fast
- Simple
- Reduces vocabulary size
- Works well for search engines and information retrieval

---

# Limitations of Stemming

- Produces invalid words
- Ignores grammar
- Less accurate than lemmatization

---

# `3. Lemmatization`

## Definition

**Lemmatization** converts words into their **lemma**, i.e., the correct dictionary base form, using vocabulary and grammar rules.

Unlike stemming, the output is always a meaningful word.

---

# Why Do We Need Lemmatization?

Words can have different forms depending on tense or grammar.

Lemmatization preserves meaning while reducing words to a common base.

---

# Examples

| Original Word | Lemma |
|---------------|--------|
| Running | Run |
| Studies | Study |
| Better | Good |
| Children | Child |
| Mice | Mouse |
| Went | Go |

Notice:

```text
Better

↓

Good
```

This is possible because lemmatization uses linguistic knowledge.

---

# POS Tagging and Lemmatization

Lemmatization works best when combined with **Part-of-Speech (POS) Tagging**.

Example

Word

```text
meeting
```

As a noun

↓

meeting

As a verb

↓

meet

The POS tag helps determine the correct lemma.

---

# Python Example

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatizer.lemmatize("running", pos="v")
```

Output

```text
run
```

---

# Advantages of Lemmatization

- Produces valid dictionary words
- Preserves grammatical meaning
- More accurate than stemming
- Better for chatbots, search engines, and language understanding

---

# Limitations of Lemmatization

- Slower than stemming
- Requires linguistic resources (e.g., WordNet)
- Often benefits from POS tagging

---

# `Stemming vs Lemmatization`

| Feature | Stemming | Lemmatization |
|----------|----------|---------------|
| Method | Rule-based | Dictionary + Grammar |
| Output | Stem | Lemma |
| Valid English Word | Not always | Yes |
| Speed | Faster | Slower |
| Accuracy | Lower | Higher |
| Uses POS Tagging | No | Often Yes |
| Example | Studies → Studi | Studies → Study |
| Best Use | Search Engines, IR | Chatbots, QA, Translation |

---

# `Tokenization vs Stemming vs Lemmatization`

| Technique | Purpose | Example |
|------------|---------|----------|
| Tokenization | Split text into tokens | "I love NLP" → ["I","love","NLP"] |
| Stemming | Reduce to stem | Playing → Play |
| Lemmatization | Reduce to dictionary form | Better → Good |

# Best Practices

- Always tokenize before further preprocessing.
- Use stemming for simple retrieval tasks where speed is important.
- Prefer lemmatization when preserving meaning and grammar is essential.
- Combine POS tagging with lemmatization for higher accuracy.
- Select preprocessing techniques based on the application rather than applying them blindly.

# Key Takeaways

- Text Preprocessing transforms cleaned text into a machine-friendly format.

# `2. Interview Questions & Answers`

## Beginner Questions

**1. What is Text Preprocessing in NLP?**
**Answer:** Text Preprocessing prepares cleaned text for machine learning by applying techniques such as tokenization, stemming, and lemmatization.

**2. What is Tokenization?**
**Answer:** Tokenization splits text into smaller units called tokens (words, sentences, characters, or subwords).

**3. What is Stemming?**
**Answer:** Stemming removes prefixes or suffixes to reduce words to a common stem, which may not be a valid dictionary word.

**4. What is Lemmatization?**
**Answer:** Lemmatization converts words into their correct dictionary base form using grammar rules and lexical resources.

**5. Why is Tokenization essential?**
**Answer:** Most NLP algorithms process tokens rather than raw paragraphs, making tokenization a foundational preprocessing step.

---

## Intermediate Questions

**1. Explain the difference between Stemming and Lemmatization.**
**Answer:** Stemming is faster and rule-based but may produce invalid words, whereas lemmatization uses grammar and dictionaries to generate valid base forms.

**2. Why is POS Tagging useful before Lemmatization?**
**Answer:** The correct lemma depends on the grammatical role of a word. POS tags help identify the correct base form.

**3. Why do Transformer models use Subword Tokenization?**
**Answer:** Subword tokenization efficiently handles rare words, unknown words, and large vocabularies by splitting words into meaningful subword units.

---

## `Scenario-Based Questions`

**Q1. You're building a search engine where speed is more important than perfect grammar. Which preprocessing technique would you choose?**
**Answer:** Stemming, because it is computationally efficient and suitable for information retrieval.

**Q2. You're developing a medical chatbot that must understand user intent accurately. Would you use stemming or lemmatization? Why?**
**Answer:** Lemmatization, because it preserves the correct dictionary forms and semantic meaning, leading to more accurate language understanding.

**Q3. Why is Subword Tokenization important for models like GPT or BERT?**
**Answer:** It allows the model to represent rare or unseen words by decomposing them into smaller meaningful units, reducing vocabulary size and improving generalization.
