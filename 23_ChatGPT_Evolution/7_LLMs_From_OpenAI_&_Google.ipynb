# Large Language Models (LLMs)

## Definition

A **Large Language Model (LLM)** is a **Transformer-based deep learning model** trained on **massive amounts of text data** to understand, generate, summarize, translate, and reason about human language.

LLMs learn language by performing a **language modeling task** (such as next-token prediction or masked-token prediction) during pretraining and then transfer this knowledge to many downstream tasks.

> **Definition:**
> A **Large Language Model (LLM)** is a Transformer-based neural network trained on large-scale text corpora using self-supervised learning to understand and generate human language.

---

```text
Rule-Based Systems
        ↓
Statistical NLP
        ↓
RNN / LSTM
        ↓
Encoder-Decoder
        ↓
Attention Mechanism
        ↓
Transformer (2017)
        ↓
Transfer Learning
        ↓
Large Language Models (LLMs)
        ↓
ChatGPT
```

LLMs represent **Stage 5** in the evolution of modern NLP.

# Architecture of an LLM

Most modern LLMs (such as GPT, LLaMA, Gemma, and Mistral) are **decoder-only Transformers**.

```text
User Prompt

↓

Tokenizer

↓

Token IDs

↓

Embedding Layer

↓

Positional Encoding / Position Embeddings

↓

Transformer Blocks

↓

Linear Layer

↓

Softmax

↓

Next Token

↓

Repeat
```

---

# Components of an LLM

## 1. Tokenizer

Converts text into tokens.

Example

```text
I love AI
```

↓

```text
["I","love","AI"]
```

↓

```text
[15,89,700]
```

---

## 2. Embedding Layer

Converts token IDs into dense vectors.

```text
15

↓

[0.23,-0.64,0.81,...]
```

Words with similar meanings tend to have similar embeddings.

---

## 3. Positional Information

Transformers process tokens in parallel.

Therefore, positional information is added.

```text
Token Embedding

+

Position Embedding

↓

Final Input Embedding
```

---

## 4. Transformer Blocks

Each Transformer block contains:

```text
Multi-Head Attention

↓

Residual Connection

↓

Layer Normalization

↓

Feed Forward Network

↓

Residual Connection

↓

Layer Normalization
```

This block is repeated many times.

Examples:

* GPT-2 Small → 12 layers
* GPT-3 → 96 layers
* Modern LLMs → dozens to hundreds of layers

---

## 5. Linear Layer

Projects the hidden representation to the vocabulary size.

Example:

Vocabulary:

```text
50000 Words
```

Output:

```text
50000 Logits
```

---

## 6. Softmax

Converts logits into probabilities.

Example:

| Word     | Probability |
| -------- | ----------: |
| learning |        0.72 |
| amazing  |        0.10 |
| future   |        0.08 |
| world    |        0.05 |
| Others   |        0.05 |

---

## 7. Token Generation

The highest-probability (or sampled) token is selected.

Example:

Prompt:

```text
Artificial Intelligence is
```

Prediction:

```text
transforming
```

The process repeats until the response is complete.

---

# How Does an LLM Work?

## Step 1

User enters a prompt.

```text
What is Machine Learning?
```

---

## Step 2

Tokenizer converts text into token IDs.

```text
"What"

↓

1023
```

---

## Step 3

Embedding layer converts IDs into vectors.

---

## Step 4

Transformer processes the sequence using self-attention.

Every token attends to relevant tokens.

---

## Step 5

The final layer predicts the probability of the next token.

---

## Step 6

The predicted token is appended to the input.

---

## Step 7

The process repeats until an end-of-sequence token or stopping condition.

---

# Training of an LLM

LLMs are trained using **self-supervised learning**.

Example:

Input:

```text
The capital of France is
```

Target:

```text
Paris
```

The model predicts:

```text
London
```

Loss is computed.

Backpropagation updates the parameters.

This process repeats billions of times.

---

# Inference

During inference:

```text
User Prompt

↓

Tokenizer

↓

Transformer

↓

Prediction

↓

Next Token

↓

Repeat
```

No parameter updates occur.

# Interview Summary

> A **Large Language Model (LLM)** is a Transformer-based neural network trained on massive text datasets using self-supervised learning. It learns language by predicting tokens, enabling it to understand context, generate text, answer questions, summarize documents, translate languages, and perform many other NLP tasks. An LLM consists of a tokenizer, embedding layer, positional information, multiple Transformer blocks (containing multi-head attention and feed-forward networks), and an output layer that predicts the next token. During training, the model learns from billions of examples using backpropagation. During inference, it generates responses autoregressively by predicting one token at a time. Modern LLMs such as GPT, LLaMA, Gemma, Mistral, and Qwen are built on this foundation and are typically adapted to specific applications through fine-tuning or parameter-efficient techniques like LoRA.
