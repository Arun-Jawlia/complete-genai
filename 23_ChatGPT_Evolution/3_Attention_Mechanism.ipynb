# `Detailed Notes 1`

# Attention Mechanism in Sequence-to-Sequence Models

## Definition

The **Attention Mechanism** is a technique introduced in **2015** to solve the major limitation of the original **Encoder-Decoder architecture**.

Instead of relying on a **single fixed-size context vector**, the decoder can dynamically **focus (attend)** to the most relevant parts of the input sentence while generating each output word.

> **Definition:**
> The **Attention Mechanism** allows the decoder to access all encoder hidden states and assign different importance (attention weights) to them while generating each output token.

---

# Why Was Attention Introduced?

## Problem with Encoder-Decoder

Recall the Encoder-Decoder architecture:

```text
Input Sentence

↓

Encoder

↓

One Context Vector

↓

Decoder

↓

Output Sentence
```

The encoder compresses the **entire input sentence** into **one fixed-size context vector**.

This works well for short sentences.

However, for longer sentences:

* Initial information starts getting lost.
* Important words are forgotten.
* Translation quality decreases.

---

## Example

Sentence:

```text
Sadly mistaken, he realized that the job offer was actually an incredible opportunity that would lead to significant personal and professional growth.
```

The encoder must compress **30+ words** into a single vector.

As the sentence becomes longer:

* Early words like **"Sadly mistaken"** may not be preserved well.
* The decoder receives an incomplete summary.
* Translation quality suffers.

This limitation is called the **Context Vector Bottleneck**.

---

# Solution: Attention Mechanism (2015)

To overcome this bottleneck, researchers introduced the **Attention Mechanism**.

Instead of sending only:

```text
One Context Vector
```

the encoder now sends:

```text
Hidden State 1
Hidden State 2
Hidden State 3
...
Hidden State n
```

The decoder can access **every encoder hidden state**, not just the final one.

---

# Encoder with Attention

Suppose the input sentence is:

```text
Transformers are great !
```

The encoder processes the sentence sequentially using RNN/LSTM/GRU.

| Input Word   | Encoder Hidden State |
| ------------ | -------------------- |
| Transformers | h₁                   |
| are          | h₂                   |
| great        | h₃                   |
| !            | h₄                   |

Unlike the original Encoder-Decoder model, **all hidden states are preserved**, not only the last one.

---

# What Does Each Hidden State Represent?

Each encoder hidden state summarizes the sentence **up to that point**.

| Hidden State | Contains Context About     |
| ------------ | -------------------------- |
| h₁           | Transformers               |
| h₂           | Transformers + are         |
| h₃           | Transformers + are + great |
| h₄           | Entire sentence            |

Notice that:

* Every hidden state contains progressively richer contextual information.
* The decoder can choose which hidden state(s) are most useful.

---

# Decoder with Attention

When generating each output word, the decoder performs two tasks:

1. Generate the next word.
2. Decide **which encoder hidden states are most relevant**.

Instead of relying on a single summary, it dynamically looks back at the encoder outputs.

```text
Encoder Hidden States

h₁
h₂
h₃
h₄

↓

Attention Layer

↓

Relevant Context

↓

Decoder

↓

Next Word
```

---

# Example

Suppose we are translating:

```text
Transformers are great !
```

into German.

While generating the German word for **"Transformers"**, the decoder should pay maximum attention to:

```text
h₁
```

While generating the German word for **"great"**, it should focus more on:

```text
h₃
```

Thus, the decoder dynamically chooses the most relevant encoder information at every step.

---

# How Attention Works

Suppose the decoder is generating the current word.

Instead of using only:

```text
Final Context Vector
```

it computes attention over:

```text
h₁
h₂
h₃
h₄
```

Each hidden state receives an **attention score**.

Example:

| Hidden State | Attention Weight |
| ------------ | ---------------: |
| h₁           |             0.10 |
| h₂           |             0.20 |
| h₃           |             0.60 |
| h₄           |             0.10 |

Since **h₃** has the highest attention weight, the decoder focuses mostly on that hidden state.

# Attention Layer

The Attention Layer performs three main tasks:

### Step 1

Compare the decoder's current hidden state with **every encoder hidden state**.

↓

### Step 2

Calculate a **similarity score**.

↓

### Step 3

Convert the scores into attention weights.

↓

### Step 4

Create a weighted context vector.

↓

### Step 5

Generate the next output word.

---

# Similarity Score

The decoder asks:

> **Which encoder hidden state is most relevant for generating the current output word?**

This is measured using a **similarity function**.

Common methods include:

* Dot Product
* General Score
* Additive (Bahdanau) Attention
* Cosine Similarity (conceptually useful for understanding similarity, though not the most common in classic attention implementations)

The higher the similarity score:

* The higher the attention weight.
* The more influence that hidden state has on the prediction.

---

# Why Is Attention Better?

## Encoder-Decoder

```text
Entire Sentence

↓

One Context Vector

↓

Decoder
```

Problem:

* Information bottleneck.
* Long sentences lose information.

---

## Attention Mechanism

```text
Entire Sentence

↓

All Hidden States

↓

Attention

↓

Decoder
```

Advantages:

* Decoder can access every encoder hidden state.
* Long sentences are translated much more accurately.
* Initial words are less likely to be forgotten.

---

# Limitation of Attention Mechanism

Although Attention significantly improved translation quality, it still had an important limitation.

The encoder still uses:

* RNN
* LSTM
* GRU

These models process words:

```text
Word1

↓

Word2

↓

Word3

↓

Word4
```

one by one.

Therefore, the encoder remains **sequential**, making training slower and limiting parallelism.

---

# Computational Complexity

Suppose:

* Input sentence has **n** words.
* Output sentence has **m** words.

For every output word, the decoder compares it with **every encoder hidden state**.

Total comparisons:

```text
n × m
```

For example:

Input:

100 words

Output:

100 words

↓

Attention calculations:

```text
100 × 100 = 10,000
```

Thus, attention improves translation quality but increases computation.

---

# Why Did We Move Beyond Attention?

Attention solved the context bottleneck.

However, two major challenges remained:

### 1. Sequential Encoder

RNN/LSTM/GRU still process tokens one at a time.

Training is slow.

---

### 2. High Attention Computation

For each output token, the decoder compares against all encoder hidden states, increasing computational cost as sequence lengths grow.

Researchers wanted:

* Parallel processing.
* Faster training.
* Better scalability.
* Better handling of long sequences.

This led to the **Transformer Architecture (2017)**.

---

# Attention Mechanism

The **Attention Mechanism** is a technique that allows the decoder to **focus on the most relevant parts of the input sequence while generating each output word**.

Instead of relying on a **single fixed-size context vector**, the decoder dynamically looks at all encoder hidden states and decides which words are most important at each decoding step.

The attention mechanism was introduced to solve the **information bottleneck** problem in the original Seq2Seq encoder-decoder architecture.

---

# Why Was Attention Introduced?

## Problem with the Original Encoder-Decoder Model

In the original Seq2Seq model:

```text
Input Sentence

↓

Encoder

↓

One Context Vector

↓

Decoder

↓

Output Sentence
```

The encoder compresses the **entire input sentence** into **one fixed-length vector**.

That single vector must contain all the information needed by the decoder.

---

# Example

Input:

```text
I am studying Artificial Intelligence at Stanford University while working on computer vision and robotics.
```

Encoder:

```text
↓

One Vector
```

Decoder:

```text
↓

French Translation
```

The problem is:

> Can one fixed-size vector accurately represent such a long sentence?

Usually **no**.

As sentence length increases, important information is lost.

This is called the **information bottleneck**.

---

# Information Bottleneck

```text
Long Sentence

↓

Encoder

↓

One Small Vector

↓

Decoder
```

Imagine compressing an entire 500-page book into **one paragraph**.

You would inevitably lose important details.

The same problem occurs in the original Seq2Seq model.

---

# Performance on Long Sentences

For short sentences:

```text
Input

↓

Encoder

↓

Context Vector

↓

Decoder

↓

Good Translation
```

For long sentences:

```text
Long Input

↓

Encoder

↓

Context Vector

↓

Decoder

↓

Missing Information

↓

Poor Translation
```

Translation quality degrades as sentence length increases.

---

# Solution: Attention Mechanism

Instead of passing only one context vector,

the encoder sends **all hidden states** to the decoder.

```text
Input

↓

Encoder

↓

h₁

h₂

h₃

h₄

h₅

↓

Attention

↓

Decoder
```

Now the decoder can access information from **every input word**.

---

Imagine reading a book and answering questions.

Question:

```text
Who won the match?
```

You don't memorize the whole book.

Instead, you **look back at the relevant page**.

The Attention Mechanism works in the same way.

The decoder looks back at the encoder outputs whenever it needs information.

Suppose the decoder is generating:

```text
Artificial
```

Instead of using only one context vector,

```text
h₃

h₄
```

because they correspond to **Artificial Intelligence**.

# Architecture

```text
             Encoder

x₁ → h₁

x₂ → h₂

x₃ → h₃

x₄ → h₄

      │

      ▼

Attention Layer

      │

      ▼

Decoder

↓

Output
```

# Step-by-Step Working

Suppose the sentence is:

```text
I love AI
```

Encoder computes

```text
h₁

↓

I

h₂

↓

love

h₃

↓

AI
```

Instead of using

```text
Context Vector
```

the decoder receives

```text
h₁

h₂

h₃
```

---

# Decoder Wants to Predict First Word

Input

```text
<BOS>
```

Decoder hidden state

```text
s₁
```

Now the decoder asks:

```text
Which encoder hidden state is most useful?
```

---

# Step 1: Calculate Attention Scores

The decoder compares its current hidden state with every encoder hidden state.

```text
Score(s₁,h₁)

↓

0.1
```

```text
Score(s₁,h₂)

↓

0.3
```

```text
Score(s₁,h₃)

↓

0.9
```

Higher score means more relevant.

---

# Step 2: Softmax

Convert scores into probabilities.

| Hidden State | Score | Attention Weight |
| ------------ | ----: | ---------------: |
| h₁           |   0.1 |             0.18 |
| h₂           |   0.3 |             0.24 |
| h₃           |   0.9 |             0.58 |

These weights sum to **1**.

---

# Step 3: Weighted Sum

The attention weights are used to compute a **context vector for the current decoding step**.

[
$(\text{Context}_t)$=
$(\sum_i
\alpha_i h_i)$
]

where:

* $(h_i)$ = encoder hidden state
* $(\alpha_i)$ = attention weight

Example:

```text
0.18 × h₁

+

0.24 × h₂

+

0.58 × h₃

↓

New Context Vector
```

Unlike the original Seq2Seq model, this context vector is **different for every decoder time step**.

---

# Step 4: Generate Output

The decoder combines:

* Current decoder hidden state
* Current attention context vector

to predict the next word.

```text
Context Vector

+

Decoder Hidden State

↓

Softmax

↓

Predicted Word
```

---

# Repeat

For the next output word:

```text
Decoder State

↓

Attention

↓

New Context Vector

↓

Prediction
```

The attention weights change at every decoding step.

---

# Complete Workflow

```text
Input Sentence

↓

Encoder

↓

h₁

h₂

h₃

h₄

↓

Attention

↓

Context Vector

↓

Decoder

↓

Output Word

↓

Next Decoder Step

↓

Attention Again

↓

Next Word
```

# Mathematical Representation

## Attention Score

One common scoring function is the dot product:

[
$(score(s_t,h_i)=s_t^T h_i)$
]

where:

* (s_t) = decoder hidden state
* (h_i) = encoder hidden state

Other scoring functions (e.g., additive/Bahdanau attention) also exist.

---

## Attention Weights

[
$(\alpha_i)$=
$(\frac{e^{score(s_t,h_i)}}
{\sum_j e^{score(s_t,h_j)}})$
]

This is simply the **Softmax** of the scores.

---

## Context Vector

[
$(c_t=
\sum_i
\alpha_i h_i)$
]

---

## Final Prediction

The decoder uses both:

* Decoder hidden state (s_t)
* Context vector (c_t)

to predict the next output token.

# Why Did We Shift from Encoder-Decoder to Attention?

The shift happened because the original encoder-decoder architecture had a **fundamental limitation**:

### Original Seq2Seq

```text
Input

↓

Encoder

↓

One Context Vector

↓

Decoder
```

Problems:

* Information bottleneck.
* Poor performance on long sentences.
* Loss of important details.
* Decoder cannot revisit the input.

### Seq2Seq + Attention

```text
Input

↓

Encoder

↓

All Hidden States

↓

Attention

↓

Decoder
```

Benefits:

* Dynamic context for every output word.
* Better handling of long sequences.
* Improved translation accuracy.
* Decoder can focus on the most relevant input words.

---

# Why Did We Later Shift to Transformers?

Although attention significantly improved Seq2Seq models, the encoder and decoder were still built on **RNNs/LSTMs/GRUs**.

These models still had limitations:

* Sequential computation (slow training).
* Difficult to parallelize.
* Long sequences remained computationally challenging.

The **Transformer** removed recurrence entirely and relied on **self-attention** for both the encoder and decoder, enabling:

* Parallel processing during training.
* Better long-range dependency modeling.
* Improved scalability to very large datasets and models.

# Interview Summary

> The **Attention Mechanism** was introduced to overcome the **information bottleneck** of the original Seq2Seq encoder-decoder architecture. In the original model, the encoder compressed the entire input sequence into a single fixed-length context vector, which caused information loss, especially for long sentences. Attention solves this by allowing the decoder to access **all encoder hidden states** and compute a **dynamic context vector** for each output token. At every decoding step, the decoder calculates attention scores, converts them into attention weights using Softmax, and forms a weighted sum of the encoder hidden states. This enables the decoder to focus on the most relevant parts of the input while generating each output word. Attention greatly improved tasks such as machine translation and directly inspired the self-attention mechanism used in Transformer architectures.
