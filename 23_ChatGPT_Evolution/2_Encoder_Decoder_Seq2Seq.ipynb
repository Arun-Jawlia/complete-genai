# Sequence-to-Sequence (Seq2Seq) Learning with Neural Networks

## What is Sequence-to-Sequence (Seq2Seq) Learning?

**Sequence-to-Sequence (Seq2Seq)** is a deep learning architecture that transforms **one sequence into another sequence**.

The input and output are both sequences, but they can have **different lengths**.

For example:

```text
English Sentence
        ↓
French Sentence
```

or

```text
Question
      ↓
Answer
```

or

```text
Long Article
      ↓
Summary
```

---

# Why Do We Need Seq2Seq Models?

Traditional neural networks assume:

```text
One Input
     ↓
One Output
```

Example:

```text
Image
   ↓
Cat
```

However, many NLP problems require:

```text
Input Sequence
        ↓
Output Sequence
```

Example:

```text
I love AI

↓

J'aime l'IA
```

A normal neural network cannot naturally handle variable-length input and output sequences.

Seq2Seq models were introduced to solve this problem.

---

# Applications of Seq2Seq

* Machine Translation
* Text Summarization
* Chatbots
* Question Answering
* Speech Recognition
* Grammar Correction
* Code Generation

---

# Basic Architecture

A Seq2Seq model consists of two parts:

1. **Encoder**
2. **Decoder**

```text
          Input Sentence

                │

                ▼

             Encoder

                │

      Context Vector (Thought Vector)

                │

                ▼

             Decoder

                │

                ▼

        Output Sentence
```

---

# Real Example

Input:

```text
I love Deep Learning
```

Output:

```text
J'aime le Deep Learning
```

The encoder understands the English sentence.

The decoder generates the French sentence.

---

# Step 1: Encoder

The encoder reads the **entire input sequence**.

Example:

```text
I → love → Deep → Learning
```

Suppose we use an LSTM.

The encoder processes one word at a time.

```text
x₁ → Encoder → h₁

x₂ → Encoder → h₂

x₃ → Encoder → h₃

x₄ → Encoder → h₄
```

where:

* (x_t) = input token
* (h_t) = hidden state

The **final hidden state** summarizes the entire sentence.

This is called the:

* Context Vector
* Thought Vector

---

# Encoder Diagram

```text
I ------►

          LSTM

love --->

          LSTM

Deep --->

          LSTM

Learning

          LSTM

              │

              ▼

       Context Vector
```

---

# What is the Context Vector?

The **Context Vector** is a fixed-length vector representing the meaning of the entire input sequence.

Example:

```text
Input Sentence

"I love AI"

↓

Encoder

↓

[0.23, -0.76, 0.91, ...]
```

This vector is passed to the decoder.

---

# Limitation of Context Vector

Suppose the sentence is:

```text
I studied Artificial Intelligence at Stanford University for five years while working on computer vision and robotics.
```

Compressing such a long sentence into **one fixed-size vector** can cause information loss.

This limitation motivated the development of the **Attention Mechanism**.

---

# Step 2: Decoder

The decoder receives:

* Context Vector
* Start token (`<SOS>` or `<BOS>`)

It generates one word at a time.

---

Example:

```text
Context Vector

↓

<BOS>

↓

J'

↓

aime

↓

le

↓

Deep

↓

Learning

↓

<EOS>
```

The decoder continues until it predicts an end-of-sequence token.

# Mathematical Representation

## Encoder

For every time step

[
h_t=f(x_t,h_{t-1})
]

where:

* (x_t) = input word
* (h_t) = hidden state

Final hidden state

[
C=h_n
]

where:

* (C) = Context Vector

---

## Decoder

Initial hidden state

[
s_0=C
]

Each decoding step

[
s_t=f(y_{t-1},s_{t-1})
]

where:

* (y_{t-1}) = previous generated word
* (s_t) = decoder hidden state

Prediction

[
P(y_t)=Softmax(Ws_t+b)
]

# Training Process

During training, the decoder does **not** use its own predicted word as the next input.

Instead, it uses the **actual target word**.

This technique is called:

## Teacher Forcing

Example

Target sentence

```text
J'

↓

aime

↓

l'IA
```

Instead of

```text
Prediction

↓

Next Input
```

we use

```text
Actual Word

↓

Next Input
```

This speeds up training and improves convergence.

---

# Training Workflow

```text
English Sentence

↓

Encoder

↓

Context Vector

↓

<BOS>

↓

Decoder

↓

Actual Word

↓

Decoder

↓

Actual Word

↓

...

↓

Loss

↓

Backpropagation
```

---

# Inference

During inference, the correct target sentence is unavailable.

Therefore, the decoder feeds **its own prediction** back as the next input.

Example:

```text
<BOS>

↓

J'

↓

aime

↓

l'IA

↓

<EOS>
```

Each generated word becomes the input for the next step.

---

# Training vs Inference

| Training               | Inference                 |
| ---------------------- | ------------------------- |
| Uses Teacher Forcing   | Uses previous prediction  |
| Ground-truth available | Ground-truth unavailable  |
| Faster learning        | Autoregressive generation |
| Parameters updated     | Parameters fixed          |

# Limitation of Basic Seq2Seq

The encoder compresses the **entire sentence into a single context vector**.

For long sentences, this fixed-size representation becomes a bottleneck.

Example:

```text
Very Long Sentence

↓

Encoder

↓

One Vector

↓

Decoder
```

Important information may be lost.

---

# Solution: Attention Mechanism

Instead of relying on one context vector, the decoder can **attend to all encoder hidden states**.

```text
Encoder Hidden States

h₁

h₂

h₃

h₄

h₅

        │

        ▼

Attention

        │

        ▼

Decoder
```

Now the decoder dynamically focuses on the most relevant parts of the input while generating each output token.

This significantly improved machine translation quality and eventually inspired the Transformer architecture.

---

# RNN Seq2Seq vs Attention Seq2Seq vs Transformer

| Feature           | Basic Seq2Seq | Seq2Seq + Attention | Transformer    |
| ----------------- | ------------- | ------------------- | -------------- |
| Encoder           | RNN/LSTM/GRU  | RNN/LSTM/GRU        | Self-Attention |
| Decoder           | RNN/LSTM/GRU  | RNN/LSTM/GRU        | Self-Attention |
| Context           | Single Vector | Dynamic Attention   | Full Attention |
| Long Sentences    | Poor          | Good                | Excellent      |
| Parallel Training | No            | No                  | Yes            |
| Performance       | Lowest        | Better              | Best           |

# Interview Summary

> **Sequence-to-Sequence (Seq2Seq)** is an encoder-decoder architecture designed to map one variable-length sequence to another. The **encoder** reads the complete input sequence and compresses it into a context vector (or thought vector), which summarizes the input. The **decoder** uses this context to generate the output sequence one token at a time. During **training**, the decoder commonly uses **teacher forcing**, where the correct previous target token is fed as input. During **inference**, the decoder feeds back its own predicted token to generate the next one. The major limitation of the original Seq2Seq model is that it compresses all input information into a single fixed-length context vector, making it difficult to handle long sequences. This limitation was addressed by the **Attention Mechanism**, which allows the decoder to dynamically focus on different encoder hidden states during generation and ultimately led to the development of the Transformer architecture.

# Encoder-Decoder Architecture for Sequence-to-Sequence (Seq2Seq) Models

## Definition

The **Encoder-Decoder** architecture is the first successful **Sequence-to-Sequence (Seq2Seq)** model introduced for tasks where both the input and output are sequences, such as **machine translation**.

It was introduced in the 2014 paper:

> **"Sequence to Sequence Learning with Neural Networks"**

The architecture consists of two main components:

* **Encoder** → Understands the input sequence.
* **Decoder** → Generates the output sequence.

---

# Why Do We Need an Encoder-Decoder?

Many NLP tasks require converting one sequence into another.

Examples:

| Input    | Output           |
| -------- | ---------------- |
| English  | German           |
| Question | Answer           |
| Article  | Summary          |
| Sentence | Translation      |
| Prompt   | ChatGPT Response |

The Encoder-Decoder architecture was designed to solve these Sequence-to-Sequence problems.

---

# Overall Architecture

```text
English Sentence

↓

Encoder (RNN/LSTM/GRU)

↓

Context Vector
(Sentence Embedding)

↓

Decoder (RNN/LSTM/GRU)

↓

German Sentence
```

---

# Step 1: Encoder

## Purpose

The encoder reads the input sentence **one word at a time** and gradually builds an understanding of the entire sentence.

Since RNNs process sequential data, each word is processed according to its **time step**.

---

## Example

Input Sentence

```text
I love Deep Learning
```

Processing inside the encoder:

```text
Time Step 1

I

↓

Hidden State h₁
```

↓

```text
Time Step 2

love

↓

Hidden State h₂
```

↓

```text
Time Step 3

Deep

↓

Hidden State h₃
```

↓

```text
Time Step 4

Learning

↓

Hidden State h₄
```

Each hidden state carries information from the previous time step.

---

# Hidden State (Context State)

The hidden state is also called the **Context State** or **Context Vector**.

It acts as a compressed summary of everything the encoder has understood so far.

At the final time step:

```text
h₄
```

contains information about the **entire input sentence**.

This final hidden state is passed to the decoder.

---

# Context Vector

The encoder produces a single vector called the **Context Vector**.

It contains a compressed representation of the whole sentence.

```text
Sentence

↓

Encoder

↓

Context Vector

↓

Decoder
```

> **Definition:**
> The **Context Vector** is the final hidden state of the encoder that summarizes the meaning of the entire input sequence.

---

# Step 2: Decoder

## Purpose

The decoder receives the context vector and starts generating the output sequence **one word at a time**.

Example:

Input:

```text
I love AI
```

↓

Context Vector

↓

Output:

```text
Ich liebe KI
```

Generation process:

```text
Ich

↓

liebe

↓

KI

↓

<EOS>
```

The decoder predicts each next word using:

* The context vector.
* The previously generated words.

# The Biggest Problem: Fixed-Length Context Vector

Although the Encoder-Decoder architecture worked well for short sentences, it struggled with **long sentences**.

Why?

Because the encoder compresses the **entire sentence into one fixed-size vector**.

No matter whether the sentence has:

* 5 words
* 20 words
* 50 words
* 100 words

the decoder still receives **only one context vector**.

---

# Example of a Long Sentence

```text
Sadly mistaken, he realized that the job offer was actually an incredible opportunity that would lead to significant personal and professional growth.
```

This sentence contains **more than 30 words**.

The encoder must compress all this information into a single vector.

As the sentence length increases:

* More information must fit into the same vector.
* Earlier information becomes harder to preserve.

---

# Why Does Information Get Lost?

Remember:

RNNs process words sequentially.

```text
Word1

↓

Word2

↓

Word3

↓

...

↓

Word32
```

Each new hidden state is influenced more by recent words.

By the time the encoder reaches the last word, information from the beginning of the sentence may become weak.

For example:

```text
Sadly mistaken,
```

is very important because it sets the tone of the sentence.

If the encoder forgets this phrase, the decoder may generate a translation with the wrong sentiment or meaning.

---

# Example

Sentence:

```text
Sadly mistaken, he realized that the job offer was actually an incredible opportunity...
```

If the model forgets:

```text
Sadly mistaken
```

the translated sentence may incorrectly suggest that the realization was entirely positive from the start, changing the intended meaning.

This is one reason translation quality decreases for long sentences in the original Encoder-Decoder architecture.

# Translation Quality

Machine Translation quality is commonly evaluated using the **BLEU (Bilingual Evaluation Understudy) Score**.

### What is BLEU?

BLEU compares the machine-generated translation with one or more high-quality human reference translations.

* Higher BLEU Score → Better translation quality.
* Lower BLEU Score → Poor translation quality.

As sentence length increases, the original Encoder-Decoder model often loses important context, causing the BLEU score to decrease.

> **Note:** BLEU is one evaluation metric. Modern research also uses metrics such as METEOR, ROUGE, and COMET for different tasks.

---

# Why Did This Problem Occur?

The encoder sends only **one context vector** to the decoder.

```text
Entire Sentence

↓

One Context Vector

↓

Decoder
```

The decoder has **no direct access** to the encoder's intermediate hidden states.

It must rely entirely on the compressed summary.

This creates a bottleneck, especially for long and complex sentences.

---

# Solution to This Problem

Researchers introduced the **Attention Mechanism**.

Instead of sending only one context vector:

```text
Encoder

↓

One Vector
```

Attention allows the decoder to access **all encoder hidden states**.

```text
Encoder

↓

All Hidden States

↓

Attention

↓

Decoder
```

Now, while generating each output word, the decoder can focus on the most relevant parts of the input sentence.

This significantly improves translation quality for long sequences.

---

# Encoder-Decoder vs Attention

| Encoder-Decoder                    | Attention-Based Seq2Seq              |
| ---------------------------------- | ------------------------------------ |
| One fixed context vector           | Access to all encoder hidden states  |
| Information bottleneck             | Dynamic focus on relevant words      |
| Weak performance on long sentences | Strong performance on long sentences |
| Translation quality degrades       | Translation quality improves         |

---

# Real-World Analogy

Imagine a student reads a **40-page chapter** and is allowed to write **only one paragraph** as a summary.

Another student can keep the entire book open while answering questions.

Who will answer better?

The second student.

Similarly:

* **Encoder-Decoder:** One summary (context vector).
* **Attention:** Access to the full sequence whenever needed.
