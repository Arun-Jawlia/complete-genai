# `Detailed Notes 1`

# Sequence-to-Sequence (Seq2Seq) Models: Evolution Towards LLMs

## Definition

**Sequence-to-Sequence (Seq2Seq)** models are deep learning architectures that take a **sequence as input** and generate another **sequence as output**.

A sequence can be:

* Text
* Audio
* Time-series data
* DNA sequence
* Video frames

> **Definition:**
> A **Sequence-to-Sequence (Seq2Seq)** model learns to transform one sequence into another, such as translating English text into German or generating a chatbot response.

---

# Evolution of Deep Learning Models

Deep Learning architectures have evolved based on the type of data they are designed to handle.

```text
Deep Learning
      │
      ├── ANN → Structured Data
      │        (Classification & Regression)
      │
      ├── CNN → Image Data
      │
      ├── RNN/LSTM/GRU → Sequential Data
      │
      └── Transformer → Modern Sequence Modeling
```

---

# 1. Artificial Neural Network (ANN)

## Purpose

ANNs are primarily used for **structured/tabular data**.

### Applications

* House Price Prediction
* Customer Churn Prediction
* Loan Approval
* Diabetes Prediction

### Tasks

* Classification
* Regression

---

# 2. Convolutional Neural Network (CNN)

## Purpose

CNNs are specialized for **image processing**.

### Applications

* Face Recognition
* Object Detection
* Medical Imaging
* OCR

---

# 3. Recurrent Neural Network (RNN)

## Purpose

RNNs are designed for **Sequential Data**.

A sequence means the order of data matters.

### Examples

* Sentences
* Speech
* Time Series
* Stock Prices

---

# Types of RNN

## 1. One-to-One

One input produces one output.

```text
Image
   ↓
Classification
```

### Example

* Image Classification

---

## 2. One-to-Many

One input generates multiple outputs.

```text
Image

↓

"A dog is playing in the park."
```

### Example

* Image Captioning

---

## 3. Many-to-One

Multiple inputs generate one output.

```text
"I love this movie"

↓

Positive
```

### Example

* Sentiment Analysis

---

## 4. Many-to-Many (Synchronous)

Input and output sequences have the **same length**.

```text
I

love

AI

↓

PRON

VERB

NOUN
```

Input Length = Output Length

### Examples

* Part-of-Speech (POS) Tagging
* Named Entity Recognition (NER) *(token-level labeling tasks typically preserve sequence length)*

---

## 5. Many-to-Many (Asynchronous)

Input and output lengths are **different**.

```text
English Sentence

↓

German Sentence
```

Input Length ≠ Output Length

### Examples

* Machine Translation
* Chatbots
* Text Summarization
* Next Word Prediction

---

# Sequence-to-Sequence (Seq2Seq) Models

## Definition

A Seq2Seq model is a Many-to-Many architecture that converts one sequence into another.

```text
Input Sequence

↓

Encoder

↓

Context Vector

↓

Decoder

↓

Output Sequence
```

---

# Synchronous vs Asynchronous Seq2Seq

## Synchronous

Input length equals output length.

```text
Input

I Love AI

↓

Output

PRON VERB NOUN
```

Input Length = Output Length

### Examples

* POS Tagging
* Token Classification

---

## Asynchronous

Input length differs from output length.

```text
English

↓

German
```

Input Length ≠ Output Length

### Examples

* Machine Translation
* ChatGPT
* Text Generation
* Question Answering
* Summarization

---

# Other Important Deep Learning Architectures

## 1. GAN (Generative Adversarial Network)

### Definition

GANs generate **new synthetic data** that resembles real data.

### Applications

* AI-generated faces
* Image generation
* Deepfakes
* Image enhancement

---

## 2. Autoencoder

### Definition

An Autoencoder is a neural network that learns to compress and reconstruct data.

```text
Input

↓

Encoder

↓

Compressed Representation

↓

Decoder

↓

Reconstructed Output
```

### Applications

* Data compression
* Image denoising
* Anomaly detection
* Feature extraction

> **Example:** Compressing a large file or learning a compact representation of an image before reconstructing it.

# Stage 1: Encoder-Decoder (2014)

## Research Paper

> **Sequence to Sequence Learning with Neural Networks**

Published by Google researchers in **2014**.

---

## Objective

The goal was to convert one language into another.

Example:

```text
English

↓

Encoder

↓

Context Vector

↓

Decoder

↓

German
```

---

## Working

### Encoder

Reads the input sequence one token at a time using:

* RNN
* LSTM
* GRU

It converts the entire input into a **fixed-size context vector**.

---

### Decoder

Receives the context vector.

Generates the translated sentence **one token at a time**.

---

## Example

Input

```text
I love AI
```

↓

Encoder

↓

Context Vector

↓

Decoder

↓

```text
Ich liebe KI
```

---

# Encoder-Decoder Workflow

```text
English Sentence

↓

RNN/LSTM/GRU Encoder

↓

Context Vector

↓

RNN/LSTM/GRU Decoder

↓

German Sentence
```

---

# Problem with Encoder-Decoder

The encoder compresses the entire input sentence into a **single fixed-length context vector**.

For short sentences:

```text
I love AI.
```

This works well.

For long sentences:

```text
I started learning Artificial Intelligence because I wanted to build intelligent applications...
```

Important information can be lost because everything must fit into one vector.

This limitation motivated the development of the **Attention Mechanism**.

---

# Why Did We Move Beyond Encoder-Decoder?

| Encoder-Decoder Limitation     | Solution            |
| ------------------------------ | ------------------- |
| Fixed-size context vector      | Attention Mechanism |
| Difficulty with long sentences | Self-Attention      |
| Sequential processing          | Transformer         |
| Limited scalability            | Modern LLMs         |

The encoder understands the source sentence, and the decoder generates the translated sentence token by token.

# `Detailed Notes 2`

# Decoder Working in Transformers

## Definition

The **Decoder** is the component of a Transformer responsible for **generating the output sequence one token at a time**.

Its primary task is **Next Token Prediction**. Based on the input and the tokens generated so far, the decoder predicts the most probable next token until the complete output is generated.

> **Definition:**
> A **Transformer Decoder** generates text by repeatedly predicting the next token using the previously generated tokens and the contextual information learned during training.

---

# Purpose of the Decoder

The decoder's main responsibility is:

* Predict the next token.
* Generate complete sentences token by token.
* Continue generation until an **End-of-Sequence (EOS)** token is produced.

### Example

**Input**

```text
Hi, How are you?
```

**Possible Output**

```text
I am doing well, thank you!
```

The decoder does **not** generate the entire sentence at once.

Instead, it predicts:

```text
I
↓

am
↓

doing
↓

well
↓

,
↓

thank
↓

you
↓

!
↓

<EOS>
```

---

# Encoder vs Decoder

| Encoder                        | Decoder                                          |
| ------------------------------ | ------------------------------------------------ |
| Understands the input sequence | Generates the output sequence                    |
| Produces contextual embeddings | Predicts the next token                          |
| Learns input context           | Uses context + previous outputs to generate text |

> **Note:**
> In an **Encoder-Decoder Transformer** (like the original Transformer used for translation), the encoder provides contextual embeddings to the decoder.
> In **decoder-only models** such as GPT, there is **no encoder**. The decoder generates text using only the prompt and previously generated tokens.

---

# Decoder Workflow

```text
Input Text
      ↓
Tokenizer
      ↓
Embeddings
      ↓
Positional Encoding
      ↓
Masked Multi-Head Self-Attention
      ↓
Feed Forward Network
      ↓
Linear Layer
      ↓
Softmax
      ↓
Next Token Prediction
```

---

# Two Phases of the Decoder

The decoder operates differently during:

1. **Training**
2. **Inference (Generation)**

---

# Phase 1: Training

During training, the model already knows the **correct output sentence**.

This allows it to compare its predictions with the actual answer and improve itself.

---

## Example

Target sentence:

```text
I love AI
```

The decoder learns using shifted inputs.

| Decoder Input       | Target Output |
| ------------------- | ------------- |
| `<START>`           | I             |
| `<START> I`         | love          |
| `<START> I love`    | AI            |
| `<START> I love AI` | `<EOS>`       |

At every step, the decoder predicts the next token.

---

## Step 1: Forward Propagation

The decoder predicts the next token.

Example:

Input:

```text
<START> I
```

Prediction:

```text
like
```

Actual Answer:

```text
love
```

The prediction is incorrect.

---

## Step 2: Loss Calculation

The prediction is compared with the actual target.

Example:

Prediction:

```text
like
```

Actual:

```text
love
```

The model computes a **loss** (prediction error).

Common loss function:

* Cross-Entropy Loss

Higher error → Higher loss.

---

## Step 3: Backpropagation

The loss is propagated backward.

The model updates:

* Attention weights
* Feed Forward weights
* Embedding weights
* Linear layer weights

The goal is to make future predictions more accurate.

---

## Training Loop

```text
Input

↓

Prediction

↓

Calculate Loss

↓

Backpropagation

↓

Update Weights

↓

Repeat
```

This process repeats millions or billions of times during training.

---

# Why Is Training Easier?

During training:

* The correct output already exists.
* The model always knows the expected next token.
* Computing the error is straightforward.

This approach is commonly called **Teacher Forcing**, where the decoder is fed the correct previous token instead of its own prediction.

> **Interview Point:** Teacher forcing speeds up training and stabilizes learning.

---

# Phase 2: Inference (Text Generation)

During inference, the correct answer is **not available**.

The decoder must generate text on its own.

---

## Example

User Input:

```text
Hello, how are you?
```

---

### Step 1

Prompt:

```text
Hello, how are you?
```

↓

Decoder predicts:

```text
I
```

---

### Step 2

New Input:

```text
Hello, how are you?

I
```

↓

Decoder predicts:

```text
am
```

---

### Step 3

Input:

```text
Hello, how are you?

I am
```

↓

Prediction:

```text
doing
```

---

The process continues:

```text
doing

↓

well

↓

,

↓

thank

↓

you

↓

!

↓

<EOS>
```

The decoder stops when it predicts the **End-of-Sequence (EOS)** token or reaches the maximum generation length.

---

# Complete Inference Flow

```text
Prompt

↓

Predict Next Token

↓

Append Token

↓

Predict Again

↓

Append Again

↓

Repeat

↓

<EOS>

↓

Stop
```

---

# How Does the Decoder Choose the Next Token?

The decoder does not directly output a word.

Instead, it produces a **score (logit)** for every token in the vocabulary.

Suppose the vocabulary contains:

```text
I
You
We
They
He
```

The Linear layer generates logits.

Example:

| Token | Logit |
| ----- | ----: |
| I     |   5.4 |
| You   |   2.1 |
| We    |   1.3 |
| They  |   0.9 |
| He    |   1.8 |

These logits are converted into probabilities using the **Softmax** function.

---

# Linear Layer + Softmax

The final layers of the decoder are:

```text
Decoder Output

↓

Linear Layer

↓

Logits

↓

Softmax

↓

Probability Distribution

↓

Next Token
```

---

## Example

After Softmax:

| Token | Probability |
| ----- | ----------: |
| I     |    **0.73** |
| You   |        0.12 |
| We    |        0.08 |
| They  |        0.04 |
| He    |        0.03 |

The model usually selects the token with the highest probability.

Prediction:

```text
I
```

---

# Temperature in Text Generation

Sometimes we don't always want the highest-probability token.

We can control creativity using a parameter called **Temperature**.

### Low Temperature (e.g., 0.2)

* Less randomness.
* More predictable.
* More factual.
* Best for coding and question answering.

Example:

```text
Probability

I → 95%

You → 3%

We → 2%
```

The model almost always chooses **I**.

---

### High Temperature (e.g., 1.2)

* More randomness.
* More creativity.
* More diverse responses.
* Better for storytelling and brainstorming.

The model may occasionally choose lower-probability words to produce more varied outputs.

> **Interview Point:** Temperature changes the probability distribution before sampling, influencing how deterministic or creative the generated text is.

---

Imagine typing on your smartphone.

You type:

The keyboard suggests:

| Word     | Probability |

| Birthday |         85% |

Your keyboard predicts the most likely next word.

# Interview Tips

> **Remember:**
>
> * The decoder performs **next-token prediction**.
> * During **training**, the correct output is available, enabling loss computation and backpropagation.
> * During **inference**, the decoder generates one token at a time using its own previous outputs.
> * The **Linear layer** converts hidden states into logits.
> * **Softmax** converts logits into probabilities.
> * **Temperature** controls the randomness and creativity of generation.
> * In GPT-style models, the decoder uses **masked self-attention**, ensuring it cannot look at future tokens during generation.

# Key Takeaways

* The decoder generates text one token at a time.
* It has two operating phases: **Training** and **Inference**.
* During training, it learns by comparing predictions with the correct output and updating its parameters.
* During inference, it predicts the next token repeatedly until an **EOS** token is generated.
* The Linear layer produces logits, and Softmax converts them into probabilities.
* Temperature influences how deterministic or creative the output becomes.

# Interview Questions & Answers

## Beginner Questions

### 1. What is the primary role of the Transformer decoder?

**Answer:**
To generate the output sequence by predicting one token at a time based on the input context and previously generated tokens.

---

### 2. What are the two phases of decoder operation?

**Answer:**

1. Training
2. Inference (Text Generation)

---

### 3. What happens during training?

**Answer:**
The decoder predicts the next token, compares it with the correct target, computes the loss, and updates its parameters using backpropagation.

---

### 4. What happens during inference?

**Answer:**
The decoder generates one token at a time, appends it to the sequence, and continues until an EOS token or maximum sequence length is reached.

---

### 5. Why is Softmax used?

**Answer:**
Softmax converts the decoder's output logits into a probability distribution over the vocabulary, allowing the model to select or sample the next token.

---

## Intermediate Questions

### 1. Why is training easier than inference?

**Answer:**
During training, the correct next token is already known (teacher forcing), making it easy to compute the loss and update the model. During inference, the model must rely solely on its own predictions.

---

### 2. What is Teacher Forcing?

**Answer:**
Teacher forcing is a training technique where the decoder receives the correct previous token as input instead of its own predicted token, leading to faster and more stable learning.

---

### 3. What is the role of the Linear layer?

**Answer:**
It maps the decoder's hidden representation to a vector of logits, with one score for each token in the vocabulary.

---

### 4. How does temperature affect generation?

**Answer:**
Lower temperatures produce more deterministic and factual outputs, while higher temperatures increase randomness and creativity by flattening the probability distribution.

---

## Scenario-Based Questions

### 1. During inference, the decoder predicts the wrong first token. What happens next?

**Answer:**
Since the decoder uses its previously generated tokens as input during inference, an incorrect early prediction can influence subsequent predictions and potentially propagate errors throughout the generated sequence.

---

### 2. A chatbot should always provide precise and consistent answers. Would you use a high or low temperature?

**Answer:**
A **low temperature** is preferable because it favors the highest-probability tokens, resulting in more consistent, accurate, and deterministic responses.
