```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from dotenv import load_dotenv

load_dotenv()

model = ChatOpenAI(model='gpt-3.5-turbo')

speech = """
Good morning, everyone. Today, I would like to talk to you about the importance of reading. 
Reading is not just a way to pass time; it is a doorway to knowledge, imagination, and personal growth. 
Books expose us to different cultures, new ideas, and a wide range of vocabulary. 
Every page we turn helps us build stronger language skills and a deeper understanding of the world. 
Moreover, reading inspires creativity. It allows us to step into the shoes of others, experiencing lives vastly different from our own. 
This empathy fosters better communication and understanding in our daily interactions. 
In an era dominated by digital screens and short-form content, the deep focus required by reading is a vital mental exercise. 
It strengthens our attention span and critical thinking abilities, helping us navigate complex information with clarity.
Furthermore, reading is a powerful tool for stress reduction. 
Immersing ourselves in a good story can lower heart rates and ease muscle tension, providing a healthy escape from daily pressures. 
Whether it is fiction, non-fiction, or poetry, the act of reading enriches our minds and souls. 
It is a lifelong habit that continues to benefit us regardless of age. 
I encourage you all to pick up a book today and discover the transformative power of reading. Thank you. 

"""

chat_message = [
    SystemMessage(content='You are an expert with expertise in summarizing in given text'), 
    HumanMessage(content=f'Please provide the short and concise summary for the provided text :{speech}')
]

print(model.get_num_tokens(speech))
```

```python
response = model.invoke(chat_message)
```

```python
print(response)
```

```python
from langchain_classic.chains import LLMChain
from langchain_core.prompts import PromptTemplate
```

```python
generic_prompt = '''
    summarize the following text: {text} to be concise in this 
    language: {language}
'''

prompt = PromptTemplate(template=generic_prompt, input_variables=['text', 'language'])
```

```python
complete_prompt = prompt.format(text=speech, language='hindi')
```

```python
print(model.get_num_tokens(complete_prompt))
```

```python
chain = LLMChain(llm = model, prompt = prompt)
```

```python
response = chain.invoke({'text':speech, 'language':"hindi"})
```
