```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_classic.chains.summarize import load_summarize_chain
from langchain_classic.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
load_dotenv()
```

```python
model = ChatOpenAI(model='gpt-3.5-turbo')
```

```python
loader = PyPDFLoader("../../Datasets/Full Stack Python Journey.pdf")
```

```python
data = loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap = 200)
```

```python
docs = splitter.split_documents(data)
```

```python
initial_prompt = PromptTemplate(template='Write a concise summary fo the following text: \n {text}', input_variables=['text'])
```

```python
refine_prompt = PromptTemplate(template=''' You current summary is {existing_answer}
Now refine and improve it using the following new text: {text}
Provide an updated, improved summary that includes imporatnt details from the old and next text
''', input_variables=['existing_answer', 'text'])
```

```python
chain = load_summarize_chain(llm=model, chain_type="refine", question_prompt = initial_prompt, refine_prompt = refine_prompt ,verbose=True)
```

```python
response = chain.run(docs)
```
