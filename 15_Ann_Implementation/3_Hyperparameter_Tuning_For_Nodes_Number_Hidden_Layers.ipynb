# Hyperparameter Tuning for Number of Nodes and Hidden Layers

## Introduction

One of the most important decisions when building an Artificial Neural Network (ANN) is choosing:

- The **number of hidden layers**
- The **number of neurons (nodes)** in each hidden layer

These are **hyperparameters**, meaning they are chosen **before training** and can significantly affect the model's performance.

Instead of selecting these values manually, **KerasTuner** can automatically find the optimal architecture.

---

# What are Hidden Layers?

## Definition

A **Hidden Layer** is the layer between the input layer and the output layer where the neural network learns patterns from the data.

Example:

```
Input Layer
      ↓
Hidden Layer 1
      ↓
Hidden Layer 2
      ↓
Output Layer
```

The more hidden layers a network has, the deeper the neural network becomes.

---

# What are Nodes (Neurons)?

## Definition

A **Node (Neuron)** is a computational unit inside a hidden layer.

Each neuron:

- Receives inputs
- Applies weights and bias
- Uses an activation function
- Produces an output

Example:

```
Hidden Layer

●  ●  ●  ●  ●

5 Neurons
```

---

# Why Tune Hidden Layers and Nodes?

The architecture of a neural network directly affects its learning capability.

Choosing too few or too many layers/neurons can lead to poor performance.

---

# Effect of Number of Hidden Layers

## Too Few Hidden Layers

```
Input
   ↓
Hidden
   ↓
Output
```

### Problems

- Underfitting
- Low accuracy

---

## Too Many Hidden Layers

```
Input
 ↓
Hidden
 ↓
Hidden
 ↓
Hidden
 ↓
Hidden
 ↓
Output
```

### Problems

- Overfitting
- Increased training time
- More memory usage
- Vanishing gradient (in very deep networks)
- Harder optimization

---

## Balanced Architecture

```
Input
 ↓
Hidden
 ↓
Hidden
 ↓
Output
```

Benefits:

- Learns complex patterns
- Good generalization
- Faster training
- Better accuracy

---

# Underfitting vs Good Fit vs Overfitting

| Model | Characteristics |
|--------|-----------------|
| Underfitting | Too few layers/neurons, poor training and testing performance |
| Good Fit | Balanced architecture with good generalization |
| Overfitting | Too many layers/neurons, excellent training performance but poor testing performance |

# Typical KerasTuner Model Building Function

The model-building function defines which hyperparameters KerasTuner is allowed to search.

Example:

```python
def build_model(hp):

    model = Sequential()

    # Number of hidden layers
    for i in range(hp.Int(
            "num_hidden_layers",
            min_value=1,
            max_value=5)):

        model.add(Dense(
            units=hp.Choice(
                f"units_{i}",
                values=[32,64,128,256]),
            activation="relu"
        ))

    model.add(Dense(1, activation="sigmoid"))

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

Here, KerasTuner searches for:

- Number of hidden layers
- Number of neurons in each layer

without manually changing the code.

# Best Practices

## Choosing Hidden Layers

General guideline:

| Problem Complexity | Hidden Layers |
|--------------------|--------------:|
| Simple | 1–2 |
| Medium | 2–4 |
| Complex | 4–10+ (depending on architecture) |

> Modern architectures like ResNet or Transformers may contain dozens or even hundreds of layers, but they use specialized designs rather than plain dense layers.

---

## Choosing Number of Neurons

Common choices:

```text
16
32
64
128
256
512
```

These powers of two are common because they often align well with hardware optimization, though they are not a strict requirement.

# Tips for Efficient Hyperparameter Tuning

- Start with a small search space.
- Tune one group of hyperparameters at a time.
- Use EarlyStopping.
- Use a validation set.
- Avoid unnecessarily large models.
- Monitor validation accuracy and validation loss.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. What is a hidden layer?

**Answer:**
A hidden layer is an intermediate layer between the input and output layers where the neural network learns useful patterns from the input data.

---

### 2. What is a neuron (node)?

**Answer:**
A neuron is a computational unit that receives inputs, applies weights and an activation function, and produces an output for the next layer.

---

### 3. Why do we tune the number of hidden layers?

**Answer:**
To find the right model complexity. Too few layers can underfit the data, while too many layers can overfit and increase computational cost.

---

### 4. Why do we tune the number of neurons?

**Answer:**
The number of neurons determines the learning capacity of a layer. Choosing the right number improves both accuracy and generalization.

---

### 5. Can KerasTuner tune hidden layers automatically?

**Answer:**
Yes. KerasTuner can search over different numbers of hidden layers and neurons to identify the architecture that performs best on validation data.

---

## Intermediate Questions

### 6. What happens if a network has too many hidden layers?

**Answer:**

* Overfitting
* Increased training time
* Higher memory usage
* Optimization challenges
* Possible vanishing gradient issues in very deep networks

---

### 7. What happens if a network has too few neurons?

**Answer:**
The model may not have enough capacity to learn complex relationships, leading to underfitting and poor predictive performance.

---

### 8. Why are values like 32, 64, and 128 commonly used for neurons?

**Answer:**
These powers of two often work efficiently with modern hardware and provide convenient scaling, although any suitable number can be used if it fits the problem.

---

### 9. Which metric should be used to select the best architecture?

**Answer:**
Validation metrics (e.g., validation accuracy or validation loss) should be used because they better reflect how the model generalizes to unseen data.

---

### 10. What other hyperparameters are commonly tuned with layers and neurons?

**Answer:**

* Learning rate
* Optimizer
* Activation function
* Batch size
* Dropout rate
* Regularization strength
