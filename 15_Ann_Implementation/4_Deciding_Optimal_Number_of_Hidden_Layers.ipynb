# Deciding the Optimal Number of Hidden Layers using KerasTuner

## Introduction

One of the most important architectural decisions in building an Artificial Neural Network (ANN) is determining the **optimal number of hidden layers**.

There is **no universal rule** for selecting the perfect number of hidden layers. The ideal architecture depends on:

- Dataset size
- Problem complexity
- Feature representation
- Model performance on validation data

Instead of manually trying different architectures, **KerasTuner** can automatically search for the optimal number of hidden layers.

---

# What is a Hidden Layer?

## Definition

A **Hidden Layer** is an intermediate layer between the input layer and output layer where the neural network learns hidden patterns and relationships from the input data.

Example:

```
Input Layer
      ↓
Hidden Layer 1
      ↓
Hidden Layer 2
      ↓
Output Layer
```

Each additional hidden layer enables the network to learn more abstract and complex representations.

---

# Why Do We Tune the Number of Hidden Layers?

Selecting the correct number of hidden layers directly affects model performance.

A poor choice can result in:

- Underfitting
- Overfitting
- Slow training
- Increased computational cost

The objective is to find a balance between **learning capacity** and **generalization**.

---

# Defining Hidden Layer Search Space

Example search range:

```python
hp.Int(
    "num_hidden_layers",
    min_value=1,
    max_value=5,
    step=1
)
```

Possible values:

```
1

2

3

4

5
```

KerasTuner evaluates each configuration and selects the one that performs best on the validation set.

---

# Building Dynamic Hidden Layers

Instead of writing fixed layers manually:

```python
Dense(64)

Dense(64)

Dense(64)
```

we use a loop.

Example:

```python
for i in range(
    hp.Int(
        "num_hidden_layers",
        1,
        5
    )
):

    model.add(
        Dense(
            units=64,
            activation="relu"
        )
    )
```

Now KerasTuner decides whether the model should have:

- 1 hidden layer
- 2 hidden layers
- 3 hidden layers
- 4 hidden layers
- 5 hidden layers

without changing the code.

# Best Practices

## Start Small

Search Example

```
1

2

3

4
```

Avoid searching through dozens of layer combinations initially.

---

## Use Validation Accuracy

Choose the architecture based on **validation performance**, not training accuracy.

---

## Combine with EarlyStopping

EarlyStopping:

- Saves training time
- Prevents overfitting
- Stops training when validation performance no longer improves

---

## Tune One Hyperparameter at a Time

Recommended order:

1. Hidden Layers
2. Number of Neurons
3. Learning Rate
4. Optimizer
5. Batch Size

This makes it easier to understand the effect of each hyperparameter.

# Common Mistakes

❌ Searching an excessively large range of hidden layers.

❌ Ignoring validation loss and validation accuracy.

❌ Not using EarlyStopping during tuning.

# 2. Interview Questions & Answers

## Beginner Questions

### 1. Why do we tune the number of hidden layers?

**Answer:**
To determine the model complexity that provides the best balance between learning capability and generalization.

---

### 2. Can adding more hidden layers always improve accuracy?

**Answer:**
No. While deeper networks can learn more complex patterns, too many layers may lead to overfitting, longer training times, and optimization difficulties.

---

### 3. How does KerasTuner help?

**Answer:**
KerasTuner automatically evaluates different numbers of hidden layers and selects the architecture that performs best on the validation set.

---

### 4. Which metric should be used to choose the best architecture?

**Answer:**
Validation metrics such as **validation accuracy** or **validation loss**, because they reflect performance on unseen data.

---

### 5. Why shouldn't training accuracy alone be used?

**Answer:**
A model may achieve very high training accuracy by memorizing the training data, but still perform poorly on new data due to overfitting.

---

## Intermediate Questions

### 6. What is `hp.Int()` in KerasTuner?

**Answer:**
`hp.Int()` defines an integer hyperparameter search space. For example, it can allow KerasTuner to search between 1 and 5 hidden layers.

---

### 7. Why is a loop used when adding hidden layers?

**Answer:**
A loop allows the model architecture to change dynamically based on the hyperparameter selected by KerasTuner, avoiding hardcoded layers.

---

### 8. Should you search between 1 and 100 hidden layers?

**Answer:**
No. Such a large search space increases computational cost and is usually unnecessary. Start with a small, realistic range.

---

### 9. How can EarlyStopping improve tuning?

**Answer:**
It stops training when validation performance no longer improves, saving time and reducing the risk of overfitting.

---

### 10. What factors influence the optimal number of hidden layers?

**Answer:**

* Dataset size
* Problem complexity
* Feature quality
* Computational resources
* Validation performance

---

## Scenario-Based Questions

### Q1. KerasTuner selects 2 hidden layers, but a manually built 5-layer model has higher training accuracy. Which model should you prefer?

**Answer:**
Prefer the model with better **validation performance**, since it is more likely to generalize well to unseen data.

---

### Q2. Validation accuracy starts decreasing as more hidden layers are added. What does this indicate?

**Answer:**
The model is likely becoming too complex and beginning to overfit.

---

### Q3. You are working with a simple binary classification dataset. Should you start with 10 hidden layers?

**Answer:**
No. Begin with a simple architecture (e.g., 1–3 hidden layers) and increase complexity only if validation performance indicates it is necessary.
