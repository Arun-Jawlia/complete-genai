## `What is an ndarray?`

An ndarray is:

- A collection of elements
- Stored in continuous memory
- Usually of the same data type
- Optimized for fast computation

### Types of Arrays
1. `1D Array`
   - Looks like a normal list.

2. `2D Array`
    - Looks like a table or matrix.
  
3. `3D Array`: 
   - Represents higher-dimensional data.
   - Used in:
       1. Image Processing
       2. Deep Learning
       3. Scientific Simulations

### `NumPy Arrays vs Python Sequences`

`Python Sequences`: 
Lists, Tuples

Python lists are:

- dynamic
- flexible
- can store different data types

But they are slower.

`NumPy Arrays`

NumPy arrays are:

- fixed-size
- homogeneous
- memory efficient
- faster


`Difference 1: Fixed Size`

1. NumPy arrays have a fixed size after creation.

   - You cannot directly increase size or manipulate like Python lists.

   - If size changes:
       - a new array is created
       - old array may be deleted
  
2. Python Lists are Dynamic

   - Lists grow dynamically.

   - This flexibility makes lists slower.
  
`Difference 2: Same Data Type`

NumPy arrays usually store elements of the same data type.
   - This allows:
      - efficient memory allocation
      - faster processing

Python Lists Can Store Mixed Types.
   - Different object types increase:
      - memory usage
      - processing overhead

`Difference 3: Memory Efficiency`

NumPy arrays store data in: contiguous memory blocks

Python lists store: references to objects

Therefore NumPy uses less memory.

`Difference 4: Faster Mathematical Operations`

NumPy performs operations directly on arrays.
