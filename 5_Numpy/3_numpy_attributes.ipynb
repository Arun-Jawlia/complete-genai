# `NumPy Attributes`
---

```python
import numpy as np
```

## 1. `ndim`
- ndim stands for n dimesions
- Returns number of dimensions

```python
one_darray = np.array([1,2,3,4,5])
print(one_darray.ndim)
```

```python
two_darray = np.array([[1,2,3],[5,6,7]])
print(two_darray.ndim)
```

```python
three_darray = np.array([[[1,2,3],[5,6,7]],[[8,9,10],[11,12,13]]])
three_darray.ndim
```

## 2. `shape` : 
- provide the idea about the rows, and column in our dataset or in array
- Returns dimensions of array.

```python
one_darray = np.array([1,2,3,4,5]) 
print(one_darray.shape)
```

```python
two_darray = np.array([[1,2,3],[5,6,7]])
print(two_darray.shape)
```

```python
three_darray = np.array([[[1,2,3],[5,6,7]],[[8,9,10],[11,12,13]]]) #
print(three_darray.shape) # row, col, z-axis ==> 2, 2 ,3
```

## 3. `size`: 
- returns total number of elements present inside our dataset or Array

```python
array_elements = np.arange(5,57)
print(array_elements.size)
```

## 4. `itemsize`: 
- provides with the idea about the size of a unique element inside our array in bytes
- Returns size of one element in bytes.

```python
print(np.array([[1,2], [3,4], [5,6]]).itemsize) # 8 byte
print(np.array([[1, 2, 10], [3, 4, 20], [5, 6, 40]]).itemsize) # 8 byte
```

```python
print(np.array([[1, 2, 10], [3, 4, 20], [5, 6, 40]], dtype=np.int32).itemsize) # 4 byte - consuming less memory
```

## 5. `dtype`: 
- provides us with the idea about the type of every item inside an array

```python
np.array([[1,2,3], [4,5,6]]).dtype # every item store up to 8 bytes
```

## 6. `astype()`
- Converts datatype

```python
arr = np.array([13,4,4,5])
arr = arr.astype(float)
print(arr)
```
