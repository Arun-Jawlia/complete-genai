# `Trigonometric Functions in NumPy`
---

```python
import numpy as np
```

## 1. `sin()`
- sin value for the value we are going to pass inside it 
- Synaxt: np.sin()

```python
print(np.sin(0))
print(np.sin(30))
print(np.sin(60))
```

```python
print(np.cos(0))
print(np.cos(30))
print(np.cos(60))
```

```python
print(np.tan(0))
print(np.tan(30))
print(np.tan(60))
```
