### `Formulas:` 

## `np.array()`: Creates a NumPy array from Python lists or tuples.
- Creating the Multidimensional Arrays
- Syntax: `np.array(object)`

1D Array or Vector Array

```python
numpy_array = np.array([1,2,3,4,5])
print(numpy_array)
print(type(numpy_array))

# ndarray --> 1d array
```

2D Array or Matrix Array

```python
# 2D array: matrix array

two_darray = np.array([[1,2,3], [4,5,6]]) # any number row and columen 
print(two_darray)
print(type(two_darray))
```

3D Array or Tensors

```python
# 3d Array: tensor array
# z-axis stand for the 3d axis
# x-axis stand for the row
# y-axis stand for the column
three_darray = np.array([[[1,2,3], [4,5,6]],[[8,9,10], [11,23,13]], [[44,55,66], [18,75,100]]])
print(three_darray)
```

## `dtype` Parameter 
- which can help us out with deciding about the size of the data we are going to store inside our array

```python
a1 = np.array([1,2,3,4,5], dtype=np.int32)
a2 = np.array([7,8,9,10,11], dtype=np.int64)

print(a1)
print(a2)
# np.float64 or np.float32
```

```python
np.array([1,2,3,4,5], dtype=int)
```

```python
np.array([7,8,9,10,11], dtype=float)
```

```python
np.array([7,8,9,10,11,0, -1], dtype=bool)
```

## `np.arange()` 
- Creates values within a range.
- Syntax: `np.arange(start,stop, step)` Note:  like range function in python

```python
# In Python
empty_list = []
for i in range(16):
    empty_list.append(i)

print(empty_list)
```

```python
# In Numpy
r = np.arange(16) # start from 0 by default and end on 16
print(r)

r1 = np.arange(1,16) # start from 1 and end 15 
print(r1)

r2 = np.arange(2,21,2) # start with 2 , end to 21 and step is 2
print(r2)
```

## `np.reshape()`
- helps us out with changing the shape of the array 
- Syntax `np.shape(row,column)`

```python
arange_data = np.arange(1,17) # start, end
reshaped_data = arange_data.reshape(4,4) # row, column
reshaped_data2 = arange_data.reshape(2,8) # row, column
# reshaped_data3 = arange_data.reshape(2,9) # ValueError: cannot reshape array of size 16 into shape (2,9
print(reshaped_data)
print(reshaped_data2)
# print(reshaped_data3)

# Note: row * column = should be equal of the length of the array
```

- Sytax: `np.ones(shape)`

```python
np.ones((4,4), dtype=int)
```

## `np.linspace()`
- it make sure that every value with the given range is going to have similar distance from each other
- or, Creates evenly spaced values between two limits.
- Syntax: `np.linspace(start, stop, num)`
 

```python
arr = np.linspace(0, 10, 5)
print(arr)
```

```python
np.linspace(-10,10,10) # necessary for machine learning
```

## `np.full()`
- Creates an array filled with a specified value
- Syntax: `np.full(shape, value)`

```python
arr = np.full((2, 2), 7)
print(arr)
```

```python
arr = np.full((5, 4), 1)
print(arr)
```

## `np.empty()`
- Creates an array without initializing entries.
- Sytax: `np.empty(shape)`


```python
arr = np.empty((2, 2))
print(arr)
```
