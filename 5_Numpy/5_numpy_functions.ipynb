# `NumPy Statistical Functions`
---

```python
import numpy as np
```

## 1. `mean()`
- mean of all values inside the array and matrix
- Syntax: np.mean()

```python
arr = np.arange(1,20)
print("mean", np.mean(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Mean of Matrix: ", np.mean(mat))
print("\n")
print("Mean of Columns: ", np.mean(mat, axis= 0))
print("\n")
print("Mean of Rows: ", np.mean(mat, axis = 1))
```

## 2. `std()`
- provides with the standard deviation of all valeus of the matrix or array
- Syntax: np.std(arr)

```python
arr = np.arange(1,20)
print("Standard Deviation", np.std(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Standard Deviation of Matrix: ", np.std(mat))
print("\n")
print("Standard Deviation of Columns: ", np.std(mat, axis= 0))
print("\n")
print("Standard Deviation of Rows: ", np.std(mat, axis = 1))
```

## 3. `median()`
- help us out calculate the Median of all values in the Array or Matrix
- Syntax: `np.median(arr)`

```python
arr = np.arange(1,20)
print("Median of an array: ", np.median(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Median of Matrix: ", np.median(mat))
print("\n")
print("Median of Columns: ", np.median(mat, axis= 0))
print("\n")
print("Median of Rows: ", np.median(mat, axis = 1))
```

## 4. `var()`
- calculating the variance of all values inside the array or matrix
- Syntax: np.var()

```python
arr = np.arange(1,20)
print("Variance of an array: ", np.median(arr))
```

```python
mat = np.round(np.random.random((4,4))* 100).astype("int32")
print(mat)
print("\n")
print("Variance of Matrix: ", np.var(mat))
print("\n")
print("Variance of Columns: ", np.var(mat, axis= 0))
print("\n")
print("Variance of Rows: ", np.var(mat, axis = 1))
```

## `round()`
- helps us rounding off the float values to its nearest possible integer value
- Syntax: np.round()

eg.
- if values is 1.4 --> round off is 0.0
- if values is 1.5 --> round off is 0.0
- if values is 1.6 --> round off is 2.0

```python
print(np.round(1.4))
print(np.round(1.5))
print(np.round(1.6))
```

```python
mat =  np.random.random((3,4))*10
print(mat)
print('\n')

print(np.round(mat))
```

## `ceil()`
- helps us rounding off the float values to its nearest max value
- Syntax: np.ceil()

eg.
- if values is 1.4 --> ceil value is 2.0
- if values is 1.5 --> ceil value is 2.0
- if values is 1.6 --> ceil value is 2.0

```python
print(np.ceil(1.4))
print(np.ceil(1.5))
print(np.ceil(1.6))
```

```python
mat =  np.random.random((3,4))*10
print(mat)
print('\n')

print(np.ceil(mat))
```

## `floor()`
- helps us rounding off the float values to its nearest in value
- Syntax: `np.ceil()`

eg.
- if values is 1.4 --> floor value is 1.0
- if values is 1.5 --> floor value is 1.0
- if values is 1.9 --> floor value is 1.0

```python
print(np.floor(1.4))
print(np.floor(1.5))
print(np.floor(1.6))
```

```python
mat =  np.random.random((3,4))*10
print(mat)
print('\n')

print(np.floor(mat))
```
